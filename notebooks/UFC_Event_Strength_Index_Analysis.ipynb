```python
# Convert the date column from text into real Python dates
ufc_fights["date"] = pd.to_datetime(ufc_fights["date"])

# Sort every UFC fight from oldest to newest
ufc_fights = ufc_fights.sort_values(
    by="date",
    ascending=True
).reset_index(drop=True)

print("Date type:", ufc_fights["date"].dtype)
print("Oldest fight date:", ufc_fights["date"].min())
print("Newest fight date:", ufc_fights["date"].max())

ufc_fights[["date", "event_name", "f1_name", "f2_name"]].head(10)
```

```python
# Create dictionaries to keep track of every fighter's UFC record
fighter_wins = {}
fighter_losses = {}

# Lists that will store each fighter's record BEFORE each fight
f1_wins_before = []
f1_losses_before = []
f2_wins_before = []
f2_losses_before = []

# Go through UFC history one fight at a time
for _, fight in ufc_fights.iterrows():

    f1 = fight["f1_name"]
    f2 = fight["f2_name"]

    # Get each fighter's record BEFORE this fight
    f1_wins_before.append(fighter_wins.get(f1, 0))
    f1_losses_before.append(fighter_losses.get(f1, 0))

    f2_wins_before.append(fighter_wins.get(f2, 0))
    f2_losses_before.append(fighter_losses.get(f2, 0))

    # Update records AFTER the fight
    if fight["f1_result"] == "W":
        fighter_wins[f1] = fighter_wins.get(f1, 0) + 1
        fighter_losses[f2] = fighter_losses.get(f2, 0) + 1

    elif fight["f2_result"] == "W":
        fighter_wins[f2] = fighter_wins.get(f2, 0) + 1
        fighter_losses[f1] = fighter_losses.get(f1, 0) + 1


# Add our new statistics to the dataset
ufc_fights["f1_ufc_wins_before"] = f1_wins_before
ufc_fights["f1_ufc_losses_before"] = f1_losses_before
ufc_fights["f2_ufc_wins_before"] = f2_wins_before
ufc_fights["f2_ufc_losses_before"] = f2_losses_before

print("Historical UFC records created!")
```

```python
from datetime import datetime

soup = BeautifulSoup(response.text, "html.parser")

event_rows = soup.select("tr.b-statistics__table-row")

new_events = []

for row in event_rows:
    link = row.select_one("a.b-link")
    date_tag = row.select_one("span.b-statistics__date")

    if link and date_tag:
        event_name = link.get_text(strip=True)
        event_url = link.get("href")
        event_date = datetime.strptime(
            date_tag.get_text(strip=True),
            "%B %d, %Y"
        )

        if event_date > datetime(2026, 6, 27):
            new_events.append({
                "date": event_date,
                "event_name": event_name,
                "event_url": event_url
            })

new_events = pd.DataFrame(new_events)

if len(new_events) > 0:
    new_events = new_events.sort_values("date").reset_index(drop=True)

print("New completed events found:", len(new_events))

new_events
```

```python
# Track every fighter's UFC career
fighter_wins = {}
fighter_losses = {}
fighter_fights = {}

# Store stats BEFORE each fight
f1_wins_before = []
f1_losses_before = []
f1_fights_before = []

f2_wins_before = []
f2_losses_before = []
f2_fights_before = []

for _, fight in ufc_fights.iterrows():

    f1 = fight["f1_name"]
    f2 = fight["f2_name"]

    # Record stats BEFORE this fight
    f1_wins_before.append(fighter_wins.get(f1, 0))
    f1_losses_before.append(fighter_losses.get(f1, 0))
    f1_fights_before.append(fighter_fights.get(f1, 0))

    f2_wins_before.append(fighter_wins.get(f2, 0))
    f2_losses_before.append(fighter_losses.get(f2, 0))
    f2_fights_before.append(fighter_fights.get(f2, 0))

    # Count the fight as UFC experience
    fighter_fights[f1] = fighter_fights.get(f1, 0) + 1
    fighter_fights[f2] = fighter_fights.get(f2, 0) + 1

    # Update wins/losses AFTER the fight
    if fight["f1_result"] == "W":
        fighter_wins[f1] = fighter_wins.get(f1, 0) + 1
        fighter_losses[f2] = fighter_losses.get(f2, 0) + 1

    elif fight["f2_result"] == "W":
        fighter_wins[f2] = fighter_wins.get(f2, 0) + 1
        fighter_losses[f1] = fighter_losses.get(f1, 0) + 1


# Add everything to our dataset
ufc_fights["f1_ufc_wins_before"] = f1_wins_before
ufc_fights["f1_ufc_losses_before"] = f1_losses_before
ufc_fights["f1_ufc_fights_before"] = f1_fights_before

ufc_fights["f2_ufc_wins_before"] = f2_wins_before
ufc_fights["f2_ufc_losses_before"] = f2_losses_before
ufc_fights["f2_ufc_fights_before"] = f2_fights_before

print("UFC record and experience calculated!")
```

```python
# Track each fighter's current UFC win streak
win_streaks = {}

# Store win streak BEFORE each fight
f1_streak_before = []
f2_streak_before = []

for _, fight in ufc_fights.iterrows():

    f1 = fight["f1_name"]
    f2 = fight["f2_name"]

    # Record streak BEFORE the fight
    f1_streak_before.append(win_streaks.get(f1, 0))
    f2_streak_before.append(win_streaks.get(f2, 0))

    # Update streaks AFTER the fight
    if fight["f1_result"] == "W":
        win_streaks[f1] = win_streaks.get(f1, 0) + 1
        win_streaks[f2] = 0

    elif fight["f2_result"] == "W":
        win_streaks[f2] = win_streaks.get(f2, 0) + 1
        win_streaks[f1] = 0

    # Draws / NCs do not add a win
    else:
        win_streaks[f1] = 0
        win_streaks[f2] = 0


# Add streaks to our dataset
ufc_fights["f1_win_streak_before"] = f1_streak_before
ufc_fights["f2_win_streak_before"] = f2_streak_before

print("Historical UFC win streaks calculated!")
```

```python
# Calculate UFC win percentage ENTERING each fight

ufc_fights["f1_ufc_win_pct_before"] = (
    ufc_fights["f1_ufc_wins_before"] /
    (
        ufc_fights["f1_ufc_wins_before"] +
        ufc_fights["f1_ufc_losses_before"]
    )
).fillna(0)

ufc_fights["f2_ufc_win_pct_before"] = (
    ufc_fights["f2_ufc_wins_before"] /
    (
        ufc_fights["f2_ufc_wins_before"] +
        ufc_fights["f2_ufc_losses_before"]
    )
).fillna(0)

print("UFC win percentages calculated!")

display(
    ufc_fights[
        [
            "f1_name",
            "f1_ufc_wins_before",
            "f1_ufc_losses_before",
            "f1_ufc_win_pct_before",
            "f2_name",
            "f2_ufc_wins_before",
            "f2_ufc_losses_before",
            "f2_ufc_win_pct_before"
        ]
    ].tail(10)
)
```

```python
# Track each fighter's recent UFC results
recent_results = {}

f1_last5_win_pct = []
f2_last5_win_pct = []

for _, fight in ufc_fights.iterrows():

    f1 = fight["f1_name"]
    f2 = fight["f2_name"]

    # Get previous results
    f1_history = recent_results.get(f1, [])
    f2_history = recent_results.get(f2, [])

    # Last 5 UFC fights BEFORE this fight
    f1_last5 = f1_history[-5:]
    f2_last5 = f2_history[-5:]

    # Calculate win percentage
    f1_pct = (
        sum(f1_last5) / len(f1_last5)
        if len(f1_last5) > 0 else 0
    )

    f2_pct = (
        sum(f2_last5) / len(f2_last5)
        if len(f2_last5) > 0 else 0
    )

    f1_last5_win_pct.append(f1_pct)
    f2_last5_win_pct.append(f2_pct)

    # Update histories AFTER the fight
    if fight["f1_result"] == "W":
        recent_results.setdefault(f1, []).append(1)
        recent_results.setdefault(f2, []).append(0)

    elif fight["f2_result"] == "W":
        recent_results.setdefault(f1, []).append(0)
        recent_results.setdefault(f2, []).append(1)

    else:
        # Draw / NC = neither fighter gets a win
        recent_results.setdefault(f1, []).append(0)
        recent_results.setdefault(f2, []).append(0)


ufc_fights["f1_last5_win_pct"] = f1_last5_win_pct
ufc_fights["f2_last5_win_pct"] = f2_last5_win_pct

print("Recent 5-fight form calculated!")
```

```python
# Track wins and finishes BEFORE each fight
career_wins = {}
career_finishes = {}

f1_finish_rate_before = []
f2_finish_rate_before = []

for _, fight in ufc_fights.iterrows():

    f1 = fight["f1_name"]
    f2 = fight["f2_name"]

    # Current stats BEFORE this fight
    f1_wins = career_wins.get(f1, 0)
    f2_wins = career_wins.get(f2, 0)

    f1_finishes = career_finishes.get(f1, 0)
    f2_finishes = career_finishes.get(f2, 0)

    # Finish rate among UFC wins
    f1_rate = f1_finishes / f1_wins if f1_wins > 0 else 0
    f2_rate = f2_finishes / f2_wins if f2_wins > 0 else 0

    f1_finish_rate_before.append(f1_rate)
    f2_finish_rate_before.append(f2_rate)

    # Determine whether THIS fight ended by finish
    method = str(fight["method"]).upper()

    is_finish = (
        "KO" in method
        or "TKO" in method
        or "SUBMISSION" in method
    )

    # Update records AFTER the fight
    if fight["f1_result"] == "W":

        career_wins[f1] = career_wins.get(f1, 0) + 1

        if is_finish:
            career_finishes[f1] = career_finishes.get(f1, 0) + 1

    elif fight["f2_result"] == "W":

        career_wins[f2] = career_wins.get(f2, 0) + 1

        if is_finish:
            career_finishes[f2] = career_finishes.get(f2, 0) + 1


ufc_fights["f1_finish_rate_before"] = f1_finish_rate_before
ufc_fights["f2_finish_rate_before"] = f2_finish_rate_before

print("Historical UFC finish rates calculated!")
```

```python
# Track total UFC finishes BEFORE each fight

fighter_finishes = {}

f1_finishes_before = []
f2_finishes_before = []

for _, fight in ufc_fights.iterrows():

    f1 = fight["f1_name"]
    f2 = fight["f2_name"]

    # Record finish totals BEFORE this fight
    f1_finishes_before.append(
        fighter_finishes.get(f1, 0)
    )

    f2_finishes_before.append(
        fighter_finishes.get(f2, 0)
    )

    # Determine whether THIS fight was a finish
    method = str(fight["method"]).upper()

    is_finish = (
        "KO" in method
        or "TKO" in method
        or "SUBMISSION" in method
    )

    # Update AFTER the fight
    if fight["f1_result"] == "W" and is_finish:
        fighter_finishes[f1] = fighter_finishes.get(f1, 0) + 1

    elif fight["f2_result"] == "W" and is_finish:
        fighter_finishes[f2] = fighter_finishes.get(f2, 0) + 1


ufc_fights["f1_ufc_finishes_before"] = f1_finishes_before
ufc_fights["f2_ufc_finishes_before"] = f2_finishes_before

print("Previous UFC finishes calculated!")
```

```python
def get_ranking_division(weight_class):

    wc = str(weight_class)

    # Women's divisions first
    if "Women's Strawweight" in wc:
        return "Women's Strawweight"

    if "Women's Flyweight" in wc:
        return "Women's Flyweight"

    if "Women's Bantamweight" in wc:
        return "Women's Bantamweight"

    if "Women's Featherweight" in wc:
        return "Women's Featherweight"

    # Men's divisions
    if "Light Heavyweight" in wc:
        return "Light Heavyweight"

    if "Heavyweight" in wc:
        return "Heavyweight"

    if "Middleweight" in wc:
        return "Middleweight"

    if "Welterweight" in wc:
        return "Welterweight"

    if "Lightweight" in wc:
        return "Lightweight"

    if "Featherweight" in wc:
        return "Featherweight"

    if "Bantamweight" in wc:
        return "Bantamweight"

    if "Flyweight" in wc:
        return "Flyweight"

    # Catchweight, Open Weight, tournaments, etc.
    return None


ufc_fights["ranking_division"] = (
    ufc_fights["weight_class"].apply(get_ranking_division)
)

print("Ranking divisions created!")

display(
    ufc_fights[
        ["weight_class", "ranking_division"]
    ]
    .drop_duplicates()
    .sort_values("ranking_division")
)
```

```python
# Make sure dates are datetime
ufc_fights["date"] = pd.to_datetime(ufc_fights["date"])
rankings["date"] = pd.to_datetime(rankings["date"])

# Create lookup dictionary:
# (date, division, fighter) -> rank
ranking_lookup = {}

for _, row in rankings.iterrows():
    key = (
        row["date"],
        row["weightclass"],
        row["fighter"]
    )
    ranking_lookup[key] = row["rank"]


def find_historical_rank(fighter, division, fight_date):

    # No normal ranking division
    if pd.isna(division):
        return None

    # Rankings did not exist before Feb. 2013
    possible_dates = rankings[
        (rankings["date"] <= fight_date) &
        (rankings["weightclass"] == division)
    ]["date"]

    if len(possible_dates) == 0:
        return None

    latest_date = possible_dates.max()

    return ranking_lookup.get(
        (latest_date, division, fighter),
        None
    )


print("Adding Fighter 1 historical rankings...")

ufc_fights["f1_rank_before"] = ufc_fights.apply(
    lambda row: find_historical_rank(
        row["f1_name"],
        row["ranking_division"],
        row["date"]
    ),
    axis=1
)

print("Adding Fighter 2 historical rankings...")

ufc_fights["f2_rank_before"] = ufc_fights.apply(
    lambda row: find_historical_rank(
        row["f2_name"],
        row["ranking_division"],
        row["date"]
    ),
    axis=1
)

print("Historical rankings added!")
```

```python
import re

def extract_headliners(event_name):

    match = re.search(
        r":\s*(.+?)\s+vs\.?\s+(.+)$",
        str(event_name),
        flags=re.IGNORECASE
    )

    if match:
        return match.group(1).strip(), match.group(2).strip()

    return None, None


# Test famous events
test_events = [
    "UFC 196: McGregor vs Diaz",
    "UFC 205: Alvarez vs McGregor",
    "UFC 229: Khabib vs McGregor",
    "UFC 300: Pereira vs Hill",
    "UFC 100"
]

for event in test_events:
    print(event)
    print("Headliners:", extract_headliners(event))
    print()
```

```python
def match_headliner_to_fighter(short_name, event_fighters):

    if short_name is None:
        return None

    short_name = short_name.lower().strip()

    matches = [
        fighter for fighter in event_fighters
        if short_name in str(fighter).lower()
    ]

    if len(matches) == 1:
        return matches[0]

    return None


headliner_results = []

for _, event in unique_events[unique_events["has_vs"]].iterrows():

    event_name = event["event_name"]
    event_date = event["date"]

    h1_short, h2_short = extract_headliners(event_name)

    event_rows = ufc_fights[
        (ufc_fights["event_name"] == event_name) &
        (ufc_fights["date"] == event_date)
    ]

    event_fighters = list(
        set(event_rows["f1_name"]) |
        set(event_rows["f2_name"])
    )

    h1_full = match_headliner_to_fighter(
        h1_short, event_fighters
    )

    h2_full = match_headliner_to_fighter(
        h2_short, event_fighters
    )

    headliner_results.append({
        "event_name": event_name,
        "date": event_date,
        "headliner_1": h1_full,
        "headliner_2": h2_full
    })


headliners = pd.DataFrame(headliner_results)

print("Events tested:", len(headliners))

print(
    "Both headliners successfully matched:",
    (
        headliners["headliner_1"].notna() &
        headliners["headliner_2"].notna()
    ).sum()
)

print(
    "Events with a matching problem:",
    (
        headliners["headliner_1"].isna() |
        headliners["headliner_2"].isna()
    ).sum()
)

print("\nSample results:")

display(
    headliners.tail(10)
)
```

```python
from difflib import SequenceMatcher
import unicodedata

def clean_name(name):
    name = str(name).lower().strip()

    # Remove accents
    name = unicodedata.normalize("NFKD", name)
    name = "".join(c for c in name if not unicodedata.combining(c))

    return name


def smart_match(short_name, event_fighters):

    if short_name is None:
        return None

    short = clean_name(short_name)

    # First try normal matching
    direct = [
        fighter for fighter in event_fighters
        if short in clean_name(fighter)
    ]

    if len(direct) == 1:
        return direct[0]

    # Otherwise find closest fighter name
    scores = []

    for fighter in event_fighters:

        full = clean_name(fighter)

        score = SequenceMatcher(
            None,
            short,
            full
        ).ratio()

        # Also compare against each individual part of name
        for part in full.split():
            score = max(
                score,
                SequenceMatcher(None, short, part).ratio()
            )

        scores.append((score, fighter))

    scores.sort(reverse=True)

    # Only accept reasonably strong matches
    if scores and scores[0][0] >= 0.70:
        return scores[0][1]

    return None


# Test some of the problem names
test_names = [
    "Rampage",
    "Shogun",
    "Aldo",
    "Korean Zombie",
    "Cro Cop"
]

for name in test_names:

    print("\n", name)

    # Show closest match from ALL UFC fighters just as a test
    all_fighters = list(
        set(ufc_fights["f1_name"]) |
        set(ufc_fights["f2_name"])
    )

    print("Match:", smart_match(name, all_fighters))
```

```python
def match_headliner_pair(event_name, event_date):

    h1, h2 = extract_headliners(event_name)

    h1 = clean_headliner_label(h1)
    h2 = clean_headliner_label(h2)

    if h1 is None or h2 is None:
        return None, None

    event_rows = ufc_fights[
        (ufc_fights["event_name"] == event_name) &
        (ufc_fights["date"] == event_date)
    ]

    h1 = clean_name(h1)
    h2 = clean_name(h2)

    for _, fight in event_rows.iterrows():

        f1 = clean_name(fight["f1_name"])
        f2 = clean_name(fight["f2_name"])

        # Event name order matches fight order
        if h1 in f1 and h2 in f2:
            return fight["f1_name"], fight["f2_name"]

        # Or fighters may be stored in reverse order
        if h1 in f2 and h2 in f1:
            return fight["f2_name"], fight["f1_name"]

    return None, None


# Test the recent problem events
for event_name in [
    "UFC Fight Night: Yan vs. Figueiredo",
    "UFC Fight Night: Blanchfield vs. Barber",
    "UFC Fight Night: Oliveira vs. Gamrot",
    "UFC Fight Night: Bonfim vs. Brown"
]:

    row = unique_events[
        unique_events["event_name"] == event_name
    ].iloc[0]

    result = match_headliner_pair(
        event_name,
        row["date"]
    )

    print(event_name)
    print("Headliners:", result)
    print("-" * 60)
```

```python
import re
import pandas as pd

def extract_headliners(event_name):
    name = str(event_name)

    # Look for "vs" or "vs."
    match = re.search(
        r'[:\-]\s*(.+?)\s+vs\.?\s+(.+)$',
        name,
        flags=re.IGNORECASE
    )

    if not match:
        return pd.Series([None, None])

    fighter1 = match.group(1).strip()
    fighter2 = match.group(2).strip()

    fighter1 = translate_nickname(fighter1)
    fighter2 = translate_nickname(fighter2)

    return pd.Series([fighter1, fighter2])


event_headliners[
    ["headline_name_1", "headline_name_2"]
] = event_headliners["event_name"].apply(extract_headliners)

print("Headliners extracted! ✅")
print(
    "Events with two extracted names:",
    (
        event_headliners["headline_name_1"].notna()
        & event_headliners["headline_name_2"].notna()
    ).sum()
)

display(
    event_headliners[
        ["event_name", "headline_name_1", "headline_name_2"]
    ].tail(15)
)
```

```python
def name_matches(label, full_name):
    if label is None or full_name is None:
        return False

    label = str(label).lower().strip()
    full_name = str(full_name).lower().strip()

    return label == full_name or label in full_name


def find_main_event(row):
    event_name = row["event_name"]
    event_date = row["date"]

    h1 = row["headline_name_1"]
    h2 = row["headline_name_2"]

    if h1 is None or h2 is None:
        return pd.Series([None, None])

    fights = ufc_fights[
        (ufc_fights["event_name"] == event_name) &
        (ufc_fights["date"] == event_date)
    ]

    for _, fight in fights.iterrows():

        f1 = fight["f1_name"]
        f2 = fight["f2_name"]

        normal_order = (
            name_matches(h1, f1) and
            name_matches(h2, f2)
        )

        reverse_order = (
            name_matches(h1, f2) and
            name_matches(h2, f1)
        )

        if normal_order or reverse_order:
            return pd.Series([f1, f2])

    return pd.Series([None, None])


event_headliners[
    ["main_event_f1", "main_event_f2"]
] = event_headliners.apply(find_main_event, axis=1)

matched = (
    event_headliners["main_event_f1"].notna() &
    event_headliners["main_event_f2"].notna()
)

print("Total events:", len(event_headliners))
print("Main events matched:", matched.sum())
print("Still unresolved:", (~matched).sum())
print("Success rate:", round(matched.mean() * 100, 1), "%")
```

```python
validation_results = []

for _, event in event_headliners.iterrows():

    fighter1 = event["main_event_f1"]
    fighter2 = event["main_event_f2"]

    # Skip events where we intentionally don't have a main event
    if pd.isna(fighter1) or pd.isna(fighter2):
        continue

    fights = ufc_fights[
        (ufc_fights["event_name"] == event["event_name"]) &
        (ufc_fights["date"] == event["date"])
    ]

    match_found = (
        (
            (fights["f1_name"] == fighter1) &
            (fights["f2_name"] == fighter2)
        )
        |
        (
            (fights["f1_name"] == fighter2) &
            (fights["f2_name"] == fighter1)
        )
    ).any()

    validation_results.append({
        "event_name": event["event_name"],
        "date": event["date"],
        "main_event_f1": fighter1,
        "main_event_f2": fighter2,
        "valid": match_found
    })

validation_df = pd.DataFrame(validation_results)

print("Main events checked:", len(validation_df))
print("Valid main events:", validation_df["valid"].sum())
print("Invalid main events:", (~validation_df["valid"]).sum())

validation_df[validation_df["valid"] == False]
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text, "html.parser")

# Find links containing bonus labels
award_links = []

for link in soup.find_all("a"):
    text = link.get_text(" ", strip=True)

    if any(x in text for x in ["FOTN", "POTN", "KOTN", "SOTN"]):
        award_links.append({
            "text": text,
            "href": link.get("href")
        })

print("Award links found:", len(award_links))

award_links[:20]
```

```python
import re

bonus_records = []

for award in award_links:

    bonus_type = award["text"].replace("$", "").strip()
    href = award["href"]

    # Pull date from URL
    date_match = re.search(r"/bout/(\d{4}-\d{2}-\d{2})-", href)

    if date_match:
        bonus_date = date_match.group(1)

        bonus_records.append({
            "date": bonus_date,
            "bonus_type": bonus_type,
            "fight_url": href
        })

bonuses = pd.DataFrame(bonus_records)

bonuses["date"] = pd.to_datetime(bonuses["date"])

print("Bonus awards extracted:", len(bonuses))
print()
print(bonuses["bonus_type"].value_counts())
print()
bonuses.head(10)
```

```python
# Remove the date and "/bout/" from each URL
bonuses["fight_slug"] = (
    bonuses["fight_url"]
    .str.replace(r"^/bout/\d{4}-\d{2}-\d{2}-", "", regex=True)
)

print("Bonus records:", len(bonuses))
print()
bonuses[
    ["date", "bonus_type", "fight_slug"]
].head(20)
```

```python
import unicodedata
import re

def make_slug(name):
    name = str(name).lower()

    # Remove accents
    name = unicodedata.normalize("NFKD", name)
    name = "".join(
        c for c in name
        if not unicodedata.combining(c)
    )

    # Replace punctuation/spaces with hyphens
    name = re.sub(r"[^a-z0-9]+", "-", name)

    return name.strip("-")


ufc_fights["f1_slug"] = ufc_fights["f1_name"].apply(make_slug)
ufc_fights["f2_slug"] = ufc_fights["f2_name"].apply(make_slug)

# Build matchup in BOTH possible orders
ufc_fights["matchup_slug_1"] = (
    ufc_fights["f1_slug"] + "-vs-" + ufc_fights["f2_slug"]
)

ufc_fights["matchup_slug_2"] = (
    ufc_fights["f2_slug"] + "-vs-" + ufc_fights["f1_slug"]
)

print("Fight slugs created:", len(ufc_fights))

ufc_fights[
    ["date", "f1_name", "f2_name", "matchup_slug_1"]
].head(10)
```

```python
bonus_match_results = []

for _, bonus in bonuses.iterrows():

    match = ufc_fights[
        (ufc_fights["date"] == bonus["date"]) &
        (
            (ufc_fights["matchup_slug_1"] == bonus["fight_slug"]) |
            (ufc_fights["matchup_slug_2"] == bonus["fight_slug"])
        )
    ]

    bonus_match_results.append(len(match) == 1)

bonuses["matched"] = bonus_match_results

print("Total bonus records:", len(bonuses))
print("Matched:", bonuses["matched"].sum())
print("Unmatched:", (~bonuses["matched"]).sum())
print()

bonuses[bonuses["matched"] == False].head(20)
```

```python
# Start with empty bonus columns
for bonus_type in ["FOTN", "POTN", "KOTN", "SOTN"]:
    ufc_fights[bonus_type] = False


# Match each bonus award to its exact fight
for _, bonus in matched_bonuses.iterrows():

    mask = (
        (ufc_fights["date"] == bonus["date"]) &
        (
            (ufc_fights["matchup_slug_1"] == bonus["fight_slug"]) |
            (ufc_fights["matchup_slug_2"] == bonus["fight_slug"])
        )
    )

    ufc_fights.loc[mask, bonus["bonus_type"]] = True


# Check results
print("Fights with FOTN:", ufc_fights["FOTN"].sum())
print("Fights with POTN:", ufc_fights["POTN"].sum())
print("Fights with KOTN:", ufc_fights["KOTN"].sum())
print("Fights with SOTN:", ufc_fights["SOTN"].sum())
```

```python
# Running bonus history for each fighter
bonus_history = {}

f1_fotn = []
f1_potn = []
f1_kotn = []
f1_sotn = []

f2_fotn = []
f2_potn = []
f2_kotn = []
f2_sotn = []


def get_bonus_history(fighter):
    if fighter not in bonus_history:
        bonus_history[fighter] = {
            "FOTN": 0,
            "POTN": 0,
            "KOTN": 0,
            "SOTN": 0
        }

    return bonus_history[fighter]


for _, fight in ufc_fights.iterrows():

    f1 = fight["f1_name"]
    f2 = fight["f2_name"]

    h1 = get_bonus_history(f1)
    h2 = get_bonus_history(f2)

    # Bonus totals BEFORE this fight
    f1_fotn.append(h1["FOTN"])
    f1_potn.append(h1["POTN"])
    f1_kotn.append(h1["KOTN"])
    f1_sotn.append(h1["SOTN"])

    f2_fotn.append(h2["FOTN"])
    f2_potn.append(h2["POTN"])
    f2_kotn.append(h2["KOTN"])
    f2_sotn.append(h2["SOTN"])

    # Add bonuses earned in this fight AFTER recording previous totals
    if fight["FOTN"]:
        h1["FOTN"] += 1
        h2["FOTN"] += 1

    if fight["POTN"]:
        if fight["f1_result"] == "W":
            h1["POTN"] += 1
        elif fight["f2_result"] == "W":
            h2["POTN"] += 1

    if fight["KOTN"]:
        if fight["f1_result"] == "W":
            h1["KOTN"] += 1
        elif fight["f2_result"] == "W":
            h2["KOTN"] += 1

    if fight["SOTN"]:
        if fight["f1_result"] == "W":
            h1["SOTN"] += 1
        elif fight["f2_result"] == "W":
            h2["SOTN"] += 1


ufc_fights["f1_previous_FOTN"] = f1_fotn
ufc_fights["f1_previous_POTN"] = f1_potn
ufc_fights["f1_previous_KOTN"] = f1_kotn
ufc_fights["f1_previous_SOTN"] = f1_sotn

ufc_fights["f2_previous_FOTN"] = f2_fotn
ufc_fights["f2_previous_POTN"] = f2_potn
ufc_fights["f2_previous_KOTN"] = f2_kotn
ufc_fights["f2_previous_SOTN"] = f2_sotn

print("Historical bonus variables created ✅")
```

```python
# Reset the old incomplete bonus columns
for bonus_type in ["FOTN", "POTN", "KOTN", "SOTN"]:
    ufc_fights[bonus_type] = False


# Attach complete FightIQ bonus data
for i, fight in ufc_fights.iterrows():

    # Bonuses did not exist before 2006
    if fight["date"] < pd.Timestamp("2006-01-01"):
        continue

    match = fightiq_bonuses[
        (fightiq_bonuses["date"] == fight["date"]) &
        (
            (fightiq_bonuses["matchup_slug_1"] == fight["matchup_slug_1"]) |
            (fightiq_bonuses["matchup_slug_2"] == fight["matchup_slug_1"])
        )
    ]

    if len(match) == 1:
        bonus_row = match.iloc[0]

        ufc_fights.at[i, "FOTN"] = bonus_row["FOTN"]
        ufc_fights.at[i, "POTN"] = bonus_row["POTN"]
        ufc_fights.at[i, "KOTN"] = bonus_row["KOTN"]
        ufc_fights.at[i, "SOTN"] = bonus_row["SOTN"]


print("Complete FightIQ bonuses attached ✅")
print()
print("FOTN fights:", ufc_fights["FOTN"].sum())
print("POTN fights:", ufc_fights["POTN"].sum())
print("KOTN fights:", ufc_fights["KOTN"].sum())
print("SOTN fights:", ufc_fights["SOTN"].sum())
```

```python
# Make sure fights are chronological
ufc_fights = ufc_fights.sort_values("date").reset_index(drop=True)

bonus_history = {}

f1_fotn = []
f1_potn = []
f1_kotn = []
f1_sotn = []

f2_fotn = []
f2_potn = []
f2_kotn = []
f2_sotn = []


def get_bonus_history(fighter):
    if fighter not in bonus_history:
        bonus_history[fighter] = {
            "FOTN": 0,
            "POTN": 0,
            "KOTN": 0,
            "SOTN": 0
        }
    return bonus_history[fighter]


for _, fight in ufc_fights.iterrows():

    f1 = fight["f1_name"]
    f2 = fight["f2_name"]

    h1 = get_bonus_history(f1)
    h2 = get_bonus_history(f2)

    # Counts ENTERING this fight
    f1_fotn.append(h1["FOTN"])
    f1_potn.append(h1["POTN"])
    f1_kotn.append(h1["KOTN"])
    f1_sotn.append(h1["SOTN"])

    f2_fotn.append(h2["FOTN"])
    f2_potn.append(h2["POTN"])
    f2_kotn.append(h2["KOTN"])
    f2_sotn.append(h2["SOTN"])

    # Add bonuses AFTER this fight
    if fight["FOTN"]:
        h1["FOTN"] += 1
        h2["FOTN"] += 1

    if fight["POTN"]:
        if fight["f1_result"] == "W":
            h1["POTN"] += 1
        elif fight["f2_result"] == "W":
            h2["POTN"] += 1

    if fight["KOTN"]:
        if fight["f1_result"] == "W":
            h1["KOTN"] += 1
        elif fight["f2_result"] == "W":
            h2["KOTN"] += 1

    if fight["SOTN"]:
        if fight["f1_result"] == "W":
            h1["SOTN"] += 1
        elif fight["f2_result"] == "W":
            h2["SOTN"] += 1


# Replace the old incomplete historical columns
ufc_fights["f1_previous_FOTN"] = f1_fotn
ufc_fights["f1_previous_POTN"] = f1_potn
ufc_fights["f1_previous_KOTN"] = f1_kotn
ufc_fights["f1_previous_SOTN"] = f1_sotn

ufc_fights["f2_previous_FOTN"] = f2_fotn
ufc_fights["f2_previous_POTN"] = f2_potn
ufc_fights["f2_previous_KOTN"] = f2_kotn
ufc_fights["f2_previous_SOTN"] = f2_sotn

print("Complete historical bonus counts rebuilt ✅")
```

```python
import time
import numpy as np

def get_fighter_popularity(fighters, start_date, end_date, anchor="Conor McGregor"):

    fighters = [f for f in fighters if f != anchor]
    search_terms = [anchor] + fighters[:4]

    pytrends.build_payload(
        search_terms,
        timeframe=f"{start_date} {end_date}",
        geo="US"
    )

    data = pytrends.interest_over_time()

    if data.empty:
        return {fighter: np.nan for fighter in fighters}

    means = data[search_terms].mean()

    anchor_mean = means[anchor]

    if anchor_mean == 0:
        return {fighter: np.nan for fighter in fighters}

    scores = {}

    for fighter in fighters:
        scores[fighter] = (means[fighter] / anchor_mean) * 100

    return scores
```

```python
def get_event_popularity(fighters, start_date, end_date, anchor="Conor McGregor"):

    fighters = list(dict.fromkeys(fighters))
    fighters = [f for f in fighters if f != anchor]

    all_scores = {}

    for i in range(0, len(fighters), 4):
        batch = fighters[i:i+4]

        scores = get_fighter_popularity(
            fighters=batch,
            start_date=start_date,
            end_date=end_date,
            anchor=anchor
        )

        all_scores.update(scores)

        time.sleep(2)

    if anchor in fighters:
        all_scores[anchor] = 100.0

    return all_scores
```

```python
def get_fighter_popularity(fighters, start_date, end_date, anchor="Conor McGregor"):

    fighters = [f for f in fighters if f != anchor]
    fighters = fighters[:4]

    search_names = [get_trends_name(f) for f in fighters]
    search_terms = [anchor] + search_names

    pytrends.build_payload(
        search_terms,
        timeframe=f"{start_date} {end_date}",
        geo="US"
    )

    data = pytrends.interest_over_time()

    if data.empty:
        return {fighter: np.nan for fighter in fighters}

    means = data[search_terms].mean()
    anchor_mean = means[anchor]

    if anchor_mean == 0:
        return {fighter: np.nan for fighter in fighters}

    scores = {}

    for fighter, search_name in zip(fighters, search_names):
        scores[fighter] = (means[search_name] / anchor_mean) * 100

    return scores
```

```python
def find_fighter_topic(fighter):
    suggestions = pytrends.suggestions(fighter)

    exact_matches = [
        item for item in suggestions
        if item.get("title", "").lower() == fighter.lower()
    ]

    # First choice: clearly combat-sports related
    preferred_types = [
        "mixed martial",
        "martial artist",
        "fighter"
    ]

    for item in exact_matches:
        item_type = item.get("type", "").lower()

        if any(term in item_type for term in preferred_types):
            return item["mid"]

    # Second choice: exact-name generic Topic
    for item in exact_matches:
        if item.get("type", "").lower() == "topic":
            return item["mid"]

    # Last resort: exact-name entity of another type
    if exact_matches:
        return exact_matches[0]["mid"]

    return None
```

```python
def find_fighter_topic(fighter):
    suggestions = pytrends.suggestions(fighter)

    exact_matches = [
        item for item in suggestions
        if item.get("title", "").strip().lower() == fighter.strip().lower()
    ]

    if not exact_matches:
        return None

    # 1. Prefer clearly combat-sports-related entities
    combat_terms = [
        "mixed martial",
        "martial artist",
        "fighter",
        "ufc"
    ]

    for item in exact_matches:
        item_type = item.get("type", "").lower()

        if any(term in item_type for term in combat_terms):
            return item["mid"]

    # 2. Prefer a specific named entity over a generic "Topic"
    for item in exact_matches:
        item_type = item.get("type", "").lower()

        if item_type != "topic":
            return item["mid"]

    # 3. Fall back to the first exact-name generic Topic
    return exact_matches[0]["mid"]
```

```python
import unicodedata

def normalize_name(name):
    name = unicodedata.normalize("NFKD", str(name))
    name = "".join(
        char for char in name
        if not unicodedata.combining(char)
    )
    return name.strip().lower()


def find_fighter_topic(fighter):
    suggestions = pytrends.suggestions(fighter)

    fighter_normalized = normalize_name(fighter)

    exact_matches = [
        item for item in suggestions
        if normalize_name(item.get("title", "")) == fighter_normalized
    ]

    if not exact_matches:
        return None

    # 1. Prefer clearly combat-sports-related entities
    combat_terms = [
        "mixed martial",
        "martial artist",
        "fighter",
        "ufc"
    ]

    for item in exact_matches:
        item_type = item.get("type", "").lower()

        if any(term in item_type for term in combat_terms):
            return item["mid"]

    # 2. Prefer a specific entity over generic "Topic"
    for item in exact_matches:
        if item.get("type", "").lower() != "topic":
            return item["mid"]

    # 3. Generic Topic fallback
    return exact_matches[0]["mid"]
```

```python
TOPIC_PROGRESS_FILE = "/content/fighter_topics_progress.csv"

def save_topic_progress():
    fighter_topics.to_csv(
        TOPIC_PROGRESS_FILE,
        index=False
    )

    checked = (fighter_topics["status"] != "not_checked").sum()
    found = (fighter_topics["status"] == "found").sum()
    missing = (fighter_topics["status"] == "missing").sum()

    print("Progress saved ✅")
    print(f"Checked: {checked}/{len(fighter_topics)}")
    print(f"Found: {found}")
    print(f"Missing: {missing}")
```

```python
def collect_fighter_topics(max_fighters=25):

    unchecked = fighter_topics[
        fighter_topics["status"] == "not_checked"
    ].index[:max_fighters]

    for number, idx in enumerate(unchecked, 1):

        fighter = fighter_topics.at[idx, "fighter"]

        try:
            topic = find_fighter_topic(fighter)

            if topic is not None:
                fighter_topics.at[idx, "topic_id"] = topic
                fighter_topics.at[idx, "status"] = "found"
                print(f"{number}. ✅ {fighter}: {topic}")

            else:
                fighter_topics.at[idx, "status"] = "missing"
                print(f"{number}. ⚠️ {fighter}: no topic")

        except Exception as e:
            fighter_topics.at[idx, "status"] = "error"
            print(f"{number}. ❌ {fighter}: {e}")

        # Save every 10 fighters
        if number % 10 == 0:
            save_topic_progress()

        time.sleep(0.5)

    save_topic_progress()
```

```python
def normalize_to_conor(trends_df, fighter_ids, anchor_id="/m/0rfgxy2"):
    """
    Convert Google Trends values from a shared batch
    into a Conor McGregor = 100 scale.
    """

    result = trends_df.copy()

    if anchor_id not in result.columns:
        raise ValueError("Conor anchor was not returned.")

    anchor = result[anchor_id].astype(float)

    for fighter_id in fighter_ids:
        if fighter_id not in result.columns:
            continue

        fighter = result[fighter_id].astype(float)

        normalized = pd.Series(
            np.nan,
            index=result.index,
            dtype=float
        )

        valid = anchor > 0
        normalized.loc[valid] = (
            fighter.loc[valid] /
            anchor.loc[valid]
        ) * 100

        result[f"{fighter_id}_conor100"] = normalized

    return result
```

```python
import time
import random
import os

TRENDS_PROGRESS_FILE = "/content/trends_collection_progress.csv"

def save_trends_progress():
    trends_collection_status.to_csv(
        TRENDS_PROGRESS_FILE,
        index=False
    )

    checked = (
        trends_collection_status["status"] != "not_collected"
    ).sum()

    successful = (
        trends_collection_status["status"] == "collected"
    ).sum()

    errors = (
        trends_collection_status["status"] == "error"
    ).sum()

    print(
        f"Progress saved ✅ | "
        f"Checked: {checked}/{len(trends_collection_status)} | "
        f"Collected: {successful} | "
        f"Errors: {errors}"
    )


def get_trends_with_retry(
    topic_id,
    timeframe,
    max_retries=4
):
    """
    Retrieve one Google Trends series with
    exponential backoff if Google rate-limits us.
    """

    for attempt in range(max_retries):

        try:
            pytrends.build_payload(
                kw_list=[
                    "/m/0rfgxy2",  # Conor anchor
                    topic_id
                ],
                timeframe=timeframe,
                geo="US"
            )

            data = pytrends.interest_over_time()

            if data is None or data.empty:
                return None

            return data

        except Exception as e:

            message = str(e).lower()

            if (
                "too_many_requests" in message
                or "429" in message
                or "rate" in message
            ):
                wait = (60 * (2 ** attempt)) + random.uniform(5, 20)

                print(
                    f"Rate limited. "
                    f"Waiting about {wait:.0f} seconds..."
                )

                time.sleep(wait)

            else:
                print(f"Trends error: {e}")
                return None

    print("Maximum retries reached.")
    return None


print("✅ Rate-limit-safe collector functions created.")
```

```python
# Build small Google Trends batches within each UFC event.
# 4 fighters + Conor anchor = 5 terms per request.

MAX_FIGHTERS_PER_BATCH = 4

request_batches = []

eligible_measurements = measurement_queue[
    measurement_queue["full_baseline_available"] &
    measurement_queue["topic_id"].notna()
].copy()

for event_name, group in eligible_measurements.groupby(
    ["event_date", "event_name"],
    sort=True
):

    group = group.drop_duplicates(
        subset=["fighter", "topic_id"]
    ).copy()

    fighters = group[
        ["fighter", "topic_id"]
    ].values.tolist()

    for start in range(
        0,
        len(fighters),
        MAX_FIGHTERS_PER_BATCH
    ):

        batch = fighters[
            start:start + MAX_FIGHTERS_PER_BATCH
        ]

        request_batches.append({
            "event_date": event_name[0],
            "event_name": event_name[1],
            "fighters": batch
        })

request_batches_df = pd.DataFrame(request_batches)

print("Eligible fighter-event measurements:",
      len(eligible_measurements))

print("Unique events:",
      eligible_measurements[
          ["event_date", "event_name"]
      ].drop_duplicates().shape[0])

print("Google Trends requests needed:",
      len(request_batches_df))

print()
print("First 10 request batches:")
print(
    request_batches_df.head(10).to_string(index=False)
)

print()
print("Average fighters per request:",
      len(eligible_measurements) / len(request_batches_df))
```

```python
import os
import time
import random
import pandas as pd
import numpy as np

TRENDS_RESULTS_FILE = "/content/ufc_trends_results.csv"

# Load existing results if we ever resume
if os.path.exists(TRENDS_RESULTS_FILE):
    trends_results = pd.read_csv(TRENDS_RESULTS_FILE)
    trends_results["event_date"] = pd.to_datetime(
        trends_results["event_date"]
    )
    print("Existing progress loaded:", len(trends_results))
else:
    trends_results = pd.DataFrame(columns=[
        "event_date",
        "event_name",
        "fighter",
        "topic_id",
        "baseline_90d",
        "fight_week_interest",
        "hype_multiplier"
    ])
    print("Starting fresh Trends collection.")


def collect_trends_batches(start_batch=0, num_batches=10):

    global trends_results

    end_batch = min(
        start_batch + num_batches,
        len(request_batches_df)
    )

    for batch_num in range(start_batch, end_batch):

        batch = request_batches_df.iloc[batch_num]

        event_date = pd.Timestamp(batch["event_date"])
        event_name = batch["event_name"]
        fighters = batch["fighters"]

        start_date = event_date - pd.Timedelta(days=97)

        timeframe = (
            f"{start_date.strftime('%Y-%m-%d')} "
            f"{event_date.strftime('%Y-%m-%d')}"
        )

        topic_ids = [
            TRENDS_CONFIG["anchor_topic_id"]
        ] + [topic_id for _, topic_id in fighters]

        try:
            pytrends.build_payload(
                kw_list=topic_ids,
                timeframe=timeframe,
                geo=TRENDS_CONFIG["geo"]
            )

            data = pytrends.interest_over_time()

            if data is None or data.empty:
                print(
                    f"⚠️ Batch {batch_num}: empty data"
                )
                continue

            if "isPartial" in data.columns:
                data = data[
                    data["isPartial"] == False
                ].copy()

            baseline_start = event_date - pd.Timedelta(days=97)
            baseline_end = event_date - pd.Timedelta(days=8)

            fight_start = event_date - pd.Timedelta(days=6)
            fight_end = event_date

            baseline_data = data.loc[
                baseline_start:baseline_end
            ]

            fight_data = data.loc[
                fight_start:fight_end
            ]

            anchor = TRENDS_CONFIG["anchor_topic_id"]

            for fighter, topic_id in fighters:

                if topic_id not in data.columns:
                    continue

                baseline_anchor_mean = baseline_data[
                    anchor
                ].mean()

                fight_anchor_mean = fight_data[
                    anchor
                ].mean()

                baseline = (
                    baseline_data[topic_id].mean()
                    / baseline_anchor_mean * 100
                    if baseline_anchor_mean > 0
                    else np.nan
                )

                fight_week = (
                    fight_data[topic_id].mean()
                    / fight_anchor_mean * 100
                    if fight_anchor_mean > 0
                    else np.nan
                )

                hype = (
                    fight_week / baseline
                    if baseline > 0
                    else np.nan
                )

                new_row = {
                    "event_date": event_date,
                    "event_name": event_name,
                    "fighter": fighter,
                    "topic_id": topic_id,
                    "baseline_90d": baseline,
                    "fight_week_interest": fight_week,
                    "hype_multiplier": hype
                }

                trends_results = pd.concat(
                    [
                        trends_results,
                        pd.DataFrame([new_row])
                    ],
                    ignore_index=True
                )

            trends_results.to_csv(
                TRENDS_RESULTS_FILE,
                index=False
            )

            print(
                f"✅ Batch {batch_num + 1}/{len(request_batches_df)} "
                f"| {event_name}"
            )

            # Gentle pacing
            time.sleep(random.uniform(3, 6))

        except Exception as e:

            print(
                f"⛔ Batch {batch_num + 1} stopped: {e}"
            )

            trends_results.to_csv(
                TRENDS_RESULTS_FILE,
                index=False
            )

            print("Progress saved ✅")
            break


print("✅ Production Trends collector ready.")
```

```python
# Build the FINAL Google Trends request queue:
# 2014+ only, max 4 fighters + Conor anchor per request

MAX_FIGHTERS_PER_BATCH = 4

eligible_measurements = measurement_queue[
    measurement_queue["trends_usable"]
].copy()

request_batches = []

for event_key, group in eligible_measurements.groupby(
    ["event_date", "event_name"],
    sort=True
):

    group = group.drop_duplicates(
        subset=["fighter", "topic_id"]
    ).copy()

    fighters = group[
        ["fighter", "topic_id"]
    ].values.tolist()

    for start in range(
        0,
        len(fighters),
        MAX_FIGHTERS_PER_BATCH
    ):

        batch = fighters[
            start:start + MAX_FIGHTERS_PER_BATCH
        ]

        request_batches.append({
            "event_date": event_key[0],
            "event_name": event_key[1],
            "fighters": batch
        })

request_batches_df = pd.DataFrame(request_batches)

print(
    "Eligible fighter-event measurements:",
    len(eligible_measurements)
)

print(
    "Unique events:",
    eligible_measurements[
        ["event_date", "event_name"]
    ].drop_duplicates().shape[0]
)

print(
    "Google Trends requests needed:",
    len(request_batches_df)
)

print()

print("First request:")
print(
    request_batches_df.iloc[0]
)
```

```python
def collect_trends_batches(start_batch=0, num_batches=10):

    global trends_results

    end_batch = min(
        start_batch + num_batches,
        len(request_batches_df)
    )

    for batch_num in range(start_batch, end_batch):

        batch = request_batches_df.iloc[batch_num]

        event_date = pd.Timestamp(batch["event_date"])
        event_name = batch["event_name"]
        fighters = batch["fighters"]

        # 90-day baseline + 7 pre-event days
        request_start = event_date - pd.Timedelta(days=97)
        request_end = event_date - pd.Timedelta(days=1)

        timeframe = (
            f"{request_start.strftime('%Y-%m-%d')} "
            f"{request_end.strftime('%Y-%m-%d')}"
        )

        # Skip fighters already successfully collected
        remaining_fighters = []

        for fighter, topic_id in fighters:

            already_done = (
                (trends_results["event_name"] == event_name) &
                (trends_results["fighter"] == fighter)
            ).any()

            if not already_done:
                remaining_fighters.append(
                    [fighter, topic_id]
                )

        if len(remaining_fighters) == 0:
            print(
                f"⏭️ Batch {batch_num + 1} already collected"
            )
            continue

        topic_ids = [
            TRENDS_CONFIG["anchor_topic_id"]
        ] + [
            topic_id
            for _, topic_id in remaining_fighters
        ]

        try:

            pytrends.build_payload(
                kw_list=topic_ids,
                timeframe=timeframe,
                geo=TRENDS_CONFIG["geo"]
            )

            data = pytrends.interest_over_time()

            if data is None or data.empty:
                print(
                    f"⚠️ Batch {batch_num + 1}: empty data"
                )
                continue

            if "isPartial" in data.columns:
                data = data[
                    data["isPartial"] == False
                ].copy()

            # 90 days BEFORE fight week
            baseline_start = (
                event_date - pd.Timedelta(days=97)
            )

            baseline_end = (
                event_date - pd.Timedelta(days=8)
            )

            # Exactly 7 days BEFORE event day
            fight_start = (
                event_date - pd.Timedelta(days=7)
            )

            fight_end = (
                event_date - pd.Timedelta(days=1)
            )

            baseline_data = data.loc[
                baseline_start:baseline_end
            ]

            fight_data = data.loc[
                fight_start:fight_end
            ]

            anchor = TRENDS_CONFIG["anchor_topic_id"]

            baseline_anchor_mean = (
                baseline_data[anchor].mean()
            )

            fight_anchor_mean = (
                fight_data[anchor].mean()
            )

            for fighter, topic_id in remaining_fighters:

                if topic_id not in data.columns:
                    continue

                baseline = (
                    baseline_data[topic_id].mean()
                    / baseline_anchor_mean * 100
                    if baseline_anchor_mean > 0
                    else np.nan
                )

                fight_week = (
                    fight_data[topic_id].mean()
                    / fight_anchor_mean * 100
                    if fight_anchor_mean > 0
                    else np.nan
                )

                hype = (
                    fight_week / baseline
                    if baseline > 0
                    else np.nan
                )

                new_row = {
                    "event_date": event_date,
                    "event_name": event_name,
                    "fighter": fighter,
                    "topic_id": topic_id,
                    "baseline_90d": baseline,
                    "fight_week_interest": fight_week,
                    "hype_multiplier": hype
                }

                trends_results = pd.concat(
                    [
                        trends_results,
                        pd.DataFrame([new_row])
                    ],
                    ignore_index=True
                )

            # Save after EVERY successful batch
            trends_results.to_csv(
                TRENDS_RESULTS_FILE,
                index=False
            )

            print(
                f"✅ Batch {batch_num + 1}/"
                f"{len(request_batches_df)} "
                f"| {event_name} "
                f"| Saved rows: {len(trends_results)}"
            )

            time.sleep(
                random.uniform(3, 6)
            )

        except Exception as e:

            print(
                f"⛔ Batch {batch_num + 1} stopped: {e}"
            )

            trends_results.to_csv(
                TRENDS_RESULTS_FILE,
                index=False
            )

            print("Progress saved ✅")
            break


print("✅ Final 2014+ collector ready")
print("Fight week = 7 days BEFORE event")
print("Production batches =", len(request_batches_df))
```

```python
import re

def make_event_search_term(event_name):

    # Numbered UFC PPVs:
    # "UFC 300: Pereira vs Hill" -> "UFC 300"
    numbered = re.match(r"^UFC\s+(\d+)", event_name)

    if numbered:
        return f"UFC {numbered.group(1)}"

    # Fight Nights / UFC on FOX / other named UFC events:
    # Keep matchup names but remove punctuation and "vs"
    if ":" in event_name:
        matchup = event_name.split(":", 1)[1]

        matchup = re.sub(
            r"\bvs\.?\b",
            " ",
            matchup,
            flags=re.IGNORECASE
        )

        matchup = re.sub(
            r"[^A-Za-z0-9\s]",
            " ",
            matchup
        )

        matchup = re.sub(
            r"\s+",
            " ",
            matchup
        ).strip()

        return f"UFC {matchup}"

    # Fallback
    return event_name


event_fanfare_queue["search_term"] = (
    event_fanfare_queue["event_name"]
    .apply(make_event_search_term)
)

print(
    event_fanfare_queue[
        [
            "event_date",
            "event_name",
            "search_term"
        ]
    ].head(20).to_string(index=False)
)
```

```python
import time
import pandas as pd
import numpy as np

CHECKPOINT_FILE = "/content/event_trends_results.csv"

def collect_event_trends(max_events=25, pause_seconds=4):

    global event_trends_results

    # Reload checkpoint from disk
    event_trends_results = pd.read_csv(CHECKPOINT_FILE)

    event_trends_results["event_date"] = pd.to_datetime(
        event_trends_results["event_date"]
    )

    completed_events = set(
        event_trends_results["event_name"]
    )

    remaining = collection_ready[
        ~collection_ready["event_name"].isin(completed_events)
    ].copy()

    print("Already saved:", len(event_trends_results))
    print("Remaining:", len(remaining))
    print("Collecting up to:", min(max_events, len(remaining)))
    print()

    collected_now = 0

    for _, event in remaining.head(max_events).iterrows():

        event_date = event["event_date"]
        topic1 = event["headliner_1_topic"]
        topic2 = event["headliner_2_topic"]

        try:
            start_date = event_date - pd.Timedelta(days=97)
            end_date = event_date - pd.Timedelta(days=1)

            pytrends.build_payload(
                kw_list=[topic1, topic2],
                timeframe=f"{start_date:%Y-%m-%d} {end_date:%Y-%m-%d}",
                geo="US"
            )

            data = pytrends.interest_over_time()

            if data.empty:
                print("EMPTY:", event["event_name"])
                time.sleep(pause_seconds)
                continue

            baseline = data.loc[
                event_date - pd.Timedelta(days=97):
                event_date - pd.Timedelta(days=8)
            ]

            fight_week = data.loc[
                event_date - pd.Timedelta(days=7):
                event_date - pd.Timedelta(days=1)
            ]

            h1_base = baseline[topic1].mean()
            h1_week = fight_week[topic1].mean()

            h2_base = baseline[topic2].mean()
            h2_week = fight_week[topic2].mean()

            h1_smoothed = (h1_week + 1) / (h1_base + 1)
            h2_smoothed = (h2_week + 1) / (h2_base + 1)

            h1_log = np.log1p(h1_smoothed)
            h2_log = np.log1p(h2_smoothed)

            new_row = pd.DataFrame([{
                "event_date": event_date,
                "event_name": event["event_name"],
                "headliner_1": event["headliner_1"],
                "headliner_2": event["headliner_2"],
                "headliner_1_topic": topic1,
                "headliner_2_topic": topic2,

                "h1_baseline": h1_base,
                "h1_fight_week": h1_week,
                "h1_smoothed_hype": h1_smoothed,
                "h1_log_hype": h1_log,

                "h2_baseline": h2_base,
                "h2_fight_week": h2_week,
                "h2_smoothed_hype": h2_smoothed,
                "h2_log_hype": h2_log,

                "event_avg_baseline": (h1_base + h2_base) / 2,
                "event_avg_fight_week": (h1_week + h2_week) / 2,
                "event_avg_log_hype": (h1_log + h2_log) / 2
            }])

            event_trends_results = pd.concat(
                [event_trends_results, new_row],
                ignore_index=True
            )

            # Save immediately
            event_trends_results.to_csv(
                CHECKPOINT_FILE,
                index=False
            )

            collected_now += 1

            print(
                f"Saved {collected_now}/{max_events}:",
                event["event_name"],
                "| Total:",
                len(event_trends_results)
            )

            time.sleep(pause_seconds)

        except Exception as e:

            print(
                "STOPPED ON ERROR:",
                event["event_name"],
                "|",
                str(e)
            )

            # Stop instead of hammering Google
            break

    print()
    print("Collection run finished ✅")
    print("Collected this run:", collected_now)
    print("Total saved:", len(event_trends_results))
    print(
        "Remaining:",
        len(collection_ready) - len(event_trends_results)
    )
```

```python
def rank_points(rank):
    if pd.isna(rank):
        return 0
    if rank == 0:
        return 16
    if 1 <= rank <= 15:
        return 16 - rank
    return 0


ranking_strength_stats = []

for event_name, group in ufc_fights.groupby("event_name"):

    f1_points = group["f1_rank_before"].apply(rank_points)
    f2_points = group["f2_rank_before"].apply(rank_points)

    all_points = pd.concat([f1_points, f2_points])

    ranking_strength_stats.append({
        "event_name": event_name,
        "total_ranking_points": all_points.sum(),
        "avg_ranking_points": all_points.mean()
    })

ranking_strength_stats = pd.DataFrame(ranking_strength_stats)

event_master = event_master.merge(
    ranking_strength_stats,
    on="event_name",
    how="left"
)

print("Ranking strength added ✅")
print()

print(
    event_master[
        [
            "event_name",
            "ranked_fighters",
            "ranked_matchups",
            "total_ranking_points",
            "avg_ranking_points"
        ]
    ]
    .sort_values("total_ranking_points", ascending=False)
    .head(15)
    .to_string(index=False)
)
```

```python
event_master["main_event_f1_rank_points"] = (
    event_master["main_event_f1_rank"].apply(rank_points)
)

event_master["main_event_f2_rank_points"] = (
    event_master["main_event_f2_rank"].apply(rank_points)
)

event_master["main_event_ranking_points"] = (
    event_master["main_event_f1_rank_points"]
    + event_master["main_event_f2_rank_points"]
)

print("Main-event ranking strength added ✅")
print()

print(
    event_master[
        [
            "event_name",
            "main_event_f1_rank",
            "main_event_f2_rank",
            "main_event_ranking_points"
        ]
    ]
    .sort_values(
        "main_event_ranking_points",
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)
```

```python
zero_check = []

for col in candidate_cols:

    non_missing = event_master[col].notna().sum()
    zeros = (event_master[col] == 0).sum()

    zero_check.append({
        "variable": col,
        "available_values": non_missing,
        "zero_values": zeros,
        "zero_percent": round(
            (zeros / non_missing) * 100, 1
        ) if non_missing > 0 else 0
    })

zero_check = pd.DataFrame(zero_check)

print("ZERO-VALUE CHECK")
print("=" * 50)

print(
    zero_check
    .sort_values("zero_percent", ascending=False)
    .to_string(index=False)
)
```

```python
def zero_preserving_percentile(series):

    result = pd.Series(index=series.index, dtype=float)

    # Missing stays missing
    result[series.isna()] = float("nan")

    # True zero gets zero points
    result[series == 0] = 0

    # Rank only positive values
    positive = series > 0

    if positive.sum() > 0:
        result.loc[positive] = (
            series.loc[positive]
            .rank(
                pct=True,
                method="average"
            ) * 100
        )

    return result


normalized = pd.DataFrame({
    "event_name": event_master["event_name"],
    "event_date": event_master["event_date"]
})

for col in candidate_cols:
    normalized[col + "_score"] = (
        zero_preserving_percentile(
            event_master[col]
        )
    )

print("Zero-preserving normalization created ✅")
print("Events:", len(normalized))
print("Normalized variables:", len(candidate_cols))

print()
print(
    normalized[
        [
            "event_name",
            "title_fights_score",
            "total_ranking_points_score",
            "avg_previous_title_fights_score",
            "main_event_ranking_points_score",
            "event_avg_log_hype_score"
        ]
    ]
    .tail(10)
    .round(1)
    .to_string(index=False)
)
```

```python
weight_tests = {
    "balanced": {
        "card_quality_score": 0.35,
        "star_experience_score": 0.25,
        "main_event_score": 0.30,
        "fanfare_score": 0.10
    },

    "card_focused": {
        "card_quality_score": 0.45,
        "star_experience_score": 0.25,
        "main_event_score": 0.25,
        "fanfare_score": 0.05
    },

    "main_event_focused": {
        "card_quality_score": 0.35,
        "star_experience_score": 0.20,
        "main_event_score": 0.40,
        "fanfare_score": 0.05
    }
}


def weighted_score_with_missing(row, weights):

    available = {
        col: weight
        for col, weight in weights.items()
        if pd.notna(row[col])
    }

    if not available:
        return float("nan")

    total_weight = sum(available.values())

    return sum(
        row[col] * weight
        for col, weight in available.items()
    ) / total_weight


for model_name, weights in weight_tests.items():

    normalized[model_name] = normalized.apply(
        lambda row: weighted_score_with_missing(
            row,
            weights
        ),
        axis=1
    )


print("WEIGHTING MODEL COMPARISON")
print("=" * 65)
print()

for model_name in weight_tests:

    print()
    print(model_name.upper())
    print("-" * 40)

    print(
        normalized[
            ["event_name", model_name]
        ]
        .sort_values(
            model_name,
            ascending=False
        )
        .head(15)
        .round(2)
        .to_string(index=False)
    )
```

```python
normalized["balanced_rank"] = (
    normalized["balanced"]
    .rank(ascending=False, method="min")
)

normalized["card_focused_rank"] = (
    normalized["card_focused"]
    .rank(ascending=False, method="min")
)

normalized["main_event_focused_rank"] = (
    normalized["main_event_focused"]
    .rank(ascending=False, method="min")
)

rank_correlations = normalized[
    [
        "balanced_rank",
        "card_focused_rank",
        "main_event_focused_rank"
    ]
].corr(method="spearman")

print("FULL RANKING STABILITY")
print("=" * 55)
print()
print(rank_correlations.round(3).to_string())

print()
print("Average rank movement vs Balanced:")

print(
    "Card Focused:",
    round(
        (
            normalized["balanced_rank"]
            - normalized["card_focused_rank"]
        ).abs().mean(),
        2
    )
)

print(
    "Main Event Focused:",
    round(
        (
            normalized["balanced_rank"]
            - normalized["main_event_focused_rank"]
        ).abs().mean(),
        2
    )
)
```

```python
def classify_event(name):
    if name.startswith("UFC Fight Night"):
        return "Fight Night"
    elif name.startswith("The Ultimate Fighter"):
        return "TUF Finale"
    elif name.startswith("UFC on"):
        return "UFC on TV"
    elif name.startswith("UFC ") and any(char.isdigit() for char in name):
        return "Numbered UFC"
    else:
        return "Other / Early UFC"


normalized["event_type"] = (
    normalized["event_name"]
    .apply(classify_event)
)

event_type_summary = (
    normalized
    .groupby("event_type")
    .agg(
        events=("event_strength_index", "count"),
        average=("event_strength_index", "mean"),
        median=("event_strength_index", "median"),
        maximum=("event_strength_index", "max"),
        minimum=("event_strength_index", "min")
    )
    .sort_values("average", ascending=False)
    .round(2)
)

print("EVENT TYPE VALIDATION")
print("=" * 65)
print()
print(event_type_summary.to_string())

print()
print("TOP 10 FIGHT NIGHTS")
print("-" * 65)

print(
    normalized[
        normalized["event_type"] == "Fight Night"
    ][
        [
            "event_name",
            "event_date",
            "event_strength_index",
            "card_quality_score",
            "star_experience_score",
            "main_event_score",
            "fanfare_score"
        ]
    ]
    .sort_values(
        "event_strength_index",
        ascending=False
    )
    .head(10)
    .round(2)
    .to_string(index=False)
)
```

```python
readme_start = """# UFC Event Strength Index

## Which UFC events were the strongest on paper?

The UFC Event Strength Index is a data analytics project that measures
the strength of UFC events using historical fighter performance,
rankings, experience, accomplishments, main-event strength, and
pre-event search interest.

The project analyzes 8,736 UFC fights across 777 events from
1994 through 2026.

Each event receives a score from 0 to 100 using four components:

- Card Quality — 40%
- Main Event Strength — 30%
- Star Experience — 25%
- Fanfare — 5%

All historical fighter statistics are calculated using only information
available before each fight, preventing future results from influencing
earlier UFC events.

## Key Result

UFC 214: Cormier vs. Jones 2 ranks #1 in the dataset with an
Event Strength Index of 90.50.
"""

readme_path = (
    "/content/UFC_Event_Strength_Portfolio/README.md"
)

with open(readme_path, "w") as file:
    file.write(readme_start)

print("README CREATED ✅🔥")
print("=" * 55)
print()
print(readme_start)

print("Saved to:")
print(readme_path)
```

```python
methodology_section = """

## Methodology

The Event Strength Index is built from four components.

### 1. Card Quality — 40%

Measures the overall strength and depth of the full event.

Inputs include:

- Number of title fights
- Historical UFC rankings
- UFC experience
- UFC win percentage
- Current UFC win streak
- Historical finishing rate

### 2. Main Event Strength — 30%

Measures the strength of the event's headlining matchup.

Inputs include:

- Combined UFC experience
- UFC win percentage
- Combined win streak
- Previous title fights
- Previous UFC main events
- Previous UFC bonuses
- Historical rankings

### 3. Star Experience — 25%

Measures the accomplishments and proven UFC experience across the card.

Inputs include:

- Previous UFC title fights
- Previous UFC main events
- Previous UFC performance and fight bonuses

### 4. Fanfare — 5%

Measures pre-event search-interest acceleration using Google Trends.

Search interest during the seven days before an event is compared with
the fighters' prior baseline interest. The metric is log-transformed to
reduce the effect of extreme spikes.

Because Google Trends data is not consistently available throughout UFC
history, missing fanfare data is not treated as a score of zero.

## Preventing Future Data Leakage

Every historical fighter statistic is calculated as it existed entering
the fight.

For example, a fighter's UFC record, win streak, title-fight experience,
main-event experience, and bonus history include only UFC fights that
occurred before the event being scored.

This prevents accomplishments earned later in a fighter's career from
artificially increasing the strength of earlier events.

## Scoring

Variables are converted to zero-preserving percentile scores.

A true value of zero remains zero, while positive values are ranked
relative to the other positive observations. Metrics that did not exist
during an earlier era, such as official UFC rankings, are treated as
unavailable rather than zero.

The final Event Strength Index uses:

- 40% Card Quality
- 30% Main Event Strength
- 25% Star Experience
- 5% Fanfare

When a component is historically unavailable, its weight is redistributed
across the available components rather than treating missing data as zero.
"""

with open(readme_path, "a") as file:
    file.write(methodology_section)

print("METHODOLOGY ADDED TO README ✅🔥")
print("=" * 60)

with open(readme_path, "r") as file:
    print(file.read())
```

```python
findings_section = """

## Key Findings

- UFC 214: Cormier vs. Jones 2 ranks as the strongest event in the
  dataset with an Event Strength Index of 90.50.

- Among the 50 highest-rated events, Star Experience is the strongest
  component for 74% of events, while Card Quality is strongest for 24%
  and Main Event Strength for 2%.

- Numbered UFC events average a 56.74 Event Strength score compared
  with 38.16 for UFC Fight Nights. Event type itself is not included
  as an input to the model.

- Strong Fight Night cards can still score highly. The highest-rated
  Fight Night in the dataset reaches 68.70.

- Event year has only a modest relationship with Event Strength:
  Pearson correlation = 0.313 and Spearman correlation = 0.283.

- The model is highly stable across alternative component-weighting
  systems, with rank correlations of approximately 0.99 between the
  tested models.

## Limitations

- Official UFC rankings are only available beginning in 2013, so
  ranking information is treated as unavailable for earlier events.

- UFC bonus history begins later than the earliest UFC events.
  Pre-bonus-era events are not penalized for the absence of bonuses.

- The Google Trends component is available for 505 of the 777 events
  and is used as a proxy for pre-event fan interest rather than a
  direct measurement of social-media engagement.

- Google Trends data is based on U.S. search interest, which may
  underrepresent internationally popular fighters and events.

- Google Trends values are relative search-interest measurements,
  not absolute audience-size measurements.

- The component weights are modeling choices rather than objectively
  correct values. Alternative weighting systems were tested to measure
  ranking sensitivity.

- The index measures the strength of an event based primarily on the
  fighters and information available around the event. It is not a
  measurement of how entertaining the fights ultimately were.
"""

with open(readme_path, "a") as file:
    file.write(findings_section)

print("FINDINGS + LIMITATIONS ADDED ✅🔥")
print("=" * 65)
print(findings_section)
```

```python
import pandas as pd

test_fights = []

for fight in fight_blocks:

    # Fighter names
    red = fight.select_one(
        ".c-listing-fight__corner-name--red"
    )
    blue = fight.select_one(
        ".c-listing-fight__corner-name--blue"
    )

    red_name = " ".join(red.stripped_strings) if red else None
    blue_name = " ".join(blue.stripped_strings) if blue else None

    # Results
    red_win = fight.select_one(
        ".c-listing-fight__corner--red .c-listing-fight__outcome--win"
    )
    blue_win = fight.select_one(
        ".c-listing-fight__corner--blue .c-listing-fight__outcome--win"
    )

    if red_win:
        red_result, blue_result = "W", "L"
    elif blue_win:
        red_result, blue_result = "L", "W"
    else:
        red_result, blue_result = None, None

    # Weight class
    wc = fight.select_one(
        ".c-listing-fight__class--desktop .c-listing-fight__class-text"
    )

    # Fight result details
    round_el = fight.select_one(".round")
    time_el = fight.select_one(".time")
    method_el = fight.select_one(".method")

    test_fights.append({
        "f1_name": red_name,
        "f1_result": red_result,
        "f2_name": blue_name,
        "f2_result": blue_result,
        "weight_class": wc.get_text(strip=True) if wc else None,
        "method": method_el.get_text(strip=True) if method_el else None,
        "round": round_el.get_text(strip=True) if round_el else None,
        "time": time_el.get_text(strip=True) if time_el else None
    })

test_event_df = pd.DataFrame(test_fights)

print("FIGHTS EXTRACTED:", len(test_event_df))
print("=" * 80)

print(test_event_df.to_string(index=False))
```

```python
import requests
from bs4 import BeautifulSoup

archive_url = "https://www.ufc.com/events"

archive_response = requests.get(
    archive_url,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=30
)

archive_soup = BeautifulSoup(
    archive_response.text,
    "html.parser"
)

print("UFC EVENT ARCHIVE TEST")
print("=" * 70)

print("Status code:", archive_response.status_code)
print("Characters received:", len(archive_response.text))

event_links = []

for a in archive_soup.select('a[href*="/event/"]'):
    href = a.get("href")
    text = " ".join(a.stripped_strings)

    if href:
        event_links.append((text, href))

print("Event links found:", len(event_links))

print()
print("FIRST 20 EVENT LINKS:")
print("=" * 70)

for text, href in event_links[:20]:
    print(text, "->", href)
```

```python
from urllib.parse import urljoin

unique_event_urls = []

for text, href in event_links:

    if "/event/" not in href:
        continue

    full_url = urljoin(
        "https://www.ufc.com",
        href.split("#")[0]
    )

    if full_url not in unique_event_urls:
        unique_event_urls.append(full_url)

print("UNIQUE EVENT PAGES FOUND:", len(unique_event_urls))
print("=" * 80)

for i, url in enumerate(unique_event_urls, start=1):
    print(i, url)
```

```python
import re
import time
import pandas as pd
import requests
from bs4 import BeautifulSoup

event_checks = []

for i, url in enumerate(update_event_urls, 1):

    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=30
    )

    soup_event = BeautifulSoup(response.text, "html.parser")

    # Event type / prefix
    prefix = soup_event.select_one(".c-hero__headline-prefix")
    prefix_text = " ".join(prefix.stripped_strings) if prefix else None

    # Headline
    headline = soup_event.select_one(".c-hero__headline")
    headline_text = " ".join(headline.stripped_strings) if headline else None

    # Date text
    suffix = soup_event.select_one(".c-hero__headline-suffix")
    suffix_text = " ".join(suffix.stripped_strings) if suffix else None

    # Completed fight blocks
    fights = soup_event.select(".c-listing-fight")

    completed = 0

    for fight in fights:
        has_result = (
            fight.select_one(".c-listing-fight__outcome--win") is not None
            or fight.select_one(".c-listing-fight__outcome--loss") is not None
        )

        if has_result:
            completed += 1

    event_checks.append({
        "url": url,
        "prefix": prefix_text,
        "headline": headline_text,
        "date_text": suffix_text,
        "fight_blocks": len(fights),
        "completed_fights": completed,
        "status": response.status_code
    })

    print(
        f"{i}/{len(update_event_urls)} checked — "
        f"{prefix_text}: {headline_text} — "
        f"{completed} completed fights"
    )

    time.sleep(0.5)

event_validation = pd.DataFrame(event_checks)

print()
print("EVENT VALIDATION COMPLETE ✅🔥")
print("=" * 110)

print(
    event_validation[
        [
            "prefix",
            "headline",
            "date_text",
            "fight_blocks",
            "completed_fights",
            "status"
        ]
    ].to_string(index=False)
)

print()
print("TOTAL COMPLETED FIGHTS:",
      event_validation["completed_fights"].sum())
```

```python
import requests
import pandas as pd
import numpy as np
import time
from bs4 import BeautifulSoup
from datetime import datetime

all_new_fights = []

for event_num, url in enumerate(completed_update_urls, 1):

    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=30
    )

    event_soup = BeautifulSoup(response.text, "html.parser")

    # Event name
    prefix = event_soup.select_one(".c-hero__headline-prefix")
    headline = event_soup.select_one(".c-hero__headline")

    prefix_text = " ".join(prefix.stripped_strings) if prefix else ""
    headline_text = " ".join(headline.stripped_strings) if headline else ""

    event_name = f"{prefix_text}: {headline_text}"

    # Event date
    suffix = event_soup.select_one(".c-hero__headline-suffix")
    date_text = " ".join(suffix.stripped_strings) if suffix else ""

    date_part = date_text.split("/")[0].strip()

    event_date = pd.to_datetime(
        f"{date_part} 2026",
        format="%a, %b %d %Y"
    )

    fight_blocks = event_soup.select(".c-listing-fight")

    event_count = 0

    for fight in fight_blocks:

        red = fight.select_one(".c-listing-fight__corner-name--red")
        blue = fight.select_one(".c-listing-fight__corner-name--blue")

        if red is None or blue is None:
            continue

        red_name = " ".join(red.stripped_strings)
        blue_name = " ".join(blue.stripped_strings)

        red_win = fight.select_one(
            ".c-listing-fight__corner--red .c-listing-fight__outcome--win"
        )

        blue_win = fight.select_one(
            ".c-listing-fight__corner--blue .c-listing-fight__outcome--win"
        )

        if red_win:
            red_result, blue_result = "W", "L"
        elif blue_win:
            red_result, blue_result = "L", "W"
        else:
            continue

        wc = fight.select_one(
            ".c-listing-fight__class--desktop .c-listing-fight__class-text"
        )

        round_el = fight.select_one(".round")
        time_el = fight.select_one(".time")
        method_el = fight.select_one(".method")

        all_new_fights.append({
            "event_name": event_name,
            "date": event_date,
            "location": np.nan,
            "f1_name": red_name,
            "f1_result": red_result,
            "f2_name": blue_name,
            "f2_result": blue_result,
            "weight_class": wc.get_text(strip=True) if wc else None,
            "method": method_el.get_text(strip=True) if method_el else None,
            "round": round_el.get_text(strip=True) if round_el else None,
            "time": time_el.get_text(strip=True) if time_el else None,
            "time_format": np.nan,
            "referee": np.nan,
            "judging_details": np.nan,
            "ufc_fmid": fight.get("data-fmid"),
            "source_url": url
        })

        event_count += 1

    print(
        f"{event_num}/{len(completed_update_urls)} — "
        f"{event_name} — {event_count} fights"
    )

    time.sleep(0.5)

new_fights = pd.DataFrame(all_new_fights)

print()
print("NEW FIGHT DATASET CREATED ✅🔥")
print("=" * 80)
print("Rows:", len(new_fights))
print("Events:", new_fights["event_name"].nunique())
print("Unique UFC fight IDs:", new_fights["ufc_fmid"].nunique())
print("Earliest date:", new_fights["date"].min())
print("Latest date:", new_fights["date"].max())
```

```python
# Create matchup keys that ignore which fighter is listed first
def matchup_key(name1, name2, date):
    fighters = sorted([
        str(name1).strip().lower(),
        str(name2).strip().lower()
    ])

    return (
        str(pd.Timestamp(date).date()),
        fighters[0],
        fighters[1]
    )

existing_keys = set(
    ufc_fights.apply(
        lambda row: matchup_key(
            row["f1_name"],
            row["f2_name"],
            row["date"]
        ),
        axis=1
    )
)

new_fights["matchup_date_key"] = new_fights.apply(
    lambda row: matchup_key(
        row["f1_name"],
        row["f2_name"],
        row["date"]
    ),
    axis=1
)

new_fights["already_in_master"] = (
    new_fights["matchup_date_key"].isin(existing_keys)
)

print("MASTER DATASET CROSS-CHECK")
print("=" * 80)

print("Existing master fights:", len(ufc_fights))
print("New fights checked:", len(new_fights))
print(
    "Already in master:",
    new_fights["already_in_master"].sum()
)
print(
    "Genuinely new:",
    (~new_fights["already_in_master"]).sum()
)

if new_fights["already_in_master"].any():
    print()
    print("MATCHES ALREADY PRESENT:")
    print(
        new_fights.loc[
            new_fights["already_in_master"],
            ["date", "event_name", "f1_name", "f2_name"]
        ].to_string(index=False)
    )
else:
    print()
    print("NO OVERLAP FOUND ✅🔥")
```

```python
# All fighters who already exist in the historical dataset
historical_fighters = set(
    pd.concat([
        ufc_fights["f1_name"],
        ufc_fights["f2_name"]
    ])
    .dropna()
    .astype(str)
    .str.strip()
)

# All fighters appearing in the 141 new fights
new_fighter_names = sorted(
    set(new_fights["f1_name"]) |
    set(new_fights["f2_name"])
)

already_known = [
    fighter for fighter in new_fighter_names
    if fighter in historical_fighters
]

not_found = [
    fighter for fighter in new_fighter_names
    if fighter not in historical_fighters
]

print("FIGHTER NAME COMPATIBILITY CHECK")
print("=" * 80)

print("Unique fighters in new fights:", len(new_fighter_names))
print("Exact historical matches:", len(already_known))
print("Not found historically:", len(not_found))

print()
print("FIGHTERS NOT FOUND IN HISTORICAL DATA:")
print("=" * 80)

for fighter in not_found:
    print(fighter)
```

```python
import unicodedata
import re

def normalize_fighter_name(name):
    name = str(name)

    # Remove accents
    name = unicodedata.normalize("NFKD", name)
    name = "".join(
        char for char in name
        if not unicodedata.combining(char)
    )

    # Lowercase
    name = name.lower()

    # Normalize punctuation
    name = name.replace("’", "'")
    name = name.replace("‘", "'")

    # Remove punctuation/spaces
    name = re.sub(r"[^a-z0-9]", "", name)

    return name


# Map normalized historical names back to their actual names
historical_normalized = {}

for fighter in historical_fighters:
    key = normalize_fighter_name(fighter)
    historical_normalized.setdefault(key, []).append(fighter)


exact_normalized_matches = {}

for fighter in remaining_not_found:
    key = normalize_fighter_name(fighter)

    if key in historical_normalized:
        exact_normalized_matches[fighter] = historical_normalized[key]


print("NORMALIZED NAME CROSS-CHECK")
print("=" * 90)

print(
    "Additional exact matches after removing accents/punctuation:",
    len(exact_normalized_matches)
)

print()

if exact_normalized_matches:
    for new_name, old_names in exact_normalized_matches.items():
        print(f"{new_name}  →  {old_names}")
else:
    print("No additional normalized matches found.")
```

```python
print("UNMATCHED FIGHTERS — HISTORICAL SURNAME CHECK")
print("=" * 100)

for fighter in remaining_not_found:

    normalized = unicodedata.normalize("NFKD", fighter)
    normalized = "".join(
        c for c in normalized
        if not unicodedata.combining(c)
    )

    surname = normalized.strip().split()[-1].lower()

    surname_matches = sorted([
        old_name
        for old_name in historical_fighters
        if normalize_fighter_name(old_name).endswith(
            normalize_fighter_name(surname)
        )
    ])

    print()
    print("NEW:", fighter)

    if surname_matches:
        for match in surname_matches[:10]:
            print("   →", match)
    else:
        print("   → No historical surname match")
```

```python
# Convert round from text to integer
new_fights["round"] = new_fights["round"].astype(int)


# Convert UFC.com time (M:SS) to historical UFCStats format
def convert_ufc_time(time_value):
    time_value = str(time_value).strip()

    if ":" not in time_value:
        return time_value

    minutes, seconds = time_value.split(":")

    minutes = int(minutes)
    seconds = int(seconds)

    if minutes == 0:
        return f"{seconds}S"

    return f"{minutes}M {seconds}S"


# Preserve original UFC.com time
new_fights["time_original"] = new_fights["time"]

# Standardize
new_fights["time"] = new_fights["time"].apply(convert_ufc_time)


print("ROUND/TIME STANDARDIZATION COMPLETE ✅🔥")
print("=" * 80)

print("Round type:", new_fights["round"].dtype)

print()
print("ROUND VALUES:")
print(new_fights["round"].value_counts().sort_index().to_string())

print()
print("TIME CONVERSION EXAMPLES:")
print(
    new_fights[
        ["time_original", "time"]
    ]
    .drop_duplicates()
    .head(20)
    .to_string(index=False)
)
```

```python
# Working copy for rebuilt fight-level dataset
ufc_fights_updated = combined_raw.copy()

# Historical records entering each fight
fighter_records = {}

record_rows = []

for _, row in ufc_fights_updated.iterrows():

    f1 = row["f1_name"]
    f2 = row["f2_name"]

    # Record entering this fight
    f1_record = fighter_records.get(f1, {"wins": 0, "losses": 0, "fights": 0})
    f2_record = fighter_records.get(f2, {"wins": 0, "losses": 0, "fights": 0})

    record_rows.append({
        "f1_ufc_wins_before": f1_record["wins"],
        "f1_ufc_losses_before": f1_record["losses"],
        "f1_ufc_fights_before": f1_record["fights"],

        "f2_ufc_wins_before": f2_record["wins"],
        "f2_ufc_losses_before": f2_record["losses"],
        "f2_ufc_fights_before": f2_record["fights"]
    })

    # Update AFTER recording pre-fight stats
    for fighter, result in [
        (f1, row["f1_result"]),
        (f2, row["f2_result"])
    ]:

        if fighter not in fighter_records:
            fighter_records[fighter] = {
                "wins": 0,
                "losses": 0,
                "fights": 0
            }

        fighter_records[fighter]["fights"] += 1

        if result == "W":
            fighter_records[fighter]["wins"] += 1
        elif result == "L":
            fighter_records[fighter]["losses"] += 1


record_df = pd.DataFrame(record_rows)

for col in record_df.columns:
    ufc_fights_updated[col] = record_df[col].values


print("UFC RECORD / EXPERIENCE REBUILT ✅🔥")
print("=" * 80)

print("Rows:", len(ufc_fights_updated))

print()
print("NEW-EVENT SAMPLE:")
print(
    ufc_fights_updated[
        ufc_fights_updated["date"] >= "2026-07-01"
    ][
        [
            "date",
            "f1_name",
            "f1_ufc_wins_before",
            "f1_ufc_losses_before",
            "f1_ufc_fights_before",
            "f2_name",
            "f2_ufc_wins_before",
            "f2_ufc_losses_before",
            "f2_ufc_fights_before"
        ]
    ]
    .head(10)
    .to_string(index=False)
)
```

```python
# Track each fighter's active UFC win streak
fighter_win_streaks = {}

f1_streaks = []
f2_streaks = []

for _, row in ufc_fights_updated.iterrows():

    f1 = row["f1_name"]
    f2 = row["f2_name"]

    # Record streak entering the fight
    f1_streaks.append(
        fighter_win_streaks.get(f1, 0)
    )

    f2_streaks.append(
        fighter_win_streaks.get(f2, 0)
    )

    # Update streak AFTER the fight
    for fighter, result in [
        (f1, row["f1_result"]),
        (f2, row["f2_result"])
    ]:

        current_streak = fighter_win_streaks.get(fighter, 0)

        if result == "W":
            fighter_win_streaks[fighter] = current_streak + 1

        else:
            fighter_win_streaks[fighter] = 0


ufc_fights_updated["f1_win_streak_before"] = f1_streaks
ufc_fights_updated["f2_win_streak_before"] = f2_streaks


print("WIN STREAKS REBUILT ✅🔥")
print("=" * 80)

print("Rows:", len(ufc_fights_updated))

print()
print("REPEAT FIGHTER CHECK:")

for fighter in [
    "Gable Steveson",
    "Jose Miguel Delgado",
    "Levi Rodrigues Jr.",
    "Axel Sola",
    "Ryan Gandra",
    "Tommy McMillen"
]:

    fights = ufc_fights_updated[
        (ufc_fights_updated["f1_name"] == fighter) |
        (ufc_fights_updated["f2_name"] == fighter)
    ].sort_values("date").tail(2)

    print()
    print(fighter)

    for _, row in fights.iterrows():

        if row["f1_name"] == fighter:
            streak = row["f1_win_streak_before"]
            result = row["f1_result"]
        else:
            streak = row["f2_win_streak_before"]
            result = row["f2_result"]

        print(
            row["date"].date(),
            "| streak entering:", streak,
            "| result:", result
        )
```

```python
# Identify divisional UFC title fights
ufc_fights_updated["is_title_fight"] = (
    ufc_fights_updated["weight_class"]
    .astype(str)
    .str.startswith("UFC ")
    &
    ufc_fights_updated["weight_class"]
    .astype(str)
    .str.contains("Title Bout", case=False, na=False)
    &
    ~ufc_fights_updated["weight_class"]
    .astype(str)
    .str.contains("Tournament", case=False, na=False)
)

# Track previous title fights entering each bout
fighter_title_fights = {}

f1_previous_titles = []
f2_previous_titles = []

for _, row in ufc_fights_updated.iterrows():

    f1 = row["f1_name"]
    f2 = row["f2_name"]

    f1_previous_titles.append(
        fighter_title_fights.get(f1, 0)
    )

    f2_previous_titles.append(
        fighter_title_fights.get(f2, 0)
    )

    # Only update AFTER the fight
    if row["is_title_fight"]:

        fighter_title_fights[f1] = (
            fighter_title_fights.get(f1, 0) + 1
        )

        fighter_title_fights[f2] = (
            fighter_title_fights.get(f2, 0) + 1
        )


ufc_fights_updated["f1_previous_title_fights"] = f1_previous_titles
ufc_fights_updated["f2_previous_title_fights"] = f2_previous_titles


print("TITLE FIGHT HISTORY REBUILT ✅🔥")
print("=" * 90)

print(
    "Total divisional title fights:",
    ufc_fights_updated["is_title_fight"].sum()
)

print()
print("NEW TITLE FIGHTS:")
print("=" * 90)

print(
    ufc_fights_updated[
        (ufc_fights_updated["date"] > "2026-06-27")
        &
        (ufc_fights_updated["is_title_fight"])
    ][
        [
            "date",
            "event_name",
            "f1_name",
            "f1_previous_title_fights",
            "f2_name",
            "f2_previous_title_fights",
            "weight_class"
        ]
    ].to_string(index=False)
)
```

```python
# UFC win percentage entering each fight
# Draws / NCs are excluded from the denominator

def calculate_win_pct(wins, losses):
    decided_fights = wins + losses

    if decided_fights == 0:
        return 0.0

    return wins / decided_fights


ufc_fights_updated["f1_ufc_win_pct_before"] = [
    calculate_win_pct(w, l)
    for w, l in zip(
        ufc_fights_updated["f1_ufc_wins_before"],
        ufc_fights_updated["f1_ufc_losses_before"]
    )
]

ufc_fights_updated["f2_ufc_win_pct_before"] = [
    calculate_win_pct(w, l)
    for w, l in zip(
        ufc_fights_updated["f2_ufc_wins_before"],
        ufc_fights_updated["f2_ufc_losses_before"]
    )
]


print("UFC WIN PERCENTAGES REBUILT ✅🔥")
print("=" * 90)

print(
    ufc_fights_updated[
        ufc_fights_updated["date"] >= "2026-07-01"
    ][
        [
            "date",
            "f1_name",
            "f1_ufc_wins_before",
            "f1_ufc_losses_before",
            "f1_ufc_win_pct_before",
            "f2_name",
            "f2_ufc_wins_before",
            "f2_ufc_losses_before",
            "f2_ufc_win_pct_before"
        ]
    ]
    .head(10)
    .to_string(index=False)
)
```

```python
# Track each fighter's previous UFC results
fighter_history = {}

f1_last5 = []
f2_last5 = []

for _, row in ufc_fights_updated.iterrows():

    f1 = row["f1_name"]
    f2 = row["f2_name"]

    # Previous results entering this fight
    f1_history = fighter_history.get(f1, [])
    f2_history = fighter_history.get(f2, [])

    # Last 5 UFC fights
    f1_recent = f1_history[-5:]
    f2_recent = f2_history[-5:]

    # Win percentage over available last 5 fights
    f1_last5.append(
        sum(result == "W" for result in f1_recent) / len(f1_recent)
        if len(f1_recent) > 0 else 0.0
    )

    f2_last5.append(
        sum(result == "W" for result in f2_recent) / len(f2_recent)
        if len(f2_recent) > 0 else 0.0
    )

    # Add current result AFTER calculating pre-fight form
    fighter_history.setdefault(f1, []).append(row["f1_result"])
    fighter_history.setdefault(f2, []).append(row["f2_result"])


ufc_fights_updated["f1_last5_win_pct"] = f1_last5
ufc_fights_updated["f2_last5_win_pct"] = f2_last5


print("LAST-5 UFC FORM REBUILT ✅🔥")
print("=" * 90)

print(
    ufc_fights_updated[
        ufc_fights_updated["date"] >= "2026-07-01"
    ][
        [
            "date",
            "f1_name",
            "f1_last5_win_pct",
            "f2_name",
            "f2_last5_win_pct"
        ]
    ]
    .head(10)
    .to_string(index=False)
)
```

```python
# Track each fighter's UFC wins and finishes
fighter_finish_history = {}

f1_finishes_before = []
f2_finishes_before = []

f1_finish_rates = []
f2_finish_rates = []

for _, row in ufc_fights_updated.iterrows():

    f1 = row["f1_name"]
    f2 = row["f2_name"]

    f1_stats = fighter_finish_history.get(
        f1, {"wins": 0, "finishes": 0}
    )

    f2_stats = fighter_finish_history.get(
        f2, {"wins": 0, "finishes": 0}
    )

    # Record raw finishes entering fight
    f1_finishes_before.append(f1_stats["finishes"])
    f2_finishes_before.append(f2_stats["finishes"])

    # Finish rate = finishes / UFC wins
    f1_finish_rates.append(
        f1_stats["finishes"] / f1_stats["wins"]
        if f1_stats["wins"] > 0 else 0.0
    )

    f2_finish_rates.append(
        f2_stats["finishes"] / f2_stats["wins"]
        if f2_stats["wins"] > 0 else 0.0
    )

    # Update AFTER current fight
    for fighter, result in [
        (f1, row["f1_result"]),
        (f2, row["f2_result"])
    ]:

        if fighter not in fighter_finish_history:
            fighter_finish_history[fighter] = {
                "wins": 0,
                "finishes": 0
            }

        if result == "W":

            fighter_finish_history[fighter]["wins"] += 1

            if row["method"] in ["KO/TKO", "Submission"]:
                fighter_finish_history[fighter]["finishes"] += 1


ufc_fights_updated["f1_ufc_finishes_before"] = f1_finishes_before
ufc_fights_updated["f2_ufc_finishes_before"] = f2_finishes_before

ufc_fights_updated["f1_finish_rate_before"] = f1_finish_rates
ufc_fights_updated["f2_finish_rate_before"] = f2_finish_rates


print("FINISH HISTORY REBUILT ✅🔥")
print("=" * 90)

print(
    ufc_fights_updated[
        ufc_fights_updated["date"] >= "2026-07-01"
    ][
        [
            "date",
            "f1_name",
            "f1_ufc_finishes_before",
            "f1_finish_rate_before",
            "f2_name",
            "f2_ufc_finishes_before",
            "f2_finish_rate_before"
        ]
    ]
    .head(10)
    .to_string(index=False)
)
```

```python
# ============================================================
# STEP 58 — VERIFIED MAIN EVENTS + MAIN EVENT HISTORY
# ============================================================

# Start with False
ufc_fights_updated["is_main_event"] = False
ufc_fights_updated["is_main_event_verified"] = False

# ------------------------------------------------------------
# 1. Preserve verified historical main events
# ------------------------------------------------------------

old_main_event_lookup = ufc_fights[
    ufc_fights["is_main_event_verified"] == True
][
    ["date", "event_name", "f1_name", "f2_name"]
].copy()

old_main_event_keys = set(
    zip(
        old_main_event_lookup["date"],
        old_main_event_lookup["event_name"],
        old_main_event_lookup["f1_name"],
        old_main_event_lookup["f2_name"]
    )
)

for idx, row in ufc_fights_updated.iterrows():

    key = (
        row["date"],
        row["event_name"],
        row["f1_name"],
        row["f2_name"]
    )

    reverse_key = (
        row["date"],
        row["event_name"],
        row["f2_name"],
        row["f1_name"]
    )

    if key in old_main_event_keys or reverse_key in old_main_event_keys:
        ufc_fights_updated.at[idx, "is_main_event"] = True
        ufc_fights_updated.at[idx, "is_main_event_verified"] = True


# ------------------------------------------------------------
# 2. Verified new main-event fighter pairs
# ------------------------------------------------------------

new_main_events = {
    "UFC 329: McGregor vs. Holloway 2":
        ("Conor McGregor", "Max Holloway"),

    "UFC Fight Night: Du Plessis vs. Usman":
        ("Dricus Du Plessis", "Kamaru Usman"),

    "UFC Fight Night: Ankalaev vs. Guskov":
        ("Magomed Ankalaev", "Bogdan Guskov"),

    "UFC Fight Night: Medic vs. Rodriguez":
        ("Uros Medic", "Daniel Rodriguez"),

    "UFC Fight Night: Gamrot vs. Salkilld":
        ("Mateusz Gamrot", "Quillan Salkilld"),

    "UFC 330: Makhachev vs. Machado Garry":
        ("Islam Makhachev", "Ian Machado Garry"),

    "UFC Fight Night: Hernandez vs. Rodrigues":
        ("Anthony Hernandez", "Gregory Rodrigues"),

    "UFC Fight Night: Nurmagomedov vs. Song":
        ("Umar Nurmagomedov", "Song Yadong"),

    "UFC Fight Night: Hooker vs. Parnasse":
        ("Dan Hooker", "Salahdine Parnasse"),

    "Noche UFC: Silva vs. Delgado":
        ("Jean Silva", "Jose Miguel Delgado"),

    "Crypto.com UFC 331: Van vs. Pantoja 2":
        ("Joshua Van", "Alexandre Pantoja")
}


# ------------------------------------------------------------
# 3. Match new verified headliners to actual fights
# ------------------------------------------------------------

for event_name, fighters in new_main_events.items():

    fighter_a, fighter_b = fighters

    mask = (
        (ufc_fights_updated["event_name"] == event_name)
        &
        (
            (
                (ufc_fights_updated["f1_name"] == fighter_a) &
                (ufc_fights_updated["f2_name"] == fighter_b)
            )
            |
            (
                (ufc_fights_updated["f1_name"] == fighter_b) &
                (ufc_fights_updated["f2_name"] == fighter_a)
            )
        )
    )

    ufc_fights_updated.loc[
        mask,
        ["is_main_event", "is_main_event_verified"]
    ] = True


# ------------------------------------------------------------
# 4. Rebuild previous-main-event counts chronologically
# ------------------------------------------------------------

fighter_main_events = {}

f1_previous_main_events = []
f2_previous_main_events = []

for _, row in ufc_fights_updated.iterrows():

    f1 = row["f1_name"]
    f2 = row["f2_name"]

    f1_previous_main_events.append(
        fighter_main_events.get(f1, 0)
    )

    f2_previous_main_events.append(
        fighter_main_events.get(f2, 0)
    )

    # Update only AFTER current fight
    if row["is_main_event_verified"]:

        fighter_main_events[f1] = (
            fighter_main_events.get(f1, 0) + 1
        )

        fighter_main_events[f2] = (
            fighter_main_events.get(f2, 0) + 1
        )


ufc_fights_updated["f1_previous_main_events"] = (
    f1_previous_main_events
)

ufc_fights_updated["f2_previous_main_events"] = (
    f2_previous_main_events
)


# ------------------------------------------------------------
# 5. VALIDATION
# ------------------------------------------------------------

new_period = ufc_fights_updated[
    ufc_fights_updated["date"] > "2026-06-27"
]

new_verified = new_period[
    new_period["is_main_event_verified"]
]


print("MAIN EVENT HISTORY REBUILT ✅🔥")
print("=" * 100)

print(
    "Historical verified main events:",
    ufc_fights_updated.loc[
        ufc_fights_updated["date"] <= "2026-06-27",
        "is_main_event_verified"
    ].sum()
)

print(
    "New verified main events:",
    new_period["is_main_event_verified"].sum()
)

print(
    "Total verified main events:",
    ufc_fights_updated["is_main_event_verified"].sum()
)

print()
print("NEW VERIFIED MAIN EVENTS:")
print("=" * 100)

print(
    new_verified[
        [
            "date",
            "event_name",
            "f1_name",
            "f1_previous_main_events",
            "f2_name",
            "f2_previous_main_events"
        ]
    ]
    .sort_values("date")
    .to_string(index=False)
)

print()
print("EVENTS WITHOUT EXACTLY ONE VERIFIED MAIN EVENT:")
print("=" * 100)

new_event_check = (
    new_period
    .groupby("event_name")["is_main_event_verified"]
    .sum()
)

problems = new_event_check[new_event_check != 1]

if len(problems) == 0:
    print("NONE — all 11 new events have exactly one verified main event ✅")
else:
    print(problems)
```

```python
import requests
from bs4 import BeautifulSoup

print("OFFICIAL UFC BONUS AUDIT 🔥")
print("=" * 100)

bonus_audit = []

for i, url in enumerate(completed_update_urls, 1):

    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=30
    )

    soup = BeautifulSoup(response.text, "html.parser")

    event_name = (
        new_fights.loc[
            new_fights["source_url"] == url,
            "event_name"
        ].iloc[0]
    )

    print()
    print(f"{i}/11 — {event_name}")
    print("-" * 100)

    fight_blocks = soup.select(".c-listing-fight")

    found = 0

    for block in fight_blocks:

        text = " ".join(block.stripped_strings)

        award_terms = [
            term for term in [
                "Fight of the Night",
                "Performance of the Night",
                "Knockout of the Night",
                "Submission of the Night"
            ]
            if term.lower() in text.lower()
        ]

        if award_terms:

            red = block.select_one(
                ".c-listing-fight__corner-name--red"
            )

            blue = block.select_one(
                ".c-listing-fight__corner-name--blue"
            )

            red_name = (
                " ".join(red.stripped_strings)
                if red else "UNKNOWN"
            )

            blue_name = (
                " ".join(blue.stripped_strings)
                if blue else "UNKNOWN"
            )

            print(
                red_name,
                "vs",
                blue_name,
                "→",
                ", ".join(award_terms)
            )

            bonus_audit.append({
                "event_name": event_name,
                "f1_name": red_name,
                "f2_name": blue_name,
                "awards": award_terms
            })

            found += 1

    if found == 0:
        print("No award labels detected")


print()
print("=" * 100)
print("BONUS-LABELED FIGHTS FOUND:", len(bonus_audit))
```

```python
# ============================================================
# STEP 60 — BONUSES + HISTORICAL BONUS EXPERIENCE
# ============================================================

# ------------------------------------------------------------
# 1. Preserve original historical bonus flags
# ------------------------------------------------------------

old_bonus_lookup = {}

for _, row in ufc_fights.iterrows():

    fighters = tuple(sorted([
        row["f1_name"],
        row["f2_name"]
    ]))

    key = (
        pd.Timestamp(row["date"]),
        fighters
    )

    old_bonus_lookup[key] = {
        "FOTN": bool(row["FOTN"]),
        "POTN": bool(row["POTN"]),
        "KOTN": bool(row["KOTN"]),
        "SOTN": bool(row["SOTN"])
    }


# Start all rebuilt bonus flags as False
for col in ["FOTN", "POTN", "KOTN", "SOTN"]:
    ufc_fights_updated[col] = False


# Restore historical bonuses
for idx, row in ufc_fights_updated.iterrows():

    if row["date"] <= pd.Timestamp("2026-06-27"):

        fighters = tuple(sorted([
            row["f1_name"],
            row["f2_name"]
        ]))

        key = (
            pd.Timestamp(row["date"]),
            fighters
        )

        if key in old_bonus_lookup:

            for col in ["FOTN", "POTN", "KOTN", "SOTN"]:
                ufc_fights_updated.at[idx, col] = (
                    old_bonus_lookup[key][col]
                )


# ------------------------------------------------------------
# 2. Normalize names for matching UFC.com bonus audit
# ------------------------------------------------------------

import unicodedata
import re

def normalize_bonus_name(name):

    name = str(name)

    # Remove accents
    name = unicodedata.normalize("NFKD", name)
    name = "".join(
        c for c in name
        if not unicodedata.combining(c)
    )

    # Standardize apostrophes
    name = (
        name
        .replace("’", "'")
        .replace("‘", "'")
    )

    # Keep letters/numbers only for matching
    name = re.sub(r"[^A-Za-z0-9]", "", name)

    return name.lower()


# ------------------------------------------------------------
# 3. Apply new UFC.com bonus labels
# ------------------------------------------------------------

bonus_matches = 0
bonus_unmatched = []

for bonus_row in bonus_audit:

    event_name = bonus_row["event_name"]

    red_norm = normalize_bonus_name(
        bonus_row["f1_name"]
    )

    blue_norm = normalize_bonus_name(
        bonus_row["f2_name"]
    )

    event_fights = ufc_fights_updated[
        ufc_fights_updated["event_name"] == event_name
    ]

    matched_idx = None

    for idx, fight in event_fights.iterrows():

        f1_norm = normalize_bonus_name(
            fight["f1_name"]
        )

        f2_norm = normalize_bonus_name(
            fight["f2_name"]
        )

        if {red_norm, blue_norm} == {f1_norm, f2_norm}:
            matched_idx = idx
            break

    if matched_idx is None:

        bonus_unmatched.append(
            (
                event_name,
                bonus_row["f1_name"],
                bonus_row["f2_name"]
            )
        )

        continue

    awards = bonus_row["awards"]

    if "Fight of the Night" in awards:
        ufc_fights_updated.at[
            matched_idx, "FOTN"
        ] = True

    if "Performance of the Night" in awards:
        ufc_fights_updated.at[
            matched_idx, "POTN"
        ] = True

    if "Knockout of the Night" in awards:
        ufc_fights_updated.at[
            matched_idx, "KOTN"
        ] = True

    if "Submission of the Night" in awards:
        ufc_fights_updated.at[
            matched_idx, "SOTN"
        ] = True

    bonus_matches += 1


# ------------------------------------------------------------
# 4. Rebuild historical bonus counts entering each fight
# ------------------------------------------------------------

fighter_bonuses = {}

bonus_types = [
    "FOTN",
    "POTN",
    "KOTN",
    "SOTN"
]

history_columns = {
    f"f1_previous_{bonus}": []
    for bonus in bonus_types
}

history_columns.update({
    f"f2_previous_{bonus}": []
    for bonus in bonus_types
})


for _, row in ufc_fights_updated.iterrows():

    f1 = row["f1_name"]
    f2 = row["f2_name"]

    if f1 not in fighter_bonuses:
        fighter_bonuses[f1] = {
            bonus: 0 for bonus in bonus_types
        }

    if f2 not in fighter_bonuses:
        fighter_bonuses[f2] = {
            bonus: 0 for bonus in bonus_types
        }

    # Record PRE-FIGHT bonus histories
    for bonus in bonus_types:

        history_columns[
            f"f1_previous_{bonus}"
        ].append(
            fighter_bonuses[f1][bonus]
        )

        history_columns[
            f"f2_previous_{bonus}"
        ].append(
            fighter_bonuses[f2][bonus]
        )


    # --------------------------------------------------------
    # Update AFTER fight
    # --------------------------------------------------------

    # Fight of the Night = BOTH fighters receive bonus
    if row["FOTN"]:

        fighter_bonuses[f1]["FOTN"] += 1
        fighter_bonuses[f2]["FOTN"] += 1


    # Individual awards = winning fighter receives bonus
    winner = None

    if row["f1_result"] == "W":
        winner = f1

    elif row["f2_result"] == "W":
        winner = f2


    if winner is not None:

        if row["POTN"]:
            fighter_bonuses[winner]["POTN"] += 1

        if row["KOTN"]:
            fighter_bonuses[winner]["KOTN"] += 1

        if row["SOTN"]:
            fighter_bonuses[winner]["SOTN"] += 1


# Add historical columns
for col, values in history_columns.items():
    ufc_fights_updated[col] = values


# ------------------------------------------------------------
# 5. Total previous bonuses
# ------------------------------------------------------------

ufc_fights_updated[
    "f1_previous_total_bonuses"
] = (
    ufc_fights_updated["f1_previous_FOTN"]
    + ufc_fights_updated["f1_previous_POTN"]
    + ufc_fights_updated["f1_previous_KOTN"]
    + ufc_fights_updated["f1_previous_SOTN"]
)

ufc_fights_updated[
    "f2_previous_total_bonuses"
] = (
    ufc_fights_updated["f2_previous_FOTN"]
    + ufc_fights_updated["f2_previous_POTN"]
    + ufc_fights_updated["f2_previous_KOTN"]
    + ufc_fights_updated["f2_previous_SOTN"]
)


# ------------------------------------------------------------
# 6. VALIDATION
# ------------------------------------------------------------

old_period = (
    ufc_fights_updated["date"]
    <= pd.Timestamp("2026-06-27")
)

new_period = (
    ufc_fights_updated["date"]
    > pd.Timestamp("2026-06-27")
)


print("BONUS HISTORY REBUILT ✅🔥")
print("=" * 100)

print("UFC.com bonus fights matched:", bonus_matches)
print("Unmatched bonus fights:", len(bonus_unmatched))

print()
print("HISTORICAL BONUS FLAGS — ORIGINAL VS REBUILT")
print("-" * 100)

for bonus in bonus_types:

    original = int(
        ufc_fights[bonus].sum()
    )

    rebuilt = int(
        ufc_fights_updated.loc[
            old_period,
            bonus
        ].sum()
    )

    print(
        f"{bonus}: original={original} | rebuilt={rebuilt}"
    )


print()
print("NEW BONUS FLAGS")
print("-" * 100)

for bonus in bonus_types:

    count = int(
        ufc_fights_updated.loc[
            new_period,
            bonus
        ].sum()
    )

    print(f"{bonus}: {count}")


print()
print("UNMATCHED:")
print("-" * 100)

if len(bonus_unmatched) == 0:
    print("NONE — all 36 UFC.com bonus fights matched ✅")
else:
    for item in bonus_unmatched:
        print(item)
```

```python
# ============================================================
# STEP 61 — FIX JAN BLACHOWICZ BONUS MATCH
# ============================================================

jan_mask = (
    (ufc_fights_updated["event_name"] ==
     "UFC Fight Night: Medic vs. Rodriguez")
    &
    (
        (ufc_fights_updated["f1_name"] == "Jan Blachowicz")
        |
        (ufc_fights_updated["f2_name"] == "Jan Blachowicz")
    )
)

print("Jan fights matched:", jan_mask.sum())

# Official UFC.com page awarded this fight a POTN
ufc_fights_updated.loc[jan_mask, "POTN"] = True


# ============================================================
# REBUILD BONUS HISTORY AFTER FIX
# ============================================================

fighter_bonuses = {}

bonus_types = ["FOTN", "POTN", "KOTN", "SOTN"]

history_columns = {}

for bonus in bonus_types:
    history_columns[f"f1_previous_{bonus}"] = []
    history_columns[f"f2_previous_{bonus}"] = []


for _, row in ufc_fights_updated.iterrows():

    f1 = row["f1_name"]
    f2 = row["f2_name"]

    if f1 not in fighter_bonuses:
        fighter_bonuses[f1] = {
            bonus: 0 for bonus in bonus_types
        }

    if f2 not in fighter_bonuses:
        fighter_bonuses[f2] = {
            bonus: 0 for bonus in bonus_types
        }

    # Record entering-fight totals
    for bonus in bonus_types:

        history_columns[
            f"f1_previous_{bonus}"
        ].append(
            fighter_bonuses[f1][bonus]
        )

        history_columns[
            f"f2_previous_{bonus}"
        ].append(
            fighter_bonuses[f2][bonus]
        )

    # FOTN credits both fighters
    if row["FOTN"]:
        fighter_bonuses[f1]["FOTN"] += 1
        fighter_bonuses[f2]["FOTN"] += 1

    # Individual bonuses credit winner
    if row["f1_result"] == "W":
        winner = f1
    elif row["f2_result"] == "W":
        winner = f2
    else:
        winner = None

    if winner is not None:

        if row["POTN"]:
            fighter_bonuses[winner]["POTN"] += 1

        if row["KOTN"]:
            fighter_bonuses[winner]["KOTN"] += 1

        if row["SOTN"]:
            fighter_bonuses[winner]["SOTN"] += 1


for col, values in history_columns.items():
    ufc_fights_updated[col] = values


ufc_fights_updated["f1_previous_total_bonuses"] = (
    ufc_fights_updated["f1_previous_FOTN"]
    + ufc_fights_updated["f1_previous_POTN"]
    + ufc_fights_updated["f1_previous_KOTN"]
    + ufc_fights_updated["f1_previous_SOTN"]
)

ufc_fights_updated["f2_previous_total_bonuses"] = (
    ufc_fights_updated["f2_previous_FOTN"]
    + ufc_fights_updated["f2_previous_POTN"]
    + ufc_fights_updated["f2_previous_KOTN"]
    + ufc_fights_updated["f2_previous_SOTN"]
)


# ============================================================
# FINAL CHECK
# ============================================================

new_mask = (
    ufc_fights_updated["date"]
    > pd.Timestamp("2026-06-27")
)

print()
print("BONUS FIX COMPLETE ✅🔥")
print("=" * 80)

print(
    "New FOTN:",
    int(ufc_fights_updated.loc[new_mask, "FOTN"].sum())
)

print(
    "New POTN:",
    int(ufc_fights_updated.loc[new_mask, "POTN"].sum())
)

print(
    "New KOTN:",
    int(ufc_fights_updated.loc[new_mask, "KOTN"].sum())
)

print(
    "New SOTN:",
    int(ufc_fights_updated.loc[new_mask, "SOTN"].sum())
)

print()
print("Jan Blachowicz fight:")
print(
    ufc_fights_updated.loc[
        jan_mask,
        [
            "date",
            "f1_name",
            "f2_name",
            "FOTN",
            "POTN"
        ]
    ].to_string(index=False)
)
```

```python
# ============================================================
# STEP 63 — EXTRACT UFC.COM EVENT-TIME RANKINGS BY CORNER
# ============================================================

import requests
from bs4 import BeautifulSoup
import re
import pandas as pd

ufc_rank_rows = []

def extract_corner_rank(corner):

    if corner is None:
        return None

    text = " ".join(corner.stripped_strings)

    # Champion = rank 0 in our existing methodology
    if re.search(r"(^|\s)C(\s|$)", text):
        return 0

    # Ranked contender
    match = re.search(r"#\s*(\d{1,2})", text)

    if match:
        return int(match.group(1))

    # Unranked
    return None


for i, url in enumerate(completed_update_urls, 1):

    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=30
    )

    soup = BeautifulSoup(response.text, "html.parser")

    event_name = (
        new_fights.loc[
            new_fights["source_url"] == url,
            "event_name"
        ].iloc[0]
    )

    fight_blocks = soup.select(".c-listing-fight")

    for block in fight_blocks:

        red_name_el = block.select_one(
            ".c-listing-fight__corner-name--red"
        )

        blue_name_el = block.select_one(
            ".c-listing-fight__corner-name--blue"
        )

        if red_name_el is None or blue_name_el is None:
            continue

        red_name = " ".join(red_name_el.stripped_strings)
        blue_name = " ".join(blue_name_el.stripped_strings)

        # Entire red/blue corner containers
        red_corner = block.select_one(
            ".c-listing-fight__corner--red"
        )

        blue_corner = block.select_one(
            ".c-listing-fight__corner--blue"
        )

        red_rank = extract_corner_rank(red_corner)
        blue_rank = extract_corner_rank(blue_corner)

        ufc_rank_rows.append({
            "event_name": event_name,
            "red_name": red_name,
            "red_rank": red_rank,
            "blue_name": blue_name,
            "blue_rank": blue_rank
        })


ufc_event_rankings = pd.DataFrame(ufc_rank_rows)


# ============================================================
# VALIDATION
# ============================================================

print("EVENT-TIME RANKINGS EXTRACTED ✅🔥")
print("=" * 100)

print("Fight rows:", len(ufc_event_rankings))

print(
    "Ranked fighter entries:",
    ufc_event_rankings["red_rank"].notna().sum()
    + ufc_event_rankings["blue_rank"].notna().sum()
)

print(
    "Champions detected:",
    (ufc_event_rankings["red_rank"] == 0).sum()
    + (ufc_event_rankings["blue_rank"] == 0).sum()
)

print()
print("RANKED FIGHTS:")
print("=" * 100)

ranked_fights = ufc_event_rankings[
    ufc_event_rankings["red_rank"].notna()
    |
    ufc_event_rankings["blue_rank"].notna()
]

print(
    ranked_fights[
        [
            "event_name",
            "red_name",
            "red_rank",
            "blue_name",
            "blue_rank"
        ]
    ].to_string(index=False)
)
```

```python
# ============================================================
# STEP 65 — CORRECT UFC.COM EVENT-TIME RANK EXTRACTION
# ============================================================

import requests
from bs4 import BeautifulSoup
import pandas as pd
import re

ranking_rows = []


def clean_rank(value):

    if value is None:
        return None

    value = value.strip()

    if value == "C":
        return 0

    match = re.search(r"#\s*(\d{1,2})", value)

    if match:
        return int(match.group(1))

    return None


for url in completed_update_urls:

    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=30
    )

    soup = BeautifulSoup(response.text, "html.parser")

    event_name = (
        new_fights.loc[
            new_fights["source_url"] == url,
            "event_name"
        ].iloc[0]
    )

    for block in soup.select(".c-listing-fight"):

        red_el = block.select_one(
            ".c-listing-fight__corner-name--red"
        )

        blue_el = block.select_one(
            ".c-listing-fight__corner-name--blue"
        )

        if red_el is None or blue_el is None:
            continue

        red_name = " ".join(red_el.stripped_strings)
        blue_name = " ".join(blue_el.stripped_strings)

        # ----------------------------------------------------
        # Extract ranking elements directly
        # ----------------------------------------------------

        rank_elements = block.select(
            ".c-listing-fight__corner-rank"
        )

        rank_values = []

        for element in rank_elements:

            text = " ".join(element.stripped_strings).strip()

            if (
                text == "C"
                or re.fullmatch(r"#\s*\d{1,2}", text)
            ):
                rank_values.append(text)

        # Remove repeated desktop/mobile copies while
        # preserving first appearance order
        unique_rank_values = []

        for value in rank_values:
            if value not in unique_rank_values:
                unique_rank_values.append(value)

        # ----------------------------------------------------
        # Determine red / blue rank
        #
        # UFC markup presents red ranking first and blue second.
        # If only one rank exists, use the dedicated mobile
        # blue-rank class to determine whether it belongs to blue.
        # ----------------------------------------------------

        red_rank = None
        blue_rank = None

        if len(unique_rank_values) >= 2:

            red_rank = clean_rank(
                unique_rank_values[0]
            )

            blue_rank = clean_rank(
                unique_rank_values[1]
            )

        elif len(unique_rank_values) == 1:

            value = unique_rank_values[0]

            blue_mobile = block.select_one(
                ".c-listing-fight__blue-corner-rank--mobile"
            )

            blue_mobile_text = (
                " ".join(blue_mobile.stripped_strings).strip()
                if blue_mobile else None
            )

            if blue_mobile_text == value:
                blue_rank = clean_rank(value)
            else:
                red_rank = clean_rank(value)


        ranking_rows.append({
            "event_name": event_name,
            "red_name": red_name,
            "red_rank": red_rank,
            "blue_name": blue_name,
            "blue_rank": blue_rank
        })


ufc_event_rankings = pd.DataFrame(ranking_rows)


# ============================================================
# VALIDATION
# ============================================================

ranked = ufc_event_rankings[
    ufc_event_rankings["red_rank"].notna()
    |
    ufc_event_rankings["blue_rank"].notna()
]


print("CORRECTED EVENT-TIME RANKINGS EXTRACTED ✅🔥")
print("=" * 100)

print("Fight rows:", len(ufc_event_rankings))

print(
    "Ranked fighter entries:",
    int(
        ufc_event_rankings["red_rank"].notna().sum()
        +
        ufc_event_rankings["blue_rank"].notna().sum()
    )
)

print(
    "Champions detected:",
    int(
        (ufc_event_rankings["red_rank"] == 0).sum()
        +
        (ufc_event_rankings["blue_rank"] == 0).sum()
    )
)

print()
print("RANKED FIGHTS:")
print("=" * 100)

print(
    ranked[
        [
            "event_name",
            "red_name",
            "red_rank",
            "blue_name",
            "blue_rank"
        ]
    ].to_string(index=False)
)
```

```python
# ============================================================
# STEP 66 — INTEGRATE HISTORICAL + NEW EVENT-TIME RANKINGS
# ============================================================

import numpy as np
import pandas as pd
import unicodedata
import re


# ------------------------------------------------------------
# 1. Restore historical ranking columns exactly
# ------------------------------------------------------------

ufc_fights_updated["ranking_division"] = np.nan
ufc_fights_updated["f1_rank_before"] = np.nan
ufc_fights_updated["f2_rank_before"] = np.nan


# Since first 8,736 rows are the historical fights in the same
# chronological order, restore their existing ranking data.
old_count = len(ufc_fights)

ufc_fights_updated.loc[
    :old_count - 1,
    "ranking_division"
] = ufc_fights["ranking_division"].values

ufc_fights_updated.loc[
    :old_count - 1,
    "f1_rank_before"
] = ufc_fights["f1_rank_before"].values

ufc_fights_updated.loc[
    :old_count - 1,
    "f2_rank_before"
] = ufc_fights["f2_rank_before"].values


# ------------------------------------------------------------
# 2. Ranking division from weight class
# ------------------------------------------------------------

def ranking_division_from_weight_class(weight_class):

    wc = str(weight_class)

    wc = (
        wc
        .replace("UFC ", "")
        .replace(" Title Bout", "")
        .replace(" Bout", "")
        .strip()
    )

    # Catchweight has no UFC divisional ranking
    if "Catchweight" in wc:
        return np.nan

    return wc


new_mask = (
    ufc_fights_updated["date"]
    > pd.Timestamp("2026-06-27")
)

ufc_fights_updated.loc[
    new_mask,
    "ranking_division"
] = (
    ufc_fights_updated.loc[
        new_mask,
        "weight_class"
    ].apply(ranking_division_from_weight_class)
)


# ------------------------------------------------------------
# 3. Normalize names only for matching
# ------------------------------------------------------------

def normalize_rank_name(name):

    name = str(name)

    name = unicodedata.normalize("NFKD", name)

    name = "".join(
        c for c in name
        if not unicodedata.combining(c)
    )

    name = (
        name
        .replace("’", "'")
        .replace("‘", "'")
    )

    name = re.sub(
        r"[^A-Za-z0-9]",
        "",
        name
    )

    return name.lower()


# ------------------------------------------------------------
# 4. Merge UFC.com ranks onto the 141 new fights
# ------------------------------------------------------------

rank_matches = 0
rank_unmatched = []

for _, rank_row in ufc_event_rankings.iterrows():

    event_name = rank_row["event_name"]

    red_norm = normalize_rank_name(
        rank_row["red_name"]
    )

    blue_norm = normalize_rank_name(
        rank_row["blue_name"]
    )

    candidates = ufc_fights_updated[
        (ufc_fights_updated["event_name"] == event_name)
        &
        new_mask
    ]

    matched_idx = None
    reversed_order = False

    for idx, fight in candidates.iterrows():

        f1_norm = normalize_rank_name(
            fight["f1_name"]
        )

        f2_norm = normalize_rank_name(
            fight["f2_name"]
        )

        if (
            red_norm == f1_norm
            and blue_norm == f2_norm
        ):
            matched_idx = idx
            reversed_order = False
            break

        if (
            red_norm == f2_norm
            and blue_norm == f1_norm
        ):
            matched_idx = idx
            reversed_order = True
            break


    if matched_idx is None:

        rank_unmatched.append(
            (
                event_name,
                rank_row["red_name"],
                rank_row["blue_name"]
            )
        )

        continue


    if not reversed_order:

        ufc_fights_updated.at[
            matched_idx,
            "f1_rank_before"
        ] = rank_row["red_rank"]

        ufc_fights_updated.at[
            matched_idx,
            "f2_rank_before"
        ] = rank_row["blue_rank"]

    else:

        ufc_fights_updated.at[
            matched_idx,
            "f1_rank_before"
        ] = rank_row["blue_rank"]

        ufc_fights_updated.at[
            matched_idx,
            "f2_rank_before"
        ] = rank_row["red_rank"]


    rank_matches += 1


# ------------------------------------------------------------
# 5. VALIDATE HISTORICAL DATA WAS PRESERVED
# ------------------------------------------------------------

historical_f1_same = (
    ufc_fights_updated.loc[
        :old_count - 1,
        "f1_rank_before"
    ]
    .reset_index(drop=True)
    .equals(
        ufc_fights["f1_rank_before"]
        .reset_index(drop=True)
    )
)

historical_f2_same = (
    ufc_fights_updated.loc[
        :old_count - 1,
        "f2_rank_before"
    ]
    .reset_index(drop=True)
    .equals(
        ufc_fights["f2_rank_before"]
        .reset_index(drop=True)
    )
)


# ------------------------------------------------------------
# 6. NEW RANKING SUMMARY
# ------------------------------------------------------------

new_ranked_entries = int(
    ufc_fights_updated.loc[
        new_mask,
        "f1_rank_before"
    ].notna().sum()
    +
    ufc_fights_updated.loc[
        new_mask,
        "f2_rank_before"
    ].notna().sum()
)

new_champions = int(
    (
        ufc_fights_updated.loc[
            new_mask,
            "f1_rank_before"
        ] == 0
    ).sum()
    +
    (
        ufc_fights_updated.loc[
            new_mask,
            "f2_rank_before"
        ] == 0
    ).sum()
)


print("RANKINGS INTEGRATED ✅🔥")
print("=" * 100)

print("New fights matched:", rank_matches)
print("Unmatched new fights:", len(rank_unmatched))

print()
print(
    "Historical F1 rankings preserved:",
    historical_f1_same
)

print(
    "Historical F2 rankings preserved:",
    historical_f2_same
)

print()
print(
    "New ranked fighter entries:",
    new_ranked_entries
)

print(
    "New champions (rank 0):",
    new_champions
)

print()
print("UNMATCHED FIGHTS:")
print("-" * 100)

if len(rank_unmatched) == 0:
    print("NONE — all 141 new fights matched ✅")
else:
    for item in rank_unmatched:
        print(item)
```

```python
# ============================================================
# STEP 67 — PATCH 4 RANKING NAME MISMATCHES + VALIDATE
# ============================================================

import numpy as np


# ------------------------------------------------------------
# 1. Fix ranking_division dtype warning for future use
# ------------------------------------------------------------

ufc_fights_updated["ranking_division"] = (
    ufc_fights_updated["ranking_division"]
    .astype("object")
)


# ------------------------------------------------------------
# 2. UFC.com name -> standardized dataset name
# ------------------------------------------------------------

ranking_name_fixes = {
    "Zachary Reese": "Zach Reese",
    "Jan Błachowicz": "Jan Blachowicz",
    "Michael Venom Page": "Michael Page",
    "Klaudia Syguła": "Klaudia Sygula"
}


# ------------------------------------------------------------
# 3. Patch the four previously unmatched fights
# ------------------------------------------------------------

patched = 0
still_unmatched = []

for event_name, red_name, blue_name in rank_unmatched:

    red_fixed = ranking_name_fixes.get(
        red_name,
        red_name
    )

    blue_fixed = ranking_name_fixes.get(
        blue_name,
        blue_name
    )

    # Find original UFC.com ranking row
    rank_row = ufc_event_rankings[
        (ufc_event_rankings["event_name"] == event_name)
        &
        (ufc_event_rankings["red_name"] == red_name)
        &
        (ufc_event_rankings["blue_name"] == blue_name)
    ]

    if len(rank_row) != 1:
        still_unmatched.append(
            (event_name, red_name, blue_name)
        )
        continue

    rank_row = rank_row.iloc[0]

    # Find standardized fight row
    fight_match = ufc_fights_updated[
        new_mask
        &
        (ufc_fights_updated["event_name"] == event_name)
        &
        (
            (
                (ufc_fights_updated["f1_name"] == red_fixed)
                &
                (ufc_fights_updated["f2_name"] == blue_fixed)
            )
            |
            (
                (ufc_fights_updated["f1_name"] == blue_fixed)
                &
                (ufc_fights_updated["f2_name"] == red_fixed)
            )
        )
    ]

    if len(fight_match) != 1:
        still_unmatched.append(
            (event_name, red_name, blue_name)
        )
        continue

    idx = fight_match.index[0]

    # Assign rankings to correct fighter
    if (
        ufc_fights_updated.at[idx, "f1_name"]
        == red_fixed
    ):

        ufc_fights_updated.at[
            idx, "f1_rank_before"
        ] = rank_row["red_rank"]

        ufc_fights_updated.at[
            idx, "f2_rank_before"
        ] = rank_row["blue_rank"]

    else:

        ufc_fights_updated.at[
            idx, "f1_rank_before"
        ] = rank_row["blue_rank"]

        ufc_fights_updated.at[
            idx, "f2_rank_before"
        ] = rank_row["red_rank"]

    patched += 1


# ------------------------------------------------------------
# 4. FINAL VALIDATION
# ------------------------------------------------------------

new_ranked_entries = int(
    ufc_fights_updated.loc[
        new_mask,
        "f1_rank_before"
    ].notna().sum()
    +
    ufc_fights_updated.loc[
        new_mask,
        "f2_rank_before"
    ].notna().sum()
)

new_champions = int(
    (
        ufc_fights_updated.loc[
            new_mask,
            "f1_rank_before"
        ] == 0
    ).sum()
    +
    (
        ufc_fights_updated.loc[
            new_mask,
            "f2_rank_before"
        ] == 0
    ).sum()
)


historical_f1_same = (
    ufc_fights_updated.loc[
        :old_count - 1,
        "f1_rank_before"
    ]
    .reset_index(drop=True)
    .equals(
        ufc_fights["f1_rank_before"]
        .reset_index(drop=True)
    )
)

historical_f2_same = (
    ufc_fights_updated.loc[
        :old_count - 1,
        "f2_rank_before"
    ]
    .reset_index(drop=True)
    .equals(
        ufc_fights["f2_rank_before"]
        .reset_index(drop=True)
    )
)


print("RANKING PATCH COMPLETE ✅🔥")
print("=" * 100)

print("Previously unmatched:", len(rank_unmatched))
print("Successfully patched:", patched)
print("Still unmatched:", len(still_unmatched))

print()
print("New ranked fighter entries:", new_ranked_entries)
print("New champions (rank 0):", new_champions)

print()
print(
    "Historical F1 rankings preserved:",
    historical_f1_same
)

print(
    "Historical F2 rankings preserved:",
    historical_f2_same
)

print()

if still_unmatched:
    print("STILL UNMATCHED:")
    for x in still_unmatched:
        print(x)
else:
    print("ALL 141 NEW FIGHTS ACCOUNTED FOR ✅🔥")
```

```python
# ============================================================
# STEP 72 — GOOGLE TRENDS FOR ALL 11 NEW EVENTS
# ============================================================

from pytrends.request import TrendReq
import pandas as pd
import numpy as np
import time


# ------------------------------------------------------------
# 1. Add confirmed Parnasse topic
# ------------------------------------------------------------

parnasse_row = pd.DataFrame([{
    "fighter": "Salahdine Parnasse",
    "topic_id": "/g/11g0kd0b7f",
    "status": "found"
}])

fighter_topics = pd.concat(
    [
        fighter_topics[
            fighter_topics["fighter"] != "Salahdine Parnasse"
        ],
        parnasse_row
    ],
    ignore_index=True
)


# ------------------------------------------------------------
# 2. Delgado stays a documented search-term fallback
# ------------------------------------------------------------

delgado_row = pd.DataFrame([{
    "fighter": "Jose Miguel Delgado",
    "topic_id": np.nan,
    "status": "search_term_fallback"
}])

fighter_topics = pd.concat(
    [
        fighter_topics[
            fighter_topics["fighter"] != "Jose Miguel Delgado"
        ],
        delgado_row
    ],
    ignore_index=True
)


# ------------------------------------------------------------
# 3. Rebuild topic lookup
# ------------------------------------------------------------

topic_lookup = (
    fighter_topics
    .dropna(subset=["topic_id"])
    .set_index("fighter")["topic_id"]
    .to_dict()
)


def trends_keyword(fighter):

    if fighter == "Jose Miguel Delgado":
        return "Jose Miguel Delgado"

    return topic_lookup.get(
        fighter,
        fighter
    )


# ------------------------------------------------------------
# 4. Google Trends client
# ------------------------------------------------------------

pytrends = TrendReq(
    hl="en-US",
    tz=360
)


# ------------------------------------------------------------
# 5. Function to calculate one fighter's hype
# ------------------------------------------------------------

def get_fighter_hype(fighter, event_date):

    event_date = pd.Timestamp(event_date)

    keyword = trends_keyword(fighter)

    # Exact original windows:
    # baseline = event -97 through event -8
    # fight week = event -7 through event -1

    start_date = (
        event_date - pd.Timedelta(days=97)
    )

    end_date = (
        event_date - pd.Timedelta(days=1)
    )

    baseline_start = start_date
    baseline_end = (
        event_date - pd.Timedelta(days=8)
    )

    fight_week_start = (
        event_date - pd.Timedelta(days=7)
    )

    fight_week_end = (
        event_date - pd.Timedelta(days=1)
    )

    timeframe = (
        f"{start_date.strftime('%Y-%m-%d')} "
        f"{end_date.strftime('%Y-%m-%d')}"
    )

    pytrends.build_payload(
        [keyword],
        timeframe=timeframe,
        geo="US"
    )

    data = pytrends.interest_over_time()

    if data.empty or keyword not in data.columns:
        return {
            "keyword": keyword,
            "baseline": np.nan,
            "fight_week": np.nan,
            "smoothed_hype": np.nan,
            "log_hype": np.nan
        }

    values = data[keyword]

    baseline_values = values[
        (values.index >= baseline_start)
        &
        (values.index <= baseline_end)
    ]

    fight_week_values = values[
        (values.index >= fight_week_start)
        &
        (values.index <= fight_week_end)
    ]

    if (
        len(baseline_values) == 0
        or len(fight_week_values) == 0
    ):
        return {
            "keyword": keyword,
            "baseline": np.nan,
            "fight_week": np.nan,
            "smoothed_hype": np.nan,
            "log_hype": np.nan
        }

    baseline = float(
        baseline_values.mean()
    )

    fight_week = float(
        fight_week_values.mean()
    )

    smoothed_hype = (
        (fight_week + 1)
        /
        (baseline + 1)
    )

    log_hype = float(
        np.log1p(smoothed_hype)
    )

    return {
        "keyword": keyword,
        "baseline": baseline,
        "fight_week": fight_week,
        "smoothed_hype": smoothed_hype,
        "log_hype": log_hype
    }


# ------------------------------------------------------------
# 6. Calculate all 11 new events
# ------------------------------------------------------------

new_trends_rows = []

print("CALCULATING NEW EVENT FANFARE 🔥")
print("=" * 100)

for _, event in new_main_event_rows.iterrows():

    event_date = pd.Timestamp(
        event["date"]
    )

    event_name = event["event_name"]

    h1 = event["f1_name"]
    h2 = event["f2_name"]

    print()
    print(
        event_date.date(),
        "|",
        event_name
    )

    try:

        h1_result = get_fighter_hype(
            h1,
            event_date
        )

        time.sleep(3)

        h2_result = get_fighter_hype(
            h2,
            event_date
        )

        time.sleep(3)

        event_avg_baseline = np.nanmean([
            h1_result["baseline"],
            h2_result["baseline"]
        ])

        event_avg_fight_week = np.nanmean([
            h1_result["fight_week"],
            h2_result["fight_week"]
        ])

        event_avg_log_hype = np.nanmean([
            h1_result["log_hype"],
            h2_result["log_hype"]
        ])

        new_trends_rows.append({
            "event_date": event_date,
            "event_name": event_name,

            "headliner_1": h1,
            "headliner_2": h2,

            "headliner_1_topic":
                h1_result["keyword"],

            "headliner_2_topic":
                h2_result["keyword"],

            "h1_baseline":
                h1_result["baseline"],

            "h1_fight_week":
                h1_result["fight_week"],

            "h1_smoothed_hype":
                h1_result["smoothed_hype"],

            "h1_log_hype":
                h1_result["log_hype"],

            "h2_baseline":
                h2_result["baseline"],

            "h2_fight_week":
                h2_result["fight_week"],

            "h2_smoothed_hype":
                h2_result["smoothed_hype"],

            "h2_log_hype":
                h2_result["log_hype"],

            "event_avg_baseline":
                event_avg_baseline,

            "event_avg_fight_week":
                event_avg_fight_week,

            "event_avg_log_hype":
                event_avg_log_hype
        })

        print(
            f"  {h1}: "
            f"baseline={h1_result['baseline']:.2f} | "
            f"week={h1_result['fight_week']:.2f} | "
            f"log={h1_result['log_hype']:.3f}"
        )

        print(
            f"  {h2}: "
            f"baseline={h2_result['baseline']:.2f} | "
            f"week={h2_result['fight_week']:.2f} | "
            f"log={h2_result['log_hype']:.3f}"
        )

        print(
            f"  EVENT LOG HYPE: "
            f"{event_avg_log_hype:.3f}"
        )

    except Exception as e:

        print(
            "  ERROR:",
            type(e).__name__,
            str(e)
        )


# ------------------------------------------------------------
# 7. Create dataframe
# ------------------------------------------------------------

new_event_trends = pd.DataFrame(
    new_trends_rows
)


# ------------------------------------------------------------
# 8. Validation
# ------------------------------------------------------------

print()
print("=" * 100)
print("NEW TRENDS CALCULATION COMPLETE")
print()

print(
    "Successful events:",
    len(new_event_trends),
    "/ 11"
)

if len(new_event_trends) > 0:

    print(
        "Missing event hype scores:",
        new_event_trends[
            "event_avg_log_hype"
        ].isna().sum()
    )

    print()
    print(
        new_event_trends[
            [
                "event_date",
                "event_name",
                "headliner_1",
                "headliner_2",
                "event_avg_log_hype"
            ]
        ].to_string(index=False)
    )
```

```python
# ============================================================
# STEP 73B — BUILD FANFARE EVENT BACKBONE
# ============================================================

fanfare_events = (
    ufc_fights_updated[
        ["date", "event_name"]
    ]
    .drop_duplicates()
    .sort_values(
        ["date", "event_name"]
    )
    .reset_index(drop=True)
)

fanfare_events["date"] = pd.to_datetime(
    fanfare_events["date"]
)


# ------------------------------------------------------------
# Basic event era
# ------------------------------------------------------------

def fanfare_era(date):

    year = date.year

    if year < 2001:
        return "1994-2000"

    elif year < 2011:
        return "2001-2010"

    elif year < 2014:
        return "2011-2013"

    elif year < 2019:
        return "2014-2018"

    else:
        return "2019-present"


fanfare_events["fanfare_era"] = (
    fanfare_events["date"]
    .apply(fanfare_era)
)


# ------------------------------------------------------------
# Add empty fields we'll eventually populate
# ------------------------------------------------------------

fanfare_events["attendance"] = np.nan
fanfare_events["gate_usd"] = np.nan

fanfare_events["ppv_buys"] = np.nan
fanfare_events["viewership"] = np.nan

fanfare_events["absolute_search_demand"] = np.nan

fanfare_events["search_hype_acceleration"] = np.nan


# ------------------------------------------------------------
# Merge existing Trends acceleration
# ------------------------------------------------------------

existing_trends = event_trends_results[
    [
        "event_date",
        "event_name",
        "event_avg_log_hype"
    ]
].copy()

existing_trends = existing_trends.rename(
    columns={
        "event_date": "date",
        "event_avg_log_hype":
            "search_hype_acceleration"
    }
)

existing_trends["date"] = pd.to_datetime(
    existing_trends["date"]
)


fanfare_events = fanfare_events.drop(
    columns=["search_hype_acceleration"]
).merge(
    existing_trends,
    on=["date", "event_name"],
    how="left"
)


# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

print("FANFARE BACKBONE CREATED 🔥")
print("=" * 100)

print(
    "Total events:",
    len(fanfare_events)
)

print(
    "Earliest:",
    fanfare_events["date"].min()
)

print(
    "Latest:",
    fanfare_events["date"].max()
)

print(
    "Existing Trends events:",
    fanfare_events[
        "search_hype_acceleration"
    ].notna().sum()
)

print()
print("EVENTS BY ERA:")
print(
    fanfare_events[
        "fanfare_era"
    ].value_counts().sort_index()
)

print()
print("Newest 15 events:")
print(
    fanfare_events[
        [
            "date",
            "event_name",
            "search_hype_acceleration"
        ]
    ]
    .tail(15)
    .to_string(index=False)
)
```

```python
# ============================================================
# STEP 76 — PATCH VERIFIED 2026 GATES + PREP BULK GATE BUILD
# ============================================================

import pandas as pd
import numpy as np


# ============================================================
# PART 1 — PATCH MORE PRECISE / NEW OFFICIAL VALUES
# ============================================================

gate_patches = [
    {
        "date": pd.Timestamp("2026-07-11"),
        "gate": 26430566,
        "source": "Reported post-event UFC 329 gate",
        "quality": "official/reported",
        "notes": "UFC 329 record gate; precise reported figure"
    },

    {
        "date": pd.Timestamp("2026-08-15"),
        "gate": 7000000,
        "source": "UFC.com post-event Dana White comments",
        "quality": "official",
        "notes": "Dana White stated UFC 330 did a $7M gate"
    }
]


for patch in gate_patches:

    mask = (
        gate_database["date"]
        == patch["date"]
    )

    if mask.sum() == 1:

        gate_database.loc[
            mask,
            "gate_usd_nominal"
        ] = patch["gate"]

        gate_database.loc[
            mask,
            "gate_source"
        ] = patch["source"]

        gate_database.loc[
            mask,
            "gate_source_quality"
        ] = patch["quality"]

        gate_database.loc[
            mask,
            "gate_notes"
        ] = patch["notes"]

    else:

        print(
            "PATCH WARNING:",
            patch["date"],
            "matched",
            mask.sum(),
            "events"
        )


# ============================================================
# PART 2 — REFRESH AVAILABILITY FLAGS
# ============================================================

gate_database["gate_available"] = (
    gate_database[
        "gate_usd_nominal"
    ].notna()
)

gate_database["gate_missing"] = (
    ~gate_database["gate_available"]
)


# ============================================================
# PART 3 — CREATE A STABLE EVENT ID
#
# This becomes our matching key during bulk collection.
# We NEVER depend solely on exact event-name spelling.
# ============================================================

gate_database["gate_event_id"] = (
    gate_database["date"]
    .dt.strftime("%Y-%m-%d")
    + "__"
    + gate_database["event_name"]
      .astype(str)
      .str.lower()
      .str.replace(
          r"[^a-z0-9]+",
          "_",
          regex=True
      )
      .str.strip("_")
)


# ============================================================
# PART 4 — IDENTIFY EVENT TYPE
#
# This will later help us understand gate availability.
# Fight Nights / APEX cards often have fundamentally
# different gate situations from arena PPVs.
# ============================================================

def classify_gate_event(name):

    name_lower = str(name).lower()

    if "ufc fight night" in name_lower:
        return "Fight Night"

    if "noche ufc" in name_lower:
        return "Noche UFC"

    if "ufc on espn" in name_lower:
        return "UFC on ESPN"

    if "ufc on fox" in name_lower:
        return "UFC on FOX"

    if "ufc on fuel" in name_lower:
        return "UFC on Fuel"

    if "ufc on fx" in name_lower:
        return "UFC on FX"

    if "ultimate fighter" in name_lower:
        return "TUF Finale"

    if (
        name_lower.startswith("ufc ")
        and any(
            char.isdigit()
            for char in name_lower[:10]
        )
    ):
        return "Numbered UFC"

    # Handles Crypto.com UFC 331 etc.
    if (
        "ufc " in name_lower
        and any(
            char.isdigit()
            for char in name_lower
        )
    ):
        return "Numbered UFC"

    return "Other"


gate_database["gate_event_type"] = (
    gate_database[
        "event_name"
    ].apply(classify_gate_event)
)


# ============================================================
# PART 5 — CREATE SAFE GATE TRANSFORM
#
# DO NOT normalize yet.
# DO NOT turn NaN into zero.
#
# log1p will reduce the extreme skew between a $500k gate
# and a $26M gate when we eventually score it.
# ============================================================

gate_database["gate_log_nominal"] = np.where(
    gate_database["gate_available"],
    np.log1p(
        gate_database[
            "gate_usd_nominal"
        ]
    ),
    np.nan
)


# ============================================================
# PART 6 — NEW 11-EVENT AUDIT
# ============================================================

new_gate_audit = (
    gate_database[
        gate_database["date"]
        > pd.Timestamp("2026-06-27")
    ]
    [
        [
            "date",
            "event_name",
            "gate_event_type",
            "gate_usd_nominal",
            "gate_available",
            "gate_source_quality"
        ]
    ]
    .copy()
)


# ============================================================
# PART 7 — CURRENT KNOWN GATES
# ============================================================

known_gate_table = (
    gate_database[
        gate_database["gate_available"]
    ]
    [
        [
            "date",
            "event_name",
            "gate_event_type",
            "gate_usd_nominal",
            "gate_source_quality"
        ]
    ]
    .sort_values(
        "gate_usd_nominal",
        ascending=False
    )
)


# ============================================================
# PART 8 — COVERAGE BY EVENT TYPE
# ============================================================

coverage_by_type = (
    gate_database
    .groupby("gate_event_type")
    .agg(
        events=("event_name", "size"),
        known_gates=("gate_available", "sum")
    )
)

coverage_by_type[
    "coverage_pct"
] = (
    coverage_by_type[
        "known_gates"
    ]
    /
    coverage_by_type[
        "events"
    ]
    * 100
).round(1)


# ============================================================
# PART 9 — HARD INTEGRITY CHECK
# ============================================================

total_events = len(gate_database)

unique_events = (
    gate_database[
        ["date", "event_name"]
    ]
    .drop_duplicates()
    .shape[0]
)

new_events = len(
    new_gate_audit
)

missing_treated_as_zero = (
    gate_database.loc[
        gate_database["gate_missing"],
        "gate_usd_nominal"
    ]
    == 0
).sum()


# ============================================================
# OUTPUT
# ============================================================

print("STEP 76 COMPLETE — GATE SYSTEM READY 🔥")
print("=" * 110)

print()
print("DATABASE INTEGRITY")
print("-" * 110)

print("Total events:", total_events)
print("Unique events:", unique_events)
print("New events:", new_events)

print(
    "Missing gates incorrectly stored as $0:",
    missing_treated_as_zero
)


print()
print("=" * 110)
print("KNOWN VERIFIED GATES")
print("-" * 110)

print(
    known_gate_table.to_string(
        index=False,
        formatters={
            "gate_usd_nominal":
                lambda x: f"${x:,.0f}"
        }
    )
)


print()
print("=" * 110)
print("ALL 11 NEW EVENTS")
print("-" * 110)

print(
    new_gate_audit.to_string(
        index=False,
        formatters={
            "gate_usd_nominal":
                lambda x:
                "MISSING"
                if pd.isna(x)
                else f"${x:,.0f}"
        }
    )
)


print()
print("=" * 110)
print("CURRENT COVERAGE BY EVENT TYPE")
print("-" * 110)

print(
    coverage_by_type.to_string()
)


print()
print("=" * 110)

if (
    total_events == 788
    and unique_events == 788
    and new_events == 11
    and missing_treated_as_zero == 0
):

    print(
        "GATE FRAMEWORK: PASS ✅🔥"
    )

else:

    print(
        "GATE FRAMEWORK: CHECK REQUIRED ❌"
    )
```

```python
# ============================================================
# STEP 77 — BULK HISTORICAL UFC GATES + INFLATION ADJUSTMENT
# ============================================================

import pandas as pd
import numpy as np
import requests
import re
from bs4 import BeautifulSoup


# ============================================================
# PART 1 — SAFETY COPY
# ============================================================

gate_database_pre_bulk = gate_database.copy()


# ============================================================
# PART 2 — DOWNLOAD UFC RECORDS PAGE
#
# We are using this as ONE bulk source, not as unquestioned
# truth. Official/reported values already in gate_database
# will NEVER be overwritten by lower-quality values.
# ============================================================

url = "https://learnsth.icu/wiki/List_of_UFC_records"

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/124.0 Safari/537.36"
    )
}

print("DOWNLOADING HISTORICAL UFC GATE DATA 🔥")
print("=" * 110)

response = requests.get(
    url,
    headers=headers,
    timeout=30
)

print("HTTP status:", response.status_code)

response.raise_for_status()

soup = BeautifulSoup(
    response.text,
    "html.parser"
)


# ============================================================
# PART 3 — FIND TABLES CONTAINING EVENT + GATE
# ============================================================

gate_tables = []

for i, table in enumerate(
    soup.find_all("table")
):

    try:

        df = pd.read_html(
            str(table)
        )[0]

        # Flatten columns
        if isinstance(
            df.columns,
            pd.MultiIndex
        ):

            df.columns = [
                str(c[-1]).strip()
                for c in df.columns
            ]

        else:

            df.columns = [
                str(c).strip()
                for c in df.columns
            ]

        column_names = [
            c.lower()
            for c in df.columns
        ]

        has_event = any(
            "event" == c
            or "event" in c
            for c in column_names
        )

        has_gate = any(
            "gate" in c
            for c in column_names
        )

        if has_event and has_gate:

            gate_tables.append(
                (i, df)
            )

            print(
                "Gate table found:",
                i,
                "| rows:",
                len(df),
                "| columns:",
                list(df.columns)
            )

    except Exception:
        continue


print()
print(
    "Gate tables discovered:",
    len(gate_tables)
)


# ============================================================
# PART 4 — EXTRACT EVENT / DATE / GATE RECORDS
# ============================================================

bulk_gate_rows = []


def money_to_float(value):

    if pd.isna(value):
        return np.nan

    text = str(value)

    # Remove citation markers etc.
    text = re.sub(
        r"\[[^\]]*\]",
        "",
        text
    )

    text = text.replace(",", "")

    match = re.search(
        r"\$?\s*([0-9]+(?:\.[0-9]+)?)",
        text
    )

    if not match:
        return np.nan

    number = float(
        match.group(1)
    )

    lower = text.lower()

    if "billion" in lower:
        number *= 1_000_000_000

    elif "million" in lower:
        number *= 1_000_000

    return number


for table_number, df in gate_tables:

    event_col = next(
        (
            c for c in df.columns
            if "event" in c.lower()
        ),
        None
    )

    date_col = next(
        (
            c for c in df.columns
            if c.lower() == "date"
            or "date" in c.lower()
        ),
        None
    )

    # Prefer raw Gate, NOT Inflation Adjusted
    gate_col = next(
        (
            c for c in df.columns
            if (
                "gate" in c.lower()
                and "inflation" not in c.lower()
                and "adjust" not in c.lower()
            )
        ),
        None
    )

    inflation_col = next(
        (
            c for c in df.columns
            if (
                "inflation" in c.lower()
                or "adjust" in c.lower()
            )
        ),
        None
    )

    if (
        event_col is None
        or gate_col is None
    ):
        continue

    for _, row in df.iterrows():

        event_text = str(
            row[event_col]
        ).strip()

        gate_value = money_to_float(
            row[gate_col]
        )

        if (
            not event_text
            or event_text.lower() == "nan"
            or pd.isna(gate_value)
        ):
            continue

        parsed_date = pd.NaT

        if date_col is not None:

            parsed_date = pd.to_datetime(
                row[date_col],
                errors="coerce"
            )

        inflation_value = np.nan

        if inflation_col is not None:

            inflation_value = money_to_float(
                row[inflation_col]
            )

        bulk_gate_rows.append({
            "source_event": event_text,
            "source_date": parsed_date,
            "gate_usd_nominal": gate_value,
            "gate_usd_adjusted_source":
                inflation_value,
            "source_table": table_number
        })


bulk_gates = pd.DataFrame(
    bulk_gate_rows
)


# ============================================================
# PART 5 — REMOVE DUPLICATES
# ============================================================

if len(bulk_gates):

    bulk_gates = (
        bulk_gates
        .drop_duplicates(
            subset=[
                "source_event",
                "source_date",
                "gate_usd_nominal"
            ]
        )
        .reset_index(drop=True)
    )


print()
print("=" * 110)

print(
    "Usable bulk gate records:",
    len(bulk_gates)
)


# ============================================================
# PART 6 — MATCH BULK RECORDS TO OUR 788 EVENTS
#
# Priority:
# 1. Exact event date
# 2. UFC number where available
#
# Existing official values NEVER get overwritten.
# ============================================================

def extract_ufc_number(name):

    text = str(name)

    match = re.search(
        r"\bUFC\s+(\d+)\b",
        text,
        flags=re.I
    )

    if match:
        return int(
            match.group(1)
        )

    return np.nan


gate_database[
    "ufc_number"
] = (
    gate_database[
        "event_name"
    ]
    .apply(extract_ufc_number)
)


if len(bulk_gates):

    bulk_gates[
        "ufc_number"
    ] = (
        bulk_gates[
            "source_event"
        ]
        .apply(extract_ufc_number)
    )


matched_bulk = 0
skipped_existing = 0
unmatched_bulk = []


for _, row in bulk_gates.iterrows():

    match_mask = pd.Series(
        False,
        index=gate_database.index
    )


    # --------------------------------------------------------
    # DATE MATCH — safest
    # --------------------------------------------------------

    if pd.notna(
        row["source_date"]
    ):

        date_matches = (
            gate_database["date"]
            == row["source_date"]
        )

        if date_matches.sum() == 1:

            match_mask = date_matches


    # --------------------------------------------------------
    # UFC NUMBER FALLBACK
    # --------------------------------------------------------

    if (
        match_mask.sum() == 0
        and pd.notna(
            row["ufc_number"]
        )
    ):

        number_matches = (
            gate_database[
                "ufc_number"
            ]
            == row["ufc_number"]
        )

        if number_matches.sum() == 1:

            match_mask = number_matches


    # --------------------------------------------------------
    # APPLY
    # --------------------------------------------------------

    if match_mask.sum() == 1:

        idx = gate_database[
            match_mask
        ].index[0]

        if pd.notna(
            gate_database.loc[
                idx,
                "gate_usd_nominal"
            ]
        ):

            skipped_existing += 1

        else:

            gate_database.loc[
                idx,
                "gate_usd_nominal"
            ] = row[
                "gate_usd_nominal"
            ]

            gate_database.loc[
                idx,
                "gate_source"
            ] = (
                "UFC records historical gate table"
            )

            gate_database.loc[
                idx,
                "gate_source_quality"
            ] = (
                "secondary/reported"
            )

            gate_database.loc[
                idx,
                "gate_notes"
            ] = (
                "Bulk historical gate record"
            )

            # Preserve source-provided inflation-adjusted
            # figure if the table has one.
            if (
                "gate_usd_adjusted_source"
                not in gate_database.columns
            ):

                gate_database[
                    "gate_usd_adjusted_source"
                ] = np.nan

            gate_database.loc[
                idx,
                "gate_usd_adjusted_source"
            ] = row[
                "gate_usd_adjusted_source"
            ]

            matched_bulk += 1

    else:

        unmatched_bulk.append(
            row["source_event"]
        )


# ============================================================
# PART 7 — MAKE SURE ADJUSTED COLUMN EXISTS
# ============================================================

if (
    "gate_usd_adjusted_source"
    not in gate_database.columns
):

    gate_database[
        "gate_usd_adjusted_source"
    ] = np.nan


# ============================================================
# PART 8 — PATCH OFFICIAL VALIDATION VALUES
#
# Higher-quality official UFC values win.
# ============================================================

official_validation_patches = [

    # UFC 281
    {
        "date": "2022-11-12",
        "gate": 11562807
    },

    # UFC 290
    {
        "date": "2023-07-08",
        "gate": 9750000
    },

    # UFC 295
    {
        "date": "2023-11-11",
        "gate": 12432563
    },

    # UFC 300
    {
        "date": "2024-04-13",
        "gate": 16500000
    },

    # UFC 329
    {
        "date": "2026-07-11",
        "gate": 26430566
    },

    # UFC 330
    {
        "date": "2026-08-15",
        "gate": 7000000
    }
]


for patch in official_validation_patches:

    patch_date = pd.Timestamp(
        patch["date"]
    )

    mask = (
        gate_database["date"]
        == patch_date
    )

    if mask.sum() == 1:

        gate_database.loc[
            mask,
            "gate_usd_nominal"
        ] = patch["gate"]

        gate_database.loc[
            mask,
            "gate_source_quality"
        ] = "official"

        gate_database.loc[
            mask,
            "gate_source"
        ] = "UFC official reporting"


# ============================================================
# PART 9 — REFRESH AVAILABILITY
# ============================================================

gate_database[
    "gate_available"
] = (
    gate_database[
        "gate_usd_nominal"
    ].notna()
)

gate_database[
    "gate_missing"
] = (
    ~gate_database[
        "gate_available"
    ]
)


# ============================================================
# PART 10 — INFLATION-ADJUSTED GATE
#
# First use a source-provided adjusted value if available.
#
# We are NOT fabricating CPI values here. Events without a
# reliable adjusted figure stay NaN until the CPI step.
# ============================================================

gate_database[
    "gate_usd_2026"
] = (
    gate_database[
        "gate_usd_adjusted_source"
    ]
)


# 2026 events need no historical adjustment.
mask_2026 = (
    gate_database["date"].dt.year
    == 2026
) & (
    gate_database[
        "gate_available"
    ]
)

gate_database.loc[
    mask_2026,
    "gate_usd_2026"
] = (
    gate_database.loc[
        mask_2026,
        "gate_usd_nominal"
    ]
)


# ============================================================
# PART 11 — LOG TRANSFORMS
# ============================================================

gate_database[
    "gate_log_nominal"
] = np.where(
    gate_database[
        "gate_available"
    ],
    np.log1p(
        gate_database[
            "gate_usd_nominal"
        ]
    ),
    np.nan
)

gate_database[
    "gate_log_2026"
] = np.where(
    gate_database[
        "gate_usd_2026"
    ].notna(),
    np.log1p(
        gate_database[
            "gate_usd_2026"
        ]
    ),
    np.nan
)


# ============================================================
# PART 12 — COVERAGE BY ERA
# ============================================================

coverage_by_era = (
    gate_database
    .groupby(
        "fanfare_era"
    )
    .agg(
        events=(
            "event_name",
            "size"
        ),
        gates_known=(
            "gate_available",
            "sum"
        )
    )
)

coverage_by_era[
    "coverage_pct"
] = (
    coverage_by_era[
        "gates_known"
    ]
    /
    coverage_by_era[
        "events"
    ]
    * 100
).round(1)


# ============================================================
# PART 13 — COVERAGE BY EVENT TYPE
# ============================================================

coverage_by_type = (
    gate_database
    .groupby(
        "gate_event_type"
    )
    .agg(
        events=(
            "event_name",
            "size"
        ),
        gates_known=(
            "gate_available",
            "sum"
        )
    )
)

coverage_by_type[
    "coverage_pct"
] = (
    coverage_by_type[
        "gates_known"
    ]
    /
    coverage_by_type[
        "events"
    ]
    * 100
).round(1)


# ============================================================
# PART 14 — TOP NOMINAL GATES
# ============================================================

top_nominal_gates = (
    gate_database[
        gate_database[
            "gate_available"
        ]
    ]
    .sort_values(
        "gate_usd_nominal",
        ascending=False
    )
    [
        [
            "date",
            "event_name",
            "gate_usd_nominal",
            "gate_source_quality"
        ]
    ]
    .head(25)
)


# ============================================================
# PART 15 — ALL 11 NEW EVENTS
# ============================================================

new_gate_audit = (
    gate_database[
        gate_database["date"]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "date",
            "event_name",
            "gate_usd_nominal",
            "gate_available",
            "gate_source_quality"
        ]
    ]
)


# ============================================================
# PART 16 — INTEGRITY CHECKS
# ============================================================

total_events = len(
    gate_database
)

unique_events = (
    gate_database[
        ["date", "event_name"]
    ]
    .drop_duplicates()
    .shape[0]
)

duplicate_events = (
    gate_database
    .duplicated(
        ["date", "event_name"]
    )
    .sum()
)

new_events = len(
    new_gate_audit
)


# ============================================================
# OUTPUT
# ============================================================

print()
print("=" * 110)
print("STEP 77 — BULK GATE BUILD COMPLETE 🔥")
print("=" * 110)

print()
print("BULK COLLECTION")
print("-" * 110)

print(
    "Bulk source records:",
    len(bulk_gates)
)

print(
    "New gates matched:",
    matched_bulk
)

print(
    "Existing higher-quality gates preserved:",
    skipped_existing
)

print(
    "Bulk records unmatched:",
    len(unmatched_bulk)
)


print()
print("=" * 110)
print("TOTAL COVERAGE")
print("-" * 110)

print(
    "Total events:",
    total_events
)

print(
    "Events with known gate:",
    int(
        gate_database[
            "gate_available"
        ].sum()
    )
)

print(
    "Events with missing gate:",
    int(
        gate_database[
            "gate_missing"
        ].sum()
    )
)

print(
    "Gate coverage:",
    round(
        gate_database[
            "gate_available"
        ].mean()
        * 100,
        1
    ),
    "%"
)


print()
print("=" * 110)
print("COVERAGE BY ERA")
print("-" * 110)

print(
    coverage_by_era.to_string()
)


print()
print("=" * 110)
print("COVERAGE BY EVENT TYPE")
print("-" * 110)

print(
    coverage_by_type.to_string()
)


print()
print("=" * 110)
print("TOP 25 NOMINAL UFC GATES")
print("-" * 110)

print(
    top_nominal_gates
    .to_string(
        index=False,
        formatters={
            "gate_usd_nominal":
                lambda x:
                f"${x:,.0f}"
        }
    )
)


print()
print("=" * 110)
print("ALL 11 NEW EVENTS")
print("-" * 110)

print(
    new_gate_audit
    .to_string(
        index=False,
        formatters={
            "gate_usd_nominal":
                lambda x:
                "MISSING"
                if pd.isna(x)
                else f"${x:,.0f}"
        }
    )
)


print()
print("=" * 110)
print("INTEGRITY")
print("-" * 110)

print(
    "Total events:",
    total_events
)

print(
    "Unique events:",
    unique_events
)

print(
    "Duplicates:",
    duplicate_events
)

print(
    "New events retained:",
    new_events
)


print()
print("=" * 110)

if (
    total_events == 788
    and unique_events == 788
    and duplicate_events == 0
    and new_events == 11
):

    print(
        "BULK GATE DATABASE: PASS ✅🔥"
    )

else:

    print(
        "BULK GATE DATABASE: CHECK REQUIRED ❌"
    )
```

```python
# ============================================================
# STEP 78 — COMMON-SCALE SEARCH DEMAND PROOF OF CONCEPT
# ============================================================

import pandas as pd
import numpy as np
import time
from pytrends.request import TrendReq


# ============================================================
# PART 1 — CONNECT TO GOOGLE TRENDS
# ============================================================

pytrends = TrendReq(
    hl="en-US",
    tz=300,
    retries=2,
    backoff_factor=0.5
)


# ============================================================
# PART 2 — HELPER TO GET STORED GOOGLE TOPIC
# ============================================================

def get_topic(fighter):

    match = fighter_topics[
        fighter_topics["fighter"]
        .str.lower()
        == fighter.lower()
    ]

    if len(match) == 0:
        return None

    topic = match.iloc[0]["topic_id"]

    if pd.isna(topic):
        return None

    return topic


# ============================================================
# PART 3 — TEST FIGHTERS
#
# Deliberately includes:
# - enormous mainstream stars
# - major UFC stars
# - solid recognizable fighters
# - lower-profile fighters
#
# This lets us sanity-check the common scale.
# ============================================================

test_fighters = [
    "Conor McGregor",
    "Jon Jones",
    "Khabib Nurmagomedov",
    "Israel Adesanya",
    "Max Holloway",
    "Sean O'Malley",
    "Islam Makhachev",
    "Uros Medic"
]


fighter_test_topics = {}

for fighter in test_fighters:

    topic = get_topic(fighter)

    fighter_test_topics[
        fighter
    ] = topic


print("TOPIC CHECK")
print("=" * 90)

for fighter, topic in fighter_test_topics.items():

    print(
        f"{fighter:25s}",
        "→",
        topic
    )


# ============================================================
# PART 4 — REMOVE ANY MISSING TOPICS
# ============================================================

valid_test = {
    fighter: topic
    for fighter, topic
    in fighter_test_topics.items()
    if topic is not None
}


print()
print(
    "Valid fighters:",
    len(valid_test),
    "/",
    len(test_fighters)
)


# ============================================================
# PART 5 — RUN OVERLAPPING COMPARISON GROUPS
#
# Google Trends comparisons have request-size limits.
#
# CONOR is our common anchor.
#
# Because Conor appears in every group, every other fighter
# can be expressed relative to Conor.
#
# THIS is fundamentally different from our old independent
# requests.
# ============================================================

anchor_fighter = "Conor McGregor"

anchor_topic = valid_test[
    anchor_fighter
]


other_fighters = [
    fighter
    for fighter in valid_test
    if fighter != anchor_fighter
]


comparison_results = []


# 4 other fighters + Conor = 5 topics/request
group_size = 4


for start in range(
    0,
    len(other_fighters),
    group_size
):

    group_fighters = (
        [anchor_fighter]
        +
        other_fighters[
            start:start + group_size
        ]
    )

    group_topics = [
        valid_test[f]
        for f in group_fighters
    ]


    print()
    print("=" * 90)

    print(
        "COMPARING:",
        " | ".join(group_fighters)
    )


    try:

        pytrends.build_payload(
            group_topics,
            timeframe="2025-01-01 2026-09-20",
            geo="US"
        )

        trends = (
            pytrends
            .interest_over_time()
        )


        if trends.empty:

            print(
                "No Trends data returned."
            )

            continue


        if "isPartial" in trends.columns:

            trends = trends.drop(
                columns=["isPartial"]
            )


        # -----------------------------------------------
        # Average search interest over entire timeframe
        # -----------------------------------------------

        averages = trends.mean()


        anchor_average = averages[
            anchor_topic
        ]


        for fighter, topic in zip(
            group_fighters,
            group_topics
        ):

            fighter_average = averages[
                topic
            ]


            if anchor_average > 0:

                relative_to_conor = (
                    fighter_average
                    /
                    anchor_average
                )

            else:

                relative_to_conor = np.nan


            comparison_results.append({

                "fighter":
                    fighter,

                "topic_id":
                    topic,

                "mean_interest":
                    fighter_average,

                "conor_mean":
                    anchor_average,

                "relative_to_conor":
                    relative_to_conor

            })


        print(
            pd.DataFrame({
                "fighter":
                    group_fighters,

                "mean_interest":
                    [
                        averages[t]
                        for t in group_topics
                    ]
            })
            .sort_values(
                "mean_interest",
                ascending=False
            )
            .to_string(
                index=False
            )
        )


    except Exception as e:

        print(
            "ERROR:",
            type(e).__name__,
            str(e)
        )


    time.sleep(3)


# ============================================================
# PART 6 — COMBINE RESULTS
# ============================================================

search_demand_test = pd.DataFrame(
    comparison_results
)


# Conor appears once per comparison group.
# Collapse duplicate Conor rows.
search_demand_test = (
    search_demand_test
    .groupby(
        ["fighter", "topic_id"],
        as_index=False
    )
    .agg(
        mean_interest=(
            "mean_interest",
            "mean"
        ),

        conor_mean=(
            "conor_mean",
            "mean"
        ),

        relative_to_conor=(
            "relative_to_conor",
            "mean"
        )
    )
)


# ============================================================
# PART 7 — CREATE EASY-TO-READ INDEX
#
# Conor = 100
#
# Someone with half Conor's calibrated demand = 50.
# Someone with 10% = 10.
#
# This is NOT yet the final fanfare score.
# ============================================================

search_demand_test[
    "search_demand_index"
] = (
    search_demand_test[
        "relative_to_conor"
    ]
    * 100
)


search_demand_test = (
    search_demand_test
    .sort_values(
        "search_demand_index",
        ascending=False
    )
    .reset_index(drop=True)
)


# ============================================================
# PART 8 — OUTPUT
# ============================================================

print()
print("=" * 90)
print(
    "COMMON-SCALE SEARCH DEMAND TEST 🔥"
)
print("=" * 90)

print(
    search_demand_test[
        [
            "fighter",
            "mean_interest",
            "relative_to_conor",
            "search_demand_index"
        ]
    ]
    .to_string(
        index=False,
        formatters={

            "relative_to_conor":
                lambda x:
                f"{x:.4f}",

            "search_demand_index":
                lambda x:
                f"{x:.2f}"
        }
    )
)


print()
print("=" * 90)

if len(search_demand_test) >= 5:

    print(
        "COMMON-SCALE TEST: COMPLETE ✅🔥"
    )

else:

    print(
        "COMMON-SCALE TEST: INCOMPLETE — CHECK OUTPUT"
    )
```

```python
# ============================================================
# STEP 78B — FIX PYTRENDS + COMMON-SCALE SEARCH DEMAND TEST
# ============================================================

import pandas as pd
import numpy as np
import time

# ------------------------------------------------------------
# PART 1 — FIX COLAB / PYTRENDS COMPATIBILITY
# ------------------------------------------------------------

# Important:
# retries=0 avoids the outdated urllib3 "method_whitelist"
# code path that caused Step 78 to fail.

from pytrends.request import TrendReq

pytrends = TrendReq(
    hl="en-US",
    tz=300,
    timeout=(10, 30),
    retries=0,
    backoff_factor=0
)

print("PYTRENDS CONNECTION OBJECT CREATED ✅")


# ------------------------------------------------------------
# PART 2 — GET TOPIC IDs FROM OUR EXISTING TOPIC DATABASE
# ------------------------------------------------------------

def get_topic(fighter):

    match = fighter_topics[
        fighter_topics["fighter"]
        .astype(str)
        .str.lower()
        == fighter.lower()
    ]

    if len(match) == 0:
        return None

    topic = match.iloc[0]["topic_id"]

    if pd.isna(topic):
        return None

    return str(topic)


test_fighters = [
    "Conor McGregor",
    "Jon Jones",
    "Khabib Nurmagomedov",
    "Israel Adesanya",
    "Max Holloway",
    "Sean O'Malley",
    "Islam Makhachev",
    "Uros Medic"
]

fighter_test_topics = {
    fighter: get_topic(fighter)
    for fighter in test_fighters
}


print()
print("=" * 100)
print("TOPIC CHECK")
print("=" * 100)

for fighter, topic in fighter_test_topics.items():
    print(f"{fighter:25s} → {topic}")


valid_test = {
    fighter: topic
    for fighter, topic in fighter_test_topics.items()
    if topic is not None
}

print()
print(
    "Valid fighters:",
    len(valid_test),
    "/",
    len(test_fighters)
)


# ------------------------------------------------------------
# PART 3 — COMMON ANCHOR SETUP
# ------------------------------------------------------------

anchor_fighter = "Conor McGregor"
anchor_topic = valid_test[anchor_fighter]

other_fighters = [
    fighter
    for fighter in valid_test
    if fighter != anchor_fighter
]

comparison_results = []

# Conor + up to 4 others = Google's 5-item comparison limit
group_size = 4


# ------------------------------------------------------------
# PART 4 — RUN OVERLAPPING GOOGLE TRENDS REQUESTS
# ------------------------------------------------------------

for start in range(
    0,
    len(other_fighters),
    group_size
):

    group_fighters = (
        [anchor_fighter]
        + other_fighters[start:start + group_size]
    )

    group_topics = [
        valid_test[fighter]
        for fighter in group_fighters
    ]

    print()
    print("=" * 100)
    print(
        "COMPARING:",
        " | ".join(group_fighters)
    )
    print("=" * 100)

    try:

        pytrends.build_payload(
            kw_list=group_topics,
            cat=0,
            timeframe="2025-01-01 2026-09-20",
            geo="US",
            gprop=""
        )

        trends = pytrends.interest_over_time()

        if trends.empty:

            print("❌ Google returned an empty table.")
            continue

        if "isPartial" in trends.columns:
            trends = trends.drop(
                columns=["isPartial"]
            )

        print(
            "Rows returned:",
            len(trends)
        )

        averages = trends.mean()

        anchor_average = float(
            averages[anchor_topic]
        )

        print(
            "Conor mean interest:",
            round(anchor_average, 3)
        )

        print()
        print("GROUP RESULTS:")

        group_output = []

        for fighter, topic in zip(
            group_fighters,
            group_topics
        ):

            fighter_average = float(
                averages[topic]
            )

            if anchor_average > 0:

                relative_to_conor = (
                    fighter_average
                    / anchor_average
                )

            else:

                relative_to_conor = np.nan

            comparison_results.append({
                "fighter": fighter,
                "topic_id": topic,
                "mean_interest": fighter_average,
                "conor_mean": anchor_average,
                "relative_to_conor": relative_to_conor
            })

            group_output.append({
                "fighter": fighter,
                "mean_interest": fighter_average,
                "relative_to_conor": relative_to_conor
            })

        group_output = (
            pd.DataFrame(group_output)
            .sort_values(
                "mean_interest",
                ascending=False
            )
        )

        print(
            group_output.to_string(
                index=False,
                formatters={
                    "mean_interest":
                        lambda x: f"{x:.3f}",
                    "relative_to_conor":
                        lambda x: f"{x:.4f}"
                }
            )
        )

    except Exception as e:

        print()
        print("❌ REQUEST FAILED")
        print("Error type:", type(e).__name__)
        print("Error:", str(e))

    # Small pause so we don't hammer Trends.
    time.sleep(5)


# ------------------------------------------------------------
# PART 5 — SAFE RESULT BUILD
#
# This prevents the second error you got.
# If Google fails, the cell reports that instead of crashing.
# ------------------------------------------------------------

print()
print("=" * 100)
print("BUILDING FINAL TEST TABLE")
print("=" * 100)

if len(comparison_results) == 0:

    search_demand_test = pd.DataFrame(
        columns=[
            "fighter",
            "topic_id",
            "mean_interest",
            "conor_mean",
            "relative_to_conor",
            "search_demand_index"
        ]
    )

    print()
    print("❌ NO COMPARISON DATA WAS RETURNED.")
    print()
    print(
        "The pytrends compatibility crash is fixed, "
        "but Google Trends itself did not return usable data."
    )

else:

    search_demand_test = pd.DataFrame(
        comparison_results
    )

    # Conor appears in both groups.
    # Collapse his duplicate observations.
    search_demand_test = (
        search_demand_test
        .groupby(
            ["fighter", "topic_id"],
            as_index=False
        )
        .agg(
            mean_interest=(
                "mean_interest",
                "mean"
            ),
            conor_mean=(
                "conor_mean",
                "mean"
            ),
            relative_to_conor=(
                "relative_to_conor",
                "mean"
            )
        )
    )

    # Conor = 100
    search_demand_test[
        "search_demand_index"
    ] = (
        search_demand_test[
            "relative_to_conor"
        ]
        * 100
    )

    search_demand_test = (
        search_demand_test
        .sort_values(
            "search_demand_index",
            ascending=False
        )
        .reset_index(drop=True)
    )

    print()
    print(
        search_demand_test[
            [
                "fighter",
                "mean_interest",
                "relative_to_conor",
                "search_demand_index"
            ]
        ]
        .to_string(
            index=False,
            formatters={
                "mean_interest":
                    lambda x: f"{x:.3f}",
                "relative_to_conor":
                    lambda x: f"{x:.4f}",
                "search_demand_index":
                    lambda x: f"{x:.2f}"
            }
        )
    )


# ------------------------------------------------------------
# PART 6 — AUTOMATIC VALIDATION
# ------------------------------------------------------------

print()
print("=" * 100)

returned_fighters = (
    search_demand_test["fighter"].nunique()
    if len(search_demand_test)
    else 0
)

print(
    "Fighters requested:",
    len(valid_test)
)

print(
    "Fighters successfully calibrated:",
    returned_fighters
)

if returned_fighters == len(valid_test):

    print()
    print(
        "COMMON-SCALE SEARCH TEST: PASS ✅🔥"
    )

elif returned_fighters > 0:

    print()
    print(
        "COMMON-SCALE SEARCH TEST: PARTIAL ⚠️"
    )

else:

    print()
    print(
        "COMMON-SCALE SEARCH TEST: FAILED ❌"
    )
```

```python
# ============================================================
# STEP 79 — BUILD + TEST MULTI-ANCHOR GOOGLE TRENDS LADDER
# ============================================================

import pandas as pd
import numpy as np
import time
from pytrends.request import TrendReq


# ============================================================
# PART 1 — FRESH PYTRENDS CONNECTION
# ============================================================

pytrends = TrendReq(
    hl="en-US",
    tz=300,
    timeout=(10, 30),
    retries=0,
    backoff_factor=0
)

print("PYTRENDS READY ✅")


# ============================================================
# PART 2 — DEFINE ANCHOR CANDIDATES
#
# We want anchors at several popularity levels.
#
# Conor = top anchor.
#
# Max / Islam = strong middle anchors from Step 78B.
#
# Israel = lower-middle.
#
# Uros = deliberately low anchor.
#
# This lets smaller fighters be compared against someone
# closer to their own search volume instead of directly
# against Conor.
# ============================================================

anchor_candidates = [
    "Conor McGregor",
    "Max Holloway",
    "Islam Makhachev",
    "Israel Adesanya",
    "Uros Medic"
]


def get_topic(fighter):

    match = fighter_topics[
        fighter_topics["fighter"]
        .astype(str)
        .str.lower()
        == fighter.lower()
    ]

    if len(match) == 0:
        return None

    topic = match.iloc[0]["topic_id"]

    if pd.isna(topic):
        return None

    return str(topic)


anchor_topics = {
    fighter: get_topic(fighter)
    for fighter in anchor_candidates
}


print()
print("=" * 100)
print("ANCHOR TOPICS")
print("=" * 100)

for fighter, topic in anchor_topics.items():

    print(
        f"{fighter:25s} → {topic}"
    )


missing_anchors = [
    fighter
    for fighter, topic
    in anchor_topics.items()
    if topic is None
]


if missing_anchors:

    print()
    print(
        "WARNING — missing anchors:",
        missing_anchors
    )


# ============================================================
# PART 3 — TEST THE ENTIRE ANCHOR LADDER TOGETHER
#
# Exactly 5 topics, so Google allows them in one comparison.
#
# This gives us direct conversion ratios between every level.
# ============================================================

valid_anchor_names = [
    fighter
    for fighter in anchor_candidates
    if anchor_topics[fighter] is not None
]

valid_anchor_topics = [
    anchor_topics[fighter]
    for fighter in valid_anchor_names
]


print()
print("=" * 100)
print("RUNNING ANCHOR LADDER TEST")
print("=" * 100)

print(
    " | ".join(valid_anchor_names)
)


anchor_ladder_raw = None
anchor_ladder = pd.DataFrame()


try:

    pytrends.build_payload(
        kw_list=valid_anchor_topics,
        cat=0,
        timeframe="2025-01-01 2026-09-20",
        geo="US",
        gprop=""
    )

    anchor_ladder_raw = (
        pytrends
        .interest_over_time()
    )

    if (
        anchor_ladder_raw is not None
        and not anchor_ladder_raw.empty
    ):

        if (
            "isPartial"
            in anchor_ladder_raw.columns
        ):

            anchor_ladder_raw = (
                anchor_ladder_raw
                .drop(
                    columns=["isPartial"]
                )
            )


        anchor_means = (
            anchor_ladder_raw.mean()
        )


        conor_topic = anchor_topics[
            "Conor McGregor"
        ]

        conor_mean = float(
            anchor_means[
                conor_topic
            ]
        )


        ladder_rows = []


        for fighter in valid_anchor_names:

            topic = anchor_topics[
                fighter
            ]

            mean_interest = float(
                anchor_means[
                    topic
                ]
            )


            relative_to_conor = (
                mean_interest
                / conor_mean
                if conor_mean > 0
                else np.nan
            )


            ladder_rows.append({

                "fighter":
                    fighter,

                "topic_id":
                    topic,

                "mean_interest":
                    mean_interest,

                "relative_to_conor":
                    relative_to_conor,

                "conor_scale":
                    relative_to_conor * 100
            })


        anchor_ladder = (
            pd.DataFrame(
                ladder_rows
            )
            .sort_values(
                "conor_scale",
                ascending=False
            )
            .reset_index(drop=True)
        )


        print()
        print(
            anchor_ladder[
                [
                    "fighter",
                    "mean_interest",
                    "relative_to_conor",
                    "conor_scale"
                ]
            ]
            .to_string(
                index=False,
                formatters={

                    "mean_interest":
                        lambda x:
                        f"{x:.3f}",

                    "relative_to_conor":
                        lambda x:
                        f"{x:.5f}",

                    "conor_scale":
                        lambda x:
                        f"{x:.2f}"
                }
            )
        )


    else:

        print(
            "❌ Google returned no anchor data."
        )


except Exception as e:

    print(
        "❌ ANCHOR REQUEST FAILED"
    )

    print(
        type(e).__name__,
        str(e)
    )


# ============================================================
# PART 4 — CREATE ANCHOR CONVERSION DICTIONARY
#
# Example:
#
# If Islam = 20 on the Conor scale,
# and another fighter = 50% of Islam in a comparison,
# that fighter becomes:
#
# 20 × 0.50 = 10 on the Conor scale.
# ============================================================

if len(anchor_ladder):

    anchor_scale = dict(
        zip(
            anchor_ladder["fighter"],
            anchor_ladder["conor_scale"]
        )
    )

else:

    anchor_scale = {}


# ============================================================
# PART 5 — SELECT TEST FIGHTERS AT DIFFERENT LEVELS
#
# These are NOT permanent anchors.
#
# They simply test whether lower anchors recover useful
# resolution that would be lost against Conor.
# ============================================================

resolution_test_fighters = [

    # Recognizable
    "Charles Oliveira",
    "Dustin Poirier",
    "Justin Gaethje",

    # Mid-level
    "Mateusz Gamrot",
    "Gregory Rodrigues",

    # Lower-volume
    "Quillan Salkilld",
    "Jose Miguel Delgado"
]


resolution_topics = {
    fighter: get_topic(fighter)
    for fighter in resolution_test_fighters
}


print()
print("=" * 100)
print("RESOLUTION TEST TOPICS")
print("=" * 100)

for fighter, topic in resolution_topics.items():

    print(
        f"{fighter:25s} → {topic}"
    )


# ============================================================
# PART 6 — CHOOSE ANCHOR BASED ON EXPECTED RESOLUTION
#
# For this TEST:
#
# recognizable fighters → Max
# mid-level fighters     → Israel
# low-volume fighters    → Uros
#
# Later, bulk calibration will assign anchors systematically.
# ============================================================

test_anchor_assignment = {

    "Charles Oliveira":
        "Max Holloway",

    "Dustin Poirier":
        "Max Holloway",

    "Justin Gaethje":
        "Max Holloway",

    "Mateusz Gamrot":
        "Israel Adesanya",

    "Gregory Rodrigues":
        "Israel Adesanya",

    "Quillan Salkilld":
        "Uros Medic",

    "Jose Miguel Delgado":
        "Uros Medic"
}


# ============================================================
# PART 7 — GROUP TEST FIGHTERS BY ANCHOR
# ============================================================

anchor_groups = {}


for fighter, anchor in (
    test_anchor_assignment.items()
):

    topic = resolution_topics.get(
        fighter
    )

    if topic is None:
        continue

    if anchor not in anchor_scale:
        continue

    anchor_groups.setdefault(
        anchor,
        []
    ).append(
        fighter
    )


# ============================================================
# PART 8 — RUN ANCHOR COMPARISONS
# ============================================================

calibration_test_rows = []


for anchor, fighters in anchor_groups.items():

    anchor_topic = anchor_topics[
        anchor
    ]

    anchor_conor_scale = (
        anchor_scale[
            anchor
        ]
    )


    # Maximum 4 fighters + anchor
    for start in range(
        0,
        len(fighters),
        4
    ):

        batch = fighters[
            start:start + 4
        ]


        names = (
            [anchor]
            + batch
        )


        topics = (
            [anchor_topic]
            + [
                resolution_topics[f]
                for f in batch
            ]
        )


        print()
        print("=" * 100)

        print(
            "CALIBRATING:",
            " | ".join(names)
        )


        try:

            pytrends.build_payload(
                kw_list=topics,
                cat=0,
                timeframe="2025-01-01 2026-09-20",
                geo="US",
                gprop=""
            )


            df = (
                pytrends
                .interest_over_time()
            )


            if df.empty:

                print(
                    "❌ Empty Trends result."
                )

                continue


            if "isPartial" in df.columns:

                df = df.drop(
                    columns=["isPartial"]
                )


            means = df.mean()


            anchor_mean = float(
                means[
                    anchor_topic
                ]
            )


            print(
                "Anchor:",
                anchor
            )

            print(
                "Anchor mean:",
                round(
                    anchor_mean,
                    3
                )
            )

            print(
                "Anchor Conor-scale:",
                round(
                    anchor_conor_scale,
                    3
                )
            )


            for fighter in batch:

                topic = resolution_topics[
                    fighter
                ]


                fighter_mean = float(
                    means[
                        topic
                    ]
                )


                relative_to_anchor = (
                    fighter_mean
                    / anchor_mean
                    if anchor_mean > 0
                    else np.nan
                )


                calibrated_scale = (
                    relative_to_anchor
                    * anchor_conor_scale
                )


                calibration_test_rows.append({

                    "fighter":
                        fighter,

                    "anchor":
                        anchor,

                    "fighter_mean":
                        fighter_mean,

                    "anchor_mean":
                        anchor_mean,

                    "relative_to_anchor":
                        relative_to_anchor,

                    "calibrated_search_demand":
                        calibrated_scale
                })


                print(
                    f"{fighter:25s}",
                    "mean=",
                    f"{fighter_mean:.3f}",
                    "| relative=",
                    f"{relative_to_anchor:.4f}",
                    "| calibrated=",
                    f"{calibrated_scale:.3f}"
                )


        except Exception as e:

            print(
                "❌ REQUEST FAILED:",
                type(e).__name__,
                str(e)
            )


        # Give Google breathing room.
        time.sleep(5)


# ============================================================
# PART 9 — BUILD TEST RESULTS
# ============================================================

calibration_resolution_test = (
    pd.DataFrame(
        calibration_test_rows
    )
)


if len(
    calibration_resolution_test
):

    calibration_resolution_test = (
        calibration_resolution_test
        .sort_values(
            "calibrated_search_demand",
            ascending=False
        )
        .reset_index(drop=True)
    )


# ============================================================
# PART 10 — OUTPUT
# ============================================================

print()
print("=" * 100)
print(
    "MULTI-ANCHOR CALIBRATION RESULTS 🔥"
)
print("=" * 100)


if len(
    calibration_resolution_test
):

    print(
        calibration_resolution_test[
            [
                "fighter",
                "anchor",
                "fighter_mean",
                "relative_to_anchor",
                "calibrated_search_demand"
            ]
        ]
        .to_string(
            index=False,
            formatters={

                "fighter_mean":
                    lambda x:
                    f"{x:.3f}",

                "relative_to_anchor":
                    lambda x:
                    f"{x:.4f}",

                "calibrated_search_demand":
                    lambda x:
                    f"{x:.3f}"
            }
        )
    )

else:

    print(
        "No calibration results returned."
    )


# ============================================================
# PART 11 — VALIDATION
# ============================================================

print()
print("=" * 100)

expected_with_topics = sum(
    topic is not None
    for topic
    in resolution_topics.values()
)

returned = (
    calibration_resolution_test[
        "fighter"
    ].nunique()
    if len(
        calibration_resolution_test
    )
    else 0
)


print(
    "Resolution-test fighters with topics:",
    expected_with_topics
)

print(
    "Successfully calibrated:",
    returned
)


nonzero = (
    (
        calibration_resolution_test[
            "calibrated_search_demand"
        ] > 0
    ).sum()
    if len(
        calibration_resolution_test
    )
    else 0
)


print(
    "Non-zero calibrated results:",
    nonzero
)


print()

if (
    returned == expected_with_topics
    and returned > 0
    and nonzero == returned
):

    print(
        "MULTI-ANCHOR TEST: PASS ✅🔥"
    )

elif returned > 0:

    print(
        "MULTI-ANCHOR TEST: PARTIAL ⚠️"
    )

else:

    print(
        "MULTI-ANCHOR TEST: FAILED ❌"
    )
```

```python
# ============================================================
# STEP 80 — PRODUCTION CALIBRATED SEARCH DEMAND
# ============================================================

import pandas as pd
import numpy as np
import time
import os
from pytrends.request import TrendReq


# ============================================================
# SETTINGS
# ============================================================

TIMEFRAME = "2025-01-01 2026-09-20"
GEO = "US"

CHECKPOINT_FILE = (
    "/content/"
    "ufc_calibrated_search_demand_checkpoint.csv"
)

REQUEST_PAUSE = 5


# ============================================================
# PART 1 — PYTRENDS CONNECTION
# ============================================================

pytrends = TrendReq(
    hl="en-US",
    tz=300,
    timeout=(10, 30),
    retries=0,
    backoff_factor=0
)


# ============================================================
# PART 2 — LOCK OUR VALIDATED ANCHOR LADDER
#
# These values come directly from Step 79.
# Conor = 100 common scale.
# ============================================================

ANCHORS = {

    "Conor McGregor": {
        "topic": "/m/0rfgxy2",
        "scale": 100.000
    },

    "Max Holloway": {
        "topic": "/m/0h_d_h5",
        "scale": 36.364
    },

    "Islam Makhachev": {
        "topic": "/m/01274x53",
        "scale": 20.000
    },

    "Israel Adesanya": {
        "topic": "/g/11c1s_3n7q",
        "scale": 5.758
    },

    "Uros Medic": {
        "topic": "/g/11p162tzjt",
        "scale": 0.606
    }
}


# ============================================================
# PART 3 — BUILD VERIFIED MAIN-EVENT HEADLINER POOL
#
# We only need the fighters who actually headline events.
# ============================================================

main_event_fights = (
    ufc_fights_updated[
        ufc_fights_updated[
            "is_main_event_verified"
        ]
    ]
    [
        [
            "date",
            "event_name",
            "f1_name",
            "f2_name"
        ]
    ]
    .copy()
)


# Absolute Trends production begins in 2014 because our
# historical Google Trends event system begins there.
main_event_fights = (
    main_event_fights[
        main_event_fights["date"]
        >= pd.Timestamp("2014-01-01")
    ]
    .copy()
)


headliner_names = sorted(
    set(
        main_event_fights[
            "f1_name"
        ].dropna()
    )
    |
    set(
        main_event_fights[
            "f2_name"
        ].dropna()
    )
)


print(
    "Verified 2014+ main-event fights:",
    len(main_event_fights)
)

print(
    "Unique 2014+ headliners:",
    len(headliner_names)
)


# ============================================================
# PART 4 — ATTACH GOOGLE TOPICS
# ============================================================

def get_topic(fighter):

    match = fighter_topics[
        fighter_topics["fighter"]
        .astype(str)
        .str.lower()
        == str(fighter).lower()
    ]

    if len(match) == 0:
        return None

    topic = match.iloc[0]["topic_id"]

    if pd.isna(topic):
        return None

    return str(topic)


headliner_topics = pd.DataFrame({
    "fighter": headliner_names
})


headliner_topics["topic_id"] = (
    headliner_topics[
        "fighter"
    ].apply(get_topic)
)


# Patch the one new verified topic we found during
# the September update.
headliner_topics.loc[
    headliner_topics["fighter"]
    == "Salahdine Parnasse",
    "topic_id"
] = "/g/11g0kd0b7f"


headliner_topics[
    "topic_available"
] = (
    headliner_topics[
        "topic_id"
    ].notna()
)


print()
print(
    "Headliners with topics:",
    int(
        headliner_topics[
            "topic_available"
        ].sum()
    )
)

print(
    "Headliners without topics:",
    int(
        (~headliner_topics[
            "topic_available"
        ]).sum()
    )
)


# ============================================================
# PART 5 — START / RESTORE CHECKPOINT
# ============================================================

checkpoint_columns = [
    "fighter",
    "topic_id",
    "anchor",
    "anchor_scale",
    "fighter_mean",
    "anchor_mean",
    "relative_to_anchor",
    "calibrated_search_demand",
    "status"
]


if os.path.exists(
    CHECKPOINT_FILE
):

    calibrated_fighters = pd.read_csv(
        CHECKPOINT_FILE
    )

    print()
    print(
        "CHECKPOINT RESTORED:",
        len(calibrated_fighters),
        "rows"
    )

else:

    calibrated_fighters = pd.DataFrame(
        columns=checkpoint_columns
    )

    print()
    print(
        "Starting new calibration checkpoint."
    )


# ============================================================
# PART 6 — INSERT OUR KNOWN ANCHORS DIRECTLY
# ============================================================

for fighter, info in ANCHORS.items():

    topic = info["topic"]
    scale = info["scale"]

    existing = (
        calibrated_fighters[
            "fighter"
        ]
        .astype(str)
        .eq(fighter)
        .any()
    )

    if not existing:

        new_row = pd.DataFrame([{
            "fighter": fighter,
            "topic_id": topic,
            "anchor": fighter,
            "anchor_scale": scale,
            "fighter_mean": np.nan,
            "anchor_mean": np.nan,
            "relative_to_anchor": 1.0,
            "calibrated_search_demand": scale,
            "status": "anchor"
        }])

        calibrated_fighters = pd.concat(
            [
                calibrated_fighters,
                new_row
            ],
            ignore_index=True
        )


calibrated_fighters.to_csv(
    CHECKPOINT_FILE,
    index=False
)


# ============================================================
# PART 7 — DETERMINE WHICH FIGHTERS STILL NEED CALIBRATION
# ============================================================

already_done = set(
    calibrated_fighters[
        calibrated_fighters[
            "status"
        ].isin(
            ["success", "anchor"]
        )
    ]["fighter"]
)


production_pool = (
    headliner_topics[
        headliner_topics[
            "topic_available"
        ]
    ]
    .copy()
)


production_pool = (
    production_pool[
        ~production_pool[
            "fighter"
        ].isin(already_done)
    ]
    .reset_index(drop=True)
)


print()
print("=" * 100)

print(
    "Production fighters still needing calibration:",
    len(production_pool)
)


# ============================================================
# PART 8 — REQUEST HELPER
#
# Compare up to four fighters against one anchor.
# ============================================================

def run_anchor_batch(
    anchor_name,
    batch_df
):

    anchor_topic = (
        ANCHORS[
            anchor_name
        ]["topic"]
    )

    anchor_scale = (
        ANCHORS[
            anchor_name
        ]["scale"]
    )

    fighters = (
        batch_df[
            "fighter"
        ].tolist()
    )

    topics = (
        [anchor_topic]
        +
        batch_df[
            "topic_id"
        ].tolist()
    )


    pytrends.build_payload(
        kw_list=topics,
        cat=0,
        timeframe=TIMEFRAME,
        geo=GEO,
        gprop=""
    )


    df = (
        pytrends
        .interest_over_time()
    )


    if df.empty:
        return []


    if "isPartial" in df.columns:

        df = df.drop(
            columns=["isPartial"]
        )


    means = df.mean()

    anchor_mean = float(
        means[
            anchor_topic
        ]
    )


    results = []


    for fighter, topic in zip(
        fighters,
        batch_df[
            "topic_id"
        ]
    ):

        fighter_mean = float(
            means[
                topic
            ]
        )


        relative = (
            fighter_mean
            / anchor_mean
            if anchor_mean > 0
            else np.nan
        )


        calibrated = (
            relative
            * anchor_scale
            if pd.notna(relative)
            else np.nan
        )


        results.append({
            "fighter":
                fighter,

            "topic_id":
                topic,

            "anchor":
                anchor_name,

            "anchor_scale":
                anchor_scale,

            "fighter_mean":
                fighter_mean,

            "anchor_mean":
                anchor_mean,

            "relative_to_anchor":
                relative,

            "calibrated_search_demand":
                calibrated,

            "status":
                "success"
        })


    return results


# ============================================================
# PART 9 — FIRST PASS
#
# Use Israel as the middle anchor.
#
# This is deliberate:
# - strong enough for recognizable fighters
# - small enough to preserve far more resolution than Conor
#
# Extremely small values will later be recalibrated using
# Uros.
# Extremely large values will later be recalibrated using
# Max.
# ============================================================

middle_anchor = "Israel Adesanya"

first_pass_results = []


print()
print("=" * 100)
print(
    "FIRST PASS — ISRAEL ADESANYA ANCHOR"
)
print("=" * 100)


for start in range(
    0,
    len(production_pool),
    4
):

    batch = (
        production_pool
        .iloc[
            start:start + 4
        ]
        .copy()
    )


    print()
    print(
        f"Batch {start + 1}-"
        f"{min(start + 4, len(production_pool))}"
        f" / {len(production_pool)}"
    )

    print(
        " | ".join(
            batch[
                "fighter"
            ].tolist()
        )
    )


    try:

        results = run_anchor_batch(
            middle_anchor,
            batch
        )


        if not results:

            print(
                "⚠️ Empty Google Trends result."
            )

        else:

            first_pass_results.extend(
                results
            )


            for result in results:

                print(
                    f"{result['fighter']:25s}",
                    "→",
                    f"{result['calibrated_search_demand']:.3f}"
                )


            # -----------------------------------------------
            # CHECKPOINT IMMEDIATELY
            # -----------------------------------------------

            calibrated_fighters = (
                pd.concat(
                    [
                        calibrated_fighters,
                        pd.DataFrame(
                            results
                        )
                    ],
                    ignore_index=True
                )
            )


            # Keep latest result per fighter
            calibrated_fighters = (
                calibrated_fighters
                .drop_duplicates(
                    subset=["fighter"],
                    keep="last"
                )
            )


            calibrated_fighters.to_csv(
                CHECKPOINT_FILE,
                index=False
            )


    except Exception as e:

        print(
            "❌ REQUEST FAILED:",
            type(e).__name__,
            str(e)
        )

        print()
        print(
            "Checkpoint is safe."
        )

        print(
            "If Google starts rate-limiting, "
            "we can resume later."
        )


    time.sleep(
        REQUEST_PAUSE
    )


# ============================================================
# PART 10 — IDENTIFY RESULTS NEEDING BETTER RESOLUTION
#
# Very small values:
# compare against Uros.
#
# Very large values:
# compare against Max.
#
# Middle values:
# Israel result is already appropriate.
# ============================================================

current_success = (
    calibrated_fighters[
        calibrated_fighters[
            "status"
        ] == "success"
    ]
    .copy()
)


low_resolution = (
    current_success[
        current_success[
            "calibrated_search_demand"
        ] < 1.5
    ]
    .copy()
)


high_resolution = (
    current_success[
        current_success[
            "calibrated_search_demand"
        ] > 15
    ]
    .copy()
)


print()
print("=" * 100)
print("RESOLUTION AUDIT")
print("=" * 100)

print(
    "Low-demand fighters to refine with Uros:",
    len(low_resolution)
)

print(
    "High-demand fighters to refine with Max:",
    len(high_resolution)
)


# ============================================================
# PART 11 — REFINEMENT FUNCTION
# ============================================================

def refine_group(
    fighters_df,
    anchor_name
):

    global calibrated_fighters

    if len(fighters_df) == 0:
        return


    topic_lookup = dict(
        zip(
            production_pool[
                "fighter"
            ],
            production_pool[
                "topic_id"
            ]
        )
    )


    refine_rows = []


    for fighter in (
        fighters_df[
            "fighter"
        ]
        .drop_duplicates()
    ):

        if fighter in ANCHORS:
            continue

        topic = topic_lookup.get(
            fighter
        )

        if topic is None:
            continue

        refine_rows.append({
            "fighter":
                fighter,

            "topic_id":
                topic
        })


    refine_df = pd.DataFrame(
        refine_rows
    )


    if len(refine_df) == 0:
        return


    print()
    print("=" * 100)

    print(
        "REFINING WITH:",
        anchor_name
    )

    print("=" * 100)


    for start in range(
        0,
        len(refine_df),
        4
    ):

        batch = (
            refine_df
            .iloc[
                start:start + 4
            ]
            .copy()
        )


        print()
        print(
            " | ".join(
                batch[
                    "fighter"
                ]
            )
        )


        try:

            results = run_anchor_batch(
                anchor_name,
                batch
            )


            for result in results:

                print(
                    f"{result['fighter']:25s}",
                    "→",
                    f"{result['calibrated_search_demand']:.3f}"
                )


            if results:

                calibrated_fighters = (
                    pd.concat(
                        [
                            calibrated_fighters,
                            pd.DataFrame(
                                results
                            )
                        ],
                        ignore_index=True
                    )
                )


                calibrated_fighters = (
                    calibrated_fighters
                    .drop_duplicates(
                        subset=[
                            "fighter"
                        ],
                        keep="last"
                    )
                )


                calibrated_fighters.to_csv(
                    CHECKPOINT_FILE,
                    index=False
                )


        except Exception as e:

            print(
                "❌ REFINEMENT FAILED:",
                type(e).__name__,
                str(e)
            )


        time.sleep(
            REQUEST_PAUSE
        )


# ============================================================
# PART 12 — RUN REFINEMENTS
# ============================================================

refine_group(
    low_resolution,
    "Uros Medic"
)


refine_group(
    high_resolution,
    "Max Holloway"
)


# ============================================================
# PART 13 — FINAL CLEAN PRODUCTION TABLE
# ============================================================

calibrated_fighters = (
    calibrated_fighters
    .drop_duplicates(
        subset=["fighter"],
        keep="last"
    )
    .sort_values(
        "calibrated_search_demand",
        ascending=False,
        na_position="last"
    )
    .reset_index(drop=True)
)


calibrated_fighters.to_csv(
    CHECKPOINT_FILE,
    index=False
)


# ============================================================
# PART 14 — COVERAGE
# ============================================================

headliner_search_demand = (
    headliner_topics
    .merge(
        calibrated_fighters[
            [
                "fighter",
                "calibrated_search_demand",
                "anchor",
                "status"
            ]
        ],
        on="fighter",
        how="left"
    )
)


headliner_search_demand[
    "calibrated_available"
] = (
    headliner_search_demand[
        "calibrated_search_demand"
    ].notna()
)


total_headliners = len(
    headliner_search_demand
)

topics_available = int(
    headliner_search_demand[
        "topic_available"
    ].sum()
)

calibrated_available = int(
    headliner_search_demand[
        "calibrated_available"
    ].sum()
)


# ============================================================
# PART 15 — OUTPUT
# ============================================================

print()
print("=" * 100)
print(
    "STEP 80 PRODUCTION CALIBRATION COMPLETE 🔥"
)
print("=" * 100)

print(
    "Total unique 2014+ headliners:",
    total_headliners
)

print(
    "Headliners with Google Topics:",
    topics_available
)

print(
    "Headliners successfully calibrated:",
    calibrated_available
)

print(
    "Topic coverage:",
    f"{topics_available / total_headliners * 100:.2f}%"
)

print(
    "Calibration coverage:",
    f"{calibrated_available / total_headliners * 100:.2f}%"
)


print()
print("=" * 100)
print("TOP 30 CALIBRATED HEADLINERS")
print("=" * 100)

print(
    headliner_search_demand[
        headliner_search_demand[
            "calibrated_available"
        ]
    ]
    .sort_values(
        "calibrated_search_demand",
        ascending=False
    )
    [
        [
            "fighter",
            "calibrated_search_demand",
            "anchor"
        ]
    ]
    .head(30)
    .to_string(
        index=False,
        formatters={
            "calibrated_search_demand":
                lambda x: f"{x:.3f}"
        }
    )
)


print()
print("=" * 100)
print("UNCALIBRATED HEADLINERS")
print("=" * 100)

uncalibrated = (
    headliner_search_demand[
        ~headliner_search_demand[
            "calibrated_available"
        ]
    ]
    [
        [
            "fighter",
            "topic_id",
            "topic_available"
        ]
    ]
)


print(
    uncalibrated.to_string(
        index=False
    )
    if len(uncalibrated)
    else
    "NONE 🔥"
)


print()
print("=" * 100)

print(
    "Checkpoint saved to:"
)

print(
    CHECKPOINT_FILE
)
```

```python
# ============================================================
# STEP 81 — SAFE RESUME OF PRODUCTION SEARCH CALIBRATION
# ============================================================

import pandas as pd
import numpy as np
import os
import time

from pytrends.request import TrendReq
from pytrends.exceptions import TooManyRequestsError


# ============================================================
# SETTINGS
# ============================================================

TIMEFRAME = "2025-01-01 2026-09-20"
GEO = "US"

CHECKPOINT_FILE = (
    "/content/"
    "ufc_calibrated_search_demand_checkpoint.csv"
)

# More conservative than Step 80
REQUEST_PAUSE = 12


# ============================================================
# PART 1 — LOAD CHECKPOINT
# ============================================================

if not os.path.exists(CHECKPOINT_FILE):

    raise FileNotFoundError(
        "Checkpoint file not found. "
        "Do NOT continue until the Step 80 checkpoint exists."
    )


calibrated_fighters = pd.read_csv(
    CHECKPOINT_FILE
)


print("=" * 105)
print("CHECKPOINT RESTORED 🔥")
print("=" * 105)

print(
    "Rows:",
    len(calibrated_fighters)
)


# ============================================================
# PART 2 — ANCHORS
# ============================================================

ANCHORS = {

    "Conor McGregor": {
        "topic": "/m/0rfgxy2",
        "scale": 100.000
    },

    "Max Holloway": {
        "topic": "/m/0h_d_h5",
        "scale": 36.364
    },

    "Islam Makhachev": {
        "topic": "/m/01274x53",
        "scale": 20.000
    },

    "Israel Adesanya": {
        "topic": "/g/11c1s_3n7q",
        "scale": 5.758
    },

    "Uros Medic": {
        "topic": "/g/11p162tzjt",
        "scale": 0.606
    }
}


# ============================================================
# PART 3 — REBUILD HEADLINER POOL
# ============================================================

main_event_fights = (
    ufc_fights_updated[
        (
            ufc_fights_updated[
                "is_main_event_verified"
            ]
        )
        &
        (
            ufc_fights_updated["date"]
            >= pd.Timestamp("2014-01-01")
        )
    ]
    [
        [
            "date",
            "event_name",
            "f1_name",
            "f2_name"
        ]
    ]
    .copy()
)


headliner_names = sorted(
    set(
        main_event_fights[
            "f1_name"
        ].dropna()
    )
    |
    set(
        main_event_fights[
            "f2_name"
        ].dropna()
    )
)


def get_topic(fighter):

    match = fighter_topics[
        fighter_topics[
            "fighter"
        ]
        .astype(str)
        .str.lower()
        ==
        str(fighter).lower()
    ]

    if len(match) == 0:
        return None

    topic = match.iloc[0][
        "topic_id"
    ]

    if pd.isna(topic):
        return None

    return str(topic)


headliner_topics = pd.DataFrame({
    "fighter": headliner_names
})


headliner_topics[
    "topic_id"
] = (
    headliner_topics[
        "fighter"
    ].apply(get_topic)
)


# Verified new topic
headliner_topics.loc[
    headliner_topics["fighter"]
    == "Salahdine Parnasse",
    "topic_id"
] = "/g/11g0kd0b7f"


headliner_topics[
    "topic_available"
] = (
    headliner_topics[
        "topic_id"
    ].notna()
)


# ============================================================
# PART 4 — FIND ONLY THE MISSING FIGHTERS
# ============================================================

completed = set(
    calibrated_fighters.loc[
        calibrated_fighters[
            "calibrated_search_demand"
        ].notna(),
        "fighter"
    ]
)


missing_pool = (
    headliner_topics[
        (
            headliner_topics[
                "topic_available"
            ]
        )
        &
        (
            ~headliner_topics[
                "fighter"
            ].isin(completed)
        )
    ]
    .copy()
    .reset_index(drop=True)
)


print()
print(
    "Total 2014+ headliners:",
    len(headliner_topics)
)

print(
    "Already calibrated:",
    len(
        set(headliner_topics["fighter"])
        & completed
    )
)

print(
    "Still needing calibration:",
    len(missing_pool)
)

print(
    "Missing because no topic:",
    int(
        (~headliner_topics[
            "topic_available"
        ]).sum()
    )
)


# ============================================================
# PART 5 — CONNECTION
# ============================================================

pytrends = TrendReq(
    hl="en-US",
    tz=300,
    timeout=(10, 30),
    retries=0,
    backoff_factor=0
)


# ============================================================
# PART 6 — REQUEST FUNCTION
#
# Use Israel for first pass.
# We'll refine low/high values AFTER coverage is complete.
# ============================================================

ANCHOR_NAME = "Israel Adesanya"

anchor_topic = (
    ANCHORS[
        ANCHOR_NAME
    ]["topic"]
)

anchor_scale = (
    ANCHORS[
        ANCHOR_NAME
    ]["scale"]
)


def calibrate_batch(batch):

    topics = (
        [anchor_topic]
        +
        batch[
            "topic_id"
        ].tolist()
    )

    pytrends.build_payload(
        kw_list=topics,
        cat=0,
        timeframe=TIMEFRAME,
        geo=GEO,
        gprop=""
    )

    df = (
        pytrends
        .interest_over_time()
    )

    if df.empty:
        return []

    if "isPartial" in df.columns:

        df = df.drop(
            columns=["isPartial"]
        )

    means = df.mean()

    anchor_mean = float(
        means[
            anchor_topic
        ]
    )

    output = []

    for _, row in batch.iterrows():

        fighter = row[
            "fighter"
        ]

        topic = row[
            "topic_id"
        ]

        fighter_mean = float(
            means[
                topic
            ]
        )

        relative = (
            fighter_mean
            / anchor_mean
            if anchor_mean > 0
            else np.nan
        )

        calibrated = (
            relative
            * anchor_scale
            if pd.notna(relative)
            else np.nan
        )

        output.append({

            "fighter":
                fighter,

            "topic_id":
                topic,

            "anchor":
                ANCHOR_NAME,

            "anchor_scale":
                anchor_scale,

            "fighter_mean":
                fighter_mean,

            "anchor_mean":
                anchor_mean,

            "relative_to_anchor":
                relative,

            "calibrated_search_demand":
                calibrated,

            "status":
                "success"
        })

    return output


# ============================================================
# PART 7 — RESUME
#
# CRITICAL DIFFERENCE:
# STOP immediately on a 429.
# ============================================================

rate_limited = False
successful_this_run = 0


print()
print("=" * 105)
print("RESUMING MISSING FIGHTERS")
print("=" * 105)


for start in range(
    0,
    len(missing_pool),
    4
):

    batch = (
        missing_pool
        .iloc[
            start:start + 4
        ]
        .copy()
    )


    print()
    print(
        f"Resume batch "
        f"{start + 1}-"
        f"{min(start + 4, len(missing_pool))}"
        f" / {len(missing_pool)}"
    )

    print(
        " | ".join(
            batch[
                "fighter"
            ].tolist()
        )
    )


    try:

        results = calibrate_batch(
            batch
        )


        if not results:

            print(
                "⚠️ Empty Trends result."
            )

        else:

            for result in results:

                print(
                    f"{result['fighter']:25s}",
                    "→",
                    f"{result['calibrated_search_demand']:.3f}"
                )


            calibrated_fighters = (
                pd.concat(
                    [
                        calibrated_fighters,
                        pd.DataFrame(
                            results
                        )
                    ],
                    ignore_index=True
                )
            )


            calibrated_fighters = (
                calibrated_fighters
                .drop_duplicates(
                    subset=["fighter"],
                    keep="last"
                )
            )


            # SAVE AFTER EVERY BATCH
            calibrated_fighters.to_csv(
                CHECKPOINT_FILE,
                index=False
            )


            successful_this_run += (
                len(results)
            )


    except TooManyRequestsError:

        print()
        print(
            "🛑 GOOGLE RATE LIMIT HIT — STOPPING SAFELY."
        )

        print(
            "No completed results were lost."
        )

        rate_limited = True

        break


    except Exception as e:

        print()
        print(
            "⚠️ OTHER REQUEST ERROR"
        )

        print(
            type(e).__name__,
            str(e)
        )

        print(
            "Skipping this batch and continuing."
        )


    time.sleep(
        REQUEST_PAUSE
    )


# ============================================================
# PART 8 — RELOAD SAVED CHECKPOINT
# ============================================================

calibrated_fighters = pd.read_csv(
    CHECKPOINT_FILE
)


completed = set(
    calibrated_fighters.loc[
        calibrated_fighters[
            "calibrated_search_demand"
        ].notna(),
        "fighter"
    ]
)


headliner_search_demand = (
    headliner_topics
    .merge(
        calibrated_fighters[
            [
                "fighter",
                "calibrated_search_demand",
                "anchor",
                "status"
            ]
        ],
        on="fighter",
        how="left"
    )
)


headliner_search_demand[
    "calibrated_available"
] = (
    headliner_search_demand[
        "calibrated_search_demand"
    ].notna()
)


# ============================================================
# PART 9 — COVERAGE AUDIT
# ============================================================

total = len(
    headliner_search_demand
)

with_topics = int(
    headliner_search_demand[
        "topic_available"
    ].sum()
)

calibrated = int(
    headliner_search_demand[
        "calibrated_available"
    ].sum()
)

remaining_with_topics = int(
    (
        headliner_search_demand[
            "topic_available"
        ]
        &
        ~headliner_search_demand[
            "calibrated_available"
        ]
    ).sum()
)


print()
print("=" * 105)
print("STEP 81 RESUME RESULTS 🔥")
print("=" * 105)

print(
    "Successful fighters added this run:",
    successful_this_run
)

print(
    "Total headliners:",
    total
)

print(
    "Headliners with topics:",
    with_topics
)

print(
    "Headliners calibrated:",
    calibrated
)

print(
    "Still missing despite having topic:",
    remaining_with_topics
)

print(
    "No-topic headliners:",
    total - with_topics
)

print(
    "Calibration coverage:",
    f"{calibrated / total * 100:.2f}%"
)

print(
    "Coverage among topic-available fighters:",
    f"{calibrated / with_topics * 100:.2f}%"
)


print()
print(
    "Rate limited during this run:",
    rate_limited
)


# ============================================================
# PART 10 — SHOW NEXT MISSING FIGHTERS
# ============================================================

still_missing = (
    headliner_search_demand[
        (
            headliner_search_demand[
                "topic_available"
            ]
        )
        &
        (
            ~headliner_search_demand[
                "calibrated_available"
            ]
        )
    ]
    [
        [
            "fighter",
            "topic_id"
        ]
    ]
)


print()
print("=" * 105)
print("NEXT MISSING TOPIC-AVAILABLE FIGHTERS")
print("=" * 105)

if len(still_missing):

    print(
        still_missing
        .head(40)
        .to_string(
            index=False
        )
    )

else:

    print(
        "NONE — ALL TOPIC-AVAILABLE HEADLINERS CALIBRATED 🔥"
    )


print()
print("=" * 105)

if remaining_with_topics == 0:

    print(
        "FIRST-PASS CALIBRATION COVERAGE: COMPLETE ✅🔥"
    )

elif rate_limited:

    print(
        "CHECKPOINT SAFE — WAIT AND RUN STEP 81 AGAIN LATER."
    )

else:

    print(
        "PARTIAL — RUN STEP 81 AGAIN TO FINISH REMAINING FIGHTERS."
    )
```

```python
# ============================================================
# STEP 82 — FINAL FIGHT-DATA STRUCTURE CLEANUP
# ============================================================

import pandas as pd
import numpy as np
import re
import unicodedata


print("=" * 105)
print("STEP 82 — UPDATED FIGHT DATA CLEANUP 🔥")
print("=" * 105)


# ============================================================
# PART 1 — SAFETY COPY
# ============================================================

ufc_fights_before_step82 = (
    ufc_fights_updated.copy()
)

print(
    "Starting fights:",
    len(ufc_fights_updated)
)

print(
    "Starting columns:",
    len(ufc_fights_updated.columns)
)


# ============================================================
# PART 2 — STANDARDIZE UFC 331 EVENT NAME
#
# UFC 331 currently came from UFC.com as:
#
# Crypto.com UFC 331: Van vs. Pantoja 2
#
# For our historical event naming/classification system,
# standardize it to:
#
# UFC 331: Van vs. Pantoja 2
#
# We preserve the UFC.com wording separately.
# ============================================================

if "event_name_original" not in ufc_fights_updated.columns:

    ufc_fights_updated[
        "event_name_original"
    ] = ufc_fights_updated[
        "event_name"
    ]


ufc_fights_updated.loc[
    ufc_fights_updated[
        "event_name"
    ].eq(
        "Crypto.com UFC 331: Van vs. Pantoja 2"
    ),
    "event_name"
] = (
    "UFC 331: Van vs. Pantoja 2"
)


# ============================================================
# PART 3 — SLUG FUNCTION
#
# Rebuild the four fields that existed in the original
# 59-column project.
# ============================================================

def make_fighter_slug(name):

    if pd.isna(name):
        return np.nan

    text = str(name).strip()

    # Convert accented characters where possible
    text = unicodedata.normalize(
        "NFKD",
        text
    )

    text = "".join(
        char
        for char in text
        if not unicodedata.combining(char)
    )

    text = text.lower()

    # Apostrophes should disappear
    text = (
        text
        .replace("’", "")
        .replace("'", "")
    )

    # Anything else non-alphanumeric becomes hyphen
    text = re.sub(
        r"[^a-z0-9]+",
        "-",
        text
    )

    text = text.strip("-")

    return text


# ============================================================
# PART 4 — RECREATE FIGHTER SLUGS
# ============================================================

ufc_fights_updated[
    "f1_slug"
] = (
    ufc_fights_updated[
        "f1_name"
    ].apply(
        make_fighter_slug
    )
)


ufc_fights_updated[
    "f2_slug"
] = (
    ufc_fights_updated[
        "f2_name"
    ].apply(
        make_fighter_slug
    )
)


# ============================================================
# PART 5 — RECREATE MATCHUP SLUGS
#
# matchup_slug_1 = f1-vs-f2
# matchup_slug_2 = f2-vs-f1
#
# Both orientations are useful when matching external data.
# ============================================================

ufc_fights_updated[
    "matchup_slug_1"
] = (
    ufc_fights_updated[
        "f1_slug"
    ].astype(str)
    +
    "-vs-"
    +
    ufc_fights_updated[
        "f2_slug"
    ].astype(str)
)


ufc_fights_updated[
    "matchup_slug_2"
] = (
    ufc_fights_updated[
        "f2_slug"
    ].astype(str)
    +
    "-vs-"
    +
    ufc_fights_updated[
        "f1_slug"
    ].astype(str)
)


# ============================================================
# PART 6 — VALIDATE UFC 331
# ============================================================

ufc331 = (
    ufc_fights_updated[
        ufc_fights_updated[
            "date"
        ].eq(
            pd.Timestamp(
                "2026-09-19"
            )
        )
    ]
)


print()
print("=" * 105)
print("UFC 331 CHECK")
print("=" * 105)

print(
    "Rows:",
    len(ufc331)
)

print(
    "Event names:",
    ufc331[
        "event_name"
    ].unique()
)


# ============================================================
# PART 7 — VALIDATE SLUGS
# ============================================================

slug_columns = [
    "f1_slug",
    "f2_slug",
    "matchup_slug_1",
    "matchup_slug_2"
]


print()
print("=" * 105)
print("SLUG VALIDATION")
print("=" * 105)


for col in slug_columns:

    print(
        f"{col:20s}",
        "missing:",
        int(
            ufc_fights_updated[
                col
            ].isna().sum()
        )
    )


print()
print("NEW-FIGHT SLUG EXAMPLES:")

print(
    ufc_fights_updated[
        ufc_fights_updated[
            "date"
        ]
        >= pd.Timestamp(
            "2026-07-11"
        )
    ]
    [
        [
            "f1_name",
            "f2_name",
            "f1_slug",
            "f2_slug",
            "matchup_slug_1"
        ]
    ]
    .head(10)
    .to_string(
        index=False
    )
)


# ============================================================
# PART 8 — CORE DATA INTEGRITY
# ============================================================

print()
print("=" * 105)
print("CORE INTEGRITY")
print("=" * 105)


total_fights = len(
    ufc_fights_updated
)

total_events = (
    ufc_fights_updated[
        [
            "date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


new_fights_count = (
    ufc_fights_updated[
        ufc_fights_updated[
            "date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    .shape[0]
)


new_events_count = (
    ufc_fights_updated[
        ufc_fights_updated[
            "date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


print(
    "Total fights:",
    total_fights
)

print(
    "Expected fights:",
    8877
)

print(
    "Total events:",
    total_events
)

print(
    "Expected events:",
    788
)

print(
    "New fights:",
    new_fights_count
)

print(
    "Expected new fights:",
    141
)

print(
    "New events:",
    new_events_count
)

print(
    "Expected new events:",
    11
)


# ============================================================
# PART 9 — IMPORTANT DERIVED VARIABLE CHECKS
# ============================================================

print()
print("=" * 105)
print("DERIVED VARIABLE CHECK")
print("=" * 105)


checks = {

    "Title fights":
        int(
            ufc_fights_updated[
                "is_title_fight"
            ].sum()
        ),

    "Verified main events":
        int(
            ufc_fights_updated[
                "is_main_event_verified"
            ].sum()
        ),

    "FOTN":
        int(
            ufc_fights_updated[
                "FOTN"
            ].sum()
        ),

    "POTN":
        int(
            ufc_fights_updated[
                "POTN"
            ].sum()
        ),

    "KOTN":
        int(
            ufc_fights_updated[
                "KOTN"
            ].sum()
        ),

    "SOTN":
        int(
            ufc_fights_updated[
                "SOTN"
            ].sum()
        )
}


for key, value in checks.items():

    print(
        f"{key:25s}",
        value
    )


# ============================================================
# PART 10 — RANKING CHECK
# ============================================================

new_period = (
    ufc_fights_updated[
        ufc_fights_updated[
            "date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
)


new_ranked_entries = (
    new_period[
        "f1_rank_before"
    ].notna().sum()
    +
    new_period[
        "f2_rank_before"
    ].notna().sum()
)


new_champions = (
    new_period[
        "f1_rank_before"
    ].eq(0).sum()
    +
    new_period[
        "f2_rank_before"
    ].eq(0).sum()
)


print()
print("=" * 105)
print("NEW RANKING CHECK")
print("=" * 105)

print(
    "Ranked fighter entries:",
    int(
        new_ranked_entries
    )
)

print(
    "Expected:",
    63
)

print(
    "Champions:",
    int(
        new_champions
    )
)

print(
    "Expected:",
    3
)


# ============================================================
# PART 11 — FINAL PASS / FAIL
# ============================================================

integrity_pass = (

    total_fights == 8877

    and

    total_events == 788

    and

    new_fights_count == 141

    and

    new_events_count == 11

    and

    int(new_ranked_entries) == 63

    and

    int(new_champions) == 3

    and

    checks[
        "Title fights"
    ] == 392

    and

    checks[
        "Verified main events"
    ] == 785
)


print()
print("=" * 105)

if integrity_pass:

    print(
        "STEP 82 DATA INTEGRITY: PASS ✅🔥"
    )

else:

    print(
        "STEP 82 DATA INTEGRITY: CHECK REQUIRED ⚠️"
    )


print("=" * 105)
```

```python
# ============================================================
# STEP 83 — REBUILD 788-EVENT MASTER
# ============================================================

import pandas as pd
import numpy as np


print("=" * 105)
print("STEP 83 — REBUILD UPDATED EVENT MASTER 🔥")
print("=" * 105)


# ============================================================
# PART 1 — EVENT BASE
# ============================================================

event_master_updated = (
    ufc_fights_updated[
        [
            "date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .rename(
        columns={
            "date": "event_date"
        }
    )
    .sort_values(
        [
            "event_date",
            "event_name"
        ]
    )
    .reset_index(drop=True)
)


print(
    "Base events:",
    len(event_master_updated)
)


# ============================================================
# PART 2 — HELPER
#
# Convert fighter-side columns into one long fighter table.
# This makes event-level averages/totals much cleaner.
# ============================================================

f1_event_fighters = pd.DataFrame({

    "event_date":
        ufc_fights_updated["date"],

    "event_name":
        ufc_fights_updated["event_name"],

    "fighter":
        ufc_fights_updated["f1_name"],

    "ufc_fights_before":
        ufc_fights_updated["f1_ufc_fights_before"],

    "ufc_win_pct_before":
        ufc_fights_updated["f1_ufc_win_pct_before"],

    "win_streak_before":
        ufc_fights_updated["f1_win_streak_before"],

    "last5_win_pct":
        ufc_fights_updated["f1_last5_win_pct"],

    "finish_rate_before":
        ufc_fights_updated["f1_finish_rate_before"],

    "ufc_finishes_before":
        ufc_fights_updated["f1_ufc_finishes_before"],

    "previous_title_fights":
        ufc_fights_updated["f1_previous_title_fights"],

    "previous_main_events":
        ufc_fights_updated["f1_previous_main_events"],

    "previous_bonuses":
        ufc_fights_updated["f1_previous_total_bonuses"],

    "rank_before":
        ufc_fights_updated["f1_rank_before"]
})


f2_event_fighters = pd.DataFrame({

    "event_date":
        ufc_fights_updated["date"],

    "event_name":
        ufc_fights_updated["event_name"],

    "fighter":
        ufc_fights_updated["f2_name"],

    "ufc_fights_before":
        ufc_fights_updated["f2_ufc_fights_before"],

    "ufc_win_pct_before":
        ufc_fights_updated["f2_ufc_win_pct_before"],

    "win_streak_before":
        ufc_fights_updated["f2_win_streak_before"],

    "last5_win_pct":
        ufc_fights_updated["f2_last5_win_pct"],

    "finish_rate_before":
        ufc_fights_updated["f2_finish_rate_before"],

    "ufc_finishes_before":
        ufc_fights_updated["f2_ufc_finishes_before"],

    "previous_title_fights":
        ufc_fights_updated["f2_previous_title_fights"],

    "previous_main_events":
        ufc_fights_updated["f2_previous_main_events"],

    "previous_bonuses":
        ufc_fights_updated["f2_previous_total_bonuses"],

    "rank_before":
        ufc_fights_updated["f2_rank_before"]
})


event_fighters_long = pd.concat(
    [
        f1_event_fighters,
        f2_event_fighters
    ],
    ignore_index=True
)


# ============================================================
# PART 3 — BASIC CARD COUNTS
# ============================================================

basic_card = (
    ufc_fights_updated
    .groupby(
        [
            "date",
            "event_name"
        ],
        as_index=False
    )
    .agg(

        total_fights=(
            "event_name",
            "size"
        ),

        title_fights=(
            "is_title_fight",
            "sum"
        )
    )
    .rename(
        columns={
            "date": "event_date"
        }
    )
)


event_master_updated = (
    event_master_updated
    .merge(
        basic_card,
        on=[
            "event_date",
            "event_name"
        ],
        how="left"
    )
)


# ============================================================
# PART 4 — RANKED FIGHTERS + RANKED MATCHUPS
# ============================================================

rank_temp = (
    ufc_fights_updated[
        [
            "date",
            "event_name",
            "f1_rank_before",
            "f2_rank_before"
        ]
    ]
    .copy()
)


rank_temp[
    "ranked_fighters"
] = (
    rank_temp[
        "f1_rank_before"
    ].notna().astype(int)
    +
    rank_temp[
        "f2_rank_before"
    ].notna().astype(int)
)


rank_temp[
    "ranked_matchups"
] = (
    rank_temp[
        "f1_rank_before"
    ].notna()
    &
    rank_temp[
        "f2_rank_before"
    ].notna()
).astype(int)


ranking_card = (
    rank_temp
    .groupby(
        [
            "date",
            "event_name"
        ],
        as_index=False
    )
    .agg(

        ranked_fighters=(
            "ranked_fighters",
            "sum"
        ),

        ranked_matchups=(
            "ranked_matchups",
            "sum"
        )
    )
    .rename(
        columns={
            "date": "event_date"
        }
    )
)


event_master_updated = (
    event_master_updated
    .merge(
        ranking_card,
        on=[
            "event_date",
            "event_name"
        ],
        how="left"
    )
)


# ============================================================
# PART 5 — CARD-WIDE FIGHTER EXPERIENCE / FORM
# ============================================================

fighter_event_stats = (
    event_fighters_long
    .groupby(
        [
            "event_date",
            "event_name"
        ],
        as_index=False
    )
    .agg(

        avg_ufc_experience=(
            "ufc_fights_before",
            "mean"
        ),

        total_prior_ufc_fights=(
            "ufc_fights_before",
            "sum"
        ),

        avg_ufc_win_pct=(
            "ufc_win_pct_before",
            "mean"
        ),

        avg_win_streak=(
            "win_streak_before",
            "mean"
        ),

        avg_last5_win_pct=(
            "last5_win_pct",
            "mean"
        ),

        avg_finish_rate=(
            "finish_rate_before",
            "mean"
        ),

        total_prior_ufc_finishes=(
            "ufc_finishes_before",
            "sum"
        ),

        avg_previous_title_fights=(
            "previous_title_fights",
            "mean"
        ),

        total_previous_title_fights=(
            "previous_title_fights",
            "sum"
        ),

        avg_previous_main_events=(
            "previous_main_events",
            "mean"
        ),

        total_previous_main_events=(
            "previous_main_events",
            "sum"
        ),

        avg_previous_bonuses=(
            "previous_bonuses",
            "mean"
        ),

        total_previous_bonuses=(
            "previous_bonuses",
            "sum"
        )
    )
)


event_master_updated = (
    event_master_updated
    .merge(
        fighter_event_stats,
        on=[
            "event_date",
            "event_name"
        ],
        how="left"
    )
)


# ============================================================
# PART 6 — VERIFIED MAIN EVENT ROWS
# ============================================================

main_events = (
    ufc_fights_updated[
        ufc_fights_updated[
            "is_main_event_verified"
        ]
    ]
    .copy()
)


main_events[
    "main_event_combined_ufc_fights"
] = (
    main_events[
        "f1_ufc_fights_before"
    ]
    +
    main_events[
        "f2_ufc_fights_before"
    ]
)


main_events[
    "main_event_avg_win_pct"
] = (
    main_events[
        [
            "f1_ufc_win_pct_before",
            "f2_ufc_win_pct_before"
        ]
    ]
    .mean(axis=1)
)


main_events[
    "main_event_combined_win_streak"
] = (
    main_events[
        "f1_win_streak_before"
    ]
    +
    main_events[
        "f2_win_streak_before"
    ]
)


main_events[
    "main_event_combined_title_fights"
] = (
    main_events[
        "f1_previous_title_fights"
    ]
    +
    main_events[
        "f2_previous_title_fights"
    ]
)


main_events[
    "main_event_combined_previous_main_events"
] = (
    main_events[
        "f1_previous_main_events"
    ]
    +
    main_events[
        "f2_previous_main_events"
    ]
)


main_events[
    "main_event_combined_bonuses"
] = (
    main_events[
        "f1_previous_total_bonuses"
    ]
    +
    main_events[
        "f2_previous_total_bonuses"
    ]
)


main_events[
    "main_event_f1_rank"
] = (
    main_events[
        "f1_rank_before"
    ]
)


main_events[
    "main_event_f2_rank"
] = (
    main_events[
        "f2_rank_before"
    ]
)


main_events[
    "main_event_ranked_fighters"
] = (
    main_events[
        "f1_rank_before"
    ].notna().astype(int)
    +
    main_events[
        "f2_rank_before"
    ].notna().astype(int)
)


main_event_event_level = (
    main_events[
        [
            "date",
            "event_name",
            "main_event_combined_ufc_fights",
            "main_event_avg_win_pct",
            "main_event_combined_win_streak",
            "main_event_combined_title_fights",
            "main_event_combined_previous_main_events",
            "main_event_combined_bonuses",
            "main_event_f1_rank",
            "main_event_f2_rank",
            "main_event_ranked_fighters"
        ]
    ]
    .rename(
        columns={
            "date": "event_date"
        }
    )
)


event_master_updated = (
    event_master_updated
    .merge(
        main_event_event_level,
        on=[
            "event_date",
            "event_name"
        ],
        how="left"
    )
)


# ============================================================
# PART 7 — RANKING POINT FUNCTION
# ============================================================

def rank_points(rank):

    if pd.isna(rank):
        return 0

    rank = float(rank)

    if rank == 0:
        return 16

    if 1 <= rank <= 15:
        return 16 - rank

    return 0


event_fighters_long[
    "ranking_points"
] = (
    event_fighters_long[
        "rank_before"
    ].apply(
        rank_points
    )
)


ranking_points_event = (
    event_fighters_long
    .groupby(
        [
            "event_date",
            "event_name"
        ],
        as_index=False
    )
    .agg(

        total_ranking_points=(
            "ranking_points",
            "sum"
        ),

        avg_ranking_points=(
            "ranking_points",
            "mean"
        )
    )
)


event_master_updated = (
    event_master_updated
    .merge(
        ranking_points_event,
        on=[
            "event_date",
            "event_name"
        ],
        how="left"
    )
)


event_master_updated[
    "main_event_f1_rank_points"
] = (
    event_master_updated[
        "main_event_f1_rank"
    ].apply(
        rank_points
    )
)


event_master_updated[
    "main_event_f2_rank_points"
] = (
    event_master_updated[
        "main_event_f2_rank"
    ].apply(
        rank_points
    )
)


event_master_updated[
    "main_event_ranking_points"
] = (
    event_master_updated[
        "main_event_f1_rank_points"
    ]
    +
    event_master_updated[
        "main_event_f2_rank_points"
    ]
)


# ============================================================
# PART 8 — AVAILABILITY FLAGS
# ============================================================

event_master_updated[
    "rankings_available"
] = (
    event_master_updated[
        "event_date"
    ]
    >= pd.Timestamp(
        "2013-02-04"
    )
)


event_master_updated[
    "bonus_history_available"
] = (
    event_master_updated[
        "event_date"
    ]
    >= pd.Timestamp(
        "2006-01-16"
    )
)


# Trends flag will be updated after we finish the new
# fanfare/search system.
event_master_updated[
    "trends_available"
] = False


# ============================================================
# PART 9 — SORT + VALIDATE
# ============================================================

event_master_updated = (
    event_master_updated
    .sort_values(
        [
            "event_date",
            "event_name"
        ]
    )
    .reset_index(drop=True)
)


duplicate_events = (
    event_master_updated
    .duplicated(
        subset=[
            "event_date",
            "event_name"
        ]
    )
    .sum()
)


verified_event_count = (
    event_master_updated[
        "main_event_combined_ufc_fights"
    ].notna().sum()
)


unverified_events = (
    event_master_updated[
        event_master_updated[
            "main_event_combined_ufc_fights"
        ].isna()
    ]
    [
        [
            "event_date",
            "event_name"
        ]
    ]
)


# ============================================================
# PART 10 — NEW EVENTS AUDIT
# ============================================================

new_event_audit = (
    event_master_updated[
        event_master_updated[
            "event_date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "total_fights",
            "title_fights",
            "ranked_fighters",
            "ranked_matchups",
            "avg_ufc_experience",
            "avg_previous_title_fights",
            "avg_previous_main_events",
            "main_event_combined_ufc_fights",
            "main_event_combined_bonuses",
            "total_ranking_points",
            "main_event_ranking_points"
        ]
    ]
)


# ============================================================
# PART 11 — OUTPUT
# ============================================================

print()
print("=" * 105)
print("EVENT MASTER VALIDATION")
print("=" * 105)

print(
    "Total events:",
    len(event_master_updated)
)

print(
    "Expected:",
    788
)

print(
    "Unique event/date combinations:",
    event_master_updated[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)

print(
    "Duplicate events:",
    int(
        duplicate_events
    )
)

print(
    "Verified main-event events:",
    int(
        verified_event_count
    )
)

print(
    "Expected verified:",
    785
)

print(
    "Unverified events:",
    len(unverified_events)
)


print()
print("UNVERIFIED EVENTS:")

print(
    unverified_events
    .to_string(
        index=False
    )
)


print()
print("=" * 105)
print("NEW 11 EVENTS")
print("=" * 105)

print(
    new_event_audit
    .to_string(
        index=False
    )
)


# ============================================================
# PART 12 — HISTORICAL PRESERVATION CHECK
#
# Compare shared numerical fields against our original
# pre-update event_master.
# ============================================================

shared_columns = [
    col
    for col in event_master_pre_update.columns
    if col in event_master_updated.columns
    and col not in [
        "event_date",
        "event_name",
        "trends_available"
    ]
]


old_compare = (
    event_master_pre_update.copy()
)


# Handle old date-column naming safely.
if (
    "date" in old_compare.columns
    and
    "event_date" not in old_compare.columns
):

    old_compare = old_compare.rename(
        columns={
            "date": "event_date"
        }
    )


comparison = (
    old_compare[
        [
            "event_date",
            "event_name"
        ]
        +
        shared_columns
    ]
    .merge(
        event_master_updated[
            [
                "event_date",
                "event_name"
            ]
            +
            shared_columns
        ],
        on=[
            "event_date",
            "event_name"
        ],
        how="inner",
        suffixes=(
            "_old",
            "_new"
        )
    )
)


historical_mismatches = {}


for col in shared_columns:

    old_col = (
        comparison[
            col + "_old"
        ]
    )

    new_col = (
        comparison[
            col + "_new"
        ]
    )


    if (
        pd.api.types
        .is_numeric_dtype(old_col)
        and
        pd.api.types
        .is_numeric_dtype(new_col)
    ):

        equal = np.isclose(
            pd.to_numeric(
                old_col,
                errors="coerce"
            ),
            pd.to_numeric(
                new_col,
                errors="coerce"
            ),
            equal_nan=True
        )

    else:

        equal = (
            old_col.fillna(
                "__NA__"
            ).astype(str)
            ==
            new_col.fillna(
                "__NA__"
            ).astype(str)
        )


    mismatch_count = int(
        (~equal).sum()
    )


    if mismatch_count > 0:

        historical_mismatches[
            col
        ] = mismatch_count


print()
print("=" * 105)
print("HISTORICAL PRESERVATION")
print("=" * 105)

print(
    "Old events matched:",
    comparison[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)

print(
    "Expected old events:",
    777
)

print(
    "Columns with historical mismatches:",
    len(
        historical_mismatches
    )
)


if historical_mismatches:

    print(
        historical_mismatches
    )

else:

    print(
        "Historical event metrics preserved exactly ✅"
    )


# ============================================================
# PART 13 — FINAL PASS / FAIL
# ============================================================

event_master_pass = (

    len(
        event_master_updated
    ) == 788

    and

    duplicate_events == 0

    and

    verified_event_count == 785

    and

    len(
        unverified_events
    ) == 3

    and

    comparison[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0] == 777

    and

    len(
        historical_mismatches
    ) == 0
)


print()
print("=" * 105)

if event_master_pass:

    print(
        "STEP 83 EVENT MASTER: PASS ✅🔥"
    )

else:

    print(
        "STEP 83 EVENT MASTER: CHECK REQUIRED ⚠️"
    )

print("=" * 105)
```

```python
# ============================================================
# STEP 84 — DIAGNOSE HISTORICAL PRESERVATION MISMATCHES
# ============================================================

import pandas as pd
import numpy as np


print("=" * 110)
print("STEP 84 — HISTORICAL MISMATCH DIAGNOSTIC")
print("=" * 110)


# ============================================================
# PART 1 — CREATE MATCH KEYS
# ============================================================

old = ufc_fights.copy()
new = ufc_fights_updated[
    ufc_fights_updated["date"] <= pd.Timestamp("2026-06-27")
].copy()


def build_fight_key(df):

    return (
        pd.to_datetime(df["date"])
        .dt.strftime("%Y-%m-%d")
        + "||"
        + df["event_name"].astype(str)
        + "||"
        + df["f1_name"].astype(str)
        + "||"
        + df["f2_name"].astype(str)
    )


old["_fight_key"] = build_fight_key(old)
new["_fight_key"] = build_fight_key(new)


print()
print("Original historical fights:", len(old))
print("Rebuilt historical fights:", len(new))

print(
    "Original unique keys:",
    old["_fight_key"].nunique()
)

print(
    "Rebuilt unique keys:",
    new["_fight_key"].nunique()
)


old_keys = set(old["_fight_key"])
new_keys = set(new["_fight_key"])

print(
    "Old fights missing from rebuilt:",
    len(old_keys - new_keys)
)

print(
    "Unexpected rebuilt historical fights:",
    len(new_keys - old_keys)
)


# ============================================================
# PART 2 — COMPARE ALL IMPORTANT DERIVED COLUMNS
# ============================================================

derived_columns = [

    "f1_ufc_wins_before",
    "f1_ufc_losses_before",
    "f2_ufc_wins_before",
    "f2_ufc_losses_before",

    "f1_ufc_fights_before",
    "f2_ufc_fights_before",

    "f1_win_streak_before",
    "f2_win_streak_before",

    "is_title_fight",

    "f1_previous_title_fights",
    "f2_previous_title_fights",

    "is_main_event",
    "is_main_event_verified",

    "f1_ufc_win_pct_before",
    "f2_ufc_win_pct_before",

    "f1_last5_win_pct",
    "f2_last5_win_pct",

    "f1_finish_rate_before",
    "f2_finish_rate_before",

    "f1_ufc_finishes_before",
    "f2_ufc_finishes_before",

    "f1_rank_before",
    "f2_rank_before",

    "f1_previous_main_events",
    "f2_previous_main_events",

    "FOTN",
    "POTN",
    "KOTN",
    "SOTN",

    "f1_previous_FOTN",
    "f1_previous_POTN",
    "f1_previous_KOTN",
    "f1_previous_SOTN",

    "f2_previous_FOTN",
    "f2_previous_POTN",
    "f2_previous_KOTN",
    "f2_previous_SOTN",

    "f1_previous_total_bonuses",
    "f2_previous_total_bonuses"
]


derived_columns = [
    col
    for col in derived_columns
    if col in old.columns
    and col in new.columns
]


compare = (
    old[
        ["_fight_key", "date", "event_name", "f1_name", "f2_name"]
        + derived_columns
    ]
    .merge(
        new[
            ["_fight_key"]
            + derived_columns
        ],
        on="_fight_key",
        how="inner",
        suffixes=("_old", "_new")
    )
)


print()
print("=" * 110)
print("FIGHT-LEVEL COLUMN MISMATCH COUNTS")
print("=" * 110)


fight_mismatch_counts = {}


for col in derived_columns:

    a = compare[col + "_old"]
    b = compare[col + "_new"]


    # Numeric / boolean safe comparison
    try:

        a_num = pd.to_numeric(
            a,
            errors="coerce"
        )

        b_num = pd.to_numeric(
            b,
            errors="coerce"
        )


        equal = np.isclose(
            a_num,
            b_num,
            equal_nan=True
        )

    except:

        equal = (
            a.fillna("__NA__").astype(str)
            ==
            b.fillna("__NA__").astype(str)
        )


    mismatch_count = int(
        (~equal).sum()
    )


    fight_mismatch_counts[col] = (
        mismatch_count
    )


    print(
        f"{col:35s}",
        mismatch_count
    )


# ============================================================
# PART 3 — FOCUS ON FINISH HISTORY
# ============================================================

print()
print("=" * 110)
print("FINISH-HISTORY DIAGNOSTIC")
print("=" * 110)


finish_cols = [

    "f1_ufc_finishes_before",
    "f2_ufc_finishes_before",

    "f1_finish_rate_before",
    "f2_finish_rate_before"
]


for col in finish_cols:

    if col not in derived_columns:
        continue

    a = pd.to_numeric(
        compare[col + "_old"],
        errors="coerce"
    )

    b = pd.to_numeric(
        compare[col + "_new"],
        errors="coerce"
    )


    mismatch = ~np.isclose(
        a,
        b,
        equal_nan=True
    )


    print()
    print(
        col,
        "mismatching fights:",
        int(mismatch.sum())
    )


    if mismatch.sum() > 0:

        sample = compare.loc[
            mismatch,
            [
                "date",
                "event_name",
                "f1_name",
                "f2_name",
                col + "_old",
                col + "_new"
            ]
        ].head(12)


        print(
            sample.to_string(
                index=False
            )
        )


# ============================================================
# PART 4 — CHECK METHOD CLASSIFICATION
#
# We want to see every historical method label because
# finish-history differences may be caused by how finishes
# were classified during the rebuild.
# ============================================================

print()
print("=" * 110)
print("HISTORICAL METHOD LABELS")
print("=" * 110)


method_counts = (
    old["method"]
    .fillna("MISSING")
    .value_counts()
)


print(
    method_counts.to_string()
)


# ============================================================
# PART 5 — IDENTIFY POSSIBLE FINISH LABELS
# ============================================================

print()
print("=" * 110)
print("METHODS CONTAINING KO / SUBMISSION / TKO")
print("=" * 110)


possible_finish_methods = (
    old.loc[
        old["method"]
        .astype(str)
        .str.contains(
            "KO|TKO|SUB",
            case=False,
            na=False
        ),
        "method"
    ]
    .value_counts()
)


print(
    possible_finish_methods.to_string()
)


# ============================================================
# PART 6 — EVENT-LEVEL SMALL MISMATCH DETAILS
# ============================================================

print()
print("=" * 110)
print("EVENT-LEVEL MISMATCH DETAILS")
print("=" * 110)


for col, count in historical_mismatches.items():

    print()
    print("-" * 110)

    print(
        col,
        "→",
        count,
        "events"
    )


    old_col = comparison[
        col + "_old"
    ]

    new_col = comparison[
        col + "_new"
    ]


    if (
        pd.api.types.is_numeric_dtype(old_col)
        and
        pd.api.types.is_numeric_dtype(new_col)
    ):

        unequal = ~np.isclose(
            pd.to_numeric(
                old_col,
                errors="coerce"
            ),
            pd.to_numeric(
                new_col,
                errors="coerce"
            ),
            equal_nan=True
        )

    else:

        unequal = (
            old_col.fillna("__NA__").astype(str)
            !=
            new_col.fillna("__NA__").astype(str)
        )


    details = comparison.loc[
        unequal,
        [
            "event_date",
            "event_name",
            col + "_old",
            col + "_new"
        ]
    ]


    print(
        details
        .head(15)
        .to_string(
            index=False
        )
    )


# ============================================================
# PART 7 — SUMMARY
# ============================================================

print()
print("=" * 110)
print("DIAGNOSTIC SUMMARY")
print("=" * 110)


nonzero_fight_mismatches = {
    col: count
    for col, count
    in fight_mismatch_counts.items()
    if count > 0
}


print(
    "Fight-level derived columns with mismatches:",
    len(nonzero_fight_mismatches)
)


print()

print(
    nonzero_fight_mismatches
)


print()
print(
    "NO DATA WAS CHANGED BY STEP 84 ✅"
)
```

```python
# ============================================================
# STEP 85 — RESTORE VALIDATED HISTORICAL DERIVED DATA
# ============================================================

import pandas as pd
import numpy as np


print("=" * 110)
print("STEP 85 — RESTORE HISTORICAL BACKBONE 🔥")
print("=" * 110)


# ============================================================
# PART 1 — SAFETY COPY
# ============================================================

ufc_fights_before_step85 = (
    ufc_fights_updated.copy()
)


CUTOFF = pd.Timestamp("2026-06-27")


# ============================================================
# PART 2 — IDENTIFY HISTORICAL VS NEW ROWS
# ============================================================

historical_mask = (
    ufc_fights_updated["date"] <= CUTOFF
)

new_mask = (
    ufc_fights_updated["date"] > CUTOFF
)


print(
    "Historical rows:",
    int(historical_mask.sum())
)

print(
    "Expected:",
    8736
)

print(
    "New rows:",
    int(new_mask.sum())
)

print(
    "Expected:",
    141
)


# ============================================================
# PART 3 — CREATE A STABLE HISTORICAL ROW ID
#
# We cannot rely only on fighter/date keys because the
# diagnostic found one duplicate fight key.
#
# cumcount gives duplicate keys an occurrence number.
# ============================================================

def add_restore_key(df):

    x = df.copy()

    x["_restore_base"] = (
        pd.to_datetime(x["date"])
        .dt.strftime("%Y-%m-%d")
        + "||"
        + x["event_name"].astype(str)
        + "||"
        + x["f1_name"].astype(str)
        + "||"
        + x["f2_name"].astype(str)
    )

    x["_restore_occurrence"] = (
        x.groupby(
            "_restore_base",
            sort=False
        )
        .cumcount()
    )

    x["_restore_key"] = (
        x["_restore_base"]
        + "||"
        + x["_restore_occurrence"].astype(str)
    )

    return x


original_history = add_restore_key(
    ufc_fights.copy()
)


rebuilt_history = add_restore_key(
    ufc_fights_updated.loc[
        historical_mask
    ].copy()
)


# ============================================================
# PART 4 — VALIDATE EXACT ROW MATCHING BEFORE TOUCHING DATA
# ============================================================

old_keys = set(
    original_history["_restore_key"]
)

rebuilt_keys = set(
    rebuilt_history["_restore_key"]
)


print()
print("=" * 110)
print("ROW-MATCH VALIDATION")
print("=" * 110)

print(
    "Original restore keys:",
    len(old_keys)
)

print(
    "Rebuilt restore keys:",
    len(rebuilt_keys)
)

print(
    "Missing old keys:",
    len(old_keys - rebuilt_keys)
)

print(
    "Unexpected rebuilt keys:",
    len(rebuilt_keys - old_keys)
)


if (
    len(old_keys - rebuilt_keys) != 0
    or
    len(rebuilt_keys - old_keys) != 0
):

    raise ValueError(
        "Historical row matching failed. "
        "Nothing should be restored until this is resolved."
    )


# ============================================================
# PART 5 — DEFINE HISTORICAL DERIVED COLUMNS TO PRESERVE
#
# Every original derived column (16–59) is authoritative
# for historical rows.
#
# Raw columns remain from our updated combined dataset.
# ============================================================

raw_columns = [
    "fight_url",
    "event_name",
    "date",
    "location",
    "f1_name",
    "f1_result",
    "f2_name",
    "f2_result",
    "weight_class",
    "method",
    "round",
    "time",
    "time_format",
    "referee",
    "judging_details"
]


historical_derived_columns = [
    col
    for col in ufc_fights.columns
    if col not in raw_columns
]


print()
print(
    "Historical derived columns to restore:",
    len(historical_derived_columns)
)


# ============================================================
# PART 6 — BUILD AUTHORITATIVE HISTORICAL LOOKUP
# ============================================================

history_lookup = (
    original_history[
        ["_restore_key"]
        +
        historical_derived_columns
    ]
    .copy()
)


restored_history = (
    rebuilt_history
    .drop(
        columns=[
            col
            for col in historical_derived_columns
            if col in rebuilt_history.columns
        ]
    )
    .merge(
        history_lookup,
        on="_restore_key",
        how="left",
        validate="one_to_one"
    )
)


# ============================================================
# PART 7 — REMOVE TEMPORARY KEYS
# ============================================================

temp_cols = [
    "_restore_base",
    "_restore_occurrence",
    "_restore_key"
]


restored_history = (
    restored_history
    .drop(
        columns=temp_cols,
        errors="ignore"
    )
)


# ============================================================
# PART 8 — KEEP NEW 141 ROWS EXACTLY AS CURRENTLY CALCULATED
# ============================================================

new_history = (
    ufc_fights_updated.loc[
        new_mask
    ]
    .copy()
)


# ============================================================
# PART 9 — ALIGN COLUMN SET
#
# event_name_original is update provenance and may not exist
# in the old historical dataset. Keep it.
# ============================================================

all_columns = list(
    dict.fromkeys(
        list(restored_history.columns)
        +
        list(new_history.columns)
    )
)


restored_history = (
    restored_history.reindex(
        columns=all_columns
    )
)


new_history = (
    new_history.reindex(
        columns=all_columns
    )
)


# ============================================================
# PART 10 — RECOMBINE
# ============================================================

ufc_fights_updated = (
    pd.concat(
        [
            restored_history,
            new_history
        ],
        ignore_index=True
    )
    .sort_values(
        [
            "date",
            "event_name"
        ],
        kind="stable"
    )
    .reset_index(drop=True)
)


# ============================================================
# PART 11 — REAPPLY UFC 331 STANDARDIZED NAME
#
# Historical rows are untouched.
# ============================================================

ufc_fights_updated.loc[
    (
        ufc_fights_updated["date"]
        == pd.Timestamp("2026-09-19")
    )
    &
    (
        ufc_fights_updated["event_name"]
        .astype(str)
        .str.contains(
            "UFC 331",
            na=False
        )
    ),
    "event_name"
] = "UFC 331: Van vs. Pantoja 2"


# ============================================================
# PART 12 — HISTORICAL EXACT-PRESERVATION AUDIT
# ============================================================

audit_history = add_restore_key(
    ufc_fights_updated[
        ufc_fights_updated["date"]
        <= CUTOFF
    ].copy()
)


audit_compare = (
    original_history[
        ["_restore_key"]
        +
        historical_derived_columns
    ]
    .merge(
        audit_history[
            ["_restore_key"]
            +
            historical_derived_columns
        ],
        on="_restore_key",
        how="inner",
        suffixes=(
            "_old",
            "_updated"
        ),
        validate="one_to_one"
    )
)


column_mismatches = {}


for col in historical_derived_columns:

    old_col = audit_compare[
        col + "_old"
    ]

    new_col = audit_compare[
        col + "_updated"
    ]


    if (
        pd.api.types.is_numeric_dtype(old_col)
        and
        pd.api.types.is_numeric_dtype(new_col)
    ):

        equal = np.isclose(
            pd.to_numeric(
                old_col,
                errors="coerce"
            ),
            pd.to_numeric(
                new_col,
                errors="coerce"
            ),
            equal_nan=True
        )

    else:

        equal = (
            old_col.fillna(
                "__NA__"
            ).astype(str)
            ==
            new_col.fillna(
                "__NA__"
            ).astype(str)
        )


    mismatch_count = int(
        (~equal).sum()
    )


    if mismatch_count > 0:

        column_mismatches[
            col
        ] = mismatch_count


# ============================================================
# PART 13 — NEW DATA PRESERVATION AUDIT
# ============================================================

new_check = (
    ufc_fights_updated[
        ufc_fights_updated["date"]
        > CUTOFF
    ]
)


new_ranked_entries = int(
    new_check["f1_rank_before"].notna().sum()
    +
    new_check["f2_rank_before"].notna().sum()
)


new_champions = int(
    new_check["f1_rank_before"].eq(0).sum()
    +
    new_check["f2_rank_before"].eq(0).sum()
)


# ============================================================
# PART 14 — OUTPUT
# ============================================================

print()
print("=" * 110)
print("RESTORATION RESULTS")
print("=" * 110)

print(
    "Total fights:",
    len(ufc_fights_updated)
)

print(
    "Expected:",
    8877
)

print(
    "Historical fights:",
    (
        ufc_fights_updated["date"]
        <= CUTOFF
    ).sum()
)

print(
    "New fights:",
    (
        ufc_fights_updated["date"]
        > CUTOFF
    ).sum()
)


print()
print(
    "Historical derived columns checked:",
    len(historical_derived_columns)
)

print(
    "Columns with historical mismatches:",
    len(column_mismatches)
)

if column_mismatches:

    print(
        column_mismatches
    )

else:

    print(
        "ALL HISTORICAL DERIVED VALUES PRESERVED EXACTLY ✅"
    )


print()
print(
    "New ranked fighter entries:",
    new_ranked_entries,
    "| expected 63"
)

print(
    "New champions:",
    new_champions,
    "| expected 3"
)

print(
    "New title fights:",
    int(
        new_check[
            "is_title_fight"
        ].sum()
    ),
    "| expected 3"
)

print(
    "New verified main events:",
    int(
        new_check[
            "is_main_event_verified"
        ].sum()
    ),
    "| expected 11"
)


# ============================================================
# PART 15 — FINAL PASS / FAIL
# ============================================================

step85_pass = (

    len(ufc_fights_updated) == 8877

    and

    (
        ufc_fights_updated["date"]
        <= CUTOFF
    ).sum() == 8736

    and

    (
        ufc_fights_updated["date"]
        > CUTOFF
    ).sum() == 141

    and

    len(column_mismatches) == 0

    and

    new_ranked_entries == 63

    and

    new_champions == 3

    and

    int(
        new_check[
            "is_title_fight"
        ].sum()
    ) == 3

    and

    int(
        new_check[
            "is_main_event_verified"
        ].sum()
    ) == 11
)


print()
print("=" * 110)

if step85_pass:

    print(
        "STEP 85 HISTORICAL RESTORATION: PASS ✅🔥"
    )

else:

    print(
        "STEP 85 HISTORICAL RESTORATION: CHECK REQUIRED ⚠️"
    )

print("=" * 110)
```

```python
# ============================================================
# STEP 86 — REBUILD EVENT MASTER AFTER HISTORICAL RESTORATION
# ============================================================

import pandas as pd
import numpy as np


print("=" * 110)
print("STEP 86 — AUTHORITATIVE 788-EVENT MASTER 🔥")
print("=" * 110)


# ============================================================
# PART 1 — EVENT BASE
# ============================================================

event_master_updated = (
    ufc_fights_updated[
        ["date", "event_name"]
    ]
    .drop_duplicates()
    .rename(columns={"date": "event_date"})
    .sort_values(
        ["event_date", "event_name"]
    )
    .reset_index(drop=True)
)


# ============================================================
# PART 2 — BUILD LONG FIGHTER TABLE
# ============================================================

def fighter_side(prefix):

    return pd.DataFrame({

        "event_date":
            ufc_fights_updated["date"],

        "event_name":
            ufc_fights_updated["event_name"],

        "fighter":
            ufc_fights_updated[f"{prefix}_name"],

        "ufc_fights_before":
            ufc_fights_updated[
                f"{prefix}_ufc_fights_before"
            ],

        "ufc_win_pct_before":
            ufc_fights_updated[
                f"{prefix}_ufc_win_pct_before"
            ],

        "win_streak_before":
            ufc_fights_updated[
                f"{prefix}_win_streak_before"
            ],

        "last5_win_pct":
            ufc_fights_updated[
                f"{prefix}_last5_win_pct"
            ],

        "finish_rate_before":
            ufc_fights_updated[
                f"{prefix}_finish_rate_before"
            ],

        "ufc_finishes_before":
            ufc_fights_updated[
                f"{prefix}_ufc_finishes_before"
            ],

        "previous_title_fights":
            ufc_fights_updated[
                f"{prefix}_previous_title_fights"
            ],

        "previous_main_events":
            ufc_fights_updated[
                f"{prefix}_previous_main_events"
            ],

        "previous_bonuses":
            ufc_fights_updated[
                f"{prefix}_previous_total_bonuses"
            ],

        "rank_before":
            ufc_fights_updated[
                f"{prefix}_rank_before"
            ]
    })


event_fighters_long = pd.concat(
    [
        fighter_side("f1"),
        fighter_side("f2")
    ],
    ignore_index=True
)


# ============================================================
# PART 3 — BASIC CARD COUNTS
# ============================================================

basic = (
    ufc_fights_updated
    .groupby(
        ["date", "event_name"],
        as_index=False
    )
    .agg(
        total_fights=("event_name", "size"),
        title_fights=("is_title_fight", "sum")
    )
    .rename(columns={"date": "event_date"})
)


event_master_updated = (
    event_master_updated.merge(
        basic,
        on=["event_date", "event_name"],
        how="left",
        validate="one_to_one"
    )
)


# ============================================================
# PART 4 — RANKED FIGHTERS / MATCHUPS
# ============================================================

rank_temp = ufc_fights_updated[
    [
        "date",
        "event_name",
        "f1_rank_before",
        "f2_rank_before"
    ]
].copy()


rank_temp["ranked_fighters"] = (
    rank_temp["f1_rank_before"]
    .notna()
    .astype(int)
    +
    rank_temp["f2_rank_before"]
    .notna()
    .astype(int)
)


rank_temp["ranked_matchups"] = (
    rank_temp["f1_rank_before"].notna()
    &
    rank_temp["f2_rank_before"].notna()
).astype(int)


rank_card = (
    rank_temp
    .groupby(
        ["date", "event_name"],
        as_index=False
    )
    .agg(
        ranked_fighters=(
            "ranked_fighters",
            "sum"
        ),
        ranked_matchups=(
            "ranked_matchups",
            "sum"
        )
    )
    .rename(columns={"date": "event_date"})
)


event_master_updated = (
    event_master_updated.merge(
        rank_card,
        on=["event_date", "event_name"],
        how="left",
        validate="one_to_one"
    )
)


# ============================================================
# PART 5 — CARD-WIDE FIGHTER METRICS
# ============================================================

card_metrics = (
    event_fighters_long
    .groupby(
        ["event_date", "event_name"],
        as_index=False
    )
    .agg(

        avg_ufc_experience=(
            "ufc_fights_before",
            "mean"
        ),

        total_prior_ufc_fights=(
            "ufc_fights_before",
            "sum"
        ),

        avg_ufc_win_pct=(
            "ufc_win_pct_before",
            "mean"
        ),

        avg_win_streak=(
            "win_streak_before",
            "mean"
        ),

        avg_last5_win_pct=(
            "last5_win_pct",
            "mean"
        ),

        avg_finish_rate=(
            "finish_rate_before",
            "mean"
        ),

        total_prior_ufc_finishes=(
            "ufc_finishes_before",
            "sum"
        ),

        avg_previous_title_fights=(
            "previous_title_fights",
            "mean"
        ),

        total_previous_title_fights=(
            "previous_title_fights",
            "sum"
        ),

        avg_previous_main_events=(
            "previous_main_events",
            "mean"
        ),

        total_previous_main_events=(
            "previous_main_events",
            "sum"
        ),

        avg_previous_bonuses=(
            "previous_bonuses",
            "mean"
        ),

        total_previous_bonuses=(
            "previous_bonuses",
            "sum"
        )
    )
)


event_master_updated = (
    event_master_updated.merge(
        card_metrics,
        on=["event_date", "event_name"],
        how="left",
        validate="one_to_one"
    )
)


# ============================================================
# PART 6 — VERIFIED MAIN EVENTS
# ============================================================

main = (
    ufc_fights_updated[
        ufc_fights_updated[
            "is_main_event_verified"
        ].fillna(False)
    ]
    .copy()
)


main["main_event_combined_ufc_fights"] = (
    main["f1_ufc_fights_before"]
    +
    main["f2_ufc_fights_before"]
)


main["main_event_avg_win_pct"] = (
    main[
        [
            "f1_ufc_win_pct_before",
            "f2_ufc_win_pct_before"
        ]
    ]
    .mean(axis=1)
)


main["main_event_combined_win_streak"] = (
    main["f1_win_streak_before"]
    +
    main["f2_win_streak_before"]
)


main["main_event_combined_title_fights"] = (
    main["f1_previous_title_fights"]
    +
    main["f2_previous_title_fights"]
)


main[
    "main_event_combined_previous_main_events"
] = (
    main["f1_previous_main_events"]
    +
    main["f2_previous_main_events"]
)


main["main_event_combined_bonuses"] = (
    main["f1_previous_total_bonuses"]
    +
    main["f2_previous_total_bonuses"]
)


main["main_event_f1_rank"] = (
    main["f1_rank_before"]
)

main["main_event_f2_rank"] = (
    main["f2_rank_before"]
)


main["main_event_ranked_fighters"] = (
    main["f1_rank_before"]
    .notna()
    .astype(int)
    +
    main["f2_rank_before"]
    .notna()
    .astype(int)
)


main_event_metrics = (
    main[
        [
            "date",
            "event_name",
            "main_event_combined_ufc_fights",
            "main_event_avg_win_pct",
            "main_event_combined_win_streak",
            "main_event_combined_title_fights",
            "main_event_combined_previous_main_events",
            "main_event_combined_bonuses",
            "main_event_f1_rank",
            "main_event_f2_rank",
            "main_event_ranked_fighters"
        ]
    ]
    .rename(columns={"date": "event_date"})
)


# There should be only one verified main event per event.
duplicate_main = (
    main_event_metrics
    .duplicated(
        subset=[
            "event_date",
            "event_name"
        ]
    )
    .sum()
)


if duplicate_main:
    raise ValueError(
        f"Found {duplicate_main} duplicate verified "
        "main-event rows."
    )


event_master_updated = (
    event_master_updated.merge(
        main_event_metrics,
        on=["event_date", "event_name"],
        how="left",
        validate="one_to_one"
    )
)


# ============================================================
# PART 7 — RANKING POINTS
# ============================================================

def rank_points(rank):

    if pd.isna(rank):
        return 0

    rank = float(rank)

    if rank == 0:
        return 16

    if 1 <= rank <= 15:
        return 16 - rank

    return 0


event_fighters_long["ranking_points"] = (
    event_fighters_long[
        "rank_before"
    ].apply(rank_points)
)


ranking_points = (
    event_fighters_long
    .groupby(
        ["event_date", "event_name"],
        as_index=False
    )
    .agg(
        total_ranking_points=(
            "ranking_points",
            "sum"
        ),
        avg_ranking_points=(
            "ranking_points",
            "mean"
        )
    )
)


event_master_updated = (
    event_master_updated.merge(
        ranking_points,
        on=["event_date", "event_name"],
        how="left",
        validate="one_to_one"
    )
)


event_master_updated[
    "main_event_f1_rank_points"
] = (
    event_master_updated[
        "main_event_f1_rank"
    ].apply(rank_points)
)


event_master_updated[
    "main_event_f2_rank_points"
] = (
    event_master_updated[
        "main_event_f2_rank"
    ].apply(rank_points)
)


event_master_updated[
    "main_event_ranking_points"
] = (
    event_master_updated[
        "main_event_f1_rank_points"
    ]
    +
    event_master_updated[
        "main_event_f2_rank_points"
    ]
)


# ============================================================
# PART 8 — AVAILABILITY FLAGS
# ============================================================

event_master_updated[
    "rankings_available"
] = (
    event_master_updated["event_date"]
    >= pd.Timestamp("2013-02-04")
)


event_master_updated[
    "bonus_history_available"
] = (
    event_master_updated["event_date"]
    >= pd.Timestamp("2006-01-16")
)


# Search/fanfare availability will be attached later.
event_master_updated[
    "trends_available"
] = False


# ============================================================
# PART 9 — SORT
# ============================================================

event_master_updated = (
    event_master_updated
    .sort_values(
        ["event_date", "event_name"]
    )
    .reset_index(drop=True)
)


# ============================================================
# PART 10 — HISTORICAL EVENT-MASTER COMPARISON
# ============================================================

old_master = (
    event_master_pre_update.copy()
)


if (
    "date" in old_master.columns
    and
    "event_date" not in old_master.columns
):

    old_master = old_master.rename(
        columns={"date": "event_date"}
    )


shared_columns = [
    col
    for col in old_master.columns
    if (
        col in event_master_updated.columns
        and col not in [
            "event_date",
            "event_name",
            "trends_available"
        ]
    )
]


historical_comparison = (
    old_master[
        ["event_date", "event_name"]
        + shared_columns
    ]
    .merge(
        event_master_updated[
            ["event_date", "event_name"]
            + shared_columns
        ],
        on=["event_date", "event_name"],
        how="inner",
        suffixes=("_old", "_new"),
        validate="one_to_one"
    )
)


historical_mismatches_86 = {}


for col in shared_columns:

    old_col = (
        historical_comparison[
            col + "_old"
        ]
    )

    new_col = (
        historical_comparison[
            col + "_new"
        ]
    )


    if (
        pd.api.types.is_numeric_dtype(old_col)
        and
        pd.api.types.is_numeric_dtype(new_col)
    ):

        equal = np.isclose(
            pd.to_numeric(
                old_col,
                errors="coerce"
            ),
            pd.to_numeric(
                new_col,
                errors="coerce"
            ),
            equal_nan=True
        )

    else:

        equal = (
            old_col.fillna("__NA__")
            .astype(str)
            ==
            new_col.fillna("__NA__")
            .astype(str)
        )


    mismatches = int(
        (~equal).sum()
    )


    if mismatches:

        historical_mismatches_86[
            col
        ] = mismatches


# ============================================================
# PART 11 — GLOBAL VALIDATION
# ============================================================

total_events = len(
    event_master_updated
)


unique_events = (
    event_master_updated[
        ["event_date", "event_name"]
    ]
    .drop_duplicates()
    .shape[0]
)


verified_events = int(
    event_master_updated[
        "main_event_combined_ufc_fights"
    ]
    .notna()
    .sum()
)


unverified = (
    event_master_updated[
        event_master_updated[
            "main_event_combined_ufc_fights"
        ].isna()
    ]
    [
        ["event_date", "event_name"]
    ]
)


new_events = (
    event_master_updated[
        event_master_updated[
            "event_date"
        ] > pd.Timestamp("2026-06-27")
    ]
)


# ============================================================
# PART 12 — OUTPUT
# ============================================================

print()
print("=" * 110)
print("GLOBAL EVENT-MASTER VALIDATION")
print("=" * 110)

print(
    "Total events:",
    total_events,
    "| expected 788"
)

print(
    "Unique events:",
    unique_events,
    "| expected 788"
)

print(
    "Verified main-event events:",
    verified_events,
    "| expected 785"
)

print(
    "Unverified events:",
    len(unverified),
    "| expected 3"
)

print(
    "New events:",
    len(new_events),
    "| expected 11"
)


print()
print("UNVERIFIED EVENTS:")

print(
    unverified.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("HISTORICAL EVENT-MASTER PRESERVATION")
print("=" * 110)

print(
    "Historical events matched:",
    len(historical_comparison),
    "| expected 777"
)

print(
    "Shared historical columns checked:",
    len(shared_columns)
)

print(
    "Columns with mismatches:",
    len(historical_mismatches_86)
)


if historical_mismatches_86:

    print(
        historical_mismatches_86
    )

else:

    print(
        "ALL HISTORICAL EVENT METRICS PRESERVED EXACTLY ✅"
    )


print()
print("=" * 110)
print("NEW EVENT QUICK CHECK")
print("=" * 110)

print(
    new_events[
        [
            "event_date",
            "event_name",
            "total_fights",
            "title_fights",
            "ranked_fighters",
            "ranked_matchups",
            "total_ranking_points",
            "main_event_ranking_points"
        ]
    ]
    .to_string(index=False)
)


# ============================================================
# PART 13 — PASS / FAIL
# ============================================================

step86_pass = (

    total_events == 788

    and

    unique_events == 788

    and

    verified_events == 785

    and

    len(unverified) == 3

    and

    len(new_events) == 11

    and

    len(historical_comparison) == 777

    and

    len(historical_mismatches_86) == 0
)


print()
print("=" * 110)

if step86_pass:

    print(
        "STEP 86 AUTHORITATIVE EVENT MASTER: PASS ✅🔥"
    )

else:

    print(
        "STEP 86 AUTHORITATIVE EVENT MASTER: CHECK REQUIRED ⚠️"
    )

print("=" * 110)
```

```python
# ============================================================
# STEP 87 — PRODUCTION GATE / COMMERCIAL DEMAND BACKBONE
# ============================================================

import pandas as pd
import numpy as np


print("=" * 110)
print("STEP 87 — GATE / COMMERCIAL DEMAND BACKBONE 🔥")
print("=" * 110)


# ============================================================
# PART 1 — SAFETY COPY
# ============================================================

gate_database_before_step87 = (
    gate_database.copy()
)


# ============================================================
# PART 2 — STANDARDIZE DATE + UFC 331 NAME
# ============================================================

# Handle either possible date-column name safely.
if (
    "date" in gate_database.columns
    and
    "event_date" not in gate_database.columns
):

    gate_database = (
        gate_database.rename(
            columns={
                "date": "event_date"
            }
        )
    )


gate_database[
    "event_date"
] = pd.to_datetime(
    gate_database[
        "event_date"
    ]
)


gate_database.loc[
    (
        gate_database[
            "event_date"
        ]
        == pd.Timestamp("2026-09-19")
    )
    &
    (
        gate_database[
            "event_name"
        ].astype(str)
        .str.contains(
            "UFC 331",
            na=False
        )
    ),
    "event_name"
] = "UFC 331: Van vs. Pantoja 2"


# ============================================================
# PART 3 — ALIGN GATE DATABASE TO AUTHORITATIVE EVENT MASTER
#
# event_master_updated is now our authoritative 788-event list.
# ============================================================

gate_columns_to_keep = [
    col
    for col in gate_database.columns
    if col not in [
        "gate_event_id",
        "gate_event_type",
        "gate_log_nominal",
        "gate_available",
        "gate_missing",
        "gate_log_score",
        "gate_era_score",
        "gate_commercial_score"
    ]
]


gate_source = (
    gate_database[
        gate_columns_to_keep
    ]
    .copy()
)


# Make sure there is exactly one source row per event.
gate_source_duplicates = (
    gate_source
    .duplicated(
        subset=[
            "event_date",
            "event_name"
        ]
    )
    .sum()
)


if gate_source_duplicates:

    raise ValueError(
        f"Gate source contains "
        f"{gate_source_duplicates} duplicate events."
    )


gate_database = (
    event_master_updated[
        [
            "event_date",
            "event_name"
        ]
    ]
    .merge(
        gate_source,
        on=[
            "event_date",
            "event_name"
        ],
        how="left",
        validate="one_to_one"
    )
)


# ============================================================
# PART 4 — REBUILD EVENT ID
# ============================================================

gate_database[
    "gate_event_id"
] = (
    gate_database[
        "event_date"
    ]
    .dt.strftime("%Y-%m-%d")
    +
    "||"
    +
    gate_database[
        "event_name"
    ].astype(str)
)


# ============================================================
# PART 5 — EVENT-TYPE CLASSIFIER
# ============================================================

def classify_gate_event(name):

    name = str(name)

    if (
        name.startswith("UFC Fight Night:")
        or
        name.startswith("UFC Fight Night ")
    ):
        return "Fight Night"

    if name.startswith("Noche UFC"):
        return "Noche UFC"

    if name.startswith("UFC on FOX"):
        return "UFC on FOX"

    if name.startswith("UFC on FX"):
        return "UFC on FX"

    if name.startswith("UFC on Fuel"):
        return "UFC on Fuel"

    if (
        "Ultimate Fighter" in name
        or
        "TUF" in name
    ):
        return "TUF Finale"

    # Numbered UFC:
    # UFC 1, UFC 100, UFC 331, etc.
    import re

    if re.match(
        r"^UFC\s+\d+",
        name
    ):
        return "Numbered UFC"

    return "Other"


gate_database[
    "gate_event_type"
] = (
    gate_database[
        "event_name"
    ].apply(
        classify_gate_event
    )
)


# ============================================================
# PART 6 — CLEAN GATE VALUES
#
# Missing stays NaN.
# Zero is NOT a legitimate substitute for missing.
# ============================================================

gate_database[
    "gate_usd"
] = pd.to_numeric(
    gate_database[
        "gate_usd"
    ],
    errors="coerce"
)


# Any accidental zero should become missing.
gate_database.loc[
    gate_database[
        "gate_usd"
    ] <= 0,
    "gate_usd"
] = np.nan


gate_database[
    "gate_available"
] = (
    gate_database[
        "gate_usd"
    ].notna()
)


gate_database[
    "gate_missing"
] = (
    ~gate_database[
        "gate_available"
    ]
)


# ============================================================
# PART 7 — LOG GATE
#
# Natural log compresses extreme dollar values while
# preserving ordering.
# ============================================================

gate_database[
    "gate_log_nominal"
] = np.where(
    gate_database[
        "gate_available"
    ],
    np.log1p(
        gate_database[
            "gate_usd"
        ]
    ),
    np.nan
)


# ============================================================
# PART 8 — DEFINE BROAD GATE ERAS
#
# These are NOT being used to force scores yet.
# They prepare us for era normalization once coverage improves.
# ============================================================

def gate_era(date):

    year = pd.Timestamp(
        date
    ).year

    if year <= 2000:
        return "1994-2000"

    if year <= 2010:
        return "2001-2010"

    if year <= 2013:
        return "2011-2013"

    if year <= 2018:
        return "2014-2018"

    return "2019-present"


gate_database[
    "gate_era"
] = (
    gate_database[
        "event_date"
    ].apply(
        gate_era
    )
)


# ============================================================
# PART 9 — COVERAGE BY ERA
# ============================================================

gate_era_coverage = (
    gate_database
    .groupby(
        "gate_era",
        as_index=False
    )
    .agg(

        events=(
            "event_name",
            "size"
        ),

        gates_available=(
            "gate_available",
            "sum"
        )
    )
)


gate_era_coverage[
    "coverage_pct"
] = (
    gate_era_coverage[
        "gates_available"
    ]
    /
    gate_era_coverage[
        "events"
    ]
    * 100
)


# ============================================================
# PART 10 — RAW LOG-GATE PERCENTILE
#
# IMPORTANT:
# This exists for diagnostics.
#
# We are NOT yet treating a percentile among only six known
# gates as our final production gate score.
# ============================================================

def zero_preserving_percentile(series):

    result = pd.Series(
        np.nan,
        index=series.index,
        dtype=float
    )

    valid = series.notna()

    if valid.sum() > 0:

        result.loc[
            valid
        ] = (
            series.loc[
                valid
            ]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )

    return result


gate_database[
    "gate_log_percentile_observed"
] = (
    zero_preserving_percentile(
        gate_database[
            "gate_log_nominal"
        ]
    )
)


# ============================================================
# PART 11 — ERA-ADJUSTED GATE
#
# Only calculate within an era if we have AT LEAST 10
# verified gates in that era.
#
# With our current six-gate database, this should mostly/all
# remain NaN. That is intentional.
# ============================================================

MIN_GATES_PER_ERA = 10


gate_database[
    "gate_era_percentile"
] = np.nan


for era, group in (
    gate_database.groupby(
        "gate_era"
    )
):

    valid = group[
        "gate_log_nominal"
    ].notna()

    n_valid = int(
        valid.sum()
    )


    if n_valid >= MIN_GATES_PER_ERA:

        idx = group.loc[
            valid
        ].index


        gate_database.loc[
            idx,
            "gate_era_percentile"
        ] = (
            group.loc[
                valid,
                "gate_log_nominal"
            ]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )


# ============================================================
# PART 12 — PRODUCTION GATE SIGNAL
#
# For now:
#
# gate_commercial_signal = log nominal gate
#
# This is the actual information we know.
#
# Later, once gate coverage is sufficient, we can combine:
# - inflation-adjusted gate
# - era percentile
# - event-type context
#
# without rebuilding the database.
# ============================================================

gate_database[
    "gate_commercial_signal"
] = (
    gate_database[
        "gate_log_nominal"
    ]
)


# ============================================================
# PART 13 — KNOWN GATE TABLE
# ============================================================

known_gates = (
    gate_database[
        gate_database[
            "gate_available"
        ]
    ]
    [
        [
            "event_date",
            "event_name",
            "gate_event_type",
            "gate_era",
            "gate_usd",
            "gate_log_nominal",
            "gate_log_percentile_observed",
            "gate_era_percentile"
        ]
    ]
    .sort_values(
        "gate_usd",
        ascending=False
    )
)


# ============================================================
# PART 14 — VALIDATE KNOWN ANCHORS
# ============================================================

expected_known = {

    (
        pd.Timestamp("2016-11-12"),
        "UFC 205: Alvarez vs McGregor"
    ):
        17740543,

    (
        pd.Timestamp("2018-10-06"),
        "UFC 229: Khabib vs. McGregor"
    ):
        17188894,

    (
        pd.Timestamp("2023-11-11"),
        "UFC 295: Prochazka vs. Pereira"
    ):
        12432563,

    (
        pd.Timestamp("2024-04-13"),
        "UFC 300: Pereira vs. Hill"
    ):
        16500000,

    (
        pd.Timestamp("2026-07-11"),
        "UFC 329: McGregor vs. Holloway 2"
    ):
        26430566,

    (
        pd.Timestamp("2026-08-15"),
        "UFC 330: Makhachev vs. Machado Garry"
    ):
        7000000
}


anchor_results = []


for (
    event_date,
    expected_name
), expected_gate in expected_known.items():

    match = gate_database[
        (
            gate_database[
                "event_date"
            ] == event_date
        )
        &
        (
            gate_database[
                "event_name"
            ] == expected_name
        )
    ]


    if len(match) == 1:

        actual_gate = (
            match.iloc[0][
                "gate_usd"
            ]
        )

        gate_match = (
            pd.notna(actual_gate)
            and
            np.isclose(
                actual_gate,
                expected_gate
            )
        )

    else:

        actual_gate = np.nan
        gate_match = False


    anchor_results.append({

        "event_date":
            event_date,

        "event_name":
            expected_name,

        "expected_gate":
            expected_gate,

        "actual_gate":
            actual_gate,

        "match":
            gate_match
    })


anchor_audit = pd.DataFrame(
    anchor_results
)


# ============================================================
# PART 15 — NEW 11 EVENT COVERAGE
# ============================================================

new_gate_audit = (
    gate_database[
        gate_database[
            "event_date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "gate_event_type",
            "gate_usd",
            "gate_available"
        ]
    ]
)


# ============================================================
# PART 16 — GLOBAL VALIDATION
# ============================================================

total_events = len(
    gate_database
)


unique_events = (
    gate_database[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


known_count = int(
    gate_database[
        "gate_available"
    ].sum()
)


zero_count = int(
    gate_database[
        "gate_usd"
    ].eq(0)
    .sum()
)


missing_count = int(
    gate_database[
        "gate_missing"
    ].sum()
)


anchor_pass = bool(
    anchor_audit[
        "match"
    ].all()
)


# ============================================================
# PART 17 — OUTPUT
# ============================================================

print()
print("=" * 110)
print("GATE DATABASE VALIDATION")
print("=" * 110)

print(
    "Total events:",
    total_events,
    "| expected 788"
)

print(
    "Unique events:",
    unique_events,
    "| expected 788"
)

print(
    "Known verified gates:",
    known_count
)

print(
    "Missing gates:",
    missing_count
)

print(
    "Gate values incorrectly stored as $0:",
    zero_count
)


print()
print("=" * 110)
print("KNOWN VERIFIED GATES")
print("=" * 110)

if len(known_gates):

    display_known = (
        known_gates.copy()
    )

    display_known[
        "gate_usd"
    ] = (
        display_known[
            "gate_usd"
        ]
        .map(
            lambda x:
            f"${x:,.0f}"
        )
    )

    display_known[
        "gate_log_nominal"
    ] = (
        display_known[
            "gate_log_nominal"
        ]
        .map(
            lambda x:
            f"{x:.4f}"
        )
    )

    display_known[
        "gate_log_percentile_observed"
    ] = (
        display_known[
            "gate_log_percentile_observed"
        ]
        .map(
            lambda x:
            f"{x:.2f}"
        )
    )

    print(
        display_known.to_string(
            index=False
        )
    )


print()
print("=" * 110)
print("GATE COVERAGE BY ERA")
print("=" * 110)

print(
    gate_era_coverage.to_string(
        index=False,
        formatters={
            "coverage_pct":
                lambda x:
                f"{x:.2f}%"
        }
    )
)


print()
print("=" * 110)
print("ANCHOR AUDIT")
print("=" * 110)

print(
    anchor_audit.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("NEW 11 EVENTS — GATE COVERAGE")
print("=" * 110)

print(
    new_gate_audit.to_string(
        index=False
    )
)


# ============================================================
# PART 18 — PASS / FAIL
# ============================================================

step87_pass = (

    total_events == 788

    and

    unique_events == 788

    and

    known_count == 6

    and

    missing_count == 782

    and

    zero_count == 0

    and

    anchor_pass

    and

    len(
        new_gate_audit
    ) == 11
)


print()
print("=" * 110)

if step87_pass:

    print(
        "STEP 87 GATE BACKBONE: PASS ✅🔥"
    )

else:

    print(
        "STEP 87 GATE BACKBONE: CHECK REQUIRED ⚠️"
    )

print("=" * 110)
```

```python
# ============================================================
# STEP 87B — CORRECTED PRODUCTION GATE BACKBONE
# ============================================================

import pandas as pd
import numpy as np
import re


print("=" * 110)
print("STEP 87B — CORRECTED GATE / COMMERCIAL DEMAND BACKBONE 🔥")
print("=" * 110)


# ============================================================
# PART 1 — RESTORE CLEAN PRE-STEP-87 COPY
# ============================================================

gate_database = (
    gate_database_before_step87.copy()
)


# Standardize date column name.
if (
    "date" in gate_database.columns
    and
    "event_date" not in gate_database.columns
):

    gate_database = gate_database.rename(
        columns={
            "date": "event_date"
        }
    )


gate_database["event_date"] = pd.to_datetime(
    gate_database["event_date"]
)


# ============================================================
# PART 2 — STANDARDIZE UFC 331 NAME
# ============================================================

gate_database.loc[
    (
        gate_database["event_date"]
        == pd.Timestamp("2026-09-19")
    )
    &
    (
        gate_database["event_name"]
        .astype(str)
        .str.contains(
            "UFC 331",
            na=False
        )
    ),
    "event_name"
] = "UFC 331: Van vs. Pantoja 2"


# ============================================================
# PART 3 — ALIGN TO AUTHORITATIVE 788-EVENT MASTER
#
# Preserve gate/provenance information from the original
# gate database while using event_master_updated as the
# authoritative event list.
# ============================================================

gate_payload_columns = [

    "event_date",
    "event_name",

    "fanfare_era",

    "gate_usd_nominal",

    "gate_source",
    "gate_source_quality",
    "gate_notes"
]


gate_payload = (
    gate_database[
        gate_payload_columns
    ]
    .copy()
)


gate_duplicates = int(
    gate_payload
    .duplicated(
        subset=[
            "event_date",
            "event_name"
        ]
    )
    .sum()
)


if gate_duplicates != 0:

    raise ValueError(
        f"Found {gate_duplicates} duplicate "
        "events in gate source."
    )


gate_database = (
    event_master_updated[
        [
            "event_date",
            "event_name"
        ]
    ]
    .merge(
        gate_payload,
        on=[
            "event_date",
            "event_name"
        ],
        how="left",
        validate="one_to_one"
    )
)


# ============================================================
# PART 4 — CLEAN NOMINAL GATE
#
# Missing gate stays NaN.
# Never turn missing into $0.
# ============================================================

gate_database[
    "gate_usd_nominal"
] = pd.to_numeric(
    gate_database[
        "gate_usd_nominal"
    ],
    errors="coerce"
)


gate_database.loc[
    gate_database[
        "gate_usd_nominal"
    ] <= 0,
    "gate_usd_nominal"
] = np.nan


gate_database[
    "gate_available"
] = (
    gate_database[
        "gate_usd_nominal"
    ].notna()
)


gate_database[
    "gate_missing"
] = (
    ~gate_database[
        "gate_available"
    ]
)


# ============================================================
# PART 5 — EVENT ID
# ============================================================

def simple_slug(text):

    text = str(text).lower()

    text = re.sub(
        r"[^a-z0-9]+",
        "_",
        text
    )

    return text.strip("_")


gate_database[
    "gate_event_id"
] = (
    gate_database[
        "event_date"
    ]
    .dt.strftime("%Y-%m-%d")
    +
    "__"
    +
    gate_database[
        "event_name"
    ].apply(simple_slug)
)


# ============================================================
# PART 6 — EVENT TYPE
# ============================================================

def classify_gate_event(name):

    name = str(name)


    if name.startswith(
        "UFC Fight Night"
    ):
        return "Fight Night"


    if name.startswith(
        "Noche UFC"
    ):
        return "Noche UFC"


    if name.startswith(
        "UFC on FOX"
    ):
        return "UFC on FOX"


    if name.startswith(
        "UFC on FX"
    ):
        return "UFC on FX"


    if (
        name.startswith(
            "UFC on Fuel"
        )
        or
        name.startswith(
            "UFC on FUEL"
        )
    ):
        return "UFC on Fuel"


    if (
        "Ultimate Fighter" in name
        or
        "TUF" in name
    ):
        return "TUF Finale"


    if re.match(
        r"^UFC\s+\d+",
        name
    ):
        return "Numbered UFC"


    return "Other"


gate_database[
    "gate_event_type"
] = (
    gate_database[
        "event_name"
    ].apply(
        classify_gate_event
    )
)


# ============================================================
# PART 7 — FANFARE ERA
#
# Preserve existing fanfare_era when available.
# Fill only if missing.
# ============================================================

def assign_fanfare_era(date):

    year = pd.Timestamp(
        date
    ).year

    if year <= 2000:
        return "1994-2000"

    if year <= 2010:
        return "2001-2010"

    if year <= 2013:
        return "2011-2013"

    if year <= 2018:
        return "2014-2018"

    return "2019-present"


missing_era = (
    gate_database[
        "fanfare_era"
    ].isna()
)


gate_database.loc[
    missing_era,
    "fanfare_era"
] = (
    gate_database.loc[
        missing_era,
        "event_date"
    ].apply(
        assign_fanfare_era
    )
)


# ============================================================
# PART 8 — LOG GATE
#
# Compresses the extreme dollar scale while retaining
# ordering and meaningful differences.
# ============================================================

gate_database[
    "gate_log_nominal"
] = np.where(

    gate_database[
        "gate_available"
    ],

    np.log1p(
        gate_database[
            "gate_usd_nominal"
        ]
    ),

    np.nan
)


# ============================================================
# PART 9 — OBSERVED-GATE PERCENTILE
#
# DIAGNOSTIC ONLY.
#
# With only six known gates, this is NOT our final historical
# gate score.
# ============================================================

gate_database[
    "gate_observed_percentile"
] = np.nan


known_mask = (
    gate_database[
        "gate_available"
    ]
)


gate_database.loc[
    known_mask,
    "gate_observed_percentile"
] = (

    gate_database.loc[
        known_mask,
        "gate_log_nominal"
    ]
    .rank(
        pct=True,
        method="average"
    )
    * 100
)


# ============================================================
# PART 10 — ERA COVERAGE
# ============================================================

gate_era_coverage = (
    gate_database
    .groupby(
        "fanfare_era",
        as_index=False
    )
    .agg(

        total_events=(
            "event_name",
            "size"
        ),

        known_gates=(
            "gate_available",
            "sum"
        )
    )
)


gate_era_coverage[
    "coverage_pct"
] = (
    gate_era_coverage[
        "known_gates"
    ]
    /
    gate_era_coverage[
        "total_events"
    ]
    * 100
)


# ============================================================
# PART 11 — ERA PERCENTILE
#
# Do NOT estimate this until an era has enough observations.
# ============================================================

MIN_GATES_PER_ERA = 10


gate_database[
    "gate_era_percentile"
] = np.nan


for era, group in gate_database.groupby(
    "fanfare_era"
):

    valid = (
        group[
            "gate_available"
        ]
    )

    n = int(
        valid.sum()
    )


    if n >= MIN_GATES_PER_ERA:

        idx = (
            group.loc[
                valid
            ].index
        )


        gate_database.loc[
            idx,
            "gate_era_percentile"
        ] = (

            group.loc[
                valid,
                "gate_log_nominal"
            ]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )


# ============================================================
# PART 12 — PRODUCTION COMMERCIAL SIGNAL
#
# At current coverage, the defensible gate signal is simply
# the transformed verified gate itself.
#
# Missing remains missing.
# ============================================================

gate_database[
    "gate_commercial_signal"
] = (
    gate_database[
        "gate_log_nominal"
    ]
)


# ============================================================
# PART 13 — KNOWN GATES
# ============================================================

known_gates = (
    gate_database[
        gate_database[
            "gate_available"
        ]
    ]
    [
        [
            "event_date",
            "event_name",
            "gate_event_type",
            "fanfare_era",
            "gate_usd_nominal",
            "gate_log_nominal",
            "gate_observed_percentile",
            "gate_source",
            "gate_source_quality"
        ]
    ]
    .sort_values(
        "gate_usd_nominal",
        ascending=False
    )
)


# ============================================================
# PART 14 — VALIDATE THE SIX VALUES
#
# Match by date + UFC number where possible rather than
# assuming every historical event-name punctuation convention.
# ============================================================

expected_gates = {

    pd.Timestamp(
        "2016-11-12"
    ): 17740543,

    pd.Timestamp(
        "2018-10-06"
    ): 17188894,

    pd.Timestamp(
        "2023-11-11"
    ): 12432563,

    pd.Timestamp(
        "2024-04-13"
    ): 16500000,

    pd.Timestamp(
        "2026-07-11"
    ): 26430566,

    pd.Timestamp(
        "2026-08-15"
    ): 7000000
}


anchor_rows = []


for event_date, expected_gate in (
    expected_gates.items()
):

    matches = (
        gate_database[
            (
                gate_database[
                    "event_date"
                ] == event_date
            )
            &
            (
                gate_database[
                    "gate_available"
                ]
            )
        ]
    )


    if len(matches) == 1:

        row = matches.iloc[0]

        actual_gate = float(
            row[
                "gate_usd_nominal"
            ]
        )

        event_name = (
            row[
                "event_name"
            ]
        )

        matches_expected = bool(
            np.isclose(
                actual_gate,
                expected_gate
            )
        )

    else:

        actual_gate = np.nan
        event_name = None
        matches_expected = False


    anchor_rows.append({

        "event_date":
            event_date,

        "event_name":
            event_name,

        "expected_gate":
            expected_gate,

        "actual_gate":
            actual_gate,

        "match":
            matches_expected
    })


gate_anchor_audit = pd.DataFrame(
    anchor_rows
)


# ============================================================
# PART 15 — NEW 11 EVENTS
# ============================================================

new_gate_audit = (
    gate_database[
        gate_database[
            "event_date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "gate_event_type",
            "gate_usd_nominal",
            "gate_available"
        ]
    ]
)


# ============================================================
# PART 16 — VALIDATION NUMBERS
# ============================================================

total_events = len(
    gate_database
)


unique_events = (
    gate_database[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


known_count = int(
    gate_database[
        "gate_available"
    ].sum()
)


missing_count = int(
    gate_database[
        "gate_missing"
    ].sum()
)


zero_count = int(
    gate_database[
        "gate_usd_nominal"
    ]
    .eq(0)
    .sum()
)


anchor_pass = bool(
    gate_anchor_audit[
        "match"
    ].all()
)


# ============================================================
# PART 17 — OUTPUT
# ============================================================

print()
print("=" * 110)
print("GATE DATABASE VALIDATION")
print("=" * 110)

print(
    "Total events:",
    total_events,
    "| expected 788"
)

print(
    "Unique events:",
    unique_events,
    "| expected 788"
)

print(
    "Known verified gates:",
    known_count,
    "| expected 6"
)

print(
    "Missing gates:",
    missing_count,
    "| expected 782"
)

print(
    "Fake $0 gates:",
    zero_count,
    "| expected 0"
)


print()
print("=" * 110)
print("KNOWN VERIFIED GATES")
print("=" * 110)


known_display = (
    known_gates.copy()
)


known_display[
    "gate_usd_nominal"
] = (
    known_display[
        "gate_usd_nominal"
    ].map(
        lambda x:
        f"${x:,.0f}"
    )
)


known_display[
    "gate_log_nominal"
] = (
    known_display[
        "gate_log_nominal"
    ].map(
        lambda x:
        f"{x:.4f}"
    )
)


known_display[
    "gate_observed_percentile"
] = (
    known_display[
        "gate_observed_percentile"
    ].map(
        lambda x:
        f"{x:.2f}"
    )
)


print(
    known_display.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("GATE COVERAGE BY ERA")
print("=" * 110)

print(
    gate_era_coverage.to_string(
        index=False,
        formatters={
            "coverage_pct":
                lambda x:
                f"{x:.2f}%"
        }
    )
)


print()
print("=" * 110)
print("ANCHOR AUDIT")
print("=" * 110)

print(
    gate_anchor_audit.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("NEW 11 EVENTS")
print("=" * 110)

print(
    new_gate_audit.to_string(
        index=False
    )
)


# ============================================================
# PART 18 — FINAL PASS / FAIL
# ============================================================

step87b_pass = (

    total_events == 788

    and

    unique_events == 788

    and

    known_count == 6

    and

    missing_count == 782

    and

    zero_count == 0

    and

    anchor_pass

    and

    len(
        new_gate_audit
    ) == 11
)


print()
print("=" * 110)

if step87b_pass:

    print(
        "STEP 87B GATE BACKBONE: PASS ✅🔥"
    )

else:

    print(
        "STEP 87B GATE BACKBONE: CHECK REQUIRED ⚠️"
    )

print("=" * 110)
```

```python
# ============================================================
# STEP 88B — APPEND NEW SEARCH-HYPE ACCELERATION RESULTS
# ============================================================

import pandas as pd
import numpy as np


print("=" * 110)
print("STEP 88B — UPDATE SEARCH HYPE ACCELERATION 🔥")
print("=" * 110)


# ============================================================
# PART 1 — SAFETY COPIES
# ============================================================

event_trends_results_before_step88b = (
    event_trends_results.copy()
)

new_event_trends_before_step88b = (
    new_event_trends.copy()
)


# ============================================================
# PART 2 — STANDARDIZE DATES
# ============================================================

event_trends_results[
    "event_date"
] = pd.to_datetime(
    event_trends_results[
        "event_date"
    ]
)


new_event_trends[
    "event_date"
] = pd.to_datetime(
    new_event_trends[
        "event_date"
    ]
)


# ============================================================
# PART 3 — STANDARDIZE UFC 331 NAME
# ============================================================

new_event_trends.loc[
    (
        new_event_trends[
            "event_date"
        ]
        == pd.Timestamp("2026-09-19")
    )
    &
    (
        new_event_trends[
            "event_name"
        ]
        .astype(str)
        .str.contains(
            "UFC 331",
            na=False
        )
    ),
    "event_name"
] = "UFC 331: Van vs. Pantoja 2"


# ============================================================
# PART 4 — PRE-APPEND VALIDATION
# ============================================================

print()
print("Historical Trends rows:", len(event_trends_results))
print("Expected historical rows: 505")

print(
    "New Trends rows:",
    len(new_event_trends)
)
print("Expected new rows: 11")


old_duplicates = int(
    event_trends_results
    .duplicated(
        subset=[
            "event_date",
            "event_name"
        ]
    )
    .sum()
)


new_duplicates = int(
    new_event_trends
    .duplicated(
        subset=[
            "event_date",
            "event_name"
        ]
    )
    .sum()
)


overlap = (
    event_trends_results[
        [
            "event_date",
            "event_name"
        ]
    ]
    .merge(
        new_event_trends[
            [
                "event_date",
                "event_name"
            ]
        ],
        on=[
            "event_date",
            "event_name"
        ],
        how="inner"
    )
)


print(
    "Historical duplicate events:",
    old_duplicates
)

print(
    "New duplicate events:",
    new_duplicates
)

print(
    "Old/new event overlap:",
    len(overlap)
)


# ============================================================
# PART 5 — REQUIRE IDENTICAL COLUMN STRUCTURE
# ============================================================

old_columns = list(
    event_trends_results.columns
)

new_columns = list(
    new_event_trends.columns
)


missing_from_new = [
    col
    for col in old_columns
    if col not in new_columns
]


extra_in_new = [
    col
    for col in new_columns
    if col not in old_columns
]


print()
print(
    "Columns missing from new Trends:",
    missing_from_new
)

print(
    "Extra columns in new Trends:",
    extra_in_new
)


if missing_from_new or extra_in_new:

    raise ValueError(
        "Old and new Trends schemas do not match. "
        "Stop before appending."
    )


# Put new data in identical column order.
new_event_trends = (
    new_event_trends[
        old_columns
    ]
    .copy()
)


# ============================================================
# PART 6 — APPEND
# ============================================================

event_trends_results = (
    pd.concat(
        [
            event_trends_results,
            new_event_trends
        ],
        ignore_index=True
    )
    .sort_values(
        [
            "event_date",
            "event_name"
        ]
    )
    .reset_index(drop=True)
)


# ============================================================
# PART 7 — VALIDATE FINAL DATASET
# ============================================================

total_rows = len(
    event_trends_results
)


unique_events = (
    event_trends_results[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


final_duplicates = int(
    event_trends_results
    .duplicated(
        subset=[
            "event_date",
            "event_name"
        ]
    )
    .sum()
)


missing_acceleration = int(
    event_trends_results[
        "event_avg_log_hype"
    ]
    .isna()
    .sum()
)


# ============================================================
# PART 8 — VALIDATE THE 11 NEW VALUES
# ============================================================

expected_new_acceleration = {

    pd.Timestamp("2026-07-11"):
        1.993343,

    pd.Timestamp("2026-07-18"):
        1.870728,

    pd.Timestamp("2026-07-25"):
        2.830986,

    pd.Timestamp("2026-08-01"):
        3.041498,

    pd.Timestamp("2026-08-08"):
        2.681017,

    pd.Timestamp("2026-08-15"):
        1.731465,

    pd.Timestamp("2026-08-22"):
        3.072657,

    pd.Timestamp("2026-08-29"):
        1.650320,

    pd.Timestamp("2026-09-05"):
        2.379491,

    pd.Timestamp("2026-09-12"):
        1.705329,

    pd.Timestamp("2026-09-19"):
        1.749946
}


new_validation_rows = []


for event_date, expected in (
    expected_new_acceleration.items()
):

    match = (
        event_trends_results[
            event_trends_results[
                "event_date"
            ] == event_date
        ]
    )


    # All these dates contain one of our update events.
    # Match only rows after old cutoff.
    match = match[
        match[
            "event_date"
        ] > pd.Timestamp(
            "2026-06-27"
        )
    ]


    if len(match) == 1:

        row = match.iloc[0]

        actual = float(
            row[
                "event_avg_log_hype"
            ]
        )

        event_name = (
            row[
                "event_name"
            ]
        )

        value_match = bool(
            np.isclose(
                actual,
                expected,
                atol=1e-5
            )
        )

    else:

        event_name = None
        actual = np.nan
        value_match = False


    new_validation_rows.append({

        "event_date":
            event_date,

        "event_name":
            event_name,

        "expected_acceleration":
            expected,

        "actual_acceleration":
            actual,

        "match":
            value_match
    })


acceleration_audit = pd.DataFrame(
    new_validation_rows
)


# ============================================================
# PART 9 — ATTACH ACCELERATION TO THE 788-EVENT MASTER
# ============================================================

acceleration_for_merge = (
    event_trends_results[
        [
            "event_date",
            "event_name",
            "event_avg_baseline",
            "event_avg_fight_week",
            "event_avg_log_hype"
        ]
    ]
    .copy()
)


event_master_updated = (
    event_master_updated
    .drop(
        columns=[
            "event_avg_baseline",
            "event_avg_fight_week",
            "event_avg_log_hype"
        ],
        errors="ignore"
    )
    .merge(
        acceleration_for_merge,
        on=[
            "event_date",
            "event_name"
        ],
        how="left",
        validate="one_to_one"
    )
)


event_master_updated[
    "trends_available"
] = (
    event_master_updated[
        "event_avg_log_hype"
    ].notna()
)


# ============================================================
# PART 10 — EVENT MASTER COVERAGE
# ============================================================

events_with_trends = int(
    event_master_updated[
        "trends_available"
    ].sum()
)


events_without_trends = int(
    (
        ~event_master_updated[
            "trends_available"
        ]
    ).sum()
)


new_master_trends = (
    event_master_updated[
        event_master_updated[
            "event_date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "event_avg_log_hype",
            "trends_available"
        ]
    ]
)


# ============================================================
# PART 11 — OUTPUT
# ============================================================

print()
print("=" * 110)
print("UPDATED TRENDS DATABASE")
print("=" * 110)

print(
    "Total Trends events:",
    total_rows,
    "| expected 516"
)

print(
    "Unique Trends events:",
    unique_events,
    "| expected 516"
)

print(
    "Duplicate Trends events:",
    final_duplicates,
    "| expected 0"
)

print(
    "Missing acceleration values:",
    missing_acceleration
)


print()
print("=" * 110)
print("NEW 11 ACCELERATION AUDIT")
print("=" * 110)

print(
    acceleration_audit.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("788-EVENT MASTER — TRENDS COVERAGE")
print("=" * 110)

print(
    "Events with Search Hype Acceleration:",
    events_with_trends,
    "| expected 516"
)

print(
    "Events without Search Hype Acceleration:",
    events_without_trends,
    "| expected 272"
)


print()
print("NEW 11 EVENTS:")

print(
    new_master_trends.to_string(
        index=False
    )
)


# ============================================================
# PART 12 — FINAL PASS / FAIL
# ============================================================

step88b_pass = (

    total_rows == 516

    and

    unique_events == 516

    and

    final_duplicates == 0

    and

    acceleration_audit[
        "match"
    ].all()

    and

    events_with_trends == 516

    and

    events_without_trends == 272

    and

    len(
        new_master_trends
    ) == 11

    and

    new_master_trends[
        "trends_available"
    ].all()
)


print()
print("=" * 110)

if step88b_pass:

    print(
        "STEP 88B SEARCH HYPE UPDATE: PASS ✅🔥"
    )

else:

    print(
        "STEP 88B SEARCH HYPE UPDATE: CHECK REQUIRED ⚠️"
    )

print("=" * 110)
```

```python
# ============================================================
# STEP 89 — FANFARE SIGNAL MASTER
# ============================================================

import pandas as pd
import numpy as np


print("=" * 110)
print("STEP 89 — FANFARE SIGNAL MASTER 🔥")
print("=" * 110)


# ============================================================
# PART 1 — START FROM AUTHORITATIVE 788 EVENTS
# ============================================================

fanfare_master = (
    event_master_updated[
        [
            "event_date",
            "event_name"
        ]
    ]
    .copy()
)


# ============================================================
# PART 2 — FANFARE ERA
# ============================================================

def assign_fanfare_era(date):

    year = pd.Timestamp(date).year

    if year <= 2000:
        return "1994-2000"

    if year <= 2010:
        return "2001-2010"

    if year <= 2013:
        return "2011-2013"

    if year <= 2018:
        return "2014-2018"

    return "2019-present"


fanfare_master[
    "fanfare_era"
] = (
    fanfare_master[
        "event_date"
    ].apply(
        assign_fanfare_era
    )
)


# ============================================================
# PART 3 — SEARCH HYPE ACCELERATION
#
# This is our existing event-time Google Trends metric.
#
# IMPORTANT:
# It measures RELATIVE acceleration from baseline to fight
# week. It is NOT absolute popularity.
# ============================================================

search_acceleration = (
    event_trends_results[
        [
            "event_date",
            "event_name",
            "event_avg_log_hype"
        ]
    ]
    .rename(
        columns={
            "event_avg_log_hype":
                "search_hype_acceleration"
        }
    )
    .copy()
)


fanfare_master = (
    fanfare_master
    .merge(
        search_acceleration,
        on=[
            "event_date",
            "event_name"
        ],
        how="left",
        validate="one_to_one"
    )
)


fanfare_master[
    "search_acceleration_available"
] = (
    fanfare_master[
        "search_hype_acceleration"
    ].notna()
)


# ============================================================
# PART 4 — VERIFIED GATE / COMMERCIAL DEMAND
# ============================================================

gate_merge = (
    gate_database[
        [
            "event_date",
            "event_name",
            "gate_usd_nominal",
            "gate_log_nominal",
            "gate_source",
            "gate_source_quality"
        ]
    ]
    .copy()
)


fanfare_master = (
    fanfare_master
    .merge(
        gate_merge,
        on=[
            "event_date",
            "event_name"
        ],
        how="left",
        validate="one_to_one"
    )
)


fanfare_master[
    "gate_available"
] = (
    fanfare_master[
        "gate_usd_nominal"
    ].notna()
)


# ============================================================
# PART 5 — ABSOLUTE SEARCH DEMAND PLACEHOLDER
#
# DO NOT populate this with the fixed 2025-26 fighter
# calibration.
#
# We need event-time/common-anchor calibration first.
# ============================================================

fanfare_master[
    "absolute_search_demand"
] = np.nan


fanfare_master[
    "absolute_search_available"
] = False


# ============================================================
# PART 6 — HISTORICAL DRAWING POWER PLACEHOLDER
#
# Future idea:
#
# Entering each event, measure what commercial/search drawing
# evidence the two headliners had already demonstrated.
#
# Must use ONLY information available BEFORE that event.
# ============================================================

fanfare_master[
    "historical_drawing_power"
] = np.nan


fanfare_master[
    "historical_drawing_power_available"
] = False


# ============================================================
# PART 7 — OPTIONAL PPV / VIEWERSHIP PLACEHOLDERS
#
# These are intentionally separate because PPV and television
# viewership are not directly interchangeable.
# ============================================================

fanfare_master[
    "ppv_buys"
] = np.nan


fanfare_master[
    "ppv_available"
] = False


fanfare_master[
    "viewership"
] = np.nan


fanfare_master[
    "viewership_available"
] = False


# ============================================================
# PART 8 — COUNT CURRENT SIGNAL AVAILABILITY
# ============================================================

availability_columns = [

    "search_acceleration_available",
    "gate_available",
    "absolute_search_available",
    "historical_drawing_power_available",
    "ppv_available",
    "viewership_available"
]


fanfare_master[
    "fanfare_signals_available"
] = (
    fanfare_master[
        availability_columns
    ]
    .sum(
        axis=1
    )
)


# ============================================================
# PART 9 — CURRENT EVIDENCE CLASSIFICATION
#
# This is NOT a score.
# It simply tells us how much fanfare evidence we currently
# possess for each event.
# ============================================================

def evidence_status(row):

    search = bool(
        row[
            "search_acceleration_available"
        ]
    )

    gate = bool(
        row[
            "gate_available"
        ]
    )


    if search and gate:
        return "Search + Gate"

    if search:
        return "Search Only"

    if gate:
        return "Gate Only"

    return "No Current Signal"


fanfare_master[
    "current_fanfare_evidence"
] = (
    fanfare_master.apply(
        evidence_status,
        axis=1
    )
)


# ============================================================
# PART 10 — COVERAGE SUMMARY
# ============================================================

coverage_summary = pd.DataFrame({

    "signal": [
        "Search Hype Acceleration",
        "Verified Gate",
        "Absolute Search Demand",
        "Historical Drawing Power",
        "PPV Buys",
        "Viewership"
    ],

    "events_available": [

        int(
            fanfare_master[
                "search_acceleration_available"
            ].sum()
        ),

        int(
            fanfare_master[
                "gate_available"
            ].sum()
        ),

        int(
            fanfare_master[
                "absolute_search_available"
            ].sum()
        ),

        int(
            fanfare_master[
                "historical_drawing_power_available"
            ].sum()
        ),

        int(
            fanfare_master[
                "ppv_available"
            ].sum()
        ),

        int(
            fanfare_master[
                "viewership_available"
            ].sum()
        )
    ]
})


coverage_summary[
    "coverage_pct"
] = (
    coverage_summary[
        "events_available"
    ]
    / len(fanfare_master)
    * 100
)


# ============================================================
# PART 11 — COVERAGE BY ERA
# ============================================================

era_coverage = (
    fanfare_master
    .groupby(
        "fanfare_era",
        as_index=False
    )
    .agg(

        events=(
            "event_name",
            "size"
        ),

        search_events=(
            "search_acceleration_available",
            "sum"
        ),

        gate_events=(
            "gate_available",
            "sum"
        )
    )
)


era_coverage[
    "search_pct"
] = (
    era_coverage[
        "search_events"
    ]
    /
    era_coverage[
        "events"
    ]
    * 100
)


era_coverage[
    "gate_pct"
] = (
    era_coverage[
        "gate_events"
    ]
    /
    era_coverage[
        "events"
    ]
    * 100
)


# ============================================================
# PART 12 — EVIDENCE DISTRIBUTION
# ============================================================

evidence_distribution = (
    fanfare_master[
        "current_fanfare_evidence"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "evidence_type"
    )
    .reset_index(
        name="events"
    )
)


evidence_distribution[
    "pct"
] = (
    evidence_distribution[
        "events"
    ]
    / len(fanfare_master)
    * 100
)


# ============================================================
# PART 13 — NEW 11 EVENTS
# ============================================================

new_fanfare_check = (
    fanfare_master[
        fanfare_master[
            "event_date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "search_hype_acceleration",
            "search_acceleration_available",
            "gate_usd_nominal",
            "gate_available",
            "fanfare_signals_available",
            "current_fanfare_evidence"
        ]
    ]
)


# ============================================================
# PART 14 — IMPORTANT CARD CHECKS
# ============================================================

important_cards = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        .str.contains(
            (
                "UFC 205|"
                "UFC 229|"
                "UFC 300|"
                "UFC 329|"
                "UFC 330"
            ),
            regex=True,
            na=False
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "search_hype_acceleration",
            "gate_usd_nominal",
            "current_fanfare_evidence"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 15 — GLOBAL VALIDATION
# ============================================================

total_events = len(
    fanfare_master
)


unique_events = (
    fanfare_master[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


search_count = int(
    fanfare_master[
        "search_acceleration_available"
    ].sum()
)


gate_count = int(
    fanfare_master[
        "gate_available"
    ].sum()
)


fake_zero_gates = int(
    fanfare_master[
        "gate_usd_nominal"
    ]
    .eq(0)
    .sum()
)


new_count = len(
    new_fanfare_check
)


new_search_count = int(
    new_fanfare_check[
        "search_acceleration_available"
    ].sum()
)


new_gate_count = int(
    new_fanfare_check[
        "gate_available"
    ].sum()
)


# ============================================================
# PART 16 — OUTPUT
# ============================================================

print()
print("=" * 110)
print("GLOBAL VALIDATION")
print("=" * 110)

print(
    "Total events:",
    total_events,
    "| expected 788"
)

print(
    "Unique events:",
    unique_events,
    "| expected 788"
)

print(
    "Search acceleration available:",
    search_count,
    "| expected 516"
)

print(
    "Verified gates available:",
    gate_count,
    "| expected 6"
)

print(
    "Fake $0 gates:",
    fake_zero_gates,
    "| expected 0"
)


print()
print("=" * 110)
print("SIGNAL COVERAGE")
print("=" * 110)

print(
    coverage_summary.to_string(
        index=False,
        formatters={
            "coverage_pct":
                lambda x:
                f"{x:.2f}%"
        }
    )
)


print()
print("=" * 110)
print("COVERAGE BY ERA")
print("=" * 110)

print(
    era_coverage.to_string(
        index=False,
        formatters={
            "search_pct":
                lambda x:
                f"{x:.2f}%",

            "gate_pct":
                lambda x:
                f"{x:.2f}%"
        }
    )
)


print()
print("=" * 110)
print("CURRENT EVIDENCE DISTRIBUTION")
print("=" * 110)

print(
    evidence_distribution.to_string(
        index=False,
        formatters={
            "pct":
                lambda x:
                f"{x:.2f}%"
        }
    )
)


print()
print("=" * 110)
print("NEW 11 EVENTS")
print("=" * 110)

print(
    new_fanfare_check.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("IMPORTANT CARD CHECK")
print("=" * 110)

print(
    important_cards.to_string(
        index=False
    )
)


# ============================================================
# PART 17 — PASS / FAIL
# ============================================================

step89_pass = (

    total_events == 788

    and

    unique_events == 788

    and

    search_count == 516

    and

    gate_count == 6

    and

    fake_zero_gates == 0

    and

    new_count == 11

    and

    new_search_count == 11

    and

    new_gate_count == 2
)


print()
print("=" * 110)

if step89_pass:

    print(
        "STEP 89 FANFARE SIGNAL MASTER: PASS ✅🔥"
    )

else:

    print(
        "STEP 89 FANFARE SIGNAL MASTER: CHECK REQUIRED ⚠️"
    )

print("=" * 110)
```

```python
# ============================================================
# STEP 90 — HISTORICAL DRAWING POWER BACKBONE
# ============================================================

import pandas as pd
import numpy as np


print("=" * 110)
print("STEP 90 — HISTORICAL DRAWING POWER BACKBONE 🔥")
print("=" * 110)


# ============================================================
# PART 1 — GET VERIFIED MAIN-EVENT HEADLINERS
# ============================================================

main_event_fights = (
    ufc_fights_updated[
        ufc_fights_updated[
            "is_main_event_verified"
        ] == True
    ]
    [
        [
            "date",
            "event_name",
            "f1_name",
            "f2_name"
        ]
    ]
    .copy()
)


main_event_fights[
    "date"
] = pd.to_datetime(
    main_event_fights[
        "date"
    ]
)


main_event_fights = (
    main_event_fights.rename(
        columns={
            "date":
                "event_date",

            "f1_name":
                "headliner_1",

            "f2_name":
                "headliner_2"
        }
    )
)


# ============================================================
# PART 2 — VALIDATE MAIN EVENTS
# ============================================================

main_event_duplicates = int(
    main_event_fights
    .duplicated(
        subset=[
            "event_date",
            "event_name"
        ]
    )
    .sum()
)


if main_event_duplicates:

    raise ValueError(
        f"Found {main_event_duplicates} duplicate "
        "verified main events."
    )


print(
    "Verified main events:",
    len(main_event_fights),
    "| expected 785"
)

print(
    "Duplicate verified main events:",
    main_event_duplicates,
    "| expected 0"
)


# ============================================================
# PART 3 — ATTACH CURRENT EVENT FANFARE SIGNALS
# ============================================================

history_source = (
    main_event_fights
    .merge(
        fanfare_master[
            [
                "event_date",
                "event_name",
                "search_hype_acceleration",
                "gate_usd_nominal"
            ]
        ],
        on=[
            "event_date",
            "event_name"
        ],
        how="left",
        validate="one_to_one"
    )
    .sort_values(
        [
            "event_date",
            "event_name"
        ]
    )
    .reset_index(
        drop=True
    )
)


# ============================================================
# PART 4 — LONG HEADLINER HISTORY TABLE
#
# Each main event becomes two fighter-event rows.
# ============================================================

h1 = (
    history_source[
        [
            "event_date",
            "event_name",
            "headliner_1",
            "search_hype_acceleration",
            "gate_usd_nominal"
        ]
    ]
    .rename(
        columns={
            "headliner_1":
                "fighter"
        }
    )
)


h1[
    "headliner_side"
] = "f1"


h2 = (
    history_source[
        [
            "event_date",
            "event_name",
            "headliner_2",
            "search_hype_acceleration",
            "gate_usd_nominal"
        ]
    ]
    .rename(
        columns={
            "headliner_2":
                "fighter"
        }
    )
)


h2[
    "headliner_side"
] = "f2"


headliner_history = (
    pd.concat(
        [
            h1,
            h2
        ],
        ignore_index=True
    )
    .sort_values(
        [
            "event_date",
            "event_name",
            "headliner_side"
        ],
        kind="stable"
    )
    .reset_index(
        drop=True
    )
)


# ============================================================
# PART 5 — PRIOR-EVENT HISTORY ENGINE
#
# IMPORTANT:
# Current event is added AFTER prior statistics are recorded.
#
# Therefore there is no future leakage and no current-event
# leakage.
# ============================================================

history_records = {}


output_rows = []


for row in (
    headliner_history
    .itertuples(
        index=False
    )
):

    fighter = row.fighter


    if fighter not in history_records:

        history_records[
            fighter
        ] = {

            "search_values": [],
            "gate_values": [],
            "evidence_events": 0
        }


    hist = history_records[
        fighter
    ]


    prior_search = (
        hist[
            "search_values"
        ]
    )


    prior_gates = (
        hist[
            "gate_values"
        ]
    )


    # ------------------------------------------
    # PRIOR SEARCH HISTORY
    # ------------------------------------------

    prior_search_events = len(
        prior_search
    )


    if prior_search_events:

        prior_avg_search = float(
            np.mean(
                prior_search
            )
        )

        prior_best_search = float(
            np.max(
                prior_search
            )
        )

    else:

        prior_avg_search = np.nan
        prior_best_search = np.nan


    # ------------------------------------------
    # PRIOR GATE HISTORY
    # ------------------------------------------

    prior_gate_events = len(
        prior_gates
    )


    if prior_gate_events:

        prior_avg_gate = float(
            np.mean(
                prior_gates
            )
        )

        prior_best_gate = float(
            np.max(
                prior_gates
            )
        )

        prior_total_gate = float(
            np.sum(
                prior_gates
            )
        )

    else:

        prior_avg_gate = np.nan
        prior_best_gate = np.nan
        prior_total_gate = np.nan


    # ------------------------------------------
    # SAVE PRIOR VALUES FOR THIS EVENT
    # ------------------------------------------

    output_rows.append({

        "event_date":
            row.event_date,

        "event_name":
            row.event_name,

        "fighter":
            fighter,

        "headliner_side":
            row.headliner_side,

        "prior_search_events":
            prior_search_events,

        "prior_avg_search_acceleration":
            prior_avg_search,

        "prior_best_search_acceleration":
            prior_best_search,

        "prior_gate_events":
            prior_gate_events,

        "prior_avg_gate_usd":
            prior_avg_gate,

        "prior_best_gate_usd":
            prior_best_gate,

        "prior_total_known_gate_usd":
            prior_total_gate,

        "prior_fanfare_evidence_events":
            hist[
                "evidence_events"
            ]
    })


    # ------------------------------------------
    # NOW ADD CURRENT EVENT TO FUTURE HISTORY
    # ------------------------------------------

    current_has_evidence = False


    if pd.notna(
        row.search_hype_acceleration
    ):

        hist[
            "search_values"
        ].append(
            float(
                row.search_hype_acceleration
            )
        )

        current_has_evidence = True


    if pd.notna(
        row.gate_usd_nominal
    ):

        hist[
            "gate_values"
        ].append(
            float(
                row.gate_usd_nominal
            )
        )

        current_has_evidence = True


    if current_has_evidence:

        hist[
            "evidence_events"
        ] += 1


historical_drawing_long = pd.DataFrame(
    output_rows
)


# ============================================================
# PART 6 — SPLIT BACK INTO HEADLINER 1 / HEADLINER 2
# ============================================================

drawing_f1 = (
    historical_drawing_long[
        historical_drawing_long[
            "headliner_side"
        ] == "f1"
    ]
    .drop(
        columns=[
            "headliner_side"
        ]
    )
    .rename(
        columns={

            "fighter":
                "headliner_1",

            "prior_search_events":
                "h1_prior_search_events",

            "prior_avg_search_acceleration":
                "h1_prior_avg_search_acceleration",

            "prior_best_search_acceleration":
                "h1_prior_best_search_acceleration",

            "prior_gate_events":
                "h1_prior_gate_events",

            "prior_avg_gate_usd":
                "h1_prior_avg_gate_usd",

            "prior_best_gate_usd":
                "h1_prior_best_gate_usd",

            "prior_total_known_gate_usd":
                "h1_prior_total_known_gate_usd",

            "prior_fanfare_evidence_events":
                "h1_prior_fanfare_evidence_events"
        }
    )
)


drawing_f2 = (
    historical_drawing_long[
        historical_drawing_long[
            "headliner_side"
        ] == "f2"
    ]
    .drop(
        columns=[
            "headliner_side"
        ]
    )
    .rename(
        columns={

            "fighter":
                "headliner_2",

            "prior_search_events":
                "h2_prior_search_events",

            "prior_avg_search_acceleration":
                "h2_prior_avg_search_acceleration",

            "prior_best_search_acceleration":
                "h2_prior_best_search_acceleration",

            "prior_gate_events":
                "h2_prior_gate_events",

            "prior_avg_gate_usd":
                "h2_prior_avg_gate_usd",

            "prior_best_gate_usd":
                "h2_prior_best_gate_usd",

            "prior_total_known_gate_usd":
                "h2_prior_total_known_gate_usd",

            "prior_fanfare_evidence_events":
                "h2_prior_fanfare_evidence_events"
        }
    )
)


# ============================================================
# PART 7 — EVENT-LEVEL HISTORICAL DRAWING TABLE
# ============================================================

historical_drawing_events = (
    main_event_fights
    .merge(
        drawing_f1,
        on=[
            "event_date",
            "event_name",
            "headliner_1"
        ],
        how="left",
        validate="one_to_one"
    )
    .merge(
        drawing_f2,
        on=[
            "event_date",
            "event_name",
            "headliner_2"
        ],
        how="left",
        validate="one_to_one"
    )
)


# ============================================================
# PART 8 — EVENT-LEVEL PRIOR SEARCH HISTORY
#
# Average both fighters where available.
# ============================================================

historical_drawing_events[
    "headliner_prior_avg_search"
] = (
    historical_drawing_events[
        [
            "h1_prior_avg_search_acceleration",
            "h2_prior_avg_search_acceleration"
        ]
    ]
    .mean(
        axis=1,
        skipna=True
    )
)


historical_drawing_events[
    "headliner_prior_best_search"
] = (
    historical_drawing_events[
        [
            "h1_prior_best_search_acceleration",
            "h2_prior_best_search_acceleration"
        ]
    ]
    .max(
        axis=1,
        skipna=True
    )
)


historical_drawing_events[
    "combined_prior_search_events"
] = (

    historical_drawing_events[
        "h1_prior_search_events"
    ].fillna(0)

    +

    historical_drawing_events[
        "h2_prior_search_events"
    ].fillna(0)
)


# ============================================================
# PART 9 — EVENT-LEVEL PRIOR COMMERCIAL HISTORY
# ============================================================

historical_drawing_events[
    "headliner_prior_best_gate"
] = (
    historical_drawing_events[
        [
            "h1_prior_best_gate_usd",
            "h2_prior_best_gate_usd"
        ]
    ]
    .max(
        axis=1,
        skipna=True
    )
)


historical_drawing_events[
    "combined_prior_gate_events"
] = (

    historical_drawing_events[
        "h1_prior_gate_events"
    ].fillna(0)

    +

    historical_drawing_events[
        "h2_prior_gate_events"
    ].fillna(0)
)


historical_drawing_events[
    "combined_prior_known_gate_usd"
] = (

    historical_drawing_events[
        "h1_prior_total_known_gate_usd"
    ].fillna(0)

    +

    historical_drawing_events[
        "h2_prior_total_known_gate_usd"
    ].fillna(0)
)


# ============================================================
# PART 10 — HISTORY AVAILABILITY
# ============================================================

historical_drawing_events[
    "prior_search_history_available"
] = (
    historical_drawing_events[
        "combined_prior_search_events"
    ] > 0
)


historical_drawing_events[
    "prior_gate_history_available"
] = (
    historical_drawing_events[
        "combined_prior_gate_events"
    ] > 0
)


historical_drawing_events[
    "historical_drawing_history_available"
] = (

    historical_drawing_events[
        "prior_search_history_available"
    ]

    |

    historical_drawing_events[
        "prior_gate_history_available"
    ]
)


# ============================================================
# PART 11 — IMPORTANT:
#
# We are NOT yet turning this into one Drawing Power score.
#
# Search acceleration history and gate history measure
# different things, and gate coverage is still extremely sparse.
# ============================================================


# ============================================================
# PART 12 — COVERAGE
# ============================================================

total_verified_main_events = len(
    historical_drawing_events
)


search_history_count = int(
    historical_drawing_events[
        "prior_search_history_available"
    ].sum()
)


gate_history_count = int(
    historical_drawing_events[
        "prior_gate_history_available"
    ].sum()
)


any_history_count = int(
    historical_drawing_events[
        "historical_drawing_history_available"
    ].sum()
)


# ============================================================
# PART 13 — CHECK IMPORTANT MCGREGOR EVENTS
# ============================================================

mcgregor_history_check = (
    historical_drawing_events[
        (
            historical_drawing_events[
                "headliner_1"
            ] == "Conor McGregor"
        )
        |
        (
            historical_drawing_events[
                "headliner_2"
            ] == "Conor McGregor"
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "headliner_1",
            "headliner_2",
            "combined_prior_search_events",
            "headliner_prior_avg_search",
            "headliner_prior_best_search",
            "combined_prior_gate_events",
            "headliner_prior_best_gate",
            "combined_prior_known_gate_usd"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 14 — NEW 11 CHECK
# ============================================================

new_drawing_check = (
    historical_drawing_events[
        historical_drawing_events[
            "event_date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "headliner_1",
            "headliner_2",
            "combined_prior_search_events",
            "headliner_prior_avg_search",
            "headliner_prior_best_search",
            "combined_prior_gate_events",
            "headliner_prior_best_gate",
            "historical_drawing_history_available"
        ]
    ]
)


# ============================================================
# PART 15 — NO CURRENT-EVENT LEAKAGE TEST
#
# A fighter's first-ever fanfare-evidence appearance must have
# zero prior evidence from that same appearance.
# ============================================================

leakage_failures = 0


for fighter, group in (
    historical_drawing_long
    .sort_values(
        "event_date"
    )
    .groupby(
        "fighter"
    )
):

    first = group.iloc[0]

    if (
        first[
            "prior_search_events"
        ] != 0
        or
        first[
            "prior_gate_events"
        ] != 0
    ):

        leakage_failures += 1


# ============================================================
# PART 16 — OUTPUT
# ============================================================

print()
print("=" * 110)
print("HISTORICAL DRAWING POWER COVERAGE")
print("=" * 110)

print(
    "Verified main events:",
    total_verified_main_events,
    "| expected 785"
)

print(
    "Events with prior search history:",
    search_history_count
)

print(
    "Events with prior gate history:",
    gate_history_count
)

print(
    "Events with any prior fanfare history:",
    any_history_count
)

print(
    "First-appearance leakage failures:",
    leakage_failures,
    "| expected 0"
)


print()
print("=" * 110)
print("MCGREGOR HISTORY CHECK")
print("=" * 110)

print(
    mcgregor_history_check.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("NEW 11 EVENTS")
print("=" * 110)

print(
    new_drawing_check.to_string(
        index=False
    )
)


# ============================================================
# PART 17 — PASS / FAIL
# ============================================================

step90_pass = (

    total_verified_main_events == 785

    and

    main_event_duplicates == 0

    and

    leakage_failures == 0

    and

    len(
        new_drawing_check
    ) == 11
)


print()
print("=" * 110)

if step90_pass:

    print(
        "STEP 90 HISTORICAL DRAWING BACKBONE: PASS ✅🔥"
    )

else:

    print(
        "STEP 90 HISTORICAL DRAWING BACKBONE: CHECK REQUIRED ⚠️"
    )

print("=" * 110)
```

```python
# ============================================================
# STEP 91 — REBUILD NON-FANFARE EVENT STRENGTH COMPONENTS
# ============================================================

import pandas as pd
import numpy as np


print("=" * 110)
print("STEP 91 — REBUILD NON-FANFARE EVENT STRENGTH COMPONENTS 🔥")
print("=" * 110)


# ============================================================
# PART 1 — START FROM AUTHORITATIVE EVENT MASTER
# ============================================================

normalized_updated = (
    event_master_updated.copy()
)


# ============================================================
# PART 2 — ORIGINAL NORMALIZATION FUNCTION
#
# Preserve zero as zero.
# Rank only positive values.
# Missing remains missing.
# ============================================================

def zero_preserving_percentile(series):

    result = pd.Series(
        index=series.index,
        dtype=float
    )

    result[
        series.isna()
    ] = np.nan

    result[
        series == 0
    ] = 0

    positive = (
        series > 0
    )

    if positive.sum() > 0:

        result.loc[
            positive
        ] = (
            series.loc[
                positive
            ]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )

    return result


# ============================================================
# PART 3 — VARIABLES USED BY THE ORIGINAL MODEL
# ============================================================

card_quality_variables = [

    "title_fights",
    "total_ranking_points",
    "avg_ufc_experience",
    "avg_ufc_win_pct",
    "avg_win_streak",
    "avg_finish_rate"
]


star_experience_variables = [

    "avg_previous_title_fights",
    "avg_previous_main_events",
    "avg_previous_bonuses"
]


main_event_variables = [

    "main_event_combined_ufc_fights",
    "main_event_avg_win_pct",
    "main_event_combined_win_streak",
    "main_event_combined_title_fights",
    "main_event_combined_previous_main_events",
    "main_event_combined_bonuses",
    "main_event_ranking_points"
]


all_model_variables = (

    card_quality_variables
    +

    star_experience_variables
    +

    main_event_variables
)


# ============================================================
# PART 4 — VERIFY ALL REQUIRED VARIABLES EXIST
# ============================================================

missing_variables = [

    col
    for col in all_model_variables
    if col not in normalized_updated.columns
]


print(
    "Required model variables:",
    len(all_model_variables)
)

print(
    "Missing required variables:",
    missing_variables
)


if missing_variables:

    raise ValueError(
        "Required Event Strength variables are missing. "
        "Stop before normalization."
    )


# ============================================================
# PART 5 — NORMALIZE EACH VARIABLE
# ============================================================

score_columns = []


for variable in all_model_variables:

    score_col = (
        variable
        +
        "_score"
    )

    normalized_updated[
        score_col
    ] = (
        zero_preserving_percentile(
            normalized_updated[
                variable
            ]
        )
    )

    score_columns.append(
        score_col
    )


# ============================================================
# PART 6 — MASK RANKING SCORES BEFORE RANKINGS EXIST
#
# Rankings begin 2013-02-04.
#
# Do NOT treat pre-ranking events as zero-quality.
# ============================================================

ranking_score_columns = [

    "total_ranking_points_score",
    "main_event_ranking_points_score"
]


for col in ranking_score_columns:

    normalized_updated.loc[
        ~normalized_updated[
            "rankings_available"
        ],
        col
    ] = np.nan


# ============================================================
# PART 7 — MASK BONUS-HISTORY SCORES BEFORE BONUS HISTORY EXISTS
#
# Bonus history begins 2006-01-16.
# ============================================================

bonus_score_columns = [

    "avg_previous_bonuses_score",
    "main_event_combined_bonuses_score"
]


for col in bonus_score_columns:

    normalized_updated.loc[
        ~normalized_updated[
            "bonus_history_available"
        ],
        col
    ] = np.nan


# ============================================================
# PART 8 — MASK MAIN-EVENT COMPONENT FOR UNVERIFIED EVENTS
#
# We have exactly three unverified main events.
# ============================================================

main_event_score_columns = [

    variable + "_score"
    for variable in main_event_variables
]


unverified_mask = (

    normalized_updated[
        "main_event_combined_ufc_fights"
    ].isna()
)


# ============================================================
# PART 9 — COMPONENT FUNCTION
#
# Mean only the metrics that are actually available.
# Missing historical systems are dynamically omitted.
# ============================================================

def component_mean(
    dataframe,
    columns
):

    return (
        dataframe[
            columns
        ]
        .mean(
            axis=1,
            skipna=True
        )
    )


# ============================================================
# PART 10 — CARD QUALITY
# ============================================================

card_quality_score_columns = [

    variable + "_score"
    for variable in card_quality_variables
]


normalized_updated[
    "card_quality_score"
] = (
    component_mean(
        normalized_updated,
        card_quality_score_columns
    )
)


# ============================================================
# PART 11 — STAR EXPERIENCE
# ============================================================

star_experience_score_columns = [

    variable + "_score"
    for variable in star_experience_variables
]


normalized_updated[
    "star_experience_score"
] = (
    component_mean(
        normalized_updated,
        star_experience_score_columns
    )
)


# ============================================================
# PART 12 — MAIN EVENT STRENGTH
# ============================================================

normalized_updated[
    "main_event_score"
] = (
    component_mean(
        normalized_updated,
        main_event_score_columns
    )
)


normalized_updated.loc[
    unverified_mask,
    "main_event_score"
] = np.nan


# ============================================================
# PART 13 — HISTORICAL PRESERVATION AUDIT
#
# IMPORTANT:
# Adding 11 new events changes percentile ranks slightly.
#
# Therefore raw historical event metrics MUST remain exact,
# which Step 86 proved.
#
# But normalized percentile scores are EXPECTED to move
# slightly because the comparison population is now 788
# instead of 777.
#
# Here we quantify that movement instead of incorrectly
# requiring exact equality.
# ============================================================

old_norm = (
    normalized.copy()
)


if "event_date" not in old_norm.columns:

    if "date" in old_norm.columns:

        old_norm = (
            old_norm.rename(
                columns={
                    "date":
                        "event_date"
                }
            )
        )


old_norm[
    "event_date"
] = pd.to_datetime(
    old_norm[
        "event_date"
    ]
)


shared_score_columns = [

    col
    for col in score_columns
    if col in old_norm.columns
]


historical_score_audit = []


if (
    "event_name" in old_norm.columns
    and
    "event_date" in old_norm.columns
):

    old_compare = (
        old_norm[
            [
                "event_date",
                "event_name"
            ]
            +
            shared_score_columns
        ]
        .copy()
    )


    new_compare = (
        normalized_updated[
            normalized_updated[
                "event_date"
            ]
            <= pd.Timestamp(
                "2026-06-27"
            )
        ]
        [
            [
                "event_date",
                "event_name"
            ]
            +
            shared_score_columns
        ]
        .copy()
    )


    score_compare = (
        old_compare
        .merge(
            new_compare,
            on=[
                "event_date",
                "event_name"
            ],
            how="inner",
            suffixes=(
                "_old",
                "_new"
            ),
            validate="one_to_one"
        )
    )


    for col in shared_score_columns:

        old_values = (
            score_compare[
                col + "_old"
            ]
        )

        new_values = (
            score_compare[
                col + "_new"
            ]
        )


        valid = (
            old_values.notna()
            &
            new_values.notna()
        )


        if valid.sum() > 0:

            differences = (
                new_values[
                    valid
                ]
                -
                old_values[
                    valid
                ]
            ).abs()


            mean_change = float(
                differences.mean()
            )

            max_change = float(
                differences.max()
            )

        else:

            mean_change = np.nan
            max_change = np.nan


        historical_score_audit.append({

            "score":
                col,

            "historical_events_compared":
                int(
                    valid.sum()
                ),

            "mean_absolute_change":
                mean_change,

            "max_absolute_change":
                max_change
        })


historical_score_audit = pd.DataFrame(
    historical_score_audit
)


# ============================================================
# PART 14 — COMPONENT VALIDATION
# ============================================================

total_events = len(
    normalized_updated
)


card_missing = int(
    normalized_updated[
        "card_quality_score"
    ]
    .isna()
    .sum()
)


star_missing = int(
    normalized_updated[
        "star_experience_score"
    ]
    .isna()
    .sum()
)


main_missing = int(
    normalized_updated[
        "main_event_score"
    ]
    .isna()
    .sum()
)


component_ranges = pd.DataFrame({

    "component": [

        "Card Quality",
        "Star Experience",
        "Main Event Strength"
    ],

    "minimum": [

        normalized_updated[
            "card_quality_score"
        ].min(),

        normalized_updated[
            "star_experience_score"
        ].min(),

        normalized_updated[
            "main_event_score"
        ].min()
    ],

    "maximum": [

        normalized_updated[
            "card_quality_score"
        ].max(),

        normalized_updated[
            "star_experience_score"
        ].max(),

        normalized_updated[
            "main_event_score"
        ].max()
    ],

    "missing": [

        card_missing,
        star_missing,
        main_missing
    ]
})


# ============================================================
# PART 15 — NEW 11 COMPONENT SCORES
# ============================================================

new_component_check = (
    normalized_updated[
        normalized_updated[
            "event_date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "card_quality_score",
            "star_experience_score",
            "main_event_score"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 16 — TOP 15 OF EACH COMPONENT
#
# Diagnostic only — NOT final Event Strength.
# ============================================================

top_card_quality = (
    normalized_updated[
        [
            "event_date",
            "event_name",
            "card_quality_score"
        ]
    ]
    .sort_values(
        "card_quality_score",
        ascending=False
    )
    .head(15)
)


top_star_experience = (
    normalized_updated[
        [
            "event_date",
            "event_name",
            "star_experience_score"
        ]
    ]
    .sort_values(
        "star_experience_score",
        ascending=False
    )
    .head(15)
)


top_main_events = (
    normalized_updated[
        [
            "event_date",
            "event_name",
            "main_event_score"
        ]
    ]
    .sort_values(
        "main_event_score",
        ascending=False
    )
    .head(15)
)


# ============================================================
# PART 17 — OUTPUT
# ============================================================

print()
print("=" * 110)
print("COMPONENT VALIDATION")
print("=" * 110)

print(
    "Total events:",
    total_events,
    "| expected 788"
)

print(
    component_ranges.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("HISTORICAL PERCENTILE MOVEMENT")
print("=" * 110)

print(
    "Adding the 11 new events changes percentile ranks slightly."
)

print(
    "This table measures that expected movement:"
)

if len(
    historical_score_audit
):

    print(
        historical_score_audit.to_string(
            index=False
        )
    )

else:

    print(
        "No comparable historical score columns found."
    )


print()
print("=" * 110)
print("NEW 11 COMPONENT SCORES")
print("=" * 110)

print(
    new_component_check.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("TOP 15 — CARD QUALITY")
print("=" * 110)

print(
    top_card_quality.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("TOP 15 — STAR EXPERIENCE")
print("=" * 110)

print(
    top_star_experience.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("TOP 15 — MAIN EVENT STRENGTH")
print("=" * 110)

print(
    top_main_events.to_string(
        index=False
    )
)


# ============================================================
# PART 18 — PASS / FAIL
# ============================================================

step91_pass = (

    total_events == 788

    and

    len(
        missing_variables
    ) == 0

    and

    card_missing == 0

    and

    star_missing == 0

    and

    main_missing == 3

    and

    len(
        new_component_check
    ) == 11

    and

    new_component_check[
        "card_quality_score"
    ].notna().all()

    and

    new_component_check[
        "star_experience_score"
    ].notna().all()

    and

    new_component_check[
        "main_event_score"
    ].notna().all()
)


print()
print("=" * 110)

if step91_pass:

    print(
        "STEP 91 NON-FANFARE COMPONENTS: PASS ✅🔥"
    )

else:

    print(
        "STEP 91 NON-FANFARE COMPONENTS: CHECK REQUIRED ⚠️"
    )

print("=" * 110)
```

```python
# ============================================================
# STEP 91B — FINISH NON-FANFARE COMPONENT VALIDATION
# ============================================================

import pandas as pd
import numpy as np


print("=" * 110)
print("STEP 91B — COMPLETE NON-FANFARE COMPONENT VALIDATION 🔥")
print("=" * 110)


# ============================================================
# PART 1 — CONFIRM STEP 91 CALCULATIONS SURVIVED
# ============================================================

required_step91_columns = [
    "card_quality_score",
    "star_experience_score",
    "main_event_score"
]

missing_step91_columns = [
    col
    for col in required_step91_columns
    if col not in normalized_updated.columns
]


print(
    "Missing Step 91 component columns:",
    missing_step91_columns
)


if missing_step91_columns:

    raise ValueError(
        "Step 91 stopped before component calculation. "
        "Do not continue."
    )


# ============================================================
# PART 2 — USE OLD FINAL DATASET FOR HISTORICAL COMPARISON
#
# We do NOT use the broken variable `normalized`.
# ============================================================

old_final = (
    final_event_strength_pre_update.copy()
)


# Standardize date column.
if (
    "event_date" not in old_final.columns
    and
    "date" in old_final.columns
):

    old_final = old_final.rename(
        columns={
            "date": "event_date"
        }
    )


old_final[
    "event_date"
] = pd.to_datetime(
    old_final[
        "event_date"
    ]
)


# ============================================================
# PART 3 — HISTORICAL COMPONENT MOVEMENT
#
# Percentile-based scores can move slightly because the
# comparison population increased from 777 to 788.
# ============================================================

component_columns = [
    "card_quality_score",
    "star_experience_score",
    "main_event_score"
]


historical_component_audit = []


old_compare = (
    old_final[
        [
            "event_date",
            "event_name"
        ]
        +
        component_columns
    ]
    .copy()
)


new_compare = (
    normalized_updated[
        normalized_updated[
            "event_date"
        ]
        <= pd.Timestamp("2026-06-27")
    ]
    [
        [
            "event_date",
            "event_name"
        ]
        +
        component_columns
    ]
    .copy()
)


component_compare = (
    old_compare
    .merge(
        new_compare,
        on=[
            "event_date",
            "event_name"
        ],
        how="inner",
        suffixes=(
            "_old",
            "_new"
        ),
        validate="one_to_one"
    )
)


for col in component_columns:

    old_values = (
        component_compare[
            col + "_old"
        ]
    )

    new_values = (
        component_compare[
            col + "_new"
        ]
    )


    both_missing = (
        old_values.isna()
        &
        new_values.isna()
    )


    both_present = (
        old_values.notna()
        &
        new_values.notna()
    )


    if both_present.sum() > 0:

        differences = (
            new_values[
                both_present
            ]
            -
            old_values[
                both_present
            ]
        ).abs()


        mean_change = float(
            differences.mean()
        )


        max_change = float(
            differences.max()
        )

    else:

        mean_change = np.nan
        max_change = np.nan


    missing_pattern_mismatches = int(
        (
            old_values.isna()
            !=
            new_values.isna()
        ).sum()
    )


    historical_component_audit.append({

        "component":
            col,

        "events_compared":
            int(
                both_present.sum()
            ),

        "both_missing":
            int(
                both_missing.sum()
            ),

        "missing_pattern_mismatches":
            missing_pattern_mismatches,

        "mean_absolute_change":
            mean_change,

        "max_absolute_change":
            max_change
    })


historical_component_audit = pd.DataFrame(
    historical_component_audit
)


# ============================================================
# PART 4 — VALIDATE COMPONENT COVERAGE
# ============================================================

total_events = len(
    normalized_updated
)


unique_events = (
    normalized_updated[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


card_missing = int(
    normalized_updated[
        "card_quality_score"
    ]
    .isna()
    .sum()
)


star_missing = int(
    normalized_updated[
        "star_experience_score"
    ]
    .isna()
    .sum()
)


main_missing = int(
    normalized_updated[
        "main_event_score"
    ]
    .isna()
    .sum()
)


# ============================================================
# PART 5 — IDENTIFY MISSING MAIN EVENTS
# ============================================================

missing_main_events = (
    normalized_updated[
        normalized_updated[
            "main_event_score"
        ].isna()
    ]
    [
        [
            "event_date",
            "event_name"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 6 — NEW 11 COMPONENT SCORES
# ============================================================

new_component_check = (
    normalized_updated[
        normalized_updated[
            "event_date"
        ]
        > pd.Timestamp("2026-06-27")
    ]
    [
        [
            "event_date",
            "event_name",
            "card_quality_score",
            "star_experience_score",
            "main_event_score"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 7 — COMPONENT RANGES
# ============================================================

component_ranges = pd.DataFrame({

    "component": [
        "Card Quality",
        "Star Experience",
        "Main Event Strength"
    ],

    "minimum": [
        normalized_updated[
            "card_quality_score"
        ].min(),

        normalized_updated[
            "star_experience_score"
        ].min(),

        normalized_updated[
            "main_event_score"
        ].min()
    ],

    "maximum": [
        normalized_updated[
            "card_quality_score"
        ].max(),

        normalized_updated[
            "star_experience_score"
        ].max(),

        normalized_updated[
            "main_event_score"
        ].max()
    ],

    "missing": [
        card_missing,
        star_missing,
        main_missing
    ]
})


# ============================================================
# PART 8 — TOP 15 DIAGNOSTICS
# ============================================================

top_card_quality = (
    normalized_updated[
        [
            "event_date",
            "event_name",
            "card_quality_score"
        ]
    ]
    .sort_values(
        "card_quality_score",
        ascending=False
    )
    .head(15)
)


top_star_experience = (
    normalized_updated[
        [
            "event_date",
            "event_name",
            "star_experience_score"
        ]
    ]
    .sort_values(
        "star_experience_score",
        ascending=False
    )
    .head(15)
)


top_main_events = (
    normalized_updated[
        [
            "event_date",
            "event_name",
            "main_event_score"
        ]
    ]
    .sort_values(
        "main_event_score",
        ascending=False
    )
    .head(15)
)


# ============================================================
# PART 9 — OUTPUT
# ============================================================

print()
print("=" * 110)
print("COMPONENT VALIDATION")
print("=" * 110)

print(
    "Total events:",
    total_events,
    "| expected 788"
)

print(
    "Unique events:",
    unique_events,
    "| expected 788"
)

print()

print(
    component_ranges.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("MISSING MAIN-EVENT SCORES")
print("=" * 110)

print(
    missing_main_events.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("HISTORICAL COMPONENT MOVEMENT")
print("=" * 110)

print(
    "Historical raw metrics were already preserved exactly in Step 86."
)

print(
    "Small score movement here is expected because percentile ranks now use 788 events."
)

print()

print(
    historical_component_audit.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("NEW 11 COMPONENT SCORES")
print("=" * 110)

print(
    new_component_check.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("TOP 15 — CARD QUALITY")
print("=" * 110)

print(
    top_card_quality.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("TOP 15 — STAR EXPERIENCE")
print("=" * 110)

print(
    top_star_experience.to_string(
        index=False
    )
)


print()
print("=" * 110)
print("TOP 15 — MAIN EVENT STRENGTH")
print("=" * 110)

print(
    top_main_events.to_string(
        index=False
    )
)


# ============================================================
# PART 10 — FINAL PASS / FAIL
# ============================================================

expected_unverified = {
    (
        pd.Timestamp("2000-03-10"),
        "UFC 24: First Defense"
    ),

    (
        pd.Timestamp("2025-05-31"),
        "UFC Fight Night: Blanchfield vs. Barber"
    ),

    (
        pd.Timestamp("2025-08-22"),
        "UFC - Road to UFC 4.6"
    )
}


actual_unverified = set(
    zip(
        missing_main_events[
            "event_date"
        ],

        missing_main_events[
            "event_name"
        ]
    )
)


historical_missing_patterns_ok = bool(
    (
        historical_component_audit[
            "missing_pattern_mismatches"
        ] == 0
    ).all()
)


step91b_pass = (

    total_events == 788

    and

    unique_events == 788

    and

    card_missing == 0

    and

    star_missing == 0

    and

    main_missing == 3

    and

    actual_unverified == expected_unverified

    and

    historical_missing_patterns_ok

    and

    len(
        new_component_check
    ) == 11

    and

    new_component_check[
        "card_quality_score"
    ].notna().all()

    and

    new_component_check[
        "star_experience_score"
    ].notna().all()

    and

    new_component_check[
        "main_event_score"
    ].notna().all()
)


print()
print("=" * 110)

if step91b_pass:

    print(
        "STEP 91B NON-FANFARE COMPONENTS: PASS ✅🔥"
    )

else:

    print(
        "STEP 91B NON-FANFARE COMPONENTS: CHECK REQUIRED ⚠️"
    )

print("=" * 110)
```

```python
# ============================================================
# STEP 93 — FANFARE SCORING FOUNDATION
# ============================================================

import pandas as pd
import numpy as np


print("=" * 115)
print("STEP 93 — FANFARE SCORING FOUNDATION 🔥")
print("=" * 115)


# ============================================================
# PART 1 — SAFETY COPIES
# ============================================================

fanfare_master_pre_step93 = fanfare_master.copy()

historical_drawing_events_pre_step93 = (
    historical_drawing_events.copy()
)


# ============================================================
# PART 2 — ZERO-PRESERVING PERCENTILE FUNCTION
# ============================================================

def fanfare_percentile(series):

    result = pd.Series(
        np.nan,
        index=series.index,
        dtype=float
    )

    valid = series.notna()

    zero = valid & (series == 0)

    positive = valid & (series > 0)

    result.loc[zero] = 0.0

    if positive.sum() > 0:

        result.loc[positive] = (
            series.loc[positive]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )

    return result


# ============================================================
# PART 3 — SEARCH HYPE ACCELERATION SCORE
#
# IMPORTANT:
#
# This is NOT absolute popularity.
#
# It measures how much search interest accelerated relative
# to the fighter's/event's prior baseline.
#
# Therefore:
#   - useful for "buzz"
#   - useful for sudden breakout interest
#   - NOT sufficient by itself for overall Fanfare
# ============================================================

fanfare_master[
    "search_acceleration_score"
] = fanfare_percentile(
    fanfare_master[
        "search_hype_acceleration"
    ]
)


# Events without Trends data remain missing.
fanfare_master.loc[
    ~fanfare_master[
        "search_acceleration_available"
    ],
    "search_acceleration_score"
] = np.nan


# ============================================================
# PART 4 — VERIFIED GATE SIGNAL
#
# Gate is one of our strongest commercial-demand signals.
#
# We use log(gate) because:
#   $26M vs $7M should matter,
# but raw dollars should not overwhelm the model.
#
# Missing gate stays NaN.
# ============================================================

fanfare_master[
    "gate_log_signal"
] = np.where(
    fanfare_master[
        "gate_usd_nominal"
    ].notna(),

    np.log1p(
        fanfare_master[
            "gate_usd_nominal"
        ]
    ),

    np.nan
)


# ============================================================
# PART 5 — GATE DIAGNOSTIC PERCENTILE
#
# WARNING:
#
# Only six gates are currently verified.
#
# This score is useful for inspection but is NOT yet a
# production-quality historical percentile.
# ============================================================

fanfare_master[
    "gate_diagnostic_score"
] = fanfare_percentile(
    fanfare_master[
        "gate_log_signal"
    ]
)


# ============================================================
# PART 6 — PREPARE HISTORICAL DRAWING SIGNALS
#
# We are deliberately NOT treating prior search acceleration
# as absolute drawing power.
#
# Instead, we preserve several separate pieces of evidence.
# ============================================================

drawing_merge_columns = [

    "event_date",
    "event_name",

    "combined_prior_search_events",
    "headliner_prior_avg_search",
    "headliner_prior_best_search",

    "combined_prior_gate_events",
    "headliner_prior_best_gate",
    "combined_prior_known_gate_usd",

    "prior_search_history_available",
    "prior_gate_history_available",
    "historical_drawing_history_available"
]


drawing_for_merge = (
    historical_drawing_events[
        drawing_merge_columns
    ]
    .copy()
)


# ============================================================
# PART 7 — ATTACH DRAWING HISTORY TO ALL 788 EVENTS
#
# The three unverified-main-event cards will naturally have
# no drawing-history row.
# ============================================================

columns_to_remove_before_merge = [

    col
    for col in drawing_merge_columns
    if (
        col not in [
            "event_date",
            "event_name"
        ]
        and
        col in fanfare_master.columns
    )
]


fanfare_master = (
    fanfare_master
    .drop(
        columns=columns_to_remove_before_merge,
        errors="ignore"
    )
    .merge(
        drawing_for_merge,
        on=[
            "event_date",
            "event_name"
        ],
        how="left",
        validate="one_to_one"
    )
)


# ============================================================
# PART 8 — CLEAN AVAILABILITY FLAGS
#
# Missing row ≠ False unless we know why it's missing.
#
# For the three unverified main events, drawing-history
# availability is unknown/not applicable.
# ============================================================

verified_event_keys = set(
    zip(
        historical_drawing_events[
            "event_date"
        ],
        historical_drawing_events[
            "event_name"
        ]
    )
)


fanfare_master[
    "main_event_verified_for_drawing"
] = [

    (
        row.event_date,
        row.event_name
    )
    in verified_event_keys

    for row in fanfare_master[
        [
            "event_date",
            "event_name"
        ]
    ].itertuples(
        index=False
    )
]


# ============================================================
# PART 9 — HISTORICAL GATE SIGNAL
#
# A fighter/event can only receive this signal if the
# headliners had a verified prior gate BEFORE the event.
#
# Missing is NOT zero.
# ============================================================

fanfare_master[
    "historical_gate_log_signal"
] = np.where(

    fanfare_master[
        "prior_gate_history_available"
    ] == True,

    np.log1p(
        fanfare_master[
            "headliner_prior_best_gate"
        ]
    ),

    np.nan
)


fanfare_master[
    "historical_gate_diagnostic_score"
] = (
    fanfare_percentile(
        fanfare_master[
            "historical_gate_log_signal"
        ]
    )
)


# ============================================================
# PART 10 — PRIOR SEARCH HISTORY SIGNALS
#
# These remain CONTEXT signals.
#
# We normalize them so we can inspect them, but they will
# NOT automatically become Historical Drawing Power.
# ============================================================

fanfare_master[
    "prior_avg_search_acceleration_score"
] = (
    fanfare_percentile(
        fanfare_master[
            "headliner_prior_avg_search"
        ]
    )
)


fanfare_master[
    "prior_best_search_acceleration_score"
] = (
    fanfare_percentile(
        fanfare_master[
            "headliner_prior_best_search"
        ]
    )
)


# ============================================================
# PART 11 — EXPERIENCE OF FANFARE EVIDENCE
#
# Number of prior main events for which we possess search
# fanfare evidence.
#
# This does NOT say the fighter was popular.
#
# It only describes the amount of historical fanfare evidence
# available for those headliners.
# ============================================================

fanfare_master[
    "prior_search_evidence_depth"
] = (
    fanfare_master[
        "combined_prior_search_events"
    ]
)


# Do not convert missing unverified-main-event cases to zero.
fanfare_master.loc[
    ~fanfare_master[
        "main_event_verified_for_drawing"
    ],
    "prior_search_evidence_depth"
] = np.nan


# ============================================================
# PART 12 — ABSOLUTE SEARCH PLACEHOLDER
#
# Still intentionally missing.
#
# DO NOT insert the fixed 2025–2026 calibration here.
# We need EVENT-TIME common-scale search demand.
# ============================================================

if (
    "absolute_search_demand"
    not in fanfare_master.columns
):

    fanfare_master[
        "absolute_search_demand"
    ] = np.nan


if (
    "absolute_search_available"
    not in fanfare_master.columns
):

    fanfare_master[
        "absolute_search_available"
    ] = False


# ============================================================
# PART 13 — PRODUCTION FANFARE SIGNAL STATUS
#
# For now:
#
# Search acceleration = production-capable buzz signal
# Gate                = high-quality but sparse signal
# Historical gate     = high-quality but sparse prior signal
# Absolute search     = waiting
#
# Prior acceleration history remains diagnostic/contextual.
# ============================================================

fanfare_master[
    "production_search_signal_available"
] = (
    fanfare_master[
        "search_acceleration_score"
    ].notna()
)


fanfare_master[
    "production_gate_signal_available"
] = (
    fanfare_master[
        "gate_log_signal"
    ].notna()
)


fanfare_master[
    "production_historical_gate_available"
] = (
    fanfare_master[
        "historical_gate_log_signal"
    ].notna()
)


fanfare_master[
    "production_absolute_search_available"
] = (
    fanfare_master[
        "absolute_search_demand"
    ].notna()
)


production_availability_columns = [

    "production_search_signal_available",
    "production_gate_signal_available",
    "production_historical_gate_available",
    "production_absolute_search_available"
]


fanfare_master[
    "production_fanfare_signals_available"
] = (
    fanfare_master[
        production_availability_columns
    ]
    .sum(
        axis=1
    )
)


# ============================================================
# PART 14 — COVERAGE SUMMARY
# ============================================================

production_coverage = pd.DataFrame({

    "signal": [

        "Search Hype Acceleration",
        "Current Verified Gate",
        "Prior Verified Gate",
        "Absolute Event-Time Search"
    ],

    "available_events": [

        int(
            fanfare_master[
                "production_search_signal_available"
            ].sum()
        ),

        int(
            fanfare_master[
                "production_gate_signal_available"
            ].sum()
        ),

        int(
            fanfare_master[
                "production_historical_gate_available"
            ].sum()
        ),

        int(
            fanfare_master[
                "production_absolute_search_available"
            ].sum()
        )
    ]
})


production_coverage[
    "coverage_pct"
] = (
    production_coverage[
        "available_events"
    ]
    /
    len(
        fanfare_master
    )
    * 100
)


# ============================================================
# PART 15 — NUMBER OF SIGNALS AVAILABLE PER EVENT
# ============================================================

signal_count_distribution = (
    fanfare_master[
        "production_fanfare_signals_available"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "signals_available"
    )
    .reset_index(
        name="events"
    )
)


signal_count_distribution[
    "pct"
] = (
    signal_count_distribution[
        "events"
    ]
    /
    len(
        fanfare_master
    )
    * 100
)


# ============================================================
# PART 16 — IMPORTANT CARD AUDIT
# ============================================================

important_pattern = (
    "UFC 189|"
    "UFC 194|"
    "UFC 196|"
    "UFC 202|"
    "UFC 205|"
    "UFC 229|"
    "UFC 246|"
    "UFC 257|"
    "UFC 264|"
    "UFC 300|"
    "UFC 329|"
    "UFC 330"
)


important_fanfare_audit = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        .str.contains(
            important_pattern,
            regex=True,
            na=False
        )
    ]
    [
        [
            "event_date",
            "event_name",

            "search_hype_acceleration",
            "search_acceleration_score",

            "gate_usd_nominal",
            "gate_diagnostic_score",

            "combined_prior_search_events",
            "headliner_prior_avg_search",
            "prior_avg_search_acceleration_score",

            "combined_prior_gate_events",
            "headliner_prior_best_gate",
            "historical_gate_diagnostic_score",

            "production_fanfare_signals_available"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 17 — NEW 11 AUDIT
# ============================================================

new_fanfare_audit = (
    fanfare_master[
        fanfare_master[
            "event_date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",

            "search_hype_acceleration",
            "search_acceleration_score",

            "gate_usd_nominal",
            "gate_diagnostic_score",

            "combined_prior_search_events",

            "combined_prior_gate_events",
            "headliner_prior_best_gate",

            "production_fanfare_signals_available"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 18 — SANITY TESTS
# ============================================================

total_events = len(
    fanfare_master
)


unique_events = (
    fanfare_master[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


search_available = int(
    fanfare_master[
        "production_search_signal_available"
    ].sum()
)


current_gate_available = int(
    fanfare_master[
        "production_gate_signal_available"
    ].sum()
)


historical_gate_available = int(
    fanfare_master[
        "production_historical_gate_available"
    ].sum()
)


absolute_search_available = int(
    fanfare_master[
        "production_absolute_search_available"
    ].sum()
)


fake_current_gate_zeros = int(
    fanfare_master[
        "gate_usd_nominal"
    ]
    .eq(0)
    .sum()
)


fake_historical_gate_zeros = int(
    fanfare_master[
        "historical_gate_log_signal"
    ]
    .eq(0)
    .sum()
)


# ============================================================
# PART 19 — UFC 329 SPECIFIC CHECK
# ============================================================

ufc329_check = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        ==
        "UFC 329: McGregor vs. Holloway 2"
    ]
)


if len(
    ufc329_check
) == 1:

    ufc329 = ufc329_check.iloc[0]

    ufc329_current_gate_ok = bool(
        np.isclose(
            ufc329[
                "gate_usd_nominal"
            ],
            26430566.0
        )
    )

    ufc329_prior_gate_ok = bool(
        np.isclose(
            ufc329[
                "headliner_prior_best_gate"
            ],
            17740543.0
        )
    )

else:

    ufc329_current_gate_ok = False
    ufc329_prior_gate_ok = False


# ============================================================
# PART 20 — OUTPUT
# ============================================================

print()
print("=" * 115)
print("GLOBAL VALIDATION")
print("=" * 115)

print(
    "Total events:",
    total_events,
    "| expected 788"
)

print(
    "Unique events:",
    unique_events,
    "| expected 788"
)

print(
    "Search signals:",
    search_available,
    "| expected 516"
)

print(
    "Current verified gates:",
    current_gate_available,
    "| expected 6"
)

print(
    "Prior verified-gate history:",
    historical_gate_available,
    "| expected 15"
)

print(
    "Absolute event-time search:",
    absolute_search_available,
    "| expected 0 for now"
)

print(
    "Fake current gate zeros:",
    fake_current_gate_zeros,
    "| expected 0"
)

print(
    "Fake historical gate zeros:",
    fake_historical_gate_zeros,
    "| expected 0"
)


print()
print("=" * 115)
print("PRODUCTION SIGNAL COVERAGE")
print("=" * 115)

print(
    production_coverage.to_string(
        index=False,
        formatters={
            "coverage_pct":
                lambda x:
                f"{x:.2f}%"
        }
    )
)


print()
print("=" * 115)
print("SIGNALS AVAILABLE PER EVENT")
print("=" * 115)

print(
    signal_count_distribution.to_string(
        index=False,
        formatters={
            "pct":
                lambda x:
                f"{x:.2f}%"
        }
    )
)


print()
print("=" * 115)
print("IMPORTANT FANFARE AUDIT")
print("=" * 115)

print(
    important_fanfare_audit.to_string(
        index=False
    )
)


print()
print("=" * 115)
print("NEW 11 EVENTS")
print("=" * 115)

print(
    new_fanfare_audit.to_string(
        index=False
    )
)


# ============================================================
# PART 21 — PASS / FAIL
# ============================================================

step93_pass = (

    total_events == 788

    and

    unique_events == 788

    and

    search_available == 516

    and

    current_gate_available == 6

    and

    historical_gate_available == 15

    and

    absolute_search_available == 0

    and

    fake_current_gate_zeros == 0

    and

    fake_historical_gate_zeros == 0

    and

    ufc329_current_gate_ok

    and

    ufc329_prior_gate_ok

    and

    len(
        new_fanfare_audit
    ) == 11
)


print()
print("=" * 115)

if step93_pass:

    print(
        "STEP 93 FANFARE SCORING FOUNDATION: PASS ✅🔥"
    )

else:

    print(
        "STEP 93 FANFARE SCORING FOUNDATION: CHECK REQUIRED ⚠️"
    )

print("=" * 115)
```

```python
# ============================================================
# STEP 94 — PRODUCTION FANFARE ENGINE ARCHITECTURE
# ============================================================

import pandas as pd
import numpy as np


print("=" * 118)
print("STEP 94 — PRODUCTION FANFARE ENGINE ARCHITECTURE 🔥")
print("=" * 118)


# ============================================================
# PART 1 — SAFETY COPY
# ============================================================

fanfare_master_pre_step94 = (
    fanfare_master.copy()
)


# ============================================================
# PART 2 — DEFINE FANFARE CONCEPTUAL WEIGHTS
#
# IMPORTANT:
#
# These are INTERNAL FANFARE weights.
#
# They are separate from the FINAL Event Strength weights:
#
#   Card Quality       37%
#   Star Experience    23%
#   Main Event         28%
#   Fanfare            12%
#
#
# FANFARE itself will eventually consist of:
#
#   Absolute Search Demand       35%
#   Current Commercial Demand    35%
#   Historical Drawing Power     20%
#   Search Hype Acceleration     10%
#
#
# Why only 10% acceleration?
#
# Because Step 93 proved that acceleration can heavily favor
# low-baseline fighters.
# ============================================================

FANFARE_WEIGHTS = {

    "absolute_search_score":
        0.35,

    "commercial_demand_score":
        0.35,

    "historical_drawing_score":
        0.20,

    "search_acceleration_score":
        0.10
}


fanfare_weight_total = sum(
    FANFARE_WEIGHTS.values()
)


print()
print("Target internal Fanfare weights:")

for key, value in (
    FANFARE_WEIGHTS.items()
):

    print(
        f"{key:<35} {value:>6.1%}"
    )


print()
print(
    "Internal Fanfare total:",
    f"{fanfare_weight_total:.1%}"
)


if not np.isclose(
    fanfare_weight_total,
    1.0
):

    raise ValueError(
        "Fanfare weights do not total 100%."
    )


# ============================================================
# PART 3 — ABSOLUTE SEARCH SCORE
#
# Still intentionally missing.
#
# We create the production column now so that later the
# event-time common-anchor data can drop directly into it.
# ============================================================

fanfare_master[
    "absolute_search_score"
] = np.nan


# ============================================================
# PART 4 — COMMERCIAL DEMAND
#
# Current verified gate is excellent evidence when available.
#
# But only six events currently have verified gate data.
#
# Therefore:
#
#   Known gate -> usable commercial evidence.
#   Missing gate -> NaN, NOT 0.
#
# For now we retain the six-event diagnostic percentile.
# It is explicitly PROVISIONAL until gate coverage improves.
# ============================================================

fanfare_master[
    "commercial_demand_score"
] = (
    fanfare_master[
        "gate_diagnostic_score"
    ]
)


fanfare_master[
    "commercial_demand_available"
] = (
    fanfare_master[
        "gate_usd_nominal"
    ].notna()
)


# ============================================================
# PART 5 — HISTORICAL DRAWING POWER
#
# IMPORTANT:
#
# We currently have two types of historical information:
#
# A) prior search acceleration
# B) prior verified gate
#
# Prior acceleration does NOT equal popularity.
#
# Therefore we DO NOT average prior acceleration into drawing
# power.
#
# For production purposes right now:
#
# Historical Drawing Power exists ONLY where prior verified
# commercial evidence exists.
#
# Later:
#   prior absolute search
#   prior PPV
#   prior viewership
# can expand this signal.
# ============================================================

fanfare_master[
    "historical_drawing_score"
] = (
    fanfare_master[
        "historical_gate_diagnostic_score"
    ]
)


fanfare_master[
    "historical_drawing_available"
] = (
    fanfare_master[
        "historical_gate_log_signal"
    ].notna()
)


# ============================================================
# PART 6 — SEARCH ACCELERATION
#
# Already calculated correctly in Step 93.
# ============================================================

fanfare_master[
    "search_acceleration_available_final"
] = (
    fanfare_master[
        "search_acceleration_score"
    ].notna()
)


# ============================================================
# PART 7 — AVAILABILITY MATRIX
# ============================================================

fanfare_master[
    "absolute_search_available_final"
] = (
    fanfare_master[
        "absolute_search_score"
    ].notna()
)


fanfare_master[
    "commercial_demand_available_final"
] = (
    fanfare_master[
        "commercial_demand_score"
    ].notna()
)


fanfare_master[
    "historical_drawing_available_final"
] = (
    fanfare_master[
        "historical_drawing_score"
    ].notna()
)


final_signal_columns = [

    "absolute_search_available_final",
    "commercial_demand_available_final",
    "historical_drawing_available_final",
    "search_acceleration_available_final"
]


fanfare_master[
    "final_fanfare_signals_available"
] = (
    fanfare_master[
        final_signal_columns
    ]
    .sum(
        axis=1
    )
)


# ============================================================
# PART 8 — CURRENT FANFARE READINESS
#
# We separate:
#
# READY FOR PROVISIONAL INSPECTION
# from
# READY FOR FINAL MODEL
#
#
# Final model should ideally have Absolute Search.
#
# Until then, NO event is marked fully final-ready.
# ============================================================

fanfare_master[
    "provisional_fanfare_ready"
] = (

    fanfare_master[
        "search_acceleration_available_final"
    ]

    &

    (
        fanfare_master[
            "commercial_demand_available_final"
        ]

        |

        fanfare_master[
            "historical_drawing_available_final"
        ]
    )
)


fanfare_master[
    "final_fanfare_ready"
] = (

    fanfare_master[
        "absolute_search_available_final"
    ]

    &

    fanfare_master[
        "search_acceleration_available_final"
    ]
)


# ============================================================
# PART 9 — DYNAMIC WEIGHT FUNCTION
#
# This is the SAME PRINCIPLE as our final Event Strength model:
#
# Missing data is omitted and remaining weights are
# redistributed.
#
# HOWEVER:
#
# This function is only used for a PROVISIONAL diagnostic
# score right now.
# ============================================================

def weighted_score_with_missing(
    row,
    weights
):

    available = {

        col: weight

        for col, weight
        in weights.items()

        if pd.notna(
            row[col]
        )
    }


    if not available:

        return np.nan


    total_available_weight = sum(
        available.values()
    )


    return sum(

        row[col] * weight

        for col, weight
        in available.items()

    ) / total_available_weight


# ============================================================
# PART 10 — PROVISIONAL FANFARE SCORE
#
# DO NOT use this in final Event Strength yet.
#
# This is purely for inspecting behavior of the architecture.
# ============================================================

fanfare_master[
    "provisional_fanfare_score"
] = (
    fanfare_master.apply(

        lambda row:
        weighted_score_with_missing(
            row,
            FANFARE_WEIGHTS
        ),

        axis=1
    )
)


# ============================================================
# PART 11 — PROVISIONAL SCORE QUALITY CLASS
#
# This tells us what actually produced the score.
# ============================================================

def provisional_quality(row):

    absolute_search = bool(
        row[
            "absolute_search_available_final"
        ]
    )

    commercial = bool(
        row[
            "commercial_demand_available_final"
        ]
    )

    historical = bool(
        row[
            "historical_drawing_available_final"
        ]
    )

    acceleration = bool(
        row[
            "search_acceleration_available_final"
        ]
    )


    if (
        absolute_search
        and
        commercial
        and
        historical
        and
        acceleration
    ):
        return "Complete"


    if (
        commercial
        and
        historical
        and
        acceleration
    ):
        return (
            "Strong Provisional "
            "(No Absolute Search)"
        )


    if (
        commercial
        and
        acceleration
    ):
        return (
            "Commercial + Acceleration"
        )


    if (
        historical
        and
        acceleration
    ):
        return (
            "Historical + Acceleration"
        )


    if acceleration:
        return (
            "Acceleration Only"
        )


    if commercial:
        return (
            "Commercial Only"
        )


    if historical:
        return (
            "Historical Only"
        )


    return "No Fanfare Evidence"


fanfare_master[
    "provisional_fanfare_quality"
] = (
    fanfare_master.apply(
        provisional_quality,
        axis=1
    )
)


# ============================================================
# PART 12 — CRITICAL SAFETY RULE
#
# Acceleration-only scores must NOT be interpreted as final
# Fanfare.
#
# This is the exact mistake we're correcting.
# ============================================================

fanfare_master[
    "acceleration_only"
] = (

    fanfare_master[
        "search_acceleration_available_final"
    ]

    &

    ~fanfare_master[
        "commercial_demand_available_final"
    ]

    &

    ~fanfare_master[
        "historical_drawing_available_final"
    ]

    &

    ~fanfare_master[
        "absolute_search_available_final"
    ]
)


# ============================================================
# PART 13 — COVERAGE SUMMARY
# ============================================================

readiness_summary = pd.DataFrame({

    "metric": [

        "Absolute Search",
        "Commercial Demand",
        "Historical Drawing",
        "Search Acceleration",
        "Provisional Ready",
        "Final Ready",
        "Acceleration Only"
    ],

    "events": [

        int(
            fanfare_master[
                "absolute_search_available_final"
            ].sum()
        ),

        int(
            fanfare_master[
                "commercial_demand_available_final"
            ].sum()
        ),

        int(
            fanfare_master[
                "historical_drawing_available_final"
            ].sum()
        ),

        int(
            fanfare_master[
                "search_acceleration_available_final"
            ].sum()
        ),

        int(
            fanfare_master[
                "provisional_fanfare_ready"
            ].sum()
        ),

        int(
            fanfare_master[
                "final_fanfare_ready"
            ].sum()
        ),

        int(
            fanfare_master[
                "acceleration_only"
            ].sum()
        )
    ]
})


readiness_summary[
    "pct"
] = (
    readiness_summary[
        "events"
    ]
    /
    len(
        fanfare_master
    )
    * 100
)


# ============================================================
# PART 14 — QUALITY DISTRIBUTION
# ============================================================

quality_distribution = (
    fanfare_master[
        "provisional_fanfare_quality"
    ]
    .value_counts()
    .rename_axis(
        "quality"
    )
    .reset_index(
        name="events"
    )
)


quality_distribution[
    "pct"
] = (
    quality_distribution[
        "events"
    ]
    /
    len(
        fanfare_master
    )
    * 100
)


# ============================================================
# PART 15 — IMPORTANT EVENT AUDIT
# ============================================================

important_pattern = (
    "UFC 189|"
    "UFC 194|"
    "UFC 196|"
    "UFC 202|"
    "UFC 205|"
    "UFC 229|"
    "UFC 246|"
    "UFC 257|"
    "UFC 264|"
    "UFC 300|"
    "UFC 329|"
    "UFC 330"
)


important_architecture_audit = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        .str.contains(
            important_pattern,
            regex=True,
            na=False
        )
    ]
    [
        [
            "event_date",
            "event_name",

            "absolute_search_score",

            "commercial_demand_score",
            "historical_drawing_score",
            "search_acceleration_score",

            "final_fanfare_signals_available",

            "provisional_fanfare_score",
            "provisional_fanfare_quality",

            "provisional_fanfare_ready",
            "final_fanfare_ready"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 16 — NEW 11
# ============================================================

new_architecture_audit = (
    fanfare_master[
        fanfare_master[
            "event_date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",

            "commercial_demand_score",
            "historical_drawing_score",
            "search_acceleration_score",

            "final_fanfare_signals_available",

            "provisional_fanfare_score",
            "provisional_fanfare_quality",

            "provisional_fanfare_ready"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 17 — UFC 329 vs LOW-BASELINE SPIKES
#
# This is one of the most important diagnostics.
# ============================================================

comparison_names = [

    "UFC 329: McGregor vs. Holloway 2",

    "UFC Fight Night: Medic vs. Rodriguez",

    "UFC Fight Night: Hernandez vs. Rodrigues",

    "UFC 330: Makhachev vs. Machado Garry"
]


problem_case_comparison = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        .isin(
            comparison_names
        )
    ]
    [
        [
            "event_name",

            "search_acceleration_score",
            "commercial_demand_score",
            "historical_drawing_score",

            "provisional_fanfare_score",
            "provisional_fanfare_quality"
        ]
    ]
)


# ============================================================
# PART 18 — TOP PROVISIONAL SCORES
#
# DIAGNOSTIC ONLY.
#
# This ranking is NOT our final Fanfare ranking.
# ============================================================

top_provisional = (
    fanfare_master[
        [
            "event_date",
            "event_name",

            "provisional_fanfare_score",
            "provisional_fanfare_quality",

            "commercial_demand_score",
            "historical_drawing_score",
            "search_acceleration_score"
        ]
    ]
    .sort_values(
        "provisional_fanfare_score",
        ascending=False
    )
    .head(25)
)


# ============================================================
# PART 19 — VALIDATION COUNTS
# ============================================================

total_events = len(
    fanfare_master
)


absolute_count = int(
    fanfare_master[
        "absolute_search_available_final"
    ].sum()
)


commercial_count = int(
    fanfare_master[
        "commercial_demand_available_final"
    ].sum()
)


historical_count = int(
    fanfare_master[
        "historical_drawing_available_final"
    ].sum()
)


acceleration_count = int(
    fanfare_master[
        "search_acceleration_available_final"
    ].sum()
)


final_ready_count = int(
    fanfare_master[
        "final_fanfare_ready"
    ].sum()
)


# ============================================================
# PART 20 — OUTPUT
# ============================================================

print()
print("=" * 118)
print("FANFARE READINESS")
print("=" * 118)

print(
    readiness_summary.to_string(
        index=False,
        formatters={
            "pct":
                lambda x:
                f"{x:.2f}%"
        }
    )
)


print()
print("=" * 118)
print("PROVISIONAL QUALITY DISTRIBUTION")
print("=" * 118)

print(
    quality_distribution.to_string(
        index=False,
        formatters={
            "pct":
                lambda x:
                f"{x:.2f}%"
        }
    )
)


print()
print("=" * 118)
print("IMPORTANT EVENT ARCHITECTURE AUDIT")
print("=" * 118)

print(
    important_architecture_audit.to_string(
        index=False
    )
)


print()
print("=" * 118)
print("NEW 11")
print("=" * 118)

print(
    new_architecture_audit.to_string(
        index=False
    )
)


print()
print("=" * 118)
print("UFC 329 vs LOW-BASELINE SEARCH SPIKES")
print("=" * 118)

print(
    problem_case_comparison.to_string(
        index=False
    )
)


print()
print("=" * 118)
print("TOP 25 PROVISIONAL FANFARE — DIAGNOSTIC ONLY")
print("=" * 118)

print(
    top_provisional.to_string(
        index=False
    )
)


# ============================================================
# PART 21 — PASS / FAIL
# ============================================================

step94_pass = (

    total_events == 788

    and

    np.isclose(
        fanfare_weight_total,
        1.0
    )

    and

    absolute_count == 0

    and

    commercial_count == 6

    and

    historical_count == 15

    and

    acceleration_count == 516

    and

    final_ready_count == 0

    and

    len(
        new_architecture_audit
    ) == 11
)


print()
print("=" * 118)

if step94_pass:

    print(
        "STEP 94 FANFARE ENGINE ARCHITECTURE: PASS ✅🔥"
    )

else:

    print(
        "STEP 94 FANFARE ENGINE ARCHITECTURE: CHECK REQUIRED ⚠️"
    )

print("=" * 118)
```

```python
# ============================================================
# STEP 95 — EVENT-TIME ABSOLUTE SEARCH: UFC ANCHOR RECOVERY
# ============================================================

from pytrends.request import TrendReq
import pandas as pd
import numpy as np
import time


print("=" * 115)
print("STEP 95 — EVENT-TIME ABSOLUTE SEARCH ANCHOR TEST 🔥")
print("=" * 115)


# ============================================================
# PART 1 — WHY WE ARE DOING THIS
# ============================================================

print()
print(
    "Goal:"
)

print(
    "Find Google's official UFC Topic so every event-time "
    "search request can share a common reference point."
)

print()

print(
    "This is what should eventually let us distinguish:"
)

print(
    "  • huge absolute search demand"
)

print(
    "from"
)

print(
    "  • a huge percentage spike from a tiny baseline"
)

print()

print(
    "ONLY ONE Google request will be attempted in this step."
)


# ============================================================
# PART 2 — CREATE FRESH PYTRENDS CONNECTION
# ============================================================

pytrends_anchor_test = TrendReq(
    hl="en-US",
    tz=300,
    timeout=(10, 30),
    retries=0,
    backoff_factor=0
)


# ============================================================
# PART 3 — ONE REQUEST ONLY
# ============================================================

ufc_anchor_request_success = False

ufc_anchor_rate_limited = False

ufc_anchor_suggestions = pd.DataFrame()


try:

    raw_suggestions = (
        pytrends_anchor_test.suggestions(
            keyword="UFC"
        )
    )


    ufc_anchor_suggestions = (
        pd.DataFrame(
            raw_suggestions
        )
    )


    ufc_anchor_request_success = True


except Exception as e:

    error_type = (
        type(e).__name__
    )

    error_message = str(e)


    print()
    print("=" * 115)
    print("GOOGLE RESPONSE")
    print("=" * 115)

    print(
        "Error type:",
        error_type
    )

    print(
        "Error message:",
        error_message
    )


    if (
        "429" in error_message
        or
        "TooManyRequests" in error_type
    ):

        ufc_anchor_rate_limited = True

        print()
        print(
            "Google is STILL rate-limiting us."
        )

        print(
            "DO NOT rerun this cell."
        )

        print(
            "We will continue the model without making "
            "more Google requests right now."
        )


# ============================================================
# PART 4 — IF GOOGLE RESPONDED, INSPECT SUGGESTIONS
# ============================================================

if ufc_anchor_request_success:

    print()
    print("=" * 115)
    print("GOOGLE UFC SUGGESTIONS")
    print("=" * 115)


    if len(
        ufc_anchor_suggestions
    ) == 0:

        print(
            "Google responded, but returned no suggestions."
        )


    else:

        print(
            ufc_anchor_suggestions.to_string(
                index=False
            )
        )


# ============================================================
# PART 5 — IDENTIFY LIKELY UFC CANDIDATES
#
# IMPORTANT:
#
# We do NOT automatically approve one.
#
# We only identify candidates whose returned title/type
# appears relevant.
# ============================================================

if (
    ufc_anchor_request_success
    and
    len(
        ufc_anchor_suggestions
    ) > 0
):

    candidate_mask = pd.Series(
        False,
        index=ufc_anchor_suggestions.index
    )


    if (
        "title"
        in ufc_anchor_suggestions.columns
    ):

        candidate_mask = (
            candidate_mask
            |
            ufc_anchor_suggestions[
                "title"
            ]
            .astype(str)
            .str.contains(
                (
                    "UFC|"
                    "Ultimate Fighting Championship"
                ),
                case=False,
                regex=True,
                na=False
            )
        )


    if (
        "type"
        in ufc_anchor_suggestions.columns
    ):

        candidate_mask = (
            candidate_mask
            |
            ufc_anchor_suggestions[
                "type"
            ]
            .astype(str)
            .str.contains(
                (
                    "sports|"
                    "organization|"
                    "league|"
                    "promotion"
                ),
                case=False,
                regex=True,
                na=False
            )
        )


    likely_ufc_candidates = (
        ufc_anchor_suggestions[
            candidate_mask
        ]
        .copy()
    )


    print()
    print("=" * 115)
    print("LIKELY UFC TOPIC CANDIDATES")
    print("=" * 115)


    if len(
        likely_ufc_candidates
    ):

        print(
            likely_ufc_candidates.to_string(
                index=False
            )
        )

    else:

        print(
            "No candidate was automatically identified."
        )


else:

    likely_ufc_candidates = (
        pd.DataFrame()
    )


# ============================================================
# PART 6 — ABSOLUTELY NO AUTOMATIC TOPIC SELECTION
#
# We need to inspect:
#
#   title
#   type
#   mid/topic ID
#
# before using the anchor.
# ============================================================

ufc_anchor_topic_verified = False

ufc_anchor_topic_id = None


# ============================================================
# PART 7 — STATUS
# ============================================================

print()
print("=" * 115)
print("STEP 95 STATUS")
print("=" * 115)


if ufc_anchor_rate_limited:

    print(
        "STATUS: RATE LIMITED ⚠️"
    )

    print(
        "No data changed."
    )

    print(
        "No additional Google requests should be made yet."
    )


elif (
    ufc_anchor_request_success
    and
    len(
        ufc_anchor_suggestions
    ) > 0
):

    print(
        "STATUS: GOOGLE RESPONDED ✅🔥"
    )

    print(
        "Suggestions returned:",
        len(
            ufc_anchor_suggestions
        )
    )

    print(
        "Likely UFC candidates:",
        len(
            likely_ufc_candidates
        )
    )

    print()
    print(
        "STOP HERE."
    )

    print(
        "Do NOT run an interest-over-time request yet."
    )

    print(
        "We will verify the correct UFC topic first."
    )


elif ufc_anchor_request_success:

    print(
        "STATUS: GOOGLE RESPONDED BUT NO SUGGESTIONS ⚠️"
    )


else:

    print(
        "STATUS: REQUEST FAILED ⚠️"
    )


print()
print("=" * 115)
print("STEP 95 COMPLETE")
print("=" * 115)
```

```python
# ============================================================
# STEP 96 — FANFARE FALLBACK + EVIDENCE QUALITY FRAMEWORK
# ============================================================

import pandas as pd
import numpy as np


print("=" * 120)
print("STEP 96 — FANFARE FALLBACK + EVIDENCE QUALITY FRAMEWORK 🔥")
print("=" * 120)


# ============================================================
# PART 1 — SAFETY COPY
# ============================================================

fanfare_master_pre_step96 = fanfare_master.copy()


# ============================================================
# PART 2 — DEFINE EVIDENCE TYPES
#
# We distinguish:
#
# 1. Absolute Search
#    Best missing piece.
#
# 2. Commercial Demand
#    Current verified gate.
#
# 3. Historical Drawing
#    Prior verified commercial performance.
#
# 4. Search Acceleration
#    Fight-week spike relative to baseline.
#
# These are NOT interchangeable.
# ============================================================

evidence_columns = {

    "absolute_search":
        "absolute_search_score",

    "commercial_demand":
        "commercial_demand_score",

    "historical_drawing":
        "historical_drawing_score",

    "search_acceleration":
        "search_acceleration_score"
}


# ============================================================
# PART 3 — EVIDENCE QUALITY POINTS
#
# These points measure CONFIDENCE / INFORMATION QUALITY.
#
# They do NOT directly increase Fanfare.
#
# This is extremely important.
#
# A card does not become more hyped merely because we have
# more data about it.
# ============================================================

EVIDENCE_QUALITY_POINTS = {

    "absolute_search":
        40,

    "commercial_demand":
        30,

    "historical_drawing":
        20,

    "search_acceleration":
        10
}


# ============================================================
# PART 4 — CALCULATE EVIDENCE QUALITY
# ============================================================

fanfare_master[
    "fanfare_evidence_quality"
] = 0.0


for evidence_name, score_column in (
    evidence_columns.items()
):

    available = (
        fanfare_master[
            score_column
        ].notna()
    )

    fanfare_master.loc[
        available,
        "fanfare_evidence_quality"
    ] += (
        EVIDENCE_QUALITY_POINTS[
            evidence_name
        ]
    )


# ============================================================
# PART 5 — EVIDENCE COUNT
# ============================================================

fanfare_master[
    "fanfare_evidence_count"
] = sum(

    fanfare_master[
        column
    ].notna().astype(int)

    for column
    in evidence_columns.values()
)


# ============================================================
# PART 6 — EVIDENCE LABEL
# ============================================================

def fanfare_evidence_label(row):

    absolute_search = pd.notna(
        row[
            "absolute_search_score"
        ]
    )

    commercial = pd.notna(
        row[
            "commercial_demand_score"
        ]
    )

    historical = pd.notna(
        row[
            "historical_drawing_score"
        ]
    )

    acceleration = pd.notna(
        row[
            "search_acceleration_score"
        ]
    )


    labels = []


    if absolute_search:
        labels.append(
            "Absolute Search"
        )


    if commercial:
        labels.append(
            "Commercial"
        )


    if historical:
        labels.append(
            "Historical Drawing"
        )


    if acceleration:
        labels.append(
            "Acceleration"
        )


    if not labels:

        return "No Evidence"


    return " + ".join(
        labels
    )


fanfare_master[
    "fanfare_evidence_label"
] = (
    fanfare_master.apply(
        fanfare_evidence_label,
        axis=1
    )
)


# ============================================================
# PART 7 — ERA CLASSIFICATION
#
# This helps us distinguish events where Trends evidence is
# structurally unavailable from events where it is merely
# missing.
# ============================================================

def classify_fanfare_era(date):

    date = pd.Timestamp(
        date
    )


    if date < pd.Timestamp(
        "2014-01-01"
    ):

        return "Pre-Trends Era"


    return "Trends Era"


fanfare_master[
    "fanfare_data_era"
] = (
    fanfare_master[
        "event_date"
    ].apply(
        classify_fanfare_era
    )
)


# ============================================================
# PART 8 — SEARCH STATUS
# ============================================================

def search_data_status(row):

    if pd.notna(
        row[
            "absolute_search_score"
        ]
    ):

        return (
            "Absolute Search Available"
        )


    if pd.notna(
        row[
            "search_acceleration_score"
        ]
    ):

        return (
            "Acceleration Available"
        )


    if (
        row[
            "fanfare_data_era"
        ]
        ==
        "Pre-Trends Era"
    ):

        return (
            "Structurally Unavailable"
        )


    return (
        "Missing Trends-Era Search"
    )


fanfare_master[
    "search_data_status"
] = (
    fanfare_master.apply(
        search_data_status,
        axis=1
    )
)


# ============================================================
# PART 9 — SAFE FALLBACK SCORE
#
# IMPORTANT:
#
# This is NOT final Fanfare.
#
# We calculate the best evidence-based estimate currently
# possible, but preserve its quality separately.
#
# We DO NOT:
#
# - turn missing signals into zero
# - give fake popularity to pre-2014 cards
# - treat evidence quality as hype
# - treat acceleration as absolute popularity
# ============================================================

def current_fanfare_estimate(row):

    values = []
    weights = []


    # Absolute Search
    if pd.notna(
        row[
            "absolute_search_score"
        ]
    ):

        values.append(
            row[
                "absolute_search_score"
            ]
        )

        weights.append(
            FANFARE_WEIGHTS[
                "absolute_search_score"
            ]
        )


    # Current Commercial Demand
    if pd.notna(
        row[
            "commercial_demand_score"
        ]
    ):

        values.append(
            row[
                "commercial_demand_score"
            ]
        )

        weights.append(
            FANFARE_WEIGHTS[
                "commercial_demand_score"
            ]
        )


    # Historical Drawing
    if pd.notna(
        row[
            "historical_drawing_score"
        ]
    ):

        values.append(
            row[
                "historical_drawing_score"
            ]
        )

        weights.append(
            FANFARE_WEIGHTS[
                "historical_drawing_score"
            ]
        )


    # Search Acceleration
    if pd.notna(
        row[
            "search_acceleration_score"
        ]
    ):

        values.append(
            row[
                "search_acceleration_score"
            ]
        )

        weights.append(
            FANFARE_WEIGHTS[
                "search_acceleration_score"
            ]
        )


    if not values:

        return np.nan


    weights = np.array(
        weights,
        dtype=float
    )


    values = np.array(
        values,
        dtype=float
    )


    return np.average(
        values,
        weights=weights
    )


fanfare_master[
    "current_fanfare_estimate"
] = (
    fanfare_master.apply(
        current_fanfare_estimate,
        axis=1
    )
)


# ============================================================
# PART 10 — SAFE-USAGE FLAG
#
# We DO NOT consider acceleration-only events safe for final
# Fanfare interpretation.
#
# At least one stronger signal must exist.
# ============================================================

fanfare_master[
    "stronger_fanfare_signal_available"
] = (

    fanfare_master[
        "absolute_search_score"
    ].notna()

    |

    fanfare_master[
        "commercial_demand_score"
    ].notna()

    |

    fanfare_master[
        "historical_drawing_score"
    ].notna()
)


fanfare_master[
    "safe_for_final_fanfare"
] = (

    fanfare_master[
        "current_fanfare_estimate"
    ].notna()

    &

    fanfare_master[
        "stronger_fanfare_signal_available"
    ]
)


# ============================================================
# PART 11 — FANFARE COVERAGE CLASS
# ============================================================

def coverage_class(row):

    quality = (
        row[
            "fanfare_evidence_quality"
        ]
    )


    if quality == 0:

        return "None"


    if quality <= 10:

        return "Very Low"


    if quality <= 30:

        return "Low"


    if quality <= 60:

        return "Moderate"


    if quality < 100:

        return "High"


    return "Complete"


fanfare_master[
    "fanfare_coverage_class"
] = (
    fanfare_master.apply(
        coverage_class,
        axis=1
    )
)


# ============================================================
# PART 12 — DISTRIBUTIONS
# ============================================================

era_summary = (
    fanfare_master
    .groupby(
        "fanfare_data_era",
        dropna=False
    )
    .agg(

        events=(
            "event_name",
            "size"
        ),

        any_fanfare_evidence=(
            "current_fanfare_estimate",
            lambda x:
            x.notna().sum()
        ),

        safe_for_final=(
            "safe_for_final_fanfare",
            "sum"
        )
    )
    .reset_index()
)


search_status_summary = (
    fanfare_master[
        "search_data_status"
    ]
    .value_counts()
    .rename_axis(
        "search_status"
    )
    .reset_index(
        name="events"
    )
)


coverage_summary = (
    fanfare_master[
        "fanfare_coverage_class"
    ]
    .value_counts()
    .rename_axis(
        "coverage_class"
    )
    .reset_index(
        name="events"
    )
)


evidence_label_summary = (
    fanfare_master[
        "fanfare_evidence_label"
    ]
    .value_counts()
    .rename_axis(
        "evidence"
    )
    .reset_index(
        name="events"
    )
)


# ============================================================
# PART 13 — IMPORTANT CARD AUDIT
# ============================================================

important_pattern = (
    "UFC 189|"
    "UFC 194|"
    "UFC 196|"
    "UFC 202|"
    "UFC 205|"
    "UFC 229|"
    "UFC 246|"
    "UFC 257|"
    "UFC 264|"
    "UFC 300|"
    "UFC 329|"
    "UFC 330"
)


important_step96 = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        .str.contains(
            important_pattern,
            regex=True,
            na=False
        )
    ]
    [
        [
            "event_date",
            "event_name",

            "search_acceleration_score",
            "commercial_demand_score",
            "historical_drawing_score",

            "current_fanfare_estimate",

            "fanfare_evidence_quality",
            "fanfare_evidence_count",
            "fanfare_evidence_label",
            "fanfare_coverage_class",

            "safe_for_final_fanfare"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 14 — NEW 11 AUDIT
# ============================================================

new_step96 = (
    fanfare_master[
        fanfare_master[
            "event_date"
        ]
        > pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",

            "search_acceleration_score",
            "commercial_demand_score",
            "historical_drawing_score",

            "current_fanfare_estimate",

            "fanfare_evidence_quality",
            "fanfare_coverage_class",

            "safe_for_final_fanfare"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 15 — SPECIFIC UFC 329 CHECK
# ============================================================

ufc329 = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        ==
        "UFC 329: McGregor vs. Holloway 2"
    ]
)


if len(
    ufc329
) == 1:

    ufc329 = (
        ufc329.iloc[0]
    )

    ufc329_quality = (
        ufc329[
            "fanfare_evidence_quality"
        ]
    )

    ufc329_safe = bool(
        ufc329[
            "safe_for_final_fanfare"
        ]
    )

else:

    ufc329_quality = np.nan
    ufc329_safe = False


# ============================================================
# PART 16 — VALIDATION
# ============================================================

total_events = len(
    fanfare_master
)


unique_events = (
    fanfare_master[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


acceleration_only_count = int(
    (
        fanfare_master[
            "fanfare_evidence_label"
        ]
        ==
        "Acceleration"
    )
    .sum()
)


no_evidence_count = int(
    (
        fanfare_master[
            "fanfare_evidence_label"
        ]
        ==
        "No Evidence"
    )
    .sum()
)


safe_count = int(
    fanfare_master[
        "safe_for_final_fanfare"
    ].sum()
)


# ============================================================
# PART 17 — OUTPUT
# ============================================================

print()
print("=" * 120)
print("ERA SUMMARY")
print("=" * 120)

print(
    era_summary.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("SEARCH STATUS")
print("=" * 120)

print(
    search_status_summary.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("COVERAGE QUALITY")
print("=" * 120)

print(
    coverage_summary.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("EVIDENCE COMBINATIONS")
print("=" * 120)

print(
    evidence_label_summary.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("IMPORTANT CARD AUDIT")
print("=" * 120)

print(
    important_step96.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("NEW 11")
print("=" * 120)

print(
    new_step96.to_string(
        index=False
    )
)


# ============================================================
# PART 18 — PASS / FAIL
# ============================================================

step96_pass = (

    total_events == 788

    and

    unique_events == 788

    and

    acceleration_only_count == 498

    and

    no_evidence_count == 272

    and

    np.isclose(
        ufc329_quality,
        60
    )

    and

    ufc329_safe

    and

    fanfare_master[
        "fanfare_evidence_quality"
    ].between(
        0,
        100
    ).all()
)


print()
print("=" * 120)

print(
    "Total events:",
    total_events
)

print(
    "Acceleration-only:",
    acceleration_only_count
)

print(
    "No evidence:",
    no_evidence_count
)

print(
    "Currently safe for stronger Fanfare interpretation:",
    safe_count
)

print(
    "UFC 329 evidence quality:",
    ufc329_quality
)


print()

if step96_pass:

    print(
        "STEP 96 FANFARE FALLBACK FRAMEWORK: PASS ✅🔥"
    )

else:

    print(
        "STEP 96 FANFARE FALLBACK FRAMEWORK: CHECK REQUIRED ⚠️"
    )

print("=" * 120)
```

```python
# ============================================================
# STEP 97 — PPV / AUDIENCE / COMMERCIAL DATA INFRASTRUCTURE
# ============================================================

import pandas as pd
import numpy as np


print("=" * 120)
print("STEP 97 — PPV / AUDIENCE INFRASTRUCTURE 🔥")
print("=" * 120)


# ============================================================
# PART 1 — SAFETY COPIES
# ============================================================

fanfare_master_pre_step97 = fanfare_master.copy()


# ============================================================
# PART 2 — CREATE AUTHORITATIVE AUDIENCE DATABASE SHELL
#
# IMPORTANT:
#
# We DO NOT mix:
#
#   PPV buys
#   television viewers
#   streaming viewers
#   global viewers
#
# into one raw number.
#
# Each observation keeps its original metric type.
# ============================================================

audience_database = (
    fanfare_master[
        [
            "event_date",
            "event_name"
        ]
    ]
    .copy()
)


# ============================================================
# PART 3 — DEFINE EVENT ERA
# ============================================================

def audience_era(date):

    date = pd.Timestamp(date)

    if date < pd.Timestamp(
        "2001-01-01"
    ):
        return "Early UFC"

    elif date < pd.Timestamp(
        "2014-01-01"
    ):
        return "Traditional PPV Era"

    elif date < pd.Timestamp(
        "2019-04-01"
    ):
        return "Late Traditional PPV Era"

    else:
        return "ESPN+ / Streaming Era"


audience_database[
    "audience_era"
] = (
    audience_database[
        "event_date"
    ].apply(
        audience_era
    )
)


# ============================================================
# PART 4 — RAW AUDIENCE FIELDS
#
# These remain NaN until VERIFIED data is added.
# ============================================================

audience_database[
    "ppv_buys"
] = np.nan


audience_database[
    "tv_viewers"
] = np.nan


audience_database[
    "streaming_viewers"
] = np.nan


audience_database[
    "global_viewers"
] = np.nan


# ============================================================
# PART 5 — METADATA / PROVENANCE
#
# Every audience observation must eventually record:
#
# metric type
# source
# source quality
# notes
#
# This prevents us from losing track of what a number means.
# ============================================================

audience_database[
    "audience_metric_type"
] = pd.NA


audience_database[
    "audience_source"
] = pd.NA


audience_database[
    "audience_source_quality"
] = pd.NA


audience_database[
    "audience_notes"
] = pd.NA


# ============================================================
# PART 6 — PPV ERA ELIGIBILITY
#
# This does NOT mean we possess PPV data.
#
# It simply tells us whether traditional PPV buys are a
# historically appropriate metric for the event.
# ============================================================

audience_database[
    "traditional_ppv_era"
] = (

    audience_database[
        "event_date"
    ]
    <
    pd.Timestamp(
        "2019-04-01"
    )
)


# ============================================================
# PART 7 — ESPN+ ERA
#
# Post-transition events are separated because their audience
# reporting is not consistently comparable to traditional
# cable/satellite PPV buy figures.
# ============================================================

audience_database[
    "espn_plus_era"
] = (

    audience_database[
        "event_date"
    ]
    >=
    pd.Timestamp(
        "2019-04-01"
    )
)


# ============================================================
# PART 8 — AVAILABILITY FLAGS
# ============================================================

audience_database[
    "ppv_available"
] = (
    audience_database[
        "ppv_buys"
    ].notna()
)


audience_database[
    "tv_available"
] = (
    audience_database[
        "tv_viewers"
    ].notna()
)


audience_database[
    "streaming_available"
] = (
    audience_database[
        "streaming_viewers"
    ].notna()
)


audience_database[
    "global_viewers_available"
] = (
    audience_database[
        "global_viewers"
    ].notna()
)


# ============================================================
# PART 9 — ANY AUDIENCE DATA
# ============================================================

audience_database[
    "audience_data_available"
] = (

    audience_database[
        [
            "ppv_available",
            "tv_available",
            "streaming_available",
            "global_viewers_available"
        ]
    ]
    .any(
        axis=1
    )
)


# ============================================================
# PART 10 — COMPARABILITY CLASS
#
# Right now everything is missing.
#
# Eventually:
#
# Traditional PPV buys:
#   comparable WITHIN traditional PPV context
#
# ESPN+ PPV estimates:
#   separate comparison class
#
# TV / streaming / global viewers:
#   separate classes
#
# We will NEVER rank 1M PPV buys directly against
# 1M television viewers as though they are equivalent.
# ============================================================

audience_database[
    "audience_comparison_class"
] = pd.NA


# ============================================================
# PART 11 — PRODUCTION AUDIENCE SCORE PLACEHOLDER
#
# This will eventually hold a normalized score AFTER
# metric-specific normalization.
#
# Missing remains NaN.
# ============================================================

audience_database[
    "audience_demand_score"
] = np.nan


# ============================================================
# PART 12 — HISTORICAL DRAWING PLACEHOLDER
#
# Later we can use prior verified PPV/audience performance
# to strengthen Historical Drawing Power.
#
# Again:
#
# STRICTLY PRIOR EVENTS ONLY.
# ============================================================

audience_database[
    "prior_audience_drawing_score"
] = np.nan


# ============================================================
# PART 13 — CHECK FOR FAKE ZEROS
# ============================================================

raw_audience_columns = [

    "ppv_buys",
    "tv_viewers",
    "streaming_viewers",
    "global_viewers"
]


fake_zero_counts = {

    column:
    int(
        audience_database[
            column
        ]
        .eq(0)
        .sum()
    )

    for column
    in raw_audience_columns
}


# ============================================================
# PART 14 — ERA SUMMARY
# ============================================================

era_summary = (
    audience_database
    .groupby(
        "audience_era",
        dropna=False
    )
    .agg(

        events=(
            "event_name",
            "size"
        ),

        audience_data_available=(
            "audience_data_available",
            "sum"
        )
    )
    .reset_index()
)


# ============================================================
# PART 15 — FANFARE MASTER PLACEHOLDERS
#
# Attach clean placeholders to fanfare_master.
#
# These columns will receive real values once we populate
# verified audience observations.
# ============================================================

for column in [

    "audience_demand_score",
    "prior_audience_drawing_score"

]:

    if column in fanfare_master.columns:

        fanfare_master = (
            fanfare_master.drop(
                columns=[
                    column
                ]
            )
        )


fanfare_master = (
    fanfare_master.merge(

        audience_database[
            [
                "event_date",
                "event_name",
                "audience_demand_score",
                "prior_audience_drawing_score"
            ]
        ],

        on=[
            "event_date",
            "event_name"
        ],

        how="left",

        validate="one_to_one"
    )
)


# ============================================================
# PART 16 — UPDATED FANFARE ROADMAP
#
# We are NOT changing our 12% final Fanfare weight.
#
# These are signal families INSIDE Fanfare.
# ============================================================

fanfare_signal_roadmap = pd.DataFrame({

    "signal_family": [

        "Absolute Search Demand",
        "Current Commercial Demand",
        "Historical Drawing Power",
        "Search Hype Acceleration",
        "Audience / PPV Demand"
    ],

    "current_status": [

        "Blocked by Google 429",
        "6 verified gates",
        "15 prior-gate histories",
        "516 events",
        "Infrastructure ready"
    ],

    "missing_means_zero": [

        False,
        False,
        False,
        False,
        False
    ],

    "production_ready": [

        False,
        True,
        True,
        True,
        False
    ]
})


# ============================================================
# PART 17 — VALIDATION
# ============================================================

total_events = len(
    audience_database
)


unique_events = (
    audience_database[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


initial_audience_observations = int(
    audience_database[
        "audience_data_available"
    ].sum()
)


all_fake_zeros = sum(
    fake_zero_counts.values()
)


fan_master_events = len(
    fanfare_master
)


# ============================================================
# PART 18 — IMPORTANT EVENTS CHECK
# ============================================================

important_names = [

    "UFC 100: Lesnar vs Mir 2",
    "UFC 189: Mendes vs McGregor",
    "UFC 194: Aldo vs McGregor",
    "UFC 196: McGregor vs Diaz",
    "UFC 202: Diaz vs. McGregor 2",
    "UFC 205: Alvarez vs McGregor",
    "UFC 229: Khabib vs. McGregor",
    "UFC 300: Pereira vs. Hill",
    "UFC 329: McGregor vs. Holloway 2"
]


important_audience_shell = (
    audience_database[
        audience_database[
            "event_name"
        ]
        .isin(
            important_names
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "audience_era",
            "traditional_ppv_era",
            "espn_plus_era",
            "ppv_buys",
            "audience_data_available"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 19 — OUTPUT
# ============================================================

print()
print("=" * 120)
print("AUDIENCE DATABASE")
print("=" * 120)

print(
    "Events:",
    total_events
)

print(
    "Unique events:",
    unique_events
)

print(
    "Initial audience observations:",
    initial_audience_observations
)


print()
print("=" * 120)
print("AUDIENCE ERA SUMMARY")
print("=" * 120)

print(
    era_summary.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("FAKE ZERO CHECK")
print("=" * 120)

for column, count in (
    fake_zero_counts.items()
):

    print(
        f"{column:<25}",
        count
    )


print()
print("=" * 120)
print("IMPORTANT EVENT SHELL")
print("=" * 120)

print(
    important_audience_shell.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("UPDATED FANFARE ROADMAP")
print("=" * 120)

print(
    fanfare_signal_roadmap.to_string(
        index=False
    )
)


# ============================================================
# PART 20 — PASS / FAIL
# ============================================================

step97_pass = (

    total_events == 788

    and

    unique_events == 788

    and

    fan_master_events == 788

    and

    initial_audience_observations == 0

    and

    all_fake_zeros == 0

    and

    fanfare_master[
        "audience_demand_score"
    ].isna().all()

    and

    fanfare_master[
        "prior_audience_drawing_score"
    ].isna().all()
)


print()
print("=" * 120)

if step97_pass:

    print(
        "STEP 97 PPV / AUDIENCE INFRASTRUCTURE: PASS ✅🔥"
    )

else:

    print(
        "STEP 97 PPV / AUDIENCE INFRASTRUCTURE: CHECK REQUIRED ⚠️"
    )

print("=" * 120)
```

```python
# ============================================================
# STEP 98 — POPULATE HIGH-CONFIDENCE REPORTED PPV DATA
# ============================================================

import pandas as pd
import numpy as np


print("=" * 120)
print("STEP 98 — VERIFIED / REPORTED PPV AUDIENCE DATA 🔥")
print("=" * 120)


# ============================================================
# PART 1 — SAFETY COPY
# ============================================================

audience_database_pre_step98 = audience_database.copy()
fanfare_master_pre_step98 = fanfare_master.copy()


# ============================================================
# PART 2 — HIGH-CONFIDENCE REPORTED PPV OBSERVATIONS
#
# IMPORTANT:
#
# These are reported industry PPV estimates, not necessarily
# official UFC disclosures.
#
# We are deliberately starting with a conservative set of
# major events rather than importing a low-quality bulk list.
#
# Post-2019 ESPN+ observations are kept in a SEPARATE
# comparison class from traditional PPV.
# ============================================================

reported_ppv_data = [

    # --------------------------------------------------------
    # TRADITIONAL PPV ERA
    # --------------------------------------------------------

    {
        "event_date": "2009-07-11",
        "event_search": "UFC 100",
        "ppv_buys": 1600000,
        "comparison_class": "Traditional PPV",
        "source_quality": "High",
        "notes": "Widely reported historical PPV estimate"
    },

    {
        "event_date": "2010-07-03",
        "event_search": "UFC 116",
        "ppv_buys": 1160000,
        "comparison_class": "Traditional PPV",
        "source_quality": "High",
        "notes": "Widely reported historical PPV estimate"
    },

    {
        "event_date": "2015-07-11",
        "event_search": "UFC 189",
        "ppv_buys": 825000,
        "comparison_class": "Traditional PPV",
        "source_quality": "High",
        "notes": "Reported historical PPV estimate"
    },

    {
        "event_date": "2015-11-15",
        "event_search": "UFC 193",
        "ppv_buys": 1100000,
        "comparison_class": "Traditional PPV",
        "source_quality": "High",
        "notes": "Widely reported historical PPV estimate"
    },

    {
        "event_date": "2015-12-12",
        "event_search": "UFC 194",
        "ppv_buys": 1200000,
        "comparison_class": "Traditional PPV",
        "source_quality": "High",
        "notes": "Widely reported historical PPV estimate"
    },

    {
        "event_date": "2016-03-05",
        "event_search": "UFC 196",
        "ppv_buys": 1317000,
        "comparison_class": "Traditional PPV",
        "source_quality": "High",
        "notes": "Reported historical PPV estimate"
    },

    {
        "event_date": "2016-07-09",
        "event_search": "UFC 200",
        "ppv_buys": 1200000,
        "comparison_class": "Traditional PPV",
        "source_quality": "High",
        "notes": "Widely reported historical PPV estimate"
    },

    {
        "event_date": "2016-08-20",
        "event_search": "UFC 202",
        "ppv_buys": 1650000,
        "comparison_class": "Traditional PPV",
        "source_quality": "High",
        "notes": "Widely reported historical PPV estimate"
    },

    {
        "event_date": "2016-11-12",
        "event_search": "UFC 205",
        "ppv_buys": 1300000,
        "comparison_class": "Traditional PPV",
        "source_quality": "High",
        "notes": "Widely reported historical PPV estimate"
    },

    {
        "event_date": "2016-12-30",
        "event_search": "UFC 207",
        "ppv_buys": 1100000,
        "comparison_class": "Traditional PPV",
        "source_quality": "High",
        "notes": "Widely reported historical PPV estimate"
    },

    {
        "event_date": "2018-10-06",
        "event_search": "UFC 229",
        "ppv_buys": 2400000,
        "comparison_class": "Traditional PPV",
        "source_quality": "High",
        "notes": "Widely reported historical PPV estimate"
    },

    # --------------------------------------------------------
    # ESPN+ ERA — SEPARATE COMPARISON CLASS
    # --------------------------------------------------------

    {
        "event_date": "2020-01-18",
        "event_search": "UFC 246",
        "ppv_buys": 1350000,
        "comparison_class": "ESPN+ Reported PPV",
        "source_quality": "Medium",
        "notes": "Reported ESPN+ era estimate; not directly pooled with traditional PPV"
    },

    {
        "event_date": "2020-07-11",
        "event_search": "UFC 251",
        "ppv_buys": 1300000,
        "comparison_class": "ESPN+ Reported PPV",
        "source_quality": "Medium",
        "notes": "Reported ESPN+ era estimate; not directly pooled with traditional PPV"
    },

    {
        "event_date": "2021-01-23",
        "event_search": "UFC 257",
        "ppv_buys": 1600000,
        "comparison_class": "ESPN+ Reported PPV",
        "source_quality": "Medium",
        "notes": "Reported ESPN+ era estimate; not directly pooled with traditional PPV"
    },

    {
        "event_date": "2021-07-10",
        "event_search": "UFC 264",
        "ppv_buys": 1800000,
        "comparison_class": "ESPN+ Reported PPV",
        "source_quality": "Medium",
        "notes": "Reported ESPN+ era estimate; not directly pooled with traditional PPV"
    }
]


reported_ppv = pd.DataFrame(
    reported_ppv_data
)


reported_ppv[
    "event_date"
] = pd.to_datetime(
    reported_ppv[
        "event_date"
    ]
)


# ============================================================
# PART 3 — MATCH OBSERVATIONS TO OUR AUTHORITATIVE EVENTS
#
# We match BOTH:
#
# date
# and event-name prefix
#
# so we don't silently attach data to the wrong event.
# ============================================================

match_results = []


for row in (
    reported_ppv.itertuples(
        index=False
    )
):

    candidates = (
        audience_database[
            audience_database[
                "event_date"
            ]
            ==
            row.event_date
        ]
    )


    candidates = (
        candidates[
            candidates[
                "event_name"
            ]
            .str.contains(
                row.event_search,
                case=False,
                regex=False,
                na=False
            )
        ]
    )


    if len(candidates) == 1:

        idx = candidates.index[0]


        audience_database.loc[
            idx,
            "ppv_buys"
        ] = row.ppv_buys


        audience_database.loc[
            idx,
            "audience_metric_type"
        ] = "PPV Buys"


        audience_database.loc[
            idx,
            "audience_source"
        ] = (
            "Reported historical PPV sources"
        )


        audience_database.loc[
            idx,
            "audience_source_quality"
        ] = row.source_quality


        audience_database.loc[
            idx,
            "audience_notes"
        ] = row.notes


        audience_database.loc[
            idx,
            "audience_comparison_class"
        ] = row.comparison_class


        match_results.append({

            "event_date":
                row.event_date,

            "event_search":
                row.event_search,

            "matched_event":
                audience_database.loc[
                    idx,
                    "event_name"
                ],

            "ppv_buys":
                row.ppv_buys,

            "status":
                "MATCHED"
        })


    else:

        match_results.append({

            "event_date":
                row.event_date,

            "event_search":
                row.event_search,

            "matched_event":
                None,

            "ppv_buys":
                row.ppv_buys,

            "status":
                f"MATCH COUNT = {len(candidates)}"
        })


ppv_match_audit = pd.DataFrame(
    match_results
)


# ============================================================
# PART 4 — REFRESH AVAILABILITY FLAGS
# ============================================================

audience_database[
    "ppv_available"
] = (
    audience_database[
        "ppv_buys"
    ].notna()
)


audience_database[
    "tv_available"
] = (
    audience_database[
        "tv_viewers"
    ].notna()
)


audience_database[
    "streaming_available"
] = (
    audience_database[
        "streaming_viewers"
    ].notna()
)


audience_database[
    "global_viewers_available"
] = (
    audience_database[
        "global_viewers"
    ].notna()
)


audience_database[
    "audience_data_available"
] = (

    audience_database[
        [
            "ppv_available",
            "tv_available",
            "streaming_available",
            "global_viewers_available"
        ]
    ]
    .any(
        axis=1
    )
)


# ============================================================
# PART 5 — NORMALIZE PPV WITHIN COMPARISON CLASS
#
# CRITICAL:
#
# Traditional PPV and ESPN+ reported PPV are NOT pooled.
#
# Each class receives its own percentile.
# ============================================================

audience_database[
    "ppv_log"
] = np.where(

    audience_database[
        "ppv_buys"
    ].notna(),

    np.log1p(
        audience_database[
            "ppv_buys"
        ]
    ),

    np.nan
)


audience_database[
    "ppv_class_percentile"
] = np.nan


for comparison_class in (

    audience_database[
        "audience_comparison_class"
    ]
    .dropna()
    .unique()

):

    mask = (

        audience_database[
            "audience_comparison_class"
        ]
        ==
        comparison_class
    )


    valid = (

        mask

        &

        audience_database[
            "ppv_log"
        ].notna()
    )


    if valid.sum() > 0:

        audience_database.loc[
            valid,
            "ppv_class_percentile"
        ] = (

            audience_database.loc[
                valid,
                "ppv_log"
            ]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )


# ============================================================
# PART 6 — CURRENT AUDIENCE DEMAND SCORE
#
# For now this is simply the within-class PPV percentile.
#
# Later, if TV / streaming / global audience data is added,
# those metric classes can be normalized separately.
# ============================================================

audience_database[
    "audience_demand_score"
] = (
    audience_database[
        "ppv_class_percentile"
    ]
)


# ============================================================
# PART 7 — BUILD PRIOR AUDIENCE DRAWING POWER
#
# STRICT NO-LEAKAGE RULE:
#
# The current event's PPV cannot affect its own historical
# drawing score.
#
# We record prior information FIRST.
# Then update headliner history AFTER the event.
#
# We use verified main-event headliners from our fight table.
# ============================================================

verified_main_events = (
    ufc_fights_updated[
        ufc_fights_updated[
            "is_main_event_verified"
        ]
        ==
        True
    ]
    [
        [
            "date",
            "event_name",
            "f1_name",
            "f2_name"
        ]
    ]
    .copy()
)


verified_main_events = (
    verified_main_events
    .rename(
        columns={
            "date":
                "event_date",
            "f1_name":
                "headliner_1",
            "f2_name":
                "headliner_2"
        }
    )
)


verified_main_events[
    "event_date"
] = pd.to_datetime(
    verified_main_events[
        "event_date"
    ]
)


verified_main_events = (
    verified_main_events
    .drop_duplicates(
        [
            "event_date",
            "event_name"
        ]
    )
    .sort_values(
        [
            "event_date",
            "event_name"
        ],
        kind="stable"
    )
    .reset_index(
        drop=True
    )
)


# Map current audience score onto event.
audience_score_lookup = {

    (
        row.event_date,
        row.event_name
    ):
    row.audience_demand_score

    for row in (
        audience_database[
            [
                "event_date",
                "event_name",
                "audience_demand_score"
            ]
        ]
        .itertuples(
            index=False
        )
    )
}


fighter_audience_history = {}

prior_audience_rows = []


for row in (
    verified_main_events.itertuples(
        index=False
    )
):

    fighters = [
        row.headliner_1,
        row.headliner_2
    ]


    prior_scores = []


    for fighter in fighters:

        fighter_history = (
            fighter_audience_history.get(
                fighter,
                []
            )
        )


        prior_scores.extend(
            fighter_history
        )


    if len(prior_scores) > 0:

        prior_best = max(
            prior_scores
        )

        prior_average = float(
            np.mean(
                prior_scores
            )
        )

        prior_count = len(
            prior_scores
        )

    else:

        prior_best = np.nan
        prior_average = np.nan
        prior_count = 0


    prior_audience_rows.append({

        "event_date":
            row.event_date,

        "event_name":
            row.event_name,

        "prior_audience_events":
            prior_count,

        "prior_best_audience_score":
            prior_best,

        "prior_avg_audience_score":
            prior_average
    })


    # --------------------------------------------------------
    # ONLY NOW add current event to fighter histories.
    # --------------------------------------------------------

    current_score = (
        audience_score_lookup.get(
            (
                row.event_date,
                row.event_name
            ),
            np.nan
        )
    )


    if pd.notna(
        current_score
    ):

        for fighter in fighters:

            fighter_audience_history.setdefault(
                fighter,
                []
            )

            fighter_audience_history[
                fighter
            ].append(
                current_score
            )


prior_audience_history = pd.DataFrame(
    prior_audience_rows
)


# ============================================================
# PART 8 — MERGE PRIOR DRAWING BACK INTO AUDIENCE DATABASE
# ============================================================

audience_database = (
    audience_database
    .drop(
        columns=[
            "prior_audience_drawing_score"
        ],
        errors="ignore"
    )
    .merge(

        prior_audience_history,

        on=[
            "event_date",
            "event_name"
        ],

        how="left",

        validate="one_to_one"
    )
)


audience_database[
    "prior_audience_drawing_score"
] = (
    audience_database[
        "prior_best_audience_score"
    ]
)


# ============================================================
# PART 9 — UPDATE FANFARE MASTER
# ============================================================

fanfare_master = (
    fanfare_master
    .drop(
        columns=[
            "audience_demand_score",
            "prior_audience_drawing_score"
        ],
        errors="ignore"
    )
    .merge(

        audience_database[
            [
                "event_date",
                "event_name",
                "audience_demand_score",
                "prior_audience_drawing_score",
                "ppv_buys",
                "audience_comparison_class",
                "audience_source_quality"
            ]
        ],

        on=[
            "event_date",
            "event_name"
        ],

        how="left",

        validate="one_to_one"
    )
)


# ============================================================
# PART 10 — AUDIENCE COVERAGE
# ============================================================

ppv_count = int(
    audience_database[
        "ppv_available"
    ].sum()
)


traditional_count = int(
    (
        audience_database[
            "audience_comparison_class"
        ]
        ==
        "Traditional PPV"
    )
    .sum()
)


espn_count = int(
    (
        audience_database[
            "audience_comparison_class"
        ]
        ==
        "ESPN+ Reported PPV"
    )
    .sum()
)


prior_audience_count = int(
    audience_database[
        "prior_audience_drawing_score"
    ].notna().sum()
)


# ============================================================
# PART 11 — PPV TABLE
# ============================================================

ppv_audit = (
    audience_database[
        audience_database[
            "ppv_available"
        ]
    ]
    [
        [
            "event_date",
            "event_name",
            "ppv_buys",
            "audience_comparison_class",
            "audience_source_quality",
            "ppv_class_percentile"
        ]
    ]
    .sort_values(
        [
            "audience_comparison_class",
            "ppv_buys"
        ],
        ascending=[
            True,
            False
        ]
    )
)


# ============================================================
# PART 12 — IMPORTANT DRAWING-POWER CHECK
# ============================================================

important_pattern = (
    "UFC 189|"
    "UFC 194|"
    "UFC 196|"
    "UFC 202|"
    "UFC 205|"
    "UFC 229|"
    "UFC 246|"
    "UFC 257|"
    "UFC 264|"
    "UFC 329"
)


drawing_audit = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        .str.contains(
            important_pattern,
            regex=True,
            na=False
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "ppv_buys",
            "audience_demand_score",
            "prior_audience_drawing_score",
            "audience_comparison_class"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 13 — NO-LEAKAGE CHECK
#
# UFC 189 is McGregor's first PPV observation in this
# conservative seed set.
#
# Its own audience score must NOT appear as prior audience
# drawing for UFC 189 itself.
# ============================================================

ufc189 = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        ==
        "UFC 189: Mendes vs McGregor"
    ]
)


if len(
    ufc189
) == 1:

    ufc189_prior_ok = pd.isna(
        ufc189.iloc[0][
            "prior_audience_drawing_score"
        ]
    )

else:

    ufc189_prior_ok = False


# UFC 194 should inherit prior McGregor audience history
# from UFC 189.
ufc194 = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        ==
        "UFC 194: Aldo vs McGregor"
    ]
)


if len(
    ufc194
) == 1:

    ufc194_prior_exists = pd.notna(
        ufc194.iloc[0][
            "prior_audience_drawing_score"
        ]
    )

else:

    ufc194_prior_exists = False


# UFC 329 should have substantial prior audience history
# through McGregor.
ufc329 = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        ==
        "UFC 329: McGregor vs. Holloway 2"
    ]
)


if len(
    ufc329
) == 1:

    ufc329_prior_audience = (
        ufc329.iloc[0][
            "prior_audience_drawing_score"
        ]
    )

else:

    ufc329_prior_audience = np.nan


# ============================================================
# PART 14 — OUTPUT
# ============================================================

print()
print("=" * 120)
print("PPV MATCH AUDIT")
print("=" * 120)

print(
    ppv_match_audit.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("PPV COVERAGE")
print("=" * 120)

print(
    "PPV observations:",
    ppv_count
)

print(
    "Traditional PPV:",
    traditional_count
)

print(
    "ESPN+ reported PPV:",
    espn_count
)

print(
    "Events with prior audience drawing evidence:",
    prior_audience_count
)


print()
print("=" * 120)
print("NORMALIZED PPV OBSERVATIONS")
print("=" * 120)

print(
    ppv_audit.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("HISTORICAL AUDIENCE DRAWING AUDIT")
print("=" * 120)

print(
    drawing_audit.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("NO-LEAKAGE CHECK")
print("=" * 120)

print(
    "UFC 189 prior audience correctly missing:",
    ufc189_prior_ok
)

print(
    "UFC 194 has prior McGregor audience evidence:",
    ufc194_prior_exists
)

print(
    "UFC 329 prior audience drawing score:",
    ufc329_prior_audience
)


# ============================================================
# PART 15 — PASS / FAIL
# ============================================================

all_matched = (
    ppv_match_audit[
        "status"
    ]
    .eq(
        "MATCHED"
    )
    .all()
)


step98_pass = (

    len(
        audience_database
    )
    ==
    788

    and

    len(
        fanfare_master
    )
    ==
    788

    and

    ppv_count
    ==
    len(
        reported_ppv
    )

    and

    traditional_count
    ==
    11

    and

    espn_count
    ==
    4

    and

    all_matched

    and

    ufc189_prior_ok

    and

    ufc194_prior_exists

    and

    pd.notna(
        ufc329_prior_audience
    )
)


print()
print("=" * 120)

if step98_pass:

    print(
        "STEP 98 PPV / AUDIENCE POPULATION: PASS ✅🔥"
    )

else:

    print(
        "STEP 98 PPV / AUDIENCE POPULATION: CHECK REQUIRED ⚠️"
    )

print("=" * 120)
```

```python
# ============================================================
# STEP 99 — UPGRADED PRODUCTION FANFARE SCORE
# ============================================================

import pandas as pd
import numpy as np

print("=" * 120)
print("STEP 99 — UPGRADED FANFARE SCORE 🔥")
print("=" * 120)


# ============================================================
# PART 1 — SAFETY COPY
# ============================================================

fanfare_master_pre_step99 = fanfare_master.copy()


# ============================================================
# PART 2 — FINAL FANFARE SIGNAL WEIGHTS
#
# Absolute Search remains our best missing signal.
#
# Commercial/Audience combines CURRENT:
#   gate + PPV/audience
#
# Historical Drawing combines PRIOR:
#   gate history + audience history
#
# Search Acceleration remains useful but deliberately small.
# ============================================================

FANFARE_FINAL_WEIGHTS = {
    "absolute_search_score": 0.35,
    "commercial_audience_score": 0.30,
    "historical_drawing_power_score": 0.25,
    "search_acceleration_score": 0.10
}

assert np.isclose(
    sum(FANFARE_FINAL_WEIGHTS.values()),
    1.0
)


# ============================================================
# PART 3 — CURRENT COMMERCIAL / AUDIENCE DEMAND
#
# If both gate and audience exist:
# average them.
#
# If only one exists:
# use the one real signal.
#
# Missing NEVER becomes zero.
# ============================================================

current_demand_cols = [
    "commercial_demand_score",
    "audience_demand_score"
]


fanfare_master[
    "commercial_audience_score"
] = (
    fanfare_master[
        current_demand_cols
    ]
    .mean(
        axis=1,
        skipna=True
    )
)


fanfare_master.loc[
    fanfare_master[
        current_demand_cols
    ]
    .isna()
    .all(axis=1),

    "commercial_audience_score"
] = np.nan


# ============================================================
# PART 4 — HISTORICAL DRAWING POWER
#
# Combine:
#
# prior verified-gate drawing
# prior PPV/audience drawing
#
# Again, missing is not zero.
# ============================================================

historical_cols = [
    "historical_drawing_score",
    "prior_audience_drawing_score"
]


fanfare_master[
    "historical_drawing_power_score"
] = (
    fanfare_master[
        historical_cols
    ]
    .mean(
        axis=1,
        skipna=True
    )
)


fanfare_master.loc[
    fanfare_master[
        historical_cols
    ]
    .isna()
    .all(axis=1),

    "historical_drawing_power_score"
] = np.nan


# ============================================================
# PART 5 — ABSOLUTE SEARCH
#
# Still unavailable.
#
# We preserve the column so the architecture is ready when
# Google eventually becomes available.
# ============================================================

if "absolute_search_score" not in fanfare_master.columns:

    fanfare_master[
        "absolute_search_score"
    ] = np.nan


# ============================================================
# PART 6 — DYNAMIC FANFARE SCORING FUNCTION
# ============================================================

def calculate_fanfare_score(row):

    weighted_sum = 0.0
    available_weight = 0.0


    for column, weight in (
        FANFARE_FINAL_WEIGHTS.items()
    ):

        value = row[column]

        if pd.notna(value):

            weighted_sum += (
                value * weight
            )

            available_weight += weight


    if available_weight == 0:

        return np.nan


    return (
        weighted_sum
        /
        available_weight
    )


fanfare_master[
    "fanfare_score"
] = (
    fanfare_master.apply(
        calculate_fanfare_score,
        axis=1
    )
)


# ============================================================
# PART 7 — AVAILABLE WEIGHT
#
# This tells us HOW MUCH of the intended Fanfare architecture
# is actually observed for each event.
# ============================================================

def available_fanfare_weight(row):

    total = 0.0

    for column, weight in (
        FANFARE_FINAL_WEIGHTS.items()
    ):

        if pd.notna(
            row[column]
        ):

            total += weight

    return total


fanfare_master[
    "fanfare_available_weight"
] = (
    fanfare_master.apply(
        available_fanfare_weight,
        axis=1
    )
)


# ============================================================
# PART 8 — FANFARE CONFIDENCE
#
# This is metadata.
#
# It does NOT alter the score.
# ============================================================

def fanfare_confidence(row):

    weight = row[
        "fanfare_available_weight"
    ]

    if weight == 0:
        return "None"

    if (
        pd.notna(
            row[
                "absolute_search_score"
            ]
        )
    ):
        return "High"

    if (
        pd.notna(
            row[
                "commercial_audience_score"
            ]
        )
        and
        pd.notna(
            row[
                "historical_drawing_power_score"
            ]
        )
    ):
        return "Moderate"

    if (
        pd.notna(
            row[
                "commercial_audience_score"
            ]
        )
        or
        pd.notna(
            row[
                "historical_drawing_power_score"
            ]
        )
    ):
        return "Low"

    return "Very Low"


fanfare_master[
    "fanfare_confidence"
] = (
    fanfare_master.apply(
        fanfare_confidence,
        axis=1
    )
)


# ============================================================
# PART 9 — ACCELERATION-ONLY FLAG
# ============================================================

fanfare_master[
    "fanfare_acceleration_only"
] = (

    fanfare_master[
        "search_acceleration_score"
    ].notna()

    &

    fanfare_master[
        "absolute_search_score"
    ].isna()

    &

    fanfare_master[
        "commercial_audience_score"
    ].isna()

    &

    fanfare_master[
        "historical_drawing_power_score"
    ].isna()
)


# ============================================================
# PART 10 — NO-EVIDENCE FLAG
# ============================================================

fanfare_master[
    "fanfare_no_evidence"
] = (

    fanfare_master[
        [
            "absolute_search_score",
            "commercial_audience_score",
            "historical_drawing_power_score",
            "search_acceleration_score"
        ]
    ]
    .isna()
    .all(axis=1)
)


# ============================================================
# PART 11 — IMPORTANT EVENT AUDIT
# ============================================================

important_pattern = (
    "UFC 100|"
    "UFC 189|"
    "UFC 193|"
    "UFC 194|"
    "UFC 196|"
    "UFC 200|"
    "UFC 202|"
    "UFC 205|"
    "UFC 207|"
    "UFC 229|"
    "UFC 246|"
    "UFC 257|"
    "UFC 264|"
    "UFC 300|"
    "UFC 329|"
    "UFC 330"
)


important_fanfare = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        .str.contains(
            important_pattern,
            regex=True,
            na=False
        )
    ]
    [
        [
            "event_date",
            "event_name",

            "search_acceleration_score",

            "commercial_demand_score",
            "audience_demand_score",
            "commercial_audience_score",

            "historical_drawing_score",
            "prior_audience_drawing_score",
            "historical_drawing_power_score",

            "fanfare_score",
            "fanfare_available_weight",
            "fanfare_confidence"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 12 — NEW 11 AUDIT
# ============================================================

new_fanfare = (
    fanfare_master[
        fanfare_master[
            "event_date"
        ]
        >
        pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",

            "search_acceleration_score",
            "commercial_audience_score",
            "historical_drawing_power_score",

            "fanfare_score",
            "fanfare_available_weight",
            "fanfare_confidence",
            "fanfare_acceleration_only"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 13 — TOP FANFARE
#
# IMPORTANT:
#
# We print TWO lists:
#
# 1. all scored events
# 2. events with stronger evidence than acceleration alone
#
# This prevents acceleration-only cards from misleading us.
# ============================================================

top_all_fanfare = (
    fanfare_master[
        fanfare_master[
            "fanfare_score"
        ].notna()
    ]
    [
        [
            "event_date",
            "event_name",
            "fanfare_score",
            "fanfare_confidence",
            "fanfare_available_weight"
        ]
    ]
    .sort_values(
        "fanfare_score",
        ascending=False
    )
    .head(20)
)


top_supported_fanfare = (
    fanfare_master[
        (
            fanfare_master[
                "fanfare_score"
            ].notna()
        )
        &
        (
            ~fanfare_master[
                "fanfare_acceleration_only"
            ]
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "fanfare_score",
            "fanfare_confidence",
            "fanfare_available_weight"
        ]
    ]
    .sort_values(
        "fanfare_score",
        ascending=False
    )
    .head(20)
)


# ============================================================
# PART 14 — COVERAGE
# ============================================================

total_events = len(
    fanfare_master
)

scored_events = int(
    fanfare_master[
        "fanfare_score"
    ].notna().sum()
)

no_evidence_events = int(
    fanfare_master[
        "fanfare_no_evidence"
    ].sum()
)

acceleration_only_events = int(
    fanfare_master[
        "fanfare_acceleration_only"
    ].sum()
)

stronger_evidence_events = int(
    (
        fanfare_master[
            "fanfare_score"
        ].notna()
        &
        ~fanfare_master[
            "fanfare_acceleration_only"
        ]
    ).sum()
)


confidence_summary = (
    fanfare_master[
        "fanfare_confidence"
    ]
    .value_counts()
    .rename_axis(
        "confidence"
    )
    .reset_index(
        name="events"
    )
)


# ============================================================
# PART 15 — UFC 329 CHECK
# ============================================================

ufc329 = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        ==
        "UFC 329: McGregor vs. Holloway 2"
    ]
)


if len(ufc329) == 1:

    ufc329_score = float(
        ufc329.iloc[0][
            "fanfare_score"
        ]
    )

    ufc329_historical = float(
        ufc329.iloc[0][
            "historical_drawing_power_score"
        ]
    )

    ufc329_current = float(
        ufc329.iloc[0][
            "commercial_audience_score"
        ]
    )

else:

    ufc329_score = np.nan
    ufc329_historical = np.nan
    ufc329_current = np.nan


# ============================================================
# PART 16 — OUTPUT
# ============================================================

print()
print("=" * 120)
print("FANFARE ARCHITECTURE")
print("=" * 120)

for key, value in (
    FANFARE_FINAL_WEIGHTS.items()
):

    print(
        f"{key:<40}",
        f"{value * 100:.1f}%"
    )


print()
print("=" * 120)
print("COVERAGE")
print("=" * 120)

print(
    "Total events:",
    total_events
)

print(
    "Events with a Fanfare score:",
    scored_events
)

print(
    "No Fanfare evidence:",
    no_evidence_events
)

print(
    "Acceleration-only:",
    acceleration_only_events
)

print(
    "Events with stronger evidence:",
    stronger_evidence_events
)


print()
print(
    confidence_summary.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("IMPORTANT EVENT AUDIT")
print("=" * 120)

print(
    important_fanfare.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("NEW 11")
print("=" * 120)

print(
    new_fanfare.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("TOP 20 — ALL CURRENT FANFARE SCORES")
print("=" * 120)

print(
    top_all_fanfare.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("TOP 20 — STRONGER-EVIDENCE FANFARE")
print("=" * 120)

print(
    top_supported_fanfare.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("UFC 329 CHECK")
print("=" * 120)

print(
    "Current commercial/audience score:",
    ufc329_current
)

print(
    "Historical drawing power:",
    ufc329_historical
)

print(
    "Final current Fanfare score:",
    ufc329_score
)


# ============================================================
# PART 17 — VALIDATION
# ============================================================

step99_pass = (

    total_events == 788

    and

    scored_events + no_evidence_events
    ==
    788

    and

    fanfare_master[
        "fanfare_score"
    ]
    .dropna()
    .between(
        0,
        100
    )
    .all()

    and

    np.isclose(
        sum(
            FANFARE_FINAL_WEIGHTS.values()
        ),
        1.0
    )

    and

    pd.notna(
        ufc329_score
    )

    and

    pd.notna(
        ufc329_historical
    )

    and

    pd.notna(
        ufc329_current
    )
)


print()
print("=" * 120)

if step99_pass:

    print(
        "STEP 99 UPGRADED FANFARE ENGINE: PASS ✅🔥"
    )

else:

    print(
        "STEP 99 UPGRADED FANFARE ENGINE: CHECK REQUIRED ⚠️"
    )

print("=" * 120)
```

```python
# ============================================================
# STEP 100 — CORRECTED PRODUCTION FANFARE SCORING
# ============================================================

import pandas as pd
import numpy as np

print("=" * 120)
print("STEP 100 — CORRECTED MISSING-SIGNAL FANFARE SCORING 🔥")
print("=" * 120)


# ============================================================
# PART 1 — SAFETY COPY
# ============================================================

fanfare_master_pre_step100 = fanfare_master.copy()


# ============================================================
# PART 2 — LOCK FANFARE ARCHITECTURE
#
# Intended full-information weights:
#
# Absolute Search          35%
# Commercial / Audience    30%
# Historical Drawing      25%
# Search Acceleration      10%
#
# KEY RULE:
#
# Search Acceleration NEVER receives redistributed weight.
#
# It always contributes:
#
#     acceleration_score * 0.10
#
# The other 90% can be redistributed among available
# SUBSTANTIVE demand signals.
# ============================================================

SUBSTANTIVE_FANFARE_WEIGHTS = {
    "absolute_search_score": 0.35,
    "commercial_audience_score": 0.30,
    "historical_drawing_power_score": 0.25
}

ACCELERATION_WEIGHT = 0.10

assert np.isclose(
    sum(SUBSTANTIVE_FANFARE_WEIGHTS.values())
    + ACCELERATION_WEIGHT,
    1.0
)


# ============================================================
# PART 3 — CORRECTED SCORING FUNCTION
#
# CASE A:
# At least one substantive signal exists.
#
# The substantive signals share the available 90-point
# substantive portion according to their intended relative
# weights.
#
# Search acceleration independently contributes up to 10 pts.
#
#
# CASE B:
# Only acceleration exists.
#
# Acceleration contributes up to 10 points.
# Missing substantive evidence contributes NOTHING.
#
#
# CASE C:
# No evidence exists.
#
# Fanfare remains NaN.
# ============================================================

def corrected_fanfare_score(row):

    substantive_values = []
    substantive_weights = []


    for column, weight in (
        SUBSTANTIVE_FANFARE_WEIGHTS.items()
    ):

        value = row[column]

        if pd.notna(value):

            substantive_values.append(
                float(value)
            )

            substantive_weights.append(
                float(weight)
            )


    acceleration = row[
        "search_acceleration_score"
    ]


    # --------------------------------------------------------
    # NO EVIDENCE AT ALL
    # --------------------------------------------------------

    if (
        len(substantive_values) == 0
        and
        pd.isna(acceleration)
    ):

        return np.nan


    score = 0.0


    # --------------------------------------------------------
    # SUBSTANTIVE DEMAND PORTION
    #
    # Available substantive signals are normalized against
    # each other, then occupy the intended 90% substantive
    # portion of Fanfare.
    # --------------------------------------------------------

    if substantive_values:

        substantive_values = np.array(
            substantive_values,
            dtype=float
        )

        substantive_weights = np.array(
            substantive_weights,
            dtype=float
        )


        substantive_average = np.average(
            substantive_values,
            weights=substantive_weights
        )


        score += (
            substantive_average
            * 0.90
        )


    # --------------------------------------------------------
    # SEARCH ACCELERATION PORTION
    #
    # NEVER redistributed.
    # Maximum contribution = 10 points.
    # --------------------------------------------------------

    if pd.notna(acceleration):

        score += (
            float(acceleration)
            * ACCELERATION_WEIGHT
        )


    return score


fanfare_master[
    "fanfare_score_corrected"
] = fanfare_master.apply(
    corrected_fanfare_score,
    axis=1
)


# ============================================================
# PART 4 — BREAK OUT CONTRIBUTIONS FOR TRANSPARENCY
# ============================================================

def substantive_component(row):

    values = []
    weights = []


    for column, weight in (
        SUBSTANTIVE_FANFARE_WEIGHTS.items()
    ):

        if pd.notna(row[column]):

            values.append(
                float(row[column])
            )

            weights.append(
                float(weight)
            )


    if not values:

        return np.nan


    return (
        np.average(
            np.array(values),
            weights=np.array(weights)
        )
        * 0.90
    )


fanfare_master[
    "fanfare_substantive_contribution"
] = fanfare_master.apply(
    substantive_component,
    axis=1
)


fanfare_master[
    "fanfare_acceleration_contribution"
] = np.where(
    fanfare_master[
        "search_acceleration_score"
    ].notna(),

    fanfare_master[
        "search_acceleration_score"
    ]
    * ACCELERATION_WEIGHT,

    0.0
)


# ============================================================
# PART 5 — PRODUCTION SCORE
#
# Replace Step 99's provisional score with corrected score.
# ============================================================

fanfare_master[
    "fanfare_score_step99"
] = fanfare_master[
    "fanfare_score"
]


fanfare_master[
    "fanfare_score"
] = fanfare_master[
    "fanfare_score_corrected"
]


# ============================================================
# PART 6 — SCORE STATUS
# ============================================================

def corrected_status(row):

    substantive_available = any(
        pd.notna(
            row[column]
        )
        for column in
        SUBSTANTIVE_FANFARE_WEIGHTS
    )

    acceleration_available = pd.notna(
        row[
            "search_acceleration_score"
        ]
    )


    if (
        substantive_available
        and
        acceleration_available
    ):

        return "Substantive + Acceleration"


    if substantive_available:

        return "Substantive Only"


    if acceleration_available:

        return "Acceleration Only"


    return "No Evidence"


fanfare_master[
    "fanfare_score_status"
] = fanfare_master.apply(
    corrected_status,
    axis=1
)


# ============================================================
# PART 7 — ACCELERATION-ONLY VALIDATION
#
# Every acceleration-only card MUST now score <= 10.
# ============================================================

acceleration_only_mask = (
    fanfare_master[
        "fanfare_score_status"
    ]
    ==
    "Acceleration Only"
)


acceleration_only_max = (
    fanfare_master.loc[
        acceleration_only_mask,
        "fanfare_score"
    ].max()
)


# ============================================================
# PART 8 — IMPORTANT CARD AUDIT
# ============================================================

important_pattern = (
    "UFC 100|"
    "UFC 189|"
    "UFC 193|"
    "UFC 194|"
    "UFC 196|"
    "UFC 200|"
    "UFC 202|"
    "UFC 205|"
    "UFC 207|"
    "UFC 229|"
    "UFC 246|"
    "UFC 257|"
    "UFC 264|"
    "UFC 300|"
    "UFC 329|"
    "UFC 330"
)


important_step100 = (
    fanfare_master[
        fanfare_master[
            "event_name"
        ]
        .str.contains(
            important_pattern,
            regex=True,
            na=False
        )
    ]
    [
        [
            "event_date",
            "event_name",

            "commercial_audience_score",
            "historical_drawing_power_score",
            "search_acceleration_score",

            "fanfare_substantive_contribution",
            "fanfare_acceleration_contribution",

            "fanfare_score_step99",
            "fanfare_score",

            "fanfare_score_status",
            "fanfare_confidence"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 9 — NEW 11
# ============================================================

new_step100 = (
    fanfare_master[
        fanfare_master[
            "event_date"
        ]
        >
        pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_date",
            "event_name",

            "commercial_audience_score",
            "historical_drawing_power_score",
            "search_acceleration_score",

            "fanfare_score_step99",
            "fanfare_score",

            "fanfare_score_status"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 10 — TOP 25 CORRECTED FANFARE
# ============================================================

top_corrected = (
    fanfare_master[
        fanfare_master[
            "fanfare_score"
        ].notna()
    ]
    [
        [
            "event_date",
            "event_name",
            "fanfare_score",
            "fanfare_score_status",
            "fanfare_confidence"
        ]
    ]
    .sort_values(
        "fanfare_score",
        ascending=False
    )
    .head(25)
)


# ============================================================
# PART 11 — TOP 25 WITH SUBSTANTIVE EVIDENCE
# ============================================================

top_substantive = (
    fanfare_master[
        fanfare_master[
            "fanfare_score_status"
        ]
        .isin(
            [
                "Substantive + Acceleration",
                "Substantive Only"
            ]
        )
    ]
    [
        [
            "event_date",
            "event_name",
            "fanfare_score",
            "fanfare_score_status",
            "fanfare_confidence"
        ]
    ]
    .sort_values(
        "fanfare_score",
        ascending=False
    )
    .head(25)
)


# ============================================================
# PART 12 — STATUS COUNTS
# ============================================================

status_summary = (
    fanfare_master[
        "fanfare_score_status"
    ]
    .value_counts()
    .rename_axis(
        "status"
    )
    .reset_index(
        name="events"
    )
)


# ============================================================
# PART 13 — SCORE DISTRIBUTION
# ============================================================

scored = fanfare_master[
    "fanfare_score"
].dropna()


distribution = {
    "count": int(
        scored.count()
    ),

    "min": float(
        scored.min()
    ),

    "median": float(
        scored.median()
    ),

    "mean": float(
        scored.mean()
    ),

    "max": float(
        scored.max()
    )
}


# ============================================================
# PART 14 — SPECIFIC CHECKS
# ============================================================

def get_event_score(name):

    rows = fanfare_master[
        fanfare_master[
            "event_name"
        ]
        ==
        name
    ]

    if len(rows) == 1:

        return float(
            rows.iloc[0][
                "fanfare_score"
            ]
        )

    return np.nan


ufc229_score = get_event_score(
    "UFC 229: Khabib vs. McGregor"
)

ufc264_score = get_event_score(
    "UFC 264: Poirier vs. McGregor 3"
)

ufc329_score = get_event_score(
    "UFC 329: McGregor vs. Holloway 2"
)


# ============================================================
# PART 15 — OUTPUT
# ============================================================

print()
print("=" * 120)
print("CORRECTED SCORING RULE")
print("=" * 120)

print(
    "Substantive demand portion: 90%"
)

print(
    "Search acceleration portion: 10%"
)

print(
    "Acceleration-only maximum:",
    acceleration_only_max
)


print()
print("=" * 120)
print("STATUS SUMMARY")
print("=" * 120)

print(
    status_summary.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("SCORE DISTRIBUTION")
print("=" * 120)

for key, value in distribution.items():

    print(
        f"{key:<10}",
        value
    )


print()
print("=" * 120)
print("IMPORTANT CARD AUDIT")
print("=" * 120)

print(
    important_step100.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("NEW 11")
print("=" * 120)

print(
    new_step100.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("TOP 25 — CORRECTED FANFARE")
print("=" * 120)

print(
    top_corrected.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("TOP 25 — SUBSTANTIVE-EVIDENCE FANFARE")
print("=" * 120)

print(
    top_substantive.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("KEY CARD SCORES")
print("=" * 120)

print(
    "UFC 229:",
    ufc229_score
)

print(
    "UFC 264:",
    ufc264_score
)

print(
    "UFC 329:",
    ufc329_score
)


# ============================================================
# PART 16 — VALIDATION
# ============================================================

step100_pass = (

    len(
        fanfare_master
    )
    ==
    788

    and

    fanfare_master[
        "fanfare_score"
    ]
    .dropna()
    .between(
        0,
        100
    )
    .all()

    and

    acceleration_only_max
    <=
    10.000001

    and

    fanfare_master.loc[
        acceleration_only_mask,
        "fanfare_score"
    ]
    .between(
        0,
        10.000001
    )
    .all()

    and

    pd.notna(
        ufc229_score
    )

    and

    pd.notna(
        ufc264_score
    )

    and

    pd.notna(
        ufc329_score
    )
)


print()
print("=" * 120)

if step100_pass:

    print(
        "STEP 100 CORRECTED FANFARE SCORING: PASS ✅🔥"
    )

else:

    print(
        "STEP 100 CORRECTED FANFARE SCORING: CHECK REQUIRED ⚠️"
    )

print("=" * 120)
```

```python
# ============================================================
# STEP 101 — FINAL UFC EVENT STRENGTH INDEX
# ============================================================

import pandas as pd
import numpy as np

print("=" * 120)
print("STEP 101 — FINAL UFC EVENT STRENGTH INDEX 🔥🔥🔥")
print("=" * 120)


# ============================================================
# PART 1 — LOCK FINAL MODEL WEIGHTS
# ============================================================

FINAL_WEIGHTS = {
    "card_quality_score": 0.37,
    "star_experience_score": 0.23,
    "main_event_score": 0.28,
    "fanfare_score": 0.12
}

assert np.isclose(
    sum(FINAL_WEIGHTS.values()),
    1.0
)


# ============================================================
# PART 2 — START FROM AUTHORITATIVE NORMALIZED EVENT TABLE
# ============================================================

final_event_strength_updated = (
    normalized_updated.copy()
)


# ============================================================
# PART 3 — MERGE CORRECTED FANFARE
# ============================================================

fanfare_merge = (
    fanfare_master[
        [
            "event_date",
            "event_name",
            "fanfare_score",
            "fanfare_confidence",
            "fanfare_score_status",
            "fanfare_available_weight",
            "commercial_audience_score",
            "historical_drawing_power_score",
            "search_acceleration_score"
        ]
    ]
    .copy()
)


final_event_strength_updated = (
    final_event_strength_updated
    .drop(
        columns=[
            "fanfare_score",
            "fanfare_confidence",
            "fanfare_score_status",
            "fanfare_available_weight",
            "commercial_audience_score",
            "historical_drawing_power_score",
            "search_acceleration_score"
        ],
        errors="ignore"
    )
    .merge(
        fanfare_merge,
        on=[
            "event_date",
            "event_name"
        ],
        how="left",
        validate="one_to_one"
    )
)


# ============================================================
# PART 4 — DYNAMIC FINAL SCORING
#
# Missing does NOT mean zero.
#
# If Fanfare is unavailable:
# remaining component weights redistribute proportionally.
#
# If Main Event is unavailable:
# remaining components redistribute proportionally.
#
# This preserves all 788 events.
# ============================================================

def calculate_final_event_strength(row):

    weighted_sum = 0.0
    available_weight = 0.0


    for column, weight in (
        FINAL_WEIGHTS.items()
    ):

        value = row[column]

        if pd.notna(value):

            weighted_sum += (
                float(value)
                * weight
            )

            available_weight += weight


    if available_weight == 0:

        return np.nan


    return (
        weighted_sum
        /
        available_weight
    )


final_event_strength_updated[
    "event_strength_score"
] = (
    final_event_strength_updated.apply(
        calculate_final_event_strength,
        axis=1
    )
)


# ============================================================
# PART 5 — AVAILABLE FINAL MODEL WEIGHT
# ============================================================

def final_available_weight(row):

    total = 0.0

    for column, weight in (
        FINAL_WEIGHTS.items()
    ):

        if pd.notna(row[column]):

            total += weight

    return total


final_event_strength_updated[
    "final_model_available_weight"
] = (
    final_event_strength_updated.apply(
        final_available_weight,
        axis=1
    )
)


# ============================================================
# PART 6 — COMPONENT CONTRIBUTIONS
#
# Raw intended contributions BEFORE dynamic redistribution.
# Useful for transparency/debugging.
# ============================================================

for column, weight in (
    FINAL_WEIGHTS.items()
):

    contribution_name = (
        column.replace(
            "_score",
            "_weighted_contribution"
        )
    )

    final_event_strength_updated[
        contribution_name
    ] = (
        final_event_strength_updated[
            column
        ]
        * weight
    )


# ============================================================
# PART 7 — FINAL SCORE RANK
# ============================================================

final_event_strength_updated[
    "event_strength_rank"
] = (
    final_event_strength_updated[
        "event_strength_score"
    ]
    .rank(
        ascending=False,
        method="min"
    )
    .astype(
        "Int64"
    )
)


# ============================================================
# PART 8 — PERCENTILE
# ============================================================

final_event_strength_updated[
    "event_strength_percentile"
] = (
    final_event_strength_updated[
        "event_strength_score"
    ]
    .rank(
        pct=True,
        method="average"
    )
    * 100
)


# ============================================================
# PART 9 — DATA QUALITY LABEL
#
# This does NOT alter the score.
# ============================================================

def final_data_quality(row):

    available = row[
        "final_model_available_weight"
    ]

    fanfare_conf = row[
        "fanfare_confidence"
    ]


    if (
        np.isclose(
            available,
            1.0
        )
        and
        fanfare_conf
        in [
            "High",
            "Moderate"
        ]
    ):

        return "Strong"


    if (
        available
        >=
        0.88
    ):

        return "Standard"


    return "Limited"


final_event_strength_updated[
    "final_data_quality"
] = (
    final_event_strength_updated.apply(
        final_data_quality,
        axis=1
    )
)


# ============================================================
# PART 10 — VALIDATION
# ============================================================

total_events = len(
    final_event_strength_updated
)


unique_events = (
    final_event_strength_updated[
        [
            "event_date",
            "event_name"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


missing_final_scores = int(
    final_event_strength_updated[
        "event_strength_score"
    ]
    .isna()
    .sum()
)


score_min = float(
    final_event_strength_updated[
        "event_strength_score"
    ].min()
)


score_max = float(
    final_event_strength_updated[
        "event_strength_score"
    ].max()
)


fanfare_missing = int(
    final_event_strength_updated[
        "fanfare_score"
    ]
    .isna()
    .sum()
)


main_event_missing = int(
    final_event_strength_updated[
        "main_event_score"
    ]
    .isna()
    .sum()
)


# ============================================================
# PART 11 — TOP 30 ALL-TIME
# ============================================================

top_30 = (
    final_event_strength_updated[
        [
            "event_strength_rank",
            "event_date",
            "event_name",

            "event_strength_score",

            "card_quality_score",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",

            "fanfare_confidence",
            "final_data_quality"
        ]
    ]
    .sort_values(
        [
            "event_strength_score",
            "event_date"
        ],
        ascending=[
            False,
            True
        ]
    )
    .head(30)
)


# ============================================================
# PART 12 — NEW 11
# ============================================================

new_final = (
    final_event_strength_updated[
        final_event_strength_updated[
            "event_date"
        ]
        >
        pd.Timestamp(
            "2026-06-27"
        )
    ]
    [
        [
            "event_strength_rank",
            "event_date",
            "event_name",

            "event_strength_score",

            "card_quality_score",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",

            "final_data_quality"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 13 — FAMOUS CARD AUDIT
# ============================================================

famous_pattern = (
    "UFC 100|"
    "UFC 189|"
    "UFC 194|"
    "UFC 196|"
    "UFC 200|"
    "UFC 202|"
    "UFC 205|"
    "UFC 229|"
    "UFC 246|"
    "UFC 257|"
    "UFC 264|"
    "UFC 300|"
    "UFC 329|"
    "UFC 330"
)


famous_audit = (
    final_event_strength_updated[
        final_event_strength_updated[
            "event_name"
        ]
        .str.contains(
            famous_pattern,
            regex=True,
            na=False
        )
    ]
    [
        [
            "event_strength_rank",
            "event_date",
            "event_name",

            "event_strength_score",

            "card_quality_score",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",

            "fanfare_confidence",
            "final_data_quality"
        ]
    ]
    .sort_values(
        "event_date"
    )
)


# ============================================================
# PART 14 — DATA QUALITY SUMMARY
# ============================================================

quality_summary = (
    final_event_strength_updated[
        "final_data_quality"
    ]
    .value_counts()
    .rename_axis(
        "quality"
    )
    .reset_index(
        name="events"
    )
)


available_weight_summary = (
    final_event_strength_updated[
        "final_model_available_weight"
    ]
    .round(2)
    .value_counts()
    .sort_index(
        ascending=False
    )
    .rename_axis(
        "available_weight"
    )
    .reset_index(
        name="events"
    )
)


# ============================================================
# PART 15 — OUTPUT
# ============================================================

print()
print("=" * 120)
print("FINAL MODEL WEIGHTS")
print("=" * 120)

for key, value in (
    FINAL_WEIGHTS.items()
):

    print(
        f"{key:<35}",
        f"{value * 100:.1f}%"
    )


print()
print("=" * 120)
print("VALIDATION")
print("=" * 120)

print(
    "Total events:",
    total_events
)

print(
    "Unique events:",
    unique_events
)

print(
    "Missing final scores:",
    missing_final_scores
)

print(
    "Missing Fanfare:",
    fanfare_missing
)

print(
    "Missing Main Event:",
    main_event_missing
)

print(
    "Final score range:",
    score_min,
    "to",
    score_max
)


print()
print("=" * 120)
print("DATA QUALITY")
print("=" * 120)

print(
    quality_summary.to_string(
        index=False
    )
)


print()
print(
    available_weight_summary.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("TOP 30 UFC EVENTS — FINAL EVENT STRENGTH INDEX")
print("=" * 120)

print(
    top_30.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("NEW 11 — FINAL SCORES")
print("=" * 120)

print(
    new_final.to_string(
        index=False
    )
)


print()
print("=" * 120)
print("FAMOUS CARD AUDIT")
print("=" * 120)

print(
    famous_audit.to_string(
        index=False
    )
)


# ============================================================
# PART 16 — PASS / FAIL
# ============================================================

step101_pass = (

    total_events == 788

    and

    unique_events == 788

    and

    missing_final_scores == 0

    and

    fanfare_missing == 265

    and

    main_event_missing == 3

    and

    score_min >= 0

    and

    score_max <= 100

    and

    np.isclose(
        sum(
            FINAL_WEIGHTS.values()
        ),
        1.0
    )
)


print()
print("=" * 120)

if step101_pass:

    print(
        "STEP 101 FINAL EVENT STRENGTH INDEX: PASS ✅🔥🔥🔥"
    )

else:

    print(
        "STEP 101 FINAL EVENT STRENGTH INDEX: CHECK REQUIRED ⚠️"
    )

print("=" * 120)
```

```python
import pandas as pd
import numpy as np

# ============================================================
# NUNES vs FIGUEIREDO — PRE-FIGHT RESUME COMPARISON
# ============================================================

df = ufc_fights_updated.copy()
df["date"] = pd.to_datetime(df["date"])

def get_fighter_before_event(fighter, event_name):

    # Find the fighter's fight at the requested event
    event_fight = df[
        (df["event_name"].eq(event_name)) &
        (
            df["f1_name"].eq(fighter) |
            df["f2_name"].eq(fighter)
        )
    ].copy()

    if len(event_fight) == 0:
        print(f"❌ Could not find {fighter} at {event_name}")
        return None

    row = event_fight.iloc[0]

    # Determine whether fighter is f1 or f2
    side = "f1" if row["f1_name"] == fighter else "f2"

    rank = row.get(f"{side}_rank_before", np.nan)

    # Same ranking-point system used by Event Strength model
    def rank_points(rank):
        if pd.isna(rank):
            return 0
        if rank == 0:
            return 16
        if 1 <= rank <= 15:
            return 16 - rank
        return 0

    result = {
        "Fighter": fighter,
        "Event": event_name,
        "Date": row["date"],

        "UFC Fights": row.get(f"{side}_ufc_fights_before", np.nan),
        "UFC Wins": row.get(f"{side}_ufc_wins_before", np.nan),
        "UFC Losses": row.get(f"{side}_ufc_losses_before", np.nan),
        "UFC Win %": row.get(f"{side}_ufc_win_pct_before", np.nan),

        "Win Streak": row.get(f"{side}_win_streak_before", np.nan),
        "Last 5 Win %": row.get(f"{side}_last5_win_pct", np.nan),

        "UFC Finishes": row.get(f"{side}_ufc_finishes_before", np.nan),
        "Finish Rate": row.get(f"{side}_finish_rate_before", np.nan),

        "Previous Title Fights": row.get(
            f"{side}_previous_title_fights", np.nan
        ),

        "Previous Main Events": row.get(
            f"{side}_previous_main_events", np.nan
        ),

        "Previous Bonuses": row.get(
            f"{side}_previous_total_bonuses", np.nan
        ),

        "Rank": rank,
        "Ranking Points": rank_points(rank)
    }

    return result


# ============================================================
# COMPARISON
# ============================================================

nunes = get_fighter_before_event(
    "Amanda Nunes",
    "UFC 269: Oliveira vs. Poirier"
)

figueiredo = get_fighter_before_event(
    "Deiveson Figueiredo",
    "UFC 283: Teixeira vs. Hill"
)

comparison = pd.DataFrame([nunes, figueiredo])

# Put fighter names as columns so comparison is easy to read
comparison_display = comparison.set_index("Fighter").T

print("\n🔥 NUNES vs FIGUEIREDO — ENTERING EVENT COMPARISON\n")
display(comparison_display)

print("\nIMPORTANT:")
print("All statistics are BEFORE the event.")
print("No future accomplishments are included.")
print("This does NOT modify the Event Strength Index.")
```

```python
import pandas as pd
import numpy as np

# ============================================================
# NUNES vs FIGUEIREDO — QUICK HISTORICAL COMPARISON
# Rebuilds ONLY what we need from the original historical data
# Does NOT modify the finished Event Strength Index
# ============================================================

# Load original historical UFC fight data
fights = pd.read_csv(
    "https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2026/2026-07-07/ufc_fights.csv"
)

fights["date"] = pd.to_datetime(fights["date"])
fights = fights.sort_values("date").reset_index(drop=True)

# ------------------------------------------------------------
# Track historical stats ENTERING every fight
# ------------------------------------------------------------

records = {}
streaks = {}
recent = {}
wins = {}
finishes = {}
title_history = {}

rows = []

for _, fight in fights.iterrows():

    f1 = fight["f1_name"]
    f2 = fight["f2_name"]

    def snapshot(fighter):
        r = records.get(fighter, {"wins": 0, "losses": 0, "fights": 0})
        w = wins.get(fighter, 0)
        fin = finishes.get(fighter, 0)
        last5 = recent.get(fighter, [])[-5:]

        return {
            "ufc_fights": r["fights"],
            "ufc_wins": r["wins"],
            "ufc_losses": r["losses"],
            "ufc_win_pct": (
                r["wins"] / (r["wins"] + r["losses"])
                if (r["wins"] + r["losses"]) > 0 else 0
            ),
            "win_streak": streaks.get(fighter, 0),
            "last5_win_pct": (
                sum(last5) / len(last5)
                if len(last5) > 0 else 0
            ),
            "ufc_finishes": fin,
            "finish_rate": fin / w if w > 0 else 0,
            "previous_title_fights": title_history.get(fighter, 0)
        }

    f1_stats = snapshot(f1)
    f2_stats = snapshot(f2)

    rows.append({
        "date": fight["date"],
        "event_name": fight["event_name"],
        "f1_name": f1,
        "f2_name": f2,
        "f1_stats": f1_stats,
        "f2_stats": f2_stats
    })

    # Determine title fight using the SAME logic as the notebook
    wc = str(fight["weight_class"])

    is_title = (
        wc.startswith("UFC ")
        and "title bout" in wc.lower()
        and "tournament" not in wc.lower()
    )

    # Determine finish
    method = str(fight["method"]).upper()

    is_finish = (
        "KO" in method
        or "TKO" in method
        or "SUBMISSION" in method
    )

    # Update fight counts
    for fighter in [f1, f2]:
        records.setdefault(
            fighter,
            {"wins": 0, "losses": 0, "fights": 0}
        )
        records[fighter]["fights"] += 1

    # Update results
    if fight["f1_result"] == "W":

        records[f1]["wins"] += 1
        records[f2]["losses"] += 1

        streaks[f1] = streaks.get(f1, 0) + 1
        streaks[f2] = 0

        recent.setdefault(f1, []).append(1)
        recent.setdefault(f2, []).append(0)

        wins[f1] = wins.get(f1, 0) + 1

        if is_finish:
            finishes[f1] = finishes.get(f1, 0) + 1

    elif fight["f2_result"] == "W":

        records[f2]["wins"] += 1
        records[f1]["losses"] += 1

        streaks[f2] = streaks.get(f2, 0) + 1
        streaks[f1] = 0

        recent.setdefault(f2, []).append(1)
        recent.setdefault(f1, []).append(0)

        wins[f2] = wins.get(f2, 0) + 1

        if is_finish:
            finishes[f2] = finishes.get(f2, 0) + 1

    else:
        streaks[f1] = 0
        streaks[f2] = 0
        recent.setdefault(f1, []).append(0)
        recent.setdefault(f2, []).append(0)

    # Update title history AFTER fight
    if is_title:
        title_history[f1] = title_history.get(f1, 0) + 1
        title_history[f2] = title_history.get(f2, 0) + 1


# ============================================================
# FIND THE TWO EXACT FIGHTER VERSIONS
# ============================================================

def find_fighter_snapshot(fighter, event_search):

    for row in rows:

        if event_search.lower() in row["event_name"].lower():

            if row["f1_name"] == fighter:
                stats = row["f1_stats"]

            elif row["f2_name"] == fighter:
                stats = row["f2_stats"]

            else:
                continue

            return {
                "Fighter": fighter,
                "Event": row["event_name"],
                "Date": row["date"].date(),
                **stats
            }

    raise ValueError(
        f"Could not find {fighter} at event containing '{event_search}'"
    )


nunes = find_fighter_snapshot(
    "Amanda Nunes",
    "UFC 269"
)

figueiredo = find_fighter_snapshot(
    "Deiveson Figueiredo",
    "UFC 283"
)

comparison = pd.DataFrame([nunes, figueiredo])

comparison = comparison.rename(columns={
    "ufc_fights": "UFC Fights",
    "ufc_wins": "UFC Wins",
    "ufc_losses": "UFC Losses",
    "ufc_win_pct": "UFC Win %",
    "win_streak": "Win Streak",
    "last5_win_pct": "Last 5 Win %",
    "ufc_finishes": "UFC Finishes",
    "finish_rate": "Finish Rate",
    "previous_title_fights": "Previous Title Fights"
})

# Make percentages easier to read
comparison["UFC Win %"] = (
    comparison["UFC Win %"] * 100
).round(1)

comparison["Last 5 Win %"] = (
    comparison["Last 5 Win %"] * 100
).round(1)

comparison["Finish Rate"] = (
    comparison["Finish Rate"] * 100
).round(1)

print("=" * 90)
print("🔥 AMANDA NUNES vs DEIVESON FIGUEIREDO")
print("ALL STATS ARE ENTERING THE EVENT — NO FUTURE LEAKAGE")
print("=" * 90)

display(
    comparison.set_index("Fighter").T
)
```

```python
# ============================================================
# BIG CELL 1
# FIGHTER CARD CONTRIBUTION — BASE FRAMEWORK
# Amanda Nunes UFC 269 vs Deiveson Figueiredo UFC 283
#
# Uses ONLY pre-event information.
# Does NOT alter the finished Event Strength Index.
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. START WITH THE COMPARISON WE JUST SUCCESSFULLY CREATED
# ------------------------------------------------------------

fighter_compare = comparison.copy()

# Make sure numeric columns are numeric
numeric_cols = [
    "UFC Fights",
    "UFC Wins",
    "UFC Losses",
    "UFC Win %",
    "Win Streak",
    "Last 5 Win %",
    "UFC Finishes",
    "Finish Rate",
    "Previous Title Fights"
]

for col in numeric_cols:
    fighter_compare[col] = pd.to_numeric(
        fighter_compare[col],
        errors="coerce"
    )


# ------------------------------------------------------------
# 2. DISPLAY RAW RESUME ADVANTAGES
# ------------------------------------------------------------

nunes = fighter_compare[
    fighter_compare["Fighter"] == "Amanda Nunes"
].iloc[0]

fig = fighter_compare[
    fighter_compare["Fighter"] == "Deiveson Figueiredo"
].iloc[0]


metrics = [
    "UFC Fights",
    "UFC Wins",
    "UFC Win %",
    "Win Streak",
    "Last 5 Win %",
    "UFC Finishes",
    "Finish Rate",
    "Previous Title Fights"
]


comparison_rows = []

for metric in metrics:

    nunes_val = nunes[metric]
    fig_val = fig[metric]

    if nunes_val > fig_val:
        leader = "Amanda Nunes"
    elif fig_val > nunes_val:
        leader = "Deiveson Figueiredo"
    else:
        leader = "Tie"

    comparison_rows.append({
        "Metric": metric,
        "Amanda Nunes": nunes_val,
        "Deiveson Figueiredo": fig_val,
        "Leader": leader
    })


raw_advantage = pd.DataFrame(comparison_rows)


# ------------------------------------------------------------
# 3. GROUP VARIABLES INTO THE SAME CONCEPTS USED BY OUR
#    EVENT STRENGTH PROJECT
# ------------------------------------------------------------

# QUALITY / CURRENT FORM
quality_metrics = [
    "UFC Win %",
    "Win Streak",
    "Last 5 Win %",
    "Finish Rate"
]

# EXPERIENCE / ACCOMPLISHMENT
experience_metrics = [
    "UFC Fights",
    "UFC Wins",
    "UFC Finishes",
    "Previous Title Fights"
]


# ------------------------------------------------------------
# 4. SIMPLE TWO-FIGHTER RELATIVE DISPLAY
#
# IMPORTANT:
# This is NOT yet the final 0-100 contribution score.
# We are deliberately NOT pretending two-person min/max
# normalization is the same as historical UFC percentile
# normalization.
# ------------------------------------------------------------

def relative_metric(a, b):

    maximum = max(a, b)

    if maximum == 0:
        return 0, 0

    return (
        (a / maximum) * 100,
        (b / maximum) * 100
    )


relative_rows = []

for metric in metrics:

    n_score, f_score = relative_metric(
        float(nunes[metric]),
        float(fig[metric])
    )

    relative_rows.append({
        "Metric": metric,
        "Nunes Relative Score": round(n_score, 1),
        "Figueiredo Relative Score": round(f_score, 1)
    })


relative_table = pd.DataFrame(relative_rows)


# ------------------------------------------------------------
# 5. CATEGORY COMPARISON
# ------------------------------------------------------------

def category_relative_score(fighter_row, opponent_row, metric_list):

    scores = []

    for metric in metric_list:

        fighter_value = float(fighter_row[metric])
        opponent_value = float(opponent_row[metric])

        maximum = max(fighter_value, opponent_value)

        if maximum == 0:
            scores.append(0)
        else:
            scores.append(
                (fighter_value / maximum) * 100
            )

    return np.mean(scores)


nunes_quality = category_relative_score(
    nunes,
    fig,
    quality_metrics
)

fig_quality = category_relative_score(
    fig,
    nunes,
    quality_metrics
)


nunes_experience = category_relative_score(
    nunes,
    fig,
    experience_metrics
)

fig_experience = category_relative_score(
    fig,
    nunes,
    experience_metrics
)


category_table = pd.DataFrame({

    "Category": [
        "Current Quality / Form",
        "Experience / Accomplishment"
    ],

    "Amanda Nunes": [
        nunes_quality,
        nunes_experience
    ],

    "Deiveson Figueiredo": [
        fig_quality,
        fig_experience
    ]
})

category_table[
    ["Amanda Nunes", "Deiveson Figueiredo"]
] = category_table[
    ["Amanda Nunes", "Deiveson Figueiredo"]
].round(1)


# ------------------------------------------------------------
# 6. OUTPUT
# ------------------------------------------------------------

print("=" * 95)
print("🔥 FIGHTER CARD CONTRIBUTION — STAGE 1")
print("Amanda Nunes entering UFC 269")
print("vs.")
print("Deiveson Figueiredo entering UFC 283")
print("=" * 95)

print("\nRAW PRE-FIGHT RESUME:\n")

display(
    fighter_compare[
        [
            "Fighter",
            "Event",
            "UFC Fights",
            "UFC Wins",
            "UFC Win %",
            "Win Streak",
            "Last 5 Win %",
            "UFC Finishes",
            "Finish Rate",
            "Previous Title Fights"
        ]
    ].set_index("Fighter").T
)


print("\n" + "=" * 95)
print("WHO LEADS EACH METRIC?")
print("=" * 95 + "\n")

display(raw_advantage)


print("\n" + "=" * 95)
print("RELATIVE METRIC COMPARISON")
print("100 = stronger fighter of these two on that metric")
print("=" * 95 + "\n")

display(relative_table)


print("\n" + "=" * 95)
print("CATEGORY COMPARISON")
print("=" * 95 + "\n")

display(category_table)


# ------------------------------------------------------------
# 7. COUNT RAW METRIC WINS
# ------------------------------------------------------------

nunes_wins = (
    raw_advantage["Leader"] == "Amanda Nunes"
).sum()

fig_wins = (
    raw_advantage["Leader"] == "Deiveson Figueiredo"
).sum()

ties = (
    raw_advantage["Leader"] == "Tie"
).sum()


print("\nSUMMARY")
print("-" * 40)

print(
    f"Amanda Nunes leads: {nunes_wins} / {len(metrics)} metrics"
)

print(
    f"Deiveson Figueiredo leads: {fig_wins} / {len(metrics)} metrics"
)

print(
    f"Ties: {ties}"
)

print("\n⚠️ THIS IS NOT THE FINAL CONTRIBUTION SCORE YET.")

print(
    "Next step adds historical UFC population normalization "
    "and the remaining résumé variables."
)

print(
    "We will NOT invent missing main-event, bonus, or ranking "
    "values just to produce a score."
)
```

```python
# ============================================================
# BIG CELL 2
# HISTORICAL UFC FIGHTER CONTRIBUTION POPULATION
#
# Purpose:
# Score Nunes and Figueiredo against ALL historical fighter
# appearances rather than against each other.
#
# Uses ONLY information available entering each fight.
# Does NOT alter the finished Event Strength Index.
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. EXPAND THE HISTORICAL SNAPSHOTS CREATED IN OUR FIRST CELL
# ------------------------------------------------------------

fighter_appearances = []

for row in rows:

    # Fighter 1
    f1 = row["f1_stats"]

    fighter_appearances.append({
        "date": row["date"],
        "event_name": row["event_name"],
        "fighter": row["f1_name"],

        "ufc_fights": f1["ufc_fights"],
        "ufc_wins": f1["ufc_wins"],
        "ufc_losses": f1["ufc_losses"],
        "ufc_win_pct": f1["ufc_win_pct"],

        "win_streak": f1["win_streak"],
        "last5_win_pct": f1["last5_win_pct"],

        "ufc_finishes": f1["ufc_finishes"],
        "finish_rate": f1["finish_rate"],

        "previous_title_fights":
            f1["previous_title_fights"]
    })

    # Fighter 2
    f2 = row["f2_stats"]

    fighter_appearances.append({
        "date": row["date"],
        "event_name": row["event_name"],
        "fighter": row["f2_name"],

        "ufc_fights": f2["ufc_fights"],
        "ufc_wins": f2["ufc_wins"],
        "ufc_losses": f2["ufc_losses"],
        "ufc_win_pct": f2["ufc_win_pct"],

        "win_streak": f2["win_streak"],
        "last5_win_pct": f2["last5_win_pct"],

        "ufc_finishes": f2["ufc_finishes"],
        "finish_rate": f2["finish_rate"],

        "previous_title_fights":
            f2["previous_title_fights"]
    })


population = pd.DataFrame(fighter_appearances)

population["date"] = pd.to_datetime(population["date"])


# ------------------------------------------------------------
# 2. ZERO-PRESERVING PERCENTILE FUNCTION
#
# Same philosophy as the Event Strength Index:
# genuine zero stays zero;
# positive values are percentile-ranked.
# ------------------------------------------------------------

def zero_preserving_percentile(series):

    series = pd.to_numeric(
        series,
        errors="coerce"
    )

    result = pd.Series(
        index=series.index,
        dtype=float
    )

    result[series.isna()] = np.nan

    result[series == 0] = 0

    positive = series > 0

    if positive.sum() > 0:

        result.loc[positive] = (
            series.loc[positive]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )

    return result


# ------------------------------------------------------------
# 3. CREATE HISTORICAL PERCENTILE SCORES
# ------------------------------------------------------------

score_variables = [
    "ufc_fights",
    "ufc_wins",
    "ufc_win_pct",
    "win_streak",
    "last5_win_pct",
    "ufc_finishes",
    "finish_rate",
    "previous_title_fights"
]


for variable in score_variables:

    population[
        variable + "_score"
    ] = zero_preserving_percentile(
        population[variable]
    )


# ------------------------------------------------------------
# 4. CREATE TWO CONCEPTUAL COMPONENTS
#
# CURRENT QUALITY / FORM
# EXPERIENCE / ACCOMPLISHMENT
#
# Still deliberately transparent — no mystery weights.
# ------------------------------------------------------------

quality_score_cols = [
    "ufc_win_pct_score",
    "win_streak_score",
    "last5_win_pct_score",
    "finish_rate_score"
]

experience_score_cols = [
    "ufc_fights_score",
    "ufc_wins_score",
    "ufc_finishes_score",
    "previous_title_fights_score"
]


population[
    "quality_form_score"
] = population[
    quality_score_cols
].mean(
    axis=1,
    skipna=True
)


population[
    "experience_accomplishment_score"
] = population[
    experience_score_cols
].mean(
    axis=1,
    skipna=True
)


# ------------------------------------------------------------
# 5. BASE FIGHTER CONTRIBUTION SCORE
#
# IMPORTANT:
# This is the BASE score from the 8 variables currently
# reconstructed.
#
# We are NOT yet pretending that missing ranking,
# previous-main-event, or bonus history is included.
# ------------------------------------------------------------

population[
    "base_fighter_contribution_score"
] = population[
    [
        "quality_form_score",
        "experience_accomplishment_score"
    ]
].mean(
    axis=1
)


# ------------------------------------------------------------
# 6. LOCATE NUNES AT UFC 269
# ------------------------------------------------------------

nunes_row = population[
    (population["fighter"] == "Amanda Nunes")
    &
    (
        population["event_name"]
        .str.contains(
            "UFC 269",
            case=False,
            na=False
        )
    )
].copy()


# ------------------------------------------------------------
# 7. LOCATE FIGUEIREDO AT UFC 283
# ------------------------------------------------------------

fig_row = population[
    (
        population["fighter"]
        == "Deiveson Figueiredo"
    )
    &
    (
        population["event_name"]
        .str.contains(
            "UFC 283",
            case=False,
            na=False
        )
    )
].copy()


if len(nunes_row) != 1:
    raise ValueError(
        f"Expected 1 Nunes UFC 269 row, found {len(nunes_row)}"
    )

if len(fig_row) != 1:
    raise ValueError(
        f"Expected 1 Figueiredo UFC 283 row, found {len(fig_row)}"
    )


# ------------------------------------------------------------
# 8. BUILD FINAL DISPLAY TABLE
# ------------------------------------------------------------

selected = pd.concat(
    [nunes_row, fig_row],
    ignore_index=True
)


display_cols = [
    "fighter",
    "event_name",

    "ufc_fights_score",
    "ufc_wins_score",
    "ufc_win_pct_score",

    "win_streak_score",
    "last5_win_pct_score",

    "ufc_finishes_score",
    "finish_rate_score",

    "previous_title_fights_score",

    "quality_form_score",
    "experience_accomplishment_score",

    "base_fighter_contribution_score"
]


result = selected[
    display_cols
].copy()


score_cols = [
    col for col in result.columns
    if col.endswith("_score")
]


result[
    score_cols
] = result[
    score_cols
].round(2)


# ------------------------------------------------------------
# 9. HISTORICAL PERCENTILE OF THE FINAL BASE SCORE
#
# This tells us where each fighter appearance sits relative
# to ALL historical UFC fighter appearances.
# ------------------------------------------------------------

population[
    "contribution_percentile"
] = (
    population[
        "base_fighter_contribution_score"
    ]
    .rank(
        pct=True,
        method="average"
    )
    * 100
)


for idx in result.index:

    fighter = result.loc[idx, "fighter"]
    event = result.loc[idx, "event_name"]

    matched = population[
        (population["fighter"] == fighter)
        &
        (population["event_name"] == event)
    ]

    result.loc[
        idx,
        "contribution_percentile"
    ] = matched[
        "contribution_percentile"
    ].iloc[0]


result[
    "contribution_percentile"
] = result[
    "contribution_percentile"
].round(2)


# ------------------------------------------------------------
# 10. OUTPUT
# ------------------------------------------------------------

print("=" * 100)
print("🔥 HISTORICAL FIGHTER CARD CONTRIBUTION — BASE MODEL")
print("=" * 100)

print(
    f"\nHistorical fighter appearances scored: "
    f"{len(population):,}"
)

print(
    f"Unique fighters: "
    f"{population['fighter'].nunique():,}"
)

print(
    f"Date range: "
    f"{population['date'].min().date()} "
    f"to "
    f"{population['date'].max().date()}"
)


print("\n" + "=" * 100)
print("NUNES vs FIGUEIREDO")
print("=" * 100 + "\n")

display(
    result.set_index("fighter").T
)


# ------------------------------------------------------------
# 11. EASY-TO-READ SUMMARY
# ------------------------------------------------------------

n = result[
    result["fighter"] == "Amanda Nunes"
].iloc[0]

f = result[
    result["fighter"] == "Deiveson Figueiredo"
].iloc[0]


print("\n" + "=" * 100)
print("SUMMARY")
print("=" * 100)

print(
    f"\nAmanda Nunes — UFC 269"
)

print(
    f"Quality/Form: "
    f"{n['quality_form_score']:.2f}"
)

print(
    f"Experience/Accomplishment: "
    f"{n['experience_accomplishment_score']:.2f}"
)

print(
    f"BASE Fighter Contribution: "
    f"{n['base_fighter_contribution_score']:.2f}"
)

print(
    f"Historical percentile: "
    f"{n['contribution_percentile']:.2f}"
)


print(
    f"\nDeiveson Figueiredo — UFC 283"
)

print(
    f"Quality/Form: "
    f"{f['quality_form_score']:.2f}"
)

print(
    f"Experience/Accomplishment: "
    f"{f['experience_accomplishment_score']:.2f}"
)

print(
    f"BASE Fighter Contribution: "
    f"{f['base_fighter_contribution_score']:.2f}"
)

print(
    f"Historical percentile: "
    f"{f['contribution_percentile']:.2f}"
)


difference = (
    n["base_fighter_contribution_score"]
    -
    f["base_fighter_contribution_score"]
)


print(
    f"\nContribution-score difference: "
    f"{difference:+.2f} points in Nunes' favor"
)


print("\n⚠️ IMPORTANT")

print(
    "This is the BASE contribution score using the "
    "8 reconstructed pre-fight variables."
)

print(
    "It is now normalized against the historical UFC "
    "fighter-appearance population, NOT against each other."
)

print(
    "Ranking/champion status, previous main events, and "
    "previous bonuses are still reserved for the final extension."
)
```

```python
# ============================================================
# BIG CELL 2
# HISTORICAL UFC FIGHTER CONTRIBUTION POPULATION
#
# Purpose:
# Score Nunes and Figueiredo against ALL historical fighter
# appearances rather than against each other.
#
# Uses ONLY information available entering each fight.
# Does NOT alter the finished Event Strength Index.
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. EXPAND THE HISTORICAL SNAPSHOTS CREATED IN OUR FIRST CELL
# ------------------------------------------------------------

fighter_appearances = []

for row in rows:

    # Fighter 1
    f1 = row["f1_stats"]

    fighter_appearances.append({
        "date": row["date"],
        "event_name": row["event_name"],
        "fighter": row["f1_name"],

        "ufc_fights": f1["ufc_fights"],
        "ufc_wins": f1["ufc_wins"],
        "ufc_losses": f1["ufc_losses"],
        "ufc_win_pct": f1["ufc_win_pct"],

        "win_streak": f1["win_streak"],
        "last5_win_pct": f1["last5_win_pct"],

        "ufc_finishes": f1["ufc_finishes"],
        "finish_rate": f1["finish_rate"],

        "previous_title_fights":
            f1["previous_title_fights"]
    })

    # Fighter 2
    f2 = row["f2_stats"]

    fighter_appearances.append({
        "date": row["date"],
        "event_name": row["event_name"],
        "fighter": row["f2_name"],

        "ufc_fights": f2["ufc_fights"],
        "ufc_wins": f2["ufc_wins"],
        "ufc_losses": f2["ufc_losses"],
        "ufc_win_pct": f2["ufc_win_pct"],

        "win_streak": f2["win_streak"],
        "last5_win_pct": f2["last5_win_pct"],

        "ufc_finishes": f2["ufc_finishes"],
        "finish_rate": f2["finish_rate"],

        "previous_title_fights":
            f2["previous_title_fights"]
    })


population = pd.DataFrame(fighter_appearances)

population["date"] = pd.to_datetime(population["date"])


# ------------------------------------------------------------
# 2. ZERO-PRESERVING PERCENTILE FUNCTION
#
# Same philosophy as the Event Strength Index:
# genuine zero stays zero;
# positive values are percentile-ranked.
# ------------------------------------------------------------

def zero_preserving_percentile(series):

    series = pd.to_numeric(
        series,
        errors="coerce"
    )

    result = pd.Series(
        index=series.index,
        dtype=float
    )

    result[series.isna()] = np.nan

    result[series == 0] = 0

    positive = series > 0

    if positive.sum() > 0:

        result.loc[positive] = (
            series.loc[positive]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )

    return result


# ------------------------------------------------------------
# 3. CREATE HISTORICAL PERCENTILE SCORES
# ------------------------------------------------------------

score_variables = [
    "ufc_fights",
    "ufc_wins",
    "ufc_win_pct",
    "win_streak",
    "last5_win_pct",
    "ufc_finishes",
    "finish_rate",
    "previous_title_fights"
]


for variable in score_variables:

    population[
        variable + "_score"
    ] = zero_preserving_percentile(
        population[variable]
    )


# ------------------------------------------------------------
# 4. CREATE TWO CONCEPTUAL COMPONENTS
#
# CURRENT QUALITY / FORM
# EXPERIENCE / ACCOMPLISHMENT
#
# Still deliberately transparent — no mystery weights.
# ------------------------------------------------------------

quality_score_cols = [
    "ufc_win_pct_score",
    "win_streak_score",
    "last5_win_pct_score",
    "finish_rate_score"
]

experience_score_cols = [
    "ufc_fights_score",
    "ufc_wins_score",
    "ufc_finishes_score",
    "previous_title_fights_score"
]


population[
    "quality_form_score"
] = population[
    quality_score_cols
].mean(
    axis=1,
    skipna=True
)


population[
    "experience_accomplishment_score"
] = population[
    experience_score_cols
].mean(
    axis=1,
    skipna=True
)


# ------------------------------------------------------------
# 5. BASE FIGHTER CONTRIBUTION SCORE
#
# IMPORTANT:
# This is the BASE score from the 8 variables currently
# reconstructed.
#
# We are NOT yet pretending that missing ranking,
# previous-main-event, or bonus history is included.
# ------------------------------------------------------------

population[
    "base_fighter_contribution_score"
] = population[
    [
        "quality_form_score",
        "experience_accomplishment_score"
    ]
].mean(
    axis=1
)


# ------------------------------------------------------------
# 6. LOCATE NUNES AT UFC 269
# ------------------------------------------------------------

nunes_row = population[
    (population["fighter"] == "Amanda Nunes")
    &
    (
        population["event_name"]
        .str.contains(
            "UFC 269",
            case=False,
            na=False
        )
    )
].copy()


# ------------------------------------------------------------
# 7. LOCATE FIGUEIREDO AT UFC 283
# ------------------------------------------------------------

fig_row = population[
    (
        population["fighter"]
        == "Deiveson Figueiredo"
    )
    &
    (
        population["event_name"]
        .str.contains(
            "UFC 283",
            case=False,
            na=False
        )
    )
].copy()


if len(nunes_row) != 1:
    raise ValueError(
        f"Expected 1 Nunes UFC 269 row, found {len(nunes_row)}"
    )

if len(fig_row) != 1:
    raise ValueError(
        f"Expected 1 Figueiredo UFC 283 row, found {len(fig_row)}"
    )


# ------------------------------------------------------------
# 8. BUILD FINAL DISPLAY TABLE
# ------------------------------------------------------------

selected = pd.concat(
    [nunes_row, fig_row],
    ignore_index=True
)


display_cols = [
    "fighter",
    "event_name",

    "ufc_fights_score",
    "ufc_wins_score",
    "ufc_win_pct_score",

    "win_streak_score",
    "last5_win_pct_score",

    "ufc_finishes_score",
    "finish_rate_score",

    "previous_title_fights_score",

    "quality_form_score",
    "experience_accomplishment_score",

    "base_fighter_contribution_score"
]


result = selected[
    display_cols
].copy()


score_cols = [
    col for col in result.columns
    if col.endswith("_score")
]


result[
    score_cols
] = result[
    score_cols
].round(2)


# ------------------------------------------------------------
# 9. HISTORICAL PERCENTILE OF THE FINAL BASE SCORE
#
# This tells us where each fighter appearance sits relative
# to ALL historical UFC fighter appearances.
# ------------------------------------------------------------

population[
    "contribution_percentile"
] = (
    population[
        "base_fighter_contribution_score"
    ]
    .rank(
        pct=True,
        method="average"
    )
    * 100
)


for idx in result.index:

    fighter = result.loc[idx, "fighter"]
    event = result.loc[idx, "event_name"]

    matched = population[
        (population["fighter"] == fighter)
        &
        (population["event_name"] == event)
    ]

    result.loc[
        idx,
        "contribution_percentile"
    ] = matched[
        "contribution_percentile"
    ].iloc[0]


result[
    "contribution_percentile"
] = result[
    "contribution_percentile"
].round(2)


# ------------------------------------------------------------
# 10. OUTPUT
# ------------------------------------------------------------

print("=" * 100)
print("🔥 HISTORICAL FIGHTER CARD CONTRIBUTION — BASE MODEL")
print("=" * 100)

print(
    f"\nHistorical fighter appearances scored: "
    f"{len(population):,}"
)

print(
    f"Unique fighters: "
    f"{population['fighter'].nunique():,}"
)

print(
    f"Date range: "
    f"{population['date'].min().date()} "
    f"to "
    f"{population['date'].max().date()}"
)


print("\n" + "=" * 100)
print("NUNES vs FIGUEIREDO")
print("=" * 100 + "\n")

display(
    result.set_index("fighter").T
)


# ------------------------------------------------------------
# 11. EASY-TO-READ SUMMARY
# ------------------------------------------------------------

n = result[
    result["fighter"] == "Amanda Nunes"
].iloc[0]

f = result[
    result["fighter"] == "Deiveson Figueiredo"
].iloc[0]


print("\n" + "=" * 100)
print("SUMMARY")
print("=" * 100)

print(
    f"\nAmanda Nunes — UFC 269"
)

print(
    f"Quality/Form: "
    f"{n['quality_form_score']:.2f}"
)

print(
    f"Experience/Accomplishment: "
    f"{n['experience_accomplishment_score']:.2f}"
)

print(
    f"BASE Fighter Contribution: "
    f"{n['base_fighter_contribution_score']:.2f}"
)

print(
    f"Historical percentile: "
    f"{n['contribution_percentile']:.2f}"
)


print(
    f"\nDeiveson Figueiredo — UFC 283"
)

print(
    f"Quality/Form: "
    f"{f['quality_form_score']:.2f}"
)

print(
    f"Experience/Accomplishment: "
    f"{f['experience_accomplishment_score']:.2f}"
)

print(
    f"BASE Fighter Contribution: "
    f"{f['base_fighter_contribution_score']:.2f}"
)

print(
    f"Historical percentile: "
    f"{f['contribution_percentile']:.2f}"
)


difference = (
    n["base_fighter_contribution_score"]
    -
    f["base_fighter_contribution_score"]
)


print(
    f"\nContribution-score difference: "
    f"{difference:+.2f} points in Nunes' favor"
)


print("\n⚠️ IMPORTANT")

print(
    "This is the BASE contribution score using the "
    "8 reconstructed pre-fight variables."
)

print(
    "It is now normalized against the historical UFC "
    "fighter-appearance population, NOT against each other."
)

print(
    "Ranking/champion status, previous main events, and "
    "previous bonuses are still reserved for the final extension."
)
```

```python
# ============================================================
# BIG CELL 3
# FINAL FIGHTER CARD CONTRIBUTION EXTENSION
#
# Adds:
#   - Event-time divisional ranking
#   - Champion status
#   - Ranking strength
#
# Preserves:
#   - All 8 historical pre-fight résumé variables
#
# Main-event history and bonus history are included ONLY if
# those exact columns already exist in the active runtime.
#
# NO FUTURE LEAKAGE
# DOES NOT MODIFY THE EVENT STRENGTH INDEX
# ============================================================

import pandas as pd
import numpy as np


# ============================================================
# 1. LOAD HISTORICAL UFC RANKINGS
# ============================================================

rankings = pd.read_csv(
    "https://raw.githubusercontent.com/rfordatascience/"
    "tidytuesday/main/data/2026/2026-07-07/"
    "ufc_rankings_dataset.csv"
)

rankings["date"] = pd.to_datetime(rankings["date"])

rankings["fighter"] = (
    rankings["fighter"]
    .astype(str)
    .str.strip()
)

rankings["rank"] = pd.to_numeric(
    rankings["rank"],
    errors="coerce"
)


print("=" * 100)
print("🔥 FINAL FIGHTER CARD CONTRIBUTION — EXTENSION")
print("=" * 100)

print(
    f"\nRanking snapshots loaded: "
    f"{len(rankings):,}"
)

print(
    f"Ranking date range: "
    f"{rankings['date'].min().date()} "
    f"to "
    f"{rankings['date'].max().date()}"
)


# ============================================================
# 2. FIND MOST RECENT RANKING ENTERING EACH FIGHT
#
# IMPORTANT:
# Only rankings published ON OR BEFORE the event are allowed.
#
# We use divisional rankings and exclude P4P because the
# Event Strength model's ranking variable is divisional.
# ============================================================

divisional_rankings = rankings[
    ~rankings["weightclass"]
    .astype(str)
    .str.contains(
        "Pound-for-Pound",
        case=False,
        na=False
    )
].copy()


# ============================================================
# 3. FUNCTION TO GET A FIGHTER'S BEST CURRENT DIVISIONAL RANK
#
# Champion = 0
# Ranked contender = 1–15
# Unranked = NaN
# ============================================================

def get_rank_entering_event(fighter, event_date):

    event_date = pd.Timestamp(event_date)

    fighter_history = divisional_rankings[
        (divisional_rankings["fighter"] == fighter)
        &
        (divisional_rankings["date"] <= event_date)
    ].copy()

    if fighter_history.empty:
        return np.nan

    # Most recent rankings publication available
    latest_date = fighter_history["date"].max()

    latest = fighter_history[
        fighter_history["date"] == latest_date
    ].copy()

    valid = latest[
        latest["rank"].between(
            0,
            15,
            inclusive="both"
        )
    ]

    if valid.empty:
        return np.nan

    # If somehow present in multiple divisions,
    # take strongest rank.
    return valid["rank"].min()


# ============================================================
# 4. GET NUNES + FIGUEIREDO EVENT-TIME RANKS
# ============================================================

nunes_event_date = pd.Timestamp("2021-12-11")
fig_event_date = pd.Timestamp("2023-01-21")

nunes_rank = get_rank_entering_event(
    "Amanda Nunes",
    nunes_event_date
)

fig_rank = get_rank_entering_event(
    "Deiveson Figueiredo",
    fig_event_date
)


print("\nEVENT-TIME DIVISIONAL RANKINGS")
print("-" * 50)

print(
    f"Amanda Nunes entering UFC 269: "
    f"{nunes_rank}"
)

print(
    f"Deiveson Figueiredo entering UFC 283: "
    f"{fig_rank}"
)


# ============================================================
# 5. SAME RANK-POINT SYSTEM AS EVENT STRENGTH PROJECT
#
# Champion = 16 points
# #1 = 15
# #2 = 14
# ...
# #15 = 1
# Unranked = 0
# ============================================================

def fighter_rank_points(rank):

    if pd.isna(rank):
        return 0

    rank = float(rank)

    if rank == 0:
        return 16

    if 1 <= rank <= 15:
        return 16 - rank

    return 0


nunes_rank_points = fighter_rank_points(
    nunes_rank
)

fig_rank_points = fighter_rank_points(
    fig_rank
)


print("\nRANKING STRENGTH")
print("-" * 50)

print(
    f"Nunes ranking points: "
    f"{nunes_rank_points}"
)

print(
    f"Figueiredo ranking points: "
    f"{fig_rank_points}"
)


# ============================================================
# 6. ADD EVENT-TIME RANKING TO EVERY HISTORICAL
#    FIGHTER APPEARANCE WHERE RANKINGS EXIST
#
# Rankings begin in 2013.
#
# This can take a little bit because we're matching historical
# fighter appearances to historical ranking snapshots.
# ============================================================

print(
    "\nBuilding historical ranking-strength population..."
)


ranking_cache = {}


def cached_rank_lookup(fighter, date):

    date = pd.Timestamp(date)

    key = (
        fighter,
        date
    )

    if key not in ranking_cache:

        ranking_cache[key] = (
            get_rank_entering_event(
                fighter,
                date
            )
        )

    return ranking_cache[key]


population[
    "rank_before"
] = population.apply(
    lambda x:
        cached_rank_lookup(
            x["fighter"],
            x["date"]
        )
        if x["date"] >= pd.Timestamp("2013-02-04")
        else np.nan,
    axis=1
)


population[
    "rank_points"
] = population[
    "rank_before"
].apply(
    lambda x:
        fighter_rank_points(x)
        if pd.notna(x)
        else np.nan
)


# ============================================================
# 7. RANKING PERCENTILE SCORE
#
# Keep true unranked fighters at zero during the rankings era.
# Pre-ranking-era appearances remain unavailable.
# ============================================================

rankings_start = pd.Timestamp(
    "2013-02-04"
)

ranking_era = (
    population["date"]
    >= rankings_start
)


population[
    "ranking_strength_score"
] = np.nan


rank_points_for_scoring = (
    population.loc[
        ranking_era,
        "rank_points"
    ]
    .fillna(0)
)


population.loc[
    ranking_era,
    "ranking_strength_score"
] = (
    zero_preserving_percentile(
        rank_points_for_scoring
    )
)


# ============================================================
# 8. REBUILD THE TWO CORE COMPONENTS
# ============================================================

quality_cols = [
    "ufc_win_pct_score",
    "win_streak_score",
    "last5_win_pct_score",
    "finish_rate_score"
]


experience_cols = [
    "ufc_fights_score",
    "ufc_wins_score",
    "ufc_finishes_score",
    "previous_title_fights_score"
]


population[
    "fighter_quality_score"
] = population[
    quality_cols
].mean(
    axis=1,
    skipna=True
)


population[
    "fighter_experience_score"
] = population[
    experience_cols
].mean(
    axis=1,
    skipna=True
)


# ============================================================
# 9. BUILD FIGHTER STATUS / RANKING COMPONENT
# ============================================================

population[
    "fighter_status_score"
] = population[
    "ranking_strength_score"
]


# ============================================================
# 10. CHECK WHETHER EXACT MAIN-EVENT / BONUS VARIABLES
#     EXIST IN THIS RUNTIME
#
# We will NOT invent them.
# ============================================================

possible_main_event_cols = [
    "previous_main_events",
    "f1_previous_main_events",
    "f2_previous_main_events"
]

possible_bonus_cols = [
    "previous_total_bonuses",
    "f1_previous_total_bonuses",
    "f2_previous_total_bonuses"
]


available_main_event_cols = [
    c for c in possible_main_event_cols
    if c in population.columns
]

available_bonus_cols = [
    c for c in possible_bonus_cols
    if c in population.columns
]


print("\nOPTIONAL HISTORICAL VARIABLES FOUND")
print("-" * 50)

print(
    "Main-event history:",
    available_main_event_cols
    if available_main_event_cols
    else "not active in restarted runtime"
)

print(
    "Bonus history:",
    available_bonus_cols
    if available_bonus_cols
    else "not active in restarted runtime"
)


# ============================================================
# 11. FINAL CURRENT CONTRIBUTION SCORE
#
# Philosophy:
#
# 40% Quality / Form
# 40% Experience / Accomplishment
# 20% Ranking / Championship Status
#
# This is a NEW fighter-level extension.
# It does NOT alter the locked Event Strength weights.
#
# Ranking is deliberately meaningful but does not overpower
# the fighter's complete UFC résumé.
# ============================================================

FIGHTER_WEIGHTS = {
    "fighter_quality_score": 0.40,
    "fighter_experience_score": 0.40,
    "fighter_status_score": 0.20
}


def weighted_available_score(
    row,
    weights
):

    available = {
        col: weight
        for col, weight in weights.items()
        if pd.notna(row[col])
    }

    if not available:
        return np.nan

    total_weight = sum(
        available.values()
    )

    return sum(
        row[col] * weight
        for col, weight in available.items()
    ) / total_weight


population[
    "fighter_card_contribution_score"
] = population.apply(
    lambda row:
        weighted_available_score(
            row,
            FIGHTER_WEIGHTS
        ),
    axis=1
)


# ============================================================
# 12. HISTORICAL PERCENTILE
# ============================================================

population[
    "fighter_card_contribution_percentile"
] = (
    population[
        "fighter_card_contribution_score"
    ]
    .rank(
        pct=True,
        method="average"
    )
    * 100
)


# ============================================================
# 13. EXTRACT NUNES + FIGUEIREDO
# ============================================================

final_nunes = population[
    (population["fighter"] == "Amanda Nunes")
    &
    (
        population["event_name"]
        .str.contains(
            "UFC 269",
            case=False,
            na=False
        )
    )
].copy()


final_fig = population[
    (
        population["fighter"]
        == "Deiveson Figueiredo"
    )
    &
    (
        population["event_name"]
        .str.contains(
            "UFC 283",
            case=False,
            na=False
        )
    )
].copy()


if len(final_nunes) != 1:
    raise ValueError(
        f"Nunes rows found: {len(final_nunes)}"
    )

if len(final_fig) != 1:
    raise ValueError(
        f"Figueiredo rows found: {len(final_fig)}"
    )


final_compare = pd.concat(
    [
        final_nunes,
        final_fig
    ],
    ignore_index=True
)


# ============================================================
# 14. DISPLAY FINAL COMPARISON
# ============================================================

final_display_cols = [
    "fighter",
    "event_name",

    "ufc_fights",
    "ufc_wins",
    "ufc_win_pct",

    "win_streak",
    "last5_win_pct",

    "ufc_finishes",
    "finish_rate",

    "previous_title_fights",

    "rank_before",
    "rank_points",

    "fighter_quality_score",
    "fighter_experience_score",
    "fighter_status_score",

    "fighter_card_contribution_score",
    "fighter_card_contribution_percentile"
]


final_display = final_compare[
    final_display_cols
].copy()


round_cols = [
    "ufc_win_pct",
    "last5_win_pct",
    "finish_rate",
    "fighter_quality_score",
    "fighter_experience_score",
    "fighter_status_score",
    "fighter_card_contribution_score",
    "fighter_card_contribution_percentile"
]


for col in round_cols:

    final_display[col] = (
        pd.to_numeric(
            final_display[col],
            errors="coerce"
        )
        .round(2)
    )


print("\n" + "=" * 100)
print("🔥 FIGHTER CARD CONTRIBUTION — FINAL CURRENT MODEL")
print("=" * 100)

display(
    final_display
    .set_index("fighter")
    .T
)


# ============================================================
# 15. EASY SUMMARY
# ============================================================

N = final_display[
    final_display.index
    == "Amanda Nunes"
]

# Because fighter is now index only after display,
# retrieve from final_compare instead for summary

n = final_compare[
    final_compare["fighter"]
    == "Amanda Nunes"
].iloc[0]

f = final_compare[
    final_compare["fighter"]
    == "Deiveson Figueiredo"
].iloc[0]


n_score = (
    n[
        "fighter_card_contribution_score"
    ]
)

f_score = (
    f[
        "fighter_card_contribution_score"
    ]
)


difference = (
    n_score
    -
    f_score
)


print("\n" + "=" * 100)
print("FINAL SUMMARY")
print("=" * 100)


print("\n🔥 AMANDA NUNES — UFC 269")

print(
    f"Quality / Form: "
    f"{n['fighter_quality_score']:.2f}"
)

print(
    f"Experience / Accomplishment: "
    f"{n['fighter_experience_score']:.2f}"
)

print(
    f"Ranking / Status: "
    f"{n['fighter_status_score']:.2f}"
)

print(
    f"FIGHTER CARD CONTRIBUTION: "
    f"{n_score:.2f} / 100"
)

print(
    f"Historical percentile: "
    f"{n['fighter_card_contribution_percentile']:.2f}"
)


print("\n🔥 DEIVESON FIGUEIREDO — UFC 283")

print(
    f"Quality / Form: "
    f"{f['fighter_quality_score']:.2f}"
)

print(
    f"Experience / Accomplishment: "
    f"{f['fighter_experience_score']:.2f}"
)

print(
    f"Ranking / Status: "
    f"{f['fighter_status_score']:.2f}"
)

print(
    f"FIGHTER CARD CONTRIBUTION: "
    f"{f_score:.2f} / 100"
)

print(
    f"Historical percentile: "
    f"{f['fighter_card_contribution_percentile']:.2f}"
)


print("\n" + "-" * 100)

print(
    f"Difference: "
    f"{difference:+.2f} points "
    f"in Nunes' favor"
)


print("\nMODEL WEIGHTS")

print(
    "40% Quality/Form"
)

print(
    "40% Experience/Accomplishment"
)

print(
    "20% Ranking/Championship Status"
)


print("\nIMPORTANT:")

print(
    "This Fighter Card Contribution metric is a NEW "
    "fighter-level extension of the project."
)

print(
    "It does NOT change the locked 37/23/28/12 "
    "Event Strength Index."
)

print(
    "All résumé statistics are calculated entering "
    "the event to prevent future leakage."
)

print(
    "Previous main-event and bonus history will only "
    "be added when their exact historical variables "
    "are available — they are not being fabricated."
)
```

```python
# ============================================================
# MODEL B — RANKING EMPHASIS SENSITIVITY TEST
# SAFE TEST ONLY — DOES NOT MODIFY ORIGINAL MODEL
# ============================================================

import pandas as pd
import numpy as np

print("=" * 100)
print("🔥 MODEL B — RANKING EMPHASIS TEST")
print("=" * 100)

# ------------------------------------------------------------
# 1. FIND THE BEST EVENT-LEVEL TABLE STILL IN MEMORY
# ------------------------------------------------------------

possible_tables = [
    "event_master_updated",
    "event_master",
    "final_event_strength",
    "final_event_strength_pre_update"
]

source_name = None
source_df = None

for name in possible_tables:
    if name in globals():
        obj = globals()[name]

        if isinstance(obj, pd.DataFrame):
            source_name = name
            source_df = obj.copy()
            break

print("\nEvent table found:", source_name)

if source_df is None:

    print("\n❌ FINAL EVENT TABLE IS NOT ACTIVE IN THIS RUNTIME.")
    print()
    print("DO NOT rerun the whole notebook.")
    print()
    print("Send me ONLY this output.")
    print("I will give you one rebuild cell using the objects")
    print("that ARE currently active.")

else:

    model_b = source_df.copy()

    print(
        f"Rows available: {len(model_b):,}"
    )

    print(
        f"Columns available: {len(model_b.columns):,}"
    )

    # --------------------------------------------------------
    # 2. SHOW WHICH REQUIRED VARIABLES EXIST
    # --------------------------------------------------------

    required = [
        "title_fights_score",
        "total_ranking_points_score",
        "avg_ufc_experience_score",
        "avg_ufc_win_pct_score",
        "avg_win_streak_score",
        "avg_finish_rate_score",
        "star_experience_score",
        "main_event_score",
        "fanfare_score"
    ]

    optional_depth = [
        "ranked_fighters_score",
        "ranked_matchups_score"
    ]

    missing = [
        c for c in required
        if c not in model_b.columns
    ]

    print("\nRequired columns missing:")
    print(missing if missing else "NONE ✅")

    if missing:

        print("\n⚠️ STOPPING SAFELY.")
        print("Nothing has been changed.")
        print("Send me this missing-column list.")

    else:

        # ----------------------------------------------------
        # 3. SAVE ORIGINAL MODEL INFORMATION
        # ----------------------------------------------------

        if "event_strength_score" in model_b.columns:
            original_score_col = "event_strength_score"

        elif "final_event_strength_score" in model_b.columns:
            original_score_col = "final_event_strength_score"

        elif "event_strength" in model_b.columns:
            original_score_col = "event_strength"

        else:
            original_score_col = None

        if original_score_col is not None:

            model_b[
                "model_a_score"
            ] = model_b[
                original_score_col
            ]

            model_b[
                "model_a_rank"
            ] = (
                model_b["model_a_score"]
                .rank(
                    ascending=False,
                    method="min"
                )
            )

        # ----------------------------------------------------
        # 4. BUILD COMPETITIVE DEPTH COMPONENT
        #
        # Ranking/depth gets MUCH more influence.
        # ----------------------------------------------------

        depth_inputs = [
            "total_ranking_points_score",
            "title_fights_score"
        ]

        # Include these if our final table contains them
        if "ranked_fighters_score" in model_b.columns:
            depth_inputs.append(
                "ranked_fighters_score"
            )

        if "ranked_matchups_score" in model_b.columns:
            depth_inputs.append(
                "ranked_matchups_score"
            )

        model_b[
            "competitive_depth_score_B"
        ] = model_b[
            depth_inputs
        ].mean(
            axis=1,
            skipna=True
        )

        # ----------------------------------------------------
        # 5. RÉSUMÉ / FORM PORTION
        # ----------------------------------------------------

        resume_inputs = [
            "avg_ufc_experience_score",
            "avg_ufc_win_pct_score",
            "avg_win_streak_score",
            "avg_finish_rate_score"
        ]

        model_b[
            "resume_form_score_B"
        ] = model_b[
            resume_inputs
        ].mean(
            axis=1,
            skipna=True
        )

        # ----------------------------------------------------
        # 6. NEW CARD QUALITY
        #
        # 50% competitive depth
        # 50% résumé/form
        # ----------------------------------------------------

        model_b[
            "card_quality_score_B"
        ] = (
            0.50
            * model_b[
                "competitive_depth_score_B"
            ]
            +
            0.50
            * model_b[
                "resume_form_score_B"
            ]
        )

        # ----------------------------------------------------
        # 7. KEEP OVERALL MODEL WEIGHTS IDENTICAL
        #
        # Card Quality       37%
        # Star Experience    23%
        # Main Event         28%
        # Fanfare            12%
        #
        # Dynamic redistribution for missing components.
        # ----------------------------------------------------

        MODEL_B_WEIGHTS = {
            "card_quality_score_B": 0.37,
            "star_experience_score": 0.23,
            "main_event_score": 0.28,
            "fanfare_score": 0.12
        }

        def weighted_missing_safe(row, weights):

            available = {
                col: weight
                for col, weight in weights.items()
                if pd.notna(row[col])
            }

            if not available:
                return np.nan

            total_weight = sum(
                available.values()
            )

            return sum(
                row[col] * weight
                for col, weight in available.items()
            ) / total_weight

        model_b[
            "model_b_score"
        ] = model_b.apply(
            lambda row:
            weighted_missing_safe(
                row,
                MODEL_B_WEIGHTS
            ),
            axis=1
        )

        model_b[
            "model_b_rank"
        ] = (
            model_b[
                "model_b_score"
            ]
            .rank(
                ascending=False,
                method="min"
            )
        )

        # ----------------------------------------------------
        # 8. IDENTIFY EVENT-NAME COLUMN
        # ----------------------------------------------------

        event_col = None

        for candidate in [
            "event_name",
            "event"
        ]:

            if candidate in model_b.columns:
                event_col = candidate
                break

        if event_col is None:
            raise ValueError(
                "Could not find event-name column."
            )

        # ----------------------------------------------------
        # 9. SCORE/RANK MOVEMENT
        # ----------------------------------------------------

        if original_score_col is not None:

            model_b[
                "score_change"
            ] = (
                model_b["model_b_score"]
                -
                model_b["model_a_score"]
            )

            model_b[
                "rank_change"
            ] = (
                model_b["model_a_rank"]
                -
                model_b["model_b_rank"]
            )

        # ----------------------------------------------------
        # 10. TARGET CARDS
        # ----------------------------------------------------

        targets = [
            "UFC 302",
            "UFC 330",
            "UFC 269",
            "UFC 300",
            "UFC 205",
            "UFC 229"
        ]

        target_mask = pd.Series(
            False,
            index=model_b.index
        )

        for target in targets:

            target_mask |= (
                model_b[event_col]
                .astype(str)
                .str.contains(
                    target,
                    case=False,
                    na=False,
                    regex=False
                )
            )

        target_results = model_b[
            target_mask
        ].copy()

        # ----------------------------------------------------
        # 11. OUTPUT TARGET COMPARISON
        # ----------------------------------------------------

        print("\n" + "=" * 100)
        print("🔥 TARGET EVENT COMPARISON")
        print("=" * 100)

        output_cols = [
            event_col
        ]

        if "ranked_fighters" in model_b.columns:
            output_cols.append(
                "ranked_fighters"
            )

        if "ranked_matchups" in model_b.columns:
            output_cols.append(
                "ranked_matchups"
            )

        if "total_ranking_points" in model_b.columns:
            output_cols.append(
                "total_ranking_points"
            )

        output_cols += [
            "competitive_depth_score_B",
            "card_quality_score_B"
        ]

        if original_score_col is not None:

            output_cols += [
                "model_a_score",
                "model_a_rank",
                "model_b_score",
                "model_b_rank",
                "score_change",
                "rank_change"
            ]

        else:

            output_cols += [
                "model_b_score",
                "model_b_rank"
            ]

        display(
            target_results[
                output_cols
            ]
            .sort_values(
                "model_b_rank"
            )
            .round(2)
        )

        # ----------------------------------------------------
        # 12. TOP 20 MODEL B
        # ----------------------------------------------------

        print("\n" + "=" * 100)
        print("🔥 MODEL B — TOP 20")
        print("=" * 100)

        top20_cols = [
            event_col,
            "competitive_depth_score_B",
            "card_quality_score_B",
            "model_b_score",
            "model_b_rank"
        ]

        if original_score_col is not None:
            top20_cols += [
                "model_a_rank",
                "rank_change"
            ]

        display(
            model_b[
                top20_cols
            ]
            .sort_values(
                "model_b_rank"
            )
            .head(20)
            .round(2)
        )

        # ----------------------------------------------------
        # 13. BIGGEST RISERS / FALLERS
        # ----------------------------------------------------

        if original_score_col is not None:

            print("\n" + "=" * 100)
            print("📈 BIGGEST RISERS")
            print("=" * 100)

            display(
                model_b[
                    [
                        event_col,
                        "model_a_rank",
                        "model_b_rank",
                        "rank_change",
                        "model_a_score",
                        "model_b_score"
                    ]
                ]
                .sort_values(
                    "rank_change",
                    ascending=False
                )
                .head(15)
                .round(2)
            )

            print("\n" + "=" * 100)
            print("📉 BIGGEST FALLERS")
            print("=" * 100)

            display(
                model_b[
                    [
                        event_col,
                        "model_a_rank",
                        "model_b_rank",
                        "rank_change",
                        "model_a_score",
                        "model_b_score"
                    ]
                ]
                .sort_values(
                    "rank_change",
                    ascending=True
                )
                .head(15)
                .round(2)
            )

        # ----------------------------------------------------
        # 14. SAFETY CONFIRMATION
        # ----------------------------------------------------

        print("\n" + "=" * 100)
        print("✅ MODEL B TEST COMPLETE")
        print("=" * 100)

        print(
            "\nOriginal source table:",
            source_name
        )

        print(
            "Test table: model_b"
        )

        print(
            "Original model overwritten: NO"
        )

        print(
            "Original exports changed: NO"
        )

        print(
            "\nCompetitive-depth inputs:",
            depth_inputs
        )

        print(
            "\nModel B tests stronger ranking/depth emphasis "
            "while keeping the overall 37/23/28/12 weights."
        )
```

```python
# ============================================================
# MODEL B — REBUILD RANKING DEPTH + SENSITIVITY TEST
#
# SAFE SANDBOX:
# - DOES NOT modify final_index
# - DOES NOT modify your finished Event Strength Index
# - DOES NOT overwrite any files
# ============================================================

import pandas as pd
import numpy as np

print("=" * 100)
print("🔥 MODEL B — RANKING DEPTH REBUILD")
print("=" * 100)


# ============================================================
# 1. COPY FINISHED MODEL
# ============================================================

model_b = final_index.copy()

model_b["event_date"] = pd.to_datetime(
    model_b["event_date"]
)

print(f"\nFinished events loaded: {len(model_b):,}")


# ============================================================
# 2. PREPARE HISTORICAL FIGHTS
# ============================================================

fight_test = fights.copy()

fight_test["date"] = pd.to_datetime(
    fight_test["date"]
)

print(f"Historical fights available: {len(fight_test):,}")


# ============================================================
# 3. TITLE-FIGHT FLAG
# SAME LOGIC AS ORIGINAL PROJECT
# ============================================================

fight_test["is_title_fight_test"] = (
    fight_test["weight_class"]
    .astype(str)
    .str.startswith("UFC ")
    &
    fight_test["weight_class"]
    .astype(str)
    .str.contains(
        "Title Bout",
        case=False,
        na=False
    )
    &
    ~fight_test["weight_class"]
    .astype(str)
    .str.contains(
        "Tournament",
        case=False,
        na=False
    )
)


# ============================================================
# 4. RANK LOOKUP
#
# Uses the ranking objects already built in the previous cell.
# Champion = 0
# #1 through #15 = ranked
# NaN = unranked
# ============================================================

rank_cache_B = {}


def rank_at_event_B(fighter, date):

    date = pd.Timestamp(date)

    key = (fighter, date)

    if key in rank_cache_B:
        return rank_cache_B[key]

    history = divisional_rankings[
        (divisional_rankings["fighter"] == fighter)
        &
        (divisional_rankings["date"] <= date)
    ]

    if history.empty:

        rank_cache_B[key] = np.nan
        return np.nan

    latest_date = history["date"].max()

    current = history[
        history["date"] == latest_date
    ]

    valid = current[
        current["rank"].between(
            0,
            15,
            inclusive="both"
        )
    ]

    if valid.empty:

        rank_cache_B[key] = np.nan
        return np.nan

    value = valid["rank"].min()

    rank_cache_B[key] = value

    return value


print("\nCalculating historical event-time rankings...")


fight_test["f1_rank_test"] = fight_test.apply(
    lambda x:
    rank_at_event_B(
        x["f1_name"],
        x["date"]
    )
    if x["date"] >= pd.Timestamp("2013-02-04")
    else np.nan,
    axis=1
)


fight_test["f2_rank_test"] = fight_test.apply(
    lambda x:
    rank_at_event_B(
        x["f2_name"],
        x["date"]
    )
    if x["date"] >= pd.Timestamp("2013-02-04")
    else np.nan,
    axis=1
)


# ============================================================
# 5. RANKING POINTS
#
# Champion = 16
# #1 = 15
# ...
# #15 = 1
# Unranked = 0
# ============================================================

def rank_points_B(rank):

    if pd.isna(rank):
        return 0

    rank = float(rank)

    if rank == 0:
        return 16

    if 1 <= rank <= 15:
        return 16 - rank

    return 0


fight_test["f1_rank_points_B"] = (
    fight_test["f1_rank_test"]
    .apply(rank_points_B)
)

fight_test["f2_rank_points_B"] = (
    fight_test["f2_rank_test"]
    .apply(rank_points_B)
)


# ============================================================
# 6. FIGHT-LEVEL RANKING DEPTH
# ============================================================

ranking_era_fight = (
    fight_test["date"]
    >= pd.Timestamp("2013-02-04")
)


fight_test["ranked_fighters_B"] = np.where(
    ranking_era_fight,
    fight_test["f1_rank_test"].notna().astype(int)
    +
    fight_test["f2_rank_test"].notna().astype(int),
    np.nan
)


fight_test["ranked_matchup_B"] = np.where(
    ranking_era_fight,
    (
        fight_test["f1_rank_test"].notna()
        &
        fight_test["f2_rank_test"].notna()
    ).astype(int),
    np.nan
)


fight_test["ranking_points_B"] = np.where(
    ranking_era_fight,
    fight_test["f1_rank_points_B"]
    +
    fight_test["f2_rank_points_B"],
    np.nan
)


# ============================================================
# 7. AGGREGATE TO EVENT LEVEL
# ============================================================

event_depth = (
    fight_test
    .groupby(
        ["date", "event_name"],
        as_index=False
    )
    .agg(
        ranked_fighters_B=(
            "ranked_fighters_B",
            lambda x:
            x.sum(min_count=1)
        ),

        ranked_matchups_B=(
            "ranked_matchup_B",
            lambda x:
            x.sum(min_count=1)
        ),

        total_ranking_points_B=(
            "ranking_points_B",
            lambda x:
            x.sum(min_count=1)
        ),

        title_fights_B=(
            "is_title_fight_test",
            "sum"
        )
    )
)


# ============================================================
# 8. MERGE WITH FINISHED 788-EVENT MODEL
# ============================================================

model_b = model_b.merge(
    event_depth,
    left_on=[
        "event_date",
        "event_name"
    ],
    right_on=[
        "date",
        "event_name"
    ],
    how="left"
)

if "date" in model_b.columns:
    model_b = model_b.drop(
        columns=["date"]
    )


# ============================================================
# 9. ZERO-PRESERVING PERCENTILE
# ============================================================

def zero_percentile_B(series):

    s = pd.to_numeric(
        series,
        errors="coerce"
    )

    result = pd.Series(
        np.nan,
        index=s.index,
        dtype=float
    )

    result[s == 0] = 0

    positive = s > 0

    if positive.sum() > 0:

        result.loc[positive] = (
            s.loc[positive]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )

    return result


for col in [
    "ranked_fighters_B",
    "ranked_matchups_B",
    "total_ranking_points_B",
    "title_fights_B"
]:

    model_b[
        col + "_score"
    ] = zero_percentile_B(
        model_b[col]
    )


# ============================================================
# 10. COMPETITIVE DEPTH SCORE
#
# Ranking information gets 85% of this component:
#
# 40% total ranking points
# 25% ranked fighters
# 20% ranked matchups
# 15% title fights
#
# This is deliberately a STRONG ranking-emphasis experiment.
# ============================================================

model_b[
    "competitive_depth_B"
] = (
    0.40
    * model_b[
        "total_ranking_points_B_score"
    ]
    +
    0.25
    * model_b[
        "ranked_fighters_B_score"
    ]
    +
    0.20
    * model_b[
        "ranked_matchups_B_score"
    ]
    +
    0.15
    * model_b[
        "title_fights_B_score"
    ]
)


# ============================================================
# 11. MODEL B CARD QUALITY
#
# IMPORTANT:
#
# We do NOT have the four original résumé/form sub-scores
# in this CSV.
#
# Therefore we use original Card Quality as the résumé/
# general-quality anchor and blend in the NEW competitive
# depth measure.
#
# 50% original Card Quality
# 50% competitive depth
#
# This is a SENSITIVITY TEST, not a replacement model.
# ============================================================

model_b[
    "card_quality_B"
] = (
    0.50
    * model_b[
        "card_quality_score"
    ]
    +
    0.50
    * model_b[
        "competitive_depth_B"
    ]
)


# ============================================================
# 12. ORIGINAL MODEL
# ============================================================

model_b[
    "model_A_score"
] = model_b[
    "event_strength_score"
]

model_b[
    "model_A_rank"
] = model_b[
    "event_strength_rank"
]


# ============================================================
# 13. MODEL B FINAL SCORE
#
# SAME OVERALL WEIGHTS:
#
# 37% Card Quality
# 23% Star Experience
# 28% Main Event
# 12% Fanfare
#
# Missing components dynamically redistribute.
# ============================================================

MODEL_B_WEIGHTS = {
    "card_quality_B": 0.37,
    "star_experience_score": 0.23,
    "main_event_score": 0.28,
    "fanfare_score": 0.12
}


def weighted_B(row):

    available = {
        col: weight
        for col, weight
        in MODEL_B_WEIGHTS.items()
        if pd.notna(row[col])
    }

    if not available:
        return np.nan

    total = sum(
        available.values()
    )

    return sum(
        row[col] * weight
        for col, weight
        in available.items()
    ) / total


model_b[
    "model_B_score"
] = model_b.apply(
    weighted_B,
    axis=1
)


model_b[
    "model_B_rank"
] = (
    model_b[
        "model_B_score"
    ]
    .rank(
        ascending=False,
        method="min"
    )
    .astype("Int64")
)


model_b[
    "rank_change"
] = (
    model_b[
        "model_A_rank"
    ]
    -
    model_b[
        "model_B_rank"
    ]
)


model_b[
    "score_change"
] = (
    model_b[
        "model_B_score"
    ]
    -
    model_b[
        "model_A_score"
    ]
)


# ============================================================
# 14. TARGET EVENTS
# ============================================================

targets = [
    "UFC 302",
    "UFC 330",
    "UFC 269",
    "UFC 300",
    "UFC 205",
    "UFC 229"
]


mask = pd.Series(
    False,
    index=model_b.index
)


for target in targets:

    mask |= (
        model_b[
            "event_name"
        ]
        .astype(str)
        .str.contains(
            target,
            case=False,
            na=False,
            regex=False
        )
    )


target_results = model_b[
    mask
].copy()


# ============================================================
# 15. OUTPUT
# ============================================================

print("\n" + "=" * 110)
print("🔥 TARGET EVENT COMPARISON")
print("=" * 110)


display(
    target_results[
        [
            "event_name",

            "ranked_fighters_B",
            "ranked_matchups_B",
            "total_ranking_points_B",
            "title_fights_B",

            "competitive_depth_B",

            "card_quality_score",
            "card_quality_B",

            "model_A_score",
            "model_A_rank",

            "model_B_score",
            "model_B_rank",

            "rank_change"
        ]
    ]
    .sort_values(
        "model_B_rank"
    )
    .round(2)
)


print("\n" + "=" * 110)
print("🔥 MODEL B TOP 20")
print("=" * 110)


display(
    model_b[
        [
            "event_name",
            "ranked_fighters_B",
            "ranked_matchups_B",
            "competitive_depth_B",
            "model_A_rank",
            "model_B_rank",
            "rank_change",
            "model_B_score"
        ]
    ]
    .sort_values(
        "model_B_rank"
    )
    .head(20)
    .round(2)
)


print("\n" + "=" * 110)
print("📈 15 BIGGEST RISERS")
print("=" * 110)


display(
    model_b[
        [
            "event_name",
            "model_A_rank",
            "model_B_rank",
            "rank_change",
            "ranked_fighters_B",
            "ranked_matchups_B"
        ]
    ]
    .sort_values(
        "rank_change",
        ascending=False
    )
    .head(15)
)


print("\n" + "=" * 110)
print("📉 15 BIGGEST FALLERS")
print("=" * 110)


display(
    model_b[
        [
            "event_name",
            "model_A_rank",
            "model_B_rank",
            "rank_change",
            "ranked_fighters_B",
            "ranked_matchups_B"
        ]
    ]
    .sort_values(
        "rank_change",
        ascending=True
    )
    .head(15)
)


print("\n" + "=" * 110)
print("✅ SAFETY CHECK")
print("=" * 110)

print(
    f"Events in finished model: "
    f"{len(final_index):,}"
)

print(
    f"Events in Model B: "
    f"{len(model_b):,}"
)

print(
    "Finished model modified: NO"
)

print(
    "Files overwritten: NO"
)

print(
    "Model B exists only in variable: model_b"
)

print(
    "\n🔥 Send me the TARGET EVENT COMPARISON table."
)
```

```python
# ============================================================
# MODEL B v2 — CLEAN RANKING SENSITIVITY TEST
#
# SAFE:
# - Uses existing model_b only
# - Does NOT alter final_index
# - Does NOT overwrite files
# - Pre-ranking era retains ORIGINAL Card Quality
# ============================================================

import pandas as pd
import numpy as np

print("=" * 105)
print("🔥 MODEL B v2 — CLEAN RANKING SENSITIVITY TEST")
print("=" * 105)

test = model_b.copy()

RANKING_START = pd.Timestamp("2013-02-04")

test["event_date"] = pd.to_datetime(test["event_date"])

# ------------------------------------------------------------
# 1. DEFINE WHERE OUR REBUILT RANKING DATA IS ACTUALLY VALID
#
# Historical fights in this runtime end 2026-06-27.
# Therefore:
#   2013-02-04 through 2026-06-27 = usable
#   before 2013 = no ranking system
#   after 2026-06-27 = incomplete reconstruction
# ------------------------------------------------------------

ranking_test_end = pd.Timestamp(
    fight_test["date"].max()
)

test["ranking_test_valid"] = (
    (test["event_date"] >= RANKING_START)
    &
    (test["event_date"] <= ranking_test_end)
    &
    (test["competitive_depth_B"].notna())
)

print("\nRanking test period:")
print(RANKING_START.date(), "through", ranking_test_end.date())

print(
    "Events with valid ranking-depth test:",
    test["ranking_test_valid"].sum()
)

print(
    "Events held at original Card Quality:",
    (~test["ranking_test_valid"]).sum()
)


# ------------------------------------------------------------
# 2. CREATE THREE CARD-QUALITY EXPERIMENTS
#
# B25 = mild
# B40 = strong
# B50 = aggressive
#
# Outside valid ranking era:
# EXACT original Card Quality.
# ------------------------------------------------------------

for pct in [25, 40, 50]:

    depth_weight = pct / 100
    original_weight = 1 - depth_weight

    new_col = f"card_quality_B{pct}"

    test[new_col] = test["card_quality_score"]

    valid = test["ranking_test_valid"]

    test.loc[
        valid,
        new_col
    ] = (
        original_weight
        * test.loc[
            valid,
            "card_quality_score"
        ]
        +
        depth_weight
        * test.loc[
            valid,
            "competitive_depth_B"
        ]
    )


# ------------------------------------------------------------
# 3. SAME FINAL MODEL WEIGHTS AS ORIGINAL
# ------------------------------------------------------------

def final_score_test(row, card_col):

    weights = {
        card_col: 0.37,
        "star_experience_score": 0.23,
        "main_event_score": 0.28,
        "fanfare_score": 0.12
    }

    available = {
        col: weight
        for col, weight in weights.items()
        if pd.notna(row[col])
    }

    if not available:
        return np.nan

    total_weight = sum(available.values())

    return sum(
        row[col] * weight
        for col, weight in available.items()
    ) / total_weight


# ------------------------------------------------------------
# 4. CALCULATE ALL THREE MODELS
# ------------------------------------------------------------

for pct in [25, 40, 50]:

    card_col = f"card_quality_B{pct}"
    score_col = f"score_B{pct}"
    rank_col = f"rank_B{pct}"
    change_col = f"rank_change_B{pct}"

    test[score_col] = test.apply(
        lambda row:
        final_score_test(
            row,
            card_col
        ),
        axis=1
    )

    test[rank_col] = (
        test[score_col]
        .rank(
            ascending=False,
            method="min"
        )
        .astype("Int64")
    )

    test[change_col] = (
        test["event_strength_rank"]
        -
        test[rank_col]
    )


# ------------------------------------------------------------
# 5. TARGET EVENTS
# ------------------------------------------------------------

targets = [
    "UFC 302",
    "UFC 300",
    "UFC 269",
    "UFC 205",
    "UFC 229",
    "UFC 259",
    "UFC 214",
    "UFC 330"
]

target_mask = pd.Series(
    False,
    index=test.index
)

for name in targets:

    target_mask |= (
        test["event_name"]
        .astype(str)
        .str.contains(
            name,
            case=False,
            na=False,
            regex=False
        )
    )

target = test[
    target_mask
].copy()


# ------------------------------------------------------------
# 6. MAIN COMPARISON
# ------------------------------------------------------------

print("\n" + "=" * 105)
print("🔥 TARGET CARDS — THREE RANKING WEIGHTS")
print("=" * 105)

display(
    target[
        [
            "event_name",
            "ranking_test_valid",

            "ranked_fighters_B",
            "ranked_matchups_B",
            "total_ranking_points_B",
            "competitive_depth_B",

            "card_quality_score",

            "event_strength_rank",
            "rank_B25",
            "rank_B40",
            "rank_B50",

            "rank_change_B25",
            "rank_change_B40",
            "rank_change_B50"
        ]
    ]
    .sort_values(
        "event_strength_rank"
    )
    .round(2)
)


# ------------------------------------------------------------
# 7. SHOW SCORE EFFECT TOO
# ------------------------------------------------------------

print("\n" + "=" * 105)
print("🔥 TARGET CARDS — SCORE CHANGES")
print("=" * 105)

display(
    target[
        [
            "event_name",
            "event_strength_score",
            "score_B25",
            "score_B40",
            "score_B50",
            "card_quality_score",
            "card_quality_B25",
            "card_quality_B40",
            "card_quality_B50"
        ]
    ]
    .sort_values(
        "event_strength_rank"
    )
    .round(2)
)


# ------------------------------------------------------------
# 8. RANKING-ERA RISERS/FALLERS ONLY
#
# This avoids the garbage pre-2013 comparisons from Model B v1.
# ------------------------------------------------------------

valid_only = test[
    test["ranking_test_valid"]
].copy()

print("\n" + "=" * 105)
print("📈 BIGGEST RISERS — B40 — VALID RANKING ERA ONLY")
print("=" * 105)

display(
    valid_only[
        [
            "event_name",
            "ranked_fighters_B",
            "ranked_matchups_B",
            "competitive_depth_B",
            "event_strength_rank",
            "rank_B40",
            "rank_change_B40"
        ]
    ]
    .sort_values(
        "rank_change_B40",
        ascending=False
    )
    .head(15)
    .round(2)
)


print("\n" + "=" * 105)
print("📉 BIGGEST FALLERS — B40 — VALID RANKING ERA ONLY")
print("=" * 105)

display(
    valid_only[
        [
            "event_name",
            "ranked_fighters_B",
            "ranked_matchups_B",
            "competitive_depth_B",
            "event_strength_rank",
            "rank_B40",
            "rank_change_B40"
        ]
    ]
    .sort_values(
        "rank_change_B40",
        ascending=True
    )
    .head(15)
    .round(2)
)


# ------------------------------------------------------------
# 9. UFC 300 vs UFC 302 DIRECT COMPARISON
# ------------------------------------------------------------

print("\n" + "=" * 105)
print("🥊 UFC 300 vs UFC 302 — DEPTH CHECK")
print("=" * 105)

comparison_mask = (
    test["event_name"]
    .astype(str)
    .str.contains(
        "UFC 300|UFC 302",
        case=False,
        na=False,
        regex=True
    )
)

display(
    test.loc[
        comparison_mask,
        [
            "event_name",
            "ranked_fighters_B",
            "ranked_matchups_B",
            "total_ranking_points_B",
            "title_fights_B",
            "competitive_depth_B",
            "event_strength_rank",
            "rank_B25",
            "rank_B40",
            "rank_B50"
        ]
    ]
    .round(2)
)


# ------------------------------------------------------------
# 10. SAFETY + SANITY CHECK
# ------------------------------------------------------------

pre_rank = test[
    test["event_date"] < RANKING_START
]

pre_rank_changed = (
    pre_rank["card_quality_B40"]
    -
    pre_rank["card_quality_score"]
).abs().max()

print("\n" + "=" * 105)
print("✅ SAFETY / SANITY CHECK")
print("=" * 105)

print("Finished model rows:", len(final_index))
print("Test rows:", len(test))

print(
    "Largest pre-2013 Card Quality change:",
    pre_rank_changed
)

print(
    "Expected:",
    0.0
)

print(
    "Original final_index modified: NO"
)

print(
    "Files overwritten: NO"
)

print(
    "\nB25 = mild ranking emphasis"
)

print(
    "B40 = strong ranking emphasis"
)

print(
    "B50 = aggressive ranking emphasis"
)

print(
    "\n🔥 MODEL B v2 COMPLETE"
)
```

```python
# ============================================================
# MODEL B v3 — CLEAN RANKING DEPTH INDEX
# SANDBOX ONLY — ORIGINAL MODEL REMAINS UNTOUCHED
# ============================================================

import pandas as pd
import numpy as np

print("=" * 105)
print("🔥 MODEL B v3 — CLEAN RANKING DEPTH INDEX")
print("=" * 105)

v3 = test.copy()

valid = v3["ranking_test_valid"]


# ------------------------------------------------------------
# 1. BUILD CLEAN RANKING DEPTH INDEX
# ------------------------------------------------------------

v3["ranking_depth_index"] = np.nan

v3.loc[valid, "ranking_depth_index"] = (
    0.45 * v3.loc[valid, "total_ranking_points_B_score"]
    +
    0.25 * v3.loc[valid, "ranked_fighters_B_score"]
    +
    0.20 * v3.loc[valid, "ranked_matchups_B_score"]
    +
    0.10 * v3.loc[valid, "title_fights_B_score"]
)


# ------------------------------------------------------------
# 2. TEST 25%, 30%, 35%, 40%
# OF CARD QUALITY
#
# IMPORTANT:
# We're testing sensitivity — not selecting a winner yet.
# ------------------------------------------------------------

DEPTH_WEIGHTS = [25, 30, 35, 40]

for pct in DEPTH_WEIGHTS:

    w = pct / 100

    card_col = f"card_v3_{pct}"

    # Default = original model
    v3[card_col] = v3["card_quality_score"]

    # Only ranking-era events get changed
    v3.loc[valid, card_col] = (
        (1 - w)
        * v3.loc[valid, "card_quality_score"]
        +
        w
        * v3.loc[valid, "ranking_depth_index"]
    )


# ------------------------------------------------------------
# 3. FINAL SCORE FUNCTION
# SAME 37 / 23 / 28 / 12 MODEL
# ------------------------------------------------------------

def final_v3(row, card_col):

    weights = {
        card_col: 0.37,
        "star_experience_score": 0.23,
        "main_event_score": 0.28,
        "fanfare_score": 0.12
    }

    available = {
        c: w
        for c, w in weights.items()
        if pd.notna(row[c])
    }

    if not available:
        return np.nan

    total = sum(available.values())

    return sum(
        row[c] * w
        for c, w in available.items()
    ) / total


# ------------------------------------------------------------
# 4. CALCULATE SCORES + RANKS
# ------------------------------------------------------------

for pct in DEPTH_WEIGHTS:

    card_col = f"card_v3_{pct}"
    score_col = f"score_v3_{pct}"
    rank_col = f"rank_v3_{pct}"
    change_col = f"change_v3_{pct}"

    v3[score_col] = v3.apply(
        lambda r:
        final_v3(
            r,
            card_col
        ),
        axis=1
    )

    v3[rank_col] = (
        v3[score_col]
        .rank(
            ascending=False,
            method="min"
        )
        .astype("Int64")
    )

    v3[change_col] = (
        v3["event_strength_rank"]
        -
        v3[rank_col]
    )


# ------------------------------------------------------------
# 5. TARGET CARDS
# ------------------------------------------------------------

targets = [
    "UFC 269",
    "UFC 214",
    "UFC 302",
    "UFC 205",
    "UFC 229",
    "UFC 259",
    "UFC 300"
]

mask = pd.Series(
    False,
    index=v3.index
)

for name in targets:

    mask |= (
        v3["event_name"]
        .astype(str)
        .str.contains(
            name,
            case=False,
            na=False,
            regex=False
        )
    )

targets_v3 = v3[mask].copy()


print("\n" + "=" * 105)
print("🔥 TARGET CARDS")
print("=" * 105)

display(
    targets_v3[
        [
            "event_name",

            "ranked_fighters_B",
            "ranked_matchups_B",
            "total_ranking_points_B",
            "title_fights_B",

            "ranking_depth_index",

            "event_strength_rank",

            "rank_v3_25",
            "rank_v3_30",
            "rank_v3_35",
            "rank_v3_40"
        ]
    ]
    .sort_values(
        "event_strength_rank"
    )
    .round(2)
)


# ------------------------------------------------------------
# 6. TOP 20 AT EACH WEIGHT
# ------------------------------------------------------------

for pct in DEPTH_WEIGHTS:

    print(
        "\n" + "=" * 105
    )

    print(
        f"🔥 TOP 20 — {pct}% RANKING DEPTH"
    )

    print(
        "=" * 105
    )

    display(
        v3[
            [
                "event_name",
                "ranking_depth_index",
                "event_strength_rank",
                f"rank_v3_{pct}",
                f"score_v3_{pct}",
                f"change_v3_{pct}"
            ]
        ]
        .sort_values(
            f"rank_v3_{pct}"
        )
        .head(20)
        .round(2)
    )


# ------------------------------------------------------------
# 7. SPECIFIC UFC 300 DIAGNOSTIC
#
# We want to know WHY it lands where it lands,
# not force it to a predetermined position.
# ------------------------------------------------------------

ufc300 = v3[
    v3["event_name"]
    .astype(str)
    .str.contains(
        "UFC 300:",
        case=False,
        na=False,
        regex=False
    )
].copy()

print("\n" + "=" * 105)
print("🥊 UFC 300 DIAGNOSTIC")
print("=" * 105)

display(
    ufc300[
        [
            "event_name",

            "card_quality_score",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",

            "ranking_depth_index",

            "card_v3_25",
            "card_v3_30",
            "card_v3_35",
            "card_v3_40",

            "event_strength_score",

            "score_v3_25",
            "score_v3_30",
            "score_v3_35",
            "score_v3_40",

            "event_strength_rank",

            "rank_v3_25",
            "rank_v3_30",
            "rank_v3_35",
            "rank_v3_40"
        ]
    ]
    .round(2)
)


# ------------------------------------------------------------
# 8. SAFETY CHECK
# ------------------------------------------------------------

pre2013 = (
    v3["event_date"]
    <
    pd.Timestamp("2013-02-04")
)

max_change = (
    v3.loc[
        pre2013,
        "card_v3_40"
    ]
    -
    v3.loc[
        pre2013,
        "card_quality_score"
    ]
).abs().max()


print("\n" + "=" * 105)
print("✅ SAFETY CHECK")
print("=" * 105)

print(
    "Pre-2013 max Card Quality change:",
    max_change
)

print("Expected: 0.0")

print(
    "Original final_index modified: NO"
)

print(
    "Files overwritten: NO"
)

print(
    "\n🔥 MODEL B v3 COMPLETE"
)
```

```python
# ============================================================
# MAIN EVENT STRENGTH DIAGNOSTIC
# SANDBOX ONLY — NOTHING IN FINAL MODEL IS MODIFIED
# ============================================================

import pandas as pd
import numpy as np

print("=" * 105)
print("🔥 MAIN EVENT STRENGTH DIAGNOSTIC")
print("=" * 105)

# ------------------------------------------------------------
# 1. TARGET EVENTS
# ------------------------------------------------------------

TARGETS = [
    "UFC 300: Pereira vs. Hill",
    "UFC 302: Makhachev vs. Poirier",
    "UFC 269: Oliveira vs. Poirier",
    "UFC 229: Khabib vs. McGregor",
    "UFC 205: Alvarez vs McGregor",
    "UFC 259: Blachowicz vs. Adesanya",
    "UFC 214: Cormier vs. Jones 2"
]

# ------------------------------------------------------------
# 2. FIND MAIN EVENT FIGHT
#
# Historical UFCStats/TidyTuesday data normally stores bouts
# in card order, but we DO NOT want to assume row order.
#
# We identify the two names from the event title and match
# them to the fight card.
# ------------------------------------------------------------

def normalize_name(x):
    return (
        str(x)
        .lower()
        .replace(".", "")
        .replace("'", "")
        .replace("-", " ")
        .strip()
    )

diagnostic_rows = []

for event in TARGETS:

    event_rows = fight_test[
        fight_test["event_name"] == event
    ].copy()

    if event_rows.empty:
        print("⚠️ Event not found:", event)
        continue

    # Pull headliner surnames from event title
    if ":" in event:
        headliner_text = event.split(":", 1)[1].strip()
    else:
        headliner_text = event

    # Manually robust enough for these selected events
    parts = (
        headliner_text
        .replace(" 2", "")
        .replace(" 3", "")
        .split(" vs")
    )

    expected = [
        normalize_name(x).split()[-1]
        for x in parts[:2]
    ]

    main_candidates = []

    for idx, row in event_rows.iterrows():

        f1 = normalize_name(row["f1_name"])
        f2 = normalize_name(row["f2_name"])

        names = f1 + " " + f2

        if all(
            surname in names
            for surname in expected
        ):
            main_candidates.append(idx)

    if len(main_candidates) != 1:

        print(
            f"⚠️ Could not uniquely identify main event: "
            f"{event} | matches={len(main_candidates)}"
        )
        continue

    row = event_rows.loc[
        main_candidates[0]
    ].copy()

    diagnostic_rows.append(row)


main_diag = pd.DataFrame(
    diagnostic_rows
).reset_index(drop=True)

print(
    "\nMain events identified:",
    len(main_diag),
    "/",
    len(TARGETS)
)

display(
    main_diag[
        [
            "date",
            "event_name",
            "f1_name",
            "f2_name"
        ]
    ]
)


# ------------------------------------------------------------
# 3. HELPER — GET HISTORICAL FIGHTER APPEARANCE
#
# 'population' was created during our Nunes/Figueiredo
# historical fighter contribution analysis.
# ------------------------------------------------------------

pop = population.copy()

pop["date"] = pd.to_datetime(
    pop["date"]
)

# Find fighter-name column robustly
fighter_col = None

for candidate in [
    "fighter",
    "fighter_name",
    "name"
]:
    if candidate in pop.columns:
        fighter_col = candidate
        break

if fighter_col is None:
    raise ValueError(
        "Could not identify fighter column in population."
    )


# ------------------------------------------------------------
# 4. FIND OUR PRE-FIGHT METRIC COLUMNS
# ------------------------------------------------------------

def first_existing(df, candidates):

    for c in candidates:
        if c in df.columns:
            return c

    return None


metric_candidates = {

    "ufc_fights": [
        "ufc_fights_before",
        "ufc_fights"
    ],

    "ufc_win_pct": [
        "ufc_win_pct_before",
        "ufc_win_pct"
    ],

    "win_streak": [
        "win_streak_before",
        "win_streak"
    ],

    "previous_title_fights": [
        "previous_title_fights",
        "title_fights_before"
    ],

    "previous_main_events": [
        "previous_main_events",
        "main_events_before"
    ],

    "previous_bonuses": [
        "previous_total_bonuses",
        "total_bonuses_before"
    ]
}


found_cols = {
    metric:
    first_existing(
        pop,
        candidates
    )
    for metric, candidates
    in metric_candidates.items()
}

print("\nHistorical columns found:")

for k, v in found_cols.items():
    print(f"{k:25s}: {v}")


# ------------------------------------------------------------
# 5. CREATE FIGHTER-LEVEL MAIN EVENT TABLE
# ------------------------------------------------------------

fighter_rows = []

for _, fight in main_diag.iterrows():

    for corner in ["f1", "f2"]:

        fighter = fight[
            f"{corner}_name"
        ]

        date = pd.Timestamp(
            fight["date"]
        )

        history = pop[
            (pop[fighter_col] == fighter)
            &
            (pop["date"] == date)
        ]

        if history.empty:

            print(
                "⚠️ Fighter appearance not found:",
                fighter,
                date.date()
            )
            continue

        r = history.iloc[0]

        out = {
            "event_name": fight["event_name"],
            "date": date,
            "fighter": fighter
        }

        for metric, col in found_cols.items():

            out[metric] = (
                r[col]
                if col is not None
                else np.nan
            )

        # Use ranking values already reconstructed
        if corner == "f1":
            out["rank"] = fight[
                "f1_rank_test"
            ]
        else:
            out["rank"] = fight[
                "f2_rank_test"
            ]

        out["ranking_points"] = (
            rank_points_B(
                out["rank"]
            )
        )

        fighter_rows.append(out)


fighters_main = pd.DataFrame(
    fighter_rows
)


# ------------------------------------------------------------
# 6. SHOW RAW MAIN-EVENT RÉSUMÉS
# ------------------------------------------------------------

print("\n" + "=" * 105)
print("🥊 RAW HEADLINER DATA")
print("=" * 105)

display(
    fighters_main[
        [
            "event_name",
            "fighter",
            "rank",
            "ranking_points",
            "ufc_fights",
            "ufc_win_pct",
            "win_streak",
            "previous_title_fights",
            "previous_main_events",
            "previous_bonuses"
        ]
    ].round(2)
)


# ------------------------------------------------------------
# 7. AGGREGATE EACH MAIN EVENT
# ------------------------------------------------------------

main_summary = (
    fighters_main
    .groupby(
        "event_name",
        as_index=False
    )
    .agg(
        combined_ufc_fights=(
            "ufc_fights",
            "sum"
        ),

        avg_win_pct=(
            "ufc_win_pct",
            "mean"
        ),

        combined_win_streak=(
            "win_streak",
            "sum"
        ),

        combined_title_fights=(
            "previous_title_fights",
            "sum"
        ),

        combined_previous_main_events=(
            "previous_main_events",
            "sum"
        ),

        combined_bonuses=(
            "previous_bonuses",
            "sum"
        ),

        combined_ranking_points=(
            "ranking_points",
            "sum"
        )
    )
)


# ------------------------------------------------------------
# 8. ADD ORIGINAL MAIN EVENT SCORE
# ------------------------------------------------------------

main_summary = main_summary.merge(
    final_index[
        [
            "event_name",
            "main_event_score",
            "event_strength_rank"
        ]
    ],
    on="event_name",
    how="left"
)


# ------------------------------------------------------------
# 9. DISPLAY DIAGNOSTIC
# ------------------------------------------------------------

print("\n" + "=" * 105)
print("🔥 WHY EACH MAIN EVENT SCORES DIFFERENTLY")
print("=" * 105)

display(
    main_summary[
        [
            "event_name",
            "combined_ufc_fights",
            "avg_win_pct",
            "combined_win_streak",
            "combined_title_fights",
            "combined_previous_main_events",
            "combined_bonuses",
            "combined_ranking_points",
            "main_event_score",
            "event_strength_rank"
        ]
    ]
    .sort_values(
        "main_event_score",
        ascending=False
    )
    .round(2)
)


# ------------------------------------------------------------
# 10. UFC 300 SPECIFIC
# ------------------------------------------------------------

print("\n" + "=" * 105)
print("🥊 UFC 300 — PEREIRA vs HILL")
print("=" * 105)

display(
    main_summary[
        main_summary["event_name"]
        .str.contains(
            "UFC 300:",
            case=False,
            na=False,
            regex=False
        )
    ].T
)


# ------------------------------------------------------------
# 11. SAFETY
# ------------------------------------------------------------

print("\n" + "=" * 105)
print("✅ SAFETY CHECK")
print("=" * 105)

print("final_index modified: NO")
print("v3 modified: NO")
print("Files overwritten: NO")
print("This cell only created diagnostic tables.")

print("\n🔥 Send me RAW HEADLINER DATA + WHY EACH MAIN EVENT SCORES DIFFERENTLY")
```

```python
# ============================================================
# MAIN EVENT v2 — REBALANCED MAIN EVENT STRENGTH
# SANDBOX ONLY
#
# 40% CURRENT ELITE STRENGTH
# 40% CHAMPIONSHIP / STAR RESUME
# 20% UFC EXPERIENCE
#
# DOES NOT MODIFY FINAL MODEL OR FILES
# ============================================================

import pandas as pd
import numpy as np

print("=" * 110)
print("🔥 MAIN EVENT v2 — REBALANCED MAIN EVENT STRENGTH")
print("=" * 110)


# ------------------------------------------------------------
# 1. WORK ON A COPY ONLY
# ------------------------------------------------------------

me2_fights = fight_test.copy()

me2_fights["date"] = pd.to_datetime(
    me2_fights["date"]
)

me2_fights = me2_fights.sort_values(
    ["date", "event_name"]
).reset_index(drop=True)


# ------------------------------------------------------------
# 2. RECONSTRUCT VERIFIED MAIN EVENTS
#
# Prefer existing verified main-event column if available.
# ------------------------------------------------------------

if "is_main_event_verified" in me2_fights.columns:

    me2_fights["ME_FLAG"] = (
        me2_fights["is_main_event_verified"]
        .fillna(False)
        .astype(bool)
    )

elif "is_main_event" in me2_fights.columns:

    me2_fights["ME_FLAG"] = (
        me2_fights["is_main_event"]
        .fillna(False)
        .astype(bool)
    )

else:

    raise ValueError(
        "No verified main-event indicator found in fight_test."
    )


print(
    "\nVerified main-event fights found:",
    me2_fights["ME_FLAG"].sum()
)


# ------------------------------------------------------------
# 3. BONUS FLAG
#
# A fighter receives one prior-bonus count for each bonus
# associated with a previous UFC appearance.
#
# FOTN counts for both fighters.
# POTN/KOTN/SOTN are winner/performance bonuses.
#
# We first create fight-level total bonus awards.
# ------------------------------------------------------------

bonus_cols = [
    c for c in
    ["FOTN", "POTN", "KOTN", "SOTN"]
    if c in me2_fights.columns
]

print(
    "Bonus columns found:",
    bonus_cols
)

if len(bonus_cols) == 0:

    raise ValueError(
        "No historical bonus columns found in fight_test."
    )


# ------------------------------------------------------------
# 4. BUILD LONG FIGHTER-APPEARANCE TABLE
# ------------------------------------------------------------

long_rows = []

for idx, row in me2_fights.iterrows():

    for corner in ["f1", "f2"]:

        fighter = row[f"{corner}_name"]

        if pd.isna(fighter):
            continue

        # --------------------------------------------
        # Determine fighter's bonus awards THIS fight
        # --------------------------------------------

        bonus_count = 0

        # FOTN belongs to both fighters
        if (
            "FOTN" in me2_fights.columns
            and row["FOTN"] == 1
        ):
            bonus_count += 1

        # Performance-style bonuses belong to winner.
        # Identify whether this corner won.
        result = str(
            row.get(
                f"{corner}_result",
                ""
            )
        ).upper()

        fighter_won = (
            result == "W"
            or result == "WIN"
        )

        if fighter_won:

            for b in ["POTN", "KOTN", "SOTN"]:

                if (
                    b in me2_fights.columns
                    and row[b] == 1
                ):
                    bonus_count += 1

        long_rows.append(
            {
                "fight_index": idx,
                "date": row["date"],
                "event_name": row["event_name"],
                "fighter": fighter,
                "corner": corner,
                "is_main_event": int(
                    row["ME_FLAG"]
                ),
                "bonuses_this_fight": bonus_count
            }
        )


fighter_history = pd.DataFrame(
    long_rows
)

fighter_history = fighter_history.sort_values(
    [
        "fighter",
        "date",
        "fight_index"
    ]
).reset_index(drop=True)


# ------------------------------------------------------------
# 5. CALCULATE ENTERING-FIGHT MAIN EVENTS + BONUSES
# ------------------------------------------------------------

fighter_history[
    "previous_main_events_rebuilt"
] = (
    fighter_history
    .groupby("fighter")["is_main_event"]
    .cumsum()
    -
    fighter_history["is_main_event"]
)


fighter_history[
    "previous_bonuses_rebuilt"
] = (
    fighter_history
    .groupby("fighter")["bonuses_this_fight"]
    .cumsum()
    -
    fighter_history["bonuses_this_fight"]
)


print(
    "\nFighter appearances rebuilt:",
    len(fighter_history)
)

print(
    "Max previous main events:",
    fighter_history[
        "previous_main_events_rebuilt"
    ].max()
)

print(
    "Max previous bonuses:",
    fighter_history[
        "previous_bonuses_rebuilt"
    ].max()
)


# ------------------------------------------------------------
# 6. ADD THE EXISTING PRE-FIGHT METRICS FROM population
# ------------------------------------------------------------

pop2 = population.copy()

pop2["date"] = pd.to_datetime(
    pop2["date"]
)

fighter_col = None

for c in [
    "fighter",
    "fighter_name",
    "name"
]:
    if c in pop2.columns:
        fighter_col = c
        break

if fighter_col is None:

    raise ValueError(
        "Could not find fighter-name column in population."
    )


# ------------------------------------------------------------
# 7. CREATE HISTORICAL MAIN-EVENT FIGHTER POPULATION
# ------------------------------------------------------------

main_fights = me2_fights[
    me2_fights["ME_FLAG"]
].copy()


me_fighter_rows = []

for idx, fight in main_fights.iterrows():

    for corner in ["f1", "f2"]:

        fighter = fight[f"{corner}_name"]
        date = pd.Timestamp(fight["date"])

        # --------------------------------------------
        # Base pre-fight metrics
        # --------------------------------------------

        p = pop2[
            (pop2[fighter_col] == fighter)
            &
            (pop2["date"] == date)
        ]

        if p.empty:
            continue

        p = p.iloc[0]

        # --------------------------------------------
        # Rebuilt historical main-event/bonus counts
        # --------------------------------------------

        h = fighter_history[
            (fighter_history["fighter"] == fighter)
            &
            (fighter_history["date"] == date)
            &
            (fighter_history["event_name"] == fight["event_name"])
        ]

        if h.empty:

            prev_me = np.nan
            prev_bonus = np.nan

        else:

            h = h.iloc[0]

            prev_me = h[
                "previous_main_events_rebuilt"
            ]

            prev_bonus = h[
                "previous_bonuses_rebuilt"
            ]

        # --------------------------------------------
        # Ranking
        # --------------------------------------------

        if corner == "f1":
            rank = fight.get(
                "f1_rank_test",
                np.nan
            )
        else:
            rank = fight.get(
                "f2_rank_test",
                np.nan
            )

        rp = rank_points_B(rank)

        me_fighter_rows.append(
            {
                "event_name":
                    fight["event_name"],

                "date":
                    date,

                "fighter":
                    fighter,

                "ufc_fights":
                    p["ufc_fights"],

                "ufc_win_pct":
                    p["ufc_win_pct"],

                "win_streak":
                    p["win_streak"],

                "previous_title_fights":
                    p["previous_title_fights"],

                "previous_main_events":
                    prev_me,

                "previous_bonuses":
                    prev_bonus,

                "rank":
                    rank,

                "ranking_points":
                    rp
            }
        )


me_population = pd.DataFrame(
    me_fighter_rows
)

print(
    "\nHistorical main-event fighter appearances:",
    len(me_population)
)


# ------------------------------------------------------------
# 8. AGGREGATE TO MAIN-EVENT LEVEL
# ------------------------------------------------------------

me_events = (
    me_population
    .groupby(
        ["event_name", "date"],
        as_index=False
    )
    .agg(

        combined_ufc_fights=(
            "ufc_fights",
            "sum"
        ),

        avg_win_pct=(
            "ufc_win_pct",
            "mean"
        ),

        combined_win_streak=(
            "win_streak",
            "sum"
        ),

        combined_title_fights=(
            "previous_title_fights",
            "sum"
        ),

        combined_previous_main_events=(
            "previous_main_events",
            "sum"
        ),

        combined_bonuses=(
            "previous_bonuses",
            "sum"
        ),

        combined_ranking_points=(
            "ranking_points",
            "sum"
        )
    )
)


# ------------------------------------------------------------
# 9. ZERO-PRESERVING PERCENTILE
# SAME PHILOSOPHY AS ORIGINAL MODEL
# ------------------------------------------------------------

def zp(series):

    result = pd.Series(
        index=series.index,
        dtype=float
    )

    result[series.isna()] = np.nan
    result[series == 0] = 0

    positive = series > 0

    if positive.sum() > 0:

        result.loc[positive] = (
            series.loc[positive]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )

    return result


# ------------------------------------------------------------
# 10. NORMALIZE EACH MAIN-EVENT INPUT
# ------------------------------------------------------------

raw_metrics = [
    "combined_ufc_fights",
    "avg_win_pct",
    "combined_win_streak",
    "combined_title_fights",
    "combined_previous_main_events",
    "combined_bonuses",
    "combined_ranking_points"
]

for metric in raw_metrics:

    me_events[
        metric + "_score"
    ] = zp(
        me_events[metric]
    )


# ------------------------------------------------------------
# 11. THREE CONCEPTUAL GROUPS
# ------------------------------------------------------------

# CURRENT ELITE STRENGTH
# Ranking + win rate + active streak

me_events[
    "current_elite_strength"
] = me_events[
    [
        "combined_ranking_points_score",
        "avg_win_pct_score",
        "combined_win_streak_score"
    ]
].mean(
    axis=1,
    skipna=True
)


# CHAMPIONSHIP / STAR RESUME
# Title-fight history + main-event history + bonuses

me_events[
    "championship_star_resume"
] = me_events[
    [
        "combined_title_fights_score",
        "combined_previous_main_events_score",
        "combined_bonuses_score"
    ]
].mean(
    axis=1,
    skipna=True
)


# UFC EXPERIENCE

me_events[
    "ufc_experience"
] = me_events[
    "combined_ufc_fights_score"
]


# ------------------------------------------------------------
# 12. MAIN EVENT v2
#
# 40% current elite
# 40% championship/star resume
# 20% UFC experience
# ------------------------------------------------------------

me_events[
    "main_event_v2"
] = (
    0.40
    * me_events[
        "current_elite_strength"
    ]
    +
    0.40
    * me_events[
        "championship_star_resume"
    ]
    +
    0.20
    * me_events[
        "ufc_experience"
    ]
)


# ------------------------------------------------------------
# 13. MERGE WITH ORIGINAL MODEL
# ------------------------------------------------------------

me_compare = me_events.merge(
    final_index[
        [
            "event_name",
            "main_event_score",
            "event_strength_score",
            "event_strength_rank"
        ]
    ],
    on="event_name",
    how="left"
)


# ------------------------------------------------------------
# 14. TARGET EVENTS
# ------------------------------------------------------------

TARGETS = [
    "UFC 300",
    "UFC 302",
    "UFC 269",
    "UFC 229",
    "UFC 205",
    "UFC 259",
    "UFC 214"
]

target_mask = pd.Series(
    False,
    index=me_compare.index
)

for name in TARGETS:

    target_mask |= (
        me_compare["event_name"]
        .astype(str)
        .str.contains(
            name,
            case=False,
            na=False,
            regex=False
        )
    )


target_me = me_compare[
    target_mask
].copy()


# ------------------------------------------------------------
# 15. SHOW THE RAW RÉSUMÉS
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🥊 TARGET MAIN EVENTS — RAW INPUTS")
print("=" * 110)

display(
    target_me[
        [
            "event_name",

            "combined_ufc_fights",
            "avg_win_pct",
            "combined_win_streak",

            "combined_title_fights",
            "combined_previous_main_events",
            "combined_bonuses",

            "combined_ranking_points"
        ]
    ]
    .round(2)
)


# ------------------------------------------------------------
# 16. SHOW GROUP SCORES + OLD VS NEW
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🔥 MAIN EVENT v1 vs MAIN EVENT v2")
print("=" * 110)

display(
    target_me[
        [
            "event_name",

            "current_elite_strength",
            "championship_star_resume",
            "ufc_experience",

            "main_event_score",
            "main_event_v2"
        ]
    ]
    .sort_values(
        "main_event_v2",
        ascending=False
    )
    .round(2)
)


# ------------------------------------------------------------
# 17. UFC 300 SPECIFIC
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🔥 UFC 300 MAIN EVENT DIAGNOSTIC")
print("=" * 110)

display(
    target_me[
        target_me["event_name"]
        .astype(str)
        .str.contains(
            "UFC 300:",
            case=False,
            na=False,
            regex=False
        )
    ][
        [
            "event_name",

            "combined_ufc_fights_score",
            "avg_win_pct_score",
            "combined_win_streak_score",

            "combined_title_fights_score",
            "combined_previous_main_events_score",
            "combined_bonuses_score",

            "combined_ranking_points_score",

            "current_elite_strength",
            "championship_star_resume",
            "ufc_experience",

            "main_event_score",
            "main_event_v2"
        ]
    ].T
)


# ------------------------------------------------------------
# 18. PLUG v2 INTO EVENT MODEL
#
# IMPORTANT:
# Keep our B40 ranking-depth Card Quality test.
# Keep Star Experience.
# Keep Fanfare.
# Keep overall weights 37 / 23 / 28 / 12.
#
# ONLY substitute Main Event v2 where available.
# ------------------------------------------------------------

sandbox_v2 = v3.copy()

sandbox_v2 = sandbox_v2.merge(
    me_events[
        [
            "event_name",
            "main_event_v2"
        ]
    ],
    on="event_name",
    how="left"
)


# For events where v2 cannot be reconstructed,
# preserve original Main Event score.

sandbox_v2[
    "main_event_v2_used"
] = sandbox_v2[
    "main_event_v2"
].combine_first(
    sandbox_v2[
        "main_event_score"
    ]
)


def overall_v2(row):

    weights = {
        "card_v3_40": 0.37,
        "star_experience_score": 0.23,
        "main_event_v2_used": 0.28,
        "fanfare_score": 0.12
    }

    available = {
        c: w
        for c, w in weights.items()
        if pd.notna(row[c])
    }

    if not available:
        return np.nan

    total = sum(
        available.values()
    )

    return sum(
        row[c] * w
        for c, w in available.items()
    ) / total


sandbox_v2[
    "event_strength_v2"
] = sandbox_v2.apply(
    overall_v2,
    axis=1
)


sandbox_v2[
    "rank_v2"
] = (
    sandbox_v2[
        "event_strength_v2"
    ]
    .rank(
        ascending=False,
        method="min"
    )
    .astype("Int64")
)


sandbox_v2[
    "rank_change_v2"
] = (
    sandbox_v2[
        "event_strength_rank"
    ]
    -
    sandbox_v2[
        "rank_v2"
    ]
)


# ------------------------------------------------------------
# 19. TARGET EVENT RESULTS
# ------------------------------------------------------------

target_event_mask = pd.Series(
    False,
    index=sandbox_v2.index
)

for name in TARGETS:

    target_event_mask |= (
        sandbox_v2[
            "event_name"
        ]
        .astype(str)
        .str.contains(
            name,
            case=False,
            na=False,
            regex=False
        )
    )


print("\n" + "=" * 110)
print("🏆 EVENT STRENGTH — ORIGINAL vs RANKING B40 + MAIN EVENT v2")
print("=" * 110)

display(
    sandbox_v2.loc[
        target_event_mask,
        [
            "event_name",

            "event_strength_score",
            "event_strength_rank",

            "card_v3_40",

            "main_event_score",
            "main_event_v2_used",

            "event_strength_v2",
            "rank_v2",
            "rank_change_v2"
        ]
    ]
    .sort_values(
        "rank_v2"
    )
    .round(2)
)


# ------------------------------------------------------------
# 20. NEW TOP 20
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🔥 TOP 20 — SANDBOX MODEL")
print("=" * 110)

display(
    sandbox_v2[
        [
            "event_name",
            "event_strength_score",
            "event_strength_rank",
            "event_strength_v2",
            "rank_v2",
            "rank_change_v2"
        ]
    ]
    .sort_values(
        "rank_v2"
    )
    .head(20)
    .round(2)
)


# ------------------------------------------------------------
# 21. SAFETY
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("✅ SAFETY CHECK")
print("=" * 110)

print("Original final_index modified: NO")
print("Original v3 modified: NO")
print("Files overwritten: NO")
print("Only new sandbox objects were created.")

print("\n🔥 MAIN EVENT v2 TEST COMPLETE")
```

```python
# ============================================================
# MODEL C — CARD QUALITY vs MAIN EVENT WEIGHT SENSITIVITY
# SANDBOX ONLY
#
# Card Quality = our B40 ranking-depth version
# Star Experience = unchanged
# Main Event Strength = ORIGINAL
# Fanfare = unchanged
#
# Tests:
# A = 37 / 23 / 28 / 12
# B = 40 / 23 / 25 / 12
# C = 43 / 23 / 22 / 12
# D = 46 / 23 / 19 / 12
# ============================================================

import pandas as pd
import numpy as np

print("=" * 110)
print("🔥 MODEL C — CARD QUALITY vs MAIN EVENT WEIGHT TEST")
print("=" * 110)

model_c = v3.copy()


# ------------------------------------------------------------
# 1. DEFINE SCENARIOS
# ------------------------------------------------------------

SCENARIOS = {
    "A_37_28": {
        "card": 0.37,
        "star": 0.23,
        "main": 0.28,
        "fanfare": 0.12
    },

    "B_40_25": {
        "card": 0.40,
        "star": 0.23,
        "main": 0.25,
        "fanfare": 0.12
    },

    "C_43_22": {
        "card": 0.43,
        "star": 0.23,
        "main": 0.22,
        "fanfare": 0.12
    },

    "D_46_19": {
        "card": 0.46,
        "star": 0.23,
        "main": 0.19,
        "fanfare": 0.12
    }
}


# ------------------------------------------------------------
# 2. DYNAMIC-MISSING SCORE FUNCTION
# Same missing-data philosophy as finished model
# ------------------------------------------------------------

def model_c_score(row, weights):

    components = {
        "card_v3_40": weights["card"],
        "star_experience_score": weights["star"],
        "main_event_score": weights["main"],
        "fanfare_score": weights["fanfare"]
    }

    available = {
        col: weight
        for col, weight in components.items()
        if pd.notna(row[col])
    }

    if not available:
        return np.nan

    total_weight = sum(
        available.values()
    )

    return sum(
        row[col] * weight
        for col, weight in available.items()
    ) / total_weight


# ------------------------------------------------------------
# 3. CALCULATE ALL FOUR SCENARIOS
# ------------------------------------------------------------

for name, weights in SCENARIOS.items():

    score_col = f"score_{name}"
    rank_col = f"rank_{name}"
    change_col = f"change_{name}"

    model_c[score_col] = model_c.apply(
        lambda row:
        model_c_score(
            row,
            weights
        ),
        axis=1
    )

    model_c[rank_col] = (
        model_c[score_col]
        .rank(
            ascending=False,
            method="min"
        )
        .astype("Int64")
    )

    model_c[change_col] = (
        model_c["event_strength_rank"]
        -
        model_c[rank_col]
    )


# ------------------------------------------------------------
# 4. TARGET EVENTS
# ------------------------------------------------------------

TARGETS = [
    "UFC 269",
    "UFC 168",
    "UFC 214",
    "UFC 217",
    "UFC 302",
    "UFC 205",
    "UFC 229",
    "UFC 259",
    "UFC 300",
    "UFC 100"
]

target_mask = pd.Series(
    False,
    index=model_c.index
)

for name in TARGETS:

    target_mask |= (
        model_c["event_name"]
        .astype(str)
        .str.contains(
            name,
            case=False,
            na=False,
            regex=False
        )
    )

target_c = model_c[
    target_mask
].copy()


# ------------------------------------------------------------
# 5. TARGET RANK COMPARISON
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🏆 TARGET EVENTS — RANK COMPARISON")
print("=" * 110)

display(
    target_c[
        [
            "event_name",

            "card_v3_40",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",

            "event_strength_rank",

            "rank_A_37_28",
            "rank_B_40_25",
            "rank_C_43_22",
            "rank_D_46_19"
        ]
    ]
    .sort_values(
        "event_strength_rank"
    )
    .round(2)
)


# ------------------------------------------------------------
# 6. TARGET SCORE COMPARISON
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🔥 TARGET EVENTS — SCORE COMPARISON")
print("=" * 110)

display(
    target_c[
        [
            "event_name",

            "event_strength_score",

            "score_A_37_28",
            "score_B_40_25",
            "score_C_43_22",
            "score_D_46_19"
        ]
    ]
    .sort_values(
        "event_strength_rank"
    )
    .round(2)
)


# ------------------------------------------------------------
# 7. UFC 300 SPECIFIC
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🥊 UFC 300 — WEIGHT SENSITIVITY")
print("=" * 110)

ufc300_c = model_c[
    model_c["event_name"]
    .astype(str)
    .str.contains(
        "UFC 300:",
        case=False,
        na=False,
        regex=False
    )
].copy()

display(
    ufc300_c[
        [
            "event_name",

            "card_v3_40",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",

            "event_strength_score",
            "event_strength_rank",

            "score_A_37_28",
            "rank_A_37_28",

            "score_B_40_25",
            "rank_B_40_25",

            "score_C_43_22",
            "rank_C_43_22",

            "score_D_46_19",
            "rank_D_46_19"
        ]
    ]
    .round(2)
)


# ------------------------------------------------------------
# 8. TOP 20 FOR EACH SCENARIO
# ------------------------------------------------------------

for name in SCENARIOS:

    print("\n" + "=" * 110)
    print(f"🔥 TOP 20 — {name}")
    print("=" * 110)

    display(
        model_c[
            [
                "event_name",
                "event_strength_rank",
                f"score_{name}",
                f"rank_{name}",
                f"change_{name}"
            ]
        ]
        .sort_values(
            f"rank_{name}"
        )
        .head(20)
        .round(2)
    )


# ------------------------------------------------------------
# 9. BIGGEST MOVERS — MIDDLE SCENARIO C
#
# C = 43 Card / 23 Star / 22 Main / 12 Fanfare
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("📈 BIGGEST RISERS — SCENARIO C")
print("=" * 110)

display(
    model_c[
        [
            "event_name",
            "card_v3_40",
            "main_event_score",
            "event_strength_rank",
            "rank_C_43_22",
            "change_C_43_22"
        ]
    ]
    .sort_values(
        "change_C_43_22",
        ascending=False
    )
    .head(15)
    .round(2)
)


print("\n" + "=" * 110)
print("📉 BIGGEST FALLERS — SCENARIO C")
print("=" * 110)

display(
    model_c[
        [
            "event_name",
            "card_v3_40",
            "main_event_score",
            "event_strength_rank",
            "rank_C_43_22",
            "change_C_43_22"
        ]
    ]
    .sort_values(
        "change_C_43_22",
        ascending=True
    )
    .head(15)
    .round(2)
)


# ------------------------------------------------------------
# 10. CORRELATION WITH ORIGINAL MODEL
#
# Helps us see whether we're improving emphasis or
# completely rewriting the model.
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("📊 CORRELATION WITH ORIGINAL EVENT STRENGTH")
print("=" * 110)

for name in SCENARIOS:

    corr = model_c[
        [
            "event_strength_score",
            f"score_{name}"
        ]
    ].corr().iloc[0, 1]

    print(
        f"{name}: {corr:.4f}"
    )


# ------------------------------------------------------------
# 11. SAFETY
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("✅ SAFETY CHECK")
print("=" * 110)

print("Original final_index modified: NO")
print("Original v3 modified: NO")
print("Main Event formula modified: NO")
print("Files overwritten: NO")
print("Only model_c was created.")

print("\n🔥 MODEL C TEST COMPLETE")
```

```python
# ============================================================
# CLEAN CARD QUALITY v1
# Replace old ranking input instead of layering on top of it
# SANDBOX ONLY
# ============================================================

import pandas as pd
import numpy as np

print("=" * 110)
print("🔥 CLEAN CARD QUALITY v1 — NO RANKING DOUBLE COUNT")
print("=" * 110)


# ------------------------------------------------------------
# 1. FIND THE EVENT-LEVEL TABLE WITH THE ORIGINAL CQ SUBSCORES
# ------------------------------------------------------------

candidate_objects = {
    "event_master": globals().get("event_master"),
    "event_master_updated": globals().get("event_master_updated"),
    "final_event_strength": globals().get("final_event_strength"),
    "v3": globals().get("v3"),
    "model_d": globals().get("model_d"),
}

needed_concepts = {
    "title": ["title_fights_score"],
    "experience": ["avg_ufc_experience_score"],
    "win_pct": ["avg_ufc_win_pct_score"],
    "streak": ["avg_win_streak_score"],
    "finish": ["avg_finish_rate_score"],
}

source = None
source_name = None

for name, obj in candidate_objects.items():

    if not isinstance(obj, pd.DataFrame):
        continue

    cols = set(obj.columns)

    if all(
        any(c in cols for c in options)
        for options in needed_concepts.values()
    ):
        source = obj.copy()
        source_name = name
        break


print("Subscore source:", source_name)

if source is None:

    print("\n❌ STOP")
    print("The five original Card Quality subscores are not currently available.")
    print("Available likely score columns:")

    for name, obj in candidate_objects.items():

        if isinstance(obj, pd.DataFrame):

            score_cols = [
                c for c in obj.columns
                if "score" in c.lower()
            ]

            print(f"\n{name}:")
            print(score_cols)

    raise RuntimeError(
        "Need to reconstruct the original Card Quality subscores before continuing."
    )


# ------------------------------------------------------------
# 2. MERGE ORIGINAL SUBSCORES INTO OUR SANDBOX
# ------------------------------------------------------------

clean = model_d.copy()

merge_cols = [
    "event_name",
    "title_fights_score",
    "avg_ufc_experience_score",
    "avg_ufc_win_pct_score",
    "avg_win_streak_score",
    "avg_finish_rate_score"
]

# Use date too if both tables have it
merge_keys = ["event_name"]

if (
    "event_date" in clean.columns
    and
    "event_date" in source.columns
):

    merge_keys = [
        "event_date",
        "event_name"
    ]

    source["event_date"] = pd.to_datetime(
        source["event_date"]
    )

    clean["event_date"] = pd.to_datetime(
        clean["event_date"]
    )


# Avoid duplicate columns if any already exist
add_cols = [
    c for c in merge_cols
    if c not in merge_keys
    and c not in clean.columns
]

clean = clean.merge(
    source[
        merge_keys + add_cols
    ].drop_duplicates(
        subset=merge_keys
    ),
    on=merge_keys,
    how="left"
)


# ------------------------------------------------------------
# 3. VALIDATE INPUT AVAILABILITY
# ------------------------------------------------------------

original_inputs = [
    "title_fights_score",
    "avg_ufc_experience_score",
    "avg_ufc_win_pct_score",
    "avg_win_streak_score",
    "avg_finish_rate_score",
    "ranking_depth_index"
]

print("\nInput availability:")

for col in original_inputs:

    print(
        f"{col}:",
        clean[col].notna().sum(),
        "/",
        len(clean)
    )


# ------------------------------------------------------------
# 4. BUILD CLEAN CARD QUALITY
#
# Original CQ had 6 equal concepts:
#
# 1 title fights
# 2 ranking strength
# 3 UFC experience
# 4 win %
# 5 win streak
# 6 finish rate
#
# CLEAN VERSION:
# ranking_depth_index REPLACES ranking strength.
#
# Therefore each concept gets 1/6.
# ------------------------------------------------------------

CLEAN_CQ_INPUTS = [
    "title_fights_score",
    "ranking_depth_index",
    "avg_ufc_experience_score",
    "avg_ufc_win_pct_score",
    "avg_win_streak_score",
    "avg_finish_rate_score"
]


def clean_card_quality(row):

    vals = [
        row[c]
        for c in CLEAN_CQ_INPUTS
        if pd.notna(row[c])
    ]

    if not vals:
        return np.nan

    return np.mean(vals)


clean["card_quality_clean_equal"] = clean.apply(
    clean_card_quality,
    axis=1
)


# ------------------------------------------------------------
# 5. SECOND VERSION:
# Give ranking depth more influence INSIDE Card Quality
#
# Ranking Depth = 30%
# Remaining five concepts = 14% each
#
# 30 + (14*5) = 100
#
# This tests your original idea WITHOUT double counting.
# ------------------------------------------------------------

CLEAN_CQ_WEIGHTS = {
    "ranking_depth_index": 0.30,
    "title_fights_score": 0.14,
    "avg_ufc_experience_score": 0.14,
    "avg_ufc_win_pct_score": 0.14,
    "avg_win_streak_score": 0.14,
    "avg_finish_rate_score": 0.14
}


def weighted_missing(row, weights):

    available = {
        col: weight
        for col, weight in weights.items()
        if pd.notna(row[col])
    }

    if not available:
        return np.nan

    total = sum(
        available.values()
    )

    return sum(
        row[col] * weight
        for col, weight in available.items()
    ) / total


clean["card_quality_clean_rank30"] = clean.apply(
    lambda row:
    weighted_missing(
        row,
        CLEAN_CQ_WEIGHTS
    ),
    axis=1
)


# ------------------------------------------------------------
# 6. IMPORTANT HISTORICAL AVAILABILITY RULE
#
# Rankings begin 2013-02-04.
# Before then, do NOT treat missing rankings as zero.
# Dynamic redistribution handles missing Ranking Depth.
# ------------------------------------------------------------

if "event_date" in clean.columns:

    pre_rank = (
        pd.to_datetime(clean["event_date"])
        <
        pd.Timestamp("2013-02-04")
    )

    print(
        "\nPre-ranking events:",
        pre_rank.sum()
    )


# ------------------------------------------------------------
# 7. COMPARE OLD B40 CARD QUALITY vs CLEAN VERSIONS
# ------------------------------------------------------------

TARGETS = [
    "UFC 269",
    "UFC 168",
    "UFC 214",
    "UFC 217",
    "UFC 302",
    "UFC 205",
    "UFC 229",
    "UFC 259",
    "UFC 300",
    "UFC 100"
]

target_mask = pd.Series(
    False,
    index=clean.index
)

for name in TARGETS:

    target_mask |= (
        clean["event_name"]
        .astype(str)
        .str.contains(
            name,
            case=False,
            na=False,
            regex=False
        )
    )


print("\n" + "=" * 110)
print("🥊 CARD QUALITY — OLD B40 vs CLEAN")
print("=" * 110)

display(
    clean.loc[
        target_mask,
        [
            "event_name",

            "card_quality_score",
            "ranking_depth_index",
            "card_v3_40",

            "card_quality_clean_equal",
            "card_quality_clean_rank30"
        ]
    ]
    .sort_values(
        "card_quality_clean_rank30",
        ascending=False
    )
    .round(2)
)


# ------------------------------------------------------------
# 8. APPLY YOUR 46 / 22 / 22 / 10 WEIGHTS
# ------------------------------------------------------------

def overall_clean(row, card_col):

    weights = {
        card_col: 0.46,
        "star_experience_score": 0.22,
        "main_event_score": 0.22,
        "fanfare_score": 0.10
    }

    return weighted_missing(
        row,
        weights
    )


clean["score_clean_equal"] = clean.apply(
    lambda row:
    overall_clean(
        row,
        "card_quality_clean_equal"
    ),
    axis=1
)

clean["score_clean_rank30"] = clean.apply(
    lambda row:
    overall_clean(
        row,
        "card_quality_clean_rank30"
    ),
    axis=1
)


clean["rank_clean_equal"] = (
    clean["score_clean_equal"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype("Int64")
)

clean["rank_clean_rank30"] = (
    clean["score_clean_rank30"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype("Int64")
)


# ------------------------------------------------------------
# 9. TARGET RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🏆 46/22/22/10 — CLEAN CARD QUALITY RESULTS")
print("=" * 110)

display(
    clean.loc[
        target_mask,
        [
            "event_name",

            "event_strength_rank",
            "rank_46_22_22_10",

            "card_quality_clean_equal",
            "rank_clean_equal",

            "card_quality_clean_rank30",
            "rank_clean_rank30"
        ]
    ]
    .sort_values(
        "rank_clean_rank30"
    )
    .round(2)
)


# ------------------------------------------------------------
# 10. TOP 25 — EQUAL CLEAN VERSION
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🔥 TOP 25 — CLEAN CQ EQUAL / 46-22-22-10")
print("=" * 110)

display(
    clean[
        [
            "event_name",
            "score_clean_equal",
            "rank_clean_equal",
            "event_strength_rank"
        ]
    ]
    .sort_values(
        "rank_clean_equal"
    )
    .head(25)
    .round(2)
)


# ------------------------------------------------------------
# 11. TOP 25 — RANKING DEPTH 30% VERSION
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🔥 TOP 25 — CLEAN CQ RANKING DEPTH 30% / 46-22-22-10")
print("=" * 110)

display(
    clean[
        [
            "event_name",
            "score_clean_rank30",
            "rank_clean_rank30",
            "event_strength_rank"
        ]
    ]
    .sort_values(
        "rank_clean_rank30"
    )
    .head(25)
    .round(2)
)


# ------------------------------------------------------------
# 12. STABILITY / CORRELATION
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("📊 CLEAN MODEL DIAGNOSTICS")
print("=" * 110)

for score_col, rank_col, label in [

    (
        "score_clean_equal",
        "rank_clean_equal",
        "Equal clean CQ"
    ),

    (
        "score_clean_rank30",
        "rank_clean_rank30",
        "30% ranking-depth CQ"
    )
]:

    corr = clean[
        [
            "event_strength_score",
            score_col
        ]
    ].corr().iloc[0, 1]

    movement = (
        clean["event_strength_rank"]
        -
        clean[rank_col]
    ).abs()

    print(f"\n{label}")
    print(
        "Correlation:",
        round(corr, 4)
    )
    print(
        "Median rank movement:",
        movement.median()
    )
    print(
        "Average rank movement:",
        round(movement.mean(), 2)
    )
    print(
        "25+ place moves:",
        (movement >= 25).sum()
    )
    print(
        "50+ place moves:",
        (movement >= 50).sum()
    )
    print(
        "100+ place moves:",
        (movement >= 100).sum()
    )


# ------------------------------------------------------------
# 13. SAFETY
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("✅ SAFETY")
print("=" * 110)

print("Published model modified: NO")
print("final_index modified: NO")
print("model_d modified: NO")
print("Files overwritten: NO")
print("New sandbox object: clean")

print("\n🔥 CLEAN CARD QUALITY TEST COMPLETE")
```

```python
# ============================================================
# RECONSTRUCT ORIGINAL CARD QUALITY SUBSCORES
# FROM HISTORICAL FIGHT-LEVEL DATA
# SANDBOX ONLY
# ============================================================

import pandas as pd
import numpy as np

print("=" * 110)
print("🔥 RECONSTRUCT CARD QUALITY INPUTS")
print("=" * 110)


# ------------------------------------------------------------
# 1. FIND AUTHORITATIVE FIGHT TABLE
# ------------------------------------------------------------

fight_candidates = {
    "ufc_fights_updated": globals().get("ufc_fights_updated"),
    "ufc_fights": globals().get("ufc_fights"),
    "fight_test": globals().get("fight_test")
}

fight_source = None
fight_source_name = None

needed = [
    "date",
    "event_name",
    "f1_name",
    "f2_name",
    "is_title_fight",
    "f1_ufc_fights_before",
    "f2_ufc_fights_before",
    "f1_ufc_win_pct_before",
    "f2_ufc_win_pct_before",
    "f1_win_streak_before",
    "f2_win_streak_before",
    "f1_finish_rate_before",
    "f2_finish_rate_before"
]

for name, obj in fight_candidates.items():

    if not isinstance(obj, pd.DataFrame):
        continue

    if all(col in obj.columns for col in needed):

        fight_source = obj.copy()
        fight_source_name = name
        break


print("Fight source:", fight_source_name)

if fight_source is None:

    print("\n❌ STOP — no fight table has all required columns.")

    for name, obj in fight_candidates.items():

        if isinstance(obj, pd.DataFrame):

            missing = [
                col for col in needed
                if col not in obj.columns
            ]

            print(
                f"\n{name}: {len(obj)} rows"
            )

            print(
                "Missing:",
                missing
            )

    raise RuntimeError(
        "Required fight-level inputs are unavailable."
    )


print(
    "Rows:",
    len(fight_source)
)

print(
    "Date range:",
    fight_source["date"].min(),
    "to",
    fight_source["date"].max()
)


# ------------------------------------------------------------
# 2. STANDARDIZE DATE
# ------------------------------------------------------------

fight_source["date"] = pd.to_datetime(
    fight_source["date"]
)


# ------------------------------------------------------------
# 3. BUILD EVENT-LEVEL RAW CARD METRICS
# ------------------------------------------------------------

event_rows = []

for (date, event), g in fight_source.groupby(
    ["date", "event_name"],
    sort=False
):

    # Every fighter appearance on the card
    experience = pd.concat(
        [
            g["f1_ufc_fights_before"],
            g["f2_ufc_fights_before"]
        ],
        ignore_index=True
    )

    win_pct = pd.concat(
        [
            g["f1_ufc_win_pct_before"],
            g["f2_ufc_win_pct_before"]
        ],
        ignore_index=True
    )

    streak = pd.concat(
        [
            g["f1_win_streak_before"],
            g["f2_win_streak_before"]
        ],
        ignore_index=True
    )

    finish_rate = pd.concat(
        [
            g["f1_finish_rate_before"],
            g["f2_finish_rate_before"]
        ],
        ignore_index=True
    )

    event_rows.append(
        {
            "event_date": date,
            "event_name": event,

            "title_fights_rebuilt":
                pd.to_numeric(
                    g["is_title_fight"],
                    errors="coerce"
                ).fillna(0).sum(),

            "avg_ufc_experience_rebuilt":
                pd.to_numeric(
                    experience,
                    errors="coerce"
                ).mean(),

            "avg_ufc_win_pct_rebuilt":
                pd.to_numeric(
                    win_pct,
                    errors="coerce"
                ).mean(),

            "avg_win_streak_rebuilt":
                pd.to_numeric(
                    streak,
                    errors="coerce"
                ).mean(),

            "avg_finish_rate_rebuilt":
                pd.to_numeric(
                    finish_rate,
                    errors="coerce"
                ).mean()
        }
    )


cq_rebuilt = pd.DataFrame(
    event_rows
)

print(
    "\nEvents reconstructed:",
    len(cq_rebuilt)
)

print(
    "Unique events:",
    cq_rebuilt[
        ["event_date", "event_name"]
    ].drop_duplicates().shape[0]
)


# ------------------------------------------------------------
# 4. ZERO-PRESERVING PERCENTILE
#
# Same normalization function used in original model.
# ------------------------------------------------------------

def zero_preserving_percentile(series):

    series = pd.to_numeric(
        series,
        errors="coerce"
    )

    result = pd.Series(
        index=series.index,
        dtype=float
    )

    result.loc[series.isna()] = np.nan

    result.loc[series == 0] = 0

    positive = series > 0

    if positive.sum() > 0:

        result.loc[positive] = (
            series.loc[positive]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )

    return result


# ------------------------------------------------------------
# 5. CREATE FIVE ORIGINAL CQ SUBSCORES
# ------------------------------------------------------------

cq_rebuilt["title_fights_score_rebuilt"] = (
    zero_preserving_percentile(
        cq_rebuilt["title_fights_rebuilt"]
    )
)

cq_rebuilt["avg_ufc_experience_score_rebuilt"] = (
    zero_preserving_percentile(
        cq_rebuilt["avg_ufc_experience_rebuilt"]
    )
)

cq_rebuilt["avg_ufc_win_pct_score_rebuilt"] = (
    zero_preserving_percentile(
        cq_rebuilt["avg_ufc_win_pct_rebuilt"]
    )
)

cq_rebuilt["avg_win_streak_score_rebuilt"] = (
    zero_preserving_percentile(
        cq_rebuilt["avg_win_streak_rebuilt"]
    )
)

cq_rebuilt["avg_finish_rate_score_rebuilt"] = (
    zero_preserving_percentile(
        cq_rebuilt["avg_finish_rate_rebuilt"]
    )
)


# ------------------------------------------------------------
# 6. MERGE INTO SANDBOX
# ------------------------------------------------------------

clean = model_d.copy()

clean["event_date"] = pd.to_datetime(
    clean["event_date"]
)

rebuild_cols = [
    "event_date",
    "event_name",

    "title_fights_rebuilt",
    "avg_ufc_experience_rebuilt",
    "avg_ufc_win_pct_rebuilt",
    "avg_win_streak_rebuilt",
    "avg_finish_rate_rebuilt",

    "title_fights_score_rebuilt",
    "avg_ufc_experience_score_rebuilt",
    "avg_ufc_win_pct_score_rebuilt",
    "avg_win_streak_score_rebuilt",
    "avg_finish_rate_score_rebuilt"
]

clean = clean.merge(
    cq_rebuilt[rebuild_cols],
    on=[
        "event_date",
        "event_name"
    ],
    how="left"
)


# ------------------------------------------------------------
# 7. VALIDATE COVERAGE
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("📊 COVERAGE")
print("=" * 110)

check_cols = [
    "title_fights_score_rebuilt",
    "avg_ufc_experience_score_rebuilt",
    "avg_ufc_win_pct_score_rebuilt",
    "avg_win_streak_score_rebuilt",
    "avg_finish_rate_score_rebuilt",
    "ranking_depth_index"
]

for col in check_cols:

    print(
        f"{col}:",
        clean[col].notna().sum(),
        "/",
        len(clean)
    )


# ------------------------------------------------------------
# 8. SHOW TARGET EVENTS
# ------------------------------------------------------------

TARGETS = [
    "UFC 269",
    "UFC 168",
    "UFC 214",
    "UFC 217",
    "UFC 302",
    "UFC 205",
    "UFC 229",
    "UFC 259",
    "UFC 300",
    "UFC 100"
]

mask = pd.Series(
    False,
    index=clean.index
)

for name in TARGETS:

    mask |= (
        clean["event_name"]
        .astype(str)
        .str.contains(
            name,
            case=False,
            na=False,
            regex=False
        )
    )


print("\n" + "=" * 110)
print("🥊 REBUILT CARD QUALITY INPUTS — TARGET EVENTS")
print("=" * 110)

display(
    clean.loc[
        mask,
        [
            "event_name",

            "title_fights_rebuilt",
            "title_fights_score_rebuilt",

            "avg_ufc_experience_rebuilt",
            "avg_ufc_experience_score_rebuilt",

            "avg_ufc_win_pct_rebuilt",
            "avg_ufc_win_pct_score_rebuilt",

            "avg_win_streak_rebuilt",
            "avg_win_streak_score_rebuilt",

            "avg_finish_rate_rebuilt",
            "avg_finish_rate_score_rebuilt",

            "ranking_depth_index",
            "card_quality_score",
            "card_v3_40"
        ]
    ]
    .sort_values(
        "card_v3_40",
        ascending=False
    )
    .round(3)
)


# ------------------------------------------------------------
# 9. IMPORTANT VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🔎 VALIDATION")
print("=" * 110)

print(
    "Duplicate rebuilt events:",
    cq_rebuilt.duplicated(
        [
            "event_date",
            "event_name"
        ]
    ).sum()
)

print(
    "Events matched to model:",
    clean[
        "avg_ufc_experience_score_rebuilt"
    ].notna().sum(),
    "/",
    len(clean)
)

print(
    "Ranking-depth available:",
    clean[
        "ranking_depth_index"
    ].notna().sum(),
    "/",
    len(clean)
)


print("\n" + "=" * 110)
print("✅ SAFETY")
print("=" * 110)

print("Published model modified: NO")
print("model_d modified: NO")
print("Files overwritten: NO")
print("New sandbox objects: cq_rebuilt, clean")

print("\n🔥 RECONSTRUCTION COMPLETE")
```

```python
# ============================================================
# CLEAN CARD QUALITY v2
# NO RANKING DOUBLE COUNT
#
# Compare:
# A) Equal clean CQ:
#    Ranking Depth = 1 of 6 equal concepts = 16.67%
#
# B) Rank30 clean CQ:
#    Ranking Depth = 30%
#    Other 5 concepts = 14% each
#
# Then apply:
# 46% Card Quality
# 22% Star Experience
# 22% Main Event
# 10% Fanfare
#
# SANDBOX ONLY
# ============================================================

import pandas as pd
import numpy as np

print("=" * 110)
print("🔥 CLEAN CARD QUALITY v2 — TRUE NO-DOUBLE-COUNT TEST")
print("=" * 110)


# ------------------------------------------------------------
# 1. INPUT COLUMNS
# ------------------------------------------------------------

NON_RANK_INPUTS = [
    "title_fights_score_rebuilt",
    "avg_ufc_experience_score_rebuilt",
    "avg_ufc_win_pct_score_rebuilt",
    "avg_win_streak_score_rebuilt",
    "avg_finish_rate_score_rebuilt"
]

ALL_CLEAN_INPUTS = [
    "ranking_depth_index"
] + NON_RANK_INPUTS


# ------------------------------------------------------------
# 2. EQUAL CLEAN CQ
#
# Rankings available:
# 6 concepts, equal weight
#
# Rankings unavailable:
# use only 5 non-ranking concepts
# No missing ranking treated as zero.
# ------------------------------------------------------------

def clean_equal_cq(row):

    vals = [
        row[col]
        for col in ALL_CLEAN_INPUTS
        if pd.notna(row[col])
    ]

    if not vals:
        return np.nan

    return np.mean(vals)


clean["card_quality_clean_equal"] = clean.apply(
    clean_equal_cq,
    axis=1
)


# ------------------------------------------------------------
# 3. RANK30 CLEAN CQ
#
# Ranking Depth = 30%
# Other five = 14% each
#
# Dynamic redistribution when ranking data does not exist.
# ------------------------------------------------------------

CQ_RANK30_WEIGHTS = {
    "ranking_depth_index": 0.30,
    "title_fights_score_rebuilt": 0.14,
    "avg_ufc_experience_score_rebuilt": 0.14,
    "avg_ufc_win_pct_score_rebuilt": 0.14,
    "avg_win_streak_score_rebuilt": 0.14,
    "avg_finish_rate_score_rebuilt": 0.14
}


def weighted_missing(row, weights):

    available = {
        col: weight
        for col, weight in weights.items()
        if pd.notna(row[col])
    }

    if not available:
        return np.nan

    total_weight = sum(available.values())

    return sum(
        row[col] * weight
        for col, weight in available.items()
    ) / total_weight


clean["card_quality_clean_rank30"] = clean.apply(
    lambda row:
    weighted_missing(
        row,
        CQ_RANK30_WEIGHTS
    ),
    axis=1
)


# ------------------------------------------------------------
# 4. IMPORTANT:
# 11 POST-JUNE-27 EVENTS DO NOT HAVE REBUILT CQ INPUTS
#
# Do NOT fabricate them.
# Hold their existing B40 Card Quality for this sandbox test.
#
# This lets all 788 events remain rankable while clearly
# separating reconstructed historical cards from held cards.
# ------------------------------------------------------------

clean["clean_cq_status"] = np.where(
    clean[
        "avg_ufc_experience_score_rebuilt"
    ].notna(),
    "reconstructed",
    "held_B40_post_cutoff"
)

clean["card_quality_clean_equal_full"] = (
    clean["card_quality_clean_equal"]
)

clean["card_quality_clean_rank30_full"] = (
    clean["card_quality_clean_rank30"]
)

post_mask = (
    clean["clean_cq_status"]
    ==
    "held_B40_post_cutoff"
)

clean.loc[
    post_mask,
    "card_quality_clean_equal_full"
] = clean.loc[
    post_mask,
    "card_v3_40"
]

clean.loc[
    post_mask,
    "card_quality_clean_rank30_full"
] = clean.loc[
    post_mask,
    "card_v3_40"
]


# ------------------------------------------------------------
# 5. APPLY 46 / 22 / 22 / 10
# ------------------------------------------------------------

def overall_score(row, card_col):

    weights = {
        card_col: 0.46,
        "star_experience_score": 0.22,
        "main_event_score": 0.22,
        "fanfare_score": 0.10
    }

    return weighted_missing(
        row,
        weights
    )


clean["score_clean_equal"] = clean.apply(
    lambda row:
    overall_score(
        row,
        "card_quality_clean_equal_full"
    ),
    axis=1
)

clean["score_clean_rank30"] = clean.apply(
    lambda row:
    overall_score(
        row,
        "card_quality_clean_rank30_full"
    ),
    axis=1
)


clean["rank_clean_equal"] = (
    clean["score_clean_equal"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype("Int64")
)

clean["rank_clean_rank30"] = (
    clean["score_clean_rank30"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype("Int64")
)


# ------------------------------------------------------------
# 6. TARGET EVENTS
# ------------------------------------------------------------

TARGETS = [
    "UFC 269",
    "UFC 168",
    "UFC 214",
    "UFC 217",
    "UFC 302",
    "UFC 205",
    "UFC 229",
    "UFC 259",
    "UFC 300",
    "UFC 100"
]

target_mask = pd.Series(
    False,
    index=clean.index
)

for name in TARGETS:

    target_mask |= (
        clean["event_name"]
        .astype(str)
        .str.contains(
            name,
            case=False,
            na=False,
            regex=False
        )
    )


print("\n" + "=" * 110)
print("🥊 CLEAN CARD QUALITY — TARGET EVENTS")
print("=" * 110)

display(
    clean.loc[
        target_mask,
        [
            "event_name",

            "ranking_depth_index",

            "card_quality_score",
            "card_v3_40",

            "card_quality_clean_equal_full",
            "card_quality_clean_rank30_full",

            "event_strength_rank",
            "rank_46_22_22_10",

            "rank_clean_equal",
            "rank_clean_rank30"
        ]
    ]
    .sort_values(
        "rank_clean_rank30"
    )
    .round(2)
)


# ------------------------------------------------------------
# 7. TOP 25 — EQUAL CLEAN CQ
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🔥 TOP 25 — EQUAL CLEAN CQ + 46/22/22/10")
print("=" * 110)

display(
    clean[
        [
            "event_name",
            "card_quality_clean_equal_full",
            "score_clean_equal",
            "rank_clean_equal",
            "event_strength_rank"
        ]
    ]
    .sort_values(
        "rank_clean_equal"
    )
    .head(25)
    .round(2)
)


# ------------------------------------------------------------
# 8. TOP 25 — RANK30 CLEAN CQ
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("🔥 TOP 25 — RANK30 CLEAN CQ + 46/22/22/10")
print("=" * 110)

display(
    clean[
        [
            "event_name",
            "ranking_depth_index",
            "card_quality_clean_rank30_full",
            "score_clean_rank30",
            "rank_clean_rank30",
            "event_strength_rank"
        ]
    ]
    .sort_values(
        "rank_clean_rank30"
    )
    .head(25)
    .round(2)
)


# ------------------------------------------------------------
# 9. DIRECT EQUAL vs RANK30 TARGET COMPARISON
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("⚔️ EQUAL CQ vs RANK30 CQ")
print("=" * 110)

display(
    clean.loc[
        target_mask,
        [
            "event_name",

            "score_clean_equal",
            "rank_clean_equal",

            "score_clean_rank30",
            "rank_clean_rank30"
        ]
    ]
    .sort_values(
        "rank_clean_rank30"
    )
    .round(2)
)


# ------------------------------------------------------------
# 10. DIAGNOSTICS
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("📊 MODEL DIAGNOSTICS")
print("=" * 110)

for score_col, rank_col, label in [

    (
        "score_clean_equal",
        "rank_clean_equal",
        "Equal clean CQ"
    ),

    (
        "score_clean_rank30",
        "rank_clean_rank30",
        "Rank30 clean CQ"
    )
]:

    correlation = clean[
        [
            "event_strength_score",
            score_col
        ]
    ].corr().iloc[0, 1]

    movement = (
        clean["event_strength_rank"]
        -
        clean[rank_col]
    ).abs()

    print("\n" + label)

    print(
        "Correlation with published:",
        round(correlation, 4)
    )

    print(
        "Median absolute rank movement:",
        movement.median()
    )

    print(
        "Average absolute rank movement:",
        round(
            movement.mean(),
            2
        )
    )

    print(
        "25+ place moves:",
        int(
            (movement >= 25).sum()
        )
    )

    print(
        "50+ place moves:",
        int(
            (movement >= 50).sum()
        )
    )

    print(
        "100+ place moves:",
        int(
            (movement >= 100).sum()
        )
    )


# ------------------------------------------------------------
# 11. RANKING IMPACT — VALID RANKING ERA ONLY
# ------------------------------------------------------------

ranking_valid = (
    clean["ranking_depth_index"].notna()
    &
    clean["card_quality_clean_equal"].notna()
)

difference = (
    clean.loc[
        ranking_valid,
        "card_quality_clean_rank30"
    ]
    -
    clean.loc[
        ranking_valid,
        "card_quality_clean_equal"
    ]
)

print("\n" + "=" * 110)
print("📈 EFFECT OF RAISING RANKING DEPTH FROM 16.7% → 30% OF CQ")
print("=" * 110)

print(
    "Events compared:",
    ranking_valid.sum()
)

print(
    "Average CQ change:",
    round(
        difference.mean(),
        3
    )
)

print(
    "Median CQ change:",
    round(
        difference.median(),
        3
    )
)

print(
    "Largest CQ increase:",
    round(
        difference.max(),
        3
    )
)

print(
    "Largest CQ decrease:",
    round(
        difference.min(),
        3
    )
)


# ------------------------------------------------------------
# 12. SAFETY
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("✅ SAFETY")
print("=" * 110)

print(
    "Historical reconstructed events:",
    (
        clean["clean_cq_status"]
        ==
        "reconstructed"
    ).sum()
)

print(
    "Post-cutoff cards held at B40:",
    (
        clean["clean_cq_status"]
        ==
        "held_B40_post_cutoff"
    ).sum()
)

print("Published model modified: NO")
print("model_d modified: NO")
print("Files overwritten: NO")

print("\n🔥 CLEAN CARD QUALITY v2 COMPLETE")
```

```python
# ============================================================
# RESTORE 11 UFC EVENTS — JULY 11 THROUGH SEPTEMBER 19, 2026
# OFFICIAL UFC.COM
# SANDBOX ONLY
# ============================================================

import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
import time

print("=" * 105)
print("🔥 RESTORING 11 COMPLETED UFC EVENTS")
print("=" * 105)


# ------------------------------------------------------------
# 1. EVENT DEFINITIONS
# ------------------------------------------------------------

EVENTS = [
    (
        "2026-07-11",
        "UFC 329: McGregor vs. Holloway 2",
        "https://www.ufc.com/event/ufc-329"
    ),
    (
        "2026-07-18",
        "UFC Fight Night: Du Plessis vs. Usman",
        "https://www.ufc.com/event/ufc-fight-night-july-18-2026"
    ),
    (
        "2026-07-25",
        "UFC Fight Night: Ankalaev vs. Guskov",
        "https://www.ufc.com/event/ufc-fight-night-july-25-2026"
    ),
    (
        "2026-08-01",
        "UFC Fight Night: Medic vs. Rodriguez",
        "https://www.ufc.com/event/ufc-fight-night-august-01-2026"
    ),
    (
        "2026-08-08",
        "UFC Fight Night: Gamrot vs. Salkilld",
        "https://www.ufc.com/event/ufc-fight-night-august-08-2026"
    ),
    (
        "2026-08-15",
        "UFC 330: Makhachev vs. Machado Garry",
        "https://www.ufc.com/event/ufc-330"
    ),
    (
        "2026-08-22",
        "UFC Fight Night: Hernandez vs. Rodrigues",
        "https://www.ufc.com/event/ufc-fight-night-august-22-2026"
    ),
    (
        "2026-08-29",
        "UFC Fight Night: Nurmagomedov vs. Song",
        "https://www.ufc.com/event/ufc-fight-night-august-29-2026"
    ),
    (
        "2026-09-05",
        "UFC Fight Night: Hooker vs. Parnasse",
        "https://www.ufc.com/event/ufc-fight-night-september-05-2026"
    ),
    (
        "2026-09-12",
        "Noche UFC: Silva vs. Delgado",
        "https://www.ufc.com/event/ufc-fight-night-september-12-2026"
    ),
    (
        "2026-09-19",
        "UFC 331: Van vs. Pantoja 2",
        "https://www.ufc.com/event/cryptocom-ufc-331"
    )
]


# ------------------------------------------------------------
# 2. HELPERS
# ------------------------------------------------------------

HEADERS = {
    "User-Agent":
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/154.0 Safari/537.36"
}


def text_or_nan(block, selector):

    node = block.select_one(selector)

    if node is None:
        return np.nan

    text = node.get_text(
        " ",
        strip=True
    )

    return text if text else np.nan


def clean_name(x):

    if pd.isna(x):
        return x

    return " ".join(
        str(x).split()
    )


# ------------------------------------------------------------
# 3. SCRAPE
# ------------------------------------------------------------

rows = []
event_summary = []

session = requests.Session()
session.headers.update(HEADERS)

for event_date, event_name, url in EVENTS:

    print(
        f"\n📥 {event_date} — {event_name}"
    )

    try:

        response = session.get(
            url,
            timeout=30
        )

        print(
            "   HTTP:",
            response.status_code
        )

        if response.status_code != 200:

            event_summary.append(
                {
                    "date": event_date,
                    "event_name": event_name,
                    "status": f"HTTP {response.status_code}",
                    "fights": 0
                }
            )

            continue

        soup = BeautifulSoup(
            response.text,
            "html.parser"
        )

        blocks = soup.select(
            ".c-listing-fight"
        )

        print(
            "   Fight blocks:",
            len(blocks)
        )

        count = 0

        for block in blocks:

            f1 = text_or_nan(
                block,
                ".c-listing-fight__corner-name--red"
            )

            f2 = text_or_nan(
                block,
                ".c-listing-fight__corner-name--blue"
            )

            # Skip non-fight/template blocks
            if pd.isna(f1) or pd.isna(f2):
                continue

            f1 = clean_name(f1)
            f2 = clean_name(f2)

            red_win = block.select_one(
                ".c-listing-fight__corner--red "
                ".c-listing-fight__outcome--win"
            )

            blue_win = block.select_one(
                ".c-listing-fight__corner--blue "
                ".c-listing-fight__outcome--win"
            )

            if red_win is not None:
                f1_result = "W"
                f2_result = "L"

            elif blue_win is not None:
                f1_result = "L"
                f2_result = "W"

            else:
                # Could be draw/NC if present
                f1_result = np.nan
                f2_result = np.nan

            weight_class = text_or_nan(
                block,
                ".c-listing-fight__class--desktop "
                ".c-listing-fight__class-text"
            )

            method = text_or_nan(
                block,
                ".method"
            )

            round_value = text_or_nan(
                block,
                ".round"
            )

            time_value = text_or_nan(
                block,
                ".time"
            )

            fight_id = block.get(
                "data-fmid",
                np.nan
            )

            rows.append(
                {
                    "fight_url": np.nan,
                    "event_name": event_name,
                    "date": pd.Timestamp(event_date),
                    "location": np.nan,

                    "f1_name": f1,
                    "f1_result": f1_result,

                    "f2_name": f2,
                    "f2_result": f2_result,

                    "weight_class": weight_class,
                    "method": method,
                    "round": round_value,
                    "time": time_value,

                    "time_format": np.nan,
                    "referee": np.nan,
                    "judging_details": np.nan,

                    "ufc_fight_id": fight_id,
                    "source_url": url
                }
            )

            count += 1

        event_summary.append(
            {
                "date": event_date,
                "event_name": event_name,
                "status": "OK",
                "fights": count
            }
        )

        print(
            "   Parsed fights:",
            count
        )

        time.sleep(1)

    except Exception as e:

        print(
            "   ❌ ERROR:",
            repr(e)
        )

        event_summary.append(
            {
                "date": event_date,
                "event_name": event_name,
                "status": repr(e),
                "fights": 0
            }
        )


# ------------------------------------------------------------
# 4. BUILD RESTORED TABLE
# ------------------------------------------------------------

new_fights_restored = pd.DataFrame(
    rows
)

restore_summary = pd.DataFrame(
    event_summary
)


# ------------------------------------------------------------
# 5. VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 105)
print("📊 EVENT RESTORE SUMMARY")
print("=" * 105)

display(
    restore_summary
)

print("\nTotal fights restored:",
      len(new_fights_restored))

print(
    "Events restored:",
    new_fights_restored[
        "event_name"
    ].nunique()
    if len(new_fights_restored)
    else 0
)

if len(new_fights_restored):

    print(
        "Date range:",
        new_fights_restored["date"].min(),
        "→",
        new_fights_restored["date"].max()
    )

    print(
        "Unique UFC fight IDs:",
        new_fights_restored[
            "ufc_fight_id"
        ].nunique()
    )

    print(
        "Missing fighter names:",
        new_fights_restored[
            ["f1_name", "f2_name"]
        ].isna().sum().sum()
    )

    print(
        "Fights without W/L result:",
        (
            new_fights_restored[
                ["f1_result", "f2_result"]
            ]
            .isna()
            .any(axis=1)
        ).sum()
    )


print("\n" + "=" * 105)
print("🔒 SAFETY")
print("=" * 105)

print("ufc_fights modified: NO")
print("clean modified: NO")
print("model_d modified: NO")
print("Files overwritten: NO")
print("New object: new_fights_restored")

print("\n🔥 RESTORE COLLECTION COMPLETE")
```

```python
# ============================================================
# RESTORE NEW FIGHT HISTORIES THROUGH SEPTEMBER 19, 2026
# CLEAN CARD QUALITY INPUTS
# SANDBOX ONLY
# ============================================================

import pandas as pd
import numpy as np

print("=" * 110)
print("🔥 BUILDING ENTERING-FIGHT HISTORIES FOR 141 RESTORED FIGHTS")
print("=" * 110)


# ------------------------------------------------------------
# 1. CONFIRMED UFC.COM -> HISTORICAL NAME MAPPINGS
# ------------------------------------------------------------

confirmed_name_map = {
    "Aleksandar Rakić": "Aleksandar Rakic",
    "Benoît Saint Denis": "Benoit Saint Denis",
    "Duško Todorović": "Dusko Todorovic",
    "Farès Ziam": "Fares Ziam",
    "Jan Błachowicz": "Jan Blachowicz",
    "Joel Álvarez": "Joel Alvarez",
    "Kauê Fernandes": "Kaue Fernandes",
    "Klaudia Syguła": "Klaudia Sygula",
    "Lone’er Kavanagh": "Lone'er Kavanagh",
    "Ludovít Klein": "Ludovit Klein",
    "Mateusz Rębecki": "Mateusz Rebecki",
    "Michael Venom Page": "Michael Page",
    "Uroš Medić": "Uros Medic",
    "Zachary Reese": "Zach Reese"
}


nf = new_fights_restored.copy()

nf["f1_name"] = nf["f1_name"].replace(
    confirmed_name_map
)

nf["f2_name"] = nf["f2_name"].replace(
    confirmed_name_map
)

nf["date"] = pd.to_datetime(nf["date"])


# ------------------------------------------------------------
# 2. PRESERVE ORIGINAL UFC.COM WEIGHT CLASS
# ------------------------------------------------------------

nf["weight_class_original"] = nf["weight_class"]


# ------------------------------------------------------------
# 3. RESTORE THE THREE CONFIRMED TITLE FIGHTS
# ------------------------------------------------------------

TITLE_FIGHTS = {
    (
        pd.Timestamp("2026-08-15"),
        "Islam Makhachev",
        "Ian Machado Garry"
    ): "UFC Welterweight Title Bout",

    (
        pd.Timestamp("2026-08-15"),
        "Mackenzie Dern",
        "Gillian Robertson"
    ): "UFC Women's Strawweight Title Bout",

    (
        pd.Timestamp("2026-09-19"),
        "Joshua Van",
        "Alexandre Pantoja"
    ): "UFC Flyweight Title Bout"
}


for i, row in nf.iterrows():

    key = (
        row["date"],
        row["f1_name"],
        row["f2_name"]
    )

    reverse_key = (
        row["date"],
        row["f2_name"],
        row["f1_name"]
    )

    if key in TITLE_FIGHTS:

        nf.at[
            i,
            "weight_class"
        ] = TITLE_FIGHTS[key]

    elif reverse_key in TITLE_FIGHTS:

        nf.at[
            i,
            "weight_class"
        ] = TITLE_FIGHTS[reverse_key]


nf["is_title_fight"] = (
    nf["weight_class"]
    .astype(str)
    .str.startswith("UFC ")
    &
    nf["weight_class"]
    .astype(str)
    .str.contains(
        "Title Bout",
        case=False,
        na=False
    )
    &
    ~nf["weight_class"]
    .astype(str)
    .str.contains(
        "Tournament",
        case=False,
        na=False
    )
)


# ------------------------------------------------------------
# 4. BUILD HISTORICAL FIGHT LEDGER
#
# Start with the existing 8,736 fights.
# Then process the 141 new fights chronologically.
#
# IMPORTANT:
# Stats are recorded BEFORE each new fight is added.
# This preserves no-future-leakage.
# ------------------------------------------------------------

history = {}


def get_fighter(name):

    if name not in history:

        history[name] = {
            "wins": 0,
            "losses": 0,
            "fights": 0,
            "win_streak": 0,
            "finishes": 0
        }

    return history[name]


def is_finish(method):

    if pd.isna(method):
        return False

    m = str(method).upper()

    return (
        "KO" in m
        or "TKO" in m
        or "SUB" in m
    )


# ------------------------------------------------------------
# 5. LOAD THE 8,736 HISTORICAL FIGHTS INTO LEDGER
# ------------------------------------------------------------

historical = ufc_fights.copy()

historical["date"] = pd.to_datetime(
    historical["date"]
)

historical = historical.sort_values(
    ["date"]
).reset_index(drop=True)


for _, row in historical.iterrows():

    f1 = row["f1_name"]
    f2 = row["f2_name"]

    s1 = get_fighter(f1)
    s2 = get_fighter(f2)

    r1 = row["f1_result"]
    r2 = row["f2_result"]

    # Record fight appearance
    s1["fights"] += 1
    s2["fights"] += 1

    # Results / streak
    if r1 == "W":

        s1["wins"] += 1
        s1["win_streak"] += 1

        s2["losses"] += 1
        s2["win_streak"] = 0

        if is_finish(row["method"]):
            s1["finishes"] += 1

    elif r2 == "W":

        s2["wins"] += 1
        s2["win_streak"] += 1

        s1["losses"] += 1
        s1["win_streak"] = 0

        if is_finish(row["method"]):
            s2["finishes"] += 1

    else:

        # Draw / NC breaks winning streak
        s1["win_streak"] = 0
        s2["win_streak"] = 0


# ------------------------------------------------------------
# 6. PROCESS NEW FIGHTS IN DATE ORDER
#
# Each event's entering stats must all be measured BEFORE
# that event. Fighters normally appear once per event, so
# date order is sufficient here.
# ------------------------------------------------------------

nf = nf.sort_values(
    ["date", "event_name"]
).reset_index(drop=True)


history_rows = []


for _, row in nf.iterrows():

    f1 = row["f1_name"]
    f2 = row["f2_name"]

    s1 = get_fighter(f1)
    s2 = get_fighter(f2)

    # ------------------------------------------
    # ENTERING-FIGHT VALUES
    # ------------------------------------------

    f1_pct = (
        s1["wins"] /
        (s1["wins"] + s1["losses"])
        if (s1["wins"] + s1["losses"]) > 0
        else np.nan
    )

    f2_pct = (
        s2["wins"] /
        (s2["wins"] + s2["losses"])
        if (s2["wins"] + s2["losses"]) > 0
        else np.nan
    )

    f1_finish_rate = (
        s1["finishes"] / s1["wins"]
        if s1["wins"] > 0
        else np.nan
    )

    f2_finish_rate = (
        s2["finishes"] / s2["wins"]
        if s2["wins"] > 0
        else np.nan
    )

    history_rows.append(
        {
            "f1_ufc_fights_before":
                s1["fights"],

            "f2_ufc_fights_before":
                s2["fights"],

            "f1_ufc_win_pct_before":
                f1_pct,

            "f2_ufc_win_pct_before":
                f2_pct,

            "f1_win_streak_before":
                s1["win_streak"],

            "f2_win_streak_before":
                s2["win_streak"],

            "f1_finish_rate_before":
                f1_finish_rate,

            "f2_finish_rate_before":
                f2_finish_rate
        }
    )

    # ------------------------------------------
    # NOW ADD THIS FIGHT TO HISTORY
    # ------------------------------------------

    s1["fights"] += 1
    s2["fights"] += 1

    if row["f1_result"] == "W":

        s1["wins"] += 1
        s1["win_streak"] += 1

        s2["losses"] += 1
        s2["win_streak"] = 0

        if is_finish(row["method"]):
            s1["finishes"] += 1

    elif row["f2_result"] == "W":

        s2["wins"] += 1
        s2["win_streak"] += 1

        s1["losses"] += 1
        s1["win_streak"] = 0

        if is_finish(row["method"]):
            s2["finishes"] += 1

    else:

        s1["win_streak"] = 0
        s2["win_streak"] = 0


history_df = pd.DataFrame(
    history_rows
)

for col in history_df.columns:

    nf[col] = history_df[col].values


# ------------------------------------------------------------
# 7. BUILD NEW EVENT-LEVEL NON-RANKING CQ INPUTS
# ------------------------------------------------------------

event_rows = []


for (date, event), g in nf.groupby(
    ["date", "event_name"],
    sort=True
):

    experience = pd.concat(
        [
            g["f1_ufc_fights_before"],
            g["f2_ufc_fights_before"]
        ],
        ignore_index=True
    )

    win_pct = pd.concat(
        [
            g["f1_ufc_win_pct_before"],
            g["f2_ufc_win_pct_before"]
        ],
        ignore_index=True
    )

    streak = pd.concat(
        [
            g["f1_win_streak_before"],
            g["f2_win_streak_before"]
        ],
        ignore_index=True
    )

    finish_rate = pd.concat(
        [
            g["f1_finish_rate_before"],
            g["f2_finish_rate_before"]
        ],
        ignore_index=True
    )

    event_rows.append(
        {
            "event_date": date,
            "event_name": event,

            "total_fights":
                len(g),

            "title_fights":
                int(
                    g["is_title_fight"].sum()
                ),

            "avg_ufc_experience":
                experience.mean(),

            "avg_ufc_win_pct":
                win_pct.mean(),

            "avg_win_streak":
                streak.mean(),

            "avg_finish_rate":
                finish_rate.mean()
        }
    )


new_event_cq_raw = pd.DataFrame(
    event_rows
)


# ------------------------------------------------------------
# 8. VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("📊 RESTORED EVENT INPUTS")
print("=" * 110)

display(
    new_event_cq_raw.round(3)
)


print("\nTotal restored fights:", len(nf))
print(
    "Restored events:",
    nf["event_name"].nunique()
)

print(
    "Title fights:",
    int(
        nf["is_title_fight"].sum()
    )
)

print(
    "Expected title fights: 3"
)


print("\n" + "=" * 110)
print("🏆 TITLE FIGHTS")
print("=" * 110)

display(
    nf.loc[
        nf["is_title_fight"],
        [
            "date",
            "event_name",
            "f1_name",
            "f2_name",
            "weight_class"
        ]
    ]
)


print("\n" + "=" * 110)
print("🔎 HISTORY COVERAGE")
print("=" * 110)

history_cols = [
    "f1_ufc_fights_before",
    "f2_ufc_fights_before",
    "f1_ufc_win_pct_before",
    "f2_ufc_win_pct_before",
    "f1_win_streak_before",
    "f2_win_streak_before",
    "f1_finish_rate_before",
    "f2_finish_rate_before"
]

for col in history_cols:

    print(
        f"{col:<30}",
        nf[col].notna().sum(),
        "/",
        len(nf)
    )


print("\n" + "=" * 110)
print("🔒 SAFETY")
print("=" * 110)

print("Original ufc_fights modified: NO")
print("Published final_index modified: NO")
print("clean modified: NO")
print("Files overwritten: NO")

print(
    "\nNew sandbox objects:",
    "nf, history, history_df, new_event_cq_raw"
)

print("\n🔥 HISTORY RESTORE COMPLETE")
```

```python
# ============================================================
# RESTORE OFFICIAL EVENT-TIME RANKINGS
# 11 EVENTS / 141 FIGHTS
# C = 0, UNRANKED = NaN
# SANDBOX ONLY
# ============================================================

import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
import re
import time

print("=" * 110)
print("🔥 RESTORING OFFICIAL EVENT-TIME UFC RANKINGS")
print("=" * 110)


def parse_rank(text):
    """
    UFC champion -> 0
    #1 through #15 -> integer
    no ranking -> NaN
    """
    if text is None:
        return np.nan

    text = str(text).strip()

    if not text:
        return np.nan

    if text.upper() == "C":
        return 0

    m = re.search(r"#\s*(\d+)", text)

    if m:
        value = int(m.group(1))

        if 1 <= value <= 15:
            return value

    return np.nan


ranking_rows = []

session = requests.Session()
session.headers.update(HEADERS)


for event_date, event_name, url in EVENTS:

    print(f"\n📥 {event_date} — {event_name}")

    response = session.get(
        url,
        timeout=30
    )

    if response.status_code != 200:
        print("   ❌ HTTP", response.status_code)
        continue

    soup = BeautifulSoup(
        response.text,
        "html.parser"
    )

    blocks = soup.select(
        ".c-listing-fight"
    )

    event_ranked = 0

    for block in blocks:

        # ----------------------------------------
        # FIGHTER NAMES
        # ----------------------------------------

        f1 = text_or_nan(
            block,
            ".c-listing-fight__corner-name--red"
        )

        f2 = text_or_nan(
            block,
            ".c-listing-fight__corner-name--blue"
        )

        if pd.isna(f1) or pd.isna(f2):
            continue

        f1 = clean_name(f1)
        f2 = clean_name(f2)

        f1 = confirmed_name_map.get(f1, f1)
        f2 = confirmed_name_map.get(f2, f2)


        # ----------------------------------------
        # CORNER-SPECIFIC RANKS
        # ----------------------------------------

        red_corner = block.select_one(
            ".c-listing-fight__corner--red"
        )

        blue_corner = block.select_one(
            ".c-listing-fight__corner--blue"
        )


        def corner_rank(corner):

            if corner is None:
                return np.nan

            # First try rank elements physically inside
            # that fighter's corner.
            nodes = corner.select(
                ".js-listing-fight__corner-rank"
            )

            values = []

            for node in nodes:

                txt = node.get_text(
                    " ",
                    strip=True
                )

                val = parse_rank(txt)

                if pd.notna(val):
                    values.append(val)

            if values:
                return values[0]

            return np.nan


        f1_rank = corner_rank(red_corner)
        f2_rank = corner_rank(blue_corner)


        # ----------------------------------------
        # FALLBACK:
        # UFC's desktop class area preserves two
        # positions even when only one is ranked.
        # Use ONLY if corner-specific extraction
        # did not identify either side.
        # ----------------------------------------

        if pd.isna(f1_rank) and pd.isna(f2_rank):

            desktop = block.select_one(
                ".c-listing-fight__class--desktop"
            )

            if desktop is not None:

                nodes = desktop.select(
                    ".js-listing-fight__corner-rank"
                )

                # Two nodes = red then blue
                if len(nodes) >= 2:

                    f1_rank = parse_rank(
                        nodes[0].get_text(
                            " ",
                            strip=True
                        )
                    )

                    f2_rank = parse_rank(
                        nodes[1].get_text(
                            " ",
                            strip=True
                        )
                    )


        # ----------------------------------------
        # TITLE-FIGHT CHAMPION OVERRIDE
        #
        # UFC page explicitly showed:
        # Makhachev C
        # Dern C
        # Pantoja C
        #
        # Their event-time rank is therefore 0.
        # ----------------------------------------

        champion_overrides = {
            (
                pd.Timestamp("2026-08-15"),
                "Islam Makhachev"
            ),
            (
                pd.Timestamp("2026-08-15"),
                "Mackenzie Dern"
            ),
            (
                pd.Timestamp("2026-09-19"),
                "Alexandre Pantoja"
            )
        }

        d = pd.Timestamp(event_date)

        if (d, f1) in champion_overrides:
            f1_rank = 0

        if (d, f2) in champion_overrides:
            f2_rank = 0


        event_ranked += (
            int(pd.notna(f1_rank))
            +
            int(pd.notna(f2_rank))
        )


        ranking_rows.append(
            {
                "event_date":
                    d,

                "event_name":
                    event_name,

                "f1_name":
                    f1,

                "f2_name":
                    f2,

                "f1_rank_before":
                    f1_rank,

                "f2_rank_before":
                    f2_rank
            }
        )


    print(
        "   Ranked fighter entries:",
        event_ranked
    )

    time.sleep(0.5)


# ============================================================
# BUILD TABLE
# ============================================================

new_rankings_restored = pd.DataFrame(
    ranking_rows
)


# ============================================================
# MERGE ONTO nf
# ============================================================

rank_merge = new_rankings_restored[
    [
        "event_date",
        "event_name",
        "f1_name",
        "f2_name",
        "f1_rank_before",
        "f2_rank_before"
    ]
].rename(
    columns={
        "event_date": "date"
    }
)


nf_ranked = nf.merge(
    rank_merge,
    on=[
        "date",
        "event_name",
        "f1_name",
        "f2_name"
    ],
    how="left",
    validate="one_to_one"
)


# ============================================================
# VALIDATION
# ============================================================

ranked_entries = (
    nf_ranked["f1_rank_before"].notna().sum()
    +
    nf_ranked["f2_rank_before"].notna().sum()
)

champions = (
    (nf_ranked["f1_rank_before"] == 0).sum()
    +
    (nf_ranked["f2_rank_before"] == 0).sum()
)


print("\n" + "=" * 110)
print("📊 RANKING RESTORE RESULTS")
print("=" * 110)

print("Fight rows:", len(nf_ranked))
print(
    "Events:",
    nf_ranked["event_name"].nunique()
)

print(
    "Ranked fighter entries:",
    ranked_entries
)

print(
    "Champions:",
    champions
)

print("\nExpected from our previous validated build:")
print("Ranked fighter entries: 63")
print("Champions: 3")


# ============================================================
# EVENT-BY-EVENT COUNTS
# ============================================================

event_rank_counts = []

for (date, event), g in nf_ranked.groupby(
    ["date", "event_name"],
    sort=True
):

    f1_count = g[
        "f1_rank_before"
    ].notna().sum()

    f2_count = g[
        "f2_rank_before"
    ].notna().sum()

    champ_count = (
        (g["f1_rank_before"] == 0).sum()
        +
        (g["f2_rank_before"] == 0).sum()
    )

    event_rank_counts.append(
        {
            "event_date": date,
            "event_name": event,
            "ranked_fighters":
                int(f1_count + f2_count),
            "champions":
                int(champ_count)
        }
    )


event_rank_counts = pd.DataFrame(
    event_rank_counts
)

print("\n" + "=" * 110)
print("🥊 EVENT RANKING COVERAGE")
print("=" * 110)

display(event_rank_counts)


# ============================================================
# SHOW EVERY RANKED FIGHTER
# ============================================================

ranked_long = pd.concat(
    [
        nf_ranked[
            [
                "date",
                "event_name",
                "f1_name",
                "f1_rank_before"
            ]
        ].rename(
            columns={
                "f1_name": "fighter",
                "f1_rank_before": "rank"
            }
        ),

        nf_ranked[
            [
                "date",
                "event_name",
                "f2_name",
                "f2_rank_before"
            ]
        ].rename(
            columns={
                "f2_name": "fighter",
                "f2_rank_before": "rank"
            }
        )
    ],
    ignore_index=True
)

ranked_long = ranked_long[
    ranked_long["rank"].notna()
].sort_values(
    [
        "date",
        "event_name",
        "rank",
        "fighter"
    ]
)


print("\n" + "=" * 110)
print("🏆 ALL RESTORED RANKED FIGHTERS")
print("=" * 110)

display(
    ranked_long.reset_index(drop=True)
)


print("\n" + "=" * 110)
print("🔒 SAFETY")
print("=" * 110)

print("Original nf modified: NO")
print("Original ufc_fights modified: NO")
print("Published final_index modified: NO")
print("clean modified: NO")
print("Files overwritten: NO")

print(
    "\nNew objects:",
    "new_rankings_restored, nf_ranked, "
    "event_rank_counts, ranked_long"
)

print("\n🔥 RANKING RESTORE COMPLETE")
```

```python
# ============================================================
# FULL 788-EVENT CLEAN RANK30 V2
# THROUGH SEPTEMBER 19, 2026
#
# SANDBOX ONLY
# Published model remains untouched.
# ============================================================

import pandas as pd
import numpy as np

print("=" * 110)
print("🔥 BUILDING FULL 788-EVENT CLEAN RANK30 V2")
print("=" * 110)


# ============================================================
# 1. HELPERS
# ============================================================

def zero_preserving_percentile(series):

    series = pd.to_numeric(
        series,
        errors="coerce"
    )

    result = pd.Series(
        np.nan,
        index=series.index,
        dtype=float
    )

    zero = series == 0
    positive = series > 0

    result.loc[zero] = 0.0

    if positive.sum() > 0:

        result.loc[positive] = (
            series.loc[positive]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )

    return result


def rank_points(rank):

    if pd.isna(rank):
        return 0.0

    rank = float(rank)

    if rank == 0:
        return 16.0

    if 1 <= rank <= 15:
        return 16.0 - rank

    return 0.0


def weighted_available(row, weights):

    available = {
        col: weight
        for col, weight in weights.items()
        if pd.notna(row.get(col, np.nan))
    }

    if not available:
        return np.nan

    total_weight = sum(
        available.values()
    )

    return sum(
        row[col] * weight
        for col, weight in available.items()
    ) / total_weight


# ============================================================
# 2. HISTORICAL RAW CARD-QUALITY INPUTS
# ============================================================

hist = cq_rebuilt.copy()

hist["event_date"] = pd.to_datetime(
    hist["event_date"]
)

needed_raw = [
    "event_date",
    "event_name",
    "title_fights",
    "avg_ufc_experience",
    "avg_ufc_win_pct",
    "avg_win_streak",
    "avg_finish_rate"
]

missing_hist = [
    c for c in needed_raw
    if c not in hist.columns
]

if missing_hist:
    raise ValueError(
        "cq_rebuilt missing columns: "
        + str(missing_hist)
    )

hist_raw = hist[
    needed_raw
].copy()


# ============================================================
# 3. NEW 11 EVENT RAW CARD-QUALITY INPUTS
# ============================================================

new_raw = new_event_cq_raw.copy()

new_raw["event_date"] = pd.to_datetime(
    new_raw["event_date"]
)

new_raw = new_raw[
    needed_raw
].copy()


# ============================================================
# 4. COMBINE 777 + 11
# ============================================================

cq_raw_788 = pd.concat(
    [
        hist_raw,
        new_raw
    ],
    ignore_index=True
)

cq_raw_788 = (
    cq_raw_788
    .sort_values(
        [
            "event_date",
            "event_name"
        ]
    )
    .reset_index(drop=True)
)

print("\n📊 CARD QUALITY RAW POPULATION")
print("Rows:", len(cq_raw_788))
print(
    "Unique events:",
    cq_raw_788[
        ["event_date", "event_name"]
    ].drop_duplicates().shape[0]
)

print(
    "Earliest:",
    cq_raw_788["event_date"].min()
)

print(
    "Latest:",
    cq_raw_788["event_date"].max()
)


# ============================================================
# 5. HISTORICAL RANKING RAW INPUTS
# ============================================================

hist_rank = ufc_fights.copy()

hist_rank["date"] = pd.to_datetime(
    hist_rank["date"]
)

required_rank_cols = [
    "date",
    "event_name",
    "f1_rank_before",
    "f2_rank_before",
    "is_title_fight"
]

missing_rank_cols = [
    c for c in required_rank_cols
    if c not in hist_rank.columns
]

if missing_rank_cols:
    raise ValueError(
        "Historical fight table missing: "
        + str(missing_rank_cols)
    )


# ============================================================
# 6. PREP NEW RANKING FIGHTS
# ============================================================

new_rank = nf_ranked_fixed.copy()

new_rank["date"] = pd.to_datetime(
    new_rank["date"]
)

# Guarantee title indicator exists
if "is_title_fight" not in new_rank.columns:

    new_rank["is_title_fight"] = (
        new_rank["weight_class"]
        .astype(str)
        .str.startswith("UFC ")
        &
        new_rank["weight_class"]
        .astype(str)
        .str.contains(
            "Title Bout",
            case=False,
            na=False
        )
        &
        ~new_rank["weight_class"]
        .astype(str)
        .str.contains(
            "Tournament",
            case=False,
            na=False
        )
    )


# ============================================================
# 7. COMBINE FIGHT-LEVEL RANKING DATA
# ============================================================

ranking_fights_788 = pd.concat(
    [
        hist_rank[
            required_rank_cols
        ].copy(),

        new_rank[
            required_rank_cols
        ].copy()
    ],
    ignore_index=True
)


# ============================================================
# 8. CALCULATE FIGHT-LEVEL RANKING POINTS
# ============================================================

ranking_fights_788[
    "f1_rank_points"
] = ranking_fights_788[
    "f1_rank_before"
].apply(rank_points)

ranking_fights_788[
    "f2_rank_points"
] = ranking_fights_788[
    "f2_rank_before"
].apply(rank_points)

ranking_fights_788[
    "fight_ranking_points"
] = (
    ranking_fights_788[
        "f1_rank_points"
    ]
    +
    ranking_fights_788[
        "f2_rank_points"
    ]
)

ranking_fights_788[
    "ranked_fighters_fight"
] = (
    ranking_fights_788[
        "f1_rank_before"
    ].notna().astype(int)
    +
    ranking_fights_788[
        "f2_rank_before"
    ].notna().astype(int)
)

ranking_fights_788[
    "ranked_matchup"
] = (
    ranking_fights_788[
        "f1_rank_before"
    ].notna()
    &
    ranking_fights_788[
        "f2_rank_before"
    ].notna()
).astype(int)


# ============================================================
# 9. EVENT-LEVEL RAW RANKING DEPTH
# ============================================================

ranking_event_raw = (
    ranking_fights_788
    .groupby(
        [
            "date",
            "event_name"
        ],
        as_index=False
    )
    .agg(
        total_ranking_points=(
            "fight_ranking_points",
            "sum"
        ),

        ranked_fighters=(
            "ranked_fighters_fight",
            "sum"
        ),

        ranked_matchups=(
            "ranked_matchup",
            "sum"
        ),

        ranking_title_fights=(
            "is_title_fight",
            "sum"
        )
    )
    .rename(
        columns={
            "date": "event_date"
        }
    )
)


# ============================================================
# 10. RANKINGS ONLY AVAILABLE FROM 2013-02-04
#
# Pre-ranking-era events must NOT receive zeros.
# ============================================================

ranking_start = pd.Timestamp(
    "2013-02-04"
)

ranking_event_raw[
    "rankings_available"
] = (
    ranking_event_raw[
        "event_date"
    ]
    >= ranking_start
)

rank_raw_cols = [
    "total_ranking_points",
    "ranked_fighters",
    "ranked_matchups",
    "ranking_title_fights"
]

ranking_event_raw.loc[
    ~ranking_event_raw[
        "rankings_available"
    ],
    rank_raw_cols
] = np.nan


# ============================================================
# 11. NORMALIZE RANKING DEPTH ACROSS FULL 788 SNAPSHOT
# ============================================================

ranking_event_raw[
    "total_ranking_points_score"
] = zero_preserving_percentile(
    ranking_event_raw[
        "total_ranking_points"
    ]
)

ranking_event_raw[
    "ranked_fighters_score"
] = zero_preserving_percentile(
    ranking_event_raw[
        "ranked_fighters"
    ]
)

ranking_event_raw[
    "ranked_matchups_score"
] = zero_preserving_percentile(
    ranking_event_raw[
        "ranked_matchups"
    ]
)

ranking_event_raw[
    "ranking_title_fights_score"
] = zero_preserving_percentile(
    ranking_event_raw[
        "ranking_title_fights"
    ]
)


RANKING_DEPTH_WEIGHTS = {
    "total_ranking_points_score": 0.45,
    "ranked_fighters_score": 0.25,
    "ranked_matchups_score": 0.20,
    "ranking_title_fights_score": 0.10
}


ranking_event_raw[
    "ranking_depth_index"
] = ranking_event_raw.apply(
    lambda row:
        weighted_available(
            row,
            RANKING_DEPTH_WEIGHTS
        )
        if row["rankings_available"]
        else np.nan,
    axis=1
)


# ============================================================
# 12. MERGE RANKING DEPTH INTO CARD QUALITY
# ============================================================

cq_full = cq_raw_788.merge(
    ranking_event_raw[
        [
            "event_date",
            "event_name",
            "rankings_available",
            "total_ranking_points",
            "ranked_fighters",
            "ranked_matchups",
            "ranking_title_fights",
            "ranking_depth_index"
        ]
    ],
    on=[
        "event_date",
        "event_name"
    ],
    how="left",
    validate="one_to_one"
)


# ============================================================
# 13. NORMALIZE THE FIVE NON-RANKING CARD CONCEPTS
# ACROSS THE FULL 788-EVENT POPULATION
# ============================================================

cq_full[
    "title_fights_score"
] = zero_preserving_percentile(
    cq_full["title_fights"]
)

cq_full[
    "avg_ufc_experience_score"
] = zero_preserving_percentile(
    cq_full["avg_ufc_experience"]
)

cq_full[
    "avg_ufc_win_pct_score"
] = zero_preserving_percentile(
    cq_full["avg_ufc_win_pct"]
)

cq_full[
    "avg_win_streak_score"
] = zero_preserving_percentile(
    cq_full["avg_win_streak"]
)

cq_full[
    "avg_finish_rate_score"
] = zero_preserving_percentile(
    cq_full["avg_finish_rate"]
)


# ============================================================
# 14. CLEAN RANK30 CARD QUALITY
#
# Ranking Depth = 30%
# Other five = 14% each
#
# Missing Ranking Depth is omitted and weights renormalize.
# ============================================================

CLEAN_RANK30_WEIGHTS = {
    "ranking_depth_index": 0.30,
    "title_fights_score": 0.14,
    "avg_ufc_experience_score": 0.14,
    "avg_ufc_win_pct_score": 0.14,
    "avg_win_streak_score": 0.14,
    "avg_finish_rate_score": 0.14
}

cq_full[
    "clean_rank30_card_quality"
] = cq_full.apply(
    lambda row:
        weighted_available(
            row,
            CLEAN_RANK30_WEIGHTS
        ),
    axis=1
)


# ============================================================
# 15. MERGE PUBLISHED NON-CARD COMPONENTS
#
# We are NOT recalculating Star/Main/Fanfare.
# We use the validated published values.
# ============================================================

published_components = final_index[
    [
        "event_date",
        "event_name",
        "star_experience_score",
        "main_event_score",
        "fanfare_score"
    ]
].copy()

published_components[
    "event_date"
] = pd.to_datetime(
    published_components[
        "event_date"
    ]
)


v2_full = cq_full.merge(
    published_components,
    on=[
        "event_date",
        "event_name"
    ],
    how="left",
    validate="one_to_one"
)


# ============================================================
# 16. FINAL V2 WEIGHTS
#
# Card 46
# Star 22
# Main 22
# Fanfare 10
# ============================================================

V2_FINAL_WEIGHTS = {
    "clean_rank30_card_quality": 0.46,
    "star_experience_score": 0.22,
    "main_event_score": 0.22,
    "fanfare_score": 0.10
}


v2_full[
    "v2_event_strength_score"
] = v2_full.apply(
    lambda row:
        weighted_available(
            row,
            V2_FINAL_WEIGHTS
        ),
    axis=1
)


# ============================================================
# 17. RANK ALL 788 EVENTS
# ============================================================

v2_full[
    "v2_rank"
] = (
    v2_full[
        "v2_event_strength_score"
    ]
    .rank(
        ascending=False,
        method="first"
    )
    .astype("Int64")
)


v2_full[
    "v2_percentile"
] = (
    v2_full[
        "v2_event_strength_score"
    ]
    .rank(
        pct=True,
        method="average"
    )
    * 100
)


# ============================================================
# 18. BRING IN PUBLISHED SCORE/RANK FOR COMPARISON
# ============================================================

published_compare = final_index[
    [
        "event_date",
        "event_name",
        "event_strength_rank",
        "event_strength_score"
    ]
].copy()

published_compare[
    "event_date"
] = pd.to_datetime(
    published_compare[
        "event_date"
    ]
)


v2_full = v2_full.merge(
    published_compare,
    on=[
        "event_date",
        "event_name"
    ],
    how="left",
    validate="one_to_one"
)

v2_full[
    "rank_change_vs_published"
] = (
    v2_full[
        "event_strength_rank"
    ]
    -
    v2_full[
        "v2_rank"
    ]
)

# Positive = moved UP in V2
# Negative = moved DOWN


# ============================================================
# 19. HARD VALIDATION
# ============================================================

event_count = len(v2_full)

unique_count = v2_full[
    [
        "event_date",
        "event_name"
    ]
].drop_duplicates().shape[0]

duplicates = (
    event_count
    -
    unique_count
)

missing_scores = v2_full[
    "v2_event_strength_score"
].isna().sum()

new_count = (
    v2_full[
        "event_date"
    ]
    > pd.Timestamp("2026-06-27")
).sum()


print("\n" + "=" * 110)
print("🔎 HARD VALIDATION")
print("=" * 110)

print(
    "Event rows:",
    event_count,
    "| TARGET: 788"
)

print(
    "Unique events:",
    unique_count,
    "| TARGET: 788"
)

print(
    "Duplicate events:",
    duplicates,
    "| TARGET: 0"
)

print(
    "Missing V2 scores:",
    missing_scores,
    "| TARGET: 0"
)

print(
    "New events after 2026-06-27:",
    new_count,
    "| TARGET: 11"
)

print(
    "Latest event:",
    v2_full[
        "event_date"
    ].max(),
    "| TARGET: 2026-09-19"
)


# ============================================================
# 20. SHOW THE 11 NEW EVENTS
# ============================================================

new_v2 = (
    v2_full[
        v2_full[
            "event_date"
        ]
        > pd.Timestamp("2026-06-27")
    ]
    .sort_values(
        "event_date"
    )
)

print("\n" + "=" * 110)
print("🔥 ALL 11 NEW EVENTS — CLEAN RANK30 V2")
print("=" * 110)

display(
    new_v2[
        [
            "event_date",
            "event_name",
            "ranked_fighters",
            "ranked_matchups",
            "total_ranking_points",
            "ranking_depth_index",
            "clean_rank30_card_quality",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",
            "v2_event_strength_score",
            "v2_rank",
            "event_strength_rank",
            "rank_change_vs_published"
        ]
    ].round(2)
)


# ============================================================
# 21. TOP 25 V2
# ============================================================

print("\n" + "=" * 110)
print("🏆 TOP 25 — CLEAN RANK30 V2")
print("=" * 110)

top25_v2 = (
    v2_full
    .sort_values(
        "v2_rank"
    )
    .head(25)
)

display(
    top25_v2[
        [
            "v2_rank",
            "event_date",
            "event_name",
            "v2_event_strength_score",
            "clean_rank30_card_quality",
            "ranking_depth_index",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",
            "event_strength_rank"
        ]
    ].round(2)
)


# ============================================================
# 22. TARGET EVENTS
# ============================================================

targets = [
    "UFC 269",
    "UFC 217",
    "UFC 214",
    "UFC 205",
    "UFC 300",
    "UFC 259",
    "UFC 229",
    "UFC 168",
    "UFC 302",
    "UFC 329",
    "UFC 330",
    "UFC 331"
]

target_mask = pd.Series(
    False,
    index=v2_full.index
)

for target in targets:

    target_mask |= (
        v2_full[
            "event_name"
        ]
        .str.contains(
            target,
            case=False,
            regex=False,
            na=False
        )
    )


print("\n" + "=" * 110)
print("🎯 TARGET EVENT CHECK")
print("=" * 110)

display(
    v2_full[
        target_mask
    ][
        [
            "v2_rank",
            "event_date",
            "event_name",
            "v2_event_strength_score",
            "clean_rank30_card_quality",
            "ranking_depth_index",
            "event_strength_rank",
            "rank_change_vs_published"
        ]
    ]
    .sort_values(
        "v2_rank"
    )
    .round(2)
)


# ============================================================
# 23. SAFETY
# ============================================================

print("\n" + "=" * 110)
print("🔒 SAFETY")
print("=" * 110)

print("Published final_index modified: NO")
print("Published files overwritten: NO")
print("Original clean modified: NO")
print("Original model_d modified: NO")
print()
print("New sandbox object: v2_full")

print("\n🔥 FULL 788-EVENT V2 BUILD COMPLETE")
```

```python
# ============================================================
# FULL 788-EVENT CLEAN RANK30 V2 — CORRECTED
# THROUGH SEPTEMBER 19, 2026
# SANDBOX ONLY
# ============================================================

import pandas as pd
import numpy as np

print("=" * 110)
print("🔥 BUILDING FULL 788-EVENT CLEAN RANK30 V2 — CORRECTED")
print("=" * 110)


# ============================================================
# 1. HELPERS
# ============================================================

def zero_preserving_percentile(series):
    s = pd.to_numeric(series, errors="coerce")

    result = pd.Series(
        np.nan,
        index=s.index,
        dtype=float
    )

    result.loc[s == 0] = 0.0

    positive = s > 0

    if positive.sum() > 0:
        result.loc[positive] = (
            s.loc[positive]
            .rank(pct=True, method="average")
            * 100
        )

    return result


def rank_points(rank):
    if pd.isna(rank):
        return 0.0

    rank = float(rank)

    if rank == 0:
        return 16.0

    if 1 <= rank <= 15:
        return 16.0 - rank

    return 0.0


def weighted_available(row, weights):
    available = {
        col: weight
        for col, weight in weights.items()
        if pd.notna(row.get(col, np.nan))
    }

    if not available:
        return np.nan

    total = sum(available.values())

    return sum(
        row[col] * weight
        for col, weight in available.items()
    ) / total


# ============================================================
# 2. HISTORICAL 777 RAW CARD INPUTS
# ============================================================

hist_raw = cq_rebuilt[
    [
        "event_date",
        "event_name",
        "title_fights_rebuilt",
        "avg_ufc_experience_rebuilt",
        "avg_ufc_win_pct_rebuilt",
        "avg_win_streak_rebuilt",
        "avg_finish_rate_rebuilt"
    ]
].copy()

hist_raw = hist_raw.rename(
    columns={
        "title_fights_rebuilt":
            "title_fights",

        "avg_ufc_experience_rebuilt":
            "avg_ufc_experience",

        "avg_ufc_win_pct_rebuilt":
            "avg_ufc_win_pct",

        "avg_win_streak_rebuilt":
            "avg_win_streak",

        "avg_finish_rate_rebuilt":
            "avg_finish_rate"
    }
)

hist_raw["event_date"] = pd.to_datetime(
    hist_raw["event_date"]
)


# ============================================================
# 3. NEW 11 RAW CARD INPUTS
# ============================================================

new_raw = new_event_cq_raw[
    [
        "event_date",
        "event_name",
        "title_fights",
        "avg_ufc_experience",
        "avg_ufc_win_pct",
        "avg_win_streak",
        "avg_finish_rate"
    ]
].copy()

new_raw["event_date"] = pd.to_datetime(
    new_raw["event_date"]
)


# ============================================================
# 4. COMBINE TO EXACTLY 788 EVENTS
# ============================================================

cq_raw_788 = pd.concat(
    [hist_raw, new_raw],
    ignore_index=True
)

cq_raw_788 = (
    cq_raw_788
    .sort_values(
        ["event_date", "event_name"]
    )
    .reset_index(drop=True)
)

unique_cq = (
    cq_raw_788[
        ["event_date", "event_name"]
    ]
    .drop_duplicates()
    .shape[0]
)

print("\n📊 CARD INPUT POPULATION")
print("Historical:", len(hist_raw))
print("New:", len(new_raw))
print("Combined:", len(cq_raw_788))
print("Unique:", unique_cq)

if len(cq_raw_788) != 788 or unique_cq != 788:
    raise ValueError(
        "Card population is not exactly 788 unique events."
    )


# ============================================================
# 5. COMBINE HISTORICAL + NEW RANKING FIGHTS
# ============================================================

required_rank_cols = [
    "date",
    "event_name",
    "f1_rank_before",
    "f2_rank_before",
    "is_title_fight"
]

hist_rank = ufc_fights[
    required_rank_cols
].copy()

hist_rank["date"] = pd.to_datetime(
    hist_rank["date"]
)

new_rank = nf_ranked_fixed.copy()

new_rank["date"] = pd.to_datetime(
    new_rank["date"]
)

# Make sure title indicator exists
if "is_title_fight" not in new_rank.columns:

    new_rank["is_title_fight"] = (
        new_rank["weight_class"]
        .astype(str)
        .str.startswith("UFC ")
        &
        new_rank["weight_class"]
        .astype(str)
        .str.contains(
            "Title Bout",
            case=False,
            na=False
        )
        &
        ~new_rank["weight_class"]
        .astype(str)
        .str.contains(
            "Tournament",
            case=False,
            na=False
        )
    )

new_rank = new_rank[
    required_rank_cols
].copy()


ranking_fights_788 = pd.concat(
    [hist_rank, new_rank],
    ignore_index=True
)

print("\n🥊 FIGHT POPULATION")
print("Historical fights:", len(hist_rank))
print("New fights:", len(new_rank))
print("Combined fights:", len(ranking_fights_788))

if len(ranking_fights_788) != 8877:
    raise ValueError(
        f"Expected 8,877 fights, got {len(ranking_fights_788)}"
    )


# ============================================================
# 6. FIGHT-LEVEL RANKING METRICS
# ============================================================

ranking_fights_788["f1_rank_points"] = (
    ranking_fights_788[
        "f1_rank_before"
    ].apply(rank_points)
)

ranking_fights_788["f2_rank_points"] = (
    ranking_fights_788[
        "f2_rank_before"
    ].apply(rank_points)
)

ranking_fights_788["fight_ranking_points"] = (
    ranking_fights_788["f1_rank_points"]
    +
    ranking_fights_788["f2_rank_points"]
)

ranking_fights_788["ranked_fighters_fight"] = (
    ranking_fights_788[
        "f1_rank_before"
    ].notna().astype(int)
    +
    ranking_fights_788[
        "f2_rank_before"
    ].notna().astype(int)
)

ranking_fights_788["ranked_matchup"] = (
    ranking_fights_788[
        "f1_rank_before"
    ].notna()
    &
    ranking_fights_788[
        "f2_rank_before"
    ].notna()
).astype(int)


# ============================================================
# 7. EVENT-LEVEL RANKING RAW METRICS
# ============================================================

ranking_event_raw = (
    ranking_fights_788
    .groupby(
        ["date", "event_name"],
        as_index=False
    )
    .agg(
        total_ranking_points=(
            "fight_ranking_points",
            "sum"
        ),
        ranked_fighters=(
            "ranked_fighters_fight",
            "sum"
        ),
        ranked_matchups=(
            "ranked_matchup",
            "sum"
        ),
        ranking_title_fights=(
            "is_title_fight",
            "sum"
        )
    )
    .rename(
        columns={"date": "event_date"}
    )
)


# ============================================================
# 8. PRE-2013 RANKING DATA = UNAVAILABLE, NOT ZERO
# ============================================================

ranking_start = pd.Timestamp("2013-02-04")

ranking_event_raw["rankings_available"] = (
    ranking_event_raw["event_date"]
    >= ranking_start
)

ranking_raw_cols = [
    "total_ranking_points",
    "ranked_fighters",
    "ranked_matchups",
    "ranking_title_fights"
]

ranking_event_raw.loc[
    ~ranking_event_raw["rankings_available"],
    ranking_raw_cols
] = np.nan


# ============================================================
# 9. FULL-SNAPSHOT RANKING NORMALIZATION
# ============================================================

ranking_event_raw["total_ranking_points_score"] = (
    zero_preserving_percentile(
        ranking_event_raw["total_ranking_points"]
    )
)

ranking_event_raw["ranked_fighters_score"] = (
    zero_preserving_percentile(
        ranking_event_raw["ranked_fighters"]
    )
)

ranking_event_raw["ranked_matchups_score"] = (
    zero_preserving_percentile(
        ranking_event_raw["ranked_matchups"]
    )
)

ranking_event_raw["ranking_title_fights_score"] = (
    zero_preserving_percentile(
        ranking_event_raw["ranking_title_fights"]
    )
)


RANKING_DEPTH_WEIGHTS = {
    "total_ranking_points_score": 0.45,
    "ranked_fighters_score": 0.25,
    "ranked_matchups_score": 0.20,
    "ranking_title_fights_score": 0.10
}

ranking_event_raw["ranking_depth_index"] = (
    ranking_event_raw.apply(
        lambda row:
            weighted_available(
                row,
                RANKING_DEPTH_WEIGHTS
            )
            if row["rankings_available"]
            else np.nan,
        axis=1
    )
)


# ============================================================
# 10. MERGE RANKING DEPTH INTO ALL 788 CARDS
# ============================================================

cq_full = cq_raw_788.merge(
    ranking_event_raw[
        [
            "event_date",
            "event_name",
            "rankings_available",
            "total_ranking_points",
            "ranked_fighters",
            "ranked_matchups",
            "ranking_title_fights",
            "ranking_depth_index"
        ]
    ],
    on=["event_date", "event_name"],
    how="left",
    validate="one_to_one"
)


# ============================================================
# 11. RE-NORMALIZE FIVE OTHER CARD CONCEPTS
# ACROSS THE FULL 788-EVENT SNAPSHOT
# ============================================================

for raw_col, score_col in [
    (
        "title_fights",
        "title_fights_score"
    ),
    (
        "avg_ufc_experience",
        "avg_ufc_experience_score"
    ),
    (
        "avg_ufc_win_pct",
        "avg_ufc_win_pct_score"
    ),
    (
        "avg_win_streak",
        "avg_win_streak_score"
    ),
    (
        "avg_finish_rate",
        "avg_finish_rate_score"
    )
]:

    cq_full[score_col] = (
        zero_preserving_percentile(
            cq_full[raw_col]
        )
    )


# ============================================================
# 12. CLEAN RANK30 CARD QUALITY
# ============================================================

CLEAN_RANK30_WEIGHTS = {
    "ranking_depth_index": 0.30,
    "title_fights_score": 0.14,
    "avg_ufc_experience_score": 0.14,
    "avg_ufc_win_pct_score": 0.14,
    "avg_win_streak_score": 0.14,
    "avg_finish_rate_score": 0.14
}

cq_full["clean_rank30_card_quality"] = (
    cq_full.apply(
        lambda row:
            weighted_available(
                row,
                CLEAN_RANK30_WEIGHTS
            ),
        axis=1
    )
)


# ============================================================
# 13. MERGE VALIDATED STAR / MAIN / FANFARE COMPONENTS
# ============================================================

published = final_index.copy()

published["event_date"] = pd.to_datetime(
    published["event_date"]
)

v2_full = cq_full.merge(
    published[
        [
            "event_date",
            "event_name",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",
            "event_strength_rank",
            "event_strength_score"
        ]
    ],
    on=["event_date", "event_name"],
    how="left",
    validate="one_to_one"
)


# ============================================================
# 14. V2 OVERALL MODEL = 46 / 22 / 22 / 10
# ============================================================

V2_FINAL_WEIGHTS = {
    "clean_rank30_card_quality": 0.46,
    "star_experience_score": 0.22,
    "main_event_score": 0.22,
    "fanfare_score": 0.10
}

v2_full["v2_event_strength_score"] = (
    v2_full.apply(
        lambda row:
            weighted_available(
                row,
                V2_FINAL_WEIGHTS
            ),
        axis=1
    )
)

v2_full["v2_rank"] = (
    v2_full["v2_event_strength_score"]
    .rank(
        ascending=False,
        method="first"
    )
    .astype("Int64")
)

v2_full["v2_percentile"] = (
    v2_full["v2_event_strength_score"]
    .rank(
        pct=True,
        method="average"
    )
    * 100
)

v2_full["rank_change_vs_published"] = (
    v2_full["event_strength_rank"]
    -
    v2_full["v2_rank"]
)


# ============================================================
# 15. HARD VALIDATION
# ============================================================

event_rows = len(v2_full)

unique_events = (
    v2_full[
        ["event_date", "event_name"]
    ]
    .drop_duplicates()
    .shape[0]
)

duplicate_events = (
    event_rows - unique_events
)

missing_v2 = (
    v2_full[
        "v2_event_strength_score"
    ]
    .isna()
    .sum()
)

new_events = (
    v2_full["event_date"]
    > pd.Timestamp("2026-06-27")
).sum()


print("\n" + "=" * 110)
print("🔎 HARD VALIDATION")
print("=" * 110)

print("Event rows:", event_rows, "| TARGET 788")
print("Unique events:", unique_events, "| TARGET 788")
print("Duplicates:", duplicate_events, "| TARGET 0")
print("Missing V2 scores:", missing_v2, "| TARGET 0")
print("New events:", new_events, "| TARGET 11")
print(
    "Latest:",
    v2_full["event_date"].max(),
    "| TARGET 2026-09-19"
)


# ============================================================
# 16. 11 NEW EVENTS
# ============================================================

new_v2 = (
    v2_full[
        v2_full["event_date"]
        > pd.Timestamp("2026-06-27")
    ]
    .sort_values("event_date")
)

print("\n" + "=" * 110)
print("🔥 11 NEW EVENTS — CLEAN RANK30 V2")
print("=" * 110)

display(
    new_v2[
        [
            "event_date",
            "event_name",
            "title_fights",
            "ranked_fighters",
            "ranked_matchups",
            "total_ranking_points",
            "ranking_depth_index",
            "clean_rank30_card_quality",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",
            "v2_event_strength_score",
            "v2_rank",
            "event_strength_rank",
            "rank_change_vs_published"
        ]
    ].round(2)
)


# ============================================================
# 17. TOP 25
# ============================================================

top25_v2 = (
    v2_full
    .sort_values("v2_rank")
    .head(25)
)

print("\n" + "=" * 110)
print("🏆 TOP 25 — FULL CLEAN RANK30 V2")
print("=" * 110)

display(
    top25_v2[
        [
            "v2_rank",
            "event_date",
            "event_name",
            "v2_event_strength_score",
            "clean_rank30_card_quality",
            "ranking_depth_index",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",
            "event_strength_rank"
        ]
    ].round(2)
)


# ============================================================
# 18. TARGET CARDS
# ============================================================

targets = [
    "UFC 269",
    "UFC 217",
    "UFC 214",
    "UFC 205",
    "UFC 300",
    "UFC 259",
    "UFC 229",
    "UFC 168",
    "UFC 302",
    "UFC 329",
    "UFC 330",
    "UFC 331"
]

target_mask = pd.Series(
    False,
    index=v2_full.index
)

for target in targets:
    target_mask |= (
        v2_full["event_name"]
        .str.contains(
            target,
            case=False,
            regex=False,
            na=False
        )
    )

print("\n" + "=" * 110)
print("🎯 TARGET EVENT CHECK")
print("=" * 110)

display(
    v2_full.loc[
        target_mask,
        [
            "v2_rank",
            "event_date",
            "event_name",
            "v2_event_strength_score",
            "clean_rank30_card_quality",
            "ranking_depth_index",
            "event_strength_rank",
            "rank_change_vs_published"
        ]
    ]
    .sort_values("v2_rank")
    .round(2)
)


# ============================================================
# 19. SAFETY
# ============================================================

print("\n" + "=" * 110)
print("🔒 SAFETY")
print("=" * 110)

print("Published final_index modified: NO")
print("Original clean modified: NO")
print("Original model_d modified: NO")
print("Files overwritten: NO")
print("New sandbox object: v2_full")

print("\n🔥 FULL 788-EVENT CLEAN RANK30 V2 COMPLETE")
```

```python
# ============================================================
# FINAL V2 MODEL AUDIT
# Clean Rank30 + 46 / 22 / 22 / 10
# ============================================================

import pandas as pd
import numpy as np

print("=" * 110)
print("🔥 FINAL V2 MODEL AUDIT")
print("=" * 110)


# ============================================================
# 1. OVERALL STABILITY
# ============================================================

audit = v2_full.copy()

audit["abs_rank_movement"] = (
    audit["v2_rank"] -
    audit["event_strength_rank"]
).abs()

corr_scores = audit[
    ["event_strength_score", "v2_event_strength_score"]
].corr().iloc[0, 1]

corr_ranks = audit[
    ["event_strength_rank", "v2_rank"]
].corr(
    method="spearman"
).iloc[0, 1]

print("\n📊 OVERALL STABILITY")
print("-" * 70)

print(
    "Score correlation:",
    round(corr_scores, 4)
)

print(
    "Rank correlation:",
    round(corr_ranks, 4)
)

print(
    "Median absolute rank movement:",
    round(audit["abs_rank_movement"].median(), 2)
)

print(
    "Average absolute rank movement:",
    round(audit["abs_rank_movement"].mean(), 2)
)

print(
    "Events moving 25+ spots:",
    int((audit["abs_rank_movement"] >= 25).sum())
)

print(
    "Events moving 50+ spots:",
    int((audit["abs_rank_movement"] >= 50).sum())
)

print(
    "Events moving 100+ spots:",
    int((audit["abs_rank_movement"] >= 100).sum())
)


# ============================================================
# 2. BIGGEST 25 RISERS
#
# Positive rank_change_vs_published = moves UP
# ============================================================

risers = (
    audit
    .sort_values(
        "rank_change_vs_published",
        ascending=False
    )
    .head(25)
)

print("\n" + "=" * 110)
print("📈 BIGGEST 25 RISERS")
print("=" * 110)

display(
    risers[
        [
            "event_date",
            "event_name",
            "event_strength_rank",
            "v2_rank",
            "rank_change_vs_published",
            "event_strength_score",
            "v2_event_strength_score",
            "clean_rank30_card_quality",
            "ranking_depth_index",
            "star_experience_score",
            "main_event_score",
            "fanfare_score"
        ]
    ].round(2)
)


# ============================================================
# 3. BIGGEST 25 FALLERS
# ============================================================

fallers = (
    audit
    .sort_values(
        "rank_change_vs_published",
        ascending=True
    )
    .head(25)
)

print("\n" + "=" * 110)
print("📉 BIGGEST 25 FALLERS")
print("=" * 110)

display(
    fallers[
        [
            "event_date",
            "event_name",
            "event_strength_rank",
            "v2_rank",
            "rank_change_vs_published",
            "event_strength_score",
            "v2_event_strength_score",
            "clean_rank30_card_quality",
            "ranking_depth_index",
            "star_experience_score",
            "main_event_score",
            "fanfare_score"
        ]
    ].round(2)
)


# ============================================================
# 4. RANKING DEPTH VS FINAL SCORE
#
# Find cards where ranking depth and overall V2 strength
# disagree the most.
# ============================================================

rank_era = audit[
    audit["ranking_depth_index"].notna()
].copy()

rank_era["ranking_depth_percentile"] = (
    rank_era["ranking_depth_index"]
    .rank(
        pct=True,
        method="average"
    )
    * 100
)

rank_era["v2_score_percentile"] = (
    rank_era["v2_event_strength_score"]
    .rank(
        pct=True,
        method="average"
    )
    * 100
)

rank_era["ranking_vs_final_gap"] = (
    rank_era["ranking_depth_percentile"]
    -
    rank_era["v2_score_percentile"]
)


# ============================================================
# 5. MUCH DEEPER CARD THAN FINAL SCORE SUGGESTS
# ============================================================

deep_underall = (
    rank_era
    .sort_values(
        "ranking_vs_final_gap",
        ascending=False
    )
    .head(20)
)

print("\n" + "=" * 110)
print("🥊 DEEPEST CARDS RELATIVE TO THEIR FINAL V2 SCORE")
print("=" * 110)

display(
    deep_underall[
        [
            "event_date",
            "event_name",
            "ranking_depth_index",
            "ranking_depth_percentile",
            "clean_rank30_card_quality",
            "v2_event_strength_score",
            "v2_score_percentile",
            "v2_rank",
            "ranking_vs_final_gap"
        ]
    ].round(2)
)


# ============================================================
# 6. HIGH FINAL SCORE DESPITE WEAKER RANKING DEPTH
# ============================================================

final_overdepth = (
    rank_era
    .sort_values(
        "ranking_vs_final_gap",
        ascending=True
    )
    .head(20)
)

print("\n" + "=" * 110)
print("⭐ STRONGEST FINAL SCORES RELATIVE TO RANKING DEPTH")
print("=" * 110)

display(
    final_overdepth[
        [
            "event_date",
            "event_name",
            "ranking_depth_index",
            "ranking_depth_percentile",
            "clean_rank30_card_quality",
            "star_experience_score",
            "main_event_score",
            "fanfare_score",
            "v2_event_strength_score",
            "v2_rank",
            "ranking_vs_final_gap"
        ]
    ].round(2)
)


# ============================================================
# 7. TOP 50 MOVEMENT CHECK
#
# This is especially important because we don't want V2
# randomly destroying the top of the model.
# ============================================================

published_top50 = audit[
    audit["event_strength_rank"] <= 50
].copy()

print("\n" + "=" * 110)
print("🏆 PUBLISHED TOP 50 — MOVEMENT UNDER V2")
print("=" * 110)

print(
    "Published Top-50 still in V2 Top-50:",
    int((published_top50["v2_rank"] <= 50).sum()),
    "/ 50"
)

print(
    "Published Top-50 falling outside V2 Top-75:",
    int((published_top50["v2_rank"] > 75).sum())
)

print(
    "Published Top-50 falling outside V2 Top-100:",
    int((published_top50["v2_rank"] > 100).sum())
)

print(
    "Median movement among published Top-50:",
    round(
        published_top50["abs_rank_movement"].median(),
        2
    )
)


# ============================================================
# 8. EXTREME MOVEMENT INSPECTION
# ============================================================

extreme = (
    audit[
        audit["abs_rank_movement"] >= 100
    ]
    .sort_values(
        "abs_rank_movement",
        ascending=False
    )
)

print("\n" + "=" * 110)
print("🚨 EVENTS MOVING 100+ POSITIONS")
print("=" * 110)

if len(extreme) == 0:
    print("✅ None")
else:
    display(
        extreme[
            [
                "event_date",
                "event_name",
                "event_strength_rank",
                "v2_rank",
                "rank_change_vs_published",
                "clean_rank30_card_quality",
                "ranking_depth_index",
                "star_experience_score",
                "main_event_score",
                "fanfare_score"
            ]
        ].round(2)
    )


# ============================================================
# 9. KEY EVENT SUMMARY
# ============================================================

key_names = [
    "UFC 269",
    "UFC 217",
    "UFC 214",
    "UFC 205",
    "UFC 300",
    "UFC 259",
    "UFC 229",
    "UFC 168",
    "UFC 302",
    "UFC 329",
    "UFC 330",
    "UFC 331"
]

key_mask = pd.Series(
    False,
    index=audit.index
)

for name in key_names:
    key_mask |= (
        audit["event_name"]
        .str.contains(
            name,
            case=False,
            regex=False,
            na=False
        )
    )

print("\n" + "=" * 110)
print("🎯 KEY EVENT FINAL AUDIT")
print("=" * 110)

display(
    audit.loc[
        key_mask,
        [
            "event_name",
            "event_strength_rank",
            "v2_rank",
            "rank_change_vs_published",
            "ranking_depth_index",
            "clean_rank30_card_quality",
            "v2_event_strength_score"
        ]
    ]
    .sort_values("v2_rank")
    .round(2)
)


# ============================================================
# 10. SAFETY
# ============================================================

print("\n" + "=" * 110)
print("🔒 SAFETY")
print("=" * 110)

print("v2_full modified: NO")
print("final_index modified: NO")
print("Published model modified: NO")
print("Files overwritten: NO")

print("\n🔥 AUDIT COMPLETE")
```

```python
# ======================================================================================
# WHOLE-CARD FANFARE V2
# BIG STEP 1 — BUILD + VALIDATE FULL 8,877-FIGHT CARD STRUCTURE
#
# GOALS:
#   1. Combine 8,736 historical fights + 141 new fights
#   2. Preserve original within-event row order
#   3. Reconstruct bout position from the TOP of every card
#   4. Identify Main Event + Co-Main Event
#   5. Create placement weights for EVERY fight
#   6. Build a fighter-level card table
#   7. Run validation / diagnostics
#
# IMPORTANT:
#   - NO published-model changes
#   - NO V2 changes
#   - NO Fanfare changes
#   - NO files overwritten
# ======================================================================================

import pandas as pd
import numpy as np

print("=" * 115)
print("🔥 WHOLE-CARD FANFARE V2 — FULL CARD STRUCTURE BUILD")
print("=" * 115)


# ======================================================================================
# 1. PREP HISTORICAL FIGHTS
# ======================================================================================

hist = ufc_fights.copy()

hist["date"] = pd.to_datetime(hist["date"])

# Preserve exact existing order.
hist["_original_order"] = np.arange(len(hist))

hist["data_period"] = "historical"


# ======================================================================================
# 2. PREP THE 141 NEW FIGHTS
# ======================================================================================

new = nf_ranked_fixed.copy()

new["date"] = pd.to_datetime(new["date"])

# Continue order after historical table.
new["_original_order"] = (
    np.arange(len(new)) + len(hist)
)

new["data_period"] = "new"


# ======================================================================================
# 3. KEEP COMMON STRUCTURAL COLUMNS
# ======================================================================================

wanted = [
    "date",
    "event_name",
    "f1_name",
    "f2_name",
    "weight_class",
    "is_title_fight",

    "f1_ufc_fights_before",
    "f2_ufc_fights_before",

    "f1_ufc_win_pct_before",
    "f2_ufc_win_pct_before",

    "f1_win_streak_before",
    "f2_win_streak_before",

    "f1_finish_rate_before",
    "f2_finish_rate_before",

    "f1_rank_before",
    "f2_rank_before",

    "is_main_event",

    "_original_order",
    "data_period"
]


# Add missing structural fields as NaN rather than inventing values.
for col in wanted:

    if col not in hist.columns:
        hist[col] = np.nan

    if col not in new.columns:
        new[col] = np.nan


hist_card = hist[wanted].copy()
new_card  = new[wanted].copy()


# ======================================================================================
# 4. COMBINE ALL 8,877 FIGHTS
# ======================================================================================

all_card_fights = pd.concat(
    [hist_card, new_card],
    ignore_index=True
)

all_card_fights = all_card_fights.sort_values(
    "_original_order",
    kind="stable"
).reset_index(drop=True)


print("\n📦 FULL FIGHT TABLE")
print("-" * 75)

print("Total fights:", len(all_card_fights))
print(
    "Historical:",
    int((all_card_fights["data_period"] == "historical").sum())
)
print(
    "New:",
    int((all_card_fights["data_period"] == "new").sum())
)

print(
    "Events:",
    all_card_fights[
        ["date", "event_name"]
    ].drop_duplicates().shape[0]
)

print(
    "Date range:",
    all_card_fights["date"].min(),
    "→",
    all_card_fights["date"].max()
)


# ======================================================================================
# 5. RECONSTRUCT WITHIN-CARD ORDER
#
# Our historical examples show rows running:
# earliest/lowest bout -> ... -> co-main -> main event
#
# Therefore:
#
# row 1 in event             = bottom of card
# final row                 = main event
# second-to-last row        = co-main
#
# bout_from_top:
# 1 = Main Event
# 2 = Co-Main
# 3 = third fight from top
# etc.
# ======================================================================================

grp = all_card_fights.groupby(
    ["date", "event_name"],
    sort=False
)

all_card_fights["total_fights_on_card"] = (
    grp["event_name"].transform("size")
)

all_card_fights["bout_from_bottom"] = (
    grp.cumcount() + 1
)

all_card_fights["bout_from_top"] = (
    all_card_fights["total_fights_on_card"]
    -
    all_card_fights["bout_from_bottom"]
    + 1
)

all_card_fights["is_main_event_rebuilt"] = (
    all_card_fights["bout_from_top"] == 1
)

all_card_fights["is_co_main_rebuilt"] = (
    all_card_fights["bout_from_top"] == 2
)


# ======================================================================================
# 6. VALIDATE REBUILT MAIN EVENTS AGAINST KNOWN HISTORICAL MAIN EVENTS
# ======================================================================================

hist_validation = all_card_fights[
    all_card_fights["data_period"] == "historical"
].copy()

# Historical is_main_event is known.
known_main = (
    hist_validation["is_main_event"] == True
)

rebuilt_main = (
    hist_validation["is_main_event_rebuilt"] == True
)

main_event_matches = (
    known_main == rebuilt_main
)

print("\n" + "=" * 115)
print("🧪 HISTORICAL MAIN-EVENT ORDER VALIDATION")
print("=" * 115)

print(
    "Historical fight rows:",
    len(hist_validation)
)

print(
    "Known main-event rows:",
    int(known_main.sum())
)

print(
    "Rebuilt main-event rows:",
    int(rebuilt_main.sum())
)

print(
    "Exact row matches:",
    int(main_event_matches.sum()),
    "/",
    len(hist_validation)
)

print(
    "Mismatching rows:",
    int((~main_event_matches).sum())
)

print(
    "Row-level agreement:",
    round(main_event_matches.mean() * 100, 4),
    "%"
)


# ======================================================================================
# 7. EVENT-LEVEL VALIDATION
#
# We care especially about whether the final row identifies
# the same main event as our verified indicator.
# ======================================================================================

event_validation = (
    hist_validation
    .groupby(
        ["date", "event_name"],
        as_index=False
    )
    .agg(
        known_main_count=(
            "is_main_event",
            lambda x: (x == True).sum()
        ),
        rebuilt_main_count=(
            "is_main_event_rebuilt",
            "sum"
        )
    )
)

event_validation["counts_match"] = (
    event_validation["known_main_count"]
    ==
    event_validation["rebuilt_main_count"]
)

print("\n📊 EVENT-LEVEL CHECK")
print("-" * 75)

print(
    "Historical events:",
    len(event_validation)
)

print(
    "Events with exactly 1 known main event:",
    int(
        (
            event_validation["known_main_count"] == 1
        ).sum()
    )
)

print(
    "Events with exactly 1 rebuilt main event:",
    int(
        (
            event_validation["rebuilt_main_count"] == 1
        ).sum()
    )
)


# ======================================================================================
# 8. SHOW ACTUAL MAIN-EVENT DISAGREEMENTS
# ======================================================================================

disagreement_events = hist_validation.loc[
    ~main_event_matches,
    ["date", "event_name"]
].drop_duplicates()

print("\n🚨 EVENTS WITH ROW-ORDER / MAIN-EVENT DISAGREEMENT")
print("-" * 75)

print(
    "Events affected:",
    len(disagreement_events)
)

if len(disagreement_events) > 0:

    disagreement_detail = (
        hist_validation.merge(
            disagreement_events,
            on=["date", "event_name"],
            how="inner"
        )
        [
            [
                "date",
                "event_name",
                "f1_name",
                "f2_name",
                "bout_from_top",
                "is_main_event",
                "is_main_event_rebuilt"
            ]
        ]
    )

    display(
        disagreement_detail.head(100)
    )

else:

    print("✅ No disagreements.")


# ======================================================================================
# 9. IMPORTANT:
# USE VERIFIED MAIN EVENT WHEN AVAILABLE.
#
# Historical:
#   verified is_main_event wins.
#
# New 11:
#   use row-order reconstruction because those rows currently
#   do not have the historical verified flag.
# ======================================================================================

all_card_fights["is_main_event_final"] = np.where(
    all_card_fights["data_period"] == "historical",
    all_card_fights["is_main_event"] == True,
    all_card_fights["is_main_event_rebuilt"]
)

# Co-main remains order-based.
all_card_fights["is_co_main_final"] = (
    all_card_fights["bout_from_top"] == 2
)


# ======================================================================================
# 10. CREATE BOUT PLACEMENT TIERS
#
# These are STRUCTURAL labels only.
# We are NOT locking final Fanfare weights yet.
# ======================================================================================

def placement_tier(row):

    pos = row["bout_from_top"]

    if pos == 1:
        return "Main Event"

    elif pos == 2:
        return "Co-Main Event"

    elif pos <= 5:
        return "Upper Card"

    elif pos <= 8:
        return "Middle Card"

    else:
        return "Lower Card"


all_card_fights["card_position_tier"] = (
    all_card_fights.apply(
        placement_tier,
        axis=1
    )
)


# ======================================================================================
# 11. TEMPORARY PLACEMENT WEIGHTS
#
# IMPORTANT:
# These are NOT final model weights.
#
# They simply let us build/test a whole-card framework.
#
# Main event gets most individual weight,
# but the rest of the card collectively matters heavily.
# ======================================================================================

TEMP_PLACEMENT_WEIGHTS = {
    "Main Event": 1.00,
    "Co-Main Event": 0.85,
    "Upper Card": 0.70,
    "Middle Card": 0.50,
    "Lower Card": 0.30
}

all_card_fights["temporary_placement_weight"] = (
    all_card_fights[
        "card_position_tier"
    ].map(
        TEMP_PLACEMENT_WEIGHTS
    )
)


# ======================================================================================
# 12. CARD-WEIGHT SHARE
#
# Normalize within each event.
#
# This is useful because a 14-fight card should not automatically
# receive a larger score simply because it has more fights.
# ======================================================================================

all_card_fights["card_weight_total"] = (
    all_card_fights.groupby(
        ["date", "event_name"]
    )[
        "temporary_placement_weight"
    ].transform("sum")
)

all_card_fights["bout_weight_share"] = (
    all_card_fights["temporary_placement_weight"]
    /
    all_card_fights["card_weight_total"]
)


# ======================================================================================
# 13. HOW MUCH OF EACH CARD'S WEIGHT BELONGS TO THE MAIN EVENT?
#
# This directly tests your concern:
# Fanfare should NOT basically equal main-event fanfare.
# ======================================================================================

weight_structure = (
    all_card_fights
    .groupby(
        ["date", "event_name"],
        as_index=False
    )
    .agg(
        total_fights=(
            "f1_name",
            "size"
        ),
        main_event_weight_share=(
            "bout_weight_share",
            lambda x: x.iloc[-1]
        )
    )
)

print("\n" + "=" * 115)
print("⚖️ WHOLE-CARD WEIGHT STRUCTURE")
print("=" * 115)

print(
    "Average main-event share:",
    round(
        weight_structure[
            "main_event_weight_share"
        ].mean() * 100,
        2
    ),
    "%"
)

print(
    "Median main-event share:",
    round(
        weight_structure[
            "main_event_weight_share"
        ].median() * 100,
        2
    ),
    "%"
)

print(
    "Average REST-OF-CARD share:",
    round(
        (
            1 -
            weight_structure[
                "main_event_weight_share"
            ].mean()
        ) * 100,
        2
    ),
    "%"
)


# ======================================================================================
# 14. CONVERT TO FIGHTER-LEVEL TABLE
#
# Every fight becomes TWO fighter rows.
#
# This is the backbone for whole-card popularity.
# ======================================================================================

fighter_1 = all_card_fights[
    [
        "date",
        "event_name",
        "f1_name",
        "bout_from_top",
        "card_position_tier",
        "bout_weight_share",
        "is_main_event_final",
        "is_co_main_final",
        "f1_rank_before"
    ]
].copy()

fighter_1 = fighter_1.rename(
    columns={
        "f1_name": "fighter_name",
        "f1_rank_before": "rank_before"
    }
)

fighter_1["corner"] = "f1"


fighter_2 = all_card_fights[
    [
        "date",
        "event_name",
        "f2_name",
        "bout_from_top",
        "card_position_tier",
        "bout_weight_share",
        "is_main_event_final",
        "is_co_main_final",
        "f2_rank_before"
    ]
].copy()

fighter_2 = fighter_2.rename(
    columns={
        "f2_name": "fighter_name",
        "f2_rank_before": "rank_before"
    }
)

fighter_2["corner"] = "f2"


whole_card_fighters = pd.concat(
    [
        fighter_1,
        fighter_2
    ],
    ignore_index=True
)

# Split each bout's weight evenly between its two fighters.
whole_card_fighters[
    "fighter_card_weight_share"
] = (
    whole_card_fighters[
        "bout_weight_share"
    ] / 2
)


# ======================================================================================
# 15. VALIDATE FIGHTER WEIGHTS SUM TO 1 PER EVENT
# ======================================================================================

fighter_weight_check = (
    whole_card_fighters
    .groupby(
        ["date", "event_name"]
    )[
        "fighter_card_weight_share"
    ]
    .sum()
)

print("\n" + "=" * 115)
print("👤 FIGHTER-LEVEL WHOLE-CARD TABLE")
print("=" * 115)

print(
    "Fighter-event rows:",
    len(whole_card_fighters)
)

print(
    "Unique fighters:",
    whole_card_fighters[
        "fighter_name"
    ].nunique()
)

print(
    "Events:",
    whole_card_fighters[
        ["date", "event_name"]
    ].drop_duplicates().shape[0]
)

print(
    "Cards whose fighter weights sum to 1:",
    int(
        np.isclose(
            fighter_weight_check.values,
            1.0
        ).sum()
    ),
    "/",
    len(fighter_weight_check)
)


# ======================================================================================
# 16. SHOW UFC 300 / 302 / 329 / 330 / 331
# ======================================================================================

targets = [
    "UFC 300",
    "UFC 302",
    "UFC 329",
    "UFC 330",
    "UFC 331"
]

target_mask = pd.Series(
    False,
    index=all_card_fights.index
)

for event in targets:

    target_mask |= (
        all_card_fights[
            "event_name"
        ]
        .str.contains(
            event,
            case=False,
            regex=False,
            na=False
        )
    )


print("\n" + "=" * 115)
print("🎯 TARGET CARD STRUCTURE")
print("=" * 115)

display(
    all_card_fights.loc[
        target_mask,
        [
            "date",
            "event_name",
            "f1_name",
            "f2_name",
            "bout_from_top",
            "card_position_tier",
            "is_title_fight",
            "is_main_event_final",
            "is_co_main_final",
            "temporary_placement_weight",
            "bout_weight_share"
        ]
    ]
    .sort_values(
        [
            "date",
            "event_name",
            "bout_from_top"
        ]
    )
    .round(4)
)


# ======================================================================================
# 17. NEW 11 EVENT MAIN / CO-MAIN CHECK
# ======================================================================================

new_top = (
    all_card_fights[
        all_card_fights[
            "data_period"
        ] == "new"
    ]
    .query(
        "bout_from_top <= 2"
    )
    [
        [
            "date",
            "event_name",
            "f1_name",
            "f2_name",
            "bout_from_top",
            "card_position_tier"
        ]
    ]
    .sort_values(
        [
            "date",
            "bout_from_top"
        ]
    )
)


print("\n" + "=" * 115)
print("🆕 NEW 11 CARDS — MAIN + CO-MAIN")
print("=" * 115)

display(new_top)


# ======================================================================================
# 18. HARD VALIDATION
# ======================================================================================

total_fights = len(all_card_fights)

total_events = (
    all_card_fights[
        ["date", "event_name"]
    ]
    .drop_duplicates()
    .shape[0]
)

new_events = (
    all_card_fights.loc[
        all_card_fights[
            "data_period"
        ] == "new",
        ["date", "event_name"]
    ]
    .drop_duplicates()
    .shape[0]
)

main_per_event = (
    all_card_fights
    .groupby(
        ["date", "event_name"]
    )[
        "is_main_event_rebuilt"
    ]
    .sum()
)


print("\n" + "=" * 115)
print("✅ HARD VALIDATION")
print("=" * 115)

checks = {
    "Total fights = 8877":
        total_fights == 8877,

    "Total events = 788":
        total_events == 788,

    "New events = 11":
        new_events == 11,

    "Every event has exactly one row-order main":
        (main_per_event == 1).all(),

    "All bout positions positive":
        (
            all_card_fights[
                "bout_from_top"
            ] >= 1
        ).all(),

    "All placement weights present":
        all_card_fights[
            "temporary_placement_weight"
        ].notna().all(),

    "All fighter card weights present":
        whole_card_fighters[
            "fighter_card_weight_share"
        ].notna().all(),

    "Every card fighter-weight sum = 1":
        np.isclose(
            fighter_weight_check.values,
            1.0
        ).all()
}


for name, passed in checks.items():

    print(
        ("✅" if passed else "❌"),
        name
    )


# ======================================================================================
# 19. SAVE SANDBOX OBJECTS
# ======================================================================================

whole_card_fanfare_base = (
    all_card_fights.copy()
)

whole_card_fanfare_fighters = (
    whole_card_fighters.copy()
)


print("\n" + "=" * 115)
print("📦 SANDBOX OBJECTS CREATED")
print("=" * 115)

print(
    "whole_card_fanfare_base:",
    whole_card_fanfare_base.shape
)

print(
    "whole_card_fanfare_fighters:",
    whole_card_fanfare_fighters.shape
)


# ======================================================================================
# 20. SAFETY
# ======================================================================================

print("\n" + "=" * 115)
print("🔒 SAFETY")
print("=" * 115)

print("Published model modified: NO")
print("V2 model modified: NO")
print("Current Fanfare modified: NO")
print("Files overwritten: NO")

print("\n🔥 BIG STEP COMPLETE")
```

```python
# ======================================================================================
# WHOLE-CARD FANFARE V2
# BIG STEP 1B — FIX NEW-CARD ORDER + REBUILD ENTIRE CARD STRUCTURE
#
# Historical 8,736:
#   existing order = bottom -> top  ✅ 100% validated
#
# New 141 UFC.com rows:
#   scrape order = top -> bottom
#
# This cell standardizes BOTH periods so:
#   bout_from_top = 1  -> MAIN EVENT
#   bout_from_top = 2  -> CO-MAIN
#   etc.
#
# NO published/V2/Fanfare changes.
# ======================================================================================

import pandas as pd
import numpy as np

print("=" * 115)
print("🔥 FIXING + FINALIZING WHOLE-CARD STRUCTURE")
print("=" * 115)


# ======================================================================================
# 1. START FRESH FROM ORIGINAL OBJECTS
# ======================================================================================

hist = ufc_fights.copy()
new  = nf_ranked_fixed.copy()

hist["date"] = pd.to_datetime(hist["date"])
new["date"]  = pd.to_datetime(new["date"])

hist["data_period"] = "historical"
new["data_period"]  = "new"


# ======================================================================================
# 2. STANDARDIZE COLUMNS
# ======================================================================================

wanted = [
    "date",
    "event_name",
    "f1_name",
    "f2_name",
    "weight_class",
    "is_title_fight",

    "f1_ufc_fights_before",
    "f2_ufc_fights_before",

    "f1_ufc_win_pct_before",
    "f2_ufc_win_pct_before",

    "f1_win_streak_before",
    "f2_win_streak_before",

    "f1_finish_rate_before",
    "f2_finish_rate_before",

    "f1_rank_before",
    "f2_rank_before",

    "is_main_event",
    "data_period"
]

for col in wanted:

    if col not in hist.columns:
        hist[col] = np.nan

    if col not in new.columns:
        new[col] = np.nan


hist = hist[wanted].copy()
new  = new[wanted].copy()


# ======================================================================================
# 3. HISTORICAL ORDER
#
# Historical rows are bottom -> top.
# Therefore reverse within event to obtain bout_from_top.
# ======================================================================================

hist["_source_row"] = np.arange(len(hist))

hist["source_position"] = (
    hist
    .groupby(
        ["date", "event_name"],
        sort=False
    )
    .cumcount()
    + 1
)

hist["total_fights_on_card"] = (
    hist
    .groupby(
        ["date", "event_name"]
    )["event_name"]
    .transform("size")
)

hist["bout_from_top"] = (
    hist["total_fights_on_card"]
    -
    hist["source_position"]
    + 1
)


# ======================================================================================
# 4. NEW UFC.COM ORDER
#
# UFC.com scrape rows are TOP -> BOTTOM.
# Therefore the first fight is already bout_from_top = 1.
# ======================================================================================

new["_source_row"] = np.arange(len(new))

new["source_position"] = (
    new
    .groupby(
        ["date", "event_name"],
        sort=False
    )
    .cumcount()
    + 1
)

new["total_fights_on_card"] = (
    new
    .groupby(
        ["date", "event_name"]
    )["event_name"]
    .transform("size")
)

new["bout_from_top"] = new["source_position"]


# ======================================================================================
# 5. COMBINE
# ======================================================================================

all_card_fights_fixed = pd.concat(
    [hist, new],
    ignore_index=True
)

all_card_fights_fixed["is_main_event_rebuilt"] = (
    all_card_fights_fixed["bout_from_top"] == 1
)

all_card_fights_fixed["is_co_main_rebuilt"] = (
    all_card_fights_fixed["bout_from_top"] == 2
)


# ======================================================================================
# 6. HISTORICAL VALIDATION AGAIN
# ======================================================================================

hv = all_card_fights_fixed[
    all_card_fights_fixed["data_period"] == "historical"
].copy()

historical_agreement = (
    (hv["is_main_event"] == True)
    ==
    hv["is_main_event_rebuilt"]
)

print("\n🧪 HISTORICAL VALIDATION")
print("-" * 75)

print(
    "Exact agreement:",
    int(historical_agreement.sum()),
    "/",
    len(historical_agreement)
)

print(
    "Agreement %:",
    round(
        historical_agreement.mean() * 100,
        4
    )
)

print(
    "Disagreements:",
    int((~historical_agreement).sum())
)


# ======================================================================================
# 7. FINAL MAIN / CO-MAIN FLAGS
# ======================================================================================

all_card_fights_fixed["is_main_event_final"] = (
    all_card_fights_fixed["bout_from_top"] == 1
)

all_card_fights_fixed["is_co_main_final"] = (
    all_card_fights_fixed["bout_from_top"] == 2
)


# ======================================================================================
# 8. PLACEMENT TIERS
#
# Structural framework only — not final Fanfare weights.
# ======================================================================================

def placement_tier(pos):

    if pos == 1:
        return "Main Event"

    elif pos == 2:
        return "Co-Main Event"

    elif pos <= 5:
        return "Upper Card"

    elif pos <= 8:
        return "Middle Card"

    else:
        return "Lower Card"


all_card_fights_fixed["card_position_tier"] = (
    all_card_fights_fixed[
        "bout_from_top"
    ].apply(
        placement_tier
    )
)


# ======================================================================================
# 9. TEMPORARY WHOLE-CARD PLACEMENT WEIGHTS
#
# Main event matters most individually,
# BUT entire rest of card collectively dominates.
#
# These remain experimental.
# ======================================================================================

TEMP_PLACEMENT_WEIGHTS = {
    "Main Event": 1.00,
    "Co-Main Event": 0.85,
    "Upper Card": 0.70,
    "Middle Card": 0.50,
    "Lower Card": 0.30
}

all_card_fights_fixed[
    "temporary_placement_weight"
] = (
    all_card_fights_fixed[
        "card_position_tier"
    ].map(
        TEMP_PLACEMENT_WEIGHTS
    )
)


# ======================================================================================
# 10. NORMALIZE WEIGHTS WITHIN EACH CARD
# ======================================================================================

all_card_fights_fixed["card_weight_total"] = (
    all_card_fights_fixed
    .groupby(
        ["date", "event_name"]
    )[
        "temporary_placement_weight"
    ]
    .transform("sum")
)

all_card_fights_fixed["bout_weight_share"] = (
    all_card_fights_fixed[
        "temporary_placement_weight"
    ]
    /
    all_card_fights_fixed[
        "card_weight_total"
    ]
)


# ======================================================================================
# 11. BUILD FIGHTER-LEVEL TABLE
# ======================================================================================

f1 = all_card_fights_fixed[
    [
        "date",
        "event_name",
        "f1_name",
        "bout_from_top",
        "card_position_tier",
        "bout_weight_share",
        "is_main_event_final",
        "is_co_main_final",
        "f1_rank_before",
        "f1_ufc_fights_before",
        "f1_ufc_win_pct_before",
        "f1_win_streak_before",
        "f1_finish_rate_before"
    ]
].copy()

f1 = f1.rename(
    columns={
        "f1_name": "fighter_name",
        "f1_rank_before": "rank_before",
        "f1_ufc_fights_before": "ufc_fights_before",
        "f1_ufc_win_pct_before": "ufc_win_pct_before",
        "f1_win_streak_before": "win_streak_before",
        "f1_finish_rate_before": "finish_rate_before"
    }
)

f1["corner"] = "f1"


f2 = all_card_fights_fixed[
    [
        "date",
        "event_name",
        "f2_name",
        "bout_from_top",
        "card_position_tier",
        "bout_weight_share",
        "is_main_event_final",
        "is_co_main_final",
        "f2_rank_before",
        "f2_ufc_fights_before",
        "f2_ufc_win_pct_before",
        "f2_win_streak_before",
        "f2_finish_rate_before"
    ]
].copy()

f2 = f2.rename(
    columns={
        "f2_name": "fighter_name",
        "f2_rank_before": "rank_before",
        "f2_ufc_fights_before": "ufc_fights_before",
        "f2_ufc_win_pct_before": "ufc_win_pct_before",
        "f2_win_streak_before": "win_streak_before",
        "f2_finish_rate_before": "finish_rate_before"
    }
)

f2["corner"] = "f2"


whole_card_fanfare_fighters_fixed = pd.concat(
    [f1, f2],
    ignore_index=True
)

whole_card_fanfare_fighters_fixed[
    "fighter_card_weight_share"
] = (
    whole_card_fanfare_fighters_fixed[
        "bout_weight_share"
    ] / 2
)


# ======================================================================================
# 12. VALIDATE FIGHTER WEIGHT SUMS
# ======================================================================================

weight_check = (
    whole_card_fanfare_fighters_fixed
    .groupby(
        ["date", "event_name"]
    )[
        "fighter_card_weight_share"
    ]
    .sum()
)


# ======================================================================================
# 13. NEW EVENT HEADLINER SANITY CHECK
# ======================================================================================

new_top = (
    all_card_fights_fixed[
        (
            all_card_fights_fixed[
                "data_period"
            ] == "new"
        )
        &
        (
            all_card_fights_fixed[
                "bout_from_top"
            ] <= 2
        )
    ]
    [
        [
            "date",
            "event_name",
            "f1_name",
            "f2_name",
            "bout_from_top",
            "card_position_tier",
            "is_title_fight"
        ]
    ]
    .sort_values(
        [
            "date",
            "bout_from_top"
        ]
    )
)


print("\n" + "=" * 115)
print("🆕 CORRECTED NEW 11 — MAIN + CO-MAIN")
print("=" * 115)

display(new_top)


# ======================================================================================
# 14. CRITICAL EVENT-NAME HEADLINER VALIDATION
#
# Check cards whose event name explicitly tells us headliners.
# ======================================================================================

checks = [
    ("UFC 329", "Conor McGregor", "Max Holloway"),
    ("Du Plessis vs. Usman", "Dricus Du Plessis", "Kamaru Usman"),
    ("Ankalaev vs. Guskov", "Magomed Ankalaev", "Bogdan Guskov"),
    ("Medic vs. Rodriguez", "Uros Medic", "Daniel Rodriguez"),
    ("Gamrot vs. Salkilld", "Mateusz Gamrot", "Quillan Salkilld"),
    ("UFC 330", "Islam Makhachev", "Ian Machado Garry"),
    ("Hernandez vs. Rodrigues", "Anthony Hernandez", "Gregory Rodrigues"),
    ("Nurmagomedov vs. Song", "Umar Nurmagomedov", "Song Yadong"),
    ("Hooker vs. Parnasse", "Dan Hooker", "Salahdine Parnasse"),
    ("Silva vs. Delgado", "Jean Silva", "Jose Miguel Delgado"),
    ("UFC 331", "Joshua Van", "Alexandre Pantoja")
]

print("\n" + "=" * 115)
print("🎯 HEADLINER VALIDATION")
print("=" * 115)

headline_passes = []

for event_piece, fighter_a, fighter_b in checks:

    candidate = all_card_fights_fixed[
        (
            all_card_fights_fixed[
                "data_period"
            ] == "new"
        )
        &
        (
            all_card_fights_fixed[
                "event_name"
            ]
            .str.contains(
                event_piece,
                case=False,
                regex=False,
                na=False
            )
        )
        &
        (
            all_card_fights_fixed[
                "bout_from_top"
            ] == 1
        )
    ]

    if len(candidate) != 1:

        passed = False

        print(
            "❌",
            event_piece,
            "— could not isolate exactly one main event"
        )

    else:

        row = candidate.iloc[0]

        names = {
            str(row["f1_name"]),
            str(row["f2_name"])
        }

        expected = {
            fighter_a,
            fighter_b
        }

        passed = names == expected

        print(
            "✅" if passed else "❌",
            event_piece,
            "→",
            row["f1_name"],
            "vs",
            row["f2_name"]
        )

    headline_passes.append(passed)


# ======================================================================================
# 15. CARD-WEIGHT STRUCTURE
# ======================================================================================

event_weight_structure = (
    all_card_fights_fixed
    .groupby(
        ["date", "event_name"],
        as_index=False
    )
    .agg(
        total_fights=(
            "f1_name",
            "size"
        ),
        main_event_share=(
            "bout_weight_share",
            lambda s:
                s[
                    all_card_fights_fixed.loc[
                        s.index,
                        "bout_from_top"
                    ] == 1
                ].sum()
        ),
        co_main_share=(
            "bout_weight_share",
            lambda s:
                s[
                    all_card_fights_fixed.loc[
                        s.index,
                        "bout_from_top"
                    ] == 2
                ].sum()
        )
    )
)

event_weight_structure[
    "rest_of_card_share"
] = (
    1
    -
    event_weight_structure[
        "main_event_share"
    ]
    -
    event_weight_structure[
        "co_main_share"
    ]
)


print("\n" + "=" * 115)
print("⚖️ WHOLE-CARD WEIGHT DISTRIBUTION")
print("=" * 115)

print(
    "Average Main Event share:",
    round(
        event_weight_structure[
            "main_event_share"
        ].mean() * 100,
        2
    ),
    "%"
)

print(
    "Average Co-Main share:",
    round(
        event_weight_structure[
            "co_main_share"
        ].mean() * 100,
        2
    ),
    "%"
)

print(
    "Average rest-of-card share:",
    round(
        event_weight_structure[
            "rest_of_card_share"
        ].mean() * 100,
        2
    ),
    "%"
)


# ======================================================================================
# 16. HARD VALIDATION
# ======================================================================================

event_count = (
    all_card_fights_fixed[
        ["date", "event_name"]
    ]
    .drop_duplicates()
    .shape[0]
)

new_event_count = (
    all_card_fights_fixed.loc[
        all_card_fights_fixed[
            "data_period"
        ] == "new",
        ["date", "event_name"]
    ]
    .drop_duplicates()
    .shape[0]
)

main_counts = (
    all_card_fights_fixed
    .groupby(
        ["date", "event_name"]
    )[
        "is_main_event_final"
    ]
    .sum()
)


hard_checks = {

    "Total fights = 8877":
        len(all_card_fights_fixed) == 8877,

    "Total events = 788":
        event_count == 788,

    "New events = 11":
        new_event_count == 11,

    "Historical main-event agreement = 100%":
        historical_agreement.all(),

    "All 11 new named headliners correct":
        all(headline_passes),

    "Exactly one main event per card":
        (main_counts == 1).all(),

    "Fighter rows = 17754":
        len(
            whole_card_fanfare_fighters_fixed
        ) == 17754,

    "Every card fighter weights = 1":
        np.isclose(
            weight_check.values,
            1.0
        ).all()
}


print("\n" + "=" * 115)
print("✅ FINAL STRUCTURE VALIDATION")
print("=" * 115)

for label, passed in hard_checks.items():

    print(
        "✅" if passed else "❌",
        label
    )


# ======================================================================================
# 17. SAVE CORRECTED SANDBOX OBJECTS
# ======================================================================================

whole_card_fanfare_base = (
    all_card_fights_fixed.copy()
)

whole_card_fanfare_fighters = (
    whole_card_fanfare_fighters_fixed.copy()
)


print("\n📦 Corrected sandbox objects saved:")
print(
    "whole_card_fanfare_base:",
    whole_card_fanfare_base.shape
)
print(
    "whole_card_fanfare_fighters:",
    whole_card_fanfare_fighters.shape
)


print("\n🔒 Published model modified: NO")
print("🔒 V2 modified: NO")
print("🔒 Existing Fanfare modified: NO")
print("🔒 Files overwritten: NO")

print("\n🔥 CORRECTION COMPLETE")
```

```python
# ======================================================================================
# WHOLE-CARD FANFARE V2
# BIG STEP 2 — BUILD PRE-EVENT WHOLE-CARD DRAWING / POPULARITY HISTORY
#
# PURPOSE:
#   Give EVERY fighter on EVERY card a historical "drawing evidence" profile
#   based ONLY on events that occurred BEFORE the current event.
#
# IMPORTANT:
#   - No future leakage
#   - Missing history stays missing
#   - We do NOT treat UFC skill/ranking as popularity
#   - We do NOT change V2
#   - We do NOT change published Fanfare
#
# OUTPUTS:
#   fighter_fanfare_history
#   whole_card_drawing_v2
# ======================================================================================

import pandas as pd
import numpy as np

print("=" * 115)
print("🔥 WHOLE-CARD FANFARE V2 — HISTORICAL DRAWING BUILD")
print("=" * 115)


# ======================================================================================
# 1. EVENT-LEVEL OBSERVED FANFARE EVIDENCE
# ======================================================================================

event_evidence = final_index[
    [
        "event_date",
        "event_name",
        "fanfare_score",
        "commercial_audience_score",
        "historical_drawing_power_score",
        "search_acceleration_score"
    ]
].copy()

event_evidence["event_date"] = pd.to_datetime(
    event_evidence["event_date"]
)

event_evidence = event_evidence.rename(
    columns={
        "event_date": "date"
    }
)


# ======================================================================================
# 2. MERGE EVENT EVIDENCE ONTO EVERY FIGHTER-EVENT ROW
# ======================================================================================

fighter_hist = whole_card_fanfare_fighters.copy()

fighter_hist["date"] = pd.to_datetime(
    fighter_hist["date"]
)

fighter_hist = fighter_hist.merge(
    event_evidence,
    on=["date", "event_name"],
    how="left",
    validate="many_to_one"
)

fighter_hist["_original_fighter_row"] = np.arange(
    len(fighter_hist)
)


# ======================================================================================
# 3. SORT CHRONOLOGICALLY FOR NO-LEAKAGE HISTORY
# ======================================================================================

fighter_hist = fighter_hist.sort_values(
    [
        "fighter_name",
        "date",
        "_original_fighter_row"
    ],
    kind="stable"
).reset_index(drop=True)


# ======================================================================================
# 4. PREVIOUS APPEARANCES
# ======================================================================================

fighter_hist["previous_ufc_event_appearances"] = (
    fighter_hist
    .groupby("fighter_name")
    .cumcount()
)


# ======================================================================================
# 5. HELPER:
# EXPANDING MEAN USING ONLY PRIOR NON-MISSING VALUES
#
# shift() ensures CURRENT event is never included.
# ======================================================================================

def prior_expanding_mean(series):

    return (
        series
        .shift(1)
        .expanding()
        .mean()
    )


def prior_nonmissing_count(series):

    return (
        series
        .notna()
        .astype(int)
        .shift(1)
        .fillna(0)
        .expanding()
        .sum()
    )


# ======================================================================================
# 6. BUILD FIGHTER HISTORY FOR EACH AVAILABLE FANFARE SIGNAL
# ======================================================================================

history_signals = [
    "fanfare_score",
    "commercial_audience_score",
    "historical_drawing_power_score",
    "search_acceleration_score"
]

for col in history_signals:

    fighter_hist[
        f"prior_avg_{col}"
    ] = (
        fighter_hist
        .groupby(
            "fighter_name",
            group_keys=False
        )[col]
        .apply(prior_expanding_mean)
        .reset_index(
            level=0,
            drop=True
        )
    )

    fighter_hist[
        f"prior_count_{col}"
    ] = (
        fighter_hist
        .groupby(
            "fighter_name",
            group_keys=False
        )[col]
        .apply(prior_nonmissing_count)
        .reset_index(
            level=0,
            drop=True
        )
    )


# ======================================================================================
# 7. IMPORTANT CORRECTION:
# SEARCH ACCELERATION IS NOT ABSOLUTE POPULARITY.
#
# Therefore fighter historical drawing is based primarily on:
#
#   A. prior commercial/audience evidence
#   B. prior historical drawing evidence
#   C. prior event Fanfare evidence
#
# Search acceleration is retained separately for diagnostics.
#
# Dynamic missing:
# only available evidence contributes.
# ======================================================================================

DRAWING_WEIGHTS = {
    "prior_avg_commercial_audience_score": 0.40,
    "prior_avg_historical_drawing_power_score": 0.35,
    "prior_avg_fanfare_score": 0.25
}


def weighted_available(row, weights):

    available = [
        (col, weight)
        for col, weight in weights.items()
        if pd.notna(row[col])
    ]

    if not available:
        return np.nan

    total_weight = sum(
        weight
        for _, weight in available
    )

    return sum(
        row[col] * weight
        for col, weight in available
    ) / total_weight


fighter_hist[
    "fighter_prior_drawing_score_raw"
] = fighter_hist.apply(
    weighted_available,
    axis=1,
    weights=DRAWING_WEIGHTS
)


# ======================================================================================
# 8. EVIDENCE COUNTS
# ======================================================================================

fighter_hist[
    "fighter_prior_drawing_evidence_count"
] = (
    fighter_hist[
        "prior_count_commercial_audience_score"
    ]
    +
    fighter_hist[
        "prior_count_historical_drawing_power_score"
    ]
    +
    fighter_hist[
        "prior_count_fanfare_score"
    ]
)


# ======================================================================================
# 9. CONFIDENCE SHRINKAGE
#
# One prior observation should not instantly make someone a
# proven historical draw.
#
# Reliability:
# 1 obs  = 0.50
# 2 obs  = 0.67
# 3 obs  = 0.75
# 5 obs  = 0.83
# 10 obs = 0.91
#
# We shrink toward neutral 50 ONLY when evidence exists.
#
# Missing remains missing.
# ======================================================================================

n = fighter_hist[
    "fighter_prior_drawing_evidence_count"
]

fighter_hist[
    "drawing_reliability"
] = np.where(
    n > 0,
    n / (n + 1),
    np.nan
)

fighter_hist[
    "fighter_prior_drawing_score"
] = np.where(
    fighter_hist[
        "fighter_prior_drawing_score_raw"
    ].notna(),

    (
        fighter_hist[
            "drawing_reliability"
        ]
        *
        fighter_hist[
            "fighter_prior_drawing_score_raw"
        ]
    )
    +
    (
        (
            1
            -
            fighter_hist[
                "drawing_reliability"
            ]
        )
        * 50
    ),

    np.nan
)


# ======================================================================================
# 10. RETURN TO ORIGINAL EVENT/CARD ORDER
# ======================================================================================

fighter_hist = fighter_hist.sort_values(
    "_original_fighter_row"
).reset_index(drop=True)


# ======================================================================================
# 11. WHOLE-CARD DRAWING SCORE
#
# KEY DESIGN:
# Missing fighter history is NOT zero.
#
# We renormalize only across fighters with available historical
# drawing evidence.
#
# Also record COVERAGE so we know whether the score represents
# 10%, 50%, 90%, etc. of the card.
# ======================================================================================

fighter_hist[
    "available_drawing_weight"
] = np.where(
    fighter_hist[
        "fighter_prior_drawing_score"
    ].notna(),

    fighter_hist[
        "fighter_card_weight_share"
    ],

    0.0
)

fighter_hist[
    "weighted_drawing_contribution"
] = np.where(
    fighter_hist[
        "fighter_prior_drawing_score"
    ].notna(),

    fighter_hist[
        "fighter_prior_drawing_score"
    ]
    *
    fighter_hist[
        "fighter_card_weight_share"
    ],

    0.0
)


whole_card_drawing_v2 = (
    fighter_hist
    .groupby(
        ["date", "event_name"],
        as_index=False
    )
    .agg(

        whole_card_drawing_numerator=(
            "weighted_drawing_contribution",
            "sum"
        ),

        whole_card_drawing_coverage=(
            "available_drawing_weight",
            "sum"
        ),

        fighters_with_drawing_history=(
            "fighter_prior_drawing_score",
            lambda x: x.notna().sum()
        ),

        total_fighters=(
            "fighter_name",
            "size"
        ),

        main_event_fighters_with_history=(
            "fighter_prior_drawing_score",
            lambda x: np.nan
        )
    )
)


whole_card_drawing_v2[
    "whole_card_historical_drawing_score"
] = np.where(

    whole_card_drawing_v2[
        "whole_card_drawing_coverage"
    ] > 0,

    whole_card_drawing_v2[
        "whole_card_drawing_numerator"
    ]
    /
    whole_card_drawing_v2[
        "whole_card_drawing_coverage"
    ],

    np.nan
)


whole_card_drawing_v2[
    "fighter_history_pct"
] = (
    whole_card_drawing_v2[
        "fighters_with_drawing_history"
    ]
    /
    whole_card_drawing_v2[
        "total_fighters"
    ]
    * 100
)


# Remove placeholder aggregation column.
whole_card_drawing_v2 = (
    whole_card_drawing_v2.drop(
        columns=[
            "main_event_fighters_with_history"
        ]
    )
)


# ======================================================================================
# 12. ADD CURRENT FANFARE FOR COMPARISON ONLY
# ======================================================================================

comparison_cols = final_index[
    [
        "event_date",
        "event_name",
        "fanfare_score",
        "commercial_audience_score",
        "historical_drawing_power_score",
        "search_acceleration_score"
    ]
].copy()

comparison_cols[
    "event_date"
] = pd.to_datetime(
    comparison_cols[
        "event_date"
    ]
)

comparison_cols = comparison_cols.rename(
    columns={
        "event_date": "date",
        "fanfare_score": "current_fanfare_score"
    }
)


whole_card_drawing_v2 = (
    whole_card_drawing_v2.merge(
        comparison_cols,
        on=["date", "event_name"],
        how="left",
        validate="one_to_one"
    )
)


# ======================================================================================
# 13. TARGET EVENTS
# ======================================================================================

targets = [
    "UFC 100",
    "UFC 205",
    "UFC 217",
    "UFC 229",
    "UFC 269",
    "UFC 300",
    "UFC 302",
    "UFC 329",
    "UFC 330",
    "UFC 331"
]

target_mask = pd.Series(
    False,
    index=whole_card_drawing_v2.index
)

for event in targets:

    target_mask |= (
        whole_card_drawing_v2[
            "event_name"
        ]
        .str.contains(
            event,
            case=False,
            regex=False,
            na=False
        )
    )


print("\n" + "=" * 115)
print("🎯 WHOLE-CARD HISTORICAL DRAWING — TARGET EVENTS")
print("=" * 115)

display(
    whole_card_drawing_v2.loc[
        target_mask,
        [
            "date",
            "event_name",
            "whole_card_historical_drawing_score",
            "whole_card_drawing_coverage",
            "fighters_with_drawing_history",
            "total_fighters",
            "fighter_history_pct",
            "current_fanfare_score",
            "commercial_audience_score",
            "historical_drawing_power_score",
            "search_acceleration_score"
        ]
    ]
    .sort_values("date")
    .round(2)
)


# ======================================================================================
# 14. TOP 25 WHOLE-CARD HISTORICAL DRAWING SCORES
# ======================================================================================

print("\n" + "=" * 115)
print("⭐ TOP 25 WHOLE-CARD HISTORICAL DRAWING SCORES")
print("=" * 115)

display(
    whole_card_drawing_v2[
        whole_card_drawing_v2[
            "whole_card_historical_drawing_score"
        ].notna()
    ]
    .sort_values(
        "whole_card_historical_drawing_score",
        ascending=False
    )
    [
        [
            "date",
            "event_name",
            "whole_card_historical_drawing_score",
            "whole_card_drawing_coverage",
            "fighters_with_drawing_history",
            "total_fighters",
            "fighter_history_pct",
            "current_fanfare_score"
        ]
    ]
    .head(25)
    .round(2)
)


# ======================================================================================
# 15. COVERAGE DIAGNOSTICS
# ======================================================================================

print("\n" + "=" * 115)
print("📊 COVERAGE")
print("=" * 115)

available = (
    whole_card_drawing_v2[
        "whole_card_historical_drawing_score"
    ].notna()
)

print(
    "Events with whole-card drawing evidence:",
    int(available.sum()),
    "/",
    len(whole_card_drawing_v2)
)

print(
    "Events missing whole-card drawing evidence:",
    int((~available).sum())
)

print(
    "Median weighted card coverage:",
    round(
        whole_card_drawing_v2.loc[
            available,
            "whole_card_drawing_coverage"
        ].median() * 100,
        2
    ),
    "%"
)

print(
    "Median fighter coverage:",
    round(
        whole_card_drawing_v2.loc[
            available,
            "fighter_history_pct"
        ].median(),
        2
    ),
    "%"
)

print(
    "Events with >= 50% weighted-card coverage:",
    int(
        (
            whole_card_drawing_v2[
                "whole_card_drawing_coverage"
            ] >= 0.50
        ).sum()
    )
)

print(
    "Events with >= 75% weighted-card coverage:",
    int(
        (
            whole_card_drawing_v2[
                "whole_card_drawing_coverage"
            ] >= 0.75
        ).sum()
    )
)


# ======================================================================================
# 16. SANITY — CURRENT EVENT MUST NOT CONTRIBUTE TO ITS OWN HISTORY
#
# First UFC appearance should have zero previous appearances.
# A fighter can only have drawing history if PRIOR events supplied
# evidence.
# ======================================================================================

first_appearances = fighter_hist[
    fighter_hist[
        "previous_ufc_event_appearances"
    ] == 0
]

bad_first_history = first_appearances[
    fighter_hist.loc[
        first_appearances.index,
        "fighter_prior_drawing_evidence_count"
    ] > 0
]

print("\n" + "=" * 115)
print("🧪 NO-LEAKAGE CHECK")
print("=" * 115)

print(
    "First UFC appearances:",
    len(first_appearances)
)

print(
    "First appearances incorrectly carrying prior evidence:",
    len(bad_first_history)
)


# ======================================================================================
# 17. SAVE
# ======================================================================================

fighter_fanfare_history = fighter_hist.copy()

print("\n📦 SANDBOX OBJECTS")
print(
    "fighter_fanfare_history:",
    fighter_fanfare_history.shape
)

print(
    "whole_card_drawing_v2:",
    whole_card_drawing_v2.shape
)


# ======================================================================================
# 18. SAFETY
# ======================================================================================

print("\n" + "=" * 115)
print("🔒 SAFETY")
print("=" * 115)

print("Published model modified: NO")
print("V2 Event Strength modified: NO")
print("Existing Fanfare modified: NO")
print("Files overwritten: NO")

print("\n🔥 WHOLE-CARD DRAWING BUILD COMPLETE")
```

```python
# ======================================================================================
# WHOLE-CARD FANFARE V2
# BIG STEP 3 — COVERAGE-CORRECT DRAWING + BUILD FANFARE V2 PROTOTYPE
#
# This does NOT pretend sparse fighter history represents the whole card.
#
# COMPONENTS:
#   1. Whole-card historical drawing evidence
#   2. Commercial / audience evidence
#   3. Search acceleration
#
# ABSOLUTE WHOLE-CARD POPULARITY remains intentionally separate / pending.
#
# NO published model changes.
# NO V2 Event Strength changes.
# ======================================================================================

import pandas as pd
import numpy as np

print("=" * 115)
print("🔥 WHOLE-CARD FANFARE V2 — PROTOTYPE BUILD")
print("=" * 115)


# ======================================================================================
# 1. START FROM WHOLE-CARD DRAWING TABLE
# ======================================================================================

fan2 = whole_card_drawing_v2.copy()

fan2["date"] = pd.to_datetime(fan2["date"])


# ======================================================================================
# 2. COVERAGE-ADJUST HISTORICAL DRAWING
#
# Important:
# We do NOT multiply score by coverage and call missing fighters zero.
#
# Instead:
# - raw drawing score remains the estimate among fighters with evidence
# - confidence is tracked separately
# - low-coverage observations shrink toward neutral 50
#
# Examples:
# 100% coverage -> keep score
# 75% coverage  -> mostly keep score
# 25% coverage  -> strongly shrink toward 50
# ======================================================================================

fan2["drawing_coverage_reliability"] = (
    fan2["whole_card_drawing_coverage"]
    .clip(lower=0, upper=1)
)

fan2["whole_card_drawing_adjusted"] = np.where(
    fan2["whole_card_historical_drawing_score"].notna(),

    (
        fan2["drawing_coverage_reliability"]
        *
        fan2["whole_card_historical_drawing_score"]
    )
    +
    (
        (1 - fan2["drawing_coverage_reliability"])
        * 50
    ),

    np.nan
)


# ======================================================================================
# 3. CREATE COMMERCIAL DEMAND COMPONENT
#
# Use existing commercial/audience score directly.
# Missing stays NaN.
# ======================================================================================

fan2["commercial_demand_v2"] = (
    fan2["commercial_audience_score"]
)


# ======================================================================================
# 4. SEARCH ACCELERATION
#
# Keep it separate.
# It measures EVENT-WEEK ACCELERATION, not absolute fame.
# ======================================================================================

fan2["search_acceleration_v2"] = (
    fan2["search_acceleration_score"]
)


# ======================================================================================
# 5. FANFARE V2 PROTOTYPE WEIGHTS
#
# IMPORTANT:
# Absolute whole-card popularity is still missing.
#
# Intended eventual architecture:
#
#   Absolute whole-card popularity   40%
#   Whole-card historical drawing    30%
#   Commercial demand                20%
#   Search acceleration              10%
#
# FOR THIS PROTOTYPE:
#
# We DO NOT redistribute the missing 40% absolute-popularity block.
#
# That prevents the remaining evidence from pretending to be a
# complete fanfare score.
#
# Each available component contributes only its intended weight.
# ======================================================================================

FANFARE_V2_WEIGHTS = {
    "absolute_whole_card_popularity": 0.40,   # intentionally unavailable
    "whole_card_drawing_adjusted":    0.30,
    "commercial_demand_v2":           0.20,
    "search_acceleration_v2":         0.10
}


# ======================================================================================
# 6. EVIDENCE-WEIGHT SUM
#
# Absolute popularity currently contributes NOTHING because
# we do not have that dataset yet.
#
# Missing components are not zero scores.
# ======================================================================================

fan2["fanfare_v2_available_weight"] = 0.0

for col, weight in [
    ("whole_card_drawing_adjusted", 0.30),
    ("commercial_demand_v2", 0.20),
    ("search_acceleration_v2", 0.10)
]:

    fan2["fanfare_v2_available_weight"] += np.where(
        fan2[col].notna(),
        weight,
        0
    )


# ======================================================================================
# 7. PARTIAL FANFARE CONTRIBUTION
#
# This is deliberately on a 0–60 scale right now because the
# missing absolute-popularity block is worth 40 points.
#
# THIS IS NOT YET THE FINAL 0–100 FANFARE SCORE.
# ======================================================================================

fan2["fanfare_v2_partial"] = 0.0

for col, weight in [
    ("whole_card_drawing_adjusted", 0.30),
    ("commercial_demand_v2", 0.20),
    ("search_acceleration_v2", 0.10)
]:

    fan2["fanfare_v2_partial"] += np.where(
        fan2[col].notna(),
        fan2[col] * weight,
        0
    )


fan2.loc[
    fan2["fanfare_v2_available_weight"] == 0,
    "fanfare_v2_partial"
] = np.nan


# ======================================================================================
# 8. EVIDENCE-NORMALIZED SCORE — DIAGNOSTIC ONLY
#
# Useful for comparing available evidence apples-to-apples.
#
# NOT our final score because this redistributes missing weight.
# ======================================================================================

fan2["fanfare_v2_evidence_normalized"] = np.where(
    fan2["fanfare_v2_available_weight"] > 0,

    fan2["fanfare_v2_partial"]
    /
    fan2["fanfare_v2_available_weight"],

    np.nan
)


# ======================================================================================
# 9. DATA QUALITY LABEL
# ======================================================================================

def fanfare_quality(row):

    w = row["fanfare_v2_available_weight"]

    coverage = row["whole_card_drawing_coverage"]

    if w == 0:
        return "No evidence"

    if (
        w >= 0.60
        and pd.notna(coverage)
        and coverage >= 0.75
    ):
        return "High"

    if (
        w >= 0.40
        and pd.notna(coverage)
        and coverage >= 0.50
    ):
        return "Medium"

    return "Low"


fan2["fanfare_v2_quality"] = fan2.apply(
    fanfare_quality,
    axis=1
)


# ======================================================================================
# 10. COMPARE UFC 300 VS UFC 302 AND OTHER KEY EVENTS
# ======================================================================================

targets = [
    "UFC 100",
    "UFC 205",
    "UFC 217",
    "UFC 229",
    "UFC 269",
    "UFC 300",
    "UFC 302",
    "UFC 329",
    "UFC 330",
    "UFC 331"
]

target_mask = pd.Series(False, index=fan2.index)

for event in targets:

    target_mask |= (
        fan2["event_name"]
        .str.contains(
            event,
            case=False,
            regex=False,
            na=False
        )
    )


print("\n" + "=" * 115)
print("🎯 KEY EVENTS — FANFARE V2 PROTOTYPE")
print("=" * 115)

display(
    fan2.loc[
        target_mask,
        [
            "date",
            "event_name",

            "whole_card_historical_drawing_score",
            "whole_card_drawing_coverage",
            "whole_card_drawing_adjusted",

            "commercial_demand_v2",
            "search_acceleration_v2",

            "fanfare_v2_available_weight",
            "fanfare_v2_partial",
            "fanfare_v2_evidence_normalized",
            "fanfare_v2_quality",

            "current_fanfare_score"
        ]
    ]
    .sort_values("date")
    .round(2)
)


# ======================================================================================
# 11. UFC 300 VS UFC 302 DIRECT COMPARISON
# ======================================================================================

compare_mask = (
    fan2["event_name"]
    .str.contains(
        "UFC 300",
        case=False,
        regex=False,
        na=False
    )
    |
    fan2["event_name"]
    .str.contains(
        "UFC 302",
        case=False,
        regex=False,
        na=False
    )
)

print("\n" + "=" * 115)
print("🥊 UFC 300 VS UFC 302")
print("=" * 115)

display(
    fan2.loc[
        compare_mask,
        [
            "event_name",
            "whole_card_drawing_adjusted",
            "commercial_demand_v2",
            "search_acceleration_v2",
            "fanfare_v2_available_weight",
            "fanfare_v2_partial",
            "fanfare_v2_evidence_normalized",
            "current_fanfare_score"
        ]
    ].round(3)
)


# ======================================================================================
# 12. TOP 30 BY EVIDENCE-NORMALIZED FANFARE
# ======================================================================================

print("\n" + "=" * 115)
print("⭐ TOP 30 — AVAILABLE FANFARE EVIDENCE")
print("⚠️ Diagnostic only — absolute popularity not added yet")
print("=" * 115)

display(
    fan2[
        fan2["fanfare_v2_evidence_normalized"].notna()
    ]
    .sort_values(
        "fanfare_v2_evidence_normalized",
        ascending=False
    )
    [
        [
            "date",
            "event_name",
            "whole_card_drawing_adjusted",
            "commercial_demand_v2",
            "search_acceleration_v2",
            "fanfare_v2_available_weight",
            "fanfare_v2_evidence_normalized",
            "fanfare_v2_quality"
        ]
    ]
    .head(30)
    .round(2)
)


# ======================================================================================
# 13. QUALITY DISTRIBUTION
# ======================================================================================

print("\n" + "=" * 115)
print("📊 FANFARE V2 EVIDENCE QUALITY")
print("=" * 115)

print(
    fan2["fanfare_v2_quality"]
    .value_counts(dropna=False)
)

print("\nAvailable-weight distribution:")

print(
    fan2["fanfare_v2_available_weight"]
    .value_counts()
    .sort_index()
)


# ======================================================================================
# 14. CRITICAL SANITY:
# EARLY SPARSE CARDS SHOULD NO LONGER GET FULL CREDIT
# ======================================================================================

sparse = fan2[
    (
        fan2["whole_card_drawing_coverage"] > 0
    )
    &
    (
        fan2["whole_card_drawing_coverage"] < 0.25
    )
].copy()

print("\n" + "=" * 115)
print("🧪 SPARSE-COVERAGE SANITY CHECK")
print("=" * 115)

display(
    sparse[
        [
            "date",
            "event_name",
            "whole_card_historical_drawing_score",
            "whole_card_drawing_coverage",
            "whole_card_drawing_adjusted",
            "fanfare_v2_evidence_normalized"
        ]
    ]
    .sort_values(
        "whole_card_historical_drawing_score",
        ascending=False
    )
    .head(20)
    .round(2)
)


# ======================================================================================
# 15. SAVE SANDBOX OBJECT
# ======================================================================================

fanfare_v2_prototype = fan2.copy()


print("\n📦 fanfare_v2_prototype:", fanfare_v2_prototype.shape)


# ======================================================================================
# 16. SAFETY
# ======================================================================================

print("\n" + "=" * 115)
print("🔒 SAFETY")
print("=" * 115)

print("Published model modified: NO")
print("V2 Event Strength modified: NO")
print("Published Fanfare modified: NO")
print("Files overwritten: NO")

print("\n🔥 FANFARE V2 PROTOTYPE COMPLETE")
```

```python
# ======================================================================================
# WHOLE-CARD FANFARE V2
# BIG STEP 4 — FIND ANY EXISTING ABSOLUTE POPULARITY / GOOGLE TRENDS DATA
#
# PURPOSE:
# Search EVERY live DataFrame / dict / Series for potentially useful
# popularity data BEFORE we waste time rebuilding anything externally.
#
# We especially want:
#   - fighter Google topic IDs
#   - absolute search interest
#   - baseline search interest
#   - fight-week search interest
#   - fighter popularity
#   - Google Trends mappings
#
# READ-ONLY. NOTHING MODIFIED.
# ======================================================================================

import pandas as pd
import numpy as np

print("=" * 120)
print("🔥 ABSOLUTE POPULARITY — FULL RUNTIME SEARCH")
print("=" * 120)


# ======================================================================================
# 1. SEARCH TERMS
# ======================================================================================

keywords = [
    "trend",
    "google",
    "topic",
    "mid",
    "search",
    "interest",
    "popular",
    "popularity",
    "baseline",
    "fight_week",
    "fightweek",
    "hype",
    "fanfare",
    "draw",
    "audience",
    "ppv",
    "gate",
    "commercial",
    "social",
    "instagram",
    "twitter",
    "youtube"
]


# ======================================================================================
# 2. SEARCH ALL DATAFRAMES
# ======================================================================================

dataframe_hits = []

print("\n" + "=" * 120)
print("📊 DATAFRAME SEARCH")
print("=" * 120)

for name, obj in list(globals().items()):

    if not isinstance(obj, pd.DataFrame):
        continue

    matching_cols = [
        col
        for col in obj.columns
        if any(
            word in str(col).lower()
            for word in keywords
        )
    ]

    if matching_cols:

        dataframe_hits.append(name)

        print(
            f"\n✅ {name}: "
            f"{obj.shape[0]} rows × {obj.shape[1]} cols"
        )

        for col in matching_cols:

            try:
                nonmissing = obj[col].notna().sum()
                unique = obj[col].nunique(dropna=True)

                print(
                    f"   {str(col):<50}"
                    f"nonmissing={nonmissing:<7} "
                    f"unique={unique}"
                )

            except Exception:

                print(
                    f"   {str(col):<50}"
                    f"[unable to summarize]"
                )


# ======================================================================================
# 3. SEARCH SERIES
# ======================================================================================

series_hits = []

print("\n" + "=" * 120)
print("📈 SERIES SEARCH")
print("=" * 120)

for name, obj in list(globals().items()):

    if not isinstance(obj, pd.Series):
        continue

    lname = name.lower()

    if any(
        word in lname
        for word in keywords
    ):

        series_hits.append(name)

        print(
            f"✅ {name:<45} "
            f"length={len(obj):<7} "
            f"nonmissing={obj.notna().sum()}"
        )


# ======================================================================================
# 4. SEARCH DICTIONARIES
#
# Topic mappings are especially likely to live here.
# ======================================================================================

dict_hits = []

print("\n" + "=" * 120)
print("📚 DICTIONARY SEARCH")
print("=" * 120)

for name, obj in list(globals().items()):

    if not isinstance(obj, dict):
        continue

    lname = name.lower()

    key_sample = list(obj.keys())[:20]

    key_text = " ".join(
        str(x).lower()
        for x in key_sample
    )

    if (
        any(word in lname for word in keywords)
        or
        any(word in key_text for word in keywords)
    ):

        dict_hits.append(name)

        print(
            f"\n✅ {name}: "
            f"{len(obj)} entries"
        )

        print(
            "   sample:",
            key_sample[:10]
        )


# ======================================================================================
# 5. SEARCH OTHER VARIABLE NAMES
#
# Sometimes an important object has useful data but isn't a
# DataFrame / Series / dict.
# ======================================================================================

other_hits = []

print("\n" + "=" * 120)
print("🔎 OTHER VARIABLE-NAME HITS")
print("=" * 120)

for name, obj in list(globals().items()):

    if (
        isinstance(obj, pd.DataFrame)
        or isinstance(obj, pd.Series)
        or isinstance(obj, dict)
    ):
        continue

    lname = name.lower()

    if any(
        word in lname
        for word in keywords
    ):

        other_hits.append(name)

        try:
            print(
                f"✅ {name:<50} "
                f"type={type(obj).__name__}"
            )

        except Exception:
            pass


# ======================================================================================
# 6. SPECIAL SEARCH:
# FIND COLUMNS THAT LOOK LIKE FIGHTER IDENTIFIERS / TOPIC IDS
# ======================================================================================

identifier_words = [
    "fighter",
    "name",
    "slug",
    "topic",
    "mid",
    "google"
]

print("\n" + "=" * 120)
print("👤 POSSIBLE FIGHTER / GOOGLE IDENTIFIER TABLES")
print("=" * 120)

for name, obj in list(globals().items()):

    if not isinstance(obj, pd.DataFrame):
        continue

    cols = [
        col
        for col in obj.columns
        if any(
            word in str(col).lower()
            for word in identifier_words
        )
    ]

    # Only print potentially interesting compact tables.
    if (
        len(cols) >= 2
        and len(obj) <= 10000
    ):

        googleish = any(
            any(
                word in str(col).lower()
                for word in [
                    "topic",
                    "mid",
                    "google",
                    "trend",
                    "search",
                    "popular"
                ]
            )
            for col in obj.columns
        )

        if googleish:

            print(
                f"\n⭐ {name}: "
                f"{obj.shape}"
            )

            print(
                "   columns:",
                cols
            )


# ======================================================================================
# 7. CHECK KNOWN HISTORICAL NAMES FROM THE ORIGINAL PROJECT
# ======================================================================================

known_possible_objects = [
    "event_trends_results",
    "fighter_topics",
    "fighter_topic_map",
    "topic_map",
    "topics",
    "trends",
    "trends_results",
    "google_trends",
    "google_topics",
    "fighter_trends",
    "trend_results",
    "popularity",
    "fighter_popularity",
    "absolute_popularity",
    "search_data"
]

print("\n" + "=" * 120)
print("🎯 KNOWN / LIKELY OBJECT NAMES")
print("=" * 120)

for name in known_possible_objects:

    if name in globals():

        obj = globals()[name]

        if isinstance(obj, pd.DataFrame):

            print(
                f"✅ {name}: "
                f"DataFrame {obj.shape}"
            )

            print(
                "   cols:",
                list(obj.columns)
            )

        elif isinstance(obj, dict):

            print(
                f"✅ {name}: "
                f"dict with {len(obj)} entries"
            )

        else:

            print(
                f"✅ {name}: "
                f"{type(obj).__name__}"
            )

    else:

        print(
            f"❌ {name}: unavailable"
        )


# ======================================================================================
# 8. SUMMARY
# ======================================================================================

print("\n" + "=" * 120)
print("📋 SEARCH SUMMARY")
print("=" * 120)

print(
    "DataFrames with popularity-related columns:",
    len(dataframe_hits)
)

print(
    "Series with popularity-related names:",
    len(series_hits)
)

print(
    "Dictionaries with possible popularity data:",
    len(dict_hits)
)

print(
    "Other matching objects:",
    len(other_hits)
)


# ======================================================================================
# 9. SAFETY
# ======================================================================================

print("\n" + "=" * 120)
print("🔒 SAFETY")
print("=" * 120)

print("Published model modified: NO")
print("V2 Event Strength modified: NO")
print("Published Fanfare modified: NO")
print("Fanfare V2 prototype modified: NO")
print("Files overwritten: NO")

print("\n🔥 RUNTIME SEARCH COMPLETE")
```

```python
# ======================================================================================
# WHOLE-CARD FANFARE V2
# BIG STEP 5 — WHOLE-CARD STAR POWER PROXY + FANFARE TEST
#
# PURPOSE:
# Build a PRE-EVENT whole-card "established star presence" proxy.
#
# IMPORTANT:
# This is NOT claimed to be true absolute popularity.
# It is a testable proxy until a historical absolute-search dataset exists.
#
# NO FUTURE LEAKAGE
# MISSING != ZERO
# NO PUBLISHED MODEL CHANGES
# ======================================================================================

import pandas as pd
import numpy as np

print("=" * 120)
print("🔥 WHOLE-CARD STAR POWER + FANFARE V2 TEST")
print("=" * 120)


# ======================================================================================
# 1. START FROM CORRECTED FIGHTER CARD TABLE
# ======================================================================================

stars = whole_card_fanfare_fighters.copy()

stars["date"] = pd.to_datetime(stars["date"])


# ======================================================================================
# 2. ADD PRIOR TITLE-FIGHT HISTORY
#
# Historical table has this directly.
# New cards require reconstruction from all previous fights.
# We rebuild universally to guarantee consistent no-leakage treatment.
# ======================================================================================

base = whole_card_fanfare_base.copy()

base["date"] = pd.to_datetime(base["date"])

base = base.sort_values(
    ["date", "event_name", "bout_from_top"],
    kind="stable"
).reset_index(drop=True)


# Fighter appearance rows carrying title-fight participation.
a = base[
    [
        "date",
        "event_name",
        "f1_name",
        "is_title_fight"
    ]
].rename(
    columns={
        "f1_name": "fighter_name"
    }
)

b = base[
    [
        "date",
        "event_name",
        "f2_name",
        "is_title_fight"
    ]
].rename(
    columns={
        "f2_name": "fighter_name"
    }
)

career = pd.concat(
    [a, b],
    ignore_index=True
)

career["is_title_fight"] = (
    career["is_title_fight"]
    .fillna(False)
    .astype(bool)
)

career = career.sort_values(
    ["fighter_name", "date"],
    kind="stable"
).reset_index(drop=True)

career["prior_title_fights_rebuilt"] = (
    career
    .groupby("fighter_name")["is_title_fight"]
    .transform(
        lambda s:
            s.astype(int)
             .cumsum()
             .shift(1)
             .fillna(0)
    )
)


# ======================================================================================
# 3. BUILD PRIOR MAIN-EVENT HISTORY
#
# Use our now-validated main-event structure.
# ======================================================================================

m1 = base[
    [
        "date",
        "event_name",
        "f1_name",
        "is_main_event_final"
    ]
].rename(
    columns={
        "f1_name": "fighter_name"
    }
)

m2 = base[
    [
        "date",
        "event_name",
        "f2_name",
        "is_main_event_final"
    ]
].rename(
    columns={
        "f2_name": "fighter_name"
    }
)

mainhist = pd.concat(
    [m1, m2],
    ignore_index=True
)

mainhist["is_main_event_final"] = (
    mainhist["is_main_event_final"]
    .fillna(False)
    .astype(bool)
)

mainhist = mainhist.sort_values(
    ["fighter_name", "date"],
    kind="stable"
).reset_index(drop=True)

mainhist["prior_main_events_rebuilt"] = (
    mainhist
    .groupby("fighter_name")["is_main_event_final"]
    .transform(
        lambda s:
            s.astype(int)
             .cumsum()
             .shift(1)
             .fillna(0)
    )
)


# ======================================================================================
# 4. MERGE CAREER HISTORY BACK TO FIGHTER CARD TABLE
# ======================================================================================

career_small = career[
    [
        "date",
        "event_name",
        "fighter_name",
        "prior_title_fights_rebuilt"
    ]
].drop_duplicates(
    ["date", "event_name", "fighter_name"]
)

main_small = mainhist[
    [
        "date",
        "event_name",
        "fighter_name",
        "prior_main_events_rebuilt"
    ]
].drop_duplicates(
    ["date", "event_name", "fighter_name"]
)

stars = stars.merge(
    career_small,
    on=["date", "event_name", "fighter_name"],
    how="left",
    validate="one_to_one"
)

stars = stars.merge(
    main_small,
    on=["date", "event_name", "fighter_name"],
    how="left",
    validate="one_to_one"
)


# ======================================================================================
# 5. RAW STAR-POWER VARIABLES
#
# These are all known entering the event.
#
# UFC experience:
# already present in fighter table.
#
# Ranking:
# Champion (0) gets 16 points.
# #1 gets 15 ... #15 gets 1.
# Unranked = 0 ONLY when rankings are available for the era.
# Before rankings existed, ranking signal = missing.
# ======================================================================================

RANKINGS_START = pd.Timestamp("2013-02-04")

stars["ranking_points"] = np.nan

rank_era_mask = stars["date"] >= RANKINGS_START

stars.loc[
    rank_era_mask,
    "ranking_points"
] = 0.0

ranked = (
    rank_era_mask
    &
    stars["rank_before"].notna()
)

stars.loc[
    ranked,
    "ranking_points"
] = stars.loc[
    ranked,
    "rank_before"
].apply(
    lambda r:
        16 if r == 0
        else max(16 - r, 0)
)


# ======================================================================================
# 6. NORMALIZE FIGHTER-LEVEL STAR VARIABLES
#
# Zero-preserving percentile.
# Zero genuinely means no prior résumé in that category.
# ======================================================================================

def zp_percentile(series):

    result = pd.Series(
        np.nan,
        index=series.index,
        dtype=float
    )

    valid = series.notna()

    result.loc[
        valid & (series == 0)
    ] = 0.0

    positive = valid & (series > 0)

    if positive.sum() > 0:

        result.loc[positive] = (
            series.loc[positive]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )

    return result


star_inputs = {
    "ufc_experience_star":
        "ufc_fights_before",

    "title_history_star":
        "prior_title_fights_rebuilt",

    "main_event_history_star":
        "prior_main_events_rebuilt",

    "ranking_star":
        "ranking_points"
}


for score_col, raw_col in star_inputs.items():

    stars[score_col] = (
        zp_percentile(
            stars[raw_col]
        )
    )


# ======================================================================================
# 7. FIGHTER STAR POWER
#
# We intentionally weight public-facing résumé indicators most:
#
# Prior Main Events   30%
# Prior Title Fights  30%
# Ranking Status      25%
# UFC Experience      15%
#
# Dynamic missing applies for pre-ranking era.
# ======================================================================================

STAR_WEIGHTS = {
    "main_event_history_star": 0.30,
    "title_history_star":      0.30,
    "ranking_star":            0.25,
    "ufc_experience_star":     0.15
}


def weighted_missing(row, weights):

    available = {
        col: weight
        for col, weight in weights.items()
        if pd.notna(row[col])
    }

    if not available:
        return np.nan

    denom = sum(
        available.values()
    )

    return sum(
        row[col] * weight
        for col, weight in available.items()
    ) / denom


stars["fighter_star_power"] = stars.apply(
    weighted_missing,
    axis=1,
    weights=STAR_WEIGHTS
)


# ======================================================================================
# 8. WHOLE-CARD STAR POWER
#
# Uses the validated placement weights:
#
# Main event matters most individually,
# but the entire rest of the card collectively matters much more.
# ======================================================================================

stars["weighted_star_power"] = (
    stars["fighter_star_power"]
    *
    stars["fighter_card_weight_share"]
)

stars["available_star_weight"] = np.where(
    stars["fighter_star_power"].notna(),
    stars["fighter_card_weight_share"],
    0.0
)


whole_card_star = (
    stars
    .groupby(
        ["date", "event_name"],
        as_index=False
    )
    .agg(
        star_numerator=(
            "weighted_star_power",
            "sum"
        ),

        star_coverage=(
            "available_star_weight",
            "sum"
        ),

        fighters_with_star_score=(
            "fighter_star_power",
            lambda s: s.notna().sum()
        ),

        total_fighters=(
            "fighter_name",
            "size"
        )
    )
)

whole_card_star[
    "whole_card_star_power"
] = np.where(
    whole_card_star[
        "star_coverage"
    ] > 0,

    whole_card_star[
        "star_numerator"
    ]
    /
    whole_card_star[
        "star_coverage"
    ],

    np.nan
)


# ======================================================================================
# 9. MERGE WITH FANFARE PROTOTYPE
# ======================================================================================

fan_test = fanfare_v2_prototype.merge(
    whole_card_star[
        [
            "date",
            "event_name",
            "whole_card_star_power",
            "star_coverage",
            "fighters_with_star_score"
        ]
    ],
    on=["date", "event_name"],
    how="left",
    validate="one_to_one"
)


# ======================================================================================
# 10. EXPERIMENTAL FANFARE V2
#
# This uses star-power proxy in the slot we eventually want
# true absolute popularity to occupy.
#
# 40% whole-card star/popularity proxy
# 30% whole-card historical drawing
# 20% commercial demand
# 10% search acceleration
#
# IMPORTANT:
# Search acceleration remains capped at its intended 10%.
# ======================================================================================

TEST_WEIGHTS = {
    "whole_card_star_power":       0.40,
    "whole_card_drawing_adjusted": 0.30,
    "commercial_demand_v2":        0.20,
    "search_acceleration_v2":      0.10
}


def final_fanfare_test(row):

    # Search acceleration NEVER gets redistributed upward.
    fixed_search = 0.0

    fixed_search_weight = 0.0

    if pd.notna(
        row["search_acceleration_v2"]
    ):

        fixed_search = (
            row["search_acceleration_v2"]
            * 0.10
        )

        fixed_search_weight = 0.10


    # Substantive 90% block can redistribute only
    # among available substantive evidence.
    substantive = {
        "whole_card_star_power": 0.40,
        "whole_card_drawing_adjusted": 0.30,
        "commercial_demand_v2": 0.20
    }

    available = {
        col: weight
        for col, weight in substantive.items()
        if pd.notna(row[col])
    }

    if not available:

        # Acceleration alone can contribute max 10 points.
        if fixed_search_weight > 0:
            return fixed_search

        return np.nan


    substantive_total_weight = sum(
        available.values()
    )

    substantive_score = sum(
        row[col] * weight
        for col, weight in available.items()
    ) / substantive_total_weight


    # Substantive evidence occupies 90 points.
    # Search occupies its fixed 10 points.
    return (
        substantive_score * 0.90
        +
        fixed_search
    )


fan_test[
    "fanfare_v2_star_proxy"
] = fan_test.apply(
    final_fanfare_test,
    axis=1
)


# ======================================================================================
# 11. TARGET EVENTS
# ======================================================================================

targets = [
    "UFC 100",
    "UFC 168",
    "UFC 205",
    "UFC 214",
    "UFC 217",
    "UFC 229",
    "UFC 259",
    "UFC 269",
    "UFC 300",
    "UFC 302",
    "UFC 329",
    "UFC 330",
    "UFC 331"
]

mask = pd.Series(
    False,
    index=fan_test.index
)

for target in targets:

    mask |= (
        fan_test["event_name"]
        .str.contains(
            target,
            case=False,
            regex=False,
            na=False
        )
    )


print("\n" + "=" * 120)
print("🎯 KEY EVENTS — WHOLE-CARD STAR POWER + FANFARE")
print("=" * 120)

display(
    fan_test.loc[
        mask,
        [
            "date",
            "event_name",
            "whole_card_star_power",
            "star_coverage",
            "whole_card_drawing_adjusted",
            "commercial_demand_v2",
            "search_acceleration_v2",
            "fanfare_v2_star_proxy",
            "current_fanfare_score"
        ]
    ]
    .sort_values("date")
    .round(2)
)


# ======================================================================================
# 12. TOP 30 WHOLE-CARD STAR POWER
# ======================================================================================

print("\n" + "=" * 120)
print("⭐ TOP 30 — WHOLE-CARD STAR POWER")
print("=" * 120)

display(
    whole_card_star
    .sort_values(
        "whole_card_star_power",
        ascending=False
    )
    .head(30)
    .round(2)
)


# ======================================================================================
# 13. TOP 30 EXPERIMENTAL FANFARE
# ======================================================================================

print("\n" + "=" * 120)
print("🔥 TOP 30 — EXPERIMENTAL WHOLE-CARD FANFARE")
print("=" * 120)

display(
    fan_test[
        [
            "date",
            "event_name",
            "whole_card_star_power",
            "whole_card_drawing_adjusted",
            "commercial_demand_v2",
            "search_acceleration_v2",
            "fanfare_v2_star_proxy"
        ]
    ]
    .sort_values(
        "fanfare_v2_star_proxy",
        ascending=False
    )
    .head(30)
    .round(2)
)


# ======================================================================================
# 14. COMPARE OLD VS TEST FANFARE
# ======================================================================================

comparison = fan_test[
    [
        "date",
        "event_name",
        "current_fanfare_score",
        "fanfare_v2_star_proxy"
    ]
].dropna(
    subset=[
        "current_fanfare_score",
        "fanfare_v2_star_proxy"
    ]
)

print("\n" + "=" * 120)
print("📊 OLD VS EXPERIMENTAL FANFARE")
print("=" * 120)

print(
    "Events compared:",
    len(comparison)
)

print(
    "Correlation:",
    round(
        comparison[
            [
                "current_fanfare_score",
                "fanfare_v2_star_proxy"
            ]
        ].corr().iloc[0,1],
        4
    )
)

comparison["change"] = (
    comparison["fanfare_v2_star_proxy"]
    -
    comparison["current_fanfare_score"]
)

print(
    "Median absolute score change:",
    round(
        comparison["change"]
        .abs()
        .median(),
        2
    )
)


# ======================================================================================
# 15. BIGGEST CHANGES
# ======================================================================================

print("\n" + "=" * 120)
print("⬆️ BIGGEST FANFARE INCREASES")
print("=" * 120)

display(
    comparison
    .sort_values(
        "change",
        ascending=False
    )
    .head(20)
    .round(2)
)


print("\n" + "=" * 120)
print("⬇️ BIGGEST FANFARE DECREASES")
print("=" * 120)

display(
    comparison
    .sort_values(
        "change"
    )
    .head(20)
    .round(2)
)


# ======================================================================================
# 16. SAVE SANDBOX ONLY
# ======================================================================================

whole_card_star_power = whole_card_star.copy()
fanfare_v2_star_test = fan_test.copy()


print("\n" + "=" * 120)
print("📦 SANDBOX OBJECTS")
print("=" * 120)

print(
    "whole_card_star_power:",
    whole_card_star_power.shape
)

print(
    "fanfare_v2_star_test:",
    fanfare_v2_star_test.shape
)


# ======================================================================================
# 17. SAFETY
# ======================================================================================

print("\n" + "=" * 120)
print("🔒 SAFETY")
print("=" * 120)

print("Published model modified: NO")
print("V2 Event Strength modified: NO")
print("Published Fanfare modified: NO")
print("Files overwritten: NO")

print("\n🔥 WHOLE-CARD STAR POWER TEST COMPLETE")
```

```python
# ======================================================================================
# WHOLE-CARD FANFARE V2 — MERGE FIX + FINISH ENTIRE STAR-POWER TEST
#
# Fixes the MergeError WITHOUT dropping fighters or changing the published model.
# ======================================================================================

import pandas as pd
import numpy as np

print("=" * 120)
print("🔧 FIXING MERGE + FINISHING WHOLE-CARD STAR POWER TEST")
print("=" * 120)


# ======================================================================================
# 1. RESTART THE FIGHTER TABLE CLEANLY
# ======================================================================================

stars = whole_card_fanfare_fighters.copy()
stars["date"] = pd.to_datetime(stars["date"])


# ======================================================================================
# 2. ADD OCCURRENCE NUMBER TO MERGE KEYS
#
# This safely distinguishes rare duplicate fighter/event keys.
# NOTHING gets dropped.
# ======================================================================================

KEYS = ["date", "event_name", "fighter_name"]

stars["_merge_occurrence"] = (
    stars.groupby(KEYS, dropna=False)
    .cumcount()
)

career["_merge_occurrence"] = (
    career.groupby(KEYS, dropna=False)
    .cumcount()
)

mainhist["_merge_occurrence"] = (
    mainhist.groupby(KEYS, dropna=False)
    .cumcount()
)


career_small = career[
    KEYS
    + [
        "_merge_occurrence",
        "prior_title_fights_rebuilt"
    ]
].copy()

main_small = mainhist[
    KEYS
    + [
        "_merge_occurrence",
        "prior_main_events_rebuilt"
    ]
].copy()


print("\n📋 DUPLICATE-KEY DIAGNOSTIC")
print("-" * 75)

print(
    "Star rows:",
    len(stars)
)

print(
    "Duplicate star keys:",
    int(stars.duplicated(KEYS, keep=False).sum())
)

print(
    "Career rows:",
    len(career_small)
)

print(
    "Main-history rows:",
    len(main_small)
)


# ======================================================================================
# 3. SAFE ONE-TO-ONE MERGES
# ======================================================================================

stars = stars.merge(
    career_small,
    on=KEYS + ["_merge_occurrence"],
    how="left",
    validate="one_to_one"
)

stars = stars.merge(
    main_small,
    on=KEYS + ["_merge_occurrence"],
    how="left",
    validate="one_to_one"
)


print("\n✅ MERGE FIX")
print("-" * 75)

print("Rows after merge:", len(stars))
print(
    "Prior-title history available:",
    stars["prior_title_fights_rebuilt"].notna().sum()
)

print(
    "Prior-main-event history available:",
    stars["prior_main_events_rebuilt"].notna().sum()
)


# ======================================================================================
# 4. RANKING STAR SIGNAL
# ======================================================================================

RANKINGS_START = pd.Timestamp("2013-02-04")

stars["ranking_points"] = np.nan

rank_era = (
    stars["date"] >= RANKINGS_START
)

# During ranking era, unranked genuinely receives 0 ranking points.
stars.loc[
    rank_era,
    "ranking_points"
] = 0.0

ranked = (
    rank_era
    &
    stars["rank_before"].notna()
)

stars.loc[
    ranked,
    "ranking_points"
] = (
    stars.loc[
        ranked,
        "rank_before"
    ]
    .apply(
        lambda r:
            16
            if r == 0
            else max(16 - r, 0)
    )
)


# ======================================================================================
# 5. ZERO-PRESERVING NORMALIZATION
# ======================================================================================

def zp_percentile(series):

    result = pd.Series(
        np.nan,
        index=series.index,
        dtype=float
    )

    valid = series.notna()

    result.loc[
        valid & (series == 0)
    ] = 0.0

    positive = (
        valid
        &
        (series > 0)
    )

    if positive.sum() > 0:

        result.loc[positive] = (
            series.loc[positive]
            .rank(
                pct=True,
                method="average"
            )
            * 100
        )

    return result


stars["ufc_experience_star"] = (
    zp_percentile(
        stars["ufc_fights_before"]
    )
)

stars["title_history_star"] = (
    zp_percentile(
        stars["prior_title_fights_rebuilt"]
    )
)

stars["main_event_history_star"] = (
    zp_percentile(
        stars["prior_main_events_rebuilt"]
    )
)

stars["ranking_star"] = (
    zp_percentile(
        stars["ranking_points"]
    )
)


# ======================================================================================
# 6. FIGHTER STAR POWER
# ======================================================================================

STAR_WEIGHTS = {
    "main_event_history_star": 0.30,
    "title_history_star": 0.30,
    "ranking_star": 0.25,
    "ufc_experience_star": 0.15
}


def weighted_missing(row, weights):

    available = {
        col: weight
        for col, weight in weights.items()
        if pd.notna(row[col])
    }

    if not available:
        return np.nan

    denominator = sum(
        available.values()
    )

    return sum(
        row[col] * weight
        for col, weight in available.items()
    ) / denominator


stars["fighter_star_power"] = stars.apply(
    weighted_missing,
    axis=1,
    weights=STAR_WEIGHTS
)


# ======================================================================================
# 7. WHOLE-CARD STAR POWER
# ======================================================================================

stars["weighted_star_power"] = (
    stars["fighter_star_power"]
    *
    stars["fighter_card_weight_share"]
)

stars["available_star_weight"] = np.where(
    stars["fighter_star_power"].notna(),
    stars["fighter_card_weight_share"],
    0.0
)


whole_card_star = (
    stars
    .groupby(
        ["date", "event_name"],
        as_index=False
    )
    .agg(
        star_numerator=(
            "weighted_star_power",
            "sum"
        ),

        star_coverage=(
            "available_star_weight",
            "sum"
        ),

        fighters_with_star_score=(
            "fighter_star_power",
            lambda s: s.notna().sum()
        ),

        total_fighters=(
            "fighter_name",
            "size"
        )
    )
)


whole_card_star["whole_card_star_power"] = np.where(
    whole_card_star["star_coverage"] > 0,

    whole_card_star["star_numerator"]
    /
    whole_card_star["star_coverage"],

    np.nan
)


# ======================================================================================
# 8. MERGE INTO EXISTING FANFARE V2 PROTOTYPE
# ======================================================================================

fan_test = fanfare_v2_prototype.merge(
    whole_card_star[
        [
            "date",
            "event_name",
            "whole_card_star_power",
            "star_coverage",
            "fighters_with_star_score"
        ]
    ],
    on=["date", "event_name"],
    how="left",
    validate="one_to_one"
)


# ======================================================================================
# 9. EXPERIMENTAL FANFARE V2
#
# 40% whole-card star proxy
# 30% whole-card historical drawing
# 20% commercial demand
# 10% search acceleration
#
# Search acceleration NEVER expands beyond 10%.
# ======================================================================================

def final_fanfare_test(row):

    # Fixed 10% search block
    search_contribution = 0.0

    if pd.notna(
        row["search_acceleration_v2"]
    ):
        search_contribution = (
            row["search_acceleration_v2"]
            * 0.10
        )

    # Substantive evidence
    substantive = {
        "whole_card_star_power": 0.40,
        "whole_card_drawing_adjusted": 0.30,
        "commercial_demand_v2": 0.20
    }

    available = {
        col: weight
        for col, weight in substantive.items()
        if pd.notna(row[col])
    }

    if not available:

        if pd.notna(
            row["search_acceleration_v2"]
        ):
            return search_contribution

        return np.nan

    denominator = sum(
        available.values()
    )

    substantive_score = sum(
        row[col] * weight
        for col, weight in available.items()
    ) / denominator

    return (
        substantive_score * 0.90
        +
        search_contribution
    )


fan_test["fanfare_v2_star_proxy"] = (
    fan_test.apply(
        final_fanfare_test,
        axis=1
    )
)


# ======================================================================================
# 10. KEY EVENTS
# ======================================================================================

targets = [
    "UFC 100",
    "UFC 168",
    "UFC 205",
    "UFC 214",
    "UFC 217",
    "UFC 229",
    "UFC 259",
    "UFC 269",
    "UFC 300",
    "UFC 302",
    "UFC 329",
    "UFC 330",
    "UFC 331"
]

mask = pd.Series(
    False,
    index=fan_test.index
)

for target in targets:

    mask |= (
        fan_test["event_name"]
        .str.contains(
            target,
            case=False,
            regex=False,
            na=False
        )
    )


print("\n" + "=" * 120)
print("🎯 KEY EVENTS — NEW FANFARE TEST")
print("=" * 120)

display(
    fan_test.loc[
        mask,
        [
            "date",
            "event_name",
            "whole_card_star_power",
            "star_coverage",
            "whole_card_drawing_adjusted",
            "commercial_demand_v2",
            "search_acceleration_v2",
            "fanfare_v2_star_proxy",
            "current_fanfare_score"
        ]
    ]
    .sort_values("date")
    .round(2)
)


# ======================================================================================
# 11. TOP 30 STAR-POWER CARDS
# ======================================================================================

print("\n" + "=" * 120)
print("⭐ TOP 30 — WHOLE-CARD STAR POWER")
print("=" * 120)

display(
    whole_card_star
    .sort_values(
        "whole_card_star_power",
        ascending=False
    )
    .head(30)
    [
        [
            "date",
            "event_name",
            "whole_card_star_power",
            "star_coverage",
            "fighters_with_star_score",
            "total_fighters"
        ]
    ]
    .round(2)
)


# ======================================================================================
# 12. TOP 30 EXPERIMENTAL FANFARE
# ======================================================================================

print("\n" + "=" * 120)
print("🔥 TOP 30 — EXPERIMENTAL WHOLE-CARD FANFARE")
print("=" * 120)

display(
    fan_test
    .sort_values(
        "fanfare_v2_star_proxy",
        ascending=False
    )
    [
        [
            "date",
            "event_name",
            "whole_card_star_power",
            "whole_card_drawing_adjusted",
            "commercial_demand_v2",
            "search_acceleration_v2",
            "fanfare_v2_star_proxy",
            "current_fanfare_score"
        ]
    ]
    .head(30)
    .round(2)
)


# ======================================================================================
# 13. OLD VS NEW FANFARE
# ======================================================================================

comparison = fan_test[
    [
        "date",
        "event_name",
        "current_fanfare_score",
        "fanfare_v2_star_proxy"
    ]
].dropna().copy()


comparison["change"] = (
    comparison["fanfare_v2_star_proxy"]
    -
    comparison["current_fanfare_score"]
)


print("\n" + "=" * 120)
print("📊 OLD VS EXPERIMENTAL FANFARE")
print("=" * 120)

print(
    "Events compared:",
    len(comparison)
)

print(
    "Correlation:",
    round(
        comparison[
            [
                "current_fanfare_score",
                "fanfare_v2_star_proxy"
            ]
        ].corr().iloc[0, 1],
        4
    )
)

print(
    "Median absolute change:",
    round(
        comparison["change"]
        .abs()
        .median(),
        2
    )
)


print("\n⬆️ BIGGEST INCREASES")

display(
    comparison
    .sort_values(
        "change",
        ascending=False
    )
    .head(15)
    .round(2)
)


print("\n⬇️ BIGGEST DECREASES")

display(
    comparison
    .sort_values(
        "change",
        ascending=True
    )
    .head(15)
    .round(2)
)


# ======================================================================================
# 14. HARD VALIDATION
# ======================================================================================

print("\n" + "=" * 120)
print("✅ HARD VALIDATION")
print("=" * 120)

checks = {
    "Fighter rows preserved = 17,754":
        len(stars) == 17754,

    "Event rows = 788":
        len(whole_card_star) == 788,

    "No duplicate event rows":
        not whole_card_star.duplicated(
            ["date", "event_name"]
        ).any(),

    "Star coverage <= 1":
        (
            whole_card_star[
                "star_coverage"
            ] <= 1.000001
        ).all(),

    "Star scores within 0-100":
        (
            whole_card_star[
                "whole_card_star_power"
            ]
            .dropna()
            .between(0, 100)
            .all()
        ),

    "Fanfare test within 0-100":
        (
            fan_test[
                "fanfare_v2_star_proxy"
            ]
            .dropna()
            .between(0, 100)
            .all()
        )
}


for label, passed in checks.items():

    print(
        "✅" if passed else "❌",
        label
    )


# ======================================================================================
# 15. SAVE SANDBOX
# ======================================================================================

whole_card_star_power = whole_card_star.copy()
fanfare_v2_star_test = fan_test.copy()


print("\n📦 SANDBOX OBJECTS")
print(
    "whole_card_star_power:",
    whole_card_star_power.shape
)

print(
    "fanfare_v2_star_test:",
    fanfare_v2_star_test.shape
)


print("\n🔒 Published model modified: NO")
print("🔒 V2 Event Strength modified: NO")
print("🔒 Published Fanfare modified: NO")
print("🔒 Files overwritten: NO")

print("\n🔥 TEST COMPLETE")
```

```python
# ======================================================================================
# UFC EVENT STRENGTH INDEX V2 — GITHUB VISUALIZATION PACK
#
# Creates:
#   1. Top 15 UFC Events by V2 Event Strength
#   2. V1 vs V2 Rankings — all 788 events
#   3. Ranking Depth vs V2 Event Strength
#   4. V2 Overall Model Weights
#   5. Average Event Strength by Year
#   6. Selected Event Rank Movement: V1 → V2
#
# Also exports:
#   TOP_15_UFC_EVENTS_V2.csv
#
# MODEL/DATA ARE NOT MODIFIED.
# ======================================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# ------------------------------------------------------------------------------
# SETUP
# ------------------------------------------------------------------------------

viz = v2_full.copy()

viz["event_date"] = pd.to_datetime(viz["event_date"])
viz["year"] = viz["event_date"].dt.year

OUT = Path("/content/UFC_V2_VISUALS")
OUT.mkdir(parents=True, exist_ok=True)

plt.rcParams.update({
    "figure.dpi": 120,
    "savefig.dpi": 220,
    "font.size": 11,
    "axes.titlesize": 17,
    "axes.labelsize": 12,
    "figure.titlesize": 18
})

print("=" * 110)
print("📊 UFC EVENT STRENGTH INDEX V2 — VISUALIZATION PACK")
print("=" * 110)


# ======================================================================================
# VISUAL 1 — TOP 15 UFC EVENTS
# ======================================================================================

top15 = (
    viz
    .sort_values("v2_rank")
    .head(15)
    .sort_values("v2_event_strength_score")
)

fig, ax = plt.subplots(figsize=(12, 8))

bars = ax.barh(
    top15["event_name"],
    top15["v2_event_strength_score"]
)

ax.set_title(
    "Top 15 UFC Events by V2 Event Strength",
    fontweight="bold",
    pad=15
)

ax.set_xlabel("V2 Event Strength Score")
ax.set_ylabel("")

ax.grid(
    axis="x",
    alpha=0.20
)

for bar, value in zip(
    bars,
    top15["v2_event_strength_score"]
):
    ax.text(
        value + 0.25,
        bar.get_y() + bar.get_height()/2,
        f"{value:.1f}",
        va="center",
        fontsize=9
    )

ax.set_xlim(
    0,
    max(top15["v2_event_strength_score"]) + 6
)

plt.tight_layout()

path1 = OUT / "top_15_ufc_events_v2.png"

plt.savefig(
    path1,
    bbox_inches="tight"
)

plt.show()


# ======================================================================================
# VISUAL 2 — V1 VS V2 RANKINGS
#
# Perfect agreement would sit on diagonal.
# This shows how stable the redesign is across all 788 events.
# ======================================================================================

fig, ax = plt.subplots(figsize=(9, 9))

ax.scatter(
    viz["event_strength_rank"],
    viz["v2_rank"],
    alpha=0.40,
    s=25
)

ax.plot(
    [1, 788],
    [1, 788],
    linestyle="--",
    linewidth=1.5
)

ax.set_title(
    "Published Model vs V2 Event Rankings",
    fontweight="bold",
    pad=15
)

ax.set_xlabel("Published Model Rank")
ax.set_ylabel("V2 Rank")

ax.set_xlim(0, 800)
ax.set_ylim(800, 0)
ax.invert_xaxis()

ax.grid(alpha=0.20)

plt.tight_layout()

path2 = OUT / "published_vs_v2_rankings.png"

plt.savefig(
    path2,
    bbox_inches="tight"
)

plt.show()


# ======================================================================================
# VISUAL 3 — RANKING DEPTH VS EVENT STRENGTH
#
# Only ranking-era events are appropriate here.
# ======================================================================================

ranking_viz = viz[
    viz["ranking_depth_index"].notna()
].copy()

corr = ranking_viz[
    [
        "ranking_depth_index",
        "v2_event_strength_score"
    ]
].corr().iloc[0, 1]

fig, ax = plt.subplots(figsize=(10, 7))

ax.scatter(
    ranking_viz["ranking_depth_index"],
    ranking_viz["v2_event_strength_score"],
    alpha=0.55,
    s=35
)

# Simple trend line
x = ranking_viz["ranking_depth_index"].values
y = ranking_viz["v2_event_strength_score"].values

coef = np.polyfit(x, y, 1)

x_line = np.linspace(
    x.min(),
    x.max(),
    100
)

y_line = (
    coef[0] * x_line
    + coef[1]
)

ax.plot(
    x_line,
    y_line,
    linewidth=2
)

ax.set_title(
    "Ranking Depth vs V2 Event Strength",
    fontweight="bold",
    pad=15
)

ax.set_xlabel("Ranking Depth Index")
ax.set_ylabel("V2 Event Strength Score")

ax.grid(alpha=0.20)

ax.text(
    0.03,
    0.95,
    f"Correlation: {corr:.3f}",
    transform=ax.transAxes,
    va="top",
    fontsize=11
)

plt.tight_layout()

path3 = OUT / "ranking_depth_vs_event_strength.png"

plt.savefig(
    path3,
    bbox_inches="tight"
)

plt.show()


# ======================================================================================
# VISUAL 4 — V2 MODEL WEIGHTS
# ======================================================================================

weights = pd.DataFrame({
    "Component": [
        "Card Quality",
        "Star Experience",
        "Main Event Strength",
        "Fanfare"
    ],
    "Weight": [
        46,
        22,
        22,
        10
    ]
})

fig, ax = plt.subplots(figsize=(10, 6))

bars = ax.barh(
    weights["Component"],
    weights["Weight"]
)

ax.set_title(
    "UFC Event Strength Index V2 — Model Weights",
    fontweight="bold",
    pad=15
)

ax.set_xlabel("Weight (%)")
ax.set_ylabel("")
ax.set_xlim(0, 52)

ax.grid(
    axis="x",
    alpha=0.20
)

for bar, value in zip(
    bars,
    weights["Weight"]
):
    ax.text(
        value + 0.5,
        bar.get_y() + bar.get_height()/2,
        f"{value}%",
        va="center",
        fontweight="bold"
    )

plt.tight_layout()

path4 = OUT / "v2_model_weights.png"

plt.savefig(
    path4,
    bbox_inches="tight"
)

plt.show()


# ======================================================================================
# VISUAL 5 — EVENT STRENGTH OVER TIME
#
# Annual mean gives a much cleaner historical trend than plotting
# all 788 individual events on one line.
# ======================================================================================

yearly = (
    viz
    .groupby("year", as_index=False)
    .agg(
        average_v2_strength=(
            "v2_event_strength_score",
            "mean"
        ),
        events=(
            "event_name",
            "size"
        )
    )
)

fig, ax = plt.subplots(figsize=(12, 7))

ax.plot(
    yearly["year"],
    yearly["average_v2_strength"],
    marker="o",
    linewidth=2
)

ax.set_title(
    "Average UFC Event Strength by Year",
    fontweight="bold",
    pad=15
)

ax.set_xlabel("Year")
ax.set_ylabel("Average V2 Event Strength")

ax.grid(alpha=0.20)

plt.tight_layout()

path5 = OUT / "event_strength_by_year_v2.png"

plt.savefig(
    path5,
    bbox_inches="tight"
)

plt.show()


# ======================================================================================
# VISUAL 6 — SELECTED V1 → V2 RANK MOVEMENT
#
# Uses real values from the dataset.
# ======================================================================================

selected_patterns = [
    "UFC 205:",
    "UFC 217:",
    "UFC 229:",
    "UFC 269:",
    "UFC 300:",
    "UFC 302:",
    "UFC 329:",
    "UFC 330:",
    "UFC 331:"
]

selected_mask = pd.Series(
    False,
    index=viz.index
)

for pattern in selected_patterns:
    selected_mask |= (
        viz["event_name"]
        .str.contains(
            pattern,
            case=False,
            regex=False,
            na=False
        )
    )

movement = (
    viz.loc[
        selected_mask,
        [
            "event_name",
            "event_strength_rank",
            "v2_rank",
            "rank_change_vs_published"
        ]
    ]
    .copy()
    .sort_values("v2_rank")
)

# Shorten labels for readability
movement["display_name"] = (
    movement["event_name"]
    .str.replace(
        r"^UFC ",
        "UFC ",
        regex=True
    )
)

fig, ax = plt.subplots(figsize=(12, 8))

ypos = np.arange(len(movement))

for i, row in enumerate(
    movement.itertuples()
):
    ax.plot(
        [
            row.event_strength_rank,
            row.v2_rank
        ],
        [i, i],
        linewidth=2
    )

    ax.scatter(
        row.event_strength_rank,
        i,
        s=70
    )

    ax.scatter(
        row.v2_rank,
        i,
        s=70
    )

ax.set_yticks(ypos)
ax.set_yticklabels(
    movement["display_name"]
)

ax.set_xlabel(
    "Event Rank (lower is better)"
)

ax.set_title(
    "Selected UFC Events: Published Rank vs V2 Rank",
    fontweight="bold",
    pad=15
)

ax.grid(
    axis="x",
    alpha=0.20
)

ax.invert_xaxis()
ax.invert_yaxis()

plt.tight_layout()

path6 = OUT / "selected_event_rank_movement.png"

plt.savefig(
    path6,
    bbox_inches="tight"
)

plt.show()


# ======================================================================================
# EXPORT TOP 15 TABLE
# ======================================================================================

top15_export = (
    viz
    .sort_values("v2_rank")
    .head(15)
    [
        [
            "v2_rank",
            "event_date",
            "event_name",
            "v2_event_strength_score",
            "ranking_depth_index",
            "clean_rank30_card_quality",
            "star_experience_score",
            "main_event_score",
            "fanfare_score"
        ]
    ]
)

top15_csv = OUT / "TOP_15_UFC_EVENTS_V2.csv"

top15_export.to_csv(
    top15_csv,
    index=False
)


# ======================================================================================
# FINAL SUMMARY
# ======================================================================================

print("\n" + "=" * 110)
print("✅ VISUALIZATION PACK COMPLETE")
print("=" * 110)

files = [
    path1,
    path2,
    path3,
    path4,
    path5,
    path6,
    top15_csv
]

for file in files:
    print("✓", file.name)

print(
    "\nRanking-era events in scatter:",
    len(ranking_viz)
)

print(
    "Ranking Depth / Event Strength correlation:",
    round(corr, 4)
)

print("\nTop 15 V2 events:")
display(
    top15_export[
        [
            "v2_rank",
            "event_name",
            "v2_event_strength_score",
            "ranking_depth_index"
        ]
    ].round(2)
)

print("\n📁 Folder:")
print(OUT)

print("\n🔒 Model modified: NO")
print("🔒 V2 scores modified: NO")
print("🔒 GitHub modified: NO")
```

```python

import pandas as pd
import numpy as np

df = ufc.copy()

models = {
    "50/25/25 (Current)": (0.50, 0.25, 0.25),
    "55/25/20": (0.55, 0.25, 0.20),
    "60/25/15": (0.60, 0.25, 0.15),
}

components = [
    "clean_rank30_card_quality",
    "star_experience_score",
    "main_event_score"
]

for name, weights in models.items():
    values = df[components]
    w = pd.Series(weights, index=components)

    # Adjust weights only for missing components
    numerator = values.mul(w).sum(axis=1)
    denominator = values.notna().mul(w).sum(axis=1)

    df[name + " Score"] = numerator.div(
        denominator.replace(0, np.nan)
    )
    df[name + " Rank"] = (
        df[name + " Score"].rank(
            ascending=False, method="min"
        ).astype("Int64")
    )

selected = df[
    df["event_name"].str.contains(
        "UFC 300|UFC 217|UFC 214|UFC Freedom 250",
        case=False, na=False
    )
]

cols = ["event_name"]
for name in models:
    cols.extend([name + " Score", name + " Rank"])

display(selected[cols].round(2))
```

```python

import requests
import pandas as pd
from bs4 import BeautifulSoup
from urllib.parse import urljoin

url = "http://ufcstats.com/statistics/events/completed?page=all"

response = requests.get(
    url.replace("http://", "https://"),
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=30
)

print("Website status:", response.status_code)
print("Response size:", len(response.content), "bytes")
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

rows = []

for tr in soup.select("tr.b-statistics__table-row"):
    link = tr.select_one("a.b-link")
    if not link:
        continue

    cells = tr.find_all("td")
    date_text = ""

    for cell in cells:
        text = cell.get_text(" ", strip=True)
        parsed = pd.to_datetime(text, errors="coerce")
        if pd.notna(parsed):
            date_text = text
            break

    if date_text:
        rows.append({
            "event_name": link.get_text(" ", strip=True),
            "event_date": date_text,
            "event_url": urljoin(url, link.get("href", ""))
        })

print("Total events extracted:", len(rows))

if not rows:
    print("\nNo events extracted.")
    print("Page title:",
          soup.title.get_text(" ", strip=True)
          if soup.title else "None")
    print("First 400 characters:")
    print(response.text[:400])
else:
    official = pd.DataFrame(rows)
    official["event_date"] = pd.to_datetime(
        official["event_date"], errors="coerce"
    )

    official = official[
        official["event_date"].between(
            "2026-06-28", "2026-10-17"
        )
    ].sort_values("event_date")

    print("\nRECENT UFC EVENTS FOUND:", len(official))
    display(official)

    official.to_csv(
        "UFC_OFFICIAL_EVENT_LINKS_2026.csv",
        index=False
    )
```

```python

import pandas as pd
import requests
from io import BytesIO

base = "https://raw.githubusercontent.com/Greco1899/scrape_ufc_stats/main/"

datasets = {}

for filename in [
    "ufc_event_details.csv",
    "ufc_fight_results.csv",
    "ufc_fight_details.csv"
]:
    url = base + filename
    try:
        response = requests.get(url, timeout=40)
        response.raise_for_status()

        df = pd.read_csv(BytesIO(response.content))
        datasets[filename] = df

        print(f"\nSUCCESS: {filename}")
        print("Rows:", len(df))
        print("Columns:", df.columns.tolist())
        display(df.head(2))

    except Exception as e:
        print(f"\nFAILED: {filename}")
        print(str(e)[:200])

print("\nFILES SUCCESSFULLY LOADED:", len(datasets))
```

```python

import pandas as pd

source_events = datasets["ufc_event_details.csv"].copy()
source_fights = datasets["ufc_fight_results.csv"].copy()

source_events["event_date"] = pd.to_datetime(
    source_events["DATE"], errors="coerce"
)

source_events = source_events[
    source_events["event_date"].between(
        "2026-06-28", "2026-10-17"
    )
].copy()

# Exclude any event with no fight results.
fight_counts = (
    source_fights.groupby("EVENT")
    .size()
    .rename("fight_count")
    .reset_index()
)

recent_events = source_events.merge(
    fight_counts,
    left_on="EVENT",
    right_on="EVENT",
    how="left"
)

recent_events["fight_count"] = (
    recent_events["fight_count"].fillna(0).astype(int)
)

recent_events = recent_events.sort_values("event_date")

print("RECENT EVENTS:", len(recent_events))
print("RECENT FIGHTS:", recent_events["fight_count"].sum())

display(
    recent_events[
        ["event_date", "EVENT", "fight_count"]
    ]
)

# Keep only fights belonging to the recent events.
recent_fights = source_fights[
    source_fights["EVENT"].isin(recent_events["EVENT"])
].copy()

print("\nRECENT FIGHT RECORDS:", len(recent_fights))
print("DUPLICATE FIGHT URLS:",
      recent_fights["URL"].duplicated().sum())

print("\nEVENTS WITHOUT FIGHT RESULTS:")
display(
    recent_events.loc[
        recent_events["fight_count"] == 0,
        ["event_date", "EVENT"]
    ]
)
```

```python

import pandas as pd
import re

events_df = datasets["ufc_event_details.csv"].copy()
fights_df = datasets["ufc_fight_results.csv"].copy()

def clean_name(value):
    value = str(value)
    value = value.replace("\xa0", " ")
    value = re.sub(r"\s+", " ", value)
    return value.strip().casefold()

events_df["match_key"] = events_df["EVENT"].map(clean_name)
fights_df["match_key"] = fights_df["EVENT"].map(clean_name)

events_df["event_date"] = pd.to_datetime(
    events_df["DATE"], errors="coerce"
)

recent = events_df[
    events_df["event_date"].between(
        "2026-06-28", "2026-10-17"
    )
].copy()

counts = fights_df.groupby("match_key").size()

recent["fight_count"] = (
    recent["match_key"].map(counts)
    .fillna(0)
    .astype(int)
)

print("RECENT EVENTS:", len(recent))
print("MATCHED FIGHTS:", recent["fight_count"].sum())

display(
    recent[
        ["event_date", "EVENT", "fight_count"]
    ].sort_values("event_date")
)

print("\nEXAMPLE EVENT NAME FROM EACH FILE:")
print("Events:", repr(events_df["EVENT"].iloc[0]))
print("Fights:", repr(fights_df["EVENT"].iloc[0]))

print("\nEXAMPLE MATCH KEYS:")
print("Events:", repr(events_df["match_key"].iloc[0]))
print("Fights:", repr(fights_df["match_key"].iloc[0]))

print("\nNUMBER OF MATCHING EVENT NAMES:",
      len(set(events_df["match_key"]) &
          set(fights_df["match_key"])))

if recent["fight_count"].sum() > 0:
    recent_fights = fights_df[
        fights_df["match_key"].isin(recent["match_key"])
    ].copy()
    print("\nSUCCESS — Recent fights recovered:",
          len(recent_fights))
else:
    print("\nStill zero matches — we'll inspect the source formatting next.")
```

```python

from collections import defaultdict
import pandas as pd
import numpy as np
import re

# Use the original historical fight data plus recent results.
history = backup["ufc_fights"].copy()
fresh = recent_fights.copy()

history["event_date"] = pd.to_datetime(history["date"])
fresh["event_date"] = pd.to_datetime(
    fresh["match_key"].map(
        recent.drop_duplicates("match_key")
              .set_index("match_key")["event_date"]
    )
)

# Standardize new fight results.
outcomes = fresh["OUTCOME"].astype(str).str.strip().str.upper()
split_outcomes = outcomes.str.split("/", n=1, expand=True)

fresh["f1_result"] = split_outcomes[0].str.strip()
fresh["f2_result"] = split_outcomes[1].str.strip()

names = fresh["BOUT"].str.split(
    r"\s+vs\.\s+", n=1, expand=True
)
fresh["f1_name"] = names[0].str.strip()
fresh["f2_name"] = names[1].str.strip()
fresh["event_name"] = fresh["EVENT"].str.strip()

# Preserve the original historical results.
history["f1_result"] = history["f1_result"].astype(str).str.upper()
history["f2_result"] = history["f2_result"].astype(str).str.upper()

all_results = pd.concat([
    history[[
        "event_date", "event_name",
        "f1_name", "f2_name", "f1_result", "f2_result"
    ]],
    fresh[[
        "event_date", "event_name",
        "f1_name", "f2_name", "f1_result", "f2_result"
    ]]
], ignore_index=True).sort_values(
    ["event_date", "event_name"]
).reset_index(drop=True)

stats = defaultdict(lambda: {
    "fights": 0, "wins": 0, "losses": 0, "streak": 0
})

records = []

# Process events as groups so fighters' records are pre-event.
for (date, event), group in all_results.groupby(
    ["event_date", "event_name"], sort=True
):
    event_records = []

    for _, fight in group.iterrows():
        row = fight.to_dict()

        for prefix in ["f1", "f2"]:
            name = row[f"{prefix}_name"]
            s = stats[name]

            row[f"{prefix}_fights_before"] = s["fights"]
            row[f"{prefix}_wins_before"] = s["wins"]
            row[f"{prefix}_losses_before"] = s["losses"]
            row[f"{prefix}_win_streak_before"] = s["streak"]
            row[f"{prefix}_win_pct_before"] = (
                s["wins"] / (s["wins"] + s["losses"])
                if s["wins"] + s["losses"] > 0
                else np.nan
            )

        event_records.append(row)

    records.extend(event_records)

    # Update fighter histories only after capturing pre-event stats.
    for row in event_records:
        for prefix in ["f1", "f2"]:
            name = row[f"{prefix}_name"]
            result = str(row[f"{prefix}_result"]).upper()
            s = stats[name]

            s["fights"] += 1

            if result == "W":
                s["wins"] += 1
                s["streak"] += 1
            elif result == "L":
                s["losses"] += 1
                s["streak"] = 0
            elif result in ["D", "DRAW"]:
                s["streak"] = 0
            # NC doesn't count as a win or loss.

calculated_fights = pd.DataFrame(records)

recent_calculated = calculated_fights[
    calculated_fights["event_date"] >= "2026-07-01"
].copy()

print("TOTAL FIGHTS PROCESSED:", len(calculated_fights))
print("RECENT FIGHTS:", len(recent_calculated))

print("\nRECENT MAIN EVENTS — FIGHTER HISTORIES")
main_event_names = set(recent["EVENT"].str.strip())

main_event_rows = recent_calculated[
    recent_calculated["event_name"].isin(main_event_names)
].groupby("event_name", sort=False).head(1)

display(main_event_rows[[
    "event_name",
    "f1_name", "f2_name",
    "f1_fights_before", "f2_fights_before",
    "f1_win_pct_before", "f2_win_pct_before",
    "f1_win_streak_before", "f2_win_streak_before"
]].round(3))
```

```python

import pandas as pd
import numpy as np

original = backup["ufc_fights"].copy()
original["event_date"] = pd.to_datetime(original["date"])

compare = original.merge(
    calculated_fights,
    on=["event_date", "event_name", "f1_name", "f2_name"],
    suffixes=("_original", "_calculated"),
    how="inner"
)

checks = {
    "UFC fights": (
        "f1_ufc_fights_before", "f1_fights_before"
    ),
    "UFC wins": (
        "f1_ufc_wins_before", "f1_wins_before"
    ),
    "Win streak": (
        "f1_win_streak_before", "f1_win_streak_before"
    ),
    "UFC win percentage": (
        "f1_ufc_win_pct_before", "f1_win_pct_before"
    )
}

print("HISTORICAL FIGHTS MATCHED:", len(compare))

for label, (original_col, calculated_col) in checks.items():
    original_values = pd.to_numeric(
        compare[original_col], errors="coerce"
    )
    calculated_values = pd.to_numeric(
        compare[calculated_col], errors="coerce"
    )

    differences = ~np.isclose(
        original_values,
        calculated_values,
        equal_nan=True,
        atol=0.0001
    )

    print(f"{label}: {differences.sum()} mismatches")

print("\nEXAMPLE MISMATCHES:")
display(
    compare.loc[
        ~np.isclose(
            pd.to_numeric(
                compare["f1_ufc_fights_before"],
                errors="coerce"
            ),
            pd.to_numeric(
                compare["f1_fights_before"],
                errors="coerce"
            ),
            equal_nan=True
        ),
        [
            "event_name", "f1_name",
            "f1_ufc_fights_before",
            "f1_fights_before"
        ]
    ].head(10)
)
```

```python

import pandas as pd
import numpy as np

original = backup["ufc_fights"].copy()
original["fight_key"] = (
    original["fight_url"].astype(str)
    .str.strip()
    .str.lower()
    .str.replace("https://", "http://", regex=False)
)

calculated = calculated_fights.copy()

# Recover fight URLs by matching the original fight order
# to the source records, without merging on fighter names.
source = combined_fights[
    ["event_date", "event_name", "fight_url",
     "f1_name", "f2_name"]
].copy()

source["fight_key"] = (
    source["fight_url"].astype(str)
    .str.strip()
    .str.lower()
    .str.replace("https://", "http://", regex=False)
)

# Identify potentially ambiguous fighter-name matches.
match_cols = [
    "event_date", "event_name", "f1_name", "f2_name"
]

ambiguous = source.duplicated(match_cols, keep=False)

print("AMBIGUOUS FIGHTER-NAME MATCHES:",
      ambiguous.sum())

# Attach unique fight URLs to calculated records.
source_unique = source.drop_duplicates(
    match_cols, keep=False
)

calculated_unique = calculated.merge(
    source_unique[match_cols + ["fight_key"]],
    on=match_cols,
    how="inner",
    validate="one_to_one"
)

compare = original.merge(
    calculated_unique,
    on="fight_key",
    suffixes=("_original", "_calculated"),
    validate="one_to_one"
)

print("HISTORICAL FIGHTS VALIDATED:", len(compare))
print("EXPECTED:", len(original))

checks = [
    ("UFC fights", "ufc_fights_before", "fights_before"),
    ("UFC wins", "ufc_wins_before", "wins_before"),
    ("Win streak", "win_streak_before", "win_streak_before"),
    ("Win percentage", "ufc_win_pct_before", "win_pct_before")
]

summary = []

for fighter in ["f1", "f2"]:
    for label, old, new in checks:
        old_col = f"{fighter}_{old}"
        new_col = f"{fighter}_{new}"

        # Resolve overlapping column names after merging.
        if old_col == new_col:
            old_col += "_original"
            new_col += "_calculated"

        a = pd.to_numeric(compare[old_col], errors="coerce")
        b = pd.to_numeric(compare[new_col], errors="coerce")

        mismatches = ~np.isclose(
            a, b, equal_nan=True, atol=0.0001
        )

        summary.append({
            "fighter": fighter,
            "measurement": label,
            "mismatches": int(mismatches.sum()),
            "total_compared": len(compare)
        })

print("\nVALIDATION RESULTS")
display(pd.DataFrame(summary))

print("\nHISTORICAL FIGHTS NOT MATCHED:",
      len(original) - len(compare))
```

```python

import pandas as pd
import numpy as np
from collections import defaultdict

history = backup["ufc_fights"].copy()
history["event_date"] = pd.to_datetime(history["date"])

fresh = recent_fights.copy()
fresh["event_date"] = pd.to_datetime(
    fresh["match_key"].map(
        recent.drop_duplicates("match_key")
              .set_index("match_key")["event_date"]
    )
)

fresh["fight_url"] = fresh["URL"].astype(str).str.strip()
fresh["event_name"] = fresh["EVENT"].astype(str).str.strip()

names = fresh["BOUT"].str.split(
    r"\s+vs\.\s+", n=1, expand=True
)
fresh["f1_name"] = names[0].str.strip()
fresh["f2_name"] = names[1].str.strip()

outcomes = fresh["OUTCOME"].astype(str).str.upper().str.strip()
parts = outcomes.str.split("/", n=1, expand=True)
fresh["f1_result"] = parts[0].str.strip()
fresh["f2_result"] = parts[1].str.strip()

cols = [
    "fight_url", "event_date", "event_name",
    "f1_name", "f2_name", "f1_result", "f2_result"
]

all_fights = pd.concat(
    [history[cols], fresh[cols]],
    ignore_index=True
)

all_fights = all_fights.sort_values(
    ["event_date", "event_name", "fight_url"]
).reset_index(drop=True)

stats = defaultdict(lambda: {
    "fights": 0, "wins": 0,
    "losses": 0, "streak": 0
})

records = []

for (date, event), group in all_fights.groupby(
    ["event_date", "event_name"], sort=True
):
    event_rows = []

    for _, fight in group.iterrows():
        row = fight.to_dict()

        for prefix in ["f1", "f2"]:
            s = stats[row[f"{prefix}_name"]]

            row[f"{prefix}_fights_before"] = s["fights"]
            row[f"{prefix}_wins_before"] = s["wins"]
            row[f"{prefix}_win_streak_before"] = s["streak"]
            row[f"{prefix}_win_pct_before"] = (
                s["wins"] / (s["wins"] + s["losses"])
                if s["wins"] + s["losses"] else np.nan
            )

        event_rows.append(row)

    records.extend(event_rows)

    for row in event_rows:
        for prefix in ["f1", "f2"]:
            s = stats[row[f"{prefix}_name"]]
            result = str(row[f"{prefix}_result"]).upper()

            s["fights"] += 1

            if result == "W":
                s["wins"] += 1
                s["streak"] += 1
            elif result == "L":
                s["losses"] += 1
                s["streak"] = 0
            elif result in ["D", "DRAW"]:
                s["streak"] = 0

recalculated = pd.DataFrame(records)

def normalize_url(series):
    return (
        series.astype(str)
        .str.strip()
        .str.lower()
        .str.replace("https://", "http://", regex=False)
    )

history["fight_key"] = normalize_url(history["fight_url"])
recalculated["fight_key"] = normalize_url(
    recalculated["fight_url"]
)

assert history["fight_key"].is_unique
assert recalculated["fight_key"].is_unique

comparison = history.merge(
    recalculated,
    on="fight_key",
    suffixes=("_original", "_new"),
    validate="one_to_one"
)

print("HISTORICAL FIGHTS MATCHED:",
      len(comparison), "/", len(history))

checks = [
    ("UFC fights", "ufc_fights_before", "fights_before"),
    ("UFC wins", "ufc_wins_before", "wins_before"),
    ("Win streak", "win_streak_before", "win_streak_before"),
    ("Win percentage", "ufc_win_pct_before", "win_pct_before")
]

results = []

for fighter in ["f1", "f2"]:
    for label, old, new in checks:
        old_col = f"{fighter}_{old}"
        new_col = f"{fighter}_{new}"

        if old_col == new_col:
            old_col += "_original"
            new_col += "_new"

        a = pd.to_numeric(
            comparison[old_col], errors="coerce"
        )
        b = pd.to_numeric(
            comparison[new_col], errors="coerce"
        )

        mismatch = ~np.isclose(
            a, b, atol=0.0001, equal_nan=True
        )

        results.append({
            "fighter": fighter,
            "measurement": label,
            "mismatches": int(mismatch.sum())
        })

print("\nVALIDATION RESULTS:")
display(pd.DataFrame(results))
```

```python

import pandas as pd
import numpy as np

# Compare historical vs reconstructed values
summary = []

for fighter in ["f1", "f2"]:
    old = pd.to_numeric(
        comparison[f"{fighter}_ufc_win_pct_before"],
        errors="coerce"
    )
    new = pd.to_numeric(
        comparison[f"{fighter}_win_pct_before"],
        errors="coerce"
    )

    both_present = old.notna() & new.notna()
    difference = (old - new).abs()

    summary.append({
        "fighter": fighter,
        "both_present": int(both_present.sum()),
        "both_missing": int((old.isna() & new.isna()).sum()),
        "original_missing_only": int((old.isna() & new.notna()).sum()),
        "reconstructed_missing_only": int((old.notna() & new.isna()).sum()),
        "difference_under_0.001": int(
            (both_present & (difference > 0.0001) &
             (difference < 0.001)).sum()
        ),
        "difference_over_0.01": int(
            (both_present & (difference > 0.01)).sum()
        )
    })

print("WIN PERCENTAGE DIAGNOSTICS:")
display(pd.DataFrame(summary))

# Show the largest meaningful discrepancies
examples = []

for fighter in ["f1", "f2"]:
    old = pd.to_numeric(
        comparison[f"{fighter}_ufc_win_pct_before"],
        errors="coerce"
    )
    new = pd.to_numeric(
        comparison[f"{fighter}_win_pct_before"],
        errors="coerce"
    )

    subset = comparison.loc[
        (old - new).abs() > 0.05,
        ["event_name_original", f"{fighter}_name_original",
         f"{fighter}_ufc_wins_before",
         f"{fighter}_ufc_fights_before",
         f"{fighter}_ufc_win_pct_before",
         f"{fighter}_win_pct_before"]
    ].copy()

    subset["fighter_side"] = fighter
    examples.append(subset.head(10))

print("\nEXAMPLES OF LARGE DIFFERENCES:")
display(pd.concat(examples, ignore_index=True))
```

```python

import pandas as pd
import numpy as np

old = backup["ufc_fights"].copy()
old["event_date"] = pd.to_datetime(old["date"])

new = fresh.copy()
new["event_date"] = pd.to_datetime(new["event_date"])

# Normalize fighter names consistently
def fighter_key(name):
    return " ".join(str(name).strip().casefold().split())

# Historical main-event experience:
# Use the most recent observed PRE-FIGHT count, then add
# that fighter's main-event appearance from that fight.
historical_main_event_counts = {}

old = old.sort_values(
    ["event_date", "event_name", "fight_url"]
)

for _, row in old.iterrows():
    for side in ["f1", "f2"]:
        fighter = fighter_key(row[f"{side}_name"])
        previous = pd.to_numeric(
            row[f"{side}_previous_main_events"],
            errors="coerce"
        )

        if pd.notna(previous):
            historical_main_event_counts[fighter] = (
                int(previous) + int(bool(row["is_main_event"]))
            )

# Calculate recent event main-event histories
new = new.sort_values(
    ["event_date", "event_name", "fight_url"]
).copy()

new["is_main_event"] = False

# Identify the main event using original UFCStats source order
main_event_urls = (
    fresh.groupby("event_name", sort=False)
    .head(1)["fight_url"]
    .astype(str)
    .str.strip()
)

new.loc[
    new["fight_url"].astype(str).str.strip().isin(main_event_urls),
    "is_main_event"
] = True

# Process events in chronological order
recent_rows = []

for (date, event), group in new.groupby(
    ["event_date", "event_name"], sort=True
):
    event_rows = []

    for _, row in group.iterrows():
        record = row.to_dict()

        for side in ["f1", "f2"]:
            fighter = fighter_key(record[f"{side}_name"])

            record[f"{side}_previous_main_events"] = (
                historical_main_event_counts.get(fighter, 0)
            )

        event_rows.append(record)

    recent_rows.extend(event_rows)

    # Update only after collecting all pre-event counts
    for record in event_rows:
        if record["is_main_event"]:
            for side in ["f1", "f2"]:
                fighter = fighter_key(record[f"{side}_name"])
                historical_main_event_counts[fighter] = (
                    historical_main_event_counts.get(fighter, 0) + 1
                )

recent_main_event_fixed = pd.DataFrame(recent_rows)

print("RECENT EVENTS:", recent_main_event_fixed["event_name"].nunique())
print("MAIN EVENTS:", recent_main_event_fixed["is_main_event"].sum())

display(
    recent_main_event_fixed.loc[
        recent_main_event_fixed["is_main_event"],
        [
            "event_name",
            "f1_name", "f2_name",
            "f1_previous_main_events",
            "f2_previous_main_events"
        ]
    ]
)
```

```python

import pandas as pd
import re

old = backup["ufc_fights"].copy()
source = datasets["ufc_fight_results.csv"].copy()

def clean_name(name):
    return " ".join(str(name).casefold().strip().split())

def pair_key(a, b):
    return tuple(sorted([clean_name(a), clean_name(b)]))

def event_key(name):
    return " ".join(str(name).casefold().strip().split())

old["event_key"] = old["event_name"].map(event_key)
source["event_key"] = source["EVENT"].map(event_key)

old["pair_key"] = [
    pair_key(a, b)
    for a, b in zip(old["f1_name"], old["f2_name"])
]

names = source["BOUT"].str.split(
    r"\s+vs\.\s+", n=1, expand=True
)

source["pair_key"] = [
    pair_key(a, b)
    for a, b in zip(names[0], names[1])
]

source_main = (
    source.groupby("event_key", sort=False)
    .head(1)[["event_key", "pair_key", "BOUT"]]
    .rename(columns={
        "pair_key": "source_main_pair",
        "BOUT": "source_main_bout"
    })
)

flagged = old[
    old["is_main_event"].fillna(False).astype(bool)
].copy()

flagged = flagged.merge(
    source_main,
    on="event_key",
    how="left",
    validate="many_to_one"
)

flagged["agrees"] = [
    a == b
    for a, b in zip(
        flagged["pair_key"],
        flagged["source_main_pair"]
    )
]

print("ORIGINAL FLAGGED MAIN EVENTS:", len(flagged))
print("MATCHING SOURCE HEADLINERS:", flagged["agrees"].sum())
print("DISAGREEMENTS:", (~flagged["agrees"]).sum())

print("\nEXAMPLE DISAGREEMENTS:")
display(
    flagged.loc[
        ~flagged["agrees"],
        [
            "event_name",
            "f1_name",
            "f2_name",
            "source_main_bout"
        ]
    ].head(20)
)
```

```python

import pandas as pd
import numpy as np

# Use our correctly URL-matched historical comparison
df = comparison.copy()

rows = []
examples = []

for side in ["f1", "f2"]:
    wins = pd.to_numeric(
        df[f"{side}_ufc_wins_before"], errors="coerce"
    )
    fights = pd.to_numeric(
        df[f"{side}_ufc_fights_before"], errors="coerce"
    )
    losses = pd.to_numeric(
        df[f"{side}_ufc_losses_before"], errors="coerce"
    )
    original_pct = pd.to_numeric(
        df[f"{side}_ufc_win_pct_before"], errors="coerce"
    )
    reconstructed_pct = pd.to_numeric(
        df[f"{side}_win_pct_before"], errors="coerce"
    )

    formulas = {
        "Wins / total UFC fights":
            wins.div(fights.replace(0, np.nan)),
        "Wins / (wins + losses)":
            wins.div((wins + losses).replace(0, np.nan)),
        "Reconstructed percentage":
            reconstructed_pct
    }

    for label, calculated in formulas.items():
        both = original_pct.notna() & calculated.notna()
        matches = both & np.isclose(
            original_pct.fillna(-999),
            calculated.fillna(-999),
            atol=0.0001
        )

        rows.append({
            "fighter": side,
            "formula": label,
            "matching_values": int(matches.sum()),
            "compared_values": int(both.sum()),
            "match_percentage": round(
                100 * matches.sum() / both.sum(), 2
            ) if both.sum() else 0
        })

    difference = (original_pct - reconstructed_pct).abs()

    sample = df.loc[
        difference > 0.05,
        [
            "event_name_original",
            f"{side}_name_original",
            f"{side}_ufc_wins_before",
            f"{side}_ufc_losses_before",
            f"{side}_ufc_fights_before",
            f"{side}_ufc_win_pct_before",
            f"{side}_win_pct_before"
        ]
    ].head(8).copy()

    sample["fighter_side"] = side
    examples.append(sample)

print("WIN PERCENTAGE FORMULA COMPARISON")
display(pd.DataFrame(rows))

print("\nLARGE DISCREPANCIES")
display(pd.concat(examples, ignore_index=True))

print("\nORIGINAL WIN PERCENTAGE MISSING VALUES")
for side in ["f1", "f2"]:
    col = f"{side}_ufc_win_pct_before"
    print(side, "missing:", df[col].isna().sum())
```

```python

import pandas as pd
import numpy as np
from collections import defaultdict

# Preserve our existing progress
updated_stats = final_fight_stats.copy()
updated_stats["event_date"] = pd.to_datetime(updated_stats["event_date"])

# Reconstruct actual losses from recorded fight outcomes.
# Process all fights from the same event before updating histories.
loss_history = defaultdict(int)
new_losses = {}

def fighter_id(name):
    return " ".join(str(name).casefold().split())

for (date, event), group in updated_stats.sort_values(
    ["event_date", "event_name"]
).groupby(["event_date", "event_name"], sort=True):

    # Record losses entering each fight
    for idx, fight in group.iterrows():
        for side in ["f1", "f2"]:
            name = fighter_id(fight[f"{side}_name"])
            new_losses[(idx, side)] = loss_history[name]

    # Update losses after the event
    for idx, fight in group.iterrows():
        for side in ["f1", "f2"]:
            result = str(fight[f"{side}_result"]).strip().upper()

            if result in ["L", "LOSS"]:
                name = fighter_id(fight[f"{side}_name"])
                loss_history[name] += 1

# Apply the original win percentage convention to recent fights
recent_mask = updated_stats["event_date"] >= pd.Timestamp("2026-07-01")

for side in ["f1", "f2"]:
    losses_col = f"{side}_losses_before"
    pct_col = f"{side}_ufc_win_pct_before"
    wins_col = f"{side}_wins_before"

    updated_stats[losses_col] = [
        new_losses[(idx, side)]
        for idx in updated_stats.index
    ]

    wins = pd.to_numeric(updated_stats[wins_col], errors="coerce")
    losses = pd.to_numeric(updated_stats[losses_col], errors="coerce")

    denominator = wins + losses
    calculated = wins / denominator.replace(0, np.nan)

    # A fighter making their UFC debut has no prior UFC win percentage.
    # Keep that undefined instead of inventing a value.
    updated_stats.loc[recent_mask, pct_col] = calculated.loc[recent_mask]

print("TOTAL FIGHTS:", len(updated_stats))
print("TOTAL EVENTS:", updated_stats["event_name"].nunique())
print("MAIN EVENTS:", int(updated_stats["is_main_event"].sum()))

print("\nRECENT FIGHTS WITH UNDEFINED WIN PERCENTAGES:")
print(
    updated_stats.loc[
        recent_mask,
        ["f1_ufc_win_pct_before", "f2_ufc_win_pct_before"]
    ].isna().sum()
)

print("\nRECENT MAIN-EVENT WIN PERCENTAGES:")
display(
    updated_stats.loc[
        recent_mask & updated_stats["is_main_event"].eq(True),
        [
            "event_name", "f1_name", "f2_name",
            "f1_ufc_win_pct_before",
            "f2_ufc_win_pct_before"
        ]
    ].round(3)
)
```

```python

import pandas as pd
import numpy as np
from collections import defaultdict

working = scoring_fights.copy()
working["event_date"] = pd.to_datetime(working["event_date"])

historical = backup["ufc_fights"].copy()
historical["date"] = pd.to_datetime(historical["date"])

def fighter_key(name):
    return " ".join(str(name).casefold().split())

# Establish each fighter's latest verified historical bonus total
bonus_baseline = {}

for _, fight in historical.sort_values(
    ["date", "event_name"]
).iterrows():
    for side in ["f1", "f2"]:
        name = fighter_key(fight[f"{side}_name"])
        value = fight[f"{side}_previous_total_bonuses"]

        if pd.notna(value):
            bonus_baseline[name] = int(value)

# Recover title-fight flags from the original historical data
title_flags = historical[
    ["fight_url", "is_title_fight"]
].copy()

# Recent title fights must be identified from reliable fight data.
# Do not assume every five-round main event is a title fight.
working["title_fight_verified"] = working["is_title_fight"]

# Rebuild historical title-fight counts from original data
title_counts = defaultdict(int)
title_records = {}

for (date, event), group in working.sort_values(
    ["event_date", "event_name"]
).groupby(["event_date", "event_name"], sort=True):

    for idx, fight in group.iterrows():
        for side in ["f1", "f2"]:
            fighter = fighter_key(fight[f"{side}_name"])
            title_records[(idx, side)] = title_counts[fighter]

    for _, fight in group.iterrows():
        if fight["title_fight_verified"] == True:
            for side in ["f1", "f2"]:
                fighter = fighter_key(fight[f"{side}_name"])
                title_counts[fighter] += 1

recent_mask = working["event_date"] >= "2026-07-01"

for side in ["f1", "f2"]:
    col = f"{side}_previous_title_fights"

    reconstructed = pd.Series(
        [
            title_records[(idx, side)]
            for idx in working.index
        ],
        index=working.index
    )

    # Preserve verified original historical counts
    working.loc[recent_mask, col] = reconstructed.loc[recent_mask]

    bonus_col = f"{side}_previous_total_bonuses"

    # Carry forward known historical totals as a baseline.
    # These are not yet complete totals for recent events.
    working.loc[
        recent_mask, f"{side}_bonus_baseline"
    ] = working.loc[recent_mask, f"{side}_name"].map(
        lambda name: bonus_baseline.get(fighter_key(name), 0)
    )

print("RECENT FIGHTS:", recent_mask.sum())

print("\nRECENT MAIN-EVENT TITLE HISTORY AND BONUS BASELINES:")
display(
    working.loc[
        recent_mask & working["is_main_event"].eq(True),
        [
            "event_name",
            "f1_name", "f2_name",
            "f1_previous_title_fights",
            "f2_previous_title_fights",
            "f1_bonus_baseline",
            "f2_bonus_baseline"
        ]
    ]
)

print("\nRECENT TITLE-FIGHT FLAGS STILL UNVERIFIED:")
print(working.loc[recent_mask, "title_fight_verified"].isna().sum())

# Save the working dataset in Colab memory
scoring_fights = working
```

```python

import pandas as pd
import requests
import time
from bs4 import BeautifulSoup

session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0 (compatible; EducationalResearch/1.0)"
})

bonus_sections = []

for _, event in bonus_pages.dropna(subset=["page_id"]).iterrows():
    title = event["best_page"]
    url = "https://en.wikipedia.org/wiki/" + requests.utils.quote(
        title.replace(" ", "_"), safe=":_"
    )

    record = {
        "event_name": event["event_name"],
        "page_title": title,
        "url": url,
        "bonus_text": None,
        "status": "Not found"
    }

    try:
        response = session.get(url, timeout=20)
        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")

        # Locate the Bonuses section
        heading = soup.find(
            lambda tag:
                tag.name in ["h2", "h3", "h4"]
                and "bonuses" in tag.get_text(" ", strip=True).lower()
        )

        if heading:
            content = []
            for element in heading.find_next_siblings():
                if element.name in ["h2", "h3", "h4"]:
                    break
                content.append(element.get_text(" ", strip=True))

            record["bonus_text"] = " ".join(content)[:2000]
            record["status"] = (
                "Section found" if record["bonus_text"]
                else "Empty section"
            )

    except Exception as error:
        record["status"] = f"Error: {str(error)[:80]}"

    bonus_sections.append(record)
    time.sleep(1.5)

bonus_sections_df = pd.DataFrame(bonus_sections)

print("PAGES CHECKED:", len(bonus_sections_df))
print("BONUS SECTIONS FOUND:",
      (bonus_sections_df["status"] == "Section found").sum())

display(
    bonus_sections_df[
        ["event_name", "status", "bonus_text"]
    ]
)
```

```python

import pandas as pd
import numpy as np

me = main_event_scoring.copy()

# Map corrected raw statistics to the original score columns
factors = {
    "corrected_main_event_ranking_points":
        ("main_event_ranking_points_score", 0.25),

    "me_combined_title_fights":
        ("main_event_combined_title_fights_score", 0.20),

    "me_avg_win_pct":
        ("main_event_avg_win_pct_score", 0.15),

    "me_combined_win_streak":
        ("main_event_combined_win_streak_score", 0.15),

    "me_combined_previous_main_events":
        ("main_event_combined_previous_main_events_score", 0.10),

    "me_combined_ufc_fights":
        ("main_event_combined_ufc_fights_score", 0.10),

    "me_combined_bonuses":
        ("main_event_combined_bonuses_score", 0.05)
}

# Percentile normalization, 0–100, across historical events.
# Historical reference prevents unverified recent data from
# changing the scoring scale.
historical = me["event_date"] < pd.Timestamp("2026-07-01")

for raw_col, (score_col, weight) in factors.items():
    reference = pd.to_numeric(
        me.loc[historical, raw_col],
        errors="coerce"
    ).dropna()

    values = pd.to_numeric(me[raw_col], errors="coerce")

    # Score based on historical percentile position.
    sorted_ref = np.sort(reference.to_numpy())

    if len(sorted_ref) == 0:
        me[score_col] = np.nan
        continue

    positions = np.searchsorted(
        sorted_ref,
        values.to_numpy(dtype=float),
        side="right"
    )

    me[score_col] = (
        positions / len(sorted_ref) * 100
    )

    # Preserve missing input values as missing scores
    me.loc[values.isna(), score_col] = np.nan

# Require all seven factor scores for a complete score
score_cols = [
    score_col for score_col, weight in factors.values()
]

weighted = sum(
    me[score_col] * weight
    for score_col, weight in factors.values()
)

me["corrected_main_event_score"] = weighted

# Track score completeness
me["main_event_score_complete"] = (
    me[score_cols].notna().all(axis=1)
    & ~me["bonus_data_provisional"]
)

print("TOTAL MAIN EVENTS:", len(me))
print("COMPLETE MAIN-EVENT SCORES:",
      me["main_event_score_complete"].sum())

print("RECENT EVENTS AWAITING RANKINGS:",
      me["rankings_provisional"].sum())

print("\nTOP 20 HISTORICAL MAIN EVENTS:")
display(
    me.loc[
        historical,
        [
            "event_name",
            "corrected_main_event_score"
        ]
    ].sort_values(
        "corrected_main_event_score",
        ascending=False
    ).head(20)
)

print("\nRECENT EVENT SCORING STATUS:")
display(
    me.loc[
        ~historical,
        [
            "event_name",
            "corrected_main_event_score",
            "main_event_score_complete"
        ]
    ]
)

main_event_scoring = me
```

```python

import pandas as pd
import numpy as np

raw = backup["event_master"].copy()
norm = backup["normalized"].copy()

factors = {
    "main_event_combined_ufc_fights":
        "main_event_combined_ufc_fights_score",
    "main_event_avg_win_pct":
        "main_event_avg_win_pct_score",
    "main_event_combined_win_streak":
        "main_event_combined_win_streak_score",
    "main_event_combined_title_fights":
        "main_event_combined_title_fights_score",
    "main_event_combined_previous_main_events":
        "main_event_combined_previous_main_events_score",
    "main_event_combined_bonuses":
        "main_event_combined_bonuses_score",
    "main_event_ranking_points":
        "main_event_ranking_points_score"
}

test = raw.merge(
    norm[["event_name"] + list(factors.values())],
    on="event_name",
    how="left",
    validate="one_to_one"
)

results = []

for raw_col, score_col in factors.items():
    x = pd.to_numeric(test[raw_col], errors="coerce")
    y = pd.to_numeric(test[score_col], errors="coerce")

    methods = {
        "percentile_rank": x.rank(pct=True) * 100,
        "min_max": (
            (x - x.min()) / (x.max() - x.min()) * 100
            if x.max() != x.min() else x * 0
        ),
        "max_value": (
            x / x.max() * 100
            if x.max() != 0 else x * 0
        )
    }

    for method, predicted in methods.items():
        valid = y.notna() & predicted.notna()

        results.append({
            "factor": raw_col,
            "method": method,
            "mean_absolute_error": (
                (y[valid] - predicted[valid]).abs().mean()
                if valid.any() else np.nan
            ),
            "max_absolute_error": (
                (y[valid] - predicted[valid]).abs().max()
                if valid.any() else np.nan
            )
        })

results_df = pd.DataFrame(results)

print("BEST NORMALIZATION METHOD FOR EACH FACTOR:")
display(
    results_df.sort_values("mean_absolute_error")
    .groupby("factor", as_index=False)
    .first()
    .round(4)
)

print("\nFULL NORMALIZATION COMPARISON:")
display(
    results_df.pivot(
        index="factor",
        columns="method",
        values="mean_absolute_error"
    ).round(3)
)
```

```python

import pandas as pd
import numpy as np

raw = backup["event_master"].copy()
norm = backup["normalized"].copy()

factors = {
    "main_event_combined_ufc_fights":
        "main_event_combined_ufc_fights_score",
    "main_event_avg_win_pct":
        "main_event_avg_win_pct_score",
    "main_event_combined_win_streak":
        "main_event_combined_win_streak_score",
    "main_event_combined_title_fights":
        "main_event_combined_title_fights_score",
    "main_event_combined_previous_main_events":
        "main_event_combined_previous_main_events_score",
    "main_event_combined_bonuses":
        "main_event_combined_bonuses_score",
    "main_event_ranking_points":
        "main_event_ranking_points_score"
}

df = raw.merge(
    norm[["event_name"] + list(factors.values())],
    on="event_name",
    validate="one_to_one"
)

results = []

for raw_col, score_col in factors.items():
    x = pd.to_numeric(df[raw_col], errors="coerce")
    y = pd.to_numeric(df[score_col], errors="coerce")

    candidates = {}

    for method in ["average", "min", "max", "dense", "first"]:
        candidates[f"{method}_original"] = (
            x.rank(method=method, pct=True) * 100
        )
        candidates[f"{method}_fill_zero"] = (
            x.fillna(0).rank(method=method, pct=True) * 100
        )

    # Also test empirical cumulative distribution
    candidates["ecdf_right"] = (
        x.map(x.value_counts(normalize=True).sort_index().cumsum())
        * 100
    )

    for name, predicted in candidates.items():
        valid = y.notna() & predicted.notna()

        if valid.any():
            errors = (y[valid] - predicted[valid]).abs()

            results.append({
                "factor": raw_col,
                "method": name,
                "mae": errors.mean(),
                "max_error": errors.max(),
                "exact_matches": (errors < 0.000001).sum(),
                "compared": valid.sum()
            })

result = pd.DataFrame(results)

best = (
    result.sort_values(
        ["factor", "mae", "max_error"]
    )
    .groupby("factor", as_index=False)
    .first()
)

print("BEST MATCHING NORMALIZATION METHODS:")
display(best.round(6))

print("\nMETHODS WITH THE MOST EXACT MATCHES:")
display(
    result.sort_values(
        ["exact_matches", "mae"],
        ascending=[False, True]
    ).head(20).round(6)
)
```

```python

import pandas as pd
import numpy as np

me = main_event_scoring.copy()

raw = backup["event_master"].copy()
norm = backup["normalized"].copy()

factors = {
    "me_combined_ufc_fights": (
        "main_event_combined_ufc_fights",
        "main_event_combined_ufc_fights_score", 0.10
    ),
    "me_avg_win_pct": (
        "main_event_avg_win_pct",
        "main_event_avg_win_pct_score", 0.15
    ),
    "me_combined_win_streak": (
        "main_event_combined_win_streak",
        "main_event_combined_win_streak_score", 0.15
    ),
    "me_combined_title_fights": (
        "main_event_combined_title_fights",
        "main_event_combined_title_fights_score", 0.20
    ),
    "me_combined_previous_main_events": (
        "main_event_combined_previous_main_events",
        "main_event_combined_previous_main_events_score", 0.10
    ),
    "me_combined_bonuses": (
        "main_event_combined_bonuses",
        "main_event_combined_bonuses_score", 0.05
    ),
    "corrected_main_event_ranking_points": (
        "main_event_ranking_points",
        "main_event_ranking_points_score", 0.25
    )
}

reference = raw.merge(
    norm[["event_name"] + [v[1] for v in factors.values()]],
    on="event_name",
    validate="one_to_one"
)

report = []

for new_raw, (old_raw, score_col, weight) in factors.items():
    lookup = (
        reference[[old_raw, score_col]]
        .apply(pd.to_numeric, errors="coerce")
        .dropna()
        .groupby(old_raw)[score_col]
        .median()
        .sort_index()
    )

    x = pd.to_numeric(me[new_raw], errors="coerce")

    # Exact lookup when possible; interpolation otherwise
    exact = x.map(lookup)

    estimated = pd.Series(
        np.interp(
            x.fillna(lookup.index.min()),
            lookup.index.to_numpy(dtype=float),
            lookup.to_numpy(dtype=float)
        ),
        index=me.index
    )

    me[score_col] = exact.fillna(estimated).clip(0, 100)
    me.loc[x.isna(), score_col] = np.nan

    report.append({
        "factor": new_raw,
        "exact_matches": int((x.notna() & exact.notna()).sum()),
        "interpolated": int((x.notna() & exact.isna()).sum()),
        "missing": int(x.isna().sum())
    })

me["corrected_main_event_score"] = sum(
    me[score_col] * weight
    for _, score_col, weight in factors.values()
)

me["main_event_score_complete"] = (
    me["corrected_main_event_score"].notna()
    & ~me["bonus_data_provisional"]
)

main_event_scoring = me

print("ORIGINAL-SCALE NORMALIZATION REPORT:")
display(pd.DataFrame(report))

print("\nCOMPLETE HISTORICAL SCORES:",
      me["main_event_score_complete"].sum())

print("\nTOP 20 CORRECTED MAIN EVENTS:")
display(
    me.loc[me["main_event_score_complete"],
           ["event_name", "corrected_main_event_score"]]
    .sort_values("corrected_main_event_score", ascending=False)
    .head(20)
    .round(2)
)
```

```python

import pandas as pd
import numpy as np

norm = backup["normalized"].copy()

categories = {
    "card_quality_score": [
        "title_fights_score",
        "total_ranking_points_score",
        "avg_ufc_experience_score",
        "avg_ufc_win_pct_score",
        "avg_win_streak_score",
        "avg_finish_rate_score"
    ],
    "star_experience_score": [
        "avg_previous_title_fights_score",
        "avg_previous_main_events_score",
        "avg_previous_bonuses_score"
    ]
}

from scipy.optimize import minimize

recovered_weights = {}

for category, factors in categories.items():

    data = norm[factors + [category]].copy()
    data = data.apply(pd.to_numeric, errors="coerce")
    data = data.dropna()

    X = data[factors].to_numpy(dtype=float)
    y = data[category].to_numpy(dtype=float)

    n = len(factors)

    # Nonnegative weights that sum to 100%
    result = minimize(
        lambda w: np.mean((X @ w - y) ** 2),
        x0=np.ones(n) / n,
        bounds=[(0, 1)] * n,
        constraints=[{
            "type": "eq",
            "fun": lambda w: np.sum(w) - 1
        }],
        method="SLSQP",
        options={"maxiter": 10000, "ftol": 1e-12}
    )

    weights = result.x
    predictions = X @ weights
    errors = np.abs(predictions - y)

    recovered_weights[category] = dict(zip(factors, weights))

    print("\n", category.upper())
    print("Optimization successful:", result.success)
    print("Events compared:", len(data))
    print("Mean absolute error:", round(errors.mean(), 8))
    print("Maximum absolute error:", round(errors.max(), 8))

    display(
        pd.DataFrame({
            "factor": factors,
            "estimated_weight": weights,
            "percentage": weights * 100
        }).round(4)
    )

print("\nRECOVERED CATEGORY WEIGHTS:")
print(recovered_weights)
```

```python

import pandas as pd
import numpy as np

fights = scoring_fights.copy()
master = backup["event_master"].copy()

historical_names = set(master["event_name"])
historical = fights[
    fights["event_name"].isin(historical_names)
].copy()

# Inspect original raw columns
print("ORIGINAL EVENT MASTER COLUMNS:")
print(master.columns.tolist())

# Compare candidate fighter-level aggregations
candidate_fields = {
    "avg_ufc_experience": (
        "f1_fights_before", "f2_fights_before"
    ),
    "avg_ufc_win_pct": (
        "f1_ufc_win_pct_before", "f2_ufc_win_pct_before"
    ),
    "avg_win_streak": (
        "f1_win_streak_before", "f2_win_streak_before"
    ),
    "avg_previous_title_fights": (
        "f1_previous_title_fights", "f2_previous_title_fights"
    ),
    "avg_previous_main_events": (
        "f1_previous_main_events", "f2_previous_main_events"
    )
}

rows = []

for event, group in historical.groupby("event_name"):

    row = {"event_name": event}

    for factor, (a, b) in candidate_fields.items():
        values = pd.concat([
            pd.to_numeric(group[a], errors="coerce"),
            pd.to_numeric(group[b], errors="coerce")
        ])

        row[factor + "_reconstructed"] = values.mean()

    rows.append(row)

reconstructed = pd.DataFrame(rows)

comparison = reconstructed.merge(
    master,
    on="event_name",
    validate="one_to_one"
)

print("\nHISTORICAL RECONSTRUCTION ERRORS:")

results = []

for factor in candidate_fields:
    predicted_col = factor + "_reconstructed"

    if factor not in comparison.columns:
        results.append({
            "factor": factor,
            "status": "Original column not found"
        })
        continue

    original = pd.to_numeric(
        comparison[factor], errors="coerce"
    )
    predicted = pd.to_numeric(
        comparison[predicted_col], errors="coerce"
    )

    valid = original.notna() & predicted.notna()

    results.append({
        "factor": factor,
        "events_compared": int(valid.sum()),
        "mean_absolute_error": (
            (original[valid] - predicted[valid]).abs().mean()
            if valid.any() else np.nan
        ),
        "max_absolute_error": (
            (original[valid] - predicted[valid]).abs().max()
            if valid.any() else np.nan
        )
    })

display(pd.DataFrame(results).round(6))
```

```python

import pandas as pd
import numpy as np

fights = scoring_fights.copy()
master = backup["event_master"].copy()

historical = fights[
    fights["event_name"].isin(master["event_name"])
].copy()

factors = {
    "avg_ufc_experience": [
        ("f1_fights_before", "f2_fights_before"),
        ("f1_ufc_fights_before", "f2_ufc_fights_before")
    ],
    "avg_win_streak": [
        ("f1_win_streak_before", "f2_win_streak_before"),
        ("f1_win_streak_before_original",
         "f2_win_streak_before_original")
    ],
    "avg_previous_main_events": [
        ("f1_previous_main_events", "f2_previous_main_events")
    ]
}

results = []
mismatches = []

for factor, candidates in factors.items():

    for col1, col2 in candidates:

        if col1 not in historical.columns or col2 not in historical.columns:
            continue

        temp = historical[
            ["event_name", col1, col2]
        ].copy()

        temp[col1] = pd.to_numeric(temp[col1], errors="coerce")
        temp[col2] = pd.to_numeric(temp[col2], errors="coerce")

        long = pd.concat([
            temp[["event_name", col1]].rename(columns={col1: "value"}),
            temp[["event_name", col2]].rename(columns={col2: "value"})
        ])

        calculated = long.groupby("event_name")["value"].mean()

        original = master.set_index("event_name")[factor]

        comparison = pd.DataFrame({
            "original": original,
            "reconstructed": calculated
        }).dropna()

        comparison["difference"] = (
            comparison["reconstructed"] - comparison["original"]
        )

        abs_diff = comparison["difference"].abs()

        results.append({
            "factor": factor,
            "source_columns": f"{col1} / {col2}",
            "events_compared": len(comparison),
            "exact_matches": int((abs_diff < 1e-8).sum()),
            "mean_absolute_error": abs_diff.mean(),
            "max_absolute_error": abs_diff.max()
        })

        biggest = comparison.loc[
            abs_diff.nlargest(5).index
        ].reset_index()

        biggest["factor"] = factor
        biggest["source_columns"] = col1

        mismatches.append(biggest)

print("CANDIDATE FORMULA COMPARISON:")
display(pd.DataFrame(results).round(6))

print("\nLARGEST HISTORICAL DIFFERENCES:")
display(
    pd.concat(mismatches, ignore_index=True)
    .sort_values(
        "difference",
        key=lambda s: s.abs(),
        ascending=False
    )
    .head(20)
    .round(4)
)
```

```python

import pandas as pd
import numpy as np
import requests
import time
from bs4 import BeautifulSoup

historical = backup["ufc_fights"].copy()
all_fights = scoring_fights.copy()

# 1. Verify the historical finish-rate formula
checks = []

for side in ["f1", "f2"]:
    finishes = pd.to_numeric(
        historical[f"{side}_ufc_finishes_before"],
        errors="coerce"
    )
    fights_before = pd.to_numeric(
        historical[f"{side}_ufc_fights_before"],
        errors="coerce"
    )
    original_rate = pd.to_numeric(
        historical[f"{side}_finish_rate_before"],
        errors="coerce"
    )

    reconstructed = (
        finishes.div(fights_before.replace(0, np.nan))
        .fillna(0)
    )

    checks.append({
        "fighter_side": side,
        "exact_matches": int(
            np.isclose(
                reconstructed,
                original_rate,
                atol=1e-8,
                equal_nan=True
            ).sum()
        ),
        "total": len(historical),
        "max_error": (
            reconstructed - original_rate
        ).abs().max()
    })

print("HISTORICAL FINISH-RATE VALIDATION:")
display(pd.DataFrame(checks))

# 2. Identify the recent fights
historical_events = set(backup["event_master"]["event_name"])

recent = all_fights[
    ~all_fights["event_name"].isin(historical_events)
].copy()

recent["event_date"] = pd.to_datetime(recent["event_date"])

# 3. Retrieve recent fight methods
session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0"
})

def get_fight_method(url):
    try:
        response = session.get(str(url), timeout=20)
        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")

        labels = soup.select("i.b-fight-details__text-item")

        for item in labels:
            text = item.get_text(" ", strip=True)

            if text.lower().startswith("method:"):
                return text.split(":", 1)[1].strip()

        return None

    except Exception:
        return None

print("\nRETRIEVING RECENT FIGHT METHODS...")

recent["method"] = [
    get_fight_method(url)
    for url in recent["fight_url"]
]

print(
    "Methods retrieved:",
    recent["method"].notna().sum(),
    "of",
    len(recent)
)

print("\nMETHOD COUNTS:")
print(recent["method"].value_counts(dropna=False))

# Save retrieved methods so we don't download again
recent[[
    "event_name", "fight_url", "method"
]].to_csv(
    "/content/ufc_recent_methods_2026.csv",
    index=False
)

print("\nSaved: /content/ufc_recent_methods_2026.csv")

# Do not build finish histories from incomplete methods
if recent["method"].isna().any():
    print("\nSTOP: Some fight methods are missing.")
    print("We must resolve those before calculating finish histories.")
else:
    print("\nAll methods retrieved. Ready to extend fighter histories.")
```

```python

import pandas as pd
import numpy as np
import requests
import re
import time
from bs4 import BeautifulSoup

historical = backup["ufc_fights"].copy()
recent = scoring_fights[
    ~scoring_fights["event_name"].isin(
        backup["event_master"]["event_name"]
    )
].copy()

# -------------------------------------------------
# PART 1: RECOVER ORIGINAL FINISH-RATE FORMULA
# -------------------------------------------------

results = []

for side in ["f1", "f2"]:

    finishes = pd.to_numeric(
        historical[f"{side}_ufc_finishes_before"],
        errors="coerce"
    )

    fights = pd.to_numeric(
        historical[f"{side}_ufc_fights_before"],
        errors="coerce"
    )

    wins = pd.to_numeric(
        historical[f"{side}_ufc_wins_before"],
        errors="coerce"
    )

    actual = pd.to_numeric(
        historical[f"{side}_finish_rate_before"],
        errors="coerce"
    )

    candidates = {
        "finishes / UFC fights":
            finishes.div(fights.replace(0, np.nan)).fillna(0),

        "finishes / UFC wins":
            finishes.div(wins.replace(0, np.nan)).fillna(0),

        "finishes / (UFC fights + 1)":
            finishes.div(fights + 1),

        "finishes / (UFC wins + 1)":
            finishes.div(wins + 1)
    }

    for name, predicted in candidates.items():

        valid = actual.notna() & predicted.notna()

        error = (
            predicted[valid] - actual[valid]
        ).abs()

        results.append({
            "side": side,
            "formula": name,
            "exact_matches": int((error < 1e-8).sum()),
            "compared": int(valid.sum()),
            "mean_absolute_error": error.mean(),
            "max_error": error.max()
        })

print("FINISH-RATE FORMULA COMPARISON:")
display(
    pd.DataFrame(results)
    .sort_values(["side", "mean_absolute_error"])
    .round(6)
)

# -------------------------------------------------
# PART 2: CLEAN UFCSTATS FIGHT URLS
# -------------------------------------------------

def clean_ufcstats_url(value):

    text = str(value).strip()

    # Handle Markdown URLs:
    # [http://...](http://...)
    match = re.search(
        r"https?://(?:www\.)?ufcstats\.com/fight-details/[a-fA-F0-9]+",
        text
    )

    return match.group(0) if match else None

recent["clean_fight_url"] = (
    recent["fight_url"].apply(clean_ufcstats_url)
)

print("\nVALID RECENT URLS:")
print(
    recent["clean_fight_url"].notna().sum(),
    "of",
    len(recent)
)

print("\nSAMPLE CLEANED URL:")
print(recent["clean_fight_url"].iloc[0])

# -------------------------------------------------
# PART 3: TEST UFCSTATS CONNECTION
# -------------------------------------------------

session = requests.Session()
session.headers.update({
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 Chrome/125.0 Safari/537.36"
    )
})

test_url = recent["clean_fight_url"].dropna().iloc[0]

try:
    response = session.get(test_url, timeout=20)

    print("\nUFCSTATS CONNECTION TEST:")
    print("HTTP status:", response.status_code)
    print("Page length:", len(response.text))
    print("Final URL:", response.url)

    soup = BeautifulSoup(response.text, "html.parser")

    print("\nPAGE TITLE:")
    print(
        soup.title.get_text(" ", strip=True)
        if soup.title else "No title"
    )

    print("\nMETHOD FIELD CANDIDATES:")
    candidates = soup.select(
        "i.b-fight-details__text-item"
    )

    for item in candidates[:12]:
        print(repr(item.get_text(" ", strip=True)))

    connection_ok = (
        response.status_code == 200
        and "ufcstats" in response.url.lower()
        and len(candidates) > 0
    )

except Exception as exc:
    print("\nCONNECTION ERROR:", repr(exc))
    connection_ok = False

# -------------------------------------------------
# PART 4: DOWNLOAD ONLY IF CONNECTION WORKS
# -------------------------------------------------

def extract_method(html):

    soup = BeautifulSoup(html, "html.parser")

    for item in soup.select("i.b-fight-details__text-item"):

        text = item.get_text(" ", strip=True)

        match = re.search(
            r"\bMETHOD\s*:\s*(.+)",
            text,
            flags=re.IGNORECASE
        )

        if match:
            return match.group(1).strip()

    return None

if connection_ok:

    print("\nDOWNLOADING RECENT FIGHT METHODS...")

    methods = {}

    for i, url in enumerate(
        recent["clean_fight_url"].dropna().unique(), 1
    ):

        try:
            r = session.get(url, timeout=20)

            methods[url] = (
                extract_method(r.text)
                if r.status_code == 200 else None
            )

        except requests.RequestException:
            methods[url] = None

        if i % 25 == 0:
            print(f"Processed {i} fights")

        time.sleep(0.15)

    recent["method"] = (
        recent["clean_fight_url"].map(methods)
    )

    print("\nMETHOD RESULTS:")
    print(
        recent["method"].value_counts(dropna=False)
    )

    recent[
        ["event_name", "clean_fight_url", "method"]
    ].to_csv(
        "/content/ufc_recent_methods_2026_v2.csv",
        index=False
    )

    print("\nSaved results to CSV.")

else:
    print(
        "\nFull scrape skipped because the "
        "connection or page structure failed."
    )
```

```python

import pandas as pd
import numpy as np

historical = backup["ufc_fights"].copy()
recent = scoring_fights[
    ~scoring_fights["event_name"].isin(
        backup["event_master"]["event_name"]
    )
].copy()

historical["event_date"] = pd.to_datetime(historical["event_date"])
recent["event_date"] = pd.to_datetime(recent["event_date"])

# Build fighter histories from the original historical fight records.
# Historical rows are kept in their existing event/fight order.
# Recent rows are processed by event date, with same-day ordering
# preserved from the current dataset.

combined = pd.concat(
    [
        historical.assign(_period="historical"),
        recent.assign(_period="recent")
    ],
    ignore_index=True,
    sort=False
)

combined["_row_order"] = np.arange(len(combined))
combined = combined.sort_values(
    ["event_date", "_row_order"],
    kind="stable"
).reset_index(drop=True)

# Build a ledger using actual completed fight results.
# UFCStats method is available for historical fights only.

ledger = {}

recent_rows = []

def number(value, default=0):
    x = pd.to_numeric(value, errors="coerce")
    return default if pd.isna(x) else float(x)

for _, fight in combined.iterrows():

    is_recent = fight["_period"] == "recent"

    updates = []

    for side in ["f1", "f2"]:

        name = str(fight[f"{side}_name"]).strip()

        if name not in ledger:
            ledger[name] = {
                "fights": 0,
                "wins": 0,
                "losses": 0,
                "streak": 0,
                "finishes": 0,
                "finish_history_complete": True
            }

        state = ledger[name]

        if is_recent:
            recent_rows.append({
                "fight_url": fight["fight_url"],
                "event_name": fight["event_name"],
                "event_date": fight["event_date"],
                "side": side,
                "fighter": name,
                "ufc_fights_before_rebuilt": state["fights"],
                "ufc_wins_before_rebuilt": state["wins"],
                "win_streak_before_rebuilt": state["streak"],
                "ufc_finishes_before_known": state["finishes"],
                "finish_history_complete": state[
                    "finish_history_complete"
                ],
                "finish_rate_before_rebuilt": (
                    state["finishes"] / state["wins"]
                    if state["wins"] > 0
                    and state["finish_history_complete"]
                    else (
                        0.0 if state["wins"] == 0 else np.nan
                    )
                ),
                "existing_fights_before": fight.get(
                    f"{side}_fights_before"
                ),
                "existing_wins_before": fight.get(
                    f"{side}_wins_before"
                ),
                "existing_streak_before": fight.get(
                    f"{side}_win_streak_before"
                )
            })

        updates.append((name, str(fight.get(f"{side}_result", ""))))

    method = str(fight.get("method", "")).upper()

    is_finish = (
        "KO" in method
        or "TKO" in method
        or "SUBMISSION" in method
    )

    for name, result in updates:

        state = ledger[name]

        if result in ("W", "L", "D", "NC"):
            state["fights"] += 1

        if result == "W":
            state["wins"] += 1
            state["streak"] += 1

            if is_recent and pd.isna(fight.get("method")):
                state["finish_history_complete"] = False

            elif is_finish:
                state["finishes"] += 1

        elif result == "L":
            state["losses"] += 1
            state["streak"] = 0

        elif result in ("D", "NC"):
            state["streak"] = 0

recent_fighter_history = pd.DataFrame(recent_rows)

print("RECENT FIGHTER APPEARANCES:")
print(len(recent_fighter_history))

print("\nREBUILT FIGHTER HISTORY SAMPLE:")
display(recent_fighter_history.head(15))

print("\nFINISH RATE COMPLETENESS:")
print(
    recent_fighter_history[
        "finish_rate_before_rebuilt"
    ].notna().sum(),
    "of",
    len(recent_fighter_history)
)

print("\nREBUILT VS EXISTING COUNTERS:")

for rebuilt, existing in [
    ("ufc_fights_before_rebuilt", "existing_fights_before"),
    ("ufc_wins_before_rebuilt", "existing_wins_before"),
    ("win_streak_before_rebuilt", "existing_streak_before")
]:
    a = pd.to_numeric(
        recent_fighter_history[rebuilt], errors="coerce"
    )
    b = pd.to_numeric(
        recent_fighter_history[existing], errors="coerce"
    )

    valid = a.notna() & b.notna()

    print(
        rebuilt,
        "| compared:", int(valid.sum()),
        "| exact:", int(np.isclose(a[valid], b[valid]).sum()),
        "| max difference:",
        (a[valid] - b[valid]).abs().max()
        if valid.any() else np.nan
    )

recent_fighter_history.to_csv(
    "/content/ufc_recent_fighter_history_rebuilt.csv",
    index=False
)

print("\nSaved reconstructed histories.")
```

```python

import pandas as pd
import numpy as np

# =====================================================
# UFC EVENT STRENGTH INDEX V2
# REBUILD RECENT FIGHTER HISTORIES — CORRECTED
# =====================================================

historical = backup["ufc_fights"].copy()
master = backup["event_master"].copy()
all_fights = scoring_fights.copy()

# 1. Recover historical event dates
date_lookup = master[
    ["event_name", "event_date"]
].drop_duplicates("event_name")

historical = historical.drop(
    columns=["event_date"], errors="ignore"
).merge(
    date_lookup,
    on="event_name",
    how="left",
    validate="many_to_one"
)

historical["event_date"] = pd.to_datetime(
    historical["event_date"], errors="coerce"
)

recent = all_fights[
    ~all_fights["event_name"].isin(master["event_name"])
].copy()

recent["event_date"] = pd.to_datetime(
    recent["event_date"], errors="coerce"
)

assert historical["event_date"].notna().all(), \
    "Some historical event dates could not be matched."

assert recent["event_date"].notna().all(), \
    "Some recent event dates are missing."

# 2. Preserve original fight ordering
historical["_period"] = "historical"
recent["_period"] = "recent"

historical["_source_order"] = np.arange(len(historical))
recent["_source_order"] = np.arange(len(recent))

combined = pd.concat(
    [historical, recent],
    ignore_index=True,
    sort=False
).sort_values(
    ["event_date", "_source_order"],
    kind="stable"
).reset_index(drop=True)

# 3. Rebuild chronological fighter histories
ledger = {}
recent_rows = []

def get_state(name):
    if name not in ledger:
        ledger[name] = {
            "fights": 0,
            "wins": 0,
            "losses": 0,
            "streak": 0,
            "finishes": 0,
            "finish_history_complete": True
        }
    return ledger[name]

for _, fight in combined.iterrows():

    is_recent = fight["_period"] == "recent"
    updates = []

    for side in ["f1", "f2"]:

        name = str(fight[f"{side}_name"]).strip()
        state = get_state(name)

        if is_recent:
            recent_rows.append({
                "fight_url": fight["fight_url"],
                "event_name": fight["event_name"],
                "event_date": fight["event_date"],
                "side": side,
                "fighter": name,
                "ufc_fights_before_rebuilt": state["fights"],
                "ufc_wins_before_rebuilt": state["wins"],
                "win_streak_before_rebuilt": state["streak"],
                "ufc_finishes_before_known": state["finishes"],
                "finish_history_complete": state[
                    "finish_history_complete"
                ],
                "finish_rate_before_rebuilt": (
                    state["finishes"] / state["wins"]
                    if state["wins"] > 0
                    and state["finish_history_complete"]
                    else (
                        0.0 if state["wins"] == 0 else np.nan
                    )
                ),
                "existing_fights_before": fight.get(
                    f"{side}_fights_before"
                ),
                "existing_wins_before": fight.get(
                    f"{side}_wins_before"
                ),
                "existing_streak_before": fight.get(
                    f"{side}_win_streak_before"
                )
            })

        updates.append((
            name,
            str(fight.get(f"{side}_result", ""))
        ))

    method = str(fight.get("method", "")).upper()

    is_finish = (
        "KO" in method
        or "TKO" in method
        or "SUBMISSION" in method
    )

    # 4. Update records after capturing pre-fight values
    for name, result in updates:

        state = get_state(name)

        if result in ("W", "L", "D", "NC"):
            state["fights"] += 1

        if result == "W":
            state["wins"] += 1
            state["streak"] += 1

            if is_recent and pd.isna(fight.get("method")):
                state["finish_history_complete"] = False
            elif is_finish:
                state["finishes"] += 1

        elif result == "L":
            state["losses"] += 1
            state["streak"] = 0

        elif result in ("D", "NC"):
            state["streak"] = 0

# 5. Save reconstructed fighter histories
recent_fighter_history = pd.DataFrame(recent_rows)

print("RECENT FIGHTER APPEARANCES:")
print(len(recent_fighter_history))

print("\nREBUILT VS EXISTING COUNTERS:")

for rebuilt, existing in [
    ("ufc_fights_before_rebuilt", "existing_fights_before"),
    ("ufc_wins_before_rebuilt", "existing_wins_before"),
    ("win_streak_before_rebuilt", "existing_streak_before")
]:
    a = pd.to_numeric(
        recent_fighter_history[rebuilt], errors="coerce"
    )
    b = pd.to_numeric(
        recent_fighter_history[existing], errors="coerce"
    )

    valid = a.notna() & b.notna()
    differences = (a[valid] - b[valid]).abs()

    print(
        rebuilt,
        "| compared:", int(valid.sum()),
        "| exact:", int((differences < 1e-8).sum()),
        "| max difference:",
        differences.max() if valid.any() else np.nan
    )

print("\nFINISH RATE COMPLETENESS:")
print(
    recent_fighter_history[
        "finish_rate_before_rebuilt"
    ].notna().sum(),
    "of",
    len(recent_fighter_history)
)

print("\nSAMPLE:")
display(recent_fighter_history.head(10))

recent_fighter_history.to_csv(
    "/content/ufc_recent_fighter_history_rebuilt.csv",
    index=False
)

print("\nReconstructed histories saved successfully.")
```

```python

import pandas as pd
import numpy as np

history = recent_fighter_history.copy()

# Preserve the existing streak when it disagrees with the reconstruction.
# This is a conservative choice pending investigation, not proof that
# the existing value is correct.

for c in [
    "ufc_fights_before_rebuilt",
    "ufc_wins_before_rebuilt",
    "win_streak_before_rebuilt",
    "existing_streak_before",
    "finish_rate_before_rebuilt"
]:
    history[c] = pd.to_numeric(history[c], errors="coerce")

history["streak_disagreement"] = (
    history["existing_streak_before"].notna()
    & ~np.isclose(
        history["win_streak_before_rebuilt"],
        history["existing_streak_before"],
        equal_nan=True
    )
)

print("WIN-STREAK DISCREPANCIES:")
display(
    history.loc[
        history["streak_disagreement"],
        [
            "event_name",
            "fighter",
            "win_streak_before_rebuilt",
            "existing_streak_before"
        ]
    ]
)

history["win_streak_selected"] = (
    history["existing_streak_before"].combine_first(
        history["win_streak_before_rebuilt"]
    )
)

# Only use finish rates where the reconstruction says
# the fighter's finish history is complete.
history["finish_rate_verified_candidate"] = (
    history["finish_rate_before_rebuilt"].where(
        history["finish_history_complete"].eq(True)
    )
)

# Event-level averages: every fighter appearance counts once.
recent_event_validated = (
    history.groupby(
        ["event_name", "event_date"], as_index=False
    )
    .agg(
        fighter_appearances=("fighter", "size"),
        avg_ufc_experience=("ufc_fights_before_rebuilt", "mean"),
        avg_win_streak=("win_streak_selected", "mean"),
        avg_finish_rate_candidate=(
            "finish_rate_verified_candidate", "mean"
        ),
        finish_rates_available=(
            "finish_rate_verified_candidate", "count"
        ),
        streak_disagreements=("streak_disagreement", "sum")
    )
    .sort_values("event_date")
)

recent_event_validated["finish_rates_missing"] = (
    recent_event_validated["fighter_appearances"]
    - recent_event_validated["finish_rates_available"]
)

recent_event_validated["finish_rate_complete"] = (
    recent_event_validated["finish_rates_missing"].eq(0)
)

# Averages with missing finish rates must not be presented
# as complete event-level values.
recent_event_validated["avg_finish_rate"] = (
    recent_event_validated["avg_finish_rate_candidate"].where(
        recent_event_validated["finish_rate_complete"]
    )
)

print("\nRECONSTRUCTED RECENT EVENT FACTORS:")
display(recent_event_validated.round(4))

print("\nEVENTS WITH COMPLETE FINISH-RATE COVERAGE:")
print(
    int(recent_event_validated["finish_rate_complete"].sum()),
    "of",
    len(recent_event_validated)
)

print("\nMISSING FINISH-RATE FIGHTER APPEARANCES:")
display(
    history.loc[
        history["finish_rate_verified_candidate"].isna(),
        ["event_date", "event_name", "fighter",
         "ufc_wins_before_rebuilt",
         "ufc_finishes_before_known",
         "finish_history_complete"]
    ]
)

# Preserve in Colab for the next stage
recent_event_validated.to_csv(
    "/content/ufc_recent_event_validated_factors.csv",
    index=False
)

print("\nSaved event-level validation results.")
```

```python

import pandas as pd
import numpy as np
import re

history = recent_fighter_history.copy()
master = backup["event_master"]
recent = scoring_fights[
    ~scoring_fights["event_name"].isin(master["event_name"])
].copy()

recent["event_date"] = pd.to_datetime(recent["event_date"])
history["event_date"] = pd.to_datetime(history["event_date"])

missing = history[
    history["finish_rate_before_rebuilt"].isna()
].copy()

affected_fighters = set(missing["fighter"])

# Find earlier recent victories for affected fighters.
# Those victories need their methods verified.
verification_rows = []

for _, fight in recent.iterrows():
    for side, other in [("f1", "f2"), ("f2", "f1")]:
        fighter = str(fight[f"{side}_name"]).strip()
        result = str(fight[f"{side}_result"]).strip()

        if fighter not in affected_fighters or result != "W":
            continue

        later_missing = missing[
            (missing["fighter"] == fighter)
            & (missing["event_date"] > fight["event_date"])
        ]

        if later_missing.empty:
            continue

        url_match = re.search(
            r"https?://(?:www\.)?ufcstats\.com/fight-details/[a-fA-F0-9]+",
            str(fight["fight_url"])
        )

        verification_rows.append({
            "event_date": fight["event_date"],
            "event_name": fight["event_name"],
            "winner": fighter,
            "opponent": fight[f"{other}_name"],
            "fight_url": url_match.group(0) if url_match else None,
            "method": None
        })

finish_method_verification = (
    pd.DataFrame(verification_rows)
    .drop_duplicates(subset=["fight_url"])
    .sort_values("event_date")
    .reset_index(drop=True)
)

print("EARLIER RECENT FIGHTS NEEDING METHOD VERIFICATION:")
print(len(finish_method_verification))

display(finish_method_verification)

print("\nMISSING FIGHTER APPEARANCES:")
display(
    missing[
        ["event_name", "fighter",
         "ufc_wins_before_rebuilt",
         "ufc_finishes_before_known"]
    ]
)

finish_method_verification.to_csv(
    "/content/ufc_finish_methods_to_verify.csv",
    index=False
)

print("\nSaved targeted verification list.")
```

```python

import pandas as pd
import numpy as np

# Verified from official UFC event results and scorecards.
verified = [
    ("2026-07-11", "King Green", "Terrance McKinney", "KO/TKO"),
    ("2026-07-11", "Gable Steveson", "Elisha Ellison", "KO/TKO"),
    ("2026-07-11", "Wang Cong", "Tracy Cortez", "Decision"),
    ("2026-07-11", "Damian Pinas", "Cesar Almeida", "KO/TKO"),
    ("2026-07-11", "Ryan Gandra", "Zach Reese", "KO/TKO"),
    ("2026-07-18", "Christian Leroy Duncan", "Jared Cannonier", "Decision"),
    ("2026-07-18", "Tommy McMillen", "Alberto Montes", "KO/TKO"),
    ("2026-07-18", "Jose Delgado", "Austin Bashi", "Decision"),
    ("2026-07-18", "Felipe Franco", "Levi Rodrigues Jr.", "KO/TKO"),
    ("2026-07-18", "Alden Coria", "Stewart Nicoll", "Decision"),
    ("2026-07-18", "RJ Harris", "Alvin Hines", "KO/TKO"),
    ("2026-07-25", "Axel Sola", "Ismael Bonfim", "Submission"),
    ("2026-08-15", "Esteban Ribovics", "Edson Barboza", "KO/TKO"),
    ("2026-08-22", "Anthony Wint", "Terrance Chatman", "Submission"),
]

verification = pd.DataFrame(
    verified, columns=["event_date", "winner", "opponent", "method"]
)
verification["event_date"] = pd.to_datetime(verification["event_date"])
verification["is_finish"] = verification["method"].isin(
    ["KO/TKO", "Submission"]
)

# Work on copies, preserving the original reconstruction.
history = recent_fighter_history.copy()
history["event_date"] = pd.to_datetime(history["event_date"])

recent = scoring_fights[
    ~scoring_fights["event_name"].isin(
        backup["event_master"]["event_name"]
    )
].copy()
recent["event_date"] = pd.to_datetime(recent["event_date"])

# Confirm all verified bouts are present and results agree.
matched = []

for _, v in verification.iterrows():
    mask = (
        recent["event_date"].eq(v["event_date"])
        & (
            (
                recent["f1_name"].eq(v["winner"])
                & recent["f2_name"].eq(v["opponent"])
                & recent["f1_result"].eq("W")
            )
            |
            (
                recent["f2_name"].eq(v["winner"])
                & recent["f1_name"].eq(v["opponent"])
                & recent["f2_result"].eq("W")
            )
        )
    )
    count = int(mask.sum())
    matched.append(count)

verification["matched_fights"] = matched

assert verification["matched_fights"].eq(1).all(), (
    "Some verified bouts do not match exactly. "
    "Review verification before updating."
)

print("VERIFIED FIGHTS MATCHED:",
      int(verification["matched_fights"].sum()), "of 14")

# Update each fighter's finish count for appearances
# strictly AFTER the verified victory.
history["additional_verified_finishes"] = 0

for _, v in verification.iterrows():
    mask = (
        history["fighter"].eq(v["winner"])
        & history["event_date"].gt(v["event_date"])
    )

    if v["is_finish"]:
        history.loc[mask, "additional_verified_finishes"] += 1

history["ufc_finishes_before_updated"] = (
    pd.to_numeric(
        history["ufc_finishes_before_known"], errors="coerce"
    )
    + history["additional_verified_finishes"]
)

wins = pd.to_numeric(
    history["ufc_wins_before_rebuilt"], errors="coerce"
)

# The 14 previously unresolved victories have now been
# classified. Do not assume unrelated unknown methods
# are verified; use the earlier completeness flag.
history["finish_rate_updated"] = np.where(
    wins.gt(0),
    history["ufc_finishes_before_updated"] / wins,
    0.0
)

# Check that every previously incomplete appearance
# was affected by a verified earlier victory.
previously_missing = history[
    history["finish_rate_before_rebuilt"].isna()
].copy()

resolved = history.loc[
    previously_missing.index,
    "fighter"
].isin(verification["winner"])

assert resolved.all(), (
    "At least one missing fighter was not resolved "
    "by the verified method list."
)

# Recalculate event-level averages.
recent_finish_rates_completed = (
    history.groupby(
        ["event_name", "event_date"], as_index=False
    )
    .agg(
        fighter_appearances=("fighter", "size"),
        avg_finish_rate=("finish_rate_updated", "mean"),
        finish_rates_available=("finish_rate_updated", "count")
    )
    .sort_values("event_date")
)

recent_finish_rates_completed["complete"] = (
    recent_finish_rates_completed["fighter_appearances"]
    .eq(recent_finish_rates_completed["finish_rates_available"])
)

print("\nUPDATED FINISH-RATE AVERAGES:")
display(recent_finish_rates_completed.round(5))

print("\nEVENTS WITH COMPLETE CALCULATED FINISH RATES:",
      int(recent_finish_rates_completed["complete"].sum()),
      "of", len(recent_finish_rates_completed))

# Save the updated data without overwriting historical scores.
history.to_csv(
    "/content/ufc_recent_fighter_finish_rates_updated.csv",
    index=False
)

recent_finish_rates_completed.to_csv(
    "/content/ufc_recent_event_finish_rates_completed.csv",
    index=False
)

print("\nSaved both updated datasets.")
```

```python

import pandas as pd
import numpy as np

hist = backup["ufc_fights"].copy()
master = backup["event_master"].copy()

rank_cols = ["f1_rank_before", "f2_rank_before"]

for c in rank_cols:
    hist[c] = pd.to_numeric(hist[c], errors="coerce")

master["total_ranking_points"] = pd.to_numeric(
    master["total_ranking_points"], errors="coerce"
)

# Aggregate the number of fighters at each rank per event.
rank_long = pd.concat([
    hist[["event_name", "f1_rank_before"]].rename(
        columns={"f1_rank_before": "rank"}
    ),
    hist[["event_name", "f2_rank_before"]].rename(
        columns={"f2_rank_before": "rank"}
    )
], ignore_index=True)

rank_long = rank_long.dropna(subset=["rank"])
rank_long["rank"] = rank_long["rank"].astype(int)

rank_counts = pd.crosstab(
    rank_long["event_name"], rank_long["rank"]
).reindex(columns=range(16), fill_value=0)

comparison = master[
    ["event_name", "event_date",
     "total_ranking_points", "rankings_available"]
].merge(
    rank_counts,
    on="event_name",
    how="left"
)

for r in range(16):
    comparison[r] = comparison[r].fillna(0)

# Test common ranking point formulas.
candidate_formulas = {
    "16 - rank (champion=16)": {
        r: 16-r for r in range(16)
    },
    "17 - rank (champion=17)": {
        r: 17-r for r in range(16)
    },
    "champion=20, rank1=15 to rank15=1": {
        r: (20 if r == 0 else 16-r)
        for r in range(16)
    },
    "champion=25, rank1=15 to rank15=1": {
        r: (25 if r == 0 else 16-r)
        for r in range(16)
    },
    "champion=30, rank1=15 to rank15=1": {
        r: (30 if r == 0 else 16-r)
        for r in range(16)
    }
}

results = []

for name, mapping in candidate_formulas.items():

    predicted = sum(
        comparison[r] * mapping[r]
        for r in range(16)
    )

    actual = comparison["total_ranking_points"]

    # Only evaluate events where historical rankings exist.
    mask = comparison["rankings_available"].eq(True)

    diff = (predicted[mask] - actual[mask]).abs()

    results.append({
        "formula": name,
        "events_compared": int(mask.sum()),
        "exact_matches": int((diff < 1e-8).sum()),
        "mean_absolute_error": diff.mean(),
        "max_error": diff.max()
    })

results_df = pd.DataFrame(results).sort_values(
    ["exact_matches", "mean_absolute_error"],
    ascending=[False, True]
)

print("RANKING POINT FORMULA COMPARISON:")
display(results_df)

# Infer per-rank point values directly from event totals.
# Least-squares is diagnostic only; the candidate formula
# must still reproduce historical event totals.
X = comparison.loc[
    comparison["rankings_available"].eq(True),
    list(range(16))
].to_numpy(dtype=float)

y = comparison.loc[
    comparison["rankings_available"].eq(True),
    "total_ranking_points"
].to_numpy(dtype=float)

if len(X):
    coefficients, residuals, rank, singular = np.linalg.lstsq(
        X, y, rcond=None
    )

    inferred = pd.DataFrame({
        "rank": list(range(16)),
        "inferred_points": coefficients,
        "rounded_points": np.rint(coefficients)
    })

    print("\nPOINT VALUES INFERRED FROM HISTORICAL TOTALS:")
    display(inferred)

    predictions = X @ coefficients

    print("\nLEAST-SQUARES FIT:")
    print("Events:", len(y))
    print("Exact within 1e-6:",
          int(np.isclose(predictions, y, atol=1e-6).sum()))
    print("Max error:",
          float(np.max(np.abs(predictions-y))))

print("\nHistorical ranking data unchanged.")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# VERIFIED PRE-EVENT RANKINGS DATABASE
# ============================================================

audit = recent_ranking_audit.copy()
audit["event_date"] = pd.to_datetime(audit["event_date"])

# One record per fighter appearance.
assert len(audit) == 358
assert audit["event_name"].nunique() == 14

# Historical scoring formula, confirmed across 553 events.
def ranking_points(rank, verified_unranked=False):
    if verified_unranked:
        return 0.0
    if pd.isna(rank):
        return np.nan
    rank = float(rank)
    if rank < 0 or rank > 15 or not rank.is_integer():
        raise ValueError(f"Invalid UFC rank: {rank}")
    return 16.0 - rank

# Preserve an existing verification file if one exists.
path = Path("/content/ufc_recent_rankings_verified.csv")

if path.exists():
    previous = pd.read_csv(path)

    # Restore previously entered rankings by fight and fighter.
    previous_cols = [
        "fight_url", "fighter", "rank_before",
        "verified_unranked", "ranking_verified",
        "ranking_source"
    ]

    previous = previous[
        [c for c in previous_cols if c in previous.columns]
    ]

    audit = audit.drop(
        columns=["rank_before", "ranking_verified",
                 "ranking_source"],
        errors="ignore"
    ).merge(
        previous,
        on=["fight_url", "fighter"],
        how="left",
        validate="one_to_one"
    )

if "verified_unranked" not in audit.columns:
    audit["verified_unranked"] = False

audit["verified_unranked"] = (
    audit["verified_unranked"]
    .fillna(False)
    .astype(bool)
)

audit["ranking_verified"] = (
    audit["ranking_verified"]
    .fillna(False)
    .astype(bool)
)

audit["rank_before"] = pd.to_numeric(
    audit["rank_before"], errors="coerce"
)

# Validation: verified ranks must be 0–15, or explicitly
# verified as unranked.
valid_ranks = (
    audit["rank_before"].isna()
    | (
        audit["rank_before"].between(0, 15)
        & audit["rank_before"].mod(1).eq(0)
    )
)

assert valid_ranks.all(), "Invalid ranking values found."

assert not (
    audit["verified_unranked"]
    & audit["rank_before"].notna()
).any(), "A fighter cannot be ranked and unranked."

audit["ranking_resolved"] = (
    audit["ranking_verified"]
    & (
        audit["rank_before"].notna()
        | audit["verified_unranked"]
    )
    & audit["ranking_source"].notna()
)

audit["ranking_points"] = audit.apply(
    lambda row: (
        ranking_points(
            row["rank_before"],
            row["verified_unranked"]
        )
        if row["ranking_resolved"]
        else np.nan
    ),
    axis=1
)

# Calculate event totals only when every appearance is resolved.
event_rankings = (
    audit.groupby(
        ["event_name", "event_date"], as_index=False
    )
    .agg(
        fighter_appearances=("fighter", "size"),
        rankings_resolved=("ranking_resolved", "sum"),
        ranking_points_known=("ranking_points", "sum")
    )
)

event_rankings["ranking_complete"] = (
    event_rankings["fighter_appearances"]
    .eq(event_rankings["rankings_resolved"])
)

event_rankings["total_ranking_points"] = np.where(
    event_rankings["ranking_complete"],
    event_rankings["ranking_points_known"],
    np.nan
)

# Main event ranking points (two fighters per event).
main_event_rankings = (
    audit[audit["is_main_event"]]
    .groupby("event_name", as_index=False)
    .agg(
        main_event_fighters=("fighter", "size"),
        main_event_rankings_resolved=("ranking_resolved", "sum"),
        main_event_points_known=("ranking_points", "sum")
    )
)

main_event_rankings["main_event_ranking_points"] = np.where(
    main_event_rankings["main_event_rankings_resolved"].eq(2),
    main_event_rankings["main_event_points_known"],
    np.nan
)

event_rankings = event_rankings.merge(
    main_event_rankings[
        ["event_name", "main_event_ranking_points"]
    ],
    on="event_name",
    how="left",
    validate="one_to_one"
)

# Save database and progress.
audit.to_csv(path, index=False)

event_rankings.to_csv(
    "/content/ufc_recent_event_ranking_progress.csv",
    index=False
)

print("RANKING VERIFICATION PROGRESS:")
print("Fighter appearances:", len(audit))
print("Verified:", int(audit["ranking_resolved"].sum()))
print("Still missing:", int((~audit["ranking_resolved"]).sum()))
print("Completed events:",
      int(event_rankings["ranking_complete"].sum()), "/ 14")

display(event_rankings.sort_values("event_date"))

print("\nSaved ranking verification database.")
```

```python

import pandas as pd
import numpy as np
import requests
import json
import time
from pathlib import Path
from datetime import timedelta
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 1: HISTORICAL RANKINGS SOURCE RECOVERY
# ============================================================

OUT = Path("/content/ufc_rankings_recovery")
OUT.mkdir(exist_ok=True)

assert "recent_ranking_audit" in globals(), (
    "Restore the project checkpoint first."
)

rankings = recent_ranking_audit.copy()
rankings["event_date"] = pd.to_datetime(rankings["event_date"])

assert len(rankings) == 358
assert rankings["event_name"].nunique() == 14

# ------------------------------------------------------------
# 1. Restore any previously saved verification progress
# ------------------------------------------------------------

verified_path = Path(
    "/content/ufc_recent_rankings_verified.csv"
)

if verified_path.exists():
    previous = pd.read_csv(verified_path)

    if {"fight_url", "fighter"}.issubset(previous.columns):
        columns_to_restore = [
            "rank_before",
            "ranking_verified",
            "verified_unranked",
            "ranking_source"
        ]

        previous = previous[
            ["fight_url", "fighter"] +
            [c for c in columns_to_restore if c in previous]
        ].drop_duplicates(["fight_url", "fighter"])

        rankings = rankings.drop(
            columns=columns_to_restore,
            errors="ignore"
        ).merge(
            previous,
            on=["fight_url", "fighter"],
            how="left",
            validate="one_to_one"
        )

for col, default in [
    ("ranking_verified", False),
    ("verified_unranked", False),
    ("ranking_source", None),
]:
    if col not in rankings:
        rankings[col] = default

# ------------------------------------------------------------
# 2. Prepare the 14 event dates
# ------------------------------------------------------------

events = (
    rankings[["event_name", "event_date"]]
    .drop_duplicates()
    .sort_values("event_date")
    .reset_index(drop=True)
)

events["pre_event_date"] = (
    events["event_date"] - pd.Timedelta(days=1)
)

events["date_string"] = (
    events["pre_event_date"].dt.strftime("%Y-%m-%d")
)

print("=" * 75)
print("UFC EVENT STRENGTH INDEX V2")
print("HISTORICAL RANKINGS RECOVERY")
print("=" * 75)

print("\nEvents to recover:", len(events))
print("Fighter appearances:", len(rankings))

display(events)

# ------------------------------------------------------------
# 3. Test publicly accessible ranking archives
# ------------------------------------------------------------

session = requests.Session()
session.headers.update({
    "User-Agent": (
        "Mozilla/5.0 UFCEventStrengthIndex/"
        "2.0 HistoricalResearch"
    )
})

def request_json(url, params=None, timeout=18):
    try:
        response = session.get(
            url,
            params=params,
            timeout=timeout
        )

        if response.status_code != 200:
            return {
                "ok": False,
                "status": response.status_code,
                "data": None,
                "error": response.text[:120]
            }

        return {
            "ok": True,
            "status": 200,
            "data": response.json(),
            "error": None
        }

    except Exception as exc:
        return {
            "ok": False,
            "status": None,
            "data": None,
            "error": str(exc)[:200]
        }

# Wayback Machine: look for snapshots of the UFC rankings
# page. These are source candidates, NOT yet verified ranks.

archive_targets = [
    "www.ufc.com/rankings",
    "ufc.com/rankings"
]

archive_results = []

for target in archive_targets:

    result = request_json(
        "https://web.archive.org/cdx/search/cdx",
        params={
            "url": target,
            "output": "json",
            "filter": "statuscode:200",
            "filter": "mimetype:text/html",
            "from": "20260701",
            "to": "20261011",
            "collapse": "timestamp:4",
            "fl": "timestamp,original,statuscode",
            "filter": "statuscode:200"
        },
        timeout=20
    )

    if result["ok"] and isinstance(result["data"], list):
        rows = result["data"]

        for item in rows[1:]:
            if len(item) >= 3:
                archive_results.append({
                    "timestamp": item[0],
                    "original": item[1],
                    "status": item[2],
                    "archive_url": (
                        "https://web.archive.org/web/"
                        + item[0] + "/"
                        + item[1]
                    )
                })

        print(
            f"Archive {target}:",
            max(0, len(rows)-1),
            "candidate snapshots"
        )
    else:
        print(
            f"Archive {target}: unavailable",
            result["status"],
            result["error"]
        )

archive_df = pd.DataFrame(
    archive_results,
    columns=[
        "timestamp", "original",
        "status", "archive_url"
    ]
).drop_duplicates()

archive_df.to_csv(
    OUT / "archive_snapshot_candidates.csv",
    index=False
)

# ------------------------------------------------------------
# 4. Optional: test UFCalendar historical rankings API
# ------------------------------------------------------------

# This service requires an API key.
# Leave the key blank to skip it.
# Do NOT paste a private API key into chat.
#
# To use it later, enter your key through getpass.
#
# Official service documentation:
# https://www.ufcalendar.com/developers/ufc-rankings-api

USE_UFCALENDAR_API = False

api_results = []

if USE_UFCALENDAR_API:
    from getpass import getpass

    key = getpass("UFCalendar API key: ").strip()

    if not key:
        print("No API key entered; skipping API.")

    else:
        for _, event in events.iterrows():

            date_string = event["date_string"]

            # API authentication header may require adjustment
            # to match the provider's current documentation.
            try:
                response = requests.get(
                    "https://api.ufcalendar.com/v1/rankings/ufc",
                    params={"date": date_string},
                    headers={
                        "Authorization": f"Bearer {key}",
                        "Accept": "application/json"
                    },
                    timeout=20
                )

                api_results.append({
                    "event_name": event["event_name"],
                    "date": date_string,
                    "http_status": response.status_code,
                    "response_length": len(response.content),
                    "response_type": response.headers.get(
                        "Content-Type", ""
                    )
                })

                if response.ok:
                    filename = (
                        "ufcalendar_"
                        + date_string
                        + ".json"
                    )

                    with open(
                        OUT / filename, "w"
                    ) as f:
                        json.dump(
                            response.json(),
                            f,
                            indent=2
                        )

            except Exception as exc:
                api_results.append({
                    "event_name": event["event_name"],
                    "date": date_string,
                    "http_status": None,
                    "response_length": 0,
                    "response_type": str(exc)[:100]
                })

            time.sleep(0.2)

# ------------------------------------------------------------
# 5. Match archive snapshots to event dates
# ------------------------------------------------------------

archive_matches = []

if len(archive_df):

    archive_df["snapshot_date"] = pd.to_datetime(
        archive_df["timestamp"].str[:8],
        format="%Y%m%d",
        errors="coerce"
    )

    for _, event in events.iterrows():

        # Historical rankings must not be taken from
        # a snapshot AFTER the event.
        candidates = archive_df[
            archive_df["snapshot_date"]
            <= event["pre_event_date"]
        ].copy()

        candidates["days_old"] = (
            event["pre_event_date"]
            - candidates["snapshot_date"]
        ).dt.days

        candidates = candidates[
            candidates["days_old"].between(0, 14)
        ].sort_values("days_old")

        if len(candidates):
            best = candidates.iloc[0]

            archive_matches.append({
                "event_name": event["event_name"],
                "event_date": event["event_date"],
                "snapshot_date": best["snapshot_date"],
                "days_old": best["days_old"],
                "archive_url": best["archive_url"],
                "candidate_found": True
            })

        else:
            archive_matches.append({
                "event_name": event["event_name"],
                "event_date": event["event_date"],
                "snapshot_date": pd.NaT,
                "days_old": np.nan,
                "archive_url": None,
                "candidate_found": False
            })

else:
    for _, event in events.iterrows():
        archive_matches.append({
            "event_name": event["event_name"],
            "event_date": event["event_date"],
            "snapshot_date": pd.NaT,
            "days_old": np.nan,
            "archive_url": None,
            "candidate_found": False
        })

archive_matches = pd.DataFrame(archive_matches)

archive_matches.to_csv(
    OUT / "event_archive_candidates.csv",
    index=False
)

# ------------------------------------------------------------
# 6. Prioritize main-event fighters
# ------------------------------------------------------------

main_event_priority = (
    rankings[rankings["is_main_event"].eq(True)]
    .copy()
    .sort_values(["event_date", "fighter"])
)

main_event_priority["rank_points_if_verified"] = np.where(
    pd.to_numeric(
        main_event_priority["rank_before"],
        errors="coerce"
    ).between(0, 15),
    16 - pd.to_numeric(
        main_event_priority["rank_before"],
        errors="coerce"
    ),
    np.nan
)

main_event_priority.to_csv(
    OUT / "main_event_fighters_priority.csv",
    index=False
)

rankings.to_csv(
    OUT / "all_358_fighters_rankings_worklist.csv",
    index=False
)

# ------------------------------------------------------------
# 7. Report findings
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("ARCHIVED UFC RANKINGS SOURCE CANDIDATES")
print("=" * 75)

print("Archived snapshots found:", len(archive_df))
print(
    "Events with nearby archive candidates:",
    int(archive_matches["candidate_found"].sum()),
    "/ 14"
)

display(archive_matches)

print("\n" + "=" * 75)
print("MAIN-EVENT RANKINGS — FIRST PRIORITY")
print("=" * 75)

display(
    main_event_priority[
        ["event_date", "event_name", "fighter",
         "rank_before", "ranking_source"]
    ]
)

print("\n" + "=" * 75)
print("FINAL STATUS")
print("=" * 75)

print("Total fighter appearances:", len(rankings))
print(
    "Rankings currently recorded:",
    int(rankings["rank_before"].notna().sum())
)

print("Files saved in:", OUT)

print("\nIMPORTANT:")
print(
    "Archive candidates are not verified ranking values."
)
print(
    "No historical or recent event scores were modified."
)

if api_results:
    print("\nAPI RESPONSE STATUS:")
    display(pd.DataFrame(api_results))
```

```python

import pandas as pd
import numpy as np
import requests
import unicodedata
import re
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 2 — FREE HISTORICAL RANKINGS RECOVERY
# Download, inspect, match, validate, save
# ============================================================

OUT = Path("/content/ufc_rankings_recovery")
OUT.mkdir(exist_ok=True)

BASE = (
    "https://raw.githubusercontent.com/"
    "martj42/ufc_rankings_history/master/"
)

SOURCES = {
    "historical": BASE + "rankings_history.csv",
    "meta": BASE + "meta_rankings.csv"
}

# ------------------------------------------------------------
# 1. Download both historical ranking files
# ------------------------------------------------------------

datasets = {}

for source_name, url in SOURCES.items():
    print(f"\nDownloading {source_name} rankings...")

    try:
        response = requests.get(url, timeout=45)
        response.raise_for_status()

        local_path = OUT / f"{source_name}_rankings_source.csv"
        local_path.write_bytes(response.content)

        df = pd.read_csv(local_path, low_memory=False)
        df.columns = df.columns.str.strip().str.lower()

        datasets[source_name] = df

        print("Downloaded rows:", len(df))
        print("Columns:", df.columns.tolist())
        display(df.head(3))

    except Exception as exc:
        print("Download failed:", exc)

if not datasets:
    raise RuntimeError(
        "Neither ranking file could be downloaded."
    )

# ------------------------------------------------------------
# 2. Standardize names and divisions
# ------------------------------------------------------------

def clean_name(value):
    if pd.isna(value):
        return ""

    text = unicodedata.normalize("NFKD", str(value))
    text = "".join(
        c for c in text if not unicodedata.combining(c)
    )

    text = text.casefold()
    text = re.sub(r"[^a-z0-9]+", " ", text)

    return " ".join(text.split())


def clean_division(value):
    if pd.isna(value):
        return ""

    text = str(value).casefold()

    women = (
        "women" in text
        or "female" in text
        or "womens" in text
    )

    divisions = [
        "strawweight",
        "flyweight",
        "bantamweight",
        "featherweight",
        "lightweight",
        "welterweight",
        "middleweight",
        "light heavyweight",
        "heavyweight"
    ]

    # Match longer names first.
    divisions = sorted(divisions, key=len, reverse=True)

    for division in divisions:
        if division in text:
            if women:
                return "women " + division
            return division

    return ""


def detect_column(df, options):
    for option in options:
        if option in df.columns:
            return option
    return None


# ------------------------------------------------------------
# 3. Inspect source coverage and standardize schema
# ------------------------------------------------------------

standardized = []
source_reports = []

for source_name, df in datasets.items():

    date_col = detect_column(
        df,
        ["date", "snapshot_date", "ranking_date",
         "updated", "updated_at"]
    )

    fighter_col = detect_column(
        df,
        ["fighter", "fighter_name", "name"]
    )

    rank_col = detect_column(
        df,
        ["rank", "ranking", "position"]
    )

    division_col = detect_column(
        df,
        ["weightclass", "weight_class",
         "division", "weight_class_name"]
    )

    if not all([
        date_col, fighter_col, rank_col, division_col
    ]):
        print(
            f"\n{source_name}: schema needs review.",
            df.columns.tolist()
        )
        continue

    standardized_df = pd.DataFrame({
        "snapshot_date": pd.to_datetime(
            df[date_col], errors="coerce"
        ),
        "fighter": df[fighter_col],
        "rank_raw": df[rank_col],
        "division_raw": df[division_col],
        "source_type": source_name,
        "source_url": SOURCES[source_name]
    })

    standardized_df["fighter_key"] = (
        standardized_df["fighter"].map(clean_name)
    )

    standardized_df["division_key"] = (
        standardized_df["division_raw"].map(clean_division)
    )

    # Champions may be represented as rank 0 or text.
    rank_text = (
        standardized_df["rank_raw"]
        .astype(str)
        .str.strip()
        .str.casefold()
    )

    standardized_df["rank"] = pd.to_numeric(
        standardized_df["rank_raw"],
        errors="coerce"
    )

    champion_mask = rank_text.isin([
        "c", "champion", "champ", "0"
    ])

    standardized_df.loc[
        champion_mask, "rank"
    ] = 0

    standardized_df = standardized_df.dropna(
        subset=["snapshot_date"]
    )

    standardized_df = standardized_df[
        standardized_df["rank"].between(0, 15)
        & standardized_df["rank"].mod(1).eq(0)
        & standardized_df["fighter_key"].ne("")
        & standardized_df["division_key"].ne("")
    ].copy()

    standardized.append(standardized_df)

    source_reports.append({
        "source": source_name,
        "rows_usable": len(standardized_df),
        "first_date": standardized_df[
            "snapshot_date"
        ].min(),
        "last_date": standardized_df[
            "snapshot_date"
        ].max(),
        "snapshot_dates": standardized_df[
            "snapshot_date"
        ].nunique(),
        "divisions": standardized_df[
            "division_key"
        ].nunique()
    })

print("\n" + "=" * 75)
print("HISTORICAL RANKING SOURCE COVERAGE")
print("=" * 75)

coverage = pd.DataFrame(source_reports)
display(coverage)

if not standardized:
    raise RuntimeError(
        "No usable ranking data. Review source schemas."
    )

all_rankings = pd.concat(
    standardized, ignore_index=True
)

# ------------------------------------------------------------
# 4. Prepare the recent fighter appearances
# ------------------------------------------------------------

worklist = recent_ranking_audit.copy()

worklist["event_date"] = pd.to_datetime(
    worklist["event_date"]
)

worklist["fighter_key"] = (
    worklist["fighter"].map(clean_name)
)

worklist["division_key"] = (
    worklist["weightclass"].map(clean_division)
)

worklist["pre_event_date"] = (
    worklist["event_date"]
    - pd.Timedelta(days=1)
)

# ------------------------------------------------------------
# 5. Match fighter, division, and pre-event date
# ------------------------------------------------------------

# This is deliberately conservative:
# - exact normalized fighter name
# - exact weight division
# - source date BEFORE the event
# - no assumption that missing means unranked
#
# Allow only a snapshot within 14 days.
# If older data exists, retain it as a candidate,
# but do not treat it as verified.

candidates = worklist.reset_index().rename(
    columns={"index": "appearance_id"}
).merge(
    all_rankings[
        ["snapshot_date", "fighter_key",
         "division_key", "rank",
         "source_type", "source_url"]
    ],
    on=["fighter_key", "division_key"],
    how="left"
)

candidates = candidates[
    candidates["snapshot_date"].notna()
    & (
        candidates["snapshot_date"]
        <= candidates["pre_event_date"]
    )
].copy()

candidates["snapshot_age_days"] = (
    candidates["pre_event_date"]
    - candidates["snapshot_date"]
).dt.days

# Prefer recent data, with Meta-era records preferred
# when two different sources have the same date.
candidates["source_priority"] = (
    candidates["source_type"].map({
        "meta": 0,
        "historical": 1
    })
)

candidates = candidates.sort_values(
    ["appearance_id",
     "snapshot_date",
     "source_priority"],
    ascending=[True, False, True]
)

# If a fighter has more than one ranking on the same
# snapshot, keep those cases for manual review rather
# than silently choosing an arbitrary one.

best_candidates = candidates.drop_duplicates(
    "appearance_id",
    keep="first"
).copy()

best_candidates["recent_enough"] = (
    best_candidates["snapshot_age_days"]
    .between(0, 14)
)

# ------------------------------------------------------------
# 6. Attach candidates WITHOUT declaring verification
# ------------------------------------------------------------

candidate_cols = [
    "appearance_id",
    "snapshot_date",
    "snapshot_age_days",
    "rank",
    "source_type",
    "source_url",
    "recent_enough"
]

results = worklist.reset_index().rename(
    columns={"index": "appearance_id"}
).merge(
    best_candidates[candidate_cols],
    on="appearance_id",
    how="left",
    validate="one_to_one"
)

results = results.rename(
    columns={
        "rank": "candidate_rank",
        "source_url": "candidate_source",
        "source_type": "candidate_source_type"
    }
)

results["candidate_points"] = np.where(
    results["candidate_rank"].between(0, 15),
    16 - results["candidate_rank"],
    np.nan
)

results["candidate_status"] = np.select(
    [
        results["candidate_rank"].isna(),
        results["recent_enough"].eq(True)
    ],
    [
        "NO_MATCH",
        "RECENT_SOURCE_CANDIDATE"
    ],
    default="STALE_SOURCE_CANDIDATE"
)

# ------------------------------------------------------------
# 7. Report progress by event
# ------------------------------------------------------------

event_progress = (
    results.groupby(
        ["event_name", "event_date"],
        as_index=False
    )
    .agg(
        fighter_appearances=("fighter", "size"),
        recent_candidates=(
            "candidate_status",
            lambda x: (x == "RECENT_SOURCE_CANDIDATE").sum()
        ),
        stale_candidates=(
            "candidate_status",
            lambda x: (x == "STALE_SOURCE_CANDIDATE").sum()
        ),
        unmatched=(
            "candidate_status",
            lambda x: (x == "NO_MATCH").sum()
        )
    )
    .sort_values("event_date")
)

print("\n" + "=" * 75)
print("RECENT EVENT RANKING MATCH PROGRESS")
print("=" * 75)

display(event_progress)

print("\n" + "=" * 75)
print("MAIN-EVENT RANKING CANDIDATES")
print("=" * 75)

display(
    results[results["is_main_event"].eq(True)][
        ["event_date", "fighter",
         "candidate_rank",
         "snapshot_date",
         "snapshot_age_days",
         "candidate_status"]
    ].sort_values("event_date")
)

print("\n" + "=" * 75)
print("OVERALL MATCH SUMMARY")
print("=" * 75)

print(
    results["candidate_status"].value_counts(
        dropna=False
    )
)

# ------------------------------------------------------------
# 8. Save everything for review
# ------------------------------------------------------------

all_rankings.to_csv(
    OUT / "all_downloaded_rankings_standardized.csv",
    index=False
)

results.to_csv(
    OUT / "recent_358_rankings_source_candidates.csv",
    index=False
)

event_progress.to_csv(
    OUT / "ranking_candidate_coverage_by_event.csv",
    index=False
)

coverage.to_csv(
    OUT / "ranking_source_date_coverage.csv",
    index=False
)

# Keep the original verified database untouched.
print("\n" + "=" * 75)
print("FILES SAVED")
print("=" * 75)

print("Folder:", OUT)
print("No verified ranks were overwritten.")
print("No event scores were changed.")
print(
    "Candidate matches require validation before scoring."
)
```

```python

import pandas as pd
import numpy as np
import re
import unicodedata
from pathlib import Path
from IPython.display import display

# ================================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 3 — SNAPSHOT-BASED RANKING RECONSTRUCTION
#
# Uses complete divisional ranking snapshots.
# Never interprets missing data as zero automatically.
# Does NOT overwrite existing scores or verified rankings.
# ================================================================

OUT = Path("/content/ufc_rankings_recovery")
OUT.mkdir(parents=True, exist_ok=True)

assert "all_rankings" in globals(), (
    "Run Big Cell 2 first to load the ranking archives."
)
assert "recent_ranking_audit" in globals(), (
    "Restore your UFC project checkpoint first."
)

# ------------------------------------------------
# 1. Normalize source records
# ------------------------------------------------

def normalize_fighter(value):
    if pd.isna(value):
        return ""
    s = unicodedata.normalize("NFKD", str(value))
    s = "".join(
        ch for ch in s if not unicodedata.combining(ch)
    )
    s = s.casefold()
    s = re.sub(r"[^a-z0-9]+", " ", s)
    return " ".join(s.split())


def normalize_division(value):
    if pd.isna(value):
        return ""

    s = str(value).casefold().replace("-", " ")
    s = re.sub(r"\s+", " ", s).strip()

    women = (
        "women" in s
        or "female" in s
        or "w straw" in s
        or "w fly" in s
        or "w bantam" in s
        or "w feather" in s
    )

    weight_names = [
        "light heavyweight",
        "strawweight",
        "flyweight",
        "bantamweight",
        "featherweight",
        "lightweight",
        "welterweight",
        "middleweight",
        "heavyweight"
    ]

    for name in weight_names:
        if name in s:
            return ("women " if women else "") + name

    return ""


source = all_rankings.copy()

source["snapshot_date"] = pd.to_datetime(
    source["snapshot_date"], errors="coerce"
).dt.normalize()

source["fighter_key"] = (
    source["fighter"].map(normalize_fighter)
)

source["division_key"] = (
    source["division_raw"].map(normalize_division)
)

source["rank"] = pd.to_numeric(
    source["rank"], errors="coerce"
)

source = source[
    source["snapshot_date"].notna()
    & source["rank"].between(0, 15)
    & source["rank"].mod(1).eq(0)
    & source["fighter_key"].ne("")
    & source["division_key"].ne("")
].copy()

source["rank"] = source["rank"].astype(int)

# ------------------------------------------------
# 2. Identify reasonably complete snapshots
# ------------------------------------------------

# A snapshot should contain the majority of a
# division's ranked fighters. Require at least
# 12 unique positions from ranks 1-15.
#
# This is a quality threshold, not proof that
# every fighter outside the snapshot is unranked.

snapshot_quality = (
    source.groupby(
        ["source_type", "snapshot_date", "division_key"],
        as_index=False
    )
    .agg(
        source_rows=("fighter_key", "size"),
        unique_fighters=("fighter_key", "nunique"),
        unique_ranks=("rank", "nunique"),
        numbered_ranks=(
            "rank",
            lambda x: x[x.between(1, 15)].nunique()
        ),
        champion_present=(
            "rank",
            lambda x: bool((x == 0).any())
        )
    )
)

snapshot_quality["usable"] = (
    snapshot_quality["numbered_ranks"] >= 12
)

# Detect conflicting rank assignments inside
# individual snapshots.
rank_conflicts = (
    source.groupby(
        ["source_type", "snapshot_date",
         "division_key", "rank"]
    )["fighter_key"]
    .nunique()
    .reset_index(name="fighters_at_rank")
)

conflict_groups = rank_conflicts[
    rank_conflicts["fighters_at_rank"] > 1
][
    ["source_type", "snapshot_date", "division_key"]
].drop_duplicates()

conflict_groups["has_conflict"] = True

snapshot_quality = snapshot_quality.merge(
    conflict_groups,
    on=["source_type", "snapshot_date", "division_key"],
    how="left"
)

snapshot_quality["has_conflict"] = (
    snapshot_quality["has_conflict"].fillna(False)
)

snapshot_quality["usable"] &= (
    ~snapshot_quality["has_conflict"]
)

# ------------------------------------------------
# 3. Build the 358 fighter appearances
# ------------------------------------------------

appearances = recent_ranking_audit.copy()
appearances = appearances.reset_index(
    drop=True
).reset_index(names="appearance_id")

appearances["event_date"] = pd.to_datetime(
    appearances["event_date"]
).dt.normalize()

appearances["pre_event_date"] = (
    appearances["event_date"]
    - pd.Timedelta(days=1)
)

appearances["fighter_key"] = (
    appearances["fighter"].map(normalize_fighter)
)

if "weightclass" not in appearances.columns:
    raise ValueError(
        "Expected weightclass in recent_ranking_audit."
    )

appearances["division_key"] = (
    appearances["weightclass"].map(normalize_division)
)

# ------------------------------------------------
# 4. Choose the latest available complete snapshot
#    for each event and division
# ------------------------------------------------

event_divisions = (
    appearances[
        ["event_name", "event_date",
         "pre_event_date", "division_key"]
    ]
    .drop_duplicates()
)

eligible = event_divisions.merge(
    snapshot_quality[
        snapshot_quality["usable"]
    ],
    on="division_key",
    how="left"
)

eligible = eligible[
    eligible["snapshot_date"].notna()
    & (
        eligible["snapshot_date"]
        <= eligible["pre_event_date"]
    )
].copy()

eligible["age_days"] = (
    eligible["pre_event_date"]
    - eligible["snapshot_date"]
).dt.days

# We prefer the Meta source when two sources
# have snapshots from the same date.
eligible["source_priority"] = (
    eligible["source_type"].map({
        "meta": 0,
        "historical": 1
    }).fillna(2)
)

eligible = eligible.sort_values(
    [
        "event_name", "division_key",
        "snapshot_date", "source_priority"
    ],
    ascending=[True, True, False, True]
)

chosen = eligible.drop_duplicates(
    ["event_name", "event_date", "division_key"]
).copy()

chosen = chosen.rename(columns={
    "snapshot_date": "chosen_snapshot_date",
    "source_type": "chosen_source",
    "age_days": "snapshot_age_days",
    "numbered_ranks": "snapshot_numbered_ranks",
    "champion_present": "snapshot_champion_present"
})

snapshot_cols = [
    "event_name", "event_date", "division_key",
    "chosen_snapshot_date", "chosen_source",
    "snapshot_age_days", "snapshot_numbered_ranks",
    "snapshot_champion_present"
]

appearances = appearances.merge(
    chosen[snapshot_cols],
    on=["event_name", "event_date", "division_key"],
    how="left",
    validate="many_to_one"
)

# ------------------------------------------------
# 5. Match fighters ONLY within their selected
#    event-specific ranking snapshot
# ------------------------------------------------

source_lookup = source[
    [
        "source_type", "snapshot_date",
        "division_key", "fighter_key", "rank"
    ]
].copy()

# If duplicate fighter entries exist in a snapshot,
# only accept them when they agree on the rank.
fighter_conflicts = (
    source_lookup.groupby(
        ["source_type", "snapshot_date",
         "division_key", "fighter_key"]
    )["rank"]
    .nunique()
    .reset_index(name="different_ranks")
)

safe_fighters = fighter_conflicts[
    fighter_conflicts["different_ranks"] == 1
][
    ["source_type", "snapshot_date",
     "division_key", "fighter_key"]
]

source_lookup = source_lookup.merge(
    safe_fighters,
    on=[
        "source_type", "snapshot_date",
        "division_key", "fighter_key"
    ],
    how="inner"
).drop_duplicates()

source_lookup = source_lookup.rename(columns={
    "source_type": "chosen_source",
    "snapshot_date": "chosen_snapshot_date",
    "rank": "snapshot_rank"
})

appearances = appearances.merge(
    source_lookup,
    on=[
        "chosen_source", "chosen_snapshot_date",
        "division_key", "fighter_key"
    ],
    how="left",
    validate="many_to_one"
)

# ------------------------------------------------
# 6. Classify without inventing rankings
# ------------------------------------------------

MAX_SNAPSHOT_AGE = 14

appearances["snapshot_recent"] = (
    appearances["snapshot_age_days"]
    .between(0, MAX_SNAPSHOT_AGE)
)

appearances["ranking_status"] = np.select(
    [
        appearances["division_key"].eq(""),
        appearances["chosen_snapshot_date"].isna(),
        ~appearances["snapshot_recent"],
        appearances["snapshot_rank"].notna()
    ],
    [
        "DIVISION_NOT_RECOGNIZED",
        "NO_COMPLETE_SNAPSHOT",
        "SNAPSHOT_TOO_OLD",
        "RANKED_SOURCE_MATCH"
    ],
    default="NOT_LISTED_IN_SNAPSHOT"
)

# Only confirmed ranked candidates receive
# candidate points. No-match stays NaN.
appearances["candidate_rank"] = np.where(
    appearances["ranking_status"].eq(
        "RANKED_SOURCE_MATCH"
    ),
    appearances["snapshot_rank"],
    np.nan
)

appearances["candidate_points"] = np.where(
    appearances["candidate_rank"].notna(),
    16 - appearances["candidate_rank"],
    np.nan
)

# These remain source candidates, not yet
# committed as verified rankings.
appearances["ready_for_manual_review"] = (
    appearances["ranking_status"].isin([
        "RANKED_SOURCE_MATCH",
        "NOT_LISTED_IN_SNAPSHOT"
    ])
)

# ------------------------------------------------
# 7. Event-level progress
# ------------------------------------------------

progress = (
    appearances.groupby(
        ["event_name", "event_date"],
        as_index=False
    )
    .agg(
        appearances=("fighter", "size"),
        ranked_matches=(
            "ranking_status",
            lambda x: (x == "RANKED_SOURCE_MATCH").sum()
        ),
        not_listed=(
            "ranking_status",
            lambda x: (
                x == "NOT_LISTED_IN_SNAPSHOT"
            ).sum()
        ),
        old_snapshots=(
            "ranking_status",
            lambda x: (
                x == "SNAPSHOT_TOO_OLD"
            ).sum()
        ),
        missing_snapshots=(
            "ranking_status",
            lambda x: (
                x == "NO_COMPLETE_SNAPSHOT"
            ).sum()
        )
    )
    .sort_values("event_date")
)

progress["potentially_resolvable"] = (
    progress["ranked_matches"]
    + progress["not_listed"]
)

progress["all_appearances_have_recent_snapshot"] = (
    progress["potentially_resolvable"]
    == progress["appearances"]
)

# ------------------------------------------------
# 8. Show results
# ------------------------------------------------

print("=" * 78)
print("UFC EVENT STRENGTH INDEX V2")
print("SNAPSHOT-BASED RANKING RECOVERY")
print("=" * 78)

print("\nSOURCE SNAPSHOT QUALITY:")
print("Total divisional snapshots:", len(snapshot_quality))
print(
    "Usable snapshots:",
    int(snapshot_quality["usable"].sum())
)

print("\nRANKING STATUS COUNTS:")
print(
    appearances["ranking_status"].value_counts(
        dropna=False
    ).to_string()
)

print("\nEVENT-LEVEL PROGRESS:")
display(progress)

print("\nMAIN-EVENT FIGHTERS:")
display(
    appearances[
        appearances["is_main_event"].eq(True)
    ][
        [
            "event_date", "fighter",
            "division_key", "candidate_rank",
            "chosen_snapshot_date",
            "snapshot_age_days", "chosen_source",
            "ranking_status"
        ]
    ].sort_values(["event_date", "fighter"])
)

print("\nNOT-LISTED FIGHTERS — REVIEW REQUIRED:")
display(
    appearances[
        appearances["ranking_status"].eq(
            "NOT_LISTED_IN_SNAPSHOT"
        )
    ][
        [
            "event_date", "fighter",
            "division_key", "chosen_snapshot_date",
            "chosen_source"
        ]
    ].head(35)
)

# ------------------------------------------------
# 9. Save reusable data
# ------------------------------------------------

snapshot_quality.to_csv(
    OUT / "snapshot_quality_audit.csv",
    index=False
)

chosen.to_csv(
    OUT / "chosen_event_division_snapshots.csv",
    index=False
)

appearances.to_csv(
    OUT / "rankings_358_snapshot_based.csv",
    index=False
)

progress.to_csv(
    OUT / "ranking_snapshot_progress_by_event.csv",
    index=False
)

# Save in notebook memory too
ranking_snapshot_results = appearances.copy()
ranking_snapshot_progress = progress.copy()

print("\n" + "=" * 78)
print("COMPLETE — SOURCE CANDIDATES SAVED")
print("=" * 78)
print("Folder:", OUT)
print("Original event scores: UNCHANGED")
print("Original verified ranking database: UNCHANGED")
print("No missing ranking was automatically assigned zero.")
```

```python

import pandas as pd
import numpy as np
import re
import unicodedata
from difflib import SequenceMatcher
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 4 — CROSS-DIVISION + NAME-ALIAS RANKING AUDIT
#
# 1. Keep existing exact division matches
# 2. Search other ranked divisions at the same event date
# 3. Flag name variations (do NOT auto-approve fuzzy matches)
# 4. Investigate unrecognized fight divisions
# 5. Check whether selected snapshots are current
# 6. Export a full review workbook
#
# NO existing event scores or verified ranks are overwritten.
# NO missing rankings are assigned zero.
# ============================================================

OUT = Path("/content/ufc_rankings_recovery")
OUT.mkdir(parents=True, exist_ok=True)

assert "ranking_snapshot_results" in globals(), (
    "Please run Big Cell 3 first."
)
assert "all_rankings" in globals(), (
    "Please run Big Cell 2 first."
)

work = ranking_snapshot_results.copy()
source = all_rankings.copy()

# ------------------------------------------------------------
# 1. Standardize source and fighter names
# ------------------------------------------------------------

def clean_fighter(x):
    if pd.isna(x):
        return ""
    x = unicodedata.normalize("NFKD", str(x))
    x = "".join(
        c for c in x if not unicodedata.combining(c)
    )
    x = x.casefold()
    x = re.sub(r"[^a-z0-9]+", " ", x)
    return " ".join(x.split())

def clean_div(x):
    if pd.isna(x):
        return ""

    x = str(x).casefold()
    x = x.replace("-", " ")
    x = re.sub(r"\s+", " ", x).strip()

    female = (
        "women" in x
        or "female" in x
        or x.startswith("w ")
    )

    names = [
        "light heavyweight",
        "strawweight",
        "flyweight",
        "bantamweight",
        "featherweight",
        "lightweight",
        "welterweight",
        "middleweight",
        "heavyweight"
    ]

    for n in names:
        if n in x:
            return ("women " if female else "") + n

    return ""

source["snapshot_date"] = pd.to_datetime(
    source["snapshot_date"], errors="coerce"
).dt.normalize()

source["fighter_key"] = (
    source["fighter"].map(clean_fighter)
)

source["division_key"] = (
    source["division_raw"].map(clean_div)
)

source["rank"] = pd.to_numeric(
    source["rank"], errors="coerce"
)

source = source[
    source["snapshot_date"].notna()
    & source["rank"].between(0, 15)
    & source["fighter_key"].ne("")
    & source["division_key"].ne("")
].copy()

source["rank"] = source["rank"].astype(int)

work["event_date"] = pd.to_datetime(
    work["event_date"]
).dt.normalize()

work["pre_event_date"] = (
    work["event_date"] - pd.Timedelta(days=1)
)

work["fighter_key"] = (
    work["fighter"].map(clean_fighter)
)

work["division_key"] = (
    work["weightclass"].map(clean_div)
)

# ------------------------------------------------------------
# 2. Select recent usable snapshots independently
#    for EACH division, source and event date
# ------------------------------------------------------------

quality = snapshot_quality.copy()

quality["snapshot_date"] = pd.to_datetime(
    quality["snapshot_date"]
).dt.normalize()

quality = quality[quality["usable"].eq(True)].copy()

events = work[
    ["event_name", "event_date", "pre_event_date"]
].drop_duplicates()

event_sources = events.merge(
    quality,
    how="cross"
)

event_sources = event_sources[
    event_sources["snapshot_date"]
    <= event_sources["pre_event_date"]
].copy()

event_sources["age_days"] = (
    event_sources["pre_event_date"]
    - event_sources["snapshot_date"]
).dt.days

event_sources = event_sources[
    event_sources["age_days"].between(0, 14)
].copy()

event_sources["priority"] = (
    event_sources["source_type"].map({
        "meta": 0,
        "historical": 1
    }).fillna(2)
)

event_sources = event_sources.sort_values(
    [
        "event_name",
        "division_key",
        "snapshot_date",
        "priority"
    ],
    ascending=[True, True, False, True]
)

# One selected snapshot per event and division.
chosen_all_divisions = event_sources.drop_duplicates(
    ["event_name", "event_date", "division_key"]
).copy()

# ------------------------------------------------------------
# 3. Build a source roster of all ranked fighters
#    available before each event
# ------------------------------------------------------------

roster = chosen_all_divisions[
    [
        "event_name",
        "event_date",
        "division_key",
        "source_type",
        "snapshot_date"
    ]
].merge(
    source[
        [
            "division_key",
            "source_type",
            "snapshot_date",
            "fighter_key",
            "fighter",
            "rank"
        ]
    ],
    on=[
        "division_key",
        "source_type",
        "snapshot_date"
    ],
    how="inner"
)

roster = roster.drop_duplicates(
    [
        "event_name",
        "event_date",
        "division_key",
        "fighter_key",
        "rank"
    ]
)

# ------------------------------------------------------------
# 4. Exact matches in OTHER divisions
# ------------------------------------------------------------

cross = work[
    [
        "appearance_id",
        "event_name",
        "event_date",
        "fighter",
        "fighter_key",
        "division_key"
    ]
].merge(
    roster,
    on=[
        "event_name",
        "event_date",
        "fighter_key"
    ],
    how="inner",
    suffixes=("_fight", "_ranking")
)

cross = cross[
    cross["division_key_fight"]
    != cross["division_key_ranking"]
].copy()

cross = cross.rename(columns={
    "division_key_ranking": "ranked_division",
    "division_key_fight": "fight_division",
    "rank": "other_division_rank",
    "source_type": "other_division_source",
    "snapshot_date": "other_division_snapshot"
})

cross["other_division_points"] = (
    16 - cross["other_division_rank"]
)

# Do not automatically add other-division points.
# Historical scoring rules need to be checked first.

# ------------------------------------------------------------
# 5. Fuzzy matching for unresolved fighters
# ------------------------------------------------------------

unresolved = work[
    work["ranking_status"].ne(
        "RANKED_SOURCE_MATCH"
    )
].copy()

rosters_by_event = {
    key: group
    for key, group in roster.groupby(
        ["event_name", "event_date"]
    )
}

fuzzy_rows = []

for _, fighter_row in unresolved.iterrows():

    key = (
        fighter_row["event_name"],
        fighter_row["event_date"]
    )

    event_roster = rosters_by_event.get(key)

    if event_roster is None:
        continue

    fighter_key = fighter_row["fighter_key"]

    # Limit to the same division where possible.
    same_division = event_roster[
        event_roster["division_key"]
        == fighter_row["division_key"]
    ]

    comparison = (
        same_division
        if len(same_division)
        else event_roster
    )

    matches = []

    for _, ranked_row in comparison.iterrows():

        similarity = SequenceMatcher(
            None,
            fighter_key,
            ranked_row["fighter_key"]
        ).ratio()

        if similarity >= 0.76:
            matches.append({
                "appearance_id": fighter_row["appearance_id"],
                "event_name": fighter_row["event_name"],
                "event_date": fighter_row["event_date"],
                "fighter": fighter_row["fighter"],
                "suggested_name": ranked_row["fighter"],
                "similarity": similarity,
                "suggested_rank": ranked_row["rank"],
                "suggested_division": ranked_row["division_key"],
                "snapshot_date": ranked_row["snapshot_date"],
                "source_type": ranked_row["source_type"]
            })

    matches.sort(
        key=lambda x: x["similarity"],
        reverse=True
    )

    fuzzy_rows.extend(matches[:3])

fuzzy = pd.DataFrame(fuzzy_rows)

# ------------------------------------------------------------
# 6. Summarize cross-division candidates
# ------------------------------------------------------------

cross_summary = (
    cross.groupby("appearance_id")
    .agg(
        cross_division_matches=(
            "ranked_division", "size"
        ),
        ranked_divisions=(
            "ranked_division",
            lambda s: ", ".join(sorted(set(s)))
        ),
        other_division_ranks=(
            "other_division_rank",
            lambda s: ", ".join(
                map(str, sorted(set(s)))
            )
        )
    )
    .reset_index()
)

work = work.merge(
    cross_summary,
    on="appearance_id",
    how="left",
    validate="one_to_one"
)

work["cross_division_matches"] = (
    work["cross_division_matches"]
    .fillna(0).astype(int)
)

work["has_cross_division_match"] = (
    work["cross_division_matches"] > 0
)

# ------------------------------------------------------------
# 7. Unrecognized divisions
# ------------------------------------------------------------

unknown_divisions = work[
    work["division_key"].eq("")
][
    [
        "event_date",
        "event_name",
        "fighter",
        "weightclass",
        "division_key"
    ]
].copy()

# ------------------------------------------------------------
# 8. Print actionable findings
# ------------------------------------------------------------

print("=" * 78)
print("UFC EVENT STRENGTH INDEX V2")
print("CROSS-DIVISION AND NAME-ALIAS RECOVERY")
print("=" * 78)

print("\nTOTAL FIGHTER APPEARANCES:", len(work))
print(
    "Existing same-division ranking matches:",
    int(
        work["ranking_status"]
        .eq("RANKED_SOURCE_MATCH")
        .sum()
    )
)

print(
    "Appearances ranked in another division:",
    int(
        work["has_cross_division_match"].sum()
    )
)

print(
    "Possible name-alias matches:",
    len(fuzzy)
)

print(
    "Unrecognized fight divisions:",
    len(unknown_divisions)
)

print("\nEXACT CROSS-DIVISION MATCHES:")
display(
    cross[
        [
            "event_date",
            "fighter_fight",
            "fight_division",
            "ranked_division",
            "other_division_rank",
            "other_division_snapshot"
        ]
    ].sort_values(
        ["event_date", "fighter_fight"]
    )
    if len(cross)
    else pd.DataFrame({
        "result": ["No exact cross-division matches"]
    })
)

print("\nPOSSIBLE NAME ALIASES:")
display(
    fuzzy.sort_values(
        ["similarity", "event_date"],
        ascending=[False, True]
    ).head(50)
    if len(fuzzy)
    else pd.DataFrame({
        "result": ["No fuzzy name candidates"]
    })
)

print("\nUNRECOGNIZED FIGHT DIVISIONS:")
display(unknown_divisions)

# ------------------------------------------------------------
# 9. Event summary
# ------------------------------------------------------------

summary = (
    work.groupby(
        ["event_name", "event_date"],
        as_index=False
    )
    .agg(
        fighter_appearances=("fighter", "size"),
        same_division_ranked=(
            "ranking_status",
            lambda x: (
                x == "RANKED_SOURCE_MATCH"
            ).sum()
        ),
        ranked_elsewhere=(
            "has_cross_division_match",
            "sum"
        ),
        unknown_division=(
            "division_key",
            lambda x: (x == "").sum()
        )
    )
    .sort_values("event_date")
)

display(summary)

# ------------------------------------------------------------
# 10. Save all review files
# ------------------------------------------------------------

cross.to_csv(
    OUT / "cross_division_ranked_fighters.csv",
    index=False
)

fuzzy.to_csv(
    OUT / "possible_fighter_name_aliases.csv",
    index=False
)

unknown_divisions.to_csv(
    OUT / "unrecognized_weightclasses.csv",
    index=False
)

work.to_csv(
    OUT / "ranking_recovery_extended_358.csv",
    index=False
)

summary.to_csv(
    OUT / "ranking_recovery_extended_summary.csv",
    index=False
)

# Save in notebook memory
ranking_recovery_extended = work.copy()
ranking_cross_division = cross.copy()
ranking_fuzzy_aliases = fuzzy.copy()

print("\n" + "=" * 78)
print("BIG CELL 4 COMPLETE")
print("=" * 78)
print("Saved to:", OUT)
print("No rankings were automatically invented.")
print("No event scores were modified.")
print("Next: review cross-division and alias results.")
```

```python

import pandas as pd
import numpy as np
import re
import unicodedata
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 5 — HISTORICAL RANKING RULE AUDIT
#
# GOALS
# 1. Audit historical cross-division ranking behavior
# 2. Investigate the 8 cross-division candidates
# 3. Resolve catchweight appearances where supported
# 4. Check duplicate and conflicting source ranks
# 5. Produce a clean, conservative ranking worklist
#
# NO scores or existing verified ranks are overwritten.
# ============================================================

OUT = Path("/content/ufc_rankings_recovery")
OUT.mkdir(parents=True, exist_ok=True)

hist = backup["ufc_fights"].copy()
recent_work = ranking_snapshot_results.copy()
source = all_rankings.copy()

# ------------------------------------------------------------
# 1. Normalize fighter names and divisions
# ------------------------------------------------------------

def norm_name(value):
    if pd.isna(value):
        return ""
    s = unicodedata.normalize("NFKD", str(value))
    s = "".join(
        ch for ch in s
        if not unicodedata.combining(ch)
    )
    s = re.sub(r"[^a-z0-9]+", " ", s.casefold())
    return " ".join(s.split())

def norm_div(value):
    if pd.isna(value):
        return ""

    s = str(value).casefold().replace("-", " ")
    s = re.sub(r"\s+", " ", s).strip()

    if "catch" in s:
        return "catchweight"

    women = (
        "women" in s
        or "female" in s
        or s.startswith("w ")
    )

    for division in [
        "light heavyweight",
        "strawweight",
        "flyweight",
        "bantamweight",
        "featherweight",
        "lightweight",
        "welterweight",
        "middleweight",
        "heavyweight"
    ]:
        if division in s:
            return ("women " if women else "") + division

    return ""

# ------------------------------------------------------------
# 2. Historical evidence: rankings and fight divisions
# ------------------------------------------------------------

print("=" * 75)
print("HISTORICAL DATA SCHEMA")
print("=" * 75)

historical_columns = [
    c for c in hist.columns
    if any(k in c.lower() for k in [
        "rank", "weight", "division", "title"
    ])
]

print("Relevant historical columns:")
print(historical_columns)

historical_rank_rows = []

for side in ["f1", "f2"]:
    rank_col = f"{side}_rank_before"
    name_col = f"{side}_name"

    if rank_col not in hist or name_col not in hist:
        continue

    temp = pd.DataFrame({
        "event_name": hist["event_name"],
        "fighter": hist[name_col],
        "rank": pd.to_numeric(
            hist[rank_col], errors="coerce"
        ),
        "side": side
    })

    if "WEIGHTCLASS" in hist:
        temp["fight_division"] = (
            hist["WEIGHTCLASS"].map(norm_div)
        )
    else:
        temp["fight_division"] = ""

    if "ranking_division" in hist:
        temp["ranking_division"] = (
            hist["ranking_division"].map(norm_div)
        )
    else:
        temp["ranking_division"] = ""

    historical_rank_rows.append(temp)

historical_ranked = pd.concat(
    historical_rank_rows, ignore_index=True
)

historical_ranked = historical_ranked[
    historical_ranked["rank"].between(0, 15)
].copy()

historical_ranked["rank_points"] = (
    16 - historical_ranked["rank"]
)

print("\nHistorical ranked fighter appearances:",
      len(historical_ranked))

has_fight_div = historical_ranked[
    "fight_division"
].ne("")

has_rank_div = historical_ranked[
    "ranking_division"
].ne("")

comparable = historical_ranked[
    has_fight_div & has_rank_div
].copy()

comparable["cross_division"] = (
    comparable["fight_division"]
    != comparable["ranking_division"]
)

print("Historical appearances with both divisions:",
      len(comparable))
print("Different divisions:",
      int(comparable["cross_division"].sum()))

print("\nHISTORICAL CROSS-DIVISION EXAMPLES:")
display(
    comparable[comparable["cross_division"]][
        [
            "event_name", "fighter",
            "fight_division", "ranking_division",
            "rank", "rank_points"
        ]
    ].head(30)
)

if len(comparable) == 0:
    print(
        "\nHistorical data does not provide enough "
        "division information to prove the "
        "cross-division rule."
    )

# ------------------------------------------------------------
# 3. Inspect original event ranking calculation
# ------------------------------------------------------------

master = backup["event_master"].copy()

rank_long = pd.concat([
    hist[["event_name", "f1_rank_before"]]
    .rename(columns={"f1_rank_before": "rank"}),

    hist[["event_name", "f2_rank_before"]]
    .rename(columns={"f2_rank_before": "rank"})
], ignore_index=True)

rank_long["rank"] = pd.to_numeric(
    rank_long["rank"], errors="coerce"
)

rank_long["points"] = np.where(
    rank_long["rank"].between(0, 15),
    16 - rank_long["rank"],
    0
)

reconstructed_totals = (
    rank_long.groupby("event_name")["points"]
    .sum()
    .rename("reconstructed_points")
    .reset_index()
)

historical_check = master[
    ["event_name", "total_ranking_points",
     "rankings_available"]
].merge(
    reconstructed_totals,
    on="event_name",
    how="left",
    validate="one_to_one"
)

historical_check["difference"] = (
    historical_check["reconstructed_points"]
    - historical_check["total_ranking_points"]
)

valid = historical_check[
    historical_check["rankings_available"].eq(True)
    & historical_check["total_ranking_points"].notna()
]

print("\nHISTORICAL RANKING FORMULA VALIDATION:")
print("Events compared:", len(valid))
print(
    "Exact matches:",
    int(np.isclose(
        valid["difference"], 0, atol=1e-8
    ).sum())
)

# ------------------------------------------------------------
# 4. Inspect cross-division candidates
# ------------------------------------------------------------

cross = ranking_cross_division.copy()

print("\n" + "=" * 75)
print("RECENT CROSS-DIVISION CANDIDATES")
print("=" * 75)

if len(cross):
    cross["candidate_points"] = (
        16 - pd.to_numeric(
            cross["other_division_rank"],
            errors="coerce"
        )
    )

    display(
        cross[
            [
                "event_date",
                "fighter_fight",
                "fight_division",
                "ranked_division",
                "other_division_rank",
                "candidate_points",
                "other_division_snapshot"
            ]
        ].sort_values("event_date")
    )

# ------------------------------------------------------------
# 5. Resolve catchweight fighter lookup candidates
# ------------------------------------------------------------

source["snapshot_date"] = pd.to_datetime(
    source["snapshot_date"]
).dt.normalize()

source["fighter_key"] = (
    source["fighter"].map(norm_name)
)

source["division_key"] = (
    source["division_raw"].map(norm_div)
)

source["rank"] = pd.to_numeric(
    source["rank"], errors="coerce"
)

recent_work["event_date"] = pd.to_datetime(
    recent_work["event_date"]
).dt.normalize()

recent_work["fighter_key"] = (
    recent_work["fighter"].map(norm_name)
)

recent_work["fight_division_clean"] = (
    recent_work["weightclass"].map(norm_div)
)

catchweight = recent_work[
    recent_work["fight_division_clean"].eq(
        "catchweight"
    )
].copy()

catch_rows = []

for _, fighter in catchweight.iterrows():

    candidates = source[
        (source["fighter_key"] == fighter["fighter_key"])
        & (
            source["snapshot_date"]
            < fighter["event_date"]
        )
        & (
            source["snapshot_date"]
            >= fighter["event_date"]
            - pd.Timedelta(days=15)
        )
    ].sort_values(
        "snapshot_date", ascending=False
    )

    if candidates.empty:
        catch_rows.append({
            "event_name": fighter["event_name"],
            "event_date": fighter["event_date"],
            "fighter": fighter["fighter"],
            "candidate_division": None,
            "candidate_rank": np.nan,
            "snapshot_date": pd.NaT,
            "status": "NO_RANK_FOUND"
        })
        continue

    for _, candidate in candidates.head(5).iterrows():
        catch_rows.append({
            "event_name": fighter["event_name"],
            "event_date": fighter["event_date"],
            "fighter": fighter["fighter"],
            "candidate_division": candidate["division_key"],
            "candidate_rank": candidate["rank"],
            "snapshot_date": candidate["snapshot_date"],
            "status": "CATCHWEIGHT_RANK_CANDIDATE"
        })

catch_audit = pd.DataFrame(catch_rows)

print("\n" + "=" * 75)
print("CATCHWEIGHT RANKING CANDIDATES")
print("=" * 75)

display(catch_audit)

# ------------------------------------------------------------
# 6. Check source contradictions
# ------------------------------------------------------------

source_conflicts = (
    source.dropna(subset=["rank"])
    .groupby([
        "source_type",
        "snapshot_date",
        "division_key",
        "fighter_key"
    ])["rank"]
    .nunique()
    .reset_index(name="distinct_ranks")
)

source_conflicts = source_conflicts[
    source_conflicts["distinct_ranks"] > 1
]

print("\nSOURCE RANKING CONFLICTS:")
print("Conflicting fighter/snapshot records:",
      len(source_conflicts))

display(source_conflicts.head(20))

# ------------------------------------------------------------
# 7. Prepare a conservative consolidated worklist
# ------------------------------------------------------------

worklist = recent_work.copy()

worklist["historical_formula_points"] = np.where(
    worklist["candidate_rank"].notna(),
    16 - worklist["candidate_rank"],
    np.nan
)

worklist["cross_division_candidate"] = (
    worklist["appearance_id"].isin(
        cross["appearance_id"]
    )
    if len(cross)
    else False
)

worklist["needs_review"] = (
    worklist["candidate_rank"].isna()
    | worklist["cross_division_candidate"]
    | worklist["fight_division_clean"].eq(
        "catchweight"
    )
)

worklist["review_reason"] = np.select(
    [
        worklist["fight_division_clean"].eq(
            "catchweight"
        ),
        worklist["cross_division_candidate"],
        worklist["candidate_rank"].isna()
    ],
    [
        "CATCHWEIGHT",
        "RANKED_IN_OTHER_DIVISION",
        "NOT_CONFIRMED_RANKED"
    ],
    default="EXACT_RANK_CANDIDATE"
)

# ------------------------------------------------------------
# 8. Final summary and exports
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CONSOLIDATED RANKING REVIEW STATUS")
print("=" * 75)

print(worklist["review_reason"].value_counts())

print("\nMain-event candidates:")
display(
    worklist[
        worklist["is_main_event"].eq(True)
    ][
        [
            "event_date", "fighter",
            "candidate_rank",
            "historical_formula_points",
            "review_reason"
        ]
    ].sort_values("event_date")
)

historical_check.to_csv(
    OUT / "historical_ranking_formula_validation.csv",
    index=False
)

comparable.to_csv(
    OUT / "historical_cross_division_evidence.csv",
    index=False
)

catch_audit.to_csv(
    OUT / "catchweight_ranking_candidates.csv",
    index=False
)

source_conflicts.to_csv(
    OUT / "source_ranking_conflicts.csv",
    index=False
)

worklist.to_csv(
    OUT / "consolidated_358_ranking_review.csv",
    index=False
)

# Persist useful objects in Colab memory
ranking_historical_audit = historical_check
ranking_catchweight_audit = catch_audit
ranking_final_review_worklist = worklist

print("\n" + "=" * 75)
print("BIG CELL 5 COMPLETE")
print("=" * 75)
print("Files saved:", OUT)
print("No original event scores changed.")
print("No missing ranks were set to zero.")
print("No cross-division candidates were auto-approved.")
```

```python

import pandas as pd
import numpy as np
import re
import unicodedata
from pathlib import Path
from IPython.display import display

# ================================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 6 — HISTORICAL DIVISION RULE VALIDATION
#
# Fixes weight_class column issue.
# Tests historical cross-division scoring.
# Audits recent rankings and catchweight candidates.
# Saves a full evidence report.
#
# Does not overwrite any final scores.
# ================================================================

OUT = Path("/content/ufc_rankings_recovery")
OUT.mkdir(parents=True, exist_ok=True)

assert "backup" in globals()
assert "ranking_snapshot_results" in globals()
assert "ranking_cross_division" in globals()

hist = backup["ufc_fights"].copy()
master = backup["event_master"].copy()
recent = ranking_snapshot_results.copy()
cross = ranking_cross_division.copy()

# ------------------------------------------------
# 1. Division normalization
# ------------------------------------------------

def normalize_division(x):
    if pd.isna(x):
        return ""

    s = str(x).lower()
    s = s.replace("-", " ")
    s = re.sub(r"\s+", " ", s).strip()

    if "catch" in s:
        return "catchweight"

    women = (
        "women" in s or "female" in s
        or s.startswith("w ")
    )

    divisions = [
        "light heavyweight",
        "strawweight",
        "flyweight",
        "bantamweight",
        "featherweight",
        "lightweight",
        "welterweight",
        "middleweight",
        "heavyweight"
    ]

    for d in divisions:
        if d in s:
            return ("women " if women else "") + d

    return ""

# ------------------------------------------------
# 2. Correct historical column names
# ------------------------------------------------

required = [
    "event_name", "weight_class",
    "ranking_division",
    "f1_rank_before", "f2_rank_before"
]

missing = [c for c in required if c not in hist.columns]

if missing:
    raise ValueError(
        "Missing historical columns: " + str(missing)
    )

print("=" * 76)
print("HISTORICAL DIVISION RULE VALIDATION")
print("=" * 76)

print("\nRAW DIVISION EXAMPLES:")
display(
    hist[
        ["weight_class", "ranking_division"]
    ].drop_duplicates().head(30)
)

hist["fight_division_clean"] = (
    hist["weight_class"].map(normalize_division)
)

hist["ranking_division_clean"] = (
    hist["ranking_division"].map(normalize_division)
)

# ------------------------------------------------
# 3. Reconstruct historical fighter-level ranks
# ------------------------------------------------

fighter_rows = []

for side in ["f1", "f2"]:

    fighter_col = next(
        (
            c for c in [
                f"{side}_name",
                f"{side}_fighter",
                f"{side}_fighter_name"
            ]
            if c in hist.columns
        ),
        None
    )

    temp = pd.DataFrame({
        "event_name": hist["event_name"],
        "fight_division": hist["fight_division_clean"],
        "ranking_division": hist["ranking_division_clean"],
        "rank": pd.to_numeric(
            hist[f"{side}_rank_before"],
            errors="coerce"
        ),
        "side": side
    })

    temp["fighter"] = (
        hist[fighter_col]
        if fighter_col else f"{side}_unknown"
    )

    fighter_rows.append(temp)

historical_fighters = pd.concat(
    fighter_rows, ignore_index=True
)

historical_fighters = historical_fighters[
    historical_fighters["rank"].between(0, 15)
].copy()

historical_fighters["ranking_points"] = (
    16 - historical_fighters["rank"]
)

comparable = historical_fighters[
    historical_fighters["fight_division"].ne("")
    & historical_fighters["ranking_division"].ne("")
    & historical_fighters["fight_division"].ne("catchweight")
].copy()

comparable["division_mismatch"] = (
    comparable["fight_division"]
    != comparable["ranking_division"]
)

historical_cross = comparable[
    comparable["division_mismatch"]
].copy()

print("\nHISTORICAL FIGHTER-LEVEL RESULTS:")
print("Ranked appearances:", len(historical_fighters))
print("Comparable division records:", len(comparable))
print("Cross-division records:", len(historical_cross))

print("\nHISTORICAL CROSS-DIVISION EXAMPLES:")
display(
    historical_cross[
        [
            "event_name", "fighter",
            "fight_division", "ranking_division",
            "rank", "ranking_points"
        ]
    ].head(50)
)

# ------------------------------------------------
# 4. Test historical point totals against master
# ------------------------------------------------

reconstructed = (
    historical_fighters
    .groupby("event_name", as_index=False)
    .agg(
        reconstructed_points=(
            "ranking_points", "sum"
        )
    )
)

checks = master[
    ["event_name", "total_ranking_points",
     "rankings_available"]
].merge(
    reconstructed,
    on="event_name",
    how="left",
    validate="one_to_one"
)

checks["reconstructed_points"] = (
    checks["reconstructed_points"].fillna(0)
)

checks["difference"] = (
    checks["reconstructed_points"]
    - checks["total_ranking_points"]
)

eligible = checks[
    checks["rankings_available"].eq(True)
    & checks["total_ranking_points"].notna()
]

exact = np.isclose(
    eligible["difference"],
    0,
    atol=1e-8
)

print("\nHISTORICAL EVENT POINT VALIDATION:")
print("Events tested:", len(eligible))
print("Exact matches:", int(exact.sum()))
print("Mismatches:", int((~exact).sum()))

# ------------------------------------------------
# 5. Cross-division contribution audit
# ------------------------------------------------

cross_event_names = set(
    historical_cross["event_name"]
)

cross_events = checks[
    checks["event_name"].isin(cross_event_names)
].copy()

cross_events = cross_events.merge(
    historical_cross.groupby(
        "event_name", as_index=False
    )["ranking_points"].sum().rename(
        columns={
            "ranking_points": "cross_division_points"
        }
    ),
    on="event_name",
    how="left"
)

cross_events["without_cross_division"] = (
    cross_events["reconstructed_points"]
    - cross_events["cross_division_points"]
)

cross_events["matches_with_all_points"] = (
    np.isclose(
        cross_events["reconstructed_points"],
        cross_events["total_ranking_points"],
        atol=1e-8
    )
)

cross_events["matches_without_cross_points"] = (
    np.isclose(
        cross_events["without_cross_division"],
        cross_events["total_ranking_points"],
        atol=1e-8
    )
)

cross_events = cross_events[
    cross_events["rankings_available"].eq(True)
    & cross_events["total_ranking_points"].notna()
]

print("\nCROSS-DIVISION HISTORICAL TEST:")
print(
    "Events with cross-division ranked fighters:",
    len(cross_events)
)
print(
    "Matching when all ranked points included:",
    int(cross_events["matches_with_all_points"].sum())
)
print(
    "Matching when cross-division points excluded:",
    int(cross_events["matches_without_cross_points"].sum())
)

display(cross_events.head(30))

# ------------------------------------------------
# 6. Recent exact and cross-division candidates
# ------------------------------------------------

recent["event_date"] = pd.to_datetime(
    recent["event_date"]
)

recent["fight_division_clean"] = (
    recent["weightclass"].map(normalize_division)
)

recent["exact_rank_candidate"] = pd.to_numeric(
    recent["candidate_rank"],
    errors="coerce"
)

recent["cross_division_rank_candidate"] = np.nan
recent["cross_division_name"] = None
recent["cross_division_source_date"] = pd.NaT

if len(cross):

    cross["other_division_snapshot"] = pd.to_datetime(
        cross["other_division_snapshot"]
    )

    # Only use one candidate when the fighter has
    # exactly one distinct eligible rank/division.
    grouped = cross.groupby("appearance_id")

    for appearance_id, rows in grouped:

        choices = rows[
            ["ranked_division", "other_division_rank"]
        ].drop_duplicates()

        if len(choices) != 1:
            continue

        choice = choices.iloc[0]
        newest = rows.sort_values(
            "other_division_snapshot",
            ascending=False
        ).iloc[0]

        mask = recent["appearance_id"].eq(appearance_id)

        recent.loc[
            mask, "cross_division_rank_candidate"
        ] = choice["other_division_rank"]

        recent.loc[
            mask, "cross_division_name"
        ] = choice["ranked_division"]

        recent.loc[
            mask, "cross_division_source_date"
        ] = newest["other_division_snapshot"]

# ------------------------------------------------
# 7. Prepare ranking candidates without approval
# ------------------------------------------------

recent["proposed_rank"] = (
    recent["exact_rank_candidate"]
    .combine_first(
        recent["cross_division_rank_candidate"]
    )
)

recent["proposed_points"] = np.where(
    recent["proposed_rank"].between(0, 15),
    16 - recent["proposed_rank"],
    np.nan
)

recent["candidate_type"] = np.select(
    [
        recent["exact_rank_candidate"].notna(),
        recent["cross_division_rank_candidate"].notna(),
        recent["fight_division_clean"].eq("catchweight")
    ],
    [
        "SAME_DIVISION",
        "CROSS_DIVISION",
        "CATCHWEIGHT_UNRESOLVED"
    ],
    default="NO_VERIFIED_RANK"
)

recent["approved_for_final_scoring"] = False

# ------------------------------------------------
# 8. Summarize recent ranking coverage
# ------------------------------------------------

recent_summary = (
    recent.groupby(
        ["event_name", "event_date"],
        as_index=False
    )
    .agg(
        fighter_appearances=("fighter", "size"),
        same_division=(
            "candidate_type",
            lambda x: (x == "SAME_DIVISION").sum()
        ),
        cross_division=(
            "candidate_type",
            lambda x: (x == "CROSS_DIVISION").sum()
        ),
        unresolved=(
            "candidate_type",
            lambda x: x.isin([
                "CATCHWEIGHT_UNRESOLVED",
                "NO_VERIFIED_RANK"
            ]).sum()
        ),
        known_candidate_points=(
            "proposed_points", "sum"
        )
    )
    .sort_values("event_date")
)

print("\n" + "=" * 76)
print("RECENT RANKING CANDIDATES")
print("=" * 76)

print(recent["candidate_type"].value_counts())

display(recent_summary)

print("\nMAIN EVENT FIGHTERS:")
display(
    recent[
        recent["is_main_event"].eq(True)
    ][
        [
            "event_date", "fighter",
            "fight_division_clean",
            "proposed_rank", "proposed_points",
            "candidate_type"
        ]
    ].sort_values(["event_date", "fighter"])
)

# ------------------------------------------------
# 9. Save all results
# ------------------------------------------------

historical_cross.to_csv(
    OUT / "historical_cross_division_corrected.csv",
    index=False
)

cross_events.to_csv(
    OUT / "historical_cross_division_event_tests.csv",
    index=False
)

checks.to_csv(
    OUT / "historical_ranking_validation_corrected.csv",
    index=False
)

recent.to_csv(
    OUT / "recent_ranking_candidates_358_v2.csv",
    index=False
)

recent_summary.to_csv(
    OUT / "recent_ranking_candidates_event_summary.csv",
    index=False
)

ranking_rule_audit_v2 = cross_events.copy()
ranking_candidate_v2 = recent.copy()
ranking_candidate_summary_v2 = recent_summary.copy()

print("\n" + "=" * 76)
print("BIG CELL 6 COMPLETE")
print("=" * 76)
print("Files saved to:", OUT)
print("No scores modified.")
print("No missing ranks automatically set to zero.")
print("All recent rankings remain candidates until approved.")
```

```python

import pandas as pd
import numpy as np
import re
import unicodedata
from difflib import SequenceMatcher
from pathlib import Path
from IPython.display import display

# ================================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 7 — BULK UNMATCHED RANKING INVESTIGATION
#
# Audits all 358 recent fighter appearances.
# Prioritizes the 278 without rank candidates.
# Checks names, other divisions, snapshot coverage,
# and historical ranking_division limitations.
#
# Does NOT overwrite verified data or final scores.
# ================================================================

OUT = Path("/content/ufc_rankings_recovery")
OUT.mkdir(parents=True, exist_ok=True)

for name in [
    "ranking_candidate_v2",
    "all_rankings",
    "snapshot_quality",
    "backup"
]:
    if name not in globals():
        raise RuntimeError(f"Missing {name}. Run earlier cells.")

work = ranking_candidate_v2.copy()
archive = all_rankings.copy()
quality = snapshot_quality.copy()
hist = backup["ufc_fights"].copy()

def norm_name(value):
    if pd.isna(value):
        return ""
    text = unicodedata.normalize("NFKD", str(value))
    text = "".join(
        c for c in text
        if not unicodedata.combining(c)
    )
    text = re.sub(r"[^a-z0-9]+", " ", text.casefold())
    return " ".join(text.split())

def norm_div(value):
    if pd.isna(value):
        return ""
    text = str(value).lower().replace("-", " ")
    text = re.sub(r"\s+", " ", text).strip()
    if "catch" in text:
        return "catchweight"
    women = (
        "women" in text or "female" in text
        or text.startswith("w ")
    )
    for div in [
        "light heavyweight", "strawweight",
        "flyweight", "bantamweight",
        "featherweight", "lightweight",
        "welterweight", "middleweight",
        "heavyweight"
    ]:
        if div in text:
            return ("women " if women else "") + div
    return ""

# ------------------------------------------------
# 1. Standardize input data
# ------------------------------------------------

work["event_date"] = pd.to_datetime(
    work["event_date"]
).dt.normalize()

work["fighter_key"] = work["fighter"].map(norm_name)
work["fight_division"] = work["weightclass"].map(norm_div)

archive["snapshot_date"] = pd.to_datetime(
    archive["snapshot_date"]
).dt.normalize()

archive["fighter_key"] = archive["fighter"].map(norm_name)
archive["division_key"] = archive["division_raw"].map(norm_div)
archive["rank"] = pd.to_numeric(
    archive["rank"], errors="coerce"
)

archive = archive[
    archive["snapshot_date"].notna()
    & archive["rank"].between(0, 15)
    & archive["fighter_key"].ne("")
    & archive["division_key"].ne("")
].copy()

quality["snapshot_date"] = pd.to_datetime(
    quality["snapshot_date"]
).dt.normalize()

quality = quality[
    quality["usable"].eq(True)
].copy()

# ------------------------------------------------
# 2. Build one event-specific snapshot per division
# ------------------------------------------------

events = work[
    ["event_name", "event_date"]
].drop_duplicates()

event_divisions = events.merge(
    quality[
        [
            "source_type", "snapshot_date",
            "division_key", "numbered_ranks",
            "champion_present"
        ]
    ],
    how="cross"
)

event_divisions["age_days"] = (
    event_divisions["event_date"]
    - event_divisions["snapshot_date"]
).dt.days

event_divisions = event_divisions[
    event_divisions["age_days"].between(1, 14)
].copy()

event_divisions["source_priority"] = (
    event_divisions["source_type"]
    .map({"meta": 0, "historical": 1})
    .fillna(2)
)

event_divisions = event_divisions.sort_values(
    [
        "event_name", "division_key",
        "snapshot_date", "source_priority"
    ],
    ascending=[True, True, False, True]
)

selected = event_divisions.drop_duplicates(
    ["event_name", "event_date", "division_key"]
).copy()

roster = selected.merge(
    archive[
        [
            "source_type", "snapshot_date",
            "division_key", "fighter_key",
            "fighter", "rank"
        ]
    ],
    on=[
        "source_type", "snapshot_date",
        "division_key"
    ],
    how="inner",
    suffixes=("", "_archive")
)

# ------------------------------------------------
# 3. Examine all currently unmatched appearances
# ------------------------------------------------

unmatched = work[
    work["proposed_rank"].isna()
].copy()

roster_by_event = {
    key: frame
    for key, frame in roster.groupby(
        ["event_name", "event_date"]
    )
}

findings = []
alias_candidates = []

for _, fighter in unmatched.iterrows():

    key = (fighter["event_name"], fighter["event_date"])
    current_roster = roster_by_event.get(key)

    result = {
        "appearance_id": fighter["appearance_id"],
        "event_name": fighter["event_name"],
        "event_date": fighter["event_date"],
        "fighter": fighter["fighter"],
        "fight_division": fighter["fight_division"],
        "exact_any_division": False,
        "possible_alias": False,
        "best_alias_similarity": np.nan,
        "best_alias_name": "",
        "best_alias_division": "",
        "same_division_snapshot": False,
        "same_division_snapshot_age": np.nan,
        "same_division_numbered_ranks": np.nan,
        "other_division_matches": "",
        "review_category": ""
    }

    if current_roster is None or current_roster.empty:
        result["review_category"] = "NO_RECENT_ROSTER"
        findings.append(result)
        continue

    same_div = current_roster[
        current_roster["division_key"]
        == fighter["fight_division"]
    ]

    if not same_div.empty:
        result["same_division_snapshot"] = True
        result["same_division_snapshot_age"] = (
            same_div["age_days"].min()
        )
        result["same_division_numbered_ranks"] = (
            same_div["numbered_ranks"].max()
        )

    exact = current_roster[
        current_roster["fighter_key"]
        == fighter["fighter_key"]
    ]

    if not exact.empty:
        result["exact_any_division"] = True
        result["other_division_matches"] = "; ".join(
            sorted(set(
                exact["division_key"].astype(str)
                + " rank "
                + exact["rank"].astype(int).astype(str)
            ))
        )
        result["review_category"] = "EXACT_RANK_RECHECK"
    else:
        comparisons = []

        for _, candidate in current_roster.iterrows():
            similarity = SequenceMatcher(
                None,
                fighter["fighter_key"],
                candidate["fighter_key"]
            ).ratio()

            if similarity >= 0.78:
                comparisons.append({
                    "appearance_id": fighter["appearance_id"],
                    "event_date": fighter["event_date"],
                    "fighter": fighter["fighter"],
                    "candidate_fighter": candidate["fighter"],
                    "candidate_rank": candidate["rank"],
                    "candidate_division": candidate["division_key"],
                    "similarity": round(similarity, 4),
                    "snapshot_date": candidate["snapshot_date"],
                    "source_type": candidate["source_type"]
                })

        comparisons.sort(
            key=lambda row: row["similarity"],
            reverse=True
        )

        alias_candidates.extend(comparisons[:3])

        if comparisons:
            best = comparisons[0]
            result["possible_alias"] = True
            result["best_alias_similarity"] = (
                best["similarity"]
            )
            result["best_alias_name"] = (
                best["candidate_fighter"]
            )
            result["best_alias_division"] = (
                best["candidate_division"]
            )
            result["review_category"] = "NAME_REVIEW"

        elif fighter["fight_division"] == "catchweight":
            result["review_category"] = "CATCHWEIGHT_REVIEW"

        elif result["same_division_snapshot"]:
            result["review_category"] = (
                "ABSENT_FROM_RECENT_SNAPSHOT"
            )

        else:
            result["review_category"] = (
                "NO_SAME_DIVISION_SNAPSHOT"
            )

    findings.append(result)

findings = pd.DataFrame(findings)
aliases = pd.DataFrame(alias_candidates)

# ------------------------------------------------
# 4. Check whether historical ranking_division
#    is just a fight-level category
# ------------------------------------------------

hist["fight_division_clean"] = (
    hist["weight_class"].map(norm_div)
)

hist["ranking_division_clean"] = (
    hist["ranking_division"].map(norm_div)
)

historical_comparison = hist[
    hist["ranking_division_clean"].ne("")
    & hist["fight_division_clean"].ne("")
].copy()

historical_comparison["same_division"] = (
    historical_comparison["fight_division_clean"]
    == historical_comparison["ranking_division_clean"]
)

print("=" * 78)
print("HISTORICAL DIVISION FIELD AUDIT")
print("=" * 78)
print(
    "Historical fights with comparable divisions:",
    len(historical_comparison)
)
print(
    "Fight and ranking division match:",
    int(historical_comparison["same_division"].sum())
)
print(
    "Different divisions:",
    int((~historical_comparison["same_division"]).sum())
)
print(
    "IMPORTANT: This checks fight-level fields, not "
    "independent fighter-level historical ranking divisions."
)

# ------------------------------------------------
# 5. Summaries
# ------------------------------------------------

print("\n" + "=" * 78)
print("UNMATCHED FIGHTER AUDIT")
print("=" * 78)

print("All appearances:", len(work))
print("Rank candidates already found:",
      int(work["proposed_rank"].notna().sum()))
print("Unmatched appearances:", len(unmatched))

print("\nREVIEW CATEGORIES:")
print(
    findings["review_category"]
    .value_counts()
    .to_string()
)

print("\nPOSSIBLE NAME MATCHES:")
if not aliases.empty:
    display(
        aliases.sort_values(
            ["similarity", "event_date"],
            ascending=[False, True]
        ).head(40)
    )
else:
    print("No close aliases found.")

print("\nMAIN-EVENT FIGHTERS STILL UNMATCHED:")
main_ids = set(
    work.loc[
        work["is_main_event"].eq(True),
        "appearance_id"
    ]
)

display(
    findings[
        findings["appearance_id"].isin(main_ids)
    ][
        [
            "event_date", "fighter",
            "fight_division",
            "review_category",
            "best_alias_name",
            "same_division_snapshot"
        ]
    ].sort_values("event_date")
)

# ------------------------------------------------
# 6. Event-by-event unresolved breakdown
# ------------------------------------------------

event_review = (
    findings.groupby(
        ["event_name", "event_date", "review_category"]
    )
    .size()
    .unstack(fill_value=0)
    .reset_index()
    .sort_values("event_date")
)

print("\nEVENT-BY-EVENT REVIEW BREAKDOWN:")
display(event_review)

# ------------------------------------------------
# 7. Export everything
# ------------------------------------------------

findings.to_csv(
    OUT / "unmatched_278_detailed_audit.csv",
    index=False
)

aliases.to_csv(
    OUT / "unmatched_name_alias_candidates.csv",
    index=False
)

event_review.to_csv(
    OUT / "unmatched_review_by_event.csv",
    index=False
)

selected.to_csv(
    OUT / "event_division_source_snapshots.csv",
    index=False
)

roster.to_csv(
    OUT / "all_selected_event_ranking_rosters.csv",
    index=False
)

ranking_unmatched_audit_v3 = findings.copy()
ranking_alias_candidates_v3 = aliases.copy()
ranking_event_review_v3 = event_review.copy()

print("\n" + "=" * 78)
print("BIG CELL 7 COMPLETE")
print("=" * 78)
print("Files saved:", OUT)
print("No unranked classifications automatically approved.")
print("No rankings or final event scores overwritten.")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ================================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 8 — RANKING COMPLETENESS & DECISION AUDIT
#
# Uses the 358-fighter worklist and pre-event snapshots.
# Checks all numbered ranks 1–15.
# Rejects misleading fuzzy-name matches.
# Separates same-division, cross-division and unlisted.
# Produces a decision-ready review dataset.
#
# IMPORTANT:
# No unranked status is automatically verified.
# No historical or recent event scores are modified.
# ================================================================

OUT = Path("/content/ufc_rankings_recovery")
OUT.mkdir(parents=True, exist_ok=True)

required_objects = [
    "ranking_candidate_v2",
    "all_rankings",
    "ranking_unmatched_audit_v3",
    "event_divisions",
    "selected"
]

for obj in required_objects:
    if obj not in globals():
        raise RuntimeError(
            f"Missing {obj}. Run Big Cells 6 and 7 first."
        )

work = ranking_candidate_v2.copy()
archive = all_rankings.copy()
selected_snapshots = selected.copy()

# ------------------------------------------------
# 1. Prepare dates and IDs
# ------------------------------------------------

work["event_date"] = pd.to_datetime(
    work["event_date"]
).dt.normalize()

archive["snapshot_date"] = pd.to_datetime(
    archive["snapshot_date"]
).dt.normalize()

selected_snapshots["snapshot_date"] = pd.to_datetime(
    selected_snapshots["snapshot_date"]
).dt.normalize()

if len(work) != 358:
    raise ValueError(
        f"Expected 358 fighter appearances; found {len(work)}."
    )

if work["appearance_id"].duplicated().any():
    raise ValueError("Duplicate appearance IDs detected.")

# ------------------------------------------------
# 2. Evaluate complete selected snapshots
# ------------------------------------------------

archive["rank"] = pd.to_numeric(
    archive["rank"], errors="coerce"
)

archive = archive[
    archive["rank"].between(0, 15)
    & archive["snapshot_date"].notna()
].copy()

archive["rank"] = archive["rank"].astype(int)

snapshot_keys = [
    "source_type",
    "snapshot_date",
    "division_key"
]

selected_keys = selected_snapshots[
    snapshot_keys
].drop_duplicates()

selected_archive = archive.merge(
    selected_keys,
    on=snapshot_keys,
    how="inner"
)

snapshot_rows = []

for key, group in selected_archive.groupby(
    snapshot_keys
):
    source_type, date, division = key

    numbered = group[
        group["rank"].between(1, 15)
    ]

    present = set(numbered["rank"].tolist())
    missing = sorted(set(range(1, 16)) - present)

    duplicate_ranks = (
        numbered.groupby("rank")["fighter_key"]
        .nunique()
    )

    duplicate_fighters = (
        group.groupby("fighter_key")["rank"]
        .nunique()
    )

    snapshot_rows.append({
        "source_type": source_type,
        "snapshot_date": date,
        "division_key": division,
        "numbered_ranks_present": len(present),
        "missing_numbered_ranks": ",".join(
            map(str, missing)
        ),
        "all_15_numbered_ranks": len(missing) == 0,
        "champion_present": bool(
            (group["rank"] == 0).any()
        ),
        "duplicate_rank_positions": int(
            (duplicate_ranks > 1).sum()
        ),
        "fighter_rank_conflicts": int(
            (duplicate_fighters > 1).sum()
        ),
        "unique_ranked_fighters": (
            group["fighter_key"].nunique()
        )
    })

snapshot_audit = pd.DataFrame(snapshot_rows)

snapshot_audit["strictly_complete"] = (
    snapshot_audit["all_15_numbered_ranks"]
    & snapshot_audit["duplicate_rank_positions"].eq(0)
    & snapshot_audit["fighter_rank_conflicts"].eq(0)
)

# ------------------------------------------------
# 3. Attach snapshot quality to each appearance
# ------------------------------------------------

snapshot_assignment = selected_snapshots[
    [
        "event_name",
        "event_date",
        "division_key",
        "source_type",
        "snapshot_date",
        "age_days"
    ]
].drop_duplicates(
    ["event_name", "event_date", "division_key"]
)

snapshot_assignment = snapshot_assignment.merge(
    snapshot_audit,
    on=snapshot_keys,
    how="left",
    validate="many_to_one"
)

# Avoid overwriting columns already in work.
quality_columns = [
    "event_name",
    "event_date",
    "division_key",
    "source_type",
    "snapshot_date",
    "age_days",
    "numbered_ranks_present",
    "missing_numbered_ranks",
    "all_15_numbered_ranks",
    "strictly_complete"
]

decisions = work.copy()

decisions["fight_division_key"] = (
    decisions["fight_division_clean"]
)

quality_lookup = snapshot_assignment[
    quality_columns
].rename(columns={
    "division_key": "fight_division_key",
    "source_type": "audit_source",
    "snapshot_date": "audit_snapshot_date",
    "age_days": "audit_snapshot_age",
    "numbered_ranks_present": "audit_numbered_ranks",
    "missing_numbered_ranks": "audit_missing_ranks",
    "all_15_numbered_ranks": "audit_all_15",
    "strictly_complete": "audit_strictly_complete"
})

decisions = decisions.merge(
    quality_lookup,
    on=[
        "event_name",
        "event_date",
        "fight_division_key"
    ],
    how="left",
    validate="many_to_one"
)

# ------------------------------------------------
# 4. Exclude three misleading fuzzy matches
# ------------------------------------------------

false_aliases = {
    "alice pereira",
    "michael oliveira",
    "imanol rodriguez"
}

decisions["false_alias_rejected"] = (
    decisions["fighter_key"].isin(false_aliases)
)

# ------------------------------------------------
# 5. Build conservative decision categories
# ------------------------------------------------

has_rank = decisions["proposed_rank"].notna()

same_div = (
    decisions["candidate_type"] == "SAME_DIVISION"
)

cross_div = (
    decisions["candidate_type"] == "CROSS_DIVISION"
)

strict_complete = (
    decisions["audit_strictly_complete"]
    .fillna(False)
    .astype(bool)
)

catchweight = (
    decisions["fight_division_key"] == "catchweight"
)

decisions["decision_category"] = np.select(
    [
        same_div & has_rank,
        cross_div & has_rank,
        catchweight,
        ~has_rank & strict_complete,
        ~has_rank & ~strict_complete
    ],
    [
        "RANKED_SAME_DIVISION_CANDIDATE",
        "RANKED_OTHER_DIVISION_REVIEW",
        "CATCHWEIGHT_REVIEW",
        "ABSENT_FROM_COMPLETE_SNAPSHOT",
        "ABSENT_BUT_SNAPSHOT_INCOMPLETE"
    ],
    default="OTHER_REVIEW"
)

# Absence from a complete snapshot supports
# an unranked candidate, but it is not the same
# as independent verification.
decisions["unranked_candidate"] = (
    decisions["decision_category"]
    == "ABSENT_FROM_COMPLETE_SNAPSHOT"
)

decisions["final_ranking_verified"] = False

decisions["final_ranking_points"] = np.nan

# ------------------------------------------------
# 6. Snapshot-level results
# ------------------------------------------------

print("=" * 78)
print("UFC EVENT STRENGTH INDEX V2")
print("STRICT RANKINGS SNAPSHOT COMPLETENESS")
print("=" * 78)

print("\nUnique selected divisional snapshots:",
      len(snapshot_audit))

print("Snapshots containing all ranks 1–15:",
      int(snapshot_audit["all_15_numbered_ranks"].sum()))

print("Strictly complete snapshots:",
      int(snapshot_audit["strictly_complete"].sum()))

print("\nSNAPSHOT COMPLETENESS EXAMPLES:")
display(
    snapshot_audit.sort_values(
        ["strictly_complete", "snapshot_date"],
        ascending=[True, False]
    ).head(30)
)

# ------------------------------------------------
# 7. Fighter-level decision results
# ------------------------------------------------

print("\n" + "=" * 78)
print("358-FIGHTER DECISION CATEGORIES")
print("=" * 78)

print(
    decisions["decision_category"]
    .value_counts(dropna=False)
    .to_string()
)

print("\nUNRANKED CANDIDATE COUNT:",
      int(decisions["unranked_candidate"].sum()))

print("\nMAIN EVENT FIGHTERS:")
display(
    decisions[
        decisions["is_main_event"].eq(True)
    ][
        [
            "event_date",
            "fighter",
            "proposed_rank",
            "decision_category",
            "audit_snapshot_date",
            "audit_numbered_ranks",
            "audit_strictly_complete"
        ]
    ].sort_values(["event_date", "fighter"])
)

# ------------------------------------------------
# 8. Event-level decision coverage
# ------------------------------------------------

event_decisions = (
    decisions.groupby(
        ["event_name", "event_date"],
        as_index=False
    )
    .agg(
        total_appearances=("fighter", "size"),
        ranked_same_division=(
            "decision_category",
            lambda x: (
                x == "RANKED_SAME_DIVISION_CANDIDATE"
            ).sum()
        ),
        ranked_other_division=(
            "decision_category",
            lambda x: (
                x == "RANKED_OTHER_DIVISION_REVIEW"
            ).sum()
        ),
        absent_complete=(
            "decision_category",
            lambda x: (
                x == "ABSENT_FROM_COMPLETE_SNAPSHOT"
            ).sum()
        ),
        absent_incomplete=(
            "decision_category",
            lambda x: (
                x == "ABSENT_BUT_SNAPSHOT_INCOMPLETE"
            ).sum()
        ),
        catchweight_review=(
            "decision_category",
            lambda x: (
                x == "CATCHWEIGHT_REVIEW"
            ).sum()
        )
    )
    .sort_values("event_date")
)

event_decisions["review_remaining"] = (
    event_decisions["ranked_other_division"]
    + event_decisions["absent_incomplete"]
    + event_decisions["catchweight_review"]
)

print("\nEVENT-BY-EVENT DECISION COVERAGE:")
display(event_decisions)

# ------------------------------------------------
# 9. Save outputs
# ------------------------------------------------

snapshot_audit.to_csv(
    OUT / "strict_snapshot_completeness.csv",
    index=False
)

decisions.to_csv(
    OUT / "ranking_decisions_358.csv",
    index=False
)

event_decisions.to_csv(
    OUT / "ranking_decisions_by_event.csv",
    index=False
)

ranking_decisions_v4 = decisions.copy()
ranking_snapshot_completeness_v4 = snapshot_audit.copy()
ranking_event_decisions_v4 = event_decisions.copy()

print("\n" + "=" * 78)
print("BIG CELL 8 COMPLETE")
print("=" * 78)
print("Files saved:", OUT)
print("No rankings automatically verified.")
print("No unranked fighter automatically assigned zero.")
print("No event scores changed.")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 9 — SOURCE RELIABILITY & FINAL EXCEPTIONS AUDIT
#
# 1. Validate all 358 candidate classifications
# 2. Check snapshot dates and age
# 3. Compare historical and meta sources where overlapping
# 4. Examine the 8 cross-division candidates
# 5. Examine the 3 catchweight cases
# 6. Flag champions and ranking conflicts
# 7. Export evidence for final decisions
#
# No scoring changes. No automatic approval.
# ============================================================

OUT = Path("/content/ufc_rankings_recovery")
OUT.mkdir(parents=True, exist_ok=True)

for obj in [
    "ranking_decisions_v4",
    "ranking_snapshot_completeness_v4",
    "all_rankings",
    "ranking_cross_division"
]:
    if obj not in globals():
        raise RuntimeError(
            f"Missing {obj}. Run Big Cells 2–8 first."
        )

decisions = ranking_decisions_v4.copy()
snapshots = ranking_snapshot_completeness_v4.copy()
archive = all_rankings.copy()
cross = ranking_cross_division.copy()

# ------------------------------------------------------------
# 1. Basic consistency checks
# ------------------------------------------------------------

decisions["event_date"] = pd.to_datetime(
    decisions["event_date"], errors="coerce"
).dt.normalize()

decisions["audit_snapshot_date"] = pd.to_datetime(
    decisions["audit_snapshot_date"], errors="coerce"
).dt.normalize()

decisions["proposed_rank"] = pd.to_numeric(
    decisions["proposed_rank"], errors="coerce"
)

assert len(decisions) == 358, (
    f"Expected 358 appearances, found {len(decisions)}."
)

assert decisions["appearance_id"].is_unique, (
    "Duplicate appearance IDs."
)

assert decisions["event_date"].notna().all(), (
    "Some event dates are missing."
)

assert decisions["decision_category"].notna().all(), (
    "Some ranking categories are missing."
)

# ------------------------------------------------------------
# 2. Source-date validation
# ------------------------------------------------------------

decisions["snapshot_days_before_event"] = (
    decisions["event_date"]
    - decisions["audit_snapshot_date"]
).dt.days

decisions["snapshot_date_valid"] = (
    decisions["snapshot_days_before_event"]
    .between(1, 14)
)

# Catchweight fights have no fight-division snapshot.
decisions["requires_division_snapshot"] = (
    decisions["fight_division_key"].ne("catchweight")
)

date_errors = decisions[
    decisions["requires_division_snapshot"]
    & ~decisions["snapshot_date_valid"]
].copy()

print("=" * 78)
print("UFC EVENT STRENGTH INDEX V2")
print("SOURCE RELIABILITY & EXCEPTIONS AUDIT")
print("=" * 78)

print("\nFIGHTER APPEARANCES:", len(decisions))
print("DIVISIONAL SNAPSHOTS:", len(snapshots))
print("INVALID OR MISSING PRE-EVENT SNAPSHOTS:",
      len(date_errors))

print("\nSNAPSHOT AGE DISTRIBUTION:")
display(
    decisions[
        decisions["requires_division_snapshot"]
    ]["snapshot_days_before_event"]
    .value_counts(dropna=False)
    .sort_index()
    .rename_axis("days_before_event")
    .reset_index(name="appearances")
)

# ------------------------------------------------------------
# 3. Archive normalization
# ------------------------------------------------------------

archive["snapshot_date"] = pd.to_datetime(
    archive["snapshot_date"], errors="coerce"
).dt.normalize()

archive["rank"] = pd.to_numeric(
    archive["rank"], errors="coerce"
)

archive = archive[
    archive["snapshot_date"].notna()
    & archive["rank"].between(0, 15)
    & archive["fighter_key"].notna()
    & archive["division_key"].notna()
].copy()

archive["rank"] = archive["rank"].astype(int)

# ------------------------------------------------------------
# 4. Compare sources on the same date
# ------------------------------------------------------------

source_comparison = (
    archive.groupby(
        [
            "snapshot_date",
            "division_key",
            "fighter_key"
        ],
        as_index=False
    )
    .agg(
        source_count=("source_type", "nunique"),
        distinct_ranks=("rank", "nunique"),
        ranks=("rank", lambda x:
               ", ".join(map(str, sorted(set(x))))),
        sources=("source_type", lambda x:
                 ", ".join(sorted(set(x))))
    )
)

overlaps = source_comparison[
    source_comparison["source_count"] >= 2
].copy()

conflicts = overlaps[
    overlaps["distinct_ranks"] > 1
].copy()

print("\nCROSS-SOURCE AGREEMENT:")
print("Same-date fighter overlaps:", len(overlaps))
print("Conflicting ranks:", len(conflicts))

if len(overlaps):
    print("Agreement rate:",
          round(
              100 * (overlaps["distinct_ranks"] == 1).mean(),
              2
          ), "%")
else:
    print(
        "No same-date overlaps found. "
        "Independent agreement cannot be established."
    )

print("\nSOURCE CONFLICT EXAMPLES:")
display(conflicts.head(30))

# ------------------------------------------------------------
# 5. Check for champion/rank-zero anomalies
# ------------------------------------------------------------

champions = archive[
    archive["rank"] == 0
].copy()

champion_counts = (
    champions.groupby(
        [
            "source_type",
            "snapshot_date",
            "division_key"
        ],
        as_index=False
    )
    .agg(
        champion_count=("fighter_key", "nunique"),
        champions=("fighter", lambda x:
                   ", ".join(sorted(set(map(str, x)))))
    )
)

multiple_champions = champion_counts[
    champion_counts["champion_count"] > 1
].copy()

print("\nCHAMPION AUDIT:")
print("Snapshots with multiple rank-zero fighters:",
      len(multiple_champions))

display(multiple_champions.head(25))

# ------------------------------------------------------------
# 6. Inspect cross-division cases
# ------------------------------------------------------------

cross_review = decisions[
    decisions["decision_category"].eq(
        "RANKED_OTHER_DIVISION_REVIEW"
    )
].copy()

cross["other_division_snapshot"] = pd.to_datetime(
    cross["other_division_snapshot"],
    errors="coerce"
).dt.normalize()

cross_details = cross[
    [
        "appearance_id",
        "ranked_division",
        "other_division_rank",
        "other_division_snapshot",
        "other_division_source"
    ]
].drop_duplicates()

cross_review = cross_review.merge(
    cross_details,
    on="appearance_id",
    how="left",
    validate="one_to_many"
)

cross_review["other_snapshot_age"] = (
    cross_review["event_date"]
    - cross_review["other_division_snapshot"]
).dt.days

cross_review["other_date_valid"] = (
    cross_review["other_snapshot_age"]
    .between(1, 14)
)

cross_review["candidate_points"] = np.where(
    pd.to_numeric(
        cross_review["other_division_rank"],
        errors="coerce"
    ).between(0, 15),
    16 - pd.to_numeric(
        cross_review["other_division_rank"],
        errors="coerce"
    ),
    np.nan
)

print("\nCROSS-DIVISION CANDIDATES:")
display(
    cross_review[
        [
            "event_date",
            "fighter",
            "fight_division_key",
            "ranked_division",
            "other_division_rank",
            "candidate_points",
            "other_division_snapshot",
            "other_snapshot_age",
            "other_date_valid"
        ]
    ].sort_values(["event_date", "fighter"])
)

# ------------------------------------------------------------
# 7. Catchweight exceptions
# ------------------------------------------------------------

catch_review = decisions[
    decisions["decision_category"].eq(
        "CATCHWEIGHT_REVIEW"
    )
].copy()

print("\nCATCHWEIGHT CASES:")
display(
    catch_review[
        [
            "event_date",
            "event_name",
            "fighter",
            "weightclass",
            "proposed_rank",
            "decision_category"
        ]
    ].sort_values(["event_date", "fighter"])
)

# ------------------------------------------------------------
# 8. Special-case decision list
# ------------------------------------------------------------

special = pd.concat(
    [
        cross_review.assign(
            exception_type="CROSS_DIVISION"
        ),
        catch_review.assign(
            exception_type="CATCHWEIGHT"
        )
    ],
    ignore_index=True
)

# Ensure only one row per fighter appearance in the
# final exception list, even if multiple sources exist.
special = special.drop_duplicates(
    subset=["appearance_id"]
)

special["manual_decision"] = "PENDING"
special["manual_notes"] = ""
special["final_approved_points"] = np.nan

# ------------------------------------------------------------
# 9. Summarize readiness by event
# ------------------------------------------------------------

event_status = (
    decisions.groupby(
        ["event_name", "event_date"],
        as_index=False
    )
    .agg(
        fighter_appearances=("fighter", "size"),
        ranked_same_division=(
            "decision_category",
            lambda x: (
                x == "RANKED_SAME_DIVISION_CANDIDATE"
            ).sum()
        ),
        unranked_candidates=(
            "decision_category",
            lambda x: (
                x == "ABSENT_FROM_COMPLETE_SNAPSHOT"
            ).sum()
        ),
        cross_division_review=(
            "decision_category",
            lambda x: (
                x == "RANKED_OTHER_DIVISION_REVIEW"
            ).sum()
        ),
        catchweight_review=(
            "decision_category",
            lambda x: (
                x == "CATCHWEIGHT_REVIEW"
            ).sum()
        )
    )
    .sort_values("event_date")
)

event_status["special_cases"] = (
    event_status["cross_division_review"]
    + event_status["catchweight_review"]
)

event_status["structurally_classified"] = (
    event_status["fighter_appearances"]
    - event_status["special_cases"]
)

event_status["needs_special_case_review"] = (
    event_status["special_cases"] > 0
)

print("\nEVENT READINESS SUMMARY:")
display(event_status)

print("\n" + "=" * 78)
print("OVERALL STATUS")
print("=" * 78)

print(
    "Same-division ranking candidates:",
    int(
        decisions["decision_category"]
        .eq("RANKED_SAME_DIVISION_CANDIDATE")
        .sum()
    )
)

print(
    "Absent from complete snapshots:",
    int(
        decisions["decision_category"]
        .eq("ABSENT_FROM_COMPLETE_SNAPSHOT")
        .sum()
    )
)

print("Cross-division exceptions:", len(
    decisions[
        decisions["decision_category"].eq(
            "RANKED_OTHER_DIVISION_REVIEW"
        )
    ]
))

print("Catchweight exceptions:", len(catch_review))

print("Events needing special-case review:",
      int(event_status["needs_special_case_review"].sum()))

# ------------------------------------------------------------
# 10. Export audit files
# ------------------------------------------------------------

overlaps.to_csv(
    OUT / "cross_source_rank_agreements.csv",
    index=False
)

conflicts.to_csv(
    OUT / "cross_source_rank_conflicts.csv",
    index=False
)

multiple_champions.to_csv(
    OUT / "multiple_champion_anomalies.csv",
    index=False
)

cross_review.to_csv(
    OUT / "cross_division_final_review.csv",
    index=False
)

catch_review.to_csv(
    OUT / "catchweight_final_review.csv",
    index=False
)

special.to_csv(
    OUT / "special_case_manual_decisions.csv",
    index=False
)

event_status.to_csv(
    OUT / "final_ranking_readiness_by_event.csv",
    index=False
)

date_errors.to_csv(
    OUT / "invalid_snapshot_dates.csv",
    index=False
)

# Save in Colab memory
ranking_source_agreement_v5 = overlaps.copy()
ranking_source_conflicts_v5 = conflicts.copy()
ranking_special_cases_v5 = special.copy()
ranking_readiness_v5 = event_status.copy()

print("\n" + "=" * 78)
print("BIG CELL 9 COMPLETE")
print("=" * 78)
print("Audit files saved to:", OUT)
print("No final rankings approved automatically.")
print("No missing data assigned zero.")
print("No event strength scores modified.")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ================================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 10 — CROSS-DIVISION RANKINGS & EVENT TOTALS
#
# Rule: A fighter keeps their highest available pre-event
# UFC ranking even when competing in another division.
#
# Ranked champion = 16 points
# Ranked #1 = 15 points ... Ranked #15 = 1 point
# Archive-supported unranked = 0 points
#
# Three unresolved catchweight cases stay missing.
# Does not overwrite historical event scores.
# ================================================================

OUT = Path("/content/ufc_rankings_recovery")
OUT.mkdir(parents=True, exist_ok=True)

assert "ranking_decisions_v4" in globals(), \
    "Run Big Cell 8 first."

assert "ranking_special_cases_v5" in globals(), \
    "Run Big Cell 9 first."

assert "recent_event_raw" in globals(), \
    "Restore your project checkpoint first."

work = ranking_decisions_v4.copy()

# ------------------------------------------------
# 1. Validate the 358 appearances
# ------------------------------------------------

assert len(work) == 358, (
    f"Expected 358 appearances, found {len(work)}"
)

assert work["appearance_id"].is_unique, \
    "Duplicate fighter appearances found."

work["event_date"] = pd.to_datetime(
    work["event_date"]
).dt.normalize()

work["proposed_rank"] = pd.to_numeric(
    work["proposed_rank"], errors="coerce"
)

valid_categories = {
    "RANKED_SAME_DIVISION_CANDIDATE",
    "RANKED_OTHER_DIVISION_REVIEW",
    "ABSENT_FROM_COMPLETE_SNAPSHOT",
    "CATCHWEIGHT_REVIEW"
}

unexpected = set(work["decision_category"]) - valid_categories

if unexpected:
    raise ValueError(
        f"Unexpected decision categories: {unexpected}"
    )

# ------------------------------------------------
# 2. Apply the approved ranking methodology
# ------------------------------------------------

work["ranking_method"] = np.select(
    [
        work["decision_category"].eq(
            "RANKED_SAME_DIVISION_CANDIDATE"
        ),
        work["decision_category"].eq(
            "RANKED_OTHER_DIVISION_REVIEW"
        ),
        work["decision_category"].eq(
            "ABSENT_FROM_COMPLETE_SNAPSHOT"
        ),
        work["decision_category"].eq(
            "CATCHWEIGHT_REVIEW"
        )
    ],
    [
        "RANKED_SAME_DIVISION",
        "RANKED_CROSS_DIVISION",
        "UNRANKED_ARCHIVE_SUPPORTED",
        "UNRESOLVED_CATCHWEIGHT"
    ],
    default="UNRESOLVED"
)

ranked_mask = work["ranking_method"].isin([
    "RANKED_SAME_DIVISION",
    "RANKED_CROSS_DIVISION"
])

unranked_mask = work["ranking_method"].eq(
    "UNRANKED_ARCHIVE_SUPPORTED"
)

work["ranking_points_v2"] = np.nan

work.loc[ranked_mask, "ranking_points_v2"] = (
    16 - work.loc[ranked_mask, "proposed_rank"]
)

work.loc[unranked_mask, "ranking_points_v2"] = 0.0

# Validate all ranked values
invalid_rank = (
    ranked_mask
    & ~work["proposed_rank"].between(0, 15)
)

if invalid_rank.any():
    raise ValueError(
        "Some ranked fighters have invalid rankings."
    )

# ------------------------------------------------
# 3. Check historical consistency
# ------------------------------------------------

historical = backup["event_master"].copy()

print("=" * 78)
print("UFC EVENT STRENGTH INDEX V2")
print("CROSS-DIVISION RANKING METHODOLOGY")
print("=" * 78)

print("\nRanking points:")
print("Champion: 16")
print("#1: 15")
print("#15: 1")
print("Unranked: 0")
print("Cross-division ranked fighters retain their points.")

print("\nHistorical events preserved:", len(historical))

# ------------------------------------------------
# 4. Fighter-level results
# ------------------------------------------------

print("\nFIGHTER-LEVEL RANKING RESULTS:")
print(
    work["ranking_method"]
    .value_counts()
    .to_string()
)

print("\nCROSS-DIVISION RANKING POINTS:")
display(
    work[
        work["ranking_method"].eq(
            "RANKED_CROSS_DIVISION"
        )
    ][
        [
            "event_date",
            "fighter",
            "fight_division_key",
            "cross_division_name",
            "proposed_rank",
            "ranking_points_v2"
        ]
    ].sort_values(["event_date", "fighter"])
)

# ------------------------------------------------
# 5. Calculate event ranking totals
# ------------------------------------------------

event_totals = (
    work.groupby(
        ["event_name", "event_date"],
        as_index=False
    )
    .agg(
        fighter_appearances=("fighter", "size"),
        ranked_same_division=(
            "ranking_method",
            lambda x: (x == "RANKED_SAME_DIVISION").sum()
        ),
        ranked_cross_division=(
            "ranking_method",
            lambda x: (x == "RANKED_CROSS_DIVISION").sum()
        ),
        archive_supported_unranked=(
            "ranking_method",
            lambda x: (
                x == "UNRANKED_ARCHIVE_SUPPORTED"
            ).sum()
        ),
        unresolved_fighters=(
            "ranking_points_v2",
            lambda x: x.isna().sum()
        ),
        known_ranking_points=(
            "ranking_points_v2", "sum"
        )
    )
    .sort_values("event_date")
)

event_totals["ranking_data_complete"] = (
    event_totals["unresolved_fighters"] == 0
)

# Important: do not call partial sums final totals.
event_totals["total_ranking_points_v2"] = np.where(
    event_totals["ranking_data_complete"],
    event_totals["known_ranking_points"],
    np.nan
)

event_totals["ranking_status"] = np.where(
    event_totals["ranking_data_complete"],
    "COMPLETE_ARCHIVE_BASED",
    "INCOMPLETE_CATCHWEIGHT"
)

# ------------------------------------------------
# 6. Display event totals
# ------------------------------------------------

print("\n" + "=" * 78)
print("RANKING TOTALS FOR ALL 14 RECENT EVENTS")
print("=" * 78)

display(
    event_totals[
        [
            "event_date",
            "event_name",
            "fighter_appearances",
            "ranked_same_division",
            "ranked_cross_division",
            "archive_supported_unranked",
            "unresolved_fighters",
            "known_ranking_points",
            "total_ranking_points_v2",
            "ranking_status"
        ]
    ]
)

# ------------------------------------------------
# 7. Identify unresolved catchweight fighters
# ------------------------------------------------

unresolved = work[
    work["ranking_points_v2"].isna()
].copy()

print("\nUNRESOLVED FIGHTERS:")
display(
    unresolved[
        [
            "event_date",
            "event_name",
            "fighter",
            "weightclass",
            "ranking_method"
        ]
    ].sort_values(["event_date", "fighter"])
)

# ------------------------------------------------
# 8. Compare with historical ranking distribution
# ------------------------------------------------

historical_points = pd.to_numeric(
    historical["total_ranking_points"],
    errors="coerce"
).dropna()

print("\nHISTORICAL RANKING POINT DISTRIBUTION:")
print(historical_points.describe())

print("\nRECENT COMPLETE EVENT RANKING DISTRIBUTION:")
print(
    event_totals.loc[
        event_totals["ranking_data_complete"],
        "total_ranking_points_v2"
    ].describe()
)

# ------------------------------------------------
# 9. Save outputs
# ------------------------------------------------

work.to_csv(
    OUT / "recent_fighter_ranking_points_v2.csv",
    index=False
)

event_totals.to_csv(
    OUT / "recent_event_ranking_totals_v2.csv",
    index=False
)

unresolved.to_csv(
    OUT / "remaining_catchweight_ranking_cases.csv",
    index=False
)

ranking_points_approved_v2 = work.copy()
ranking_event_totals_v2 = event_totals.copy()
ranking_remaining_v2 = unresolved.copy()

print("\n" + "=" * 78)
print("BIG CELL 10 COMPLETE")
print("=" * 78)

print("Fighter appearances:", len(work))
print("Events:", len(event_totals))
print(
    "Events with complete ranking totals:",
    int(event_totals["ranking_data_complete"].sum())
)
print(
    "Events still missing ranking information:",
    int((~event_totals["ranking_data_complete"]).sum())
)

print("Files saved to:", OUT)
print("Historical scores unchanged.")
print("Recent final scores unchanged.")
```

```python

import pandas as pd
import numpy as np
import unicodedata
import re
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 11 — FINAL CATCHWEIGHT RANKINGS RECONSTRUCTION
#
# Search ALL pre-event ranked divisions for:
# Charles Johnson, Eduardo Chapolin, Edgar Chairez.
#
# Do not assume the catchweight bout division.
# Retain cross-division ranking value.
# Zero only when absent from available complete
# pre-event divisional snapshots.
#
# No final event-strength scores are changed.
# ============================================================

OUT = Path("/content/ufc_rankings_recovery")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "ranking_points_approved_v2",
    "all_rankings",
    "selected",
    "ranking_snapshot_completeness_v4"
]

for obj in required:
    if obj not in globals():
        raise RuntimeError(f"Missing {obj}; restore earlier cells.")

work = ranking_points_approved_v2.copy()
archive = all_rankings.copy()
selected_work = selected.copy()
quality = ranking_snapshot_completeness_v4.copy()

def clean_name(value):
    if pd.isna(value):
        return ""
    s = unicodedata.normalize("NFKD", str(value))
    s = "".join(
        c for c in s if not unicodedata.combining(c)
    )
    return " ".join(
        re.sub(r"[^a-z0-9]+", " ", s.lower()).split()
    )

work["event_date"] = pd.to_datetime(
    work["event_date"]
).dt.normalize()

archive["snapshot_date"] = pd.to_datetime(
    archive["snapshot_date"]
).dt.normalize()

selected_work["snapshot_date"] = pd.to_datetime(
    selected_work["snapshot_date"]
).dt.normalize()

quality["snapshot_date"] = pd.to_datetime(
    quality["snapshot_date"]
).dt.normalize()

archive["fighter_key"] = archive["fighter"].map(clean_name)
archive["rank"] = pd.to_numeric(
    archive["rank"], errors="coerce"
)

work["fighter_key_check"] = work["fighter"].map(clean_name)

# ------------------------------------------------------------
# 1. Find the three unresolved fighter appearances
# ------------------------------------------------------------

unresolved = work[
    work["ranking_points_v2"].isna()
].copy()

print("=" * 76)
print("FINAL CATCHWEIGHT RANKINGS INVESTIGATION")
print("=" * 76)

print("\nUnresolved fighter appearances:", len(unresolved))

display(
    unresolved[
        [
            "event_date",
            "fighter",
            "event_name",
            "weightclass"
        ]
    ]
)

# ------------------------------------------------------------
# 2. Select only complete pre-event snapshots
# ------------------------------------------------------------

complete = quality[
    quality["strictly_complete"].eq(True)
][
    ["source_type", "snapshot_date", "division_key"]
].drop_duplicates()

eligible = selected_work.merge(
    complete,
    on=["source_type", "snapshot_date", "division_key"],
    how="inner"
)

eligible["event_date"] = pd.to_datetime(
    eligible["event_date"]
).dt.normalize()

eligible["snapshot_age"] = (
    eligible["event_date"] - eligible["snapshot_date"]
).dt.days

eligible = eligible[
    eligible["snapshot_age"].between(1, 14)
].copy()

# ------------------------------------------------------------
# 3. Search every selected division for each fighter
# ------------------------------------------------------------

all_matches = []
coverage_rows = []
resolved_rows = []

for _, fighter in unresolved.iterrows():

    event_name = fighter["event_name"]
    event_date = fighter["event_date"]
    name_key = fighter["fighter_key_check"]

    event_sources = eligible[
        eligible["event_name"].eq(event_name)
        & eligible["event_date"].eq(event_date)
    ].copy()

    event_roster = archive.merge(
        event_sources[
            ["source_type", "snapshot_date", "division_key"]
        ].drop_duplicates(),
        on=["source_type", "snapshot_date", "division_key"],
        how="inner"
    )

    matches = event_roster[
        event_roster["fighter_key"].eq(name_key)
        & event_roster["rank"].between(0, 15)
    ].copy()

    for _, match in matches.iterrows():
        all_matches.append({
            "appearance_id": fighter["appearance_id"],
            "event_date": event_date,
            "fighter": fighter["fighter"],
            "ranked_division": match["division_key"],
            "rank": int(match["rank"]),
            "ranking_points": 16 - int(match["rank"]),
            "snapshot_date": match["snapshot_date"],
            "source_type": match["source_type"]
        })

    divisions_checked = sorted(
        event_sources["division_key"].dropna().unique()
    )

    coverage_rows.append({
        "appearance_id": fighter["appearance_id"],
        "fighter": fighter["fighter"],
        "event_name": event_name,
        "event_date": event_date,
        "complete_divisions_checked": len(divisions_checked),
        "divisions": ", ".join(divisions_checked),
        "exact_rank_matches": len(matches)
    })

    if not matches.empty:
        # Highest-ranked position means most points.
        best = matches.sort_values(
            ["rank", "snapshot_date"],
            ascending=[True, False]
        ).iloc[0]

        resolved_rows.append({
            "appearance_id": fighter["appearance_id"],
            "fighter": fighter["fighter"],
            "rank": int(best["rank"]),
            "points": float(16 - best["rank"]),
            "method": "RANKED_ANY_DIVISION",
            "division": best["division_key"],
            "snapshot_date": best["snapshot_date"]
        })

    elif len(divisions_checked) >= 10:
        # An archive-supported zero, not independent
        # confirmation that the fighter was unranked.
        resolved_rows.append({
            "appearance_id": fighter["appearance_id"],
            "fighter": fighter["fighter"],
            "rank": np.nan,
            "points": 0.0,
            "method": "UNRANKED_ALL_AVAILABLE_SNAPSHOTS",
            "division": "",
            "snapshot_date": pd.NaT
        })

    else:
        resolved_rows.append({
            "appearance_id": fighter["appearance_id"],
            "fighter": fighter["fighter"],
            "rank": np.nan,
            "points": np.nan,
            "method": "INSUFFICIENT_DIVISION_COVERAGE",
            "division": "",
            "snapshot_date": pd.NaT
        })

coverage = pd.DataFrame(coverage_rows)
resolved = pd.DataFrame(resolved_rows)
matches_df = pd.DataFrame(all_matches)

print("\nALL-DIVISION SOURCE COVERAGE:")
display(coverage)

print("\nRANKED MATCHES FOUND:")
if len(matches_df):
    display(matches_df)
else:
    print("No exact-name ranked matches found.")

print("\nPROPOSED CATCHWEIGHT RESOLUTIONS:")
display(resolved)

# ------------------------------------------------------------
# 4. Update only the three unresolved appearances
# ------------------------------------------------------------

if len(resolved):
    assert resolved["appearance_id"].is_unique

    point_lookup = resolved.set_index(
        "appearance_id"
    )["points"]

    method_lookup = resolved.set_index(
        "appearance_id"
    )["method"]

    mask = work["appearance_id"].isin(
        resolved["appearance_id"]
    )

    work.loc[mask, "ranking_points_v2"] = (
        work.loc[mask, "appearance_id"].map(point_lookup)
        .to_numpy()
    )

    work.loc[mask, "ranking_method"] = (
        work.loc[mask, "appearance_id"].map(method_lookup)
        .to_numpy()
    )

# ------------------------------------------------------------
# 5. Recalculate all 14 event totals
# ------------------------------------------------------------

final_totals = (
    work.groupby(
        ["event_name", "event_date"],
        as_index=False
    )
    .agg(
        appearances=("fighter", "size"),
        unresolved=(
            "ranking_points_v2",
            lambda s: s.isna().sum()
        ),
        ranking_points_known=(
            "ranking_points_v2", "sum"
        )
    )
    .sort_values("event_date")
)

final_totals["ranking_data_complete"] = (
    final_totals["unresolved"].eq(0)
)

final_totals["total_ranking_points_v2"] = np.where(
    final_totals["ranking_data_complete"],
    final_totals["ranking_points_known"],
    np.nan
)

final_totals["status"] = np.where(
    final_totals["ranking_data_complete"],
    "COMPLETE_ARCHIVE_BASED",
    "STILL_INCOMPLETE"
)

assert len(final_totals) == 14
assert len(work) == 358

print("\n" + "=" * 76)
print("UPDATED RANKING TOTALS — ALL 14 EVENTS")
print("=" * 76)

display(
    final_totals[
        [
            "event_date",
            "event_name",
            "appearances",
            "unresolved",
            "total_ranking_points_v2",
            "status"
        ]
    ]
)

print("\nEVENTS WITH COMPLETE ARCHIVE-BASED TOTALS:",
      int(final_totals["ranking_data_complete"].sum()),
      "/ 14")

# ------------------------------------------------------------
# 6. Save results without overwriting prior versions
# ------------------------------------------------------------

coverage.to_csv(
    OUT / "catchweight_all_divisions_coverage.csv",
    index=False
)

resolved.to_csv(
    OUT / "catchweight_resolution_evidence.csv",
    index=False
)

work.to_csv(
    OUT / "recent_fighter_ranking_points_v3.csv",
    index=False
)

final_totals.to_csv(
    OUT / "recent_event_ranking_totals_v3.csv",
    index=False
)

ranking_points_final_candidate_v3 = work.copy()
ranking_event_totals_final_candidate_v3 = final_totals.copy()
ranking_catchweight_evidence_v3 = coverage.copy()
ranking_catchweight_resolutions_v3 = resolved.copy()

print("\nBIG CELL 11 COMPLETE")
print("No historical scores modified.")
print("No final event-strength scores modified.")
print("Archive-based classifications remain documented.")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ================================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 12 — FINAL SCORING DATA READINESS AUDIT
#
# Audits:
# 1. All 14 completed archive-based ranking totals
# 2. Historical and recent factor definitions
# 3. Previous fighter bonus data
# 4. Main-event scoring completeness
# 5. Card Quality and Star Experience factors
# 6. Remaining requirements for 791-event leaderboard
#
# DOES NOT fabricate bonuses, overwrite original data,
# or change any final scores.
# ================================================================

OUT = Path("/content/ufc_final_scoring_audit")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "backup",
    "scoring_fights",
    "main_event_scoring",
    "event_strength_v2",
    "recent_event_raw",
    "ranking_points_final_candidate_v3",
    "ranking_event_totals_final_candidate_v3"
]

missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError(
        "Missing Colab objects: " + ", ".join(missing)
    )

historical_raw = backup["event_master"].copy()
historical_normalized = backup["normalized"].copy()
fights = scoring_fights.copy()
main = main_event_scoring.copy()
strength = event_strength_v2.copy()
recent = recent_event_raw.copy()
ranking_fighters = ranking_points_final_candidate_v3.copy()
ranking_events = ranking_event_totals_final_candidate_v3.copy()

print("=" * 80)
print("UFC EVENT STRENGTH INDEX V2")
print("FINAL SCORING READINESS AUDIT")
print("=" * 80)

# ------------------------------------------------------------
# 1. Dataset inventory
# ------------------------------------------------------------

datasets = {
    "Historical raw event factors": historical_raw,
    "Historical normalized factors": historical_normalized,
    "All scored fights": fights,
    "Main event scoring": main,
    "Event Strength V2": strength,
    "Recent raw event factors": recent,
    "Recent fighter ranking decisions": ranking_fighters,
    "Recent event ranking totals": ranking_events
}

inventory = pd.DataFrame([
    {
        "dataset": name,
        "rows": len(df),
        "columns": len(df.columns),
        "duplicate_columns": int(
            df.columns.duplicated().sum()
        )
    }
    for name, df in datasets.items()
])

display(inventory)

# ------------------------------------------------------------
# 2. Confirm ranking completeness
# ------------------------------------------------------------

ranking_events["event_date"] = pd.to_datetime(
    ranking_events["event_date"],
    errors="coerce"
).dt.normalize()

assert len(ranking_events) == 14
assert len(ranking_fighters) == 358

assert ranking_events["event_name"].is_unique, \
    "Duplicate event names in ranking totals."

assert ranking_events["unresolved"].eq(0).all(), \
    "Some fighter rankings remain unresolved."

assert ranking_events[
    "total_ranking_points_v2"
].notna().all(), "Missing event ranking totals."

ranking_sum = ranking_fighters.groupby(
    "event_name"
)["ranking_points_v2"].sum()

comparison = ranking_events.set_index(
    "event_name"
)["total_ranking_points_v2"].to_frame(
    "event_total"
).join(
    ranking_sum.rename("fighter_sum")
)

comparison["matches"] = np.isclose(
    comparison["event_total"],
    comparison["fighter_sum"],
    atol=1e-9
)

assert comparison["matches"].all(), \
    "Fighter ranking totals do not match events."

print("\nRANKING POINT RECONCILIATION:")
print(
    int(comparison["matches"].sum()),
    "/14 event totals match their fighter sums."
)

# ------------------------------------------------------------
# 3. Discover factor columns
# ------------------------------------------------------------

def factor_catalog(df, dataset_name):
    rows = []

    keywords = [
        "bonus", "rank", "win", "streak",
        "finish", "title", "main",
        "experience", "quality", "star",
        "score", "weight", "fight"
    ]

    for column in df.columns:
        if any(
            word in str(column).lower()
            for word in keywords
        ):
            series = df[column]
            numeric = pd.to_numeric(
                series, errors="coerce"
            )

            rows.append({
                "dataset": dataset_name,
                "column": column,
                "dtype": str(series.dtype),
                "nonmissing": int(series.notna().sum()),
                "missing": int(series.isna().sum()),
                "numeric_values": int(numeric.notna().sum()),
                "minimum": (
                    numeric.min()
                    if numeric.notna().any() else np.nan
                ),
                "maximum": (
                    numeric.max()
                    if numeric.notna().any() else np.nan
                )
            })

    return pd.DataFrame(rows)

catalogs = []

for name, df in datasets.items():
    result = factor_catalog(df, name)
    if not result.empty:
        catalogs.append(result)

factor_inventory = pd.concat(
    catalogs, ignore_index=True
)

print("\n" + "=" * 80)
print("HISTORICAL AND RECENT FACTOR CATALOG")
print("=" * 80)

display(
    factor_inventory[
        [
            "dataset", "column", "nonmissing",
            "missing", "minimum", "maximum"
        ]
    ]
)

# ------------------------------------------------------------
# 4. Bonus data specifically
# ------------------------------------------------------------

bonus_columns = factor_inventory[
    factor_inventory["column"].str.contains(
        "bonus", case=False, na=False
    )
].copy()

print("\n" + "=" * 80)
print("PREVIOUS BONUS DATA AUDIT")
print("=" * 80)

if bonus_columns.empty:
    print("No bonus-related columns discovered.")
else:
    display(bonus_columns)

# Show recent bonus fields and actual values.
recent_bonus_cols = [
    c for c in recent.columns
    if "bonus" in str(c).lower()
]

recent_identifier = [
    c for c in ["event_name", "event_date"]
    if c in recent.columns
]

print("\nRECENT EVENT BONUS FACTORS:")
if recent_bonus_cols:
    display(
        recent[
            recent_identifier + recent_bonus_cols
        ]
    )
else:
    print("No recent bonus columns found.")

# ------------------------------------------------------------
# 5. Main event scoring completeness
# ------------------------------------------------------------

main_cols = [
    c for c in main.columns
    if any(
        word in str(c).lower()
        for word in [
            "bonus", "rank", "win",
            "streak", "title", "experience",
            "score", "complete", "provisional"
        ]
    )
]

main_event_audit = pd.DataFrame([
    {
        "column": c,
        "nonmissing": int(main[c].notna().sum()),
        "missing": int(main[c].isna().sum()),
        "dtype": str(main[c].dtype)
    }
    for c in main_cols
])

print("\n" + "=" * 80)
print("MAIN EVENT STRENGTH COMPLETENESS")
print("=" * 80)

display(main_event_audit)

# ------------------------------------------------------------
# 6. Card Quality / Star Experience readiness
# ------------------------------------------------------------

score_cols = [
    c for c in strength.columns
    if any(
        word in str(c).lower()
        for word in [
            "card_quality",
            "star_experience",
            "main_event",
            "final",
            "overall",
            "score"
        ]
    )
]

score_audit = pd.DataFrame([
    {
        "column": c,
        "nonmissing": int(strength[c].notna().sum()),
        "missing": int(strength[c].isna().sum()),
        "dtype": str(strength[c].dtype)
    }
    for c in score_cols
])

print("\n" + "=" * 80)
print("OVERALL EVENT SCORING READINESS")
print("=" * 80)

display(score_audit)

# ------------------------------------------------------------
# 7. Compare historical vs recent raw factor names
# ------------------------------------------------------------

historical_factors = set(
    map(str, historical_raw.columns)
)

recent_factors = set(
    map(str, recent.columns)
)

shared = sorted(
    historical_factors.intersection(recent_factors)
)

historical_only = sorted(
    historical_factors - recent_factors
)

recent_only = sorted(
    recent_factors - historical_factors
)

print("\n" + "=" * 80)
print("RAW FACTOR COMPATIBILITY")
print("=" * 80)

print("\nSHARED COLUMNS:")
print(shared)

print("\nHISTORICAL-ONLY COLUMNS:")
print(historical_only)

print("\nRECENT-ONLY COLUMNS:")
print(recent_only)

# ------------------------------------------------------------
# 8. Build a 14-event readiness report
# ------------------------------------------------------------

readiness = ranking_events[
    [
        "event_name", "event_date",
        "total_ranking_points_v2"
    ]
].copy()

recent_cols = [
    c for c in recent.columns
    if c not in ["event_name", "event_date"]
]

# Avoid column collisions by adding a prefix.
recent_join = recent[
    ["event_name"] + recent_cols
].copy()

recent_join = recent_join.rename(
    columns={
        c: "raw__" + str(c)
        for c in recent_cols
    }
)

if recent_join["event_name"].duplicated().any():
    raise ValueError(
        "Duplicate recent event rows detected."
    )

readiness = readiness.merge(
    recent_join,
    on="event_name",
    how="left",
    validate="one_to_one"
)

readiness["ranking_factor_complete"] = True

readiness["bonus_factor_present"] = (
    readiness[
        [
            c for c in readiness.columns
            if "bonus" in c.lower()
        ]
    ].notna().any(axis=1)
    if any(
        "bonus" in c.lower()
        for c in readiness.columns
    )
    else False
)

# Presence is NOT the same as historical accuracy.
readiness["bonus_history_verified"] = False

readiness["all_factors_verified"] = False

print("\n" + "=" * 80)
print("RECENT EVENT SCORING READINESS")
print("=" * 80)

display(
    readiness[
        [
            "event_date",
            "event_name",
            "total_ranking_points_v2",
            "ranking_factor_complete",
            "bonus_factor_present",
            "bonus_history_verified",
            "all_factors_verified"
        ]
    ].sort_values("event_date")
)

# ------------------------------------------------------------
# 9. Export audit evidence
# ------------------------------------------------------------

inventory.to_csv(
    OUT / "dataset_inventory.csv",
    index=False
)

factor_inventory.to_csv(
    OUT / "all_scoring_factor_inventory.csv",
    index=False
)

bonus_columns.to_csv(
    OUT / "bonus_data_column_audit.csv",
    index=False
)

main_event_audit.to_csv(
    OUT / "main_event_completeness.csv",
    index=False
)

score_audit.to_csv(
    OUT / "overall_score_completeness.csv",
    index=False
)

readiness.to_csv(
    OUT / "recent_14_event_scoring_readiness.csv",
    index=False
)

comparison.to_csv(
    OUT / "ranking_point_reconciliation.csv"
)

# Save objects in the current Colab session.
final_scoring_inventory_v1 = inventory.copy()
final_scoring_factor_audit_v1 = factor_inventory.copy()
final_scoring_bonus_audit_v1 = bonus_columns.copy()
final_scoring_main_audit_v1 = main_event_audit.copy()
final_scoring_score_audit_v1 = score_audit.copy()
final_scoring_readiness_v1 = readiness.copy()

print("\n" + "=" * 80)
print("BIG CELL 12 COMPLETE")
print("=" * 80)
print("Ranking totals validated for 14/14 recent events.")
print("Historical event scores unchanged.")
print("No provisional bonuses treated as verified.")
print("No final event scores changed.")
print("Audit files saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
import re
from pathlib import Path
from IPython.display import display

# ================================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 13 — BONUS HISTORY RECONSTRUCTION & VALIDATION
#
# 1. Audit historical bonus definitions
# 2. Reconcile recent fighter-level bonus baselines
# 3. Rebuild recent event bonus averages
# 4. Compare recent main-event bonus totals
# 5. Identify inconsistent or missing source evidence
# 6. Save safe candidate outputs
#
# No historical scores overwritten.
# No recent final scores overwritten.
# No bonus counts invented.
# ================================================================

OUT = Path("/content/ufc_bonus_reconstruction")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "backup",
    "scoring_fights",
    "main_event_scoring",
    "recent_event_raw",
    "ranking_event_totals_final_candidate_v3"
]

missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError(
        "Missing objects: " + ", ".join(missing)
    )

fights = scoring_fights.copy()
historical = backup["event_master"].copy()
recent = recent_event_raw.copy()
main = main_event_scoring.copy()
ranking = ranking_event_totals_final_candidate_v3.copy()

def find_col(df, options, required=True):
    for name in options:
        if name in df.columns:
            return name
    if required:
        raise KeyError(
            "Could not find any of: " + str(options)
        )
    return None

def numeric(df, col):
    return pd.to_numeric(df[col], errors="coerce")

def normalized_date(df):
    col = find_col(
        df, ["event_date", "date", "Date"], False
    )
    if col is None:
        return pd.Series(pd.NaT, index=df.index)
    return pd.to_datetime(
        df[col], errors="coerce"
    ).dt.normalize()

def safe_mean(values):
    s = pd.to_numeric(
        pd.Series(values), errors="coerce"
    )
    return s.mean() if s.notna().any() else np.nan

print("=" * 85)
print("UFC EVENT STRENGTH INDEX V2")
print("BIG CELL 13 — PREVIOUS BONUS HISTORY AUDIT")
print("=" * 85)

# ------------------------------------------------------------
# 1. Identify recent fights using exact event names
# ------------------------------------------------------------

event_col = find_col(
    fights, ["event_name", "event", "Event"]
)

recent_names = set(
    recent["event_name"].astype(str)
)

recent_fights = fights[
    fights[event_col].astype(str).isin(recent_names)
].copy()

recent_fights["audit_event_name"] = (
    recent_fights[event_col].astype(str)
)

recent_fights["audit_event_date"] = (
    normalized_date(recent_fights)
)

print("\nRECENT FIGHT INVENTORY")
print("Expected recent fights: 179")
print("Found recent fights:", len(recent_fights))
print("Recent events:", recent_fights["audit_event_name"].nunique())

if len(recent_fights) != 179:
    print(
        "WARNING: Fight count differs from 179. "
        "Review event matching before approving calculations."
    )

# ------------------------------------------------------------
# 2. Compare bonus columns in all datasets
# ------------------------------------------------------------

print("\nBONUS-RELATED COLUMNS")

for name, df in [
    ("HISTORICAL RAW", historical),
    ("ALL FIGHTS", fights),
    ("MAIN EVENTS", main),
    ("RECENT RAW", recent)
]:
    cols = [
        c for c in df.columns
        if "bonus" in str(c).lower()
    ]
    print("\n", name)
    print(cols)

# ------------------------------------------------------------
# 3. Identify fighter bonus fields
# ------------------------------------------------------------

fighter_bonus_columns = [
    "f1_previous_total_bonuses",
    "f2_previous_total_bonuses",
    "f1_bonus_baseline",
    "f2_bonus_baseline"
]

for col in fighter_bonus_columns:
    if col not in recent_fights.columns:
        recent_fights[col] = np.nan

for col in fighter_bonus_columns:
    recent_fights[col] = numeric(recent_fights, col)

# Important: a baseline is a candidate value, not a
# confirmed complete historical bonus count.

recent_fights["f1_bonus_candidate"] = (
    recent_fights["f1_previous_total_bonuses"]
    .combine_first(recent_fights["f1_bonus_baseline"])
)

recent_fights["f2_bonus_candidate"] = (
    recent_fights["f2_previous_total_bonuses"]
    .combine_first(recent_fights["f2_bonus_baseline"])
)

recent_fights["f1_bonus_source"] = np.select(
    [
        recent_fights["f1_previous_total_bonuses"].notna(),
        recent_fights["f1_bonus_baseline"].notna()
    ],
    [
        "PREVIOUS_TOTAL_BONUSES",
        "BASELINE_ONLY"
    ],
    default="MISSING"
)

recent_fights["f2_bonus_source"] = np.select(
    [
        recent_fights["f2_previous_total_bonuses"].notna(),
        recent_fights["f2_bonus_baseline"].notna()
    ],
    [
        "PREVIOUS_TOTAL_BONUSES",
        "BASELINE_ONLY"
    ],
    default="MISSING"
)

# ------------------------------------------------------------
# 4. Compare overlapping values when available
# ------------------------------------------------------------

comparison_rows = []

for fighter_num in [1, 2]:
    old_col = f"f{fighter_num}_previous_total_bonuses"
    base_col = f"f{fighter_num}_bonus_baseline"

    overlap = recent_fights[
        recent_fights[old_col].notna()
        & recent_fights[base_col].notna()
    ].copy()

    if len(overlap):
        overlap["difference"] = (
            overlap[old_col] - overlap[base_col]
        )

        comparison_rows.append({
            "fighter_position": fighter_num,
            "overlapping_values": len(overlap),
            "exact_matches": int(
                np.isclose(
                    overlap[old_col],
                    overlap[base_col]
                ).sum()
            ),
            "disagreements": int(
                (~np.isclose(
                    overlap[old_col],
                    overlap[base_col]
                )).sum()
            ),
            "maximum_absolute_difference": (
                overlap["difference"].abs().max()
            )
        })
    else:
        comparison_rows.append({
            "fighter_position": fighter_num,
            "overlapping_values": 0,
            "exact_matches": 0,
            "disagreements": 0,
            "maximum_absolute_difference": np.nan
        })

bonus_overlap_audit_v1 = pd.DataFrame(comparison_rows)

print("\nBONUS BASELINE VS PREVIOUS-TOTAL OVERLAP")
display(bonus_overlap_audit_v1)

# ------------------------------------------------------------
# 5. Reconstruct event-level average bonuses
# ------------------------------------------------------------

recent_fights["bonus_appearance_count"] = (
    recent_fights[
        ["f1_bonus_candidate", "f2_bonus_candidate"]
    ].notna().sum(axis=1)
)

recent_fights["bonus_fight_sum"] = (
    recent_fights[
        ["f1_bonus_candidate", "f2_bonus_candidate"]
    ].sum(axis=1, min_count=1)
)

event_rows = []

for event_name, group in recent_fights.groupby(
    "audit_event_name", sort=False
):
    values = pd.concat([
        group["f1_bonus_candidate"],
        group["f2_bonus_candidate"]
    ], ignore_index=True)

    expected = len(group) * 2
    available = int(values.notna().sum())

    event_rows.append({
        "event_name": event_name,
        "fights_found": len(group),
        "expected_fighter_appearances": expected,
        "bonus_appearances_available": available,
        "bonus_appearances_missing": expected - available,
        "total_previous_bonuses_candidate": (
            values.sum(min_count=1)
        ),
        "avg_previous_bonuses_candidate": (
            values.mean()
            if available else np.nan
        ),
        "all_bonus_values_present": (
            available == expected
        )
    })

bonus_events = pd.DataFrame(event_rows)

recent_bonus_col = find_col(
    recent,
    [
        "avg_previous_bonuses_provisional",
        "avg_previous_bonuses"
    ],
    False
)

if recent_bonus_col is not None:
    baseline_events = recent[
        ["event_name", recent_bonus_col]
    ].copy()

    baseline_events = baseline_events.rename(
        columns={
            recent_bonus_col:
            "existing_provisional_event_average"
        }
    )

    bonus_events = bonus_events.merge(
        baseline_events,
        on="event_name",
        how="left",
        validate="one_to_one"
    )

    bonus_events["average_difference"] = (
        bonus_events["avg_previous_bonuses_candidate"]
        - bonus_events["existing_provisional_event_average"]
    )

    bonus_events["matches_existing_average"] = (
        np.isclose(
            bonus_events["avg_previous_bonuses_candidate"],
            bonus_events["existing_provisional_event_average"],
            atol=1e-8,
            rtol=0,
            equal_nan=False
        )
    )
else:
    bonus_events["existing_provisional_event_average"] = np.nan
    bonus_events["average_difference"] = np.nan
    bonus_events["matches_existing_average"] = False

print("\n" + "=" * 85)
print("RECONSTRUCTED BONUSES — 14 RECENT EVENTS")
print("=" * 85)

display(
    bonus_events[
        [
            "event_name",
            "fights_found",
            "bonus_appearances_missing",
            "total_previous_bonuses_candidate",
            "avg_previous_bonuses_candidate",
            "existing_provisional_event_average",
            "average_difference",
            "matches_existing_average"
        ]
    ]
)

# ------------------------------------------------------------
# 6. Check bonus values for invalid data
# ------------------------------------------------------------

invalid_rows = []

for col in [
    "f1_bonus_candidate",
    "f2_bonus_candidate"
]:
    bad = recent_fights[
        recent_fights[col].notna()
        & (
            (recent_fights[col] < 0)
            | (~np.isclose(
                recent_fights[col],
                np.round(recent_fights[col])
            ))
        )
    ].copy()

    for _, row in bad.iterrows():
        invalid_rows.append({
            "event_name": row["audit_event_name"],
            "column": col,
            "value": row[col]
        })

bonus_invalid_values_v1 = pd.DataFrame(
    invalid_rows,
    columns=["event_name", "column", "value"]
)

print("\nINVALID BONUS VALUES:")
print(len(bonus_invalid_values_v1))

if len(bonus_invalid_values_v1):
    display(bonus_invalid_values_v1)

# ------------------------------------------------------------
# 7. Compare main-event bonus values to fight records
# ------------------------------------------------------------

main_event_col = find_col(
    recent_fights,
    ["is_main_event"],
    False
)

if main_event_col is not None:
    flag = recent_fights[main_event_col]

    if pd.api.types.is_bool_dtype(flag):
        is_main = flag.fillna(False)
    else:
        is_main = (
            flag.astype(str).str.lower()
            .str.strip().isin(["true", "1", "yes"])
        )

    recent_main_fights = recent_fights[
        is_main
    ].copy()

else:
    recent_main_fights = recent_fights.iloc[0:0].copy()
    print(
        "\nWARNING: No is_main_event field found "
        "in recent fight records."
    )

recent_main_fights["reconstructed_me_bonuses"] = (
    recent_main_fights[
        ["f1_bonus_candidate", "f2_bonus_candidate"]
    ].sum(axis=1, min_count=2)
)

main_recent = main[
    main["event_name"].astype(str).isin(recent_names)
].copy()

if "me_combined_bonuses" in main_recent.columns:
    main_recent["stored_me_bonuses"] = pd.to_numeric(
        main_recent["me_combined_bonuses"],
        errors="coerce"
    )
else:
    main_recent["stored_me_bonuses"] = np.nan

main_comparison = main_recent[
    ["event_name", "stored_me_bonuses"]
].merge(
    recent_main_fights[
        [
            "audit_event_name",
            "reconstructed_me_bonuses"
        ]
    ].rename(
        columns={"audit_event_name": "event_name"}
    ),
    on="event_name",
    how="left",
    validate="one_to_one"
)

main_comparison["difference"] = (
    main_comparison["reconstructed_me_bonuses"]
    - main_comparison["stored_me_bonuses"]
)

main_comparison["matches"] = np.isclose(
    main_comparison["reconstructed_me_bonuses"],
    main_comparison["stored_me_bonuses"],
    atol=1e-8,
    rtol=0,
    equal_nan=False
)

print("\n" + "=" * 85)
print("MAIN-EVENT PREVIOUS BONUS RECONCILIATION")
print("=" * 85)

display(main_comparison)

# ------------------------------------------------------------
# 8. Historical bonus distribution and data availability
# ------------------------------------------------------------

historical_bonus_cols = [
    c for c in [
        "avg_previous_bonuses",
        "total_previous_bonuses",
        "main_event_combined_bonuses",
        "bonus_history_available"
    ]
    if c in historical.columns
]

historical_bonus_audit = pd.DataFrame([
    {
        "column": c,
        "nonmissing": int(historical[c].notna().sum()),
        "missing": int(historical[c].isna().sum()),
        "zero_values": int(
            (pd.to_numeric(
                historical[c], errors="coerce"
            ) == 0).sum()
        )
    }
    for c in historical_bonus_cols
])

print("\nHISTORICAL BONUS DATA AVAILABILITY")
display(historical_bonus_audit)

if "bonus_history_available" in historical.columns:
    print("\nHISTORICAL BONUS HISTORY FLAGS:")
    print(
        historical["bonus_history_available"]
        .value_counts(dropna=False)
        .to_string()
    )

# ------------------------------------------------------------
# 9. Bonus scoring readiness — do not equate
#    internal agreement with independent verification
# ------------------------------------------------------------

bonus_events["internal_average_reconciled"] = (
    bonus_events["all_bonus_values_present"]
    & bonus_events["matches_existing_average"]
)

bonus_events["independently_verified"] = False

bonus_events["ready_for_provisional_scoring"] = (
    bonus_events["internal_average_reconciled"]
    & (len(bonus_invalid_values_v1) == 0)
)

print("\n" + "=" * 85)
print("BONUS SCORING READINESS")
print("=" * 85)

display(
    bonus_events[
        [
            "event_name",
            "all_bonus_values_present",
            "internal_average_reconciled",
            "ready_for_provisional_scoring",
            "independently_verified"
        ]
    ]
)

# ------------------------------------------------------------
# 10. Save everything without modifying original tables
# ------------------------------------------------------------

recent_fights.to_csv(
    OUT / "recent_fight_bonus_candidates.csv",
    index=False
)

bonus_events.to_csv(
    OUT / "recent_event_bonus_reconciliation.csv",
    index=False
)

main_comparison.to_csv(
    OUT / "recent_main_event_bonus_reconciliation.csv",
    index=False
)

historical_bonus_audit.to_csv(
    OUT / "historical_bonus_availability.csv",
    index=False
)

bonus_overlap_audit_v1.to_csv(
    OUT / "bonus_baseline_overlap_audit.csv",
    index=False
)

bonus_invalid_values_v1.to_csv(
    OUT / "invalid_bonus_values.csv",
    index=False
)

bonus_fight_candidates_v1 = recent_fights.copy()
bonus_event_candidates_v1 = bonus_events.copy()
bonus_main_event_comparison_v1 = main_comparison.copy()
bonus_historical_audit_v1 = historical_bonus_audit.copy()

print("\n" + "=" * 85)
print("BIG CELL 13 COMPLETE")
print("=" * 85)

print("Recent fights analyzed:", len(recent_fights))
print("Recent events analyzed:", len(bonus_events))
print(
    "Event averages matching existing data:",
    int(bonus_events["matches_existing_average"].sum()),
    "/", len(bonus_events)
)
print(
    "Main-event bonuses matching:",
    int(main_comparison["matches"].sum()),
    "/", len(main_comparison)
)
print(
    "Events internally ready for provisional bonus scoring:",
    int(
        bonus_events[
            "ready_for_provisional_scoring"
        ].sum()
    )
)
print("Original scores unchanged.")
print("Audit files saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ==============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 14 — HISTORICAL SCORING RECONSTRUCTION
# Card Quality + Star Experience
#
# - Audits historical raw-to-normalized factor relationships
# - Rebuilds recent raw factors from fighter records
# - Detects historical formula compatibility
# - Generates candidate normalized recent factor values
# - Reconciles historical category scoring
# - Preserves all original score tables
# ==============================================================

OUT = Path("/content/ufc_category_reconstruction")
OUT.mkdir(parents=True, exist_ok=True)

needed = [
    "backup", "scoring_fights", "recent_event_raw",
    "event_strength_v2",
    "ranking_event_totals_final_candidate_v3",
    "bonus_event_candidates_v1"
]
missing = [x for x in needed if x not in globals()]
if missing:
    raise RuntimeError("Missing objects: " + ", ".join(missing))

hist = backup["event_master"].copy()
norm = backup["normalized"].copy()
fights = scoring_fights.copy()
recent = recent_event_raw.copy()
scores = event_strength_v2.copy()
rankings = ranking_event_totals_final_candidate_v3.copy()
bonuses = bonus_event_candidates_v1.copy()

def num(s):
    return pd.to_numeric(s, errors="coerce")

def check_key(df, name):
    if "event_name" not in df:
        raise KeyError(name + " lacks event_name")
    if df["event_name"].duplicated().any():
        raise ValueError(name + " has duplicate event names")

for name, df in [
    ("Historical raw", hist),
    ("Historical normalized", norm),
    ("Recent raw", recent),
    ("Scores", scores),
    ("Ranking totals", rankings),
    ("Bonus candidates", bonuses)
]:
    check_key(df, name)

recent_names = set(recent["event_name"].astype(str))
rf = fights[
    fights["event_name"].astype(str).isin(recent_names)
].copy()

print("=" * 85)
print("BIG CELL 14 — CARD QUALITY AND STAR EXPERIENCE")
print("=" * 85)
print("Historical raw events:", len(hist))
print("Historical normalized events:", len(norm))
print("Recent events:", len(recent))
print("Recent fights:", len(rf))

if len(hist) != 777 or len(norm) != 777:
    print("WARNING: Historical event counts differ from expected.")
if len(recent) != 14 or len(rf) != 179:
    print("WARNING: Recent counts differ from expected.")

# --------------------------------------------------------------
# 1. Discover historical factor-score pairs
# --------------------------------------------------------------

score_pairs = []
for col in hist.columns:
    candidate = str(col) + "_score"
    if candidate in norm.columns:
        score_pairs.append((col, candidate))

print("\nHISTORICAL RAW -> NORMALIZED FACTOR PAIRS")
for a, b in score_pairs:
    print(f"{a:42} -> {b}")

print("\nALL HISTORICAL NORMALIZED COLUMNS")
print(list(norm.columns))

print("\nALL RECENT RAW COLUMNS")
print(list(recent.columns))

print("\nALL RECENT FIGHT COLUMNS")
print(list(rf.columns))

# --------------------------------------------------------------
# 2. Historical normalized factor audit
# --------------------------------------------------------------

h = hist.merge(
    norm,
    on="event_name",
    suffixes=("_raw", "_normalized"),
    validate="one_to_one"
)

pair_rows = []
lookup_tables = {}

for raw_col, score_col in score_pairs:
    x = num(h[raw_col])
    y = num(h[score_col])
    valid = x.notna() & y.notna()

    temp = pd.DataFrame({
        "raw": x[valid],
        "score": y[valid]
    })

    if len(temp):
        grouped = temp.groupby("raw")["score"].agg(
            ["min", "max", "mean", "count"]
        ).reset_index()

        conflicts = int(
            (~np.isclose(
                grouped["min"], grouped["max"],
                atol=1e-7
            )).sum()
        )

        lookup_tables[raw_col] = grouped
    else:
        conflicts = 0
        grouped = pd.DataFrame()

    pair_rows.append({
        "raw_factor": raw_col,
        "normalized_factor": score_col,
        "historical_observations": int(valid.sum()),
        "unique_raw_values": int(x[valid].nunique()),
        "same_raw_different_score_conflicts": conflicts,
        "raw_min": x[valid].min(),
        "raw_max": x[valid].max(),
        "score_min": y[valid].min(),
        "score_max": y[valid].max()
    })

pair_audit = pd.DataFrame(pair_rows)

print("\n" + "=" * 85)
print("HISTORICAL NORMALIZATION CONSISTENCY")
print("=" * 85)
display(pair_audit)

# --------------------------------------------------------------
# 3. Build recent raw factor candidates
# --------------------------------------------------------------

candidate = recent.copy()

# The ranking points have been independently reconstructed
# from archived ranking snapshots. Use the latest candidate.
ranking_small = rankings[
    ["event_name", "total_ranking_points_v2"]
].copy()

candidate = candidate.merge(
    ranking_small,
    on="event_name",
    validate="one_to_one"
)

candidate["total_ranking_points_candidate"] = num(
    candidate["total_ranking_points_v2"]
)

bonus_small = bonuses[
    [
        "event_name",
        "avg_previous_bonuses_candidate",
        "total_previous_bonuses_candidate",
        "all_bonus_values_present"
    ]
].copy()

candidate = candidate.merge(
    bonus_small,
    on="event_name",
    validate="one_to_one"
)

# Explicitly map candidate values to historical factor names.
candidate["avg_previous_bonuses"] = num(
    candidate["avg_previous_bonuses_candidate"]
)

candidate["total_previous_bonuses"] = num(
    candidate["total_previous_bonuses_candidate"]
)

candidate["total_ranking_points"] = num(
    candidate["total_ranking_points_candidate"]
)

# Historical total_fights is the number of fights, not
# fighter appearances.
candidate["total_fights"] = (
    rf.groupby("event_name").size()
    .reindex(candidate["event_name"])
    .to_numpy()
)

# Reconstruct additive event-level raw factors where the
# underlying per-fighter values are present.
additive_fighter_fields = {
    "total_previous_main_events": [
        "f1_previous_main_events",
        "f2_previous_main_events"
    ],
    "total_previous_title_fights": [
        "f1_previous_title_fights",
        "f2_previous_title_fights"
    ],
    "total_prior_ufc_fights": [
        "f1_ufc_fights_before",
        "f2_ufc_fights_before"
    ],
    "total_prior_ufc_finishes": [
        "f1_ufc_finishes_before",
        "f2_ufc_finishes_before"
    ]
}

factor_sources = []

for target, columns in additive_fighter_fields.items():
    available = [c for c in columns if c in rf.columns]

    if len(available) == 2:
        tmp = rf[
            ["event_name"] + available
        ].copy()

        for col in available:
            tmp[col] = num(tmp[col])

        # No silent filling of missing individual values.
        tmp["_sum"] = tmp[available].sum(
            axis=1, min_count=2
        )

        complete = tmp.groupby("event_name")[
            available
        ].apply(lambda x: x.notna().all().all())

        sums = tmp.groupby("event_name")["_sum"].sum(
            min_count=1
        )

        sums = sums.where(complete, np.nan)

        candidate[target] = (
            candidate["event_name"].map(sums)
        )

        factor_sources.append({
            "factor": target,
            "source": "fighter-level reconstructed",
            "events_complete": int(
                candidate[target].notna().sum()
            )
        })

    else:
        factor_sources.append({
            "factor": target,
            "source": "unavailable fighter columns",
            "events_complete": 0
        })

# For average ranking points, historical denominator needs
# validation. Do not assume it is fighters or fights.
candidate["avg_ranking_points_candidate_per_fighter"] = (
    candidate["total_ranking_points"]
    / (candidate["total_fights"] * 2)
)

# --------------------------------------------------------------
# 4. Check historical average ranking denominator
# --------------------------------------------------------------

if all(c in hist for c in [
    "avg_ranking_points",
    "total_ranking_points",
    "total_fights"
]):
    test = hist[
        [
            "event_name",
            "avg_ranking_points",
            "total_ranking_points",
            "total_fights"
        ]
    ].copy()

    test["predicted_per_fighter"] = (
        num(test["total_ranking_points"])
        / (num(test["total_fights"]) * 2)
    )

    test["matches_per_fighter"] = np.isclose(
        num(test["avg_ranking_points"]),
        test["predicted_per_fighter"],
        atol=1e-7,
        rtol=0
    )

    print("\nHISTORICAL AVERAGE RANKING DENOMINATOR")
    print(
        "Matches total ranking points / (2 * fights):",
        int(test["matches_per_fighter"].sum()),
        "/", len(test)
    )

    if test["matches_per_fighter"].all():
        candidate["avg_ranking_points"] = (
            candidate[
                "avg_ranking_points_candidate_per_fighter"
            ]
        )
        print("Candidate average ranking points approved.")
    else:
        print(
            "Historical denominator is not fully confirmed. "
            "Average ranking points will remain unassigned."
        )

# --------------------------------------------------------------
# 5. Generate normalized candidates from historical lookups
# --------------------------------------------------------------

def interpolate_from_history(values, lookup):
    """
    Reproduce known historical raw -> score values.
    Linearly interpolate only inside observed raw range.
    Return NaN outside the observed range.
    """
    if lookup is None or len(lookup) == 0:
        return pd.Series(np.nan, index=values.index)

    table = lookup.sort_values("raw").copy()
    x = table["raw"].to_numpy(dtype=float)
    y = table["mean"].to_numpy(dtype=float)

    result = []
    for value in num(values):
        if pd.isna(value):
            result.append(np.nan)
        elif len(x) == 1:
            result.append(
                float(y[0])
                if np.isclose(value, x[0])
                else np.nan
            )
        elif value < x.min() or value > x.max():
            result.append(np.nan)
        else:
            result.append(
                float(np.interp(value, x, y))
            )

    return pd.Series(result, index=values.index)

normalization_rows = []

for raw_col, score_col in score_pairs:
    outcol = "candidate__" + score_col

    if raw_col not in candidate.columns:
        candidate[outcol] = np.nan
        status = "RAW_FACTOR_MISSING"
    else:
        candidate[outcol] = interpolate_from_history(
            candidate[raw_col],
            lookup_tables.get(raw_col)
        )
        status = (
            "LOOKUP_CONFLICT_REVIEW"
            if pair_audit.set_index("raw_factor").loc[
                raw_col,
                "same_raw_different_score_conflicts"
            ] > 0
            else "HISTORICAL_LOOKUP"
        )

    normalization_rows.append({
        "raw_factor": raw_col,
        "normalized_factor": score_col,
        "method": status,
        "recent_events_with_raw": (
            int(candidate[raw_col].notna().sum())
            if raw_col in candidate else 0
        ),
        "recent_events_with_candidate_score": int(
            candidate[outcol].notna().sum()
        )
    })

normalization_audit = pd.DataFrame(
    normalization_rows
)

print("\n" + "=" * 85)
print("RECENT FACTOR NORMALIZATION COVERAGE")
print("=" * 85)
display(normalization_audit)

# --------------------------------------------------------------
# 6. Discover historical category scoring formula
# --------------------------------------------------------------

historical_scores = hist[
    ["event_name"]
].merge(
    norm,
    on="event_name",
    validate="one_to_one"
).merge(
    scores[
        [
            "event_name",
            "card_quality_score",
            "star_experience_score"
        ]
    ],
    on="event_name",
    validate="one_to_one"
)

factor_score_columns = [
    c for c in norm.columns
    if str(c).endswith("_score")
    and c not in [
        "card_quality_score",
        "star_experience_score",
        "corrected_main_event_score"
    ]
]

category_tests = []

# The original scoring may average subsets of factors.
# First test natural keyword-based groupings.
card_keywords = [
    "fight", "finish", "ranking",
    "ranked", "title"
]
star_keywords = [
    "experience", "win", "streak",
    "main_events", "bonuses", "hype"
]

for category, keywords, target in [
    (
        "CARD_QUALITY",
        card_keywords,
        "card_quality_score"
    ),
    (
        "STAR_EXPERIENCE",
        star_keywords,
        "star_experience_score"
    )
]:
    cols = [
        c for c in factor_score_columns
        if any(
            k in c.lower() for k in keywords
        )
    ]

    if cols:
        predicted = historical_scores[
            cols
        ].apply(pd.to_numeric, errors="coerce").mean(
            axis=1, skipna=True
        )

        actual = num(historical_scores[target])
        valid = actual.notna() & predicted.notna()

        errors = (
            actual[valid] - predicted[valid]
        ).abs()

        category_tests.append({
            "category": category,
            "candidate_factor_count": len(cols),
            "factors": ", ".join(cols),
            "tested_events": int(valid.sum()),
            "exact_matches": int(
                np.isclose(
                    actual[valid],
                    predicted[valid],
                    atol=1e-6
                ).sum()
            ),
            "mean_absolute_error": (
                float(errors.mean())
                if len(errors) else np.nan
            ),
            "maximum_absolute_error": (
                float(errors.max())
                if len(errors) else np.nan
            )
        })

category_formula_audit = pd.DataFrame(
    category_tests
)

print("\n" + "=" * 85)
print("HISTORICAL CATEGORY FORMULA TESTS")
print("=" * 85)
display(category_formula_audit)

print(
    "\nIMPORTANT: Keyword groupings above are DIAGNOSTIC "
    "TESTS, not approved category formulas."
)
print(
    "We will only use a category formula if it "
    "reproduces historical scores."
)

# --------------------------------------------------------------
# 7. Recent factor completeness
# --------------------------------------------------------------

recent_factor_cols = [
    c for c in candidate.columns
    if c.startswith("candidate__")
]

recent_coverage = pd.DataFrame({
    "event_name": candidate["event_name"],
    "event_date": candidate["event_date"],
    "fights": candidate["total_fights"],
    "ranking_points": candidate[
        "total_ranking_points"
    ],
    "average_previous_bonuses": candidate[
        "avg_previous_bonuses"
    ],
    "normalized_factors_available": candidate[
        recent_factor_cols
    ].notna().sum(axis=1),
    "normalized_factors_total": len(
        recent_factor_cols
    )
})

print("\n" + "=" * 85)
print("14 RECENT EVENTS — NORMALIZATION READINESS")
print("=" * 85)
display(recent_coverage)

# --------------------------------------------------------------
# 8. Export all candidates and evidence
# --------------------------------------------------------------

pair_audit.to_csv(
    OUT / "historical_normalization_consistency.csv",
    index=False
)

normalization_audit.to_csv(
    OUT / "recent_factor_normalization_coverage.csv",
    index=False
)

category_formula_audit.to_csv(
    OUT / "historical_category_formula_tests.csv",
    index=False
)

recent_coverage.to_csv(
    OUT / "recent_normalization_readiness.csv",
    index=False
)

candidate.to_csv(
    OUT / "recent_raw_and_normalized_candidates.csv",
    index=False
)

pd.DataFrame(factor_sources).to_csv(
    OUT / "fighter_factor_source_audit.csv",
    index=False
)

category_recent_candidates_v1 = candidate.copy()
category_historical_pair_audit_v1 = pair_audit.copy()
category_normalization_audit_v1 = normalization_audit.copy()
category_formula_audit_v1 = category_formula_audit.copy()
category_recent_coverage_v1 = recent_coverage.copy()

print("\n" + "=" * 85)
print("BIG CELL 14 COMPLETE")
print("=" * 85)
print("Original 777 historical scores unchanged.")
print("Original 14 recent scores unchanged.")
print("No final leaderboard published prematurely.")
print("Saved files:", OUT)
```

```python

import pandas as pd
import numpy as np
import itertools
from pathlib import Path
from IPython.display import display

# ==============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 15 — STAR EXPERIENCE FORMULA DISCOVERY
#
# Exhaustively tests equal-weight factor combinations.
# Uses original 777 historical events as ground truth.
# Checks missing-data behavior and possible formulas.
# Calculates recent Star Experience only if fully verified.
#
# DOES NOT change original scores or publish leaderboard.
# ==============================================================

OUT = Path("/content/ufc_star_formula_discovery")
OUT.mkdir(parents=True, exist_ok=True)

if "backup" not in globals():
    raise RuntimeError("Historical backup is missing.")

if "category_recent_candidates_v2" not in globals():
    raise RuntimeError(
        "Run the Big Cell 14 fix first."
    )

hist = backup["normalized"].copy()
recent = category_recent_candidates_v2.copy()

assert len(hist) == 777
assert len(recent) == 14
assert hist["event_name"].is_unique
assert recent["event_name"].is_unique

TARGET = "star_experience_score"

if TARGET not in hist.columns:
    raise KeyError("Historical Star Experience scores missing.")

# These are potential Star Experience factors.
# We test rather than assume which belong in the formula.
potential = [
    "avg_ufc_experience_score",
    "avg_ufc_win_pct_score",
    "avg_win_streak_score",
    "avg_previous_title_fights_score",
    "avg_previous_main_events_score",
    "avg_previous_bonuses_score",
    "event_avg_log_hype_score",
    "total_ranking_points_score",
    "title_fights_score",
    "avg_finish_rate_score"
]

potential = [c for c in potential if c in hist.columns]

print("=" * 82)
print("BIG CELL 15 — STAR EXPERIENCE FORMULA DISCOVERY")
print("=" * 82)
print("Historical events:", len(hist))
print("Candidate factors:", len(potential))
print("Combinations to test:", 2**len(potential) - 1)

actual = pd.to_numeric(
    hist[TARGET], errors="coerce"
).to_numpy(dtype=float)

X = hist[potential].apply(
    pd.to_numeric, errors="coerce"
)

results = []
perfect_formulas = []

# --------------------------------------------------------------
# 1. Test every nonempty equal-weight subset
# --------------------------------------------------------------

for size in range(1, len(potential) + 1):
    for combo in itertools.combinations(potential, size):

        predicted = X[list(combo)].mean(
            axis=1, skipna=True
        ).to_numpy(dtype=float)

        comparable = (
            np.isfinite(actual)
            & np.isfinite(predicted)
        )

        if comparable.sum() == 0:
            continue

        errors = np.abs(
            actual[comparable] - predicted[comparable]
        )

        exact = np.isclose(
            actual[comparable],
            predicted[comparable],
            atol=1e-6,
            rtol=0
        )

        # Missing predictions do not count as matches.
        full_match = (
            comparable.sum() == np.isfinite(actual).sum()
            and exact.all()
        )

        row = {
            "factor_count": size,
            "factors": " | ".join(combo),
            "historical_events_tested": int(
                comparable.sum()
            ),
            "exact_matches": int(exact.sum()),
            "mean_absolute_error": float(errors.mean()),
            "maximum_absolute_error": float(errors.max()),
            "fully_verified": bool(full_match)
        }

        results.append(row)

        if full_match:
            perfect_formulas.append(combo)

formula_results = pd.DataFrame(results).sort_values(
    [
        "fully_verified",
        "exact_matches",
        "mean_absolute_error",
        "factor_count"
    ],
    ascending=[False, False, True, True]
).reset_index(drop=True)

print("\n" + "=" * 82)
print("TOP 20 HISTORICAL STAR EXPERIENCE FORMULAS")
print("=" * 82)

display(formula_results.head(20))

print("\nEXACT HISTORICAL FORMULAS FOUND:",
      len(perfect_formulas))

if perfect_formulas:
    for i, combo in enumerate(perfect_formulas, 1):
        print(f"\nFORMULA {i}:")
        for factor in combo:
            print("  ", factor)
else:
    print(
        "\nNo tested equal-weight combination reproduces "
        "all historical scores. We will not invent one."
    )

# --------------------------------------------------------------
# 2. Audit historical missingness
# --------------------------------------------------------------

missingness = pd.DataFrame({
    "factor": potential,
    "historical_nonmissing": [
        int(X[c].notna().sum()) for c in potential
    ],
    "historical_missing": [
        int(X[c].isna().sum()) for c in potential
    ]
})

print("\n" + "=" * 82)
print("HISTORICAL STAR FACTOR DATA AVAILABILITY")
print("=" * 82)

display(missingness)

# --------------------------------------------------------------
# 3. Check historical scores against the strongest formula
# --------------------------------------------------------------

best_factors = (
    list(perfect_formulas[0])
    if perfect_formulas
    else formula_results.iloc[0]["factors"].split(" | ")
)

historical_check = hist[
    ["event_name", "event_date", TARGET]
].copy()

historical_check["reconstructed_score"] = (
    X[best_factors].mean(axis=1, skipna=True)
)

historical_check["absolute_error"] = (
    pd.to_numeric(
        historical_check[TARGET],
        errors="coerce"
    )
    - historical_check["reconstructed_score"]
).abs()

historical_check["exact_match"] = np.isclose(
    pd.to_numeric(
        historical_check[TARGET],
        errors="coerce"
    ),
    historical_check["reconstructed_score"],
    atol=1e-6,
    rtol=0,
    equal_nan=True
)

print("\n" + "=" * 82)
print("HISTORICAL RECONSTRUCTION CHECK")
print("=" * 82)

print("Best factors:", best_factors)
print(
    "Exact historical matches:",
    int(historical_check["exact_match"].sum()),
    "/", len(historical_check)
)

if not historical_check["exact_match"].all():
    print("\nLARGEST HISTORICAL MISMATCHES:")
    display(
        historical_check.sort_values(
            "absolute_error",
            ascending=False
        ).head(15)
    )

# --------------------------------------------------------------
# 4. Calculate recent Star Experience candidates
#    ONLY if the formula is verified.
# --------------------------------------------------------------

recent["star_experience_candidate_v3"] = np.nan
recent["star_experience_formula_verified"] = False
recent["star_experience_factor_coverage"] = 0

if perfect_formulas:

    # Prefer the verified formula with the fewest factors.
    chosen = min(perfect_formulas, key=len)

    recent_factor_cols = [
        "candidate__" + c for c in chosen
    ]

    missing_recent_cols = [
        c for c in recent_factor_cols
        if c not in recent.columns
    ]

    if not missing_recent_cols:

        recent_matrix = recent[
            recent_factor_cols
        ].apply(pd.to_numeric, errors="coerce")

        recent["star_experience_factor_coverage"] = (
            recent_matrix.notna().sum(axis=1)
        )

        # Historical formula uses available-factor mean.
        # Require every selected recent factor to be present
        # for a fully populated recent candidate.
        complete = recent_matrix.notna().all(axis=1)

        recent.loc[
            complete,
            "star_experience_candidate_v3"
        ] = recent_matrix.loc[complete].mean(axis=1)

        recent["star_experience_formula_verified"] = (
            complete
        )

        print("\nVERIFIED STAR EXPERIENCE FORMULA:")
        print(" + ".join(chosen))
        print("Equal-weight mean of these normalized factors.")

    else:
        print("\nVerified formula found, but recent columns missing:")
        print(missing_recent_cols)

else:
    print(
        "\nNo verified formula found. "
        "Recent Star Experience scores remain missing."
    )

# --------------------------------------------------------------
# 5. Display recent category score status
# --------------------------------------------------------------

print("\n" + "=" * 82)
print("14 RECENT EVENTS — CATEGORY SCORE STATUS")
print("=" * 82)

display(
    recent[
        [
            "event_name",
            "card_quality_candidate",
            "star_experience_candidate_v3",
            "star_experience_factor_coverage",
            "star_experience_formula_verified"
        ]
    ]
)

# --------------------------------------------------------------
# 6. Save results
# --------------------------------------------------------------

formula_results.to_csv(
    OUT / "all_star_formula_combinations.csv",
    index=False
)

historical_check.to_csv(
    OUT / "historical_star_formula_reconstruction.csv",
    index=False
)

missingness.to_csv(
    OUT / "historical_star_factor_missingness.csv",
    index=False
)

recent.to_csv(
    OUT / "recent_star_experience_candidates.csv",
    index=False
)

star_formula_search_v1 = formula_results.copy()
star_historical_check_v1 = historical_check.copy()
star_recent_candidates_v1 = recent.copy()
star_verified_formulas_v1 = perfect_formulas

print("\n" + "=" * 82)
print("BIG CELL 15 COMPLETE")
print("=" * 82)
print("Exact formulas found:", len(perfect_formulas))
print(
    "Recent Star Experience candidates available:",
    int(
        recent["star_experience_candidate_v3"]
        .notna().sum()
    ),
    "/14"
)
print("Original scores unchanged.")
print("Files saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ===========================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 16 — UFC 330 CARD QUALITY INVESTIGATION
#
# Inspect factor contributions, rankings, title fights,
# and raw data consistency. NO score changes.
# ===========================================================

OUT = Path("/content/ufc_330_card_audit")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "star_recent_candidates_v1",
    "scoring_fights",
    "ranking_points_final_candidate_v3",
    "ranking_event_totals_final_candidate_v3"
]
missing = [v for v in required if v not in globals()]
if missing:
    raise RuntimeError("Missing objects: " + ", ".join(missing))

events = star_recent_candidates_v1.copy()
fights = scoring_fights.copy()
fighter_ranks = ranking_points_final_candidate_v3.copy()
ranking_events = ranking_event_totals_final_candidate_v3.copy()

TARGET = "UFC 330: Makhachev vs. Machado Garry"

card_factors = [
    "title_fights_score",
    "total_ranking_points_score",
    "avg_ufc_experience_score",
    "avg_ufc_win_pct_score",
    "avg_win_streak_score",
    "avg_finish_rate_score"
]

print("=" * 85)
print("BIG CELL 16 — UFC 330 CARD QUALITY INVESTIGATION")
print("=" * 85)

if TARGET not in set(events["event_name"]):
    raise RuntimeError("UFC 330 not found in recent events.")

# -----------------------------------------------------------
# 1. Six-factor Card Quality breakdown
# -----------------------------------------------------------

factor_rows = []

for factor in card_factors:
    col = "candidate__" + factor
    if col not in events:
        raise KeyError("Missing normalized factor: " + col)

    vals = pd.to_numeric(events[col], errors="coerce")
    target_val = vals.loc[events["event_name"].eq(TARGET)].iloc[0]

    factor_rows.append({
        "factor": factor,
        "UFC_330_normalized_score": target_val,
        "contribution_to_card_quality": target_val / 6,
        "recent_event_median": vals.median(),
        "rank_among_14_highest_first": int(
            vals.rank(ascending=False, method="min")
            .loc[events["event_name"].eq(TARGET)].iloc[0]
        )
    })

breakdown = pd.DataFrame(factor_rows)

print("\nUFC 330 — SIX CARD QUALITY FACTORS")
display(breakdown.round(3))

print("\nSum of contributions:",
      round(breakdown["contribution_to_card_quality"].sum(), 6))

print("Stored Card Quality:",
      round(float(events.loc[
          events["event_name"].eq(TARGET),
          "card_quality_candidate"
      ].iloc[0]), 6))

# -----------------------------------------------------------
# 2. Compare all recent events
# -----------------------------------------------------------

compare_cols = [
    "event_name", "card_quality_candidate",
    "star_experience_candidate_v3"
] + ["candidate__" + x for x in card_factors]

comparison = events[compare_cols].copy()
comparison = comparison.sort_values(
    "card_quality_candidate", ascending=False
)

print("\nALL 14 RECENT EVENTS — CARD QUALITY COMPARISON")
display(comparison.round(2))

# -----------------------------------------------------------
# 3. Inspect UFC 330's actual stored fight records
# -----------------------------------------------------------

ufc330 = fights[
    fights["event_name"].eq(TARGET)
].copy()

print("\nUFC 330 FIGHTS FOUND:", len(ufc330))

fight_cols = [
    "f1_name", "f2_name", "is_main_event",
    "is_title_fight", "title_fight_verified",
    "WEIGHTCLASS",
    "f1_rank_before", "f2_rank_before",
    "f1_ufc_fights_before", "f2_ufc_fights_before",
    "f1_ufc_win_pct_before", "f2_ufc_win_pct_before",
    "f1_win_streak_before", "f2_win_streak_before"
]

available = [c for c in fight_cols if c in ufc330]

print("\nUFC 330 — FIGHT-BY-FIGHT DATA")
display(ufc330[available].reset_index(drop=True))

# -----------------------------------------------------------
# 4. Inspect independently reconstructed ranking points
# -----------------------------------------------------------

ranked = fighter_ranks[
    fighter_ranks["event_name"].eq(TARGET)
].copy()

ranking_columns = [
    "fighter", "fight_division_key",
    "proposed_rank", "ranking_points_v2",
    "decision_category", "is_main_event"
]

ranking_columns = [
    c for c in ranking_columns if c in ranked
]

print("\nUFC 330 — FIGHTER RANKING AUDIT")
display(ranked[ranking_columns].reset_index(drop=True))

print("\nRanking decisions by category:")
if "decision_category" in ranked:
    print(ranked["decision_category"].value_counts(
        dropna=False
    ).to_string())

if "ranking_points_v2" in ranked:
    points = pd.to_numeric(
        ranked["ranking_points_v2"], errors="coerce"
    )
    print("\nTotal ranking points:", points.sum())
    print("Fighter appearances with ranking points > 0:",
          int(points.gt(0).sum()))
    print("Fighter appearances with zero ranking points:",
          int(points.eq(0).sum()))

# -----------------------------------------------------------
# 5. Raw event factor comparison
# -----------------------------------------------------------

raw_cols = [
    "event_name", "fight_count", "title_fights",
    "total_ranking_points", "avg_ufc_experience",
    "avg_ufc_win_pct", "avg_win_streak",
    "avg_finish_rate"
]

raw_cols = [c for c in raw_cols if c in events]

print("\nUFC 330 — RAW FACTORS")
display(events.loc[
    events["event_name"].eq(TARGET),
    raw_cols
])

# -----------------------------------------------------------
# 6. Reconcile raw title fight count and fighter experience
# -----------------------------------------------------------

checks = []

if "is_title_fight" in ufc330:
    title_flags = ufc330["is_title_fight"].astype(
        str
    ).str.lower().str.strip().isin(["true", "1", "yes"])

    raw_titles = pd.to_numeric(
        events.loc[
            events["event_name"].eq(TARGET),
            "title_fights"
        ], errors="coerce"
    ).iloc[0]

    checks.append({
        "check": "Title fights: fight records vs event raw",
        "reconstructed": int(title_flags.sum()),
        "stored": raw_titles,
        "matches": bool(np.isclose(
            title_flags.sum(), raw_titles
        ))
    })

if all(c in ufc330 for c in [
    "f1_ufc_fights_before", "f2_ufc_fights_before"
]):
    experience_values = pd.concat([
        pd.to_numeric(
            ufc330["f1_ufc_fights_before"],
            errors="coerce"
        ),
        pd.to_numeric(
            ufc330["f2_ufc_fights_before"],
            errors="coerce"
        )
    ])

    raw_experience = pd.to_numeric(
        events.loc[
            events["event_name"].eq(TARGET),
            "avg_ufc_experience"
        ], errors="coerce"
    ).iloc[0]

    checks.append({
        "check": "Average UFC experience",
        "reconstructed": experience_values.mean(),
        "stored": raw_experience,
        "matches": bool(
            experience_values.notna().all()
            and np.isclose(
                experience_values.mean(),
                raw_experience, atol=1e-7
            )
        )
    })

raw_checks = pd.DataFrame(checks)

print("\nUFC 330 — RAW DATA RECONCILIATION")
display(raw_checks)

# -----------------------------------------------------------
# 7. Preserve evidence
# -----------------------------------------------------------

breakdown.to_csv(
    OUT / "ufc330_card_quality_breakdown.csv",
    index=False
)
comparison.to_csv(
    OUT / "recent_card_quality_comparison.csv",
    index=False
)
ufc330.to_csv(
    OUT / "ufc330_fight_records.csv",
    index=False
)
ranked.to_csv(
    OUT / "ufc330_fighter_rankings.csv",
    index=False
)
raw_checks.to_csv(
    OUT / "ufc330_raw_factor_checks.csv",
    index=False
)

ufc330_card_breakdown_v1 = breakdown.copy()
ufc330_card_comparison_v1 = comparison.copy()
ufc330_raw_checks_v1 = raw_checks.copy()

print("\n" + "=" * 85)
print("BIG CELL 16 COMPLETE")
print("=" * 85)
print("No historical or recent scores modified.")
print("Evidence saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2 — BIG CELL 17
# UFC 330 EXPERIENCE, RANKING AND NORMALIZATION AUDIT
#
# Diagnostic only:
# - Never overwrites existing scores
# - Never treats missing values as zero
# - Checks raw historical normalization
# - Identifies possible fighter-level experience sources
# - Tests whether 85.06 can currently be approved
# ============================================================

OUT = Path("/content/ufc_330_deep_audit")
OUT.mkdir(parents=True, exist_ok=True)

required = ["backup", "scoring_fights", "star_recent_candidates_v1"]
missing = [v for v in required if v not in globals()]
if missing:
    raise RuntimeError("Missing variables: " + ", ".join(missing))

TARGET = "UFC 330: Makhachev vs. Machado Garry"

fights = scoring_fights.copy()
recent = star_recent_candidates_v1.copy()
hist_raw = backup["event_master"].copy()
hist_norm = backup["normalized"].copy()

ufc330 = fights.loc[fights["event_name"].eq(TARGET)].copy()
event = recent.loc[recent["event_name"].eq(TARGET)].copy()

if len(event) != 1 or len(ufc330) != 12:
    raise RuntimeError(
        f"Unexpected event/fight count: {len(event)}, {len(ufc330)}"
    )

def number(x):
    return pd.to_numeric(x, errors="coerce")

def show(label, df):
    print("\n" + "=" * 78)
    print(label)
    print("=" * 78)
    display(df)

print("BIG CELL 17 — UFC 330 DEEP DATA AUDIT")
print("Event:", TARGET)
print("Fights:", len(ufc330))
print("Fighter appearances:", 2 * len(ufc330))

# ------------------------------------------------------------
# 1. Discover possible fighter experience columns
# ------------------------------------------------------------

terms = [
    "experience", "ufc_fights", "fight_count",
    "prior_fights", "fights_before", "career"
]

candidate_cols = [
    c for c in fights.columns
    if any(t in str(c).lower() for t in terms)
]

inventory = pd.DataFrame([
    {
        "column": c,
        "nonmissing_ufc330": int(ufc330[c].notna().sum()),
        "sample_values": str(
            ufc330[c].dropna().head(3).tolist()
        )
    }
    for c in candidate_cols
])

show("POSSIBLE UFC EXPERIENCE SOURCE COLUMNS", inventory)

# ------------------------------------------------------------
# 2. Inspect both fighters and available experience fields
# ------------------------------------------------------------

identity_cols = [
    c for c in ["f1_name", "f2_name", "event_name"]
    if c in ufc330.columns
]

experience_cols = [
    c for c in candidate_cols
    if str(c).startswith(("f1_", "f2_"))
]

fighter_experience = ufc330[
    identity_cols + experience_cols
].copy()

show("UFC 330 FIGHTER-LEVEL EXPERIENCE RECORDS",
     fighter_experience)

# ------------------------------------------------------------
# 3. Test paired f1/f2 experience fields
# ------------------------------------------------------------

pair_results = []

for c1 in experience_cols:
    if not c1.startswith("f1_"):
        continue

    c2 = "f2_" + c1[3:]

    if c2 not in ufc330.columns:
        continue

    values = pd.concat([
        number(ufc330[c1]),
        number(ufc330[c2])
    ], ignore_index=True)

    nonmissing = int(values.notna().sum())
    avg = values.mean() if nonmissing else np.nan

    pair_results.append({
        "f1_column": c1,
        "f2_column": c2,
        "fighter_values_present": nonmissing,
        "fighter_values_expected": 24,
        "calculated_mean": avg,
        "complete": nonmissing == 24,
        "matches_stored_11_291667": (
            bool(np.isclose(avg, 11.291667, atol=1e-5))
            if pd.notna(avg) else False
        )
    })

pair_audit = pd.DataFrame(pair_results)
show("EXPERIENCE PAIR RECONCILIATION", pair_audit)

# ------------------------------------------------------------
# 4. Compare historical experience normalization
# ------------------------------------------------------------

raw_name = "avg_ufc_experience"
score_name = "avg_ufc_experience_score"

if raw_name not in hist_raw or score_name not in hist_norm:
    print("Historical experience fields not found.")
    historical_lookup = pd.DataFrame()
else:
    historical_lookup = hist_raw[
        ["event_name", raw_name]
    ].merge(
        hist_norm[["event_name", score_name]],
        on="event_name",
        validate="one_to_one"
    )

    historical_lookup[raw_name] = number(
        historical_lookup[raw_name]
    )
    historical_lookup[score_name] = number(
        historical_lookup[score_name]
    )

    historical_lookup = historical_lookup.sort_values(
        raw_name
    )

    raw_value = number(event[raw_name]).iloc[0]
    score_value = number(
        event["candidate__" + score_name]
    ).iloc[0]

    nearby = historical_lookup.copy()
    nearby["distance"] = (
        nearby[raw_name] - raw_value
    ).abs()

    nearby = nearby.sort_values("distance").head(15)

    print("\nUFC 330 stored average experience:", raw_value)
    print("UFC 330 normalized experience score:", score_value)

    show("15 CLOSEST HISTORICAL EXPERIENCE VALUES",
         nearby)

# ------------------------------------------------------------
# 5. Check ranking points and ranked fighter concentration
# ------------------------------------------------------------

if "ranking_points_final_candidate_v3" in globals():
    ranks = ranking_points_final_candidate_v3.copy()
    ranks = ranks.loc[ranks["event_name"].eq(TARGET)].copy()

    point_col = next(
        (
            c for c in ["ranking_points_v2", "ranking_points"]
            if c in ranks.columns
        ),
        None
    )

    if point_col:
        ranks["_audit_points"] = number(ranks[point_col])
        ranked_count = int(ranks["_audit_points"].gt(0).sum())
        total_points = ranks["_audit_points"].sum(
            min_count=1
        )

        print("\nRANKING AUDIT")
        print("Ranked fighter appearances:", ranked_count)
        print("Total ranking points:", total_points)
        print("Share ranked:", f"{ranked_count / len(ranks):.1%}")

        cols = [
            c for c in [
                "fighter", "proposed_rank",
                "decision_category", point_col
            ] if c in ranks.columns
        ]
        show("UFC 330 RANKING EVIDENCE", ranks[cols])
    else:
        print("No ranking point field available.")
else:
    print("Ranking candidate table not found.")

# ------------------------------------------------------------
# 6. Evaluate whether the experience factor is verified
# ------------------------------------------------------------

fully_matched = (
    pair_audit[
        pair_audit["complete"]
        & pair_audit["matches_stored_11_291667"]
    ]
    if len(pair_audit) else pd.DataFrame()
)

experience_reconciled = len(fully_matched) > 0

summary = pd.DataFrame([
    {
        "check": "Card Quality arithmetic",
        "result": "PASS",
        "detail": "85.06202 reproduced in Big Cell 16"
    },
    {
        "check": "Title fight count",
        "result": "PASS_IN_DATASET",
        "detail": "Two stored title fights matched fight flags"
    },
    {
        "check": "Ranking concentration",
        "result": "REVIEW",
        "detail": "Only four of 24 appearances ranked"
    },
    {
        "check": "Average UFC experience",
        "result": (
            "INTERNALLY_RECONCILED"
            if experience_reconciled else
            "NOT_YET_RECONCILED"
        ),
        "detail": (
            "Complete fighter-level pair matches"
            if experience_reconciled else
            "No complete matching fighter-level pair found"
        )
    },
    {
        "check": "Independent external source verification",
        "result": "NOT_DONE",
        "detail": "Requires original fight and fighter records"
    },
    {
        "check": "Approve UFC 330 score as final",
        "result": "NO",
        "detail": "External validation still outstanding"
    }
])

show("FINAL UFC 330 DATA READINESS", summary)

# ------------------------------------------------------------
# 7. Export audit
# ------------------------------------------------------------

inventory.to_csv(
    OUT / "experience_column_inventory.csv", index=False
)
fighter_experience.to_csv(
    OUT / "ufc330_fighter_experience.csv", index=False
)
pair_audit.to_csv(
    OUT / "experience_pair_reconciliation.csv", index=False
)
summary.to_csv(
    OUT / "ufc330_readiness.csv", index=False
)

if len(historical_lookup):
    historical_lookup.to_csv(
        OUT / "historical_experience_lookup.csv",
        index=False
    )

ufc330_experience_audit_v2 = pair_audit.copy()
ufc330_readiness_v2 = summary.copy()

print("\nBIG CELL 17 COMPLETE")
print("Experience internally reconciled:", experience_reconciled)
print("Original event scores unchanged.")
print("Files saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 18 — MAIN EVENT STRENGTH RECONSTRUCTION AUDIT
#
# 1. Verify historical 7-factor scoring formula
# 2. Identify exact factor columns
# 3. Audit recent main-event inputs
# 4. Recover candidate normalized factor scores
# 5. Calculate recent Main Event Strength only if complete
#
# NO ORIGINAL SCORES ARE MODIFIED
# ============================================================

OUT = Path("/content/ufc_main_event_reconstruction")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "backup",
    "main_event_scoring",
    "star_recent_candidates_v1",
    "scoring_fights",
    "ranking_points_final_candidate_v3"
]

missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError("Missing Colab variables: " + ", ".join(missing))

historical_raw = backup["event_master"].copy()
historical_norm = backup["normalized"].copy()
main = main_event_scoring.copy()
recent = star_recent_candidates_v1.copy()
fights = scoring_fights.copy()
rankings = ranking_points_final_candidate_v3.copy()

print("=" * 85)
print("BIG CELL 18 — MAIN EVENT STRENGTH RECONSTRUCTION")
print("=" * 85)

print("Historical events:", len(historical_norm))
print("Main event records:", len(main))
print("Recent events:", len(recent))

# ------------------------------------------------------------
# 1. Discover main event factor columns
# ------------------------------------------------------------

main_columns = [
    c for c in main.columns
    if any(t in c.lower() for t in [
        "main_event", "me_", "ranking", "bonus",
        "title_fight", "win_streak", "win_pct",
        "ufc_fights"
    ])
]

normalized_main_columns = [
    c for c in historical_norm.columns
    if c.startswith("main_event_") and c.endswith("_score")
]

print("\nMAIN EVENT SCORING COLUMNS:")
for c in main_columns:
    print(" -", c)

print("\nHISTORICAL NORMALIZED MAIN EVENT FACTORS:")
for c in normalized_main_columns:
    print(" -", c)

# ------------------------------------------------------------
# 2. Verify historical seven-factor formula
# ------------------------------------------------------------

factor_weights = {
    "main_event_ranking_points_score": 0.25,
    "main_event_combined_title_fights_score": 0.20,
    "main_event_avg_win_pct_score": 0.15,
    "main_event_combined_win_streak_score": 0.15,
    "main_event_combined_previous_main_events_score": 0.10,
    "main_event_combined_ufc_fights_score": 0.10,
    "main_event_combined_bonuses_score": 0.05
}

missing_historical = [
    c for c in factor_weights
    if c not in historical_norm.columns
]

if missing_historical:
    raise RuntimeError(
        "Missing historical main event factors: "
        + str(missing_historical)
    )

historical_main = historical_norm[
    ["event_name"] + list(factor_weights)
].copy()

historical_main["weighted_prediction"] = sum(
    pd.to_numeric(historical_main[c], errors="coerce")
    * w
    for c, w in factor_weights.items()
)

# Compare with stored main-event scoring values.
score_candidates = [
    "corrected_main_event_score",
    "main_event_score",
    "main_event_strength_score"
]

score_col = next(
    (c for c in score_candidates if c in main.columns),
    None
)

if score_col is None:
    raise RuntimeError("No main event target score found.")

historical_main = historical_main.merge(
    main[["event_name", score_col]],
    on="event_name",
    how="left",
    validate="one_to_one"
)

historical_main["actual"] = pd.to_numeric(
    historical_main[score_col], errors="coerce"
)

historical_main["absolute_error"] = (
    historical_main["weighted_prediction"]
    - historical_main["actual"]
).abs()

historical_main["matches"] = np.isclose(
    historical_main["weighted_prediction"],
    historical_main["actual"],
    atol=1e-6,
    rtol=0
)

valid = historical_main[
    historical_main["actual"].notna()
].copy()

print("\n" + "=" * 85)
print("HISTORICAL MAIN EVENT FORMULA VALIDATION")
print("=" * 85)

print("Historical main event scores available:", len(valid))
print("Exact matches:", int(valid["matches"].sum()))
print("Mismatch count:", int((~valid["matches"]).sum()))

if len(valid):
    print("Mean absolute error:",
          valid["absolute_error"].mean())

    print("Maximum absolute error:",
          valid["absolute_error"].max())

    display(
        valid.sort_values(
            "absolute_error", ascending=False
        ).head(12)
    )

# ------------------------------------------------------------
# 3. Determine recent main event rows
# ------------------------------------------------------------

recent_names = set(recent["event_name"])
recent_main = main[
    main["event_name"].isin(recent_names)
].copy()

print("\n" + "=" * 85)
print("RECENT MAIN EVENT INPUT COVERAGE")
print("=" * 85)

print("Recent main events found:", len(recent_main))

input_cols = [
    c for c in recent_main.columns
    if any(t in c.lower() for t in [
        "me_", "main_event_", "ranking",
        "bonus", "win_pct", "win_streak"
    ])
]

coverage = pd.DataFrame([
    {
        "column": c,
        "nonmissing": int(recent_main[c].notna().sum()),
        "missing": int(recent_main[c].isna().sum()),
        "example": str(
            recent_main[c].dropna().head(2).tolist()
        )
    }
    for c in input_cols
])

display(coverage)

# ------------------------------------------------------------
# 4. Historical raw/normalized factor mapping
# ------------------------------------------------------------

historical_factor_pairs = []

for score_factor in factor_weights:

    raw_factor = score_factor.removesuffix("_score")

    if raw_factor not in historical_raw.columns:
        historical_factor_pairs.append({
            "score_factor": score_factor,
            "raw_factor": raw_factor,
            "available": False,
            "unique_raw_values": 0,
            "conflicts": np.nan
        })
        continue

    joined = historical_raw[
        ["event_name", raw_factor]
    ].merge(
        historical_norm[
            ["event_name", score_factor]
        ],
        on="event_name",
        validate="one_to_one"
    )

    joined[raw_factor] = pd.to_numeric(
        joined[raw_factor], errors="coerce"
    )

    joined[score_factor] = pd.to_numeric(
        joined[score_factor], errors="coerce"
    )

    clean = joined.dropna()

    conflicts = (
        clean.groupby(raw_factor)[score_factor]
        .nunique()
        .gt(1)
        .sum()
    )

    historical_factor_pairs.append({
        "score_factor": score_factor,
        "raw_factor": raw_factor,
        "available": True,
        "unique_raw_values": clean[raw_factor].nunique(),
        "conflicts": int(conflicts)
    })

mapping_audit = pd.DataFrame(historical_factor_pairs)

print("\n" + "=" * 85)
print("HISTORICAL RAW-TO-NORMALIZED FACTOR MAPPING")
print("=" * 85)

display(mapping_audit)

# ------------------------------------------------------------
# 5. Check which recent raw main event values exist
# ------------------------------------------------------------

recent_raw_rows = []

for score_factor in factor_weights:
    raw_factor = score_factor.removesuffix("_score")

    candidates = [
        raw_factor,
        raw_factor.replace("main_event_", "me_"),
        "me_" + raw_factor.replace("main_event_", "")
    ]

    found = [
        c for c in candidates
        if c in recent_main.columns
    ]

    best = (
        max(found, key=lambda c:
            recent_main[c].notna().sum())
        if found else None
    )

    recent_raw_rows.append({
        "factor": score_factor,
        "candidate_raw_column": best,
        "recent_nonmissing": (
            int(recent_main[best].notna().sum())
            if best else 0
        ),
        "recent_total": len(recent_main)
    })

recent_factor_coverage = pd.DataFrame(recent_raw_rows)

print("\n" + "=" * 85)
print("RECENT MAIN EVENT FACTOR READINESS")
print("=" * 85)

display(recent_factor_coverage)

# ------------------------------------------------------------
# 6. Produce candidate scores only if fully justified
# ------------------------------------------------------------

recent_main_candidates = recent_main[
    ["event_name"]
].copy()

for factor in factor_weights:
    recent_main_candidates[
        "candidate__" + factor
    ] = np.nan

recent_main_candidates[
    "main_event_strength_candidate"
] = np.nan

print("\nRecent Main Event Strength candidates withheld until")
print("all seven raw factors and their normalization are")
print("verified against historical records.")

# ------------------------------------------------------------
# 7. Save audit files and Colab variables
# ------------------------------------------------------------

historical_main.to_csv(
    OUT / "historical_main_event_formula_audit.csv",
    index=False
)

coverage.to_csv(
    OUT / "recent_main_event_column_coverage.csv",
    index=False
)

mapping_audit.to_csv(
    OUT / "historical_main_event_normalization_mapping.csv",
    index=False
)

recent_factor_coverage.to_csv(
    OUT / "recent_main_event_factor_readiness.csv",
    index=False
)

recent_main_candidates.to_csv(
    OUT / "recent_main_event_candidates.csv",
    index=False
)

main_formula_audit_v1 = historical_main.copy()
main_factor_mapping_v1 = mapping_audit.copy()
main_recent_coverage_v1 = recent_factor_coverage.copy()
main_recent_candidates_v1 = recent_main_candidates.copy()

print("\n" + "=" * 85)
print("BIG CELL 18 COMPLETE")
print("=" * 85)
print("Historical scores unchanged.")
print("Recent scores unchanged.")
print("Audit files saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 19 — MAIN EVENT FORMULA DIAGNOSTICS
#
# Historical formula reconstruction only.
# No original scores or datasets are overwritten.
# ============================================================

OUT = Path("/content/ufc_main_event_formula_diagnostics")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "backup",
    "main_event_scoring",
    "ranking_points_final_candidate_v3"
]
missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError("Missing variables: " + ", ".join(missing))

hist = backup["normalized"].copy()
main = main_event_scoring.copy()
ranks = ranking_points_final_candidate_v3.copy()

weights = {
    "main_event_ranking_points_score": .25,
    "main_event_combined_title_fights_score": .20,
    "main_event_avg_win_pct_score": .15,
    "main_event_combined_win_streak_score": .15,
    "main_event_combined_previous_main_events_score": .10,
    "main_event_combined_ufc_fights_score": .10,
    "main_event_combined_bonuses_score": .05
}

factor_cols = list(weights)
missing_cols = [c for c in factor_cols if c not in hist]
if missing_cols:
    raise RuntimeError("Missing factor columns: " + str(missing_cols))

h = hist[["event_name"] + factor_cols].copy()

for c in factor_cols:
    h[c] = pd.to_numeric(h[c], errors="coerce")

# Restrict comparison to the historical 777 event names.
historical_names = set(h["event_name"])

target_cols = [
    "event_name",
    "corrected_main_event_score",
    "main_event_score_complete"
]
target_cols = [c for c in target_cols if c in main.columns]

targets = main[
    main["event_name"].isin(historical_names)
][target_cols].copy()

if targets["event_name"].duplicated().any():
    raise RuntimeError("Duplicate historical main-event names.")

h = h.merge(
    targets,
    on="event_name",
    how="left",
    validate="one_to_one"
)

if "main_event_score" in hist.columns:
    original = hist[["event_name", "main_event_score"]].copy()
    h = h.merge(
        original,
        on="event_name",
        how="left",
        validate="one_to_one"
    )

# ------------------------------------------------------------
# 1. Reconstruct scores using alternative missing-data rules
# ------------------------------------------------------------

w = pd.Series(weights)
matrix = h[factor_cols]

weighted_parts = matrix.mul(w, axis=1)

# Method A: missing factors make result NaN
h["strict_weighted"] = weighted_parts.sum(
    axis=1, min_count=len(weights)
)

# Method B: missing contributions contribute nothing
# Diagnostic only; NOT an approved imputation rule.
h["zero_contribution_diagnostic"] = weighted_parts.sum(
    axis=1, min_count=1
)

# Method C: redistribute weight across available factors
available_weights = matrix.notna().mul(w, axis=1).sum(axis=1)
h["renormalized_available_weights"] = (
    weighted_parts.sum(axis=1, min_count=1)
    / available_weights
)

# Method D: simple mean of available normalized factors
h["equal_mean_available"] = matrix.mean(axis=1)

h["missing_factor_count"] = matrix.isna().sum(axis=1)
h["available_weight_total"] = available_weights

methods = [
    "strict_weighted",
    "zero_contribution_diagnostic",
    "renormalized_available_weights",
    "equal_mean_available"
]

targets_to_test = [
    c for c in [
        "corrected_main_event_score",
        "main_event_score"
    ] if c in h.columns
]

results = []

for target in targets_to_test:
    actual = pd.to_numeric(h[target], errors="coerce")

    for method in methods:
        predicted = pd.to_numeric(
            h[method], errors="coerce"
        )

        eligible = actual.notna() & predicted.notna()
        diff = (predicted - actual).abs()

        results.append({
            "target": target,
            "method": method,
            "eligible_events": int(eligible.sum()),
            "exact_matches": int(
                (eligible & diff.le(1e-6)).sum()
            ),
            "mismatches": int(
                (eligible & diff.gt(1e-6)).sum()
            ),
            "mean_absolute_error": (
                float(diff[eligible].mean())
                if eligible.any() else np.nan
            ),
            "maximum_error": (
                float(diff[eligible].max())
                if eligible.any() else np.nan
            )
        })

formula_results = pd.DataFrame(results).sort_values(
    ["target", "exact_matches"],
    ascending=[True, False]
)

print("\n" + "=" * 80)
print("FORMULA RECONSTRUCTION COMPARISON")
print("=" * 80)
display(formula_results.round(6))

# ------------------------------------------------------------
# 2. Missingness and mismatches
# ------------------------------------------------------------

if "corrected_main_event_score" in h.columns:
    actual = pd.to_numeric(
        h["corrected_main_event_score"],
        errors="coerce"
    )
    h["strict_error"] = (
        h["strict_weighted"] - actual
    ).abs()
    h["strict_matches"] = h["strict_error"].le(1e-6)

    missingness = h.groupby(
        "missing_factor_count", dropna=False
    ).agg(
        events=("event_name", "size"),
        strict_matches=("strict_matches", "sum"),
        mean_strict_error=("strict_error", "mean")
    ).reset_index()

    print("\nMISMATCHES BY NUMBER OF MISSING FACTORS")
    display(missingness)

    print("\nLARGEST 20 HISTORICAL MISMATCHES")
    display(
        h.sort_values(
            "strict_error", ascending=False
        )[
            ["event_name", "missing_factor_count",
             "available_weight_total",
             "corrected_main_event_score",
             "strict_weighted",
             "renormalized_available_weights",
             "strict_error"] + factor_cols
        ].head(20).round(4)
    )

    if "main_event_score" in h.columns:
        h["historical_vs_corrected_difference"] = (
            pd.to_numeric(
                h["main_event_score"], errors="coerce"
            ) - actual
        ).abs()

        print("\nORIGINAL VS CORRECTED HISTORICAL SCORES")
        print(
            "Exact matches:",
            int(
                h["historical_vs_corrected_difference"]
                .le(1e-6).sum()
            ),
            "/", len(h)
        )
        print(
            "Maximum difference:",
            h["historical_vs_corrected_difference"].max()
        )

# ------------------------------------------------------------
# 3. Inspect historical missing-factor patterns
# ------------------------------------------------------------

pattern = matrix.isna().copy()
pattern.columns = [
    c.replace("main_event_", "").replace("_score", "")
    for c in pattern.columns
]

patterns = (
    pattern.value_counts(dropna=False)
    .rename("event_count")
    .reset_index()
    .sort_values("event_count", ascending=False)
)

print("\nHISTORICAL MISSING FACTOR PATTERNS")
display(patterns.head(20))

# ------------------------------------------------------------
# 4. Discover recent ranking fields and fighter identities
# ------------------------------------------------------------

recent_names = set(ranks["event_name"])

recent_main = main[
    main["event_name"].isin(recent_names)
].copy()

ranking_columns = [
    c for c in ranks.columns
    if any(t in c.lower() for t in [
        "fighter", "rank", "point", "division",
        "decision", "snapshot"
    ])
]

main_identity_columns = [
    c for c in recent_main.columns
    if any(t in c.lower() for t in [
        "f1_name", "f2_name", "fighter",
        "ranking", "rank"
    ])
]

print("\nRANKING AUDIT COLUMNS")
print(ranking_columns)

print("\nMAIN EVENT FIGHTER / RANKING COLUMNS")
print(main_identity_columns)

print("\nRECENT MAIN EVENT IDENTITIES")
show_cols = [
    c for c in [
        "event_name", "f1_name", "f2_name",
        "corrected_main_event_ranking_points",
        "main_event_ranking_points_score"
    ] if c in recent_main.columns
]
display(recent_main[show_cols].reset_index(drop=True))

print("\nRANKING AUDIT SAMPLE")
display(
    ranks[
        ["event_name"] + ranking_columns
    ].head(12)
)

# ------------------------------------------------------------
# 5. Save all evidence
# ------------------------------------------------------------

formula_results.to_csv(
    OUT / "formula_comparison.csv", index=False
)
h.to_csv(
    OUT / "historical_formula_diagnostics.csv",
    index=False
)
patterns.to_csv(
    OUT / "historical_missingness_patterns.csv",
    index=False
)

main_formula_diagnostics_v2 = h.copy()
main_formula_comparison_v2 = formula_results.copy()
main_missingness_patterns_v2 = patterns.copy()

print("\n" + "=" * 80)
print("BIG CELL 19 COMPLETE")
print("=" * 80)
print("No original data modified.")
print("Files saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 20 — CORRECTED MAIN EVENT SCORE FORENSICS
#
# Diagnose historical differences and recent rankings.
# No original scores are modified.
# ============================================================

OUT = Path("/content/ufc_main_event_forensics")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "main_formula_diagnostics_v2",
    "main_event_scoring",
    "ranking_points_final_candidate_v3",
    "backup"
]
missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError("Missing objects: " + ", ".join(missing))

h = main_formula_diagnostics_v2.copy()
main = main_event_scoring.copy()
ranks = ranking_points_final_candidate_v3.copy()
raw = backup["event_master"].copy()

weights = {
    "main_event_ranking_points_score": .25,
    "main_event_combined_title_fights_score": .20,
    "main_event_avg_win_pct_score": .15,
    "main_event_combined_win_streak_score": .15,
    "main_event_combined_previous_main_events_score": .10,
    "main_event_combined_ufc_fights_score": .10,
    "main_event_combined_bonuses_score": .05
}

factors = list(weights)

print("=" * 85)
print("BIG CELL 20 — MAIN EVENT SCORE FORENSICS")
print("=" * 85)

# ------------------------------------------------------------
# 1. Focus on complete historical rows
# ------------------------------------------------------------

complete = h[
    h[factors].notna().all(axis=1)
    & h["corrected_main_event_score"].notna()
].copy()

complete["expected"] = sum(
    complete[c] * weight
    for c, weight in weights.items()
)

complete["signed_difference"] = (
    complete["corrected_main_event_score"]
    - complete["expected"]
)

complete["absolute_difference"] = (
    complete["signed_difference"].abs()
)

complete["mismatch"] = (
    complete["absolute_difference"] > 1e-6
)

print("\nCOMPLETE HISTORICAL ROWS")
print("Rows:", len(complete))
print("Matches:", int((~complete["mismatch"]).sum()))
print("Mismatches:", int(complete["mismatch"].sum()))

print("\nDIFFERENCE DISTRIBUTION")
display(
    complete["signed_difference"]
    .describe()
    .to_frame("difference")
)

print("\nMOST COMMON NONZERO DIFFERENCES")
display(
    complete.loc[
        complete["mismatch"], "signed_difference"
    ].round(6).value_counts().head(20)
)

# ------------------------------------------------------------
# 2. Identify which factors are associated with mismatches
# ------------------------------------------------------------

factor_comparison = []

for c in factors:
    matched = complete.loc[~complete["mismatch"], c]
    mismatched = complete.loc[complete["mismatch"], c]

    factor_comparison.append({
        "factor": c,
        "matched_mean": matched.mean(),
        "mismatched_mean": mismatched.mean(),
        "matched_median": matched.median(),
        "mismatched_median": mismatched.median()
    })

factor_comparison = pd.DataFrame(factor_comparison)

print("\nFACTOR COMPARISON")
display(factor_comparison.round(4))

# ------------------------------------------------------------
# 3. Compare raw main-event data for mismatched events
# ------------------------------------------------------------

raw_main_cols = [
    c for c in raw.columns
    if c.startswith("main_event_")
    and not c.endswith("_score")
]

raw_subset = raw[
    ["event_name"] + raw_main_cols
].copy()

detail = complete.merge(
    raw_subset,
    on="event_name",
    how="left",
    validate="one_to_one"
)

print("\nLARGEST 25 COMPLETE-DATA MISMATCHES")
display(
    detail.sort_values(
        "absolute_difference", ascending=False
    )[
        ["event_name", "corrected_main_event_score",
         "expected", "signed_difference"]
        + factors
    ].head(25).round(5)
)

# ------------------------------------------------------------
# 4. Check corrected score against main_event_scoring
# ------------------------------------------------------------

historical_main = main[
    main["event_name"].isin(
        set(complete["event_name"])
    )
].copy()

extra_cols = [
    "event_name",
    "corrected_main_event_ranking_points",
    "corrected_f1_rank_points",
    "corrected_f2_rank_points",
    "me_combined_win_streak",
    "me_combined_title_fights",
    "me_combined_previous_main_events",
    "me_combined_ufc_fights",
    "me_combined_bonuses",
    "main_event_score_complete"
]

extra_cols = [
    c for c in extra_cols
    if c in historical_main.columns
]

extra = historical_main[extra_cols].copy()

detail = detail.merge(
    extra,
    on="event_name",
    how="left",
    validate="one_to_one",
    suffixes=("", "_main")
)

print("\nRAW VALUES FOR LARGEST MISMATCHES")
raw_show = [
    c for c in [
        "event_name",
        "signed_difference",
        "main_event_ranking_points",
        "corrected_main_event_ranking_points",
        "me_combined_win_streak",
        "me_combined_title_fights",
        "me_combined_previous_main_events",
        "me_combined_ufc_fights",
        "me_combined_bonuses"
    ] if c in detail.columns
]

display(
    detail.sort_values(
        "absolute_difference", ascending=False
    )[raw_show].head(25)
)

# ------------------------------------------------------------
# 5. Recent main-event ranking reconstruction
# ------------------------------------------------------------

recent_names = (
    set(ranks["event_name"])
    - set(raw["event_name"])
)

recent_main = main[
    main["event_name"].isin(recent_names)
].copy()

rank_points_col = next(
    (
        c for c in [
            "ranking_points_v2",
            "final_ranking_points",
            "proposed_points"
        ]
        if c in ranks.columns
        and pd.to_numeric(
            ranks[c], errors="coerce"
        ).notna().any()
    ),
    None
)

if rank_points_col is None:
    raise RuntimeError(
        "No usable fighter ranking points column found."
    )

ranks["points_for_audit"] = pd.to_numeric(
    ranks[rank_points_col], errors="coerce"
)

recent_ranks = ranks[
    ranks["event_name"].isin(recent_names)
].copy()

def normalize_name(value):
    if pd.isna(value):
        return ""
    import unicodedata
    import re
    value = unicodedata.normalize(
        "NFKD", str(value)
    )
    value = "".join(
        ch for ch in value
        if not unicodedata.combining(ch)
    )
    return re.sub(
        r"[^a-z0-9]+", " ", value.lower()
    ).strip()

recent_ranks["name_key"] = (
    recent_ranks["fighter"].map(normalize_name)
)

if recent_ranks.duplicated(
    ["event_name", "name_key"]
).any():
    raise RuntimeError(
        "Duplicate event/fighter ranking entries."
    )

rank_lookup = recent_ranks.set_index(
    ["event_name", "name_key"]
)["points_for_audit"]

recent_results = []

for _, row in recent_main.iterrows():
    event = row["event_name"]
    f1 = row.get("f1_name")
    f2 = row.get("f2_name")

    key1 = (event, normalize_name(f1))
    key2 = (event, normalize_name(f2))

    p1 = (
        rank_lookup.loc[key1]
        if key1 in rank_lookup.index
        else np.nan
    )
    p2 = (
        rank_lookup.loc[key2]
        if key2 in rank_lookup.index
        else np.nan
    )

    valid = pd.notna(p1) and pd.notna(p2)

    recent_results.append({
        "event_name": event,
        "fighter_1": f1,
        "fighter_2": f2,
        "fighter_1_points": p1,
        "fighter_2_points": p2,
        "combined_ranking_points": (
            p1 + p2 if valid else np.nan
        ),
        "both_fighters_resolved": valid
    })

recent_ranking_audit = pd.DataFrame(recent_results)

print("\nRECENT MAIN EVENT RANKING POINTS")
display(recent_ranking_audit)

print(
    "\nResolved main events:",
    int(
        recent_ranking_audit[
            "both_fighters_resolved"
        ].sum()
    ),
    "/",
    len(recent_ranking_audit)
)

print(
    "\nRanking source remains provisional:",
    "third-party historical archive"
)

# ------------------------------------------------------------
# 6. Save audit evidence
# ------------------------------------------------------------

complete.to_csv(
    OUT / "complete_historical_differences.csv",
    index=False
)

factor_comparison.to_csv(
    OUT / "mismatch_factor_comparison.csv",
    index=False
)

detail.to_csv(
    OUT / "historical_mismatch_details.csv",
    index=False
)

recent_ranking_audit.to_csv(
    OUT / "recent_main_event_ranking_points.csv",
    index=False
)

main_complete_mismatch_v3 = complete.copy()
main_mismatch_details_v3 = detail.copy()
main_recent_ranking_audit_v3 = recent_ranking_audit.copy()

print("\n" + "=" * 85)
print("BIG CELL 20 COMPLETE")
print("=" * 85)
print("No original scores changed.")
print("Evidence saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 21 — HISTORICAL FACTOR DIFFERENCE DETECTIVE
# Read-only audit. Does not modify existing scores.
# ============================================================

OUT = Path("/content/ufc_main_factor_detective")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "main_complete_mismatch_v3",
    "main_event_scoring",
    "backup"
]
missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError("Missing objects: " + ", ".join(missing))

h = main_complete_mismatch_v3.copy()
main = main_event_scoring.copy()
norm = backup["normalized"].copy()

weights = {
    "main_event_ranking_points_score": .25,
    "main_event_combined_title_fights_score": .20,
    "main_event_avg_win_pct_score": .15,
    "main_event_combined_win_streak_score": .15,
    "main_event_combined_previous_main_events_score": .10,
    "main_event_combined_ufc_fights_score": .10,
    "main_event_combined_bonuses_score": .05
}

factors = list(weights)

print("=" * 85)
print("BIG CELL 21 — HISTORICAL FACTOR DIFFERENCE DETECTIVE")
print("=" * 85)

# ------------------------------------------------------------
# 1. Recalculate baseline and identify affected events
# ------------------------------------------------------------

h["baseline_recalculated"] = sum(
    pd.to_numeric(h[c], errors="coerce") * w
    for c, w in weights.items()
)

h["residual"] = (
    pd.to_numeric(
        h["corrected_main_event_score"], errors="coerce"
    ) - h["baseline_recalculated"]
)

h["affected"] = h["residual"].abs().gt(1e-6)

print("\nCOMPLETE HISTORICAL EVENTS:", len(h))
print("MATCHED:", int((~h["affected"]).sum()))
print("DIFFERENT:", int(h["affected"].sum()))

# ------------------------------------------------------------
# 2. Find all alternative normalized score columns
# ------------------------------------------------------------

score_columns = [
    c for c in norm.columns
    if c.endswith("_score")
    and c not in factors
    and c not in [
        "main_event_score",
        "card_quality_score",
        "star_experience_score",
        "overall_score",
        "event_strength_score"
    ]
]

alternative_tests = []

for factor, weight in weights.items():
    base = pd.to_numeric(h[factor], errors="coerce")
    baseline = h["baseline_recalculated"]

    for alternative in score_columns:
        alt_map = (
            norm[["event_name", alternative]]
            .drop_duplicates("event_name")
            .set_index("event_name")[alternative]
        )

        replacement = pd.to_numeric(
            h["event_name"].map(alt_map),
            errors="coerce"
        )

        eligible = replacement.notna() & base.notna()
        if not eligible.any():
            continue

        prediction = (
            baseline + weight * (replacement - base)
        )

        error = (
            prediction
            - h["corrected_main_event_score"]
        ).abs()

        alternative_tests.append({
            "replaced_factor": factor,
            "alternative_factor": alternative,
            "eligible": int(eligible.sum()),
            "exact_matches": int(
                (eligible & error.le(1e-6)).sum()
            ),
            "remaining_mismatches": int(
                (eligible & error.gt(1e-6)).sum()
            ),
            "mean_absolute_error": float(
                error[eligible].mean()
            ),
            "max_error": float(error[eligible].max())
        })

alternative_results = pd.DataFrame(alternative_tests)

if not alternative_results.empty:
    alternative_results = alternative_results.sort_values(
        ["exact_matches", "mean_absolute_error"],
        ascending=[False, True]
    )

    print("\nTOP 25 SINGLE-FACTOR REPLACEMENT TESTS")
    display(alternative_results.head(25).round(6))
else:
    print("\nNo alternative score columns found.")

# ------------------------------------------------------------
# 3. Test whether the score contains an extra bonus
# ------------------------------------------------------------

test_rows = []

for c in factors:
    values = pd.to_numeric(h[c], errors="coerce")

    for sign in [-1, 1]:
        for weight in [.05, .10, .15, .20, .25]:
            prediction = (
                h["baseline_recalculated"]
                + sign * weight * values
            )

            error = (
                prediction
                - h["corrected_main_event_score"]
            ).abs()

            test_rows.append({
                "factor": c,
                "operation": (
                    "add" if sign == 1 else "subtract"
                ),
                "extra_weight": weight,
                "exact_matches": int(
                    error.le(1e-6).sum()
                ),
                "mean_error": float(error.mean())
            })

extra_weight_tests = (
    pd.DataFrame(test_rows)
    .sort_values(
        ["exact_matches", "mean_error"],
        ascending=[False, True]
    )
)

print("\nTOP 15 EXTRA-WEIGHT TESTS")
display(extra_weight_tests.head(15).round(6))

# ------------------------------------------------------------
# 4. Check whether differences align with title history
# ------------------------------------------------------------

main_cols = [
    c for c in main.columns
    if any(word in c.lower() for word in [
        "title", "champion", "streak", "corrected"
    ])
    and c != "event_name"
]

main_unique = (
    main[["event_name"] + main_cols]
    .drop_duplicates("event_name")
)

diagnostic = h[
    ["event_name", "residual", "affected"]
].merge(
    main_unique,
    on="event_name",
    how="left",
    validate="one_to_one"
)

numeric_tests = []

for c in main_cols:
    vals = pd.to_numeric(
        diagnostic[c], errors="coerce"
    )

    eligible = vals.notna() & diagnostic["residual"].notna()

    if eligible.sum() < 20 or vals[eligible].nunique() < 2:
        continue

    corr = vals[eligible].corr(
        diagnostic.loc[eligible, "residual"]
    )

    numeric_tests.append({
        "column": c,
        "correlation_with_difference": corr,
        "available_rows": int(eligible.sum()),
        "affected_mean": vals[
            eligible & diagnostic["affected"]
        ].mean(),
        "unaffected_mean": vals[
            eligible & ~diagnostic["affected"]
        ].mean()
    })

correlations = pd.DataFrame(numeric_tests)

if not correlations.empty:
    correlations["absolute_correlation"] = (
        correlations["correlation_with_difference"].abs()
    )

    correlations = correlations.sort_values(
        "absolute_correlation", ascending=False
    )

    print("\nTITLE / STREAK FACTOR ASSOCIATIONS")
    display(correlations.head(20).round(5))

# ------------------------------------------------------------
# 5. Detect mathematical patterns in residuals
# ------------------------------------------------------------

affected = h[h["affected"]].copy()

print("\nALL 51 DIFFERENCES, SORTED")
display(
    affected[
        ["event_name", "residual"]
    ].sort_values(
        "residual", ascending=False
    ).round(6)
)

print("\nRESIDUALS AS MULTIPLES OF EACH FACTOR WEIGHT")
multiples = affected[
    ["event_name", "residual"]
].copy()

for factor, weight in weights.items():
    label = factor.replace("main_event_", "")
    multiples["implied_" + label] = (
        affected["residual"] / weight
    )

display(multiples.head(15).round(4))

# ------------------------------------------------------------
# 6. Preserve outputs
# ------------------------------------------------------------

if not alternative_results.empty:
    alternative_results.to_csv(
        OUT / "alternative_factor_tests.csv",
        index=False
    )

extra_weight_tests.to_csv(
    OUT / "extra_weight_tests.csv",
    index=False
)

if not correlations.empty:
    correlations.to_csv(
        OUT / "factor_correlations.csv",
        index=False
    )

affected.to_csv(
    OUT / "affected_historical_events.csv",
    index=False
)

main_factor_detective_v4 = affected.copy()
main_alternative_tests_v4 = alternative_results.copy()
main_factor_correlations_v4 = correlations.copy()

print("\n" + "=" * 85)
print("BIG CELL 21 COMPLETE")
print("=" * 85)
print("Original scores unchanged.")
print("Audit files saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 22 — WIN STREAK CORRECTION INVESTIGATION
# READ-ONLY: no original scores changed
# ============================================================

OUT = Path("/content/ufc_win_streak_investigation")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "main_complete_mismatch_v3",
    "main_event_scoring",
    "backup"
]
missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError("Missing: " + ", ".join(missing))

h = main_complete_mismatch_v3.copy()
main = main_event_scoring.copy()
norm = backup["normalized"].copy()

weights = {
    "main_event_ranking_points_score": .25,
    "main_event_combined_title_fights_score": .20,
    "main_event_avg_win_pct_score": .15,
    "main_event_combined_win_streak_score": .15,
    "main_event_combined_previous_main_events_score": .10,
    "main_event_combined_ufc_fights_score": .10,
    "main_event_combined_bonuses_score": .05
}

factors = list(weights)

h["baseline"] = sum(
    pd.to_numeric(h[c], errors="coerce") * w
    for c, w in weights.items()
)

h["residual"] = (
    pd.to_numeric(
        h["corrected_main_event_score"],
        errors="coerce"
    ) - h["baseline"]
)

h["affected"] = h["residual"].abs() > 1e-6

print("=" * 85)
print("BIG CELL 22 — WIN STREAK CORRECTION INVESTIGATION")
print("=" * 85)

# ------------------------------------------------------------
# 1. Collect original and corrected fighter streaks
# ------------------------------------------------------------

streak_columns = [
    c for c in main.columns
    if "streak" in c.lower()
]

print("\nAVAILABLE STREAK COLUMNS")
for c in streak_columns:
    print(c)

selected = [
    "event_name",
    "f1_win_streak_before",
    "f2_win_streak_before",
    "f1_win_streak_before_original",
    "f2_win_streak_before_original",
    "me_combined_win_streak"
]

selected = [c for c in selected if c in main.columns]

streak = main[selected].copy()

if streak["event_name"].duplicated().any():
    raise RuntimeError("Duplicate event names in main_event_scoring.")

d = h.merge(
    streak,
    on="event_name",
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------------------
# 2. Reconstruct original and corrected combined streaks
# ------------------------------------------------------------

required_streaks = [
    "f1_win_streak_before",
    "f2_win_streak_before",
    "f1_win_streak_before_original",
    "f2_win_streak_before_original"
]

missing_streaks = [
    c for c in required_streaks if c not in d.columns
]

if missing_streaks:
    raise RuntimeError(
        "Missing streak fields: " + str(missing_streaks)
    )

for c in required_streaks:
    d[c] = pd.to_numeric(d[c], errors="coerce")

d["combined_corrected_streak"] = (
    d["f1_win_streak_before"]
    + d["f2_win_streak_before"]
)

d["combined_original_streak"] = (
    d["f1_win_streak_before_original"]
    + d["f2_win_streak_before_original"]
)

d["streak_raw_difference"] = (
    d["combined_corrected_streak"]
    - d["combined_original_streak"]
)

print("\nSTREAK CHANGE SUMMARY")
print(
    "Events with changed streaks:",
    int(d["streak_raw_difference"].abs().gt(1e-9).sum())
)
print(
    "Affected events with changed streaks:",
    int(
        (
            d["affected"]
            & d["streak_raw_difference"].abs().gt(1e-9)
        ).sum()
    )
)
print(
    "Unaffected events with changed streaks:",
    int(
        (
            ~d["affected"]
            & d["streak_raw_difference"].abs().gt(1e-9)
        ).sum()
    )
)

print("\nSTREAK CHANGES VS SCORE MISMATCHES")
display(
    pd.crosstab(
        d["affected"],
        d["streak_raw_difference"].abs().gt(1e-9),
        rownames=["Score mismatch"],
        colnames=["Streak changed"]
    )
)

# ------------------------------------------------------------
# 3. Historical normalization of combined streaks
# ------------------------------------------------------------

raw_col = "me_combined_win_streak"
score_col = "main_event_combined_win_streak_score"

if raw_col not in main.columns:
    raise RuntimeError("Missing historical raw streak column.")

lookup = main[
    ["event_name", raw_col]
].merge(
    norm[["event_name", score_col]],
    on="event_name",
    how="inner",
    validate="one_to_one"
)

lookup[raw_col] = pd.to_numeric(
    lookup[raw_col], errors="coerce"
)
lookup[score_col] = pd.to_numeric(
    lookup[score_col], errors="coerce"
)

lookup = lookup.dropna().sort_values(raw_col)

conflicts = (
    lookup.groupby(raw_col)[score_col]
    .agg(lambda x: x.max() - x.min())
)

print("\nRAW-TO-SCORE NORMALIZATION")
print("Observed raw values:", lookup[raw_col].nunique())
print(
    "Conflicting scores for identical raw values:",
    int((conflicts > 1e-6).sum())
)

curve = (
    lookup.groupby(raw_col, as_index=False)[score_col]
    .mean()
    .sort_values(raw_col)
)

def normalize_streak(values):
    vals = pd.to_numeric(values, errors="coerce")
    result = pd.Series(np.nan, index=vals.index, dtype=float)
    valid = vals.notna()
    if valid.any():
        result.loc[valid] = np.interp(
            vals.loc[valid].to_numpy(dtype=float),
            curve[raw_col].to_numpy(dtype=float),
            curve[score_col].to_numpy(dtype=float)
        )
    return result

d["original_streak_normalized_candidate"] = (
    normalize_streak(d["combined_original_streak"])
)

d["corrected_streak_normalized_candidate"] = (
    normalize_streak(d["combined_corrected_streak"])
)

# ------------------------------------------------------------
# 4. Test whether original streak normalization explains
#    the historical corrected score
# ------------------------------------------------------------

streak_weight = weights[score_col]

d["prediction_using_original_streak"] = (
    d["baseline"]
    + streak_weight * (
        d["original_streak_normalized_candidate"]
        - d[score_col]
    )
)

d["prediction_using_corrected_streak"] = (
    d["baseline"]
    + streak_weight * (
        d["corrected_streak_normalized_candidate"]
        - d[score_col]
    )
)

target = pd.to_numeric(
    d["corrected_main_event_score"],
    errors="coerce"
)

results = []

for label, prediction_col in [
    ("Stored factor baseline", "baseline"),
    ("Original streak candidate",
     "prediction_using_original_streak"),
    ("Corrected streak candidate",
     "prediction_using_corrected_streak")
]:
    predicted = pd.to_numeric(
        d[prediction_col], errors="coerce"
    )
    eligible = predicted.notna() & target.notna()
    error = (predicted - target).abs()

    results.append({
        "method": label,
        "eligible": int(eligible.sum()),
        "matches": int(
            (eligible & error.le(1e-6)).sum()
        ),
        "mismatches": int(
            (eligible & error.gt(1e-6)).sum()
        ),
        "mean_error": float(error[eligible].mean()),
        "maximum_error": float(error[eligible].max())
    })

results = pd.DataFrame(results)

print("\nFORMULA COMPARISON")
display(results.round(6))

# ------------------------------------------------------------
# 5. Inspect the largest mismatches
# ------------------------------------------------------------

d["original_streak_error"] = (
    d["prediction_using_original_streak"] - target
).abs()

print("\nTOP 25 MISMATCHES — ORIGINAL STREAK TEST")
display(
    d.sort_values(
        "original_streak_error", ascending=False
    )[
        [
            "event_name",
            "residual",
            "combined_original_streak",
            "combined_corrected_streak",
            "streak_raw_difference",
            score_col,
            "original_streak_normalized_candidate",
            "prediction_using_original_streak",
            "corrected_main_event_score",
            "original_streak_error"
        ]
    ].head(25).round(5)
)

# ------------------------------------------------------------
# 6. Save audit
# ------------------------------------------------------------

d.to_csv(
    OUT / "historical_streak_comparison.csv",
    index=False
)

results.to_csv(
    OUT / "streak_formula_results.csv",
    index=False
)

curve.to_csv(
    OUT / "historical_streak_normalization_curve.csv",
    index=False
)

main_streak_diagnostics_v5 = d.copy()
main_streak_results_v5 = results.copy()
main_streak_curve_v5 = curve.copy()

print("\n" + "=" * 85)
print("BIG CELL 22 COMPLETE")
print("=" * 85)
print("Historical scores unchanged.")
print("Audit files:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 23 — MAIN EVENT CORRECTION FORENSICS
# Read-only. No original scores are modified.
# ============================================================

OUT = Path("/content/ufc_main_event_correction_forensics")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "main_streak_diagnostics_v5",
    "main_event_scoring",
    "backup"
]

missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError("Missing Colab variables: " + ", ".join(missing))

d = main_streak_diagnostics_v5.copy()
main = main_event_scoring.copy()
norm = backup["normalized"].copy()
raw = backup["event_master"].copy()

TOL = 1e-6

print("=" * 85)
print("BIG CELL 23 — MAIN EVENT CORRECTION FORENSICS")
print("=" * 85)

# ------------------------------------------------------------
# 1. Separate the historical mismatch groups
# ------------------------------------------------------------

d["score_mismatch"] = d["residual"].abs() > TOL
d["streak_changed"] = (
    d["streak_raw_difference"].abs() > TOL
)

d["mismatch_group"] = np.select(
    [
        d["score_mismatch"] & d["streak_changed"],
        d["score_mismatch"] & ~d["streak_changed"],
        ~d["score_mismatch"] & d["streak_changed"]
    ],
    [
        "Mismatch + streak changed",
        "Mismatch + streak unchanged",
        "Match + streak changed"
    ],
    default="Match + streak unchanged"
)

print("\nHISTORICAL EVENT GROUPS")
display(
    d["mismatch_group"]
    .value_counts()
    .rename_axis("group")
    .reset_index(name="events")
)

# ------------------------------------------------------------
# 2. Check original normalized streak factor against
#    the copy stored in main_event_scoring
# ------------------------------------------------------------

score_col = "main_event_combined_win_streak_score"

if score_col in main.columns:
    comparison = main[
        ["event_name", score_col]
    ].rename(columns={
        score_col: "score_from_main_event_scoring"
    })

    d = d.merge(
        comparison,
        on="event_name",
        how="left",
        validate="one_to_one"
    )

    d["stored_score_difference"] = (
        pd.to_numeric(
            d["score_from_main_event_scoring"],
            errors="coerce"
        )
        - pd.to_numeric(d[score_col], errors="coerce")
    )

    print("\nSTORED NORMALIZED STREAK SCORE COMPARISON")
    print(
        "Different score values:",
        int(
            d["stored_score_difference"]
            .abs().gt(TOL).sum()
        )
    )

    display(
        d.groupby("mismatch_group")[
            "stored_score_difference"
        ].agg(["count", "min", "max", "mean"])
    )

# ------------------------------------------------------------
# 3. Check raw streak consistency
# ------------------------------------------------------------

if "me_combined_win_streak" in d.columns:
    d["stored_raw_streak"] = pd.to_numeric(
        d["me_combined_win_streak"],
        errors="coerce"
    )

    d["stored_vs_corrected_raw"] = (
        d["stored_raw_streak"]
        - d["combined_corrected_streak"]
    )

    d["stored_vs_original_raw"] = (
        d["stored_raw_streak"]
        - d["combined_original_streak"]
    )

    print("\nSTORED RAW STREAK AGREEMENT")
    print(
        "Matches corrected streak:",
        int(
            d["stored_vs_corrected_raw"]
            .abs().le(TOL).sum()
        )
    )
    print(
        "Matches original streak:",
        int(
            d["stored_vs_original_raw"]
            .abs().le(TOL).sum()
        )
    )

# ------------------------------------------------------------
# 4. Determine whether streak changes explain residuals
#    statistically (diagnostic only, not a new model)
# ------------------------------------------------------------

changed = d[
    d["score_mismatch"] & d["streak_changed"]
].copy()

if len(changed) >= 3:
    x = changed["streak_raw_difference"].to_numpy(float)
    y = changed["residual"].to_numpy(float)

    slope, intercept = np.polyfit(x, y, 1)
    fitted = slope * x + intercept

    sse = float(np.sum((y - fitted) ** 2))
    sst = float(np.sum((y - np.mean(y)) ** 2))
    r2 = 1 - sse / sst if sst > 0 else np.nan

    print("\nSTREAK CHANGE VS SCORING DIFFERENCE")
    print("Changed-streak mismatches:", len(changed))
    print("Correlation:", round(float(np.corrcoef(x, y)[0, 1]), 6))
    print("Diagnostic slope:", round(float(slope), 6))
    print("Diagnostic intercept:", round(float(intercept), 6))
    print("Diagnostic R-squared:", round(float(r2), 6))

    changed["diagnostic_fitted_residual"] = fitted
    changed["diagnostic_fit_error"] = y - fitted

    print("\nCHANGED-STREAK EVENTS")
    display(
        changed[
            [
                "event_name",
                "combined_original_streak",
                "combined_corrected_streak",
                "streak_raw_difference",
                "residual",
                "diagnostic_fitted_residual",
                "diagnostic_fit_error"
            ]
        ].sort_values(
            "residual", ascending=False
        ).round(5)
    )

# ------------------------------------------------------------
# 5. Inspect all 21 mismatches without a streak change
# ------------------------------------------------------------

unchanged = d[
    d["score_mismatch"] & ~d["streak_changed"]
].copy()

print("\nMISMATCHES WITHOUT STREAK CHANGES")
print("Count:", len(unchanged))

show_cols = [
    "event_name",
    "residual",
    "combined_original_streak",
    "combined_corrected_streak",
    "main_event_ranking_points_score",
    "main_event_combined_title_fights_score",
    "main_event_avg_win_pct_score",
    "main_event_combined_win_streak_score",
    "main_event_combined_previous_main_events_score",
    "main_event_combined_ufc_fights_score",
    "main_event_combined_bonuses_score"
]

show_cols = [c for c in show_cols if c in unchanged.columns]

display(
    unchanged[show_cols]
    .sort_values("residual", ascending=False)
    .round(5)
)

# ------------------------------------------------------------
# 6. Inspect source columns that could reflect revisions
# ------------------------------------------------------------

revision_terms = [
    "original", "corrected", "verified",
    "adjusted", "revised", "streak",
    "title", "ranking", "score"
]

source_columns = [
    c for c in main.columns
    if any(term in c.lower() for term in revision_terms)
]

print("\nPOSSIBLE REVISION / CORRECTION COLUMNS")
for c in source_columns:
    print(" -", c)

# ------------------------------------------------------------
# 7. Compare residuals to raw win-streak changes
#    without fitting or replacing the historical scores
# ------------------------------------------------------------

summary = (
    d.groupby("mismatch_group", dropna=False)
    .agg(
        events=("event_name", "size"),
        mean_residual=("residual", "mean"),
        min_residual=("residual", "min"),
        max_residual=("residual", "max"),
        mean_streak_change=("streak_raw_difference", "mean")
    )
    .reset_index()
)

print("\nFINAL DIAGNOSTIC SUMMARY")
display(summary.round(6))

# ------------------------------------------------------------
# 8. Save results and keep variables in Colab
# ------------------------------------------------------------

d.to_csv(
    OUT / "all_historical_correction_groups.csv",
    index=False
)

changed.to_csv(
    OUT / "changed_streak_mismatches.csv",
    index=False
)

unchanged.to_csv(
    OUT / "unchanged_streak_mismatches.csv",
    index=False
)

summary.to_csv(
    OUT / "correction_summary.csv",
    index=False
)

main_correction_groups_v6 = d.copy()
main_changed_streak_mismatches_v6 = changed.copy()
main_unchanged_streak_mismatches_v6 = unchanged.copy()
main_correction_summary_v6 = summary.copy()

print("\n" + "=" * 85)
print("BIG CELL 23 COMPLETE")
print("=" * 85)
print("No original data or scores modified.")
print("Saved audit files:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 24 — CORRECTED FACTOR RECONSTRUCTION
# Read-only investigation; no existing scores modified.
# ============================================================

OUT = Path("/content/ufc_corrected_factor_reconstruction")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "backup",
    "main_event_scoring",
    "main_correction_groups_v6"
]
missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError("Missing variables: " + ", ".join(missing))

old = backup["normalized"].copy()
new = main_event_scoring.copy()
groups = main_correction_groups_v6.copy()

weights = {
    "main_event_ranking_points_score": 0.25,
    "main_event_combined_title_fights_score": 0.20,
    "main_event_avg_win_pct_score": 0.15,
    "main_event_combined_win_streak_score": 0.15,
    "main_event_combined_previous_main_events_score": 0.10,
    "main_event_combined_ufc_fights_score": 0.10,
    "main_event_combined_bonuses_score": 0.05
}

factors = list(weights)
TOL = 1e-6

print("=" * 85)
print("BIG CELL 24 — CORRECTED FACTOR RECONSTRUCTION")
print("=" * 85)

# ------------------------------------------------------------
# 1. Validate event keys
# ------------------------------------------------------------

for label, df in [
    ("Historical normalized", old),
    ("Main event scoring", new),
    ("Previous diagnostic", groups)
]:
    if "event_name" not in df.columns:
        raise RuntimeError(f"{label} lacks event_name.")
    if df["event_name"].isna().any():
        raise RuntimeError(f"{label} contains null event names.")
    if df["event_name"].duplicated().any():
        raise RuntimeError(f"{label} has duplicate event names.")

missing_old = [c for c in factors if c not in old.columns]
missing_new = [c for c in factors if c not in new.columns]

if missing_old or missing_new:
    raise RuntimeError(
        f"Missing historical columns: {missing_old}\n"
        f"Missing corrected columns: {missing_new}"
    )

# ------------------------------------------------------------
# 2. Match historical records from both datasets
# ------------------------------------------------------------

historical_names = set(old["event_name"])

historical_new = new[
    new["event_name"].isin(historical_names)
].copy()

old_cols = ["event_name"] + factors

old_data = old[old_cols].rename(
    columns={c: c + "_old" for c in factors}
)

new_cols = (
    ["event_name"] + factors
    + ["corrected_main_event_score"]
)

new_data = historical_new[new_cols].rename(
    columns={c: c + "_new" for c in factors}
)

audit = old_data.merge(
    new_data,
    on="event_name",
    how="outer",
    validate="one_to_one",
    indicator=True
)

print("\nEVENT JOIN STATUS")
display(audit["_merge"].value_counts())

audit = audit[audit["_merge"] == "both"].copy()

audit = audit.merge(
    groups[["event_name", "mismatch_group"]],
    on="event_name",
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------------------
# 3. Compare old vs corrected factors
# ------------------------------------------------------------

changes = []

for factor in factors:
    old_col = factor + "_old"
    new_col = factor + "_new"

    audit[old_col] = pd.to_numeric(
        audit[old_col], errors="coerce"
    )
    audit[new_col] = pd.to_numeric(
        audit[new_col], errors="coerce"
    )

    both = audit[old_col].notna() & audit[new_col].notna()

    delta = audit[new_col] - audit[old_col]

    audit[factor + "_delta"] = delta

    changes.append({
        "factor": factor,
        "both_available": int(both.sum()),
        "changed_values": int(
            (both & delta.abs().gt(TOL)).sum()
        ),
        "old_only": int(
            (
                audit[old_col].notna()
                & audit[new_col].isna()
            ).sum()
        ),
        "new_only": int(
            (
                audit[new_col].notna()
                & audit[old_col].isna()
            ).sum()
        ),
        "largest_absolute_change": (
            float(delta[both].abs().max())
            if both.any() else np.nan
        )
    })

changes_df = pd.DataFrame(changes)

print("\nOLD VS CORRECTED FACTOR DIFFERENCES")
display(changes_df.round(6))

# ------------------------------------------------------------
# 4. Calculate scores using both sets of factors
#    Only complete seven-factor records are compared.
# ------------------------------------------------------------

old_matrix = pd.DataFrame({
    c: audit[c + "_old"]
    for c in factors
}, index=audit.index)

new_matrix = pd.DataFrame({
    c: audit[c + "_new"]
    for c in factors
}, index=audit.index)

w = pd.Series(weights)

audit["old_complete"] = old_matrix.notna().all(axis=1)
audit["new_complete"] = new_matrix.notna().all(axis=1)

audit["score_using_old_factors"] = (
    old_matrix.mul(w, axis=1)
    .sum(axis=1, min_count=len(factors))
)

audit["score_using_new_factors"] = (
    new_matrix.mul(w, axis=1)
    .sum(axis=1, min_count=len(factors))
)

target = pd.to_numeric(
    audit["corrected_main_event_score"],
    errors="coerce"
)

audit["old_error"] = (
    audit["score_using_old_factors"] - target
).abs()

audit["new_error"] = (
    audit["score_using_new_factors"] - target
).abs()

audit["old_match"] = audit["old_error"].le(TOL)
audit["new_match"] = audit["new_error"].le(TOL)

def summarize(label, error, complete):
    eligible = complete & target.notna()
    return {
        "method": label,
        "eligible": int(eligible.sum()),
        "matches": int(
            (eligible & error.le(TOL)).sum()
        ),
        "mismatches": int(
            (eligible & error.gt(TOL)).sum()
        ),
        "mean_absolute_error": (
            float(error[eligible].mean())
            if eligible.any() else np.nan
        ),
        "max_absolute_error": (
            float(error[eligible].max())
            if eligible.any() else np.nan
        )
    }

results = pd.DataFrame([
    summarize(
        "Original normalized factors",
        audit["old_error"],
        audit["old_complete"]
    ),
    summarize(
        "Corrected main_event_scoring factors",
        audit["new_error"],
        audit["new_complete"]
    )
])

print("\nHISTORICAL FORMULA RECONSTRUCTION")
display(results.round(8))

# ------------------------------------------------------------
# 5. Did the corrected factors fix the original 51?
# ------------------------------------------------------------

known = audit[
    audit["mismatch_group"].notna()
].copy()

known["resolved_by_new_factors"] = (
    ~known["old_match"] & known["new_match"]
)

group_summary = known.groupby(
    "mismatch_group", dropna=False
).agg(
    events=("event_name", "size"),
    old_matches=("old_match", "sum"),
    new_matches=("new_match", "sum"),
    resolved=("resolved_by_new_factors", "sum"),
    largest_remaining_error=("new_error", "max")
).reset_index()

print("\nRESULTS BY ORIGINAL MISMATCH GROUP")
display(group_summary.round(8))

# ------------------------------------------------------------
# 6. Identify exactly which factors changed
# ------------------------------------------------------------

delta_cols = [
    factor + "_delta"
    for factor in factors
]

known["changed_factor_count"] = (
    known[delta_cols].abs().gt(TOL).sum(axis=1)
)

print("\nCHANGED FACTOR COUNTS")
display(
    known.groupby("mismatch_group")[
        "changed_factor_count"
    ].value_counts().rename("events")
)

print("\nEVENTS WITH CHANGED FACTORS")
changed_events = known[
    known["changed_factor_count"] > 0
].copy()

display(
    changed_events[
        [
            "event_name",
            "mismatch_group",
            "old_error",
            "new_error",
            "changed_factor_count"
        ] + delta_cols
    ].sort_values(
        "old_error", ascending=False
    ).head(40).round(6)
)

# ------------------------------------------------------------
# 7. Remaining complete-data mismatches
# ------------------------------------------------------------

unresolved = known[
    known["new_complete"]
    & target.loc[known.index].notna()
    & ~known["new_match"]
].copy()

print("\nREMAINING COMPLETE-DATA MISMATCHES")
print("Count:", len(unresolved))

display(
    unresolved[
        [
            "event_name",
            "mismatch_group",
            "corrected_main_event_score",
            "score_using_new_factors",
            "new_error"
        ] + [
            c + "_new" for c in factors
        ]
    ].sort_values(
        "new_error", ascending=False
    ).head(35).round(6)
)

# ------------------------------------------------------------
# 8. Missing-data cases, kept separate
# ------------------------------------------------------------

incomplete = audit[
    ~audit["new_complete"]
].copy()

print("\nHISTORICAL EVENTS MISSING NEW FACTORS")
print("Count:", len(incomplete))

if not incomplete.empty:
    missing_counts = new_matrix.loc[
        incomplete.index
    ].isna().sum().sort_values(ascending=False)

    display(
        missing_counts.to_frame(
            "events_missing_factor"
        )
    )

# ------------------------------------------------------------
# 9. Save results
# ------------------------------------------------------------

audit.to_csv(
    OUT / "full_corrected_factor_audit.csv",
    index=False
)

changes_df.to_csv(
    OUT / "factor_change_summary.csv",
    index=False
)

results.to_csv(
    OUT / "formula_reconstruction_results.csv",
    index=False
)

group_summary.to_csv(
    OUT / "original_mismatch_group_results.csv",
    index=False
)

unresolved.to_csv(
    OUT / "remaining_complete_mismatches.csv",
    index=False
)

main_corrected_factor_audit_v7 = audit.copy()
main_corrected_formula_results_v7 = results.copy()
main_corrected_group_results_v7 = group_summary.copy()
main_corrected_unresolved_v7 = unresolved.copy()

print("\n" + "=" * 85)
print("BIG CELL 24 COMPLETE")
print("=" * 85)
print("No existing scores or datasets were modified.")
print("Audit saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 25 — RECENT MAIN EVENT SCORING READINESS
# No original datasets or scores modified
# ============================================================

OUT = Path("/content/ufc_recent_main_scoring")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "main_event_scoring",
    "main_corrected_formula_results_v7"
]
missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError("Missing variables: " + ", ".join(missing))

main = main_event_scoring.copy()

weights = {
    "main_event_ranking_points_score": 0.25,
    "main_event_combined_title_fights_score": 0.20,
    "main_event_avg_win_pct_score": 0.15,
    "main_event_combined_win_streak_score": 0.15,
    "main_event_combined_previous_main_events_score": 0.10,
    "main_event_combined_ufc_fights_score": 0.10,
    "main_event_combined_bonuses_score": 0.05
}

factors = list(weights)
TOL = 1e-6

print("=" * 85)
print("BIG CELL 25 — RECENT MAIN EVENT SCORING READINESS")
print("=" * 85)

# 1. Confirm historical formula was fully reproduced
check = main_corrected_formula_results_v7.copy()
verified = check[
    check["method"] == "Corrected main_event_scoring factors"
]

if len(verified) != 1:
    raise RuntimeError("Historical verification row missing.")

verified = verified.iloc[0]

if (
    int(verified["eligible"]) != 777
    or int(verified["matches"]) != 777
    or int(verified["mismatches"]) != 0
):
    raise RuntimeError(
        "Historical scoring verification did not pass."
    )

print("\nHISTORICAL FORMULA: VERIFIED INTERNALLY (777/777)")

# 2. Validate main-event table
if main["event_name"].duplicated().any():
    raise RuntimeError("Duplicate event names detected.")

if main["event_name"].isna().any():
    raise RuntimeError("Missing event names detected.")

if "corrected_main_event_score" not in main.columns:
    raise RuntimeError("Corrected score column missing.")

main["corrected_main_event_score"] = pd.to_numeric(
    main["corrected_main_event_score"],
    errors="coerce"
)

recent = main[
    main["corrected_main_event_score"].isna()
].copy()

historical = main[
    main["corrected_main_event_score"].notna()
].copy()

print("\nHISTORICAL EVENTS:", len(historical))
print("RECENT UNSCORED EVENTS:", len(recent))

if len(recent) != 14:
    print(
        "WARNING: Expected 14 recent events; found",
        len(recent)
    )

# 3. Show recent event names
event_cols = [
    c for c in ["event_name", "event_date"]
    if c in recent.columns
]

print("\nRECENT EVENTS")
display(recent[event_cols].reset_index(drop=True))

# 4. Inspect the actual seven factor columns
readiness = []

for factor, weight in weights.items():
    if factor not in recent.columns:
        recent[factor] = np.nan
        column_exists = False
    else:
        column_exists = True

    recent[factor] = pd.to_numeric(
        recent[factor], errors="coerce"
    )

    available = recent[factor].notna()

    readiness.append({
        "factor": factor,
        "weight": weight,
        "column_exists": column_exists,
        "available_events": int(available.sum()),
        "missing_events": int((~available).sum()),
        "min_available": (
            float(recent.loc[available, factor].min())
            if available.any() else np.nan
        ),
        "max_available": (
            float(recent.loc[available, factor].max())
            if available.any() else np.nan
        )
    })

readiness = pd.DataFrame(readiness)

print("\nSEVEN-FACTOR READINESS")
display(readiness.round(5))

# 5. Per-event factor availability
recent["factors_available"] = recent[factors].notna().sum(axis=1)
recent["factors_missing"] = len(factors) - recent["factors_available"]

recent["missing_factor_names"] = recent[factors].apply(
    lambda row: ", ".join(row.index[row.isna()]),
    axis=1
)

print("\nEVENT-BY-EVENT READINESS")
display(
    recent[
        event_cols
        + [
            "factors_available",
            "factors_missing",
            "missing_factor_names"
        ]
    ].reset_index(drop=True)
)

# 6. Calculate only genuinely complete scores
# No missing value is replaced with zero.
factor_matrix = recent[factors].copy()
w = pd.Series(weights)

recent["main_event_candidate_score"] = (
    factor_matrix.mul(w, axis=1)
    .sum(axis=1, min_count=len(factors))
)

recent["score_ready"] = (
    recent["factors_missing"].eq(0)
    & recent["main_event_candidate_score"].notna()
)

# Flag suspicious normalized values
recent["factor_out_of_range"] = (
    factor_matrix.lt(-TOL)
    | factor_matrix.gt(100 + TOL)
).any(axis=1)

recent.loc[
    recent["factor_out_of_range"],
    "score_ready"
] = False

recent.loc[
    ~recent["score_ready"],
    "main_event_candidate_score"
] = np.nan

print("\nRECENT MAIN EVENT SCORE CANDIDATES")
display(
    recent[
        event_cols
        + [
            "factors_available",
            "score_ready",
            "factor_out_of_range",
            "main_event_candidate_score"
        ]
    ].reset_index(drop=True).round(5)
)

# 7. Identify raw columns that can help finish missing factors
keywords = [
    "rank", "title", "win_pct", "streak",
    "previous_main", "ufc_fights", "bonus"
]

raw_candidates = [
    c for c in main.columns
    if any(k in c.lower() for k in keywords)
    and c not in factors
]

raw_availability = []

for c in raw_candidates:
    values = recent[c]
    numeric = pd.to_numeric(values, errors="coerce")

    raw_availability.append({
        "column": c,
        "nonmissing_recent": int(values.notna().sum()),
        "numeric_recent": int(numeric.notna().sum()),
        "sample_value": (
            str(values.dropna().iloc[0])[:65]
            if values.notna().any() else ""
        )
    })

raw_availability = pd.DataFrame(raw_availability)

print("\nAVAILABLE RAW INPUTS FOR RECONSTRUCTION")
display(
    raw_availability.sort_values(
        ["numeric_recent", "nonmissing_recent"],
        ascending=False
    ).reset_index(drop=True)
)

# 8. Historical factor ranges for comparison
historical_ranges = []

for factor in factors:
    vals = pd.to_numeric(
        historical[factor], errors="coerce"
    )

    historical_ranges.append({
        "factor": factor,
        "historical_min": vals.min(),
        "historical_max": vals.max(),
        "historical_unique_values": vals.nunique()
    })

historical_ranges = pd.DataFrame(historical_ranges)

print("\nHISTORICAL NORMALIZED FACTOR RANGES")
display(historical_ranges.round(5))

# 9. Save audit only
recent.to_csv(
    OUT / "recent_main_event_readiness.csv",
    index=False
)

readiness.to_csv(
    OUT / "seven_factor_readiness.csv",
    index=False
)

raw_availability.to_csv(
    OUT / "available_raw_inputs.csv",
    index=False
)

historical_ranges.to_csv(
    OUT / "historical_factor_ranges.csv",
    index=False
)

main_recent_readiness_v8 = recent.copy()
main_recent_factor_status_v8 = readiness.copy()
main_recent_raw_inputs_v8 = raw_availability.copy()
main_recent_historical_ranges_v8 = historical_ranges.copy()

print("\n" + "=" * 85)
print("BIG CELL 25 COMPLETE")
print("=" * 85)
print("Recent events:", len(recent))
print("Fully score-ready:", int(recent["score_ready"].sum()))
print("Still missing factors:", int((~recent["score_ready"]).sum()))
print("Original datasets unchanged.")
print("Saved to:", OUT)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# UFC EVENT STRENGTH INDEX V2 — BIG CELL 26 (FIXED)

OUT = Path("/content/ufc_recent_main_event_completed")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "backup",
    "main_event_scoring",
    "ranking_event_totals_final_candidate_v3",
    "main_recent_readiness_v8"
]

missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError(f"Missing variables: {missing}")

raw = backup["event_master"].copy()
norm = backup["normalized"].copy()
recent = main_recent_readiness_v8.copy()
rankings = ranking_event_totals_final_candidate_v3.copy()

weights = {
    "main_event_ranking_points_score": 0.25,
    "main_event_combined_title_fights_score": 0.20,
    "main_event_avg_win_pct_score": 0.15,
    "main_event_combined_win_streak_score": 0.15,
    "main_event_combined_previous_main_events_score": 0.10,
    "main_event_combined_ufc_fights_score": 0.10,
    "main_event_combined_bonuses_score": 0.05
}

factors = list(weights)
rank_col = "total_ranking_points_v2"
raw_col = "main_event_ranking_points"
score_col = "main_event_ranking_points_score"

print("=" * 75)
print("BIG CELL 26 — RECENT MAIN EVENT RANKING NORMALIZATION")
print("=" * 75)

# 1. Verify columns and unique event names

for label, df, cols in [
    ("Historical raw", raw, ["event_name", raw_col]),
    ("Historical normalized", norm, ["event_name", score_col]),
    ("Recent events", recent, ["event_name"] + factors),
    ("Recent rankings", rankings, ["event_name", rank_col])
]:
    absent = [c for c in cols if c not in df.columns]
    if absent:
        raise RuntimeError(f"{label} missing columns: {absent}")
    if df["event_name"].isna().any():
        raise RuntimeError(f"{label} has missing event names")
    if df["event_name"].duplicated().any():
        raise RuntimeError(f"{label} has duplicate event names")

# 2. Historical ranking normalization

historical = raw[["event_name", raw_col]].merge(
    norm[["event_name", score_col]],
    on="event_name",
    how="inner",
    validate="one_to_one"
)

historical[raw_col] = pd.to_numeric(
    historical[raw_col], errors="coerce"
)
historical[score_col] = pd.to_numeric(
    historical[score_col], errors="coerce"
)

historical = historical.dropna()

curve = (
    historical.groupby(raw_col)[score_col]
    .agg(["min", "max", "mean"])
    .reset_index()
    .sort_values(raw_col)
)

if ((curve["max"] - curve["min"]) > 1e-6).any():
    raise RuntimeError(
        "Historical ranking normalization has conflicting values"
    )

curve = curve[[raw_col, "mean"]].rename(
    columns={"mean": score_col}
)

print("\nHistorical ranking normalization verified")
print("Historical pairs:", len(historical))
print("Unique ranking totals:", len(curve))

# 3. Join recent ranking totals

rank_data = rankings[
    ["event_name", rank_col]
].rename(
    columns={rank_col: "recent_ranking_points"}
)

rank_data["recent_ranking_points"] = pd.to_numeric(
    rank_data["recent_ranking_points"],
    errors="coerce"
)

recent = recent.drop(
    columns=["recent_ranking_points"],
    errors="ignore"
)

recent = recent.merge(
    rank_data,
    on="event_name",
    how="left",
    validate="one_to_one"
)

# 4. Exact mapping, then interpolation only if necessary

lookup = curve.set_index(raw_col)[score_col]

recent[score_col] = (
    recent["recent_ranking_points"].map(lookup)
)

recent["ranking_normalization_method"] = np.where(
    recent[score_col].notna(),
    "Exact historical mapping",
    "Unavailable"
)

needs_interpolation = (
    recent[score_col].isna()
    & recent["recent_ranking_points"].between(
        curve[raw_col].min(),
        curve[raw_col].max()
    )
)

if needs_interpolation.any():
    recent.loc[
        needs_interpolation, score_col
    ] = np.interp(
        recent.loc[
            needs_interpolation,
            "recent_ranking_points"
        ].to_numpy(float),
        curve[raw_col].to_numpy(float),
        curve[score_col].to_numpy(float)
    )

    recent.loc[
        needs_interpolation,
        "ranking_normalization_method"
    ] = "Interpolated historical mapping"

print("\nRECENT RANKING POINTS")
display(
    recent[
        [
            "event_name",
            "recent_ranking_points",
            score_col,
            "ranking_normalization_method"
        ]
    ].round(5)
)

# 5. Calculate complete seven-factor main event scores

for factor in factors:
    recent[factor] = pd.to_numeric(
        recent[factor], errors="coerce"
    )

matrix = recent[factors]

recent["factors_available_final"] = (
    matrix.notna().sum(axis=1)
)

recent["score_ready_final"] = (
    recent["factors_available_final"].eq(7)
    & matrix.ge(0).all(axis=1)
    & matrix.le(100).all(axis=1)
)

recent["main_event_score_candidate_v2"] = (
    matrix.mul(pd.Series(weights), axis=1)
    .sum(axis=1, min_count=7)
)

recent.loc[
    ~recent["score_ready_final"],
    "main_event_score_candidate_v2"
] = np.nan

# 6. Display leaderboard

leaderboard = recent[
    [
        "event_name",
        "event_date",
        "recent_ranking_points",
        "main_event_score_candidate_v2",
        "score_ready_final",
        "ranking_normalization_method"
    ]
].sort_values(
    "main_event_score_candidate_v2",
    ascending=False,
    na_position="last"
).reset_index(drop=True)

print("\nPROVISIONAL MAIN EVENT STRENGTH LEADERBOARD")
display(leaderboard.round(4))

# 7. Preserve results

recent.to_csv(
    OUT / "recent_main_event_scored_candidates.csv",
    index=False
)

leaderboard.to_csv(
    OUT / "recent_main_event_leaderboard.csv",
    index=False
)

curve.to_csv(
    OUT / "historical_ranking_normalization_curve.csv",
    index=False
)

main_recent_scored_candidates_v9 = recent.copy()
main_recent_leaderboard_v9 = leaderboard.copy()
main_ranking_curve_v9 = curve.copy()

print("\n" + "=" * 75)
print("BIG CELL 26 COMPLETE")
print("=" * 75)
print("Recent main events scored:", int(recent["score_ready_final"].sum()), "of", len(recent))
print("Original datasets unchanged.")
print("Files saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 27 — CORRECT MAIN EVENT RANKING INPUTS
# Uses fighter-level rankings, NOT full-card totals
# Does not overwrite historical or original datasets
# ============================================================

OUT = Path("/content/ufc_main_event_rankings_corrected")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "main_event_scoring",
    "ranking_points_final_candidate_v3",
    "backup",
    "main_recent_readiness_v8"
]

missing = [x for x in required if x not in globals()]
if missing:
    raise RuntimeError("Missing variables: " + ", ".join(missing))

main = main_event_scoring.copy()
fighter_rankings = ranking_points_final_candidate_v3.copy()
recent = main_recent_readiness_v8.copy()

historical_raw = backup["event_master"].copy()
historical_norm = backup["normalized"].copy()

weights = {
    "main_event_ranking_points_score": 0.25,
    "main_event_combined_title_fights_score": 0.20,
    "main_event_avg_win_pct_score": 0.15,
    "main_event_combined_win_streak_score": 0.15,
    "main_event_combined_previous_main_events_score": 0.10,
    "main_event_combined_ufc_fights_score": 0.10,
    "main_event_combined_bonuses_score": 0.05
}

factors = list(weights)
TOL = 1e-6

print("=" * 85)
print("BIG CELL 27 — CORRECT MAIN EVENT RANKING INPUTS")
print("=" * 85)

# ------------------------------------------------------------
# 1. Inspect fighter ranking source columns
# ------------------------------------------------------------

print("\nFIGHTER RANKING SOURCE COLUMNS")
print(fighter_rankings.columns.tolist())

required_rank_cols = [
    "event_name",
    "fighter",
    "ranking_points_v2"
]

missing_rank_cols = [
    c for c in required_rank_cols
    if c not in fighter_rankings.columns
]

if missing_rank_cols:
    raise RuntimeError(
        f"Missing fighter ranking columns: {missing_rank_cols}"
    )

# ------------------------------------------------------------
# 2. Find main-event fighters
# ------------------------------------------------------------

main_fighter_cols = ["event_name", "f1_name", "f2_name"]

missing_main_cols = [
    c for c in main_fighter_cols
    if c not in main.columns
]

if missing_main_cols:
    raise RuntimeError(
        f"Missing main-event fighter columns: {missing_main_cols}"
    )

main_names = main[main_fighter_cols].copy()

recent = recent.merge(
    main_names,
    on="event_name",
    how="left",
    validate="one_to_one",
    suffixes=("", "_source")
)

for c in ["f1_name", "f2_name"]:
    source_col = c + "_source"
    if source_col in recent.columns:
        recent[c] = recent[c].combine_first(
            recent[source_col]
        )
        recent = recent.drop(columns=[source_col])

print("\nMAIN EVENT FIGHTERS")
display(
    recent[
        ["event_name", "f1_name", "f2_name"]
    ]
)

# ------------------------------------------------------------
# 3. Normalize fighter names for matching
# ------------------------------------------------------------

def clean_name(series):
    return (
        series.astype("string")
        .str.strip()
        .str.lower()
        .str.replace(r"\s+", " ", regex=True)
    )

fighter_rankings["fighter_key"] = clean_name(
    fighter_rankings["fighter"]
)

recent["f1_key"] = clean_name(recent["f1_name"])
recent["f2_key"] = clean_name(recent["f2_name"])

fighter_rankings["ranking_points_v2"] = pd.to_numeric(
    fighter_rankings["ranking_points_v2"],
    errors="coerce"
)

if fighter_rankings[
    ["event_name", "fighter_key"]
].duplicated().any():
    raise RuntimeError(
        "Duplicate fighter rankings for the same event."
    )

lookup = fighter_rankings.set_index(
    ["event_name", "fighter_key"]
)["ranking_points_v2"]

# ------------------------------------------------------------
# 4. Match each main-event fighter individually
# ------------------------------------------------------------

recent["f1_ranking_points"] = [
    lookup.get((event, fighter), np.nan)
    for event, fighter in zip(
        recent["event_name"], recent["f1_key"]
    )
]

recent["f2_ranking_points"] = [
    lookup.get((event, fighter), np.nan)
    for event, fighter in zip(
        recent["event_name"], recent["f2_key"]
    )
]

recent["main_event_ranking_points_corrected"] = (
    recent["f1_ranking_points"]
    + recent["f2_ranking_points"]
)

recent["ranking_pair_complete"] = (
    recent["f1_ranking_points"].notna()
    & recent["f2_ranking_points"].notna()
)

print("\nMAIN EVENT RANKING POINTS — TWO FIGHTERS ONLY")
display(
    recent[
        [
            "event_name",
            "f1_name",
            "f1_ranking_points",
            "f2_name",
            "f2_ranking_points",
            "main_event_ranking_points_corrected",
            "ranking_pair_complete"
        ]
    ]
)

# ------------------------------------------------------------
# 5. Reconstruct exact historical normalization
# ------------------------------------------------------------

raw_col = "main_event_ranking_points"
score_col = "main_event_ranking_points_score"

historical = historical_raw[
    ["event_name", raw_col]
].merge(
    historical_norm[["event_name", score_col]],
    on="event_name",
    how="inner",
    validate="one_to_one"
)

historical[raw_col] = pd.to_numeric(
    historical[raw_col], errors="coerce"
)

historical[score_col] = pd.to_numeric(
    historical[score_col], errors="coerce"
)

historical = historical.dropna()

curve = (
    historical.groupby(raw_col)[score_col]
    .agg(["min", "max", "mean"])
    .reset_index()
)

if ((curve["max"] - curve["min"]) > TOL).any():
    raise RuntimeError(
        "Historical normalization has conflicting values."
    )

curve = curve[
    [raw_col, "mean"]
].rename(columns={"mean": score_col})

lookup_score = curve.set_index(raw_col)[score_col]

recent[score_col] = (
    recent["main_event_ranking_points_corrected"]
    .map(lookup_score)
)

recent["ranking_normalization_method"] = np.where(
    recent[score_col].notna(),
    "Exact historical mapping",
    "Unmatched historical value"
)

print("\nCORRECTED MAIN EVENT NORMALIZATION")
display(
    recent[
        [
            "event_name",
            "main_event_ranking_points_corrected",
            score_col,
            "ranking_normalization_method"
        ]
    ].round(6)
)

# ------------------------------------------------------------
# 6. Calculate corrected Main Event Strength
# ------------------------------------------------------------

for c in factors:
    recent[c] = pd.to_numeric(
        recent[c], errors="coerce"
    )

matrix = recent[factors]

recent["factors_available_final"] = (
    matrix.notna().sum(axis=1)
)

recent["score_ready_final"] = (
    recent["ranking_pair_complete"]
    & recent["factors_available_final"].eq(7)
    & matrix.ge(-TOL).all(axis=1)
    & matrix.le(100 + TOL).all(axis=1)
)

recent["main_event_score_candidate_v3"] = (
    matrix.mul(pd.Series(weights), axis=1)
    .sum(axis=1, min_count=7)
)

recent.loc[
    ~recent["score_ready_final"],
    "main_event_score_candidate_v3"
] = np.nan

# ------------------------------------------------------------
# 7. Display corrected leaderboard
# ------------------------------------------------------------

leaderboard = recent[
    [
        "event_name",
        "event_date",
        "f1_name",
        "f2_name",
        "main_event_ranking_points_corrected",
        "main_event_score_candidate_v3",
        "score_ready_final"
    ]
].sort_values(
    "main_event_score_candidate_v3",
    ascending=False,
    na_position="last"
).reset_index(drop=True)

print("\nCORRECTED MAIN EVENT STRENGTH LEADERBOARD")
display(leaderboard.round(5))

# ------------------------------------------------------------
# 8. Save corrected candidates
# ------------------------------------------------------------

recent.to_csv(
    OUT / "corrected_recent_main_event_scores.csv",
    index=False
)

leaderboard.to_csv(
    OUT / "corrected_recent_main_event_leaderboard.csv",
    index=False
)

main_recent_scored_candidates_v10 = recent.copy()
main_recent_leaderboard_v10 = leaderboard.copy()
main_recent_ranking_curve_v10 = curve.copy()

print("\n" + "=" * 85)
print("BIG CELL 27 COMPLETE")
print("=" * 85)
print(
    "Fully scored recent main events:",
    int(recent["score_ready_final"].sum()),
    "of", len(recent)
)
print("Original datasets unchanged.")
print("Saved to:", OUT)
```

```python

import pandas as pd
import numpy as np
import pickle
from pathlib import Path
from IPython.display import display

# ============================================================
# UFC EVENT STRENGTH INDEX V2
# BIG CELL 28 — COMPLETE OVERALL LEADERBOARD (CORRECTED)
#
# Card Quality:        50%
# Star Experience:     25%
# Main Event Strength: 25%
#
# Uses exact verified column names from saved data.
# Does not overwrite original historical datasets.
# ============================================================

print("=" * 85)
print("BIG CELL 28 — OVERALL UFC EVENT STRENGTH INDEX V2")
print("=" * 85)

# ------------------------------------------------------------
# STEP 1 — Load saved scoring tables
# ------------------------------------------------------------

required = [
    "category_recent_candidates_v2",
    "star_recent_candidates_v1",
    "main_recent_scored_candidates_v10"
]

missing = [name for name in required if name not in globals()]

if missing:
    raise RuntimeError(
        "Missing variables: " + ", ".join(missing)
    )

card = category_recent_candidates_v2.copy()
star = star_recent_candidates_v1.copy()
main = main_recent_scored_candidates_v10.copy()

card_col = "card_quality_candidate"
star_col = "star_experience_candidate_v3"
main_col = "main_event_score_candidate_v3"

for label, df, score_col in [
    ("Card Quality", card, card_col),
    ("Star Experience", star, star_col),
    ("Main Event Strength", main, main_col)
]:
    if "event_name" not in df.columns:
        raise RuntimeError(
            f"{label}: event_name column missing"
        )

    if score_col not in df.columns:
        raise RuntimeError(
            f"{label}: {score_col} column missing"
        )

    if df["event_name"].isna().any():
        raise RuntimeError(
            f"{label}: missing event names"
        )

    if df["event_name"].duplicated().any():
        raise RuntimeError(
            f"{label}: duplicate event names"
        )

    print(
        f"{label}: {len(df)} events | "
        f"Score column: {score_col}"
    )

# ------------------------------------------------------------
# STEP 2 — Check Star Experience verification
# ------------------------------------------------------------

if "star_experience_formula_verified" in star.columns:
    verified = star[
        "star_experience_formula_verified"
    ].fillna(False).astype(bool)

    print(
        "\nStar Experience formulas verified:",
        int(verified.sum()),
        "of",
        len(verified)
    )

    if not verified.all():
        raise RuntimeError(
            "Some Star Experience formulas failed verification."
        )

# ------------------------------------------------------------
# STEP 3 — Prepare clean component tables
# ------------------------------------------------------------

card_data = card[
    ["event_name", card_col]
].copy()

card_data = card_data.rename(
    columns={card_col: "card_quality_score"}
)

star_data = star[
    ["event_name", star_col]
].copy()

star_data = star_data.rename(
    columns={star_col: "star_experience_score"}
)

main_columns = ["event_name", main_col]

if "event_date" in main.columns:
    main_columns.append("event_date")

if "score_ready_final" in main.columns:
    main_columns.append("score_ready_final")

main_data = main[main_columns].copy()

main_data = main_data.rename(
    columns={
        main_col: "main_event_strength_score"
    }
)

# ------------------------------------------------------------
# STEP 4 — Combine all three categories
# ------------------------------------------------------------

overall = main_data.merge(
    card_data,
    on="event_name",
    how="outer",
    validate="one_to_one",
    indicator="_card_join"
)

overall = overall.merge(
    star_data,
    on="event_name",
    how="outer",
    validate="one_to_one",
    indicator="_star_join"
)

print("\n" + "-" * 85)
print("EVENT MATCHING AUDIT")
print("-" * 85)

print("\nMain Event / Card Quality:")
print(overall["_card_join"].value_counts().to_string())

print("\nStar Experience:")
print(overall["_star_join"].value_counts().to_string())

if len(overall) != 14:
    raise RuntimeError(
        f"Expected 14 events, found {len(overall)}"
    )

if (
    (overall["_card_join"] != "both").any()
    or (overall["_star_join"] != "both").any()
):
    display(
        overall[
            ["event_name", "_card_join", "_star_join"]
        ]
    )

    raise RuntimeError(
        "Some events did not match across categories."
    )

print("\nAll 14 events matched successfully.")

# ------------------------------------------------------------
# STEP 5 — Validate component scores
# ------------------------------------------------------------

components = [
    "card_quality_score",
    "star_experience_score",
    "main_event_strength_score"
]

for column in components:
    overall[column] = pd.to_numeric(
        overall[column],
        errors="coerce"
    )

overall["component_count"] = (
    overall[components].notna().sum(axis=1)
)

overall["all_components_valid"] = (
    overall["component_count"].eq(3)
    & overall[components].ge(0).all(axis=1)
    & overall[components].le(100).all(axis=1)
)

if "score_ready_final" in overall.columns:
    overall["all_components_valid"] = (
        overall["all_components_valid"]
        & overall["score_ready_final"].eq(True)
    )

print("\n" + "-" * 85)
print("COMPONENT VALIDATION")
print("-" * 85)

print(
    "Complete, valid events:",
    int(overall["all_components_valid"].sum()),
    "of",
    len(overall)
)

if not overall["all_components_valid"].all():
    display(
        overall[
            ["event_name"] + components
            + ["component_count", "all_components_valid"]
        ]
    )

    raise RuntimeError(
        "Some component scores are missing or invalid."
    )

# ------------------------------------------------------------
# STEP 6 — Calculate overall Event Strength Index
# ------------------------------------------------------------

overall["overall_event_strength_v2"] = (
    overall["card_quality_score"] * 0.50
    + overall["star_experience_score"] * 0.25
    + overall["main_event_strength_score"] * 0.25
)

# Independent arithmetic verification
manual_check = (
    overall[components].to_numpy(dtype=float)
    @ np.array([0.50, 0.25, 0.25])
)

if not np.allclose(
    overall["overall_event_strength_v2"].to_numpy(float),
    manual_check,
    atol=1e-9,
    rtol=0
):
    raise RuntimeError(
        "Overall weighted score calculation failed."
    )

print("\nWeighted score formula verified.")

# ------------------------------------------------------------
# STEP 7 — Rank all 14 recent events
# ------------------------------------------------------------

overall = overall.sort_values(
    ["overall_event_strength_v2", "event_name"],
    ascending=[False, True]
).reset_index(drop=True)

overall.insert(
    0,
    "recent_rank",
    np.arange(1, len(overall) + 1)
)

overall["source_status"] = (
    "PROVISIONAL — source verification pending"
)

display_columns = [
    "recent_rank",
    "event_name"
]

if "event_date" in overall.columns:
    display_columns.append("event_date")

display_columns += [
    "card_quality_score",
    "star_experience_score",
    "main_event_strength_score",
    "overall_event_strength_v2"
]

print("\n" + "=" * 85)
print("PROVISIONAL UFC EVENT STRENGTH INDEX V2 LEADERBOARD")
print("=" * 85)

display(
    overall[display_columns].round(3)
)

# ------------------------------------------------------------
# STEP 8 — Show category contributions
# ------------------------------------------------------------

overall["card_contribution"] = (
    overall["card_quality_score"] * 0.50
)

overall["star_contribution"] = (
    overall["star_experience_score"] * 0.25
)

overall["main_contribution"] = (
    overall["main_event_strength_score"] * 0.25
)

print("\n" + "-" * 85)
print("OVERALL SCORE CONTRIBUTIONS")
print("-" * 85)

display(
    overall[
        [
            "recent_rank",
            "event_name",
            "card_contribution",
            "star_contribution",
            "main_contribution",
            "overall_event_strength_v2"
        ]
    ].round(3)
)

# ------------------------------------------------------------
# STEP 9 — Save results locally
# ------------------------------------------------------------

OUT = Path("/content/ufc_overall_leaderboard_v2")
OUT.mkdir(parents=True, exist_ok=True)

overall.to_csv(
    OUT / "provisional_recent_overall_leaderboard.csv",
    index=False
)

overall[display_columns].to_csv(
    OUT / "provisional_recent_leaderboard_clean.csv",
    index=False
)

overall_recent_leaderboard_v11 = overall.copy()

print("\nLocal CSV files saved successfully.")

# ------------------------------------------------------------
# STEP 10 — Save latest progress to Google Drive
# ------------------------------------------------------------

drive_dir = Path("/content/drive/MyDrive")

if drive_dir.exists():
    checkpoint_path = (
        drive_dir / "UFC_Event_Strength_V2_Checkpoint.pkl"
    )

    checkpoint = {}

    if checkpoint_path.exists():
        with open(checkpoint_path, "rb") as file:
            checkpoint = pickle.load(file)

    checkpoint.update({
        "main_recent_scored_candidates_v10":
            main_recent_scored_candidates_v10,
        "main_recent_leaderboard_v10":
            main_recent_leaderboard_v10,
        "main_recent_ranking_curve_v10":
            main_recent_ranking_curve_v10,
        "overall_recent_leaderboard_v11":
            overall_recent_leaderboard_v11
    })

    temporary_path = checkpoint_path.with_suffix(".tmp")

    with open(temporary_path, "wb") as file:
        pickle.dump(
            checkpoint,
            file,
            protocol=pickle.HIGHEST_PROTOCOL
        )

    temporary_path.replace(checkpoint_path)

    overall.to_csv(
        drive_dir / "UFC_Event_Strength_V2_Recent_Leaderboard.csv",
        index=False
    )

    print("\nGoogle Drive checkpoint updated!")
    print("Total saved variables:", len(checkpoint))

else:
    print(
        "\nGoogle Drive not mounted. "
        "Results saved locally only."
    )

# ------------------------------------------------------------
# STEP 11 — Final summary
# ------------------------------------------------------------

print("\n" + "=" * 85)
print("BIG CELL 28 COMPLETE")
print("=" * 85)

print("Recent events analyzed:", len(overall))
print(
    "Events with complete valid scores:",
    int(overall["all_components_valid"].sum())
)
print("Overall score calculation: VERIFIED")
print("Historical datasets: UNCHANGED")
print("Recent source verification: STILL PENDING")

print("\nTOP 5 RECENT EVENTS")

display(
    overall[
        [
            "recent_rank",
            "event_name",
            "overall_event_strength_v2"
        ]
    ].head(5).round(3)
)

print("\nReady for the next stage of the project!")
```

```python

import pandas as pd
import numpy as np
import re
import unicodedata
from pathlib import Path
from IPython.display import display

print("=" * 80)
print("BIG CELL 31 — UFC 333 PRE-EVENT SCORING READINESS")
print("=" * 80)

if "ufc333_verified_card_v13" not in globals():
    raise RuntimeError("Please run Big Cell 30 first.")

if "scoring_fights" not in globals():
    raise RuntimeError("Missing scoring_fights.")

card333 = ufc333_verified_card_v13.copy()
history = scoring_fights.copy()
cutoff = pd.Timestamp("2026-10-24")

def clean_fighter_name(value):
    if pd.isna(value):
        return ""
    value = unicodedata.normalize("NFKD", str(value))
    value = "".join(
        c for c in value
        if not unicodedata.combining(c)
    )
    return re.sub(
        r"[^a-z0-9]+",
        "",
        value.lower().replace("’", "'")
    )

# Check available columns
print("\nSCORING DATASET")
print("Rows:", len(history))
print("Columns:", len(history.columns))

print("\nAVAILABLE DATE AND FIGHTER COLUMNS")
print([
    c for c in history.columns
    if any(x in c.lower() for x in [
        "date", "fighter", "f1_name", "f2_name"
    ])
])

# Identify useful pre-fight statistics
stat_keywords = [
    "rank", "streak", "bonus", "title",
    "main_event", "win_pct", "finish",
    "fights_before", "wins_before"
]

stat_columns = [
    c for c in history.columns
    if any(k in c.lower() for k in stat_keywords)
]

print("\nPOTENTIAL SCORING STATISTICS")
for c in stat_columns:
    print(" -", c)

if "event_date" not in history.columns:
    raise RuntimeError(
        "Cannot safely identify pre-event records: "
        "event_date column is missing."
    )

history["event_date_parsed"] = pd.to_datetime(
    history["event_date"],
    errors="coerce"
)

history = history[
    history["event_date_parsed"] < cutoff
].copy()

print("\nHistorical fight rows before UFC 333:", len(history))

# Create long-format fighter history
fighter_frames = []

for side in ["f1", "f2"]:
    name_col = f"{side}_name"

    if name_col not in history.columns:
        continue

    part = history.copy()
    part["fighter"] = part[name_col]
    part["fighter_key"] = part["fighter"].map(
        clean_fighter_name
    )
    part["fighter_side"] = side

    fighter_frames.append(part)

if not fighter_frames:
    raise RuntimeError("No fighter name columns found.")

fighter_history = pd.concat(
    fighter_frames,
    ignore_index=True
)

fighter_history = fighter_history.sort_values(
    "event_date_parsed",
    ascending=False
)

# Extract latest available prior fight row
target_fighters = pd.unique(
    pd.concat([
        card333["f1_name"],
        card333["f2_name"]
    ])
)

results = []

for fighter in target_fighters:
    key = clean_fighter_name(fighter)

    matches = fighter_history[
        fighter_history["fighter_key"] == key
    ]

    if len(matches):
        latest = matches.iloc[0]

        results.append({
            "fighter": fighter,
            "historical_records": len(matches),
            "latest_record_date":
                latest["event_date_parsed"],
            "latest_event": latest.get(
                "event_name", None
            ),
            "data_available": True
        })
    else:
        results.append({
            "fighter": fighter,
            "historical_records": 0,
            "latest_record_date": pd.NaT,
            "latest_event": None,
            "data_available": False
        })

coverage = pd.DataFrame(results)

print("\nFIGHTER COVERAGE BEFORE UFC 333")
display(coverage)

print("\nFIGHTERS WITHOUT PRIOR FIGHT RECORDS")
missing = coverage[
    ~coverage["data_available"]
]

if missing.empty:
    print("None")
else:
    display(missing)

# Audit which statistics are stored for the main event
main_event = card333[
    card333["is_main_event"]
].iloc[0]

main_names = [
    main_event["f1_name"],
    main_event["f2_name"]
]

print("\nMAIN EVENT — LATEST HISTORICAL RECORDS")

main_rows = fighter_history[
    fighter_history["fighter_key"].isin(
        [clean_fighter_name(x) for x in main_names]
    )
].drop_duplicates("fighter_key")

useful_columns = [
    "fighter",
    "event_name",
    "event_date",
    "fighter_side"
]

useful_columns += [
    c for c in stat_columns
    if c not in useful_columns
][:20]

useful_columns = [
    c for c in useful_columns
    if c in main_rows.columns
]

display(
    main_rows[useful_columns].T
)

# Save audit
OUT = Path("/content/ufc_portfolio_v2")
OUT.mkdir(parents=True, exist_ok=True)

coverage.to_csv(
    OUT / "ufc333_pre_event_fighter_coverage.csv",
    index=False
)

ufc333_pre_event_coverage_v14 = coverage.copy()

print("\n" + "=" * 80)
print("BIG CELL 31 COMPLETE")
print("=" * 80)

print(
    "Fighters with pre-event history:",
    int(coverage["data_available"].sum()),
    "/",
    len(coverage)
)

print(
    "Missing fighters:",
    len(missing)
)

print("UFC 333 score: pending verified inputs")
print("Existing 791-event leaderboard: unchanged")
```

```python

import pandas as pd
import numpy as np
import re
import unicodedata
from IPython.display import display

print("=" * 75)
print("BIG CELL 32 — UFC 333 FIGHTER STATISTICS")
print("=" * 75)

assert "scoring_fights" in globals(), "Missing scoring_fights"
assert "ufc333_verified_card_v13" in globals(), "Run Cell 30"

hist = scoring_fights.copy()
card = ufc333_verified_card_v13.copy()
cutoff = pd.Timestamp("2026-10-24")

def key(name):
    if pd.isna(name):
        return ""
    name = unicodedata.normalize("NFKD", str(name))
    name = "".join(
        c for c in name if not unicodedata.combining(c)
    )
    return re.sub(r"[^a-z0-9]", "", name.lower())

hist["event_date"] = pd.to_datetime(
    hist["event_date"], errors="coerce"
)

hist = hist[
    hist["event_date"].notna() &
    (hist["event_date"] < cutoff)
].copy()

# Collect the latest pre-fight statistics for every fighter.
# We deliberately do NOT assume these equal the final
# pre-UFC-333 statistics until the last fight is accounted for.

fighter_rows = []

for _, fight in card.iterrows():
    for side in ["f1", "f2"]:
        fighter = fight[f"{side}_name"]
        fighter_key = key(fighter)

        matches = []

        for historical_side in ["f1", "f2"]:
            name_col = f"{historical_side}_name"

            mask = hist[name_col].map(key).eq(fighter_key)
            part = hist.loc[mask].copy()

            part["_side"] = historical_side
            matches.append(part)

        prior = pd.concat(matches)

        if prior.empty:
            fighter_rows.append({
                "fighter": fighter,
                "fight_on_card": fight["f1_name"] +
                    " vs. " + fight["f2_name"],
                "historical_fights": 0,
                "last_fight_date": pd.NaT,
                "ufc_fights_before_last": np.nan,
                "ufc_wins_before_last": np.nan,
                "win_streak_before_last": np.nan,
                "previous_title_fights_before_last": np.nan,
                "previous_main_events_before_last": np.nan,
                "previous_bonuses_before_last": np.nan,
                "needs_last_result_update": True
            })
            continue

        prior = prior.sort_values(
            "event_date", ascending=False
        )

        latest = prior.iloc[0]
        s = latest["_side"]

        def get_stat(suffix):
            col = f"{s}_{suffix}"
            return latest.get(col, np.nan)

        fighter_rows.append({
            "fighter": fighter,
            "fight_on_card": fight["f1_name"] +
                " vs. " + fight["f2_name"],
            "historical_fights": len(prior),
            "last_fight_date": latest["event_date"],
            "last_event": latest.get("event_name"),
            "ufc_fights_before_last":
                get_stat("ufc_fights_before"),
            "ufc_wins_before_last":
                get_stat("ufc_wins_before"),
            "win_streak_before_last":
                get_stat("win_streak_before"),
            "previous_title_fights_before_last":
                get_stat("previous_title_fights"),
            "previous_main_events_before_last":
                get_stat("previous_main_events"),
            "previous_bonuses_before_last":
                get_stat("previous_total_bonuses"),
            "needs_last_result_update": True
        })

ufc333_fighter_stats_v15 = pd.DataFrame(fighter_rows)

print("\nFIGHTER STATISTICS — MOST RECENT HISTORICAL RECORD")
display(ufc333_fighter_stats_v15)

print("\nDATA COVERAGE")
print("Fighters:", len(ufc333_fighter_stats_v15))
print(
    "Fighters with historical data:",
    int(
        (ufc333_fighter_stats_v15["historical_fights"] > 0)
        .sum()
    )
)

print("\nHISTORICAL DATA COLUMNS AVAILABLE FOR RESULT UPDATES")
print([
    c for c in hist.columns
    if any(word in c.lower() for word in [
        "winner", "result", "outcome", "method",
        "finish", "bonus"
    ])
])

print("\nHISTORICAL DATA COLUMNS AVAILABLE FOR RANKING")
print([
    c for c in hist.columns
    if "rank" in c.lower()
])

print("\nBIG CELL 32 COMPLETE")
print("Next: finalize fighter inputs and calculate UFC 333.")
```

```python

import pandas as pd
import numpy as np
import re
import unicodedata
from IPython.display import display

print("=" * 75)
print("BIG CELL 33 — UFC 333 PRE-EVENT FIGHTER DATA")
print("=" * 75)

assert "scoring_fights" in globals()
assert "ufc333_verified_card_v13" in globals()

history = scoring_fights.copy()
card = ufc333_verified_card_v13.copy()

history["event_date"] = pd.to_datetime(
    history["event_date"], errors="coerce"
)

history = history[
    history["event_date"] < pd.Timestamp("2026-10-24")
].copy()

def normalize_name(value):
    if pd.isna(value):
        return ""
    value = unicodedata.normalize("NFKD", str(value))
    value = "".join(
        c for c in value if not unicodedata.combining(c)
    )
    return re.sub(r"[^a-z0-9]", "", value.lower())

def number(value):
    return pd.to_numeric(value, errors="coerce")

def is_win(value):
    return str(value).strip().lower() in [
        "win", "w"
    ]

def is_loss(value):
    return str(value).strip().lower() in [
        "loss", "l"
    ]

fighter_list = pd.unique(pd.concat([
    card["f1_name"],
    card["f2_name"]
]))

records = []

for fighter in fighter_list:

    pieces = []

    for side in ["f1", "f2"]:
        matches = history[
            history[f"{side}_name"].map(
                normalize_name
            ).eq(normalize_name(fighter))
        ].copy()

        matches["_side"] = side
        pieces.append(matches)

    fighter_history = pd.concat(
        pieces, ignore_index=True
    ).sort_values("event_date")

    if fighter_history.empty:
        records.append({
            "fighter": fighter,
            "ufc_fights": np.nan,
            "ufc_wins": np.nan,
            "ufc_win_pct": np.nan,
            "win_streak": np.nan,
            "previous_title_fights": np.nan,
            "previous_main_events": np.nan,
            "previous_bonuses": np.nan,
            "data_status": "No historical record"
        })
        continue

    last = fighter_history.iloc[-1]
    side = last["_side"]

    def latest_stat(suffix):
        return number(
            last.get(f"{side}_{suffix}", np.nan)
        )

    result = last.get(
        f"{side}_result", np.nan
    )

    wins = latest_stat("ufc_wins_before")
    fights = latest_stat("ufc_fights_before")
    streak = latest_stat("win_streak_before")

    # Advance the stored BEFORE-LAST-FIGHT statistics
    # using the result of the fighter's latest appearance.
    if pd.notna(fights):
        fights += 1

    if pd.notna(wins) and is_win(result):
        wins += 1

    if pd.notna(streak):
        if is_win(result):
            streak += 1
        elif is_loss(result):
            streak = 0
        else:
            streak = np.nan

    title_count = latest_stat(
        "previous_title_fights"
    )
    main_count = latest_stat(
        "previous_main_events"
    )
    bonus_count = latest_stat(
        "previous_total_bonuses"
    )

    if pd.notna(title_count):
        title_count += int(
            bool(last.get("is_title_fight", False))
        )

    if pd.notna(main_count):
        main_count += int(
            bool(last.get("is_main_event", False))
        )

    # Bonus data for the latest fight may not be available.
    # Do not invent bonus awards.

    win_pct = (
        wins / fights
        if pd.notna(wins)
        and pd.notna(fights)
        and fights > 0
        else np.nan
    )

    records.append({
        "fighter": fighter,
        "ufc_fights": fights,
        "ufc_wins": wins,
        "ufc_win_pct": win_pct,
        "win_streak": streak,
        "previous_title_fights": title_count,
        "previous_main_events": main_count,
        "previous_bonuses": bonus_count,
        "last_fight_date": last["event_date"],
        "last_fight_result": result,
        "data_status": (
            "Partial — latest bonus unverified"
            if pd.notna(fights)
            else "Partial — missing UFC totals"
        )
    })

ufc333_fighter_inputs_v16 = pd.DataFrame(records)

print("\nUPDATED FIGHTER STATISTICS")
display(ufc333_fighter_inputs_v16)

print("\nMAIN EVENT")
display(
    ufc333_fighter_inputs_v16[
        ufc333_fighter_inputs_v16["fighter"].isin([
            "Alexander Volkanovski",
            "Movsar Evloev"
        ])
    ]
)

print("\nMISSING DATA BY COLUMN")
print(
    ufc333_fighter_inputs_v16.isna().sum()
    .sort_values(ascending=False)
)

print("\n" + "=" * 75)
print("BIG CELL 33 COMPLETE")
print("=" * 75)
print("Fighters processed:", len(ufc333_fighter_inputs_v16))
print("Next: UFC 333 scoring and ranking.")
```

```python

import pandas as pd
import numpy as np
from IPython.display import display

print("=" * 75)
print("BIG CELL 34 — UFC 333 DATA RECOVERY")
print("=" * 75)

assert "ufc333_fighter_inputs_v16" in globals()
assert "scoring_fights" in globals()

fighters = ufc333_fighter_inputs_v16.copy()
history = scoring_fights.copy()

# Recover UFC fight totals from the historical fight records
# where the existing pre-fight totals are missing.
# This is a separate estimate and is NOT silently substituted
# for the original data.

def clean_name(value):
    import re, unicodedata
    if pd.isna(value):
        return ""
    value = unicodedata.normalize("NFKD", str(value))
    value = "".join(
        c for c in value
        if not unicodedata.combining(c)
    )
    return re.sub(r"[^a-z0-9]", "", value.lower())

history["event_date"] = pd.to_datetime(
    history["event_date"], errors="coerce"
)

history = history[
    history["event_date"] < pd.Timestamp("2026-10-24")
].copy()

recovery = []

for _, fighter_row in fighters.iterrows():
    name = fighter_row["fighter"]
    key = clean_name(name)

    all_fights = []

    for side in ["f1", "f2"]:
        matches = history[
            history[f"{side}_name"].map(clean_name).eq(key)
        ].copy()

        if matches.empty:
            continue

        matches["fighter_result"] = matches[
            f"{side}_result"
        ]
        all_fights.append(matches)

    if all_fights:
        fighter_history = pd.concat(
            all_fights, ignore_index=True
        )

        results = (
            fighter_history["fighter_result"]
            .astype(str).str.upper().str.strip()
        )

        recorded_fights = len(fighter_history)
        recorded_wins = int(results.isin(["W", "WIN"]).sum())
        recorded_losses = int(results.isin(["L", "LOSS"]).sum())
        other_results = recorded_fights - recorded_wins - recorded_losses

    else:
        recorded_fights = 0
        recorded_wins = 0
        other_results = 0

    recovery.append({
        "fighter": name,
        "original_ufc_fights": fighter_row["ufc_fights"],
        "original_ufc_wins": fighter_row["ufc_wins"],
        "recorded_fights": recorded_fights,
        "recorded_wins": recorded_wins,
        "other_results": other_results,
        "needs_recovery": pd.isna(
            fighter_row["ufc_fights"]
        )
    })

ufc333_recovery_audit_v17 = pd.DataFrame(recovery)

print("\nFIGHTERS NEEDING RECOVERY")
display(
    ufc333_recovery_audit_v17[
        ufc333_recovery_audit_v17["needs_recovery"]
    ]
)

print("\nHISTORICAL SCORING REFERENCES")
for variable in [
    "event_strength_v2",
    "category_recent_candidates_v2",
    "star_recent_candidates_v1",
    "main_recent_scored_candidates_v10",
    "overall_recent_leaderboard_v11",
    "ranking_points_final_candidate_v3"
]:
    if variable in globals():
        obj = globals()[variable]
        print(
            f"\n{variable}:",
            getattr(obj, "shape", type(obj))
        )
        if isinstance(obj, pd.DataFrame):
            print("Columns:", list(obj.columns))
    else:
        print(f"\n{variable}: NOT FOUND")

print("\n" + "=" * 75)
print("BIG CELL 34 COMPLETE")
print("=" * 75)
print("Next: calculate UFC 333 using the existing model.")
```

```python

import pandas as pd
import numpy as np
from IPython.display import display

print("=" * 75)
print("BIG CELL 35 — UFC 333 FINAL INPUT PREPARATION")
print("=" * 75)

assert "ufc333_fighter_inputs_v16" in globals()
assert "ufc333_recovery_audit_v17" in globals()

fighters = ufc333_fighter_inputs_v16.copy()
recovery = ufc333_recovery_audit_v17.copy()

merged = fighters.merge(
    recovery[
        ["fighter", "recorded_fights",
         "recorded_wins", "other_results"]
    ],
    on="fighter",
    how="left",
    validate="one_to_one"
)

merged["ufc_fights_final"] = merged[
    "ufc_fights"
].fillna(merged["recorded_fights"])

merged["ufc_wins_final"] = merged[
    "ufc_wins"
].fillna(merged["recorded_wins"])

# No historical appearances is not proof of zero UFC fights.
missing_history = (
    merged["fighter"].eq("Asaf Chopurov") &
    merged["ufc_fights"].isna()
)

merged.loc[
    missing_history,
    ["ufc_fights_final", "ufc_wins_final"]
] = np.nan

merged["ufc_win_pct_final"] = np.where(
    merged["ufc_fights_final"] > 0,
    merged["ufc_wins_final"] /
    merged["ufc_fights_final"],
    np.nan
)

merged["fight_totals_source"] = np.where(
    merged["ufc_fights"].notna(),
    "Existing fighter statistics",
    np.where(
        merged["recorded_fights"] > 0,
        "Recovered from fight records — provisional",
        "Missing"
    )
)

# Keep bonus history unchanged when available.
# Missing bonus counts are NOT automatically zero.
merged["bonuses_final"] = merged[
    "previous_bonuses"
]

# Flag missing information for the next scoring step.
required = [
    "ufc_fights_final",
    "ufc_win_pct_final",
    "win_streak",
    "previous_title_fights",
    "previous_main_events",
    "bonuses_final"
]

merged["missing_required_factors"] = (
    merged[required].isna().sum(axis=1)
)

ufc333_fighter_inputs_v18 = merged.copy()

print("\nFINAL FIGHTER DATA")
display(
    merged[
        ["fighter", "ufc_fights_final",
         "ufc_wins_final", "ufc_win_pct_final",
         "win_streak", "previous_title_fights",
         "previous_main_events", "bonuses_final",
         "fight_totals_source",
         "missing_required_factors"]
    ]
)

print("\nCOVERAGE SUMMARY")
print("Fighters:", len(merged))
print(
    "Fighters with fight totals:",
    int(merged["ufc_fights_final"].notna().sum())
)
print(
    "Fighters with win percentages:",
    int(merged["ufc_win_pct_final"].notna().sum())
)
print(
    "Fighters with bonus history:",
    int(merged["bonuses_final"].notna().sum())
)

print("\nMISSING FACTORS BY FIGHTER")
display(
    merged.loc[
        merged["missing_required_factors"] > 0,
        ["fighter", "missing_required_factors",
         "fight_totals_source"]
    ]
)

print("\nBIG CELL 35 COMPLETE")
print("Next: calculate the provisional UFC 333 score.")
```

```python

import pandas as pd
import numpy as np
from IPython.display import display

print("=" * 76)
print("BIG CELL 36 — UFC 333 MODEL RECONSTRUCTION")
print("=" * 76)

assert "ufc333_fighter_inputs_v18" in globals()
assert "ufc333_verified_card_v13" in globals()
assert "category_recent_candidates_v2" in globals()
assert "star_recent_candidates_v1" in globals()
assert "main_recent_scored_candidates_v10" in globals()
assert "overall_recent_leaderboard_v11" in globals()

fighters = ufc333_fighter_inputs_v18.copy()
card = ufc333_verified_card_v13.copy()
category = category_recent_candidates_v2.copy()
star = star_recent_candidates_v1.copy()
main = main_recent_scored_candidates_v10.copy()
leaderboard = overall_recent_leaderboard_v11.copy()

# Build event-level UFC 333 statistics without turning
# unknown values into zero.

def mean_complete(series):
    values = pd.to_numeric(series, errors="coerce")
    if values.isna().any():
        return np.nan
    return float(values.mean())

def sum_complete(series):
    values = pd.to_numeric(series, errors="coerce")
    if values.isna().any():
        return np.nan
    return float(values.sum())

event_stats = {
    "event_name": "UFC 333: Volkanovski vs. Evloev",
    "event_date": pd.Timestamp("2026-10-24"),
    "fight_count": len(card),
    "fighter_appearances": len(fighters),
    "title_fights": 2,
    "avg_ufc_experience": mean_complete(
        fighters["ufc_fights_final"]
    ),
    "avg_ufc_win_pct": mean_complete(
        fighters["ufc_win_pct_final"]
    ),
    "avg_win_streak": mean_complete(
        fighters["win_streak"]
    ),
    "avg_previous_title_fights": mean_complete(
        fighters["previous_title_fights"]
    ),
    "avg_previous_main_events": mean_complete(
        fighters["previous_main_events"]
    ),
    "avg_previous_bonuses": mean_complete(
        fighters["bonuses_final"]
    ),
    "total_ranking_points": np.nan,
    "avg_finish_rate": np.nan
}

ufc333_event_raw_v19 = pd.DataFrame([event_stats])

print("\nUFC 333 EVENT FACTORS")
display(ufc333_event_raw_v19.T)

print("\nEXISTING MODEL NORMALIZATION EXAMPLES")
factor_pairs = [
    ("title_fights", "candidate__title_fights_score"),
    ("avg_ufc_experience", "candidate__avg_ufc_experience_score"),
    ("avg_ufc_win_pct", "candidate__avg_ufc_win_pct_score"),
    ("avg_win_streak", "candidate__avg_win_streak_score"),
    ("avg_finish_rate", "candidate__avg_finish_rate_score"),
    ("avg_previous_title_fights", "candidate__avg_previous_title_fights_score"),
    ("avg_previous_main_events", "candidate__avg_previous_main_events_score"),
    ("avg_previous_bonuses", "candidate__avg_previous_bonuses_score"),
    ("total_ranking_points", "candidate__total_ranking_points_score")
]

normalization_rows = []

for raw_col, score_col in factor_pairs:
    if raw_col not in category.columns or score_col not in category.columns:
        continue

    sample = category[[raw_col, score_col]].copy()
    sample[raw_col] = pd.to_numeric(
        sample[raw_col], errors="coerce"
    )
    sample[score_col] = pd.to_numeric(
        sample[score_col], errors="coerce"
    )
    sample = sample.dropna()

    normalization_rows.append({
        "factor": raw_col,
        "score_column": score_col,
        "observed_raw_min": sample[raw_col].min(),
        "observed_raw_max": sample[raw_col].max(),
        "observed_score_min": sample[score_col].min(),
        "observed_score_max": sample[score_col].max(),
        "sample_count": len(sample)
    })

ufc333_normalization_audit_v19 = pd.DataFrame(
    normalization_rows
)

display(ufc333_normalization_audit_v19)

print("\nMAIN EVENT INPUTS")
main_fighters = fighters[
    fighters["fighter"].isin([
        "Alexander Volkanovski",
        "Movsar Evloev"
    ])
]

display(
    main_fighters[
        ["fighter", "ufc_fights_final",
         "ufc_win_pct_final", "win_streak",
         "previous_title_fights",
         "previous_main_events", "bonuses_final"]
    ]
)

print("\nMISSING UFC 333 FACTORS")
missing = ufc333_event_raw_v19.columns[
    ufc333_event_raw_v19.iloc[0].isna()
].tolist()
print(missing)

print("\nEXISTING RECENT MODEL SCORES")
display(
    leaderboard[
        ["event_name", "card_quality_score",
         "star_experience_score",
         "main_event_strength_score",
         "overall_event_strength_v2"]
    ].head(5)
)

print("\n" + "=" * 76)
print("BIG CELL 36 COMPLETE")
print("=" * 76)
print("Next: resolve remaining inputs and rank UFC 333.")
```

```python

import pandas as pd
import numpy as np
from IPython.display import display

print("=" * 75)
print("BIG CELL 37 — UFC 333 PROVISIONAL SCORE RANGE")
print("=" * 75)

fighters = ufc333_fighter_inputs_v18.copy()
category = category_recent_candidates_v2.copy()
main = main_recent_scored_candidates_v10.copy()

# Calculate available-card averages.
# Missing fighters are excluded ONLY for the provisional estimate.
raw = {
    "title_fights": 2.0,
    "avg_ufc_experience": fighters["ufc_fights_final"].mean(),
    "avg_ufc_win_pct": fighters["ufc_win_pct_final"].mean(),
    "avg_win_streak": fighters["win_streak"].mean(),
    "avg_previous_title_fights": fighters["previous_title_fights"].mean(),
    "avg_previous_main_events": fighters["previous_main_events"].mean(),
    "avg_previous_bonuses": fighters["bonuses_final"].mean(),
    "total_ranking_points": np.nan,
    "avg_finish_rate": np.nan
}

factor_map = {
    "title_fights": "candidate__title_fights_score",
    "avg_ufc_experience": "candidate__avg_ufc_experience_score",
    "avg_ufc_win_pct": "candidate__avg_ufc_win_pct_score",
    "avg_win_streak": "candidate__avg_win_streak_score",
    "avg_finish_rate": "candidate__avg_finish_rate_score",
    "total_ranking_points": "candidate__total_ranking_points_score",
    "avg_previous_title_fights": "candidate__avg_previous_title_fights_score",
    "avg_previous_main_events": "candidate__avg_previous_main_events_score",
    "avg_previous_bonuses": "candidate__avg_previous_bonuses_score"
}

# Fit the observed raw-to-score mapping for each factor.
# This is an approximation of the existing normalization,
# NOT a claim to have recovered its original parameters.

factor_results = {}

for factor, score_col in factor_map.items():
    reference = category[[factor, score_col]].copy()
    reference = reference.apply(
        pd.to_numeric, errors="coerce"
    ).dropna()

    if len(reference) < 2:
        continue

    x = reference[factor].to_numpy()
    y = reference[score_col].to_numpy()

    slope, intercept = np.polyfit(x, y, 1)

    observed_min = float(y.min())
    observed_max = float(y.max())

    if pd.notna(raw[factor]):
        estimated = float(np.clip(
            slope * raw[factor] + intercept, 0, 100
        ))
        low = estimated
        high = estimated
    else:
        estimated = np.nan
        low = observed_min
        high = observed_max

    factor_results[factor] = {
        "raw_value": raw[factor],
        "estimated_score": estimated,
        "low_score": low,
        "high_score": high
    }

factor_table = pd.DataFrame(factor_results).T

card_factors = [
    "title_fights",
    "total_ranking_points",
    "avg_ufc_experience",
    "avg_ufc_win_pct",
    "avg_win_streak",
    "avg_finish_rate"
]

star_factors = [
    "avg_previous_title_fights",
    "avg_previous_main_events",
    "avg_previous_bonuses"
]

card_low = factor_table.loc[
    card_factors, "low_score"
].mean()

card_high = factor_table.loc[
    card_factors, "high_score"
].mean()

star_low = factor_table.loc[
    star_factors, "low_score"
].mean()

star_high = factor_table.loc[
    star_factors, "high_score"
].mean()

# The main event's full score also depends on rankings.
# Use observed recent main-event scores as a conservative
# illustrative range rather than inventing a UFC 333 score.

main_scores = pd.to_numeric(
    main["main_event_score_candidate_v3"],
    errors="coerce"
).dropna()

main_low = float(main_scores.min())
main_high = float(main_scores.max())

overall_low = (
    .50 * card_low +
    .25 * star_low +
    .25 * main_low
)

overall_high = (
    .50 * card_high +
    .25 * star_high +
    .25 * main_high
)

ufc333_score_range_v20 = pd.DataFrame([{
    "event_name": "UFC 333: Volkanovski vs. Evloev",
    "card_quality_low": card_low,
    "card_quality_high": card_high,
    "star_experience_low": star_low,
    "star_experience_high": star_high,
    "main_event_low": main_low,
    "main_event_high": main_high,
    "overall_low": overall_low,
    "overall_high": overall_high,
    "status": "Illustrative provisional range"
}])

print("\nFACTOR ESTIMATES")
display(factor_table.round(3))

print("\nUFC 333 PROVISIONAL RANGE")
display(ufc333_score_range_v20.round(3))

print("\nEXISTING LEADERBOARD COMPARISON")
if "complete_event_leaderboard_v12" in globals():
    board = complete_event_leaderboard_v12.copy()
    scores = pd.to_numeric(
        board["event_strength_final"],
        errors="coerce"
    ).dropna()

    possible_best_rank = int(
        (scores > overall_high).sum() + 1
    )
    possible_worst_rank = int(
        (scores > overall_low).sum() + 1
    )

    print(
        "Possible rank range:",
        possible_best_rank,
        "to",
        possible_worst_rank,
        "out of",
        len(scores) + 1
    )

print("\nIMPORTANT: This is a provisional range, not an exact score.")
print("BIG CELL 37 COMPLETE")
```

```python

import pandas as pd
import numpy as np
import re
import unicodedata
from IPython.display import display

print("=" * 75)
print("BIG CELL 39 — UFC 333 FINISH RATE RECOVERY")
print("=" * 75)

assert "backup" in globals()
assert "ufc333_fighter_inputs_v18" in globals()

historical = backup["ufc_fights"].copy()
fighters = ufc333_fighter_inputs_v18.copy()

def fighter_key(name):
    if pd.isna(name):
        return ""
    name = unicodedata.normalize("NFKD", str(name))
    name = "".join(
        c for c in name if not unicodedata.combining(c)
    )
    return re.sub(r"[^a-z0-9]", "", name.lower())

historical["event_date"] = pd.to_datetime(
    historical["event_date"], errors="coerce"
)

historical = historical[
    historical["event_date"] < pd.Timestamp("2026-10-24")
].copy()

records = []

for _, row in fighters.iterrows():
    name = row["fighter"]
    key = fighter_key(name)

    appearances = []

    for side in ["f1", "f2"]:
        name_col = f"{side}_name"
        finish_col = f"{side}_finish_rate_before"
        finishes_col = f"{side}_ufc_finishes_before"

        if not all(
            col in historical.columns
            for col in [name_col, finish_col, finishes_col]
        ):
            continue

        matches = historical[
            historical[name_col].map(fighter_key).eq(key)
        ].copy()

        if matches.empty:
            continue

        matches["prior_finish_rate"] = pd.to_numeric(
            matches[finish_col], errors="coerce"
        )
        matches["prior_ufc_finishes"] = pd.to_numeric(
            matches[finishes_col], errors="coerce"
        )
        matches["fighter_side"] = side

        appearances.append(matches)

    if appearances:
        all_appearances = pd.concat(
            appearances, ignore_index=True
        ).sort_values("event_date", ascending=False)

        latest = all_appearances.iloc[0]

        records.append({
            "fighter": name,
            "last_historical_date": latest["event_date"],
            "finish_rate_before_last_fight":
                latest["prior_finish_rate"],
            "ufc_finishes_before_last_fight":
                latest["prior_ufc_finishes"],
            "last_fight_result":
                row.get("last_fight_result", np.nan),
            "finish_rate_status":
                "Historical baseline — needs latest-fight update"
        })
    else:
        records.append({
            "fighter": name,
            "last_historical_date": pd.NaT,
            "finish_rate_before_last_fight": np.nan,
            "ufc_finishes_before_last_fight": np.nan,
            "last_fight_result": np.nan,
            "finish_rate_status": "No historical record"
        })

ufc333_finish_recovery_v22 = pd.DataFrame(records)

print("\nRECOVERED FIGHTER FINISH STATISTICS")
display(ufc333_finish_recovery_v22)

print("\nFINISH RATE COVERAGE")
print(
    "Available historical finish rates:",
    int(
        ufc333_finish_recovery_v22[
            "finish_rate_before_last_fight"
        ].notna().sum()
    ),
    "of",
    len(ufc333_finish_recovery_v22)
)

print("\nLATEST-FIGHT METHOD COVERAGE")
print(
    "Method column available:",
    "method" in historical.columns
)

print("\nBIG CELL 39 COMPLETE")
print("Next: update finish rates and calculate UFC 333.")
```

```python

import pandas as pd
import numpy as np
import re
import unicodedata
from IPython.display import display

print("=" * 75)
print("BIG CELL 40 — UFC 333 UPDATED FINISH RATES")
print("=" * 75)

fighters = ufc333_fighter_inputs_v18.copy()
baselines = ufc333_finish_recovery_v22.copy()
history = backup["ufc_fights"].copy()
recent = scoring_fights.copy()

def clean_name(value):
    if pd.isna(value):
        return ""
    value = unicodedata.normalize("NFKD", str(value))
    value = "".join(
        c for c in value if not unicodedata.combining(c)
    )
    return re.sub(r"[^a-z0-9]", "", value.lower())

history["_date"] = pd.to_datetime(
    history["date"], errors="coerce"
)

recent["event_date"] = pd.to_datetime(
    recent["event_date"], errors="coerce"
)

history = history[
    history["_date"] < pd.Timestamp("2026-10-24")
].copy()

recent = recent[
    recent["event_date"] < pd.Timestamp("2026-10-24")
].copy()

# Identify each fighter's UFC wins by finish in the original
# historical data. Only known methods count as finishes.
finish_methods = [
    "KO/TKO",
    "KO",
    "TKO",
    "SUB",
    "SUBMISSION"
]

records = []

for _, fighter in fighters.iterrows():
    name = fighter["fighter"]
    key = clean_name(name)

    matches = []

    for side in ["f1", "f2"]:
        subset = history[
            history[f"{side}_name"].map(clean_name).eq(key)
        ].copy()

        if subset.empty:
            continue

        subset["_result"] = subset[
            f"{side}_result"
        ] if f"{side}_result" in subset.columns else np.nan

        matches.append(subset)

    if matches:
        all_history = pd.concat(
            matches, ignore_index=True
        ).drop_duplicates(
            subset=["fight_url"]
        ) if "fight_url" in history.columns else pd.concat(
            matches, ignore_index=True
        )

        methods = (
            all_history["method"]
            .astype(str).str.upper().str.strip()
        )

        results = (
            all_history["_result"]
            .astype(str).str.upper().str.strip()
        )

        wins_by_finish = (
            results.isin(["W", "WIN"]) &
            methods.apply(
                lambda x: any(
                    term in x for term in finish_methods
                )
            )
        ).sum()

        historical_finish_count = int(wins_by_finish)

    else:
        historical_finish_count = 0

    baseline_row = baselines[
        baselines["fighter"].eq(name)
    ]

    baseline_finishes = (
        baseline_row["ufc_finishes_before_last_fight"].iloc[0]
        if not baseline_row.empty else np.nan
    )

    # Historical fight method counts are an independent
    # cross-check, not automatically a complete career total.
    records.append({
        "fighter": name,
        "ufc_fights_final": fighter["ufc_fights_final"],
        "historical_recorded_finishes":
            historical_finish_count,
        "baseline_prior_finishes": baseline_finishes,
        "finish_rate_baseline":
            baseline_row["finish_rate_before_last_fight"].iloc[0]
            if not baseline_row.empty else np.nan,
        "finish_data_status":
            "Historical evidence — provisional"
            if not pd.isna(baseline_finishes)
            else "Incomplete historical coverage"
    })

ufc333_finish_audit_v23 = pd.DataFrame(records)

print("\nUPDATED FINISH INFORMATION")
display(ufc333_finish_audit_v23)

print("\nMISSING BASELINE FINISH DATA")
display(
    ufc333_finish_audit_v23[
        ufc333_finish_audit_v23[
            "baseline_prior_finishes"
        ].isna()
    ]
)

print("\nBIG CELL 40 COMPLETE")
print("Next: calculate the final provisional UFC 333 estimate.")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

print("=" * 76)
print("BIG CELL 41 — UFC 333 PROVISIONAL SCORE AND RANK")
print("=" * 76)

fighters = ufc333_fighter_inputs_v18.copy()
finish = ufc333_finish_audit_v23.copy()
reference = category_recent_candidates_v2.copy()
main_reference = main_recent_scored_candidates_v10.copy()

# Preserve source records and label every assumption.
data = fighters.merge(
    finish[["fighter", "historical_recorded_finishes"]],
    on="fighter", how="left"
)

data["estimated_finish_rate"] = (
    pd.to_numeric(
        data["historical_recorded_finishes"], errors="coerce"
    ) /
    pd.to_numeric(
        data["ufc_wins_final"], errors="coerce"
    ).replace(0, np.nan)
)

# For missing values, use the observed median of the
# other fighters on this card. This is an imputation,
# NOT a verified UFC statistic.
raw_columns = [
    "ufc_fights_final", "ufc_win_pct_final",
    "win_streak", "previous_title_fights",
    "previous_main_events", "bonuses_final",
    "estimated_finish_rate"
]

imputation_log = []

for col in raw_columns:
    data[col] = pd.to_numeric(data[col], errors="coerce")
    missing = int(data[col].isna().sum())

    if missing:
        median = data[col].median()
        data[col] = data[col].fillna(median)
        imputation_log.append({
            "factor": col,
            "imputed_fighters": missing,
            "imputation_value": median
        })

# Missing rankings are imputed at the median of
# recent UFC events, not treated as zero.
ranking_median = pd.to_numeric(
    reference["total_ranking_points"],
    errors="coerce"
).median()

raw = {
    "title_fights": 2.0,
    "avg_ufc_experience": data["ufc_fights_final"].mean(),
    "avg_ufc_win_pct": data["ufc_win_pct_final"].mean(),
    "avg_win_streak": data["win_streak"].mean(),
    "avg_finish_rate": data["estimated_finish_rate"].mean(),
    "total_ranking_points": ranking_median,
    "avg_previous_title_fights":
        data["previous_title_fights"].mean(),
    "avg_previous_main_events":
        data["previous_main_events"].mean(),
    "avg_previous_bonuses":
        data["bonuses_final"].mean()
}

factor_map = {
    "title_fights": "candidate__title_fights_score",
    "avg_ufc_experience": "candidate__avg_ufc_experience_score",
    "avg_ufc_win_pct": "candidate__avg_ufc_win_pct_score",
    "avg_win_streak": "candidate__avg_win_streak_score",
    "avg_finish_rate": "candidate__avg_finish_rate_score",
    "total_ranking_points": "candidate__total_ranking_points_score",
    "avg_previous_title_fights":
        "candidate__avg_previous_title_fights_score",
    "avg_previous_main_events":
        "candidate__avg_previous_main_events_score",
    "avg_previous_bonuses":
        "candidate__avg_previous_bonuses_score"
}

def estimated_factor_score(raw_col, score_col, value):
    sample = reference[[raw_col, score_col]].copy()
    sample = sample.apply(
        pd.to_numeric, errors="coerce"
    ).dropna()

    x = sample[raw_col].to_numpy()
    y = sample[score_col].to_numpy()

    slope, intercept = np.polyfit(x, y, 1)

    return float(np.clip(
        slope * value + intercept, 0, 100
    ))

factor_scores = {
    key: estimated_factor_score(
        key, score_col, raw[key]
    )
    for key, score_col in factor_map.items()
}

card_keys = [
    "title_fights", "total_ranking_points",
    "avg_ufc_experience", "avg_ufc_win_pct",
    "avg_win_streak", "avg_finish_rate"
]

star_keys = [
    "avg_previous_title_fights",
    "avg_previous_main_events",
    "avg_previous_bonuses"
]

card_score = np.mean([
    factor_scores[k] for k in card_keys
])

star_score = np.mean([
    factor_scores[k] for k in star_keys
])

# Main-event score: reference-based estimate.
# The true model needs verified pre-event rankings.
main_event_estimate = float(
    pd.to_numeric(
        main_reference["main_event_score_candidate_v3"],
        errors="coerce"
    ).median()
)

overall = (
    0.50 * card_score +
    0.25 * star_score +
    0.25 * main_event_estimate
)

# Compare with the saved historical/recent leaderboard.
if "complete_event_leaderboard_v12" in globals():
    leaderboard = complete_event_leaderboard_v12.copy()
else:
    leaderboard = pd.read_csv(
        "/content/ufc_portfolio_v2/"
        "complete_event_strength_leaderboard.csv"
    )

existing_scores = pd.to_numeric(
    leaderboard["event_strength_final"],
    errors="coerce"
).dropna()

rank = int((existing_scores > overall).sum() + 1)

ufc333_provisional_result_v24 = pd.DataFrame([{
    "event_name": "UFC 333: Volkanovski vs. Evloev",
    "event_date": "2026-10-24",
    "card_quality_score": card_score,
    "star_experience_score": star_score,
    "main_event_strength_score": main_event_estimate,
    "event_strength_estimate": overall,
    "provisional_rank": rank,
    "comparison_events": len(existing_scores) + 1,
    "data_status": "PROVISIONAL — IMPUTED"
}])

print("\nUFC 333 ESTIMATED RESULT")
display(ufc333_provisional_result_v24.round(3))

print("\nIMPUTATION AUDIT")
display(pd.DataFrame(imputation_log))

print("\nFACTOR SCORES")
display(
    pd.DataFrame(
        [{"factor": k, "raw": raw[k], "score": v}
         for k, v in factor_scores.items()]
    ).round(3)
)

# Save separate provisional files, leaving verified
# leaderboard files unchanged.
output_dir = Path("/content/ufc_portfolio_v2")
output_dir.mkdir(parents=True, exist_ok=True)

ufc333_provisional_result_v24.to_csv(
    output_dir / "ufc333_provisional_score.csv",
    index=False
)

data.to_csv(
    output_dir / "ufc333_provisional_fighter_inputs.csv",
    index=False
)

print("\nFILES SAVED:")
print(output_dir / "ufc333_provisional_score.csv")
print(output_dir / "ufc333_provisional_fighter_inputs.csv")

print("\nBIG CELL 41 COMPLETE")
```

```python

# BIG CELL 44 — RANKINGS-FOCUSED CARD QUALITY TEST
# Original scores remain unchanged.

import pandas as pd
import numpy as np
from IPython.display import display

weights = {
    "title_fights": 0.20,
    "total_ranking_points": 0.40,
    "avg_ufc_experience": 0.10,
    "avg_ufc_win_pct": 0.10,
    "avg_win_streak": 0.10,
    "avg_finish_rate": 0.10,
}

def test_card_quality(df, prefix, group):
    result = df.copy()

    factor_cols = {
        factor: prefix + factor + "_score"
        for factor in weights
    }

    missing_cols = [
        col for col in factor_cols.values()
        if col not in result.columns
    ]

    if missing_cols:
        raise ValueError(
            f"{group} missing columns: {missing_cols}"
        )

    values = result[list(factor_cols.values())].apply(
        pd.to_numeric, errors="coerce"
    )

    # Exclude incomplete rows from this experiment only.
    incomplete = values.isna().any(axis=1)
    print(
        f"{group}: {len(result)} total, "
        f"{incomplete.sum()} incomplete"
    )

    result = result.loc[~incomplete].copy()
    values = values.loc[~incomplete].copy()

    if result.empty:
        raise ValueError(f"{group}: no complete rows")

    if not values.ge(0).all().all() or not values.le(100).all().all():
        raise ValueError(
            f"{group}: factor outside 0–100"
        )

    result["old_card_quality_rebuilt"] = (
        values.mean(axis=1)
    )

    result["new_card_quality_test"] = sum(
        values[factor_cols[factor]] * weight
        for factor, weight in weights.items()
    )

    result["card_quality_change"] = (
        result["new_card_quality_test"]
        - result["old_card_quality_rebuilt"]
    )

    result["dataset"] = group

    return result

historical_test = test_card_quality(
    historical_norm, "", "Historical"
)

recent_test = test_card_quality(
    category_recent_candidates_v2,
    "candidate__",
    "Recent"
)

comparison = pd.concat(
    [historical_test, recent_test],
    ignore_index=True
)

comparison["new_card_rank"] = (
    comparison["new_card_quality_test"]
    .rank(method="min", ascending=False)
    .astype(int)
)

show_cols = [
    "event_name",
    "dataset",
    "old_card_quality_rebuilt",
    "new_card_quality_test",
    "card_quality_change",
    "new_card_rank"
]

print("\nTOTAL EVENTS TESTED:", len(comparison))

print("\nUFC 330 RESULTS:")
display(
    comparison.loc[
        comparison["event_name"].str.contains(
            "UFC 330", case=False, na=False
        ),
        show_cols
    ].round(2)
)

print("\nTOP 15 NEW CARD QUALITY SCORES:")
display(
    comparison.nlargest(
        15, "new_card_quality_test"
    )[show_cols].round(2)
)

print("\nBIGGEST SCORE DROPS:")
display(
    comparison.nsmallest(
        10, "card_quality_change"
    )[show_cols].round(2)
)

print("\nIMPORTANT:")
print("Men's 1.5x ranking multiplier not applied yet.")
print("No original scores changed.")

card_quality_experiment_v44 = comparison.copy()

print("\nBIG CELL 44 COMPLETE")
```

```python

# BIG CELL 46 — 1.5x MEN'S RANKING EXPERIMENT
# Does not overwrite original model scores.

import pandas as pd
import numpy as np
from IPython.display import display

base = card_quality_experiment_v44.copy()
fights = scoring_fights.copy()

required = [
    "event_name", "f1_rank_before",
    "f2_rank_before"
]
missing = [c for c in required if c not in fights]
if missing:
    raise ValueError("Missing fight columns: " + str(missing))

division_col = next(
    (c for c in [
        "WEIGHTCLASS", "weight_class",
        "weightclass", "ranking_division"
    ] if c in fights.columns),
    None
)
if division_col is None:
    raise ValueError("No weight-class column available")

def rank_points(rank):
    rank = pd.to_numeric(rank, errors="coerce")
    if pd.isna(rank) or rank < 0 or rank > 15:
        return 0.0
    return 16.0 - rank

rows = []

for _, fight in fights.iterrows():
    division = str(fight[division_col]).lower()

    if "women" in division or "female" in division:
        multiplier = 1.0
    elif "catch" in division or division in ["nan", ""]:
        multiplier = np.nan
    else:
        multiplier = 1.5

    for side in ["f1", "f2"]:
        rank = fight[f"{side}_rank_before"]
        points = rank_points(rank)

        rows.append({
            "event_name": fight["event_name"],
            "base_points": points,
            "adjusted_points": (
                points * multiplier
                if pd.notna(multiplier)
                else np.nan
            ),
            "ranked": points > 0,
            "division_unknown": pd.isna(multiplier)
        })

fighter_points = pd.DataFrame(rows)

summary = fighter_points.groupby(
    "event_name", as_index=False
).agg(
    original_fighter_points=("base_points", "sum"),
    adjusted_fighter_points=("adjusted_points", "sum"),
    ranked_appearances=("ranked", "sum"),
    unknown_divisions=("division_unknown", "sum")
)

summary["ranking_multiplier_ratio"] = np.where(
    summary["original_fighter_points"] > 0,
    summary["adjusted_fighter_points"]
    / summary["original_fighter_points"],
    np.nan
)

test = base.merge(
    summary, on="event_name", how="left",
    validate="many_to_one"
)

# Preserve 0–100 scale by using the maximum 1.5x adjustment.
# This is a sensitivity test, not final historical normalization.
ranking_col = (
    "candidate__total_ranking_points_score"
    if False else None
)

def existing_ranking_score(row):
    col = (
        "candidate__total_ranking_points_score"
        if row["dataset"] == "Recent"
        else "total_ranking_points_score"
    )
    return row[col]

test["ranking_score_before"] = test.apply(
    existing_ranking_score, axis=1
)

test["ranking_score_after"] = (
    test["ranking_score_before"]
    * test["ranking_multiplier_ratio"]
    / 1.5
).clip(0, 100)

test["adjusted_card_quality"] = (
    test["new_card_quality_test"]
    + 0.40 * (
        test["ranking_score_after"]
        - test["ranking_score_before"]
    )
)

test["adjusted_card_quality"] = (
    test["adjusted_card_quality"].clip(0, 100)
)

test["adjusted_rank_all"] = (
    test["adjusted_card_quality"]
    .rank(ascending=False, method="min")
)

# Identify 2026 events from the earlier comparison.
if "cards_2026" in globals():
    names_2026 = set(cards_2026["event_name"])
    year_test = test[
        test["event_name"].isin(names_2026)
    ].copy()
else:
    year_test = test[
        test["dataset"] == "Recent"
    ].copy()

year_test = year_test.sort_values(
    "adjusted_card_quality", ascending=False
)
year_test["new_2026_rank"] = range(
    1, len(year_test) + 1
)

cols = [
    "new_2026_rank", "event_name",
    "new_card_quality_test",
    "adjusted_card_quality",
    "ranked_appearances",
    "unknown_divisions"
]

print("2026 LEADERBOARD — MULTIPLIER EXPERIMENT")
display(year_test[cols].round(2).reset_index(drop=True))

print("\nUFC 330:")
display(
    year_test[
        year_test["event_name"].str.contains(
            r"UFC\s+330\b", case=False, na=False
        )
    ][cols].round(2)
)

print("\nDATA CHECK:")
print("Events with missing fighter-level ranking totals:",
      int(test["ranking_multiplier_ratio"].isna().sum()))
print("Events with unknown fight divisions:",
      int(test["unknown_divisions"].fillna(0).gt(0).sum()))
print("Original scores unchanged.")

ranking_multiplier_experiment_v46 = test.copy()

print("\nBIG CELL 46 COMPLETE")
```

```python

# BIG CELL 49 — VERIFIED RANKING MULTIPLIER TEST
# Uses existing fighter-level ranking records.
# Does not overwrite original model.

import pandas as pd
import numpy as np
from IPython.display import display

rankings = ranking_points_final_candidate_v3.copy()
base = card_quality_experiment_v44.copy()

required = [
    "event_name", "fighter",
    "rank_before", "ranking_division"
]

missing = [c for c in required if c not in rankings.columns]
if missing:
    raise ValueError(f"Missing ranking columns: {missing}")

rankings["rank_numeric"] = pd.to_numeric(
    rankings["rank_before"], errors="coerce"
)

rankings["ranking_points"] = np.where(
    rankings["rank_numeric"].between(0, 15),
    16 - rankings["rank_numeric"],
    0
)

division = rankings["ranking_division"].fillna("").str.lower()

rankings["gender_multiplier"] = np.where(
    division.str.contains("women"),
    1.0,
    np.where(division.eq(""), np.nan, 1.5)
)

rankings["adjusted_ranking_points"] = (
    rankings["ranking_points"]
    * rankings["gender_multiplier"]
)

event_rankings = rankings.groupby(
    "event_name", as_index=False
).agg(
    original_points=("ranking_points", "sum"),
    adjusted_points=("adjusted_ranking_points", "sum"),
    ranked_fighters=("rank_numeric", lambda x: x.between(0,15).sum()),
    unknown_divisions=("gender_multiplier", lambda x: x.isna().sum())
)

event_rankings["ranking_boost_ratio"] = np.where(
    event_rankings["original_points"] > 0,
    event_rankings["adjusted_points"]
    / event_rankings["original_points"],
    np.nan
)

experiment = base.merge(
    event_rankings,
    on="event_name",
    how="left",
    validate="many_to_one"
)

experiment["ranking_score_before"] = np.where(
    experiment["dataset"].eq("Recent"),
    experiment["candidate__total_ranking_points_score"],
    experiment["total_ranking_points_score"]
)

# Preserve the original normalized ranking score and
# apply a bounded multiplier without arbitrary downscaling.
experiment["ranking_score_adjusted"] = (
    experiment["ranking_score_before"]
    * experiment["ranking_boost_ratio"]
).clip(0, 100)

experiment["card_quality_adjusted"] = (
    experiment["new_card_quality_test"]
    + 0.40 * (
        experiment["ranking_score_adjusted"]
        - experiment["ranking_score_before"]
    )
)

# Only rank events with complete multiplier data.
eligible = experiment[
    experiment["ranking_boost_ratio"].notna()
    & experiment["unknown_divisions"].eq(0)
].copy()

if "cards_2026" in globals():
    names_2026 = set(cards_2026["event_name"])
    eligible_2026 = eligible[
        eligible["event_name"].isin(names_2026)
    ].copy()
else:
    eligible_2026 = eligible[
        eligible["dataset"].eq("Recent")
    ].copy()

eligible_2026 = eligible_2026.sort_values(
    "card_quality_adjusted", ascending=False
)

eligible_2026["2026_rank"] = range(
    1, len(eligible_2026) + 1
)

print("EVENTS WITH COMPLETE RANKING MULTIPLIER:",
      len(eligible))
print("2026 EVENTS ELIGIBLE:",
      len(eligible_2026))

cols = [
    "2026_rank", "event_name",
    "new_card_quality_test",
    "card_quality_adjusted",
    "ranked_fighters",
    "ranking_boost_ratio"
]

print("\n2026 MULTIPLIER TEST:")
display(
    eligible_2026[cols].round(2).head(20)
)

print("\nUFC 330:")
display(
    eligible_2026[
        eligible_2026["event_name"].str.contains(
            r"UFC\s+330\b", case=False, na=False
        )
    ][cols].round(2)
)

print("\nBIG CELL 49 COMPLETE")
```

```python

# BIG CELL 52 — FINAL CORRECTED V3 TEST
# Rewards ranked depth and ranked-vs-ranked fights.
# Does not change original scores.

import pandas as pd
import numpy as np
from IPython.display import display

r = ranking_points_final_candidate_v3.copy()
base = card_quality_experiment_v44.copy()

# ---------- RANKING DATA ----------

r["points"] = pd.to_numeric(
    r["ranking_points_v2"], errors="coerce"
)

wc = r["weightclass"].fillna("").astype(str).str.lower()

r["multiplier"] = np.where(
    wc.str.contains("women"), 1.0,
    np.where(wc.str.contains("catch"), np.nan, 1.5)
)

r["weighted_points"] = np.where(
    r["points"].eq(0), 0.0,
    r["points"] * r["multiplier"]
)

r["is_ranked"] = r["points"].gt(0)

# Every fighter appearance needs a valid fight ID.
r["fight_id"] = r["fight_url"].astype(str)

fight_level = r.groupby(
    ["event_name", "fight_id"], as_index=False
).agg(
    ranked_fighters=("is_ranked", "sum"),
    ranking_points=("weighted_points", "sum"),
    appearances=("fighter", "count"),
    missing_points=("points", lambda x: x.isna().sum()),
    missing_weighted=("weighted_points", lambda x: x.isna().sum())
)

fight_level["ranked_matchup"] = (
    fight_level["ranked_fighters"].eq(2)
    & fight_level["appearances"].eq(2)
)

event_depth = fight_level.groupby(
    "event_name", as_index=False
).agg(
    v3_total_fights=("fight_id", "count"),
    v3_ranked_fighters=("ranked_fighters", "sum"),
    v3_ranked_matchups=("ranked_matchup", "sum"),
    v3_weighted_points=("ranking_points", "sum"),
    v3_missing_points=("missing_points", "sum"),
    v3_missing_weighted=("missing_weighted", "sum"),
    v3_appearances=("appearances", "sum")
)

event_depth["ranking_depth_score"] = (
    100 * event_depth["v3_weighted_points"]
    / (event_depth["v3_total_fights"] * 48)
).clip(0, 100)

event_depth["ranked_matchup_score"] = (
    100 * event_depth["v3_ranked_matchups"]
    / (event_depth["v3_total_fights"] * 0.5)
).clip(0, 100)

event_depth["ranking_data_complete"] = (
    event_depth["v3_missing_points"].eq(0)
    & event_depth["v3_missing_weighted"].eq(0)
    & event_depth["v3_appearances"].eq(
        event_depth["v3_total_fights"] * 2
    )
)

v3 = base.merge(
    event_depth, on="event_name",
    how="left", validate="many_to_one"
)

# ---------- EXISTING FACTORS ----------

# Search score columns by keywords, excluding main-event scores.
def factor_by_keywords(frame, keywords):
    matches = [
        c for c in frame.columns
        if "score" in c.lower()
        and all(k in c.lower() for k in keywords)
        and not any(x in c.lower() for x in [
            "main_event", "card_quality",
            "adjusted", "rebuilt"
        ])
    ]

    hist = [c for c in matches if not c.startswith("candidate__")]
    recent = [c for c in matches if c.startswith("candidate__")]

    hcol = hist[0] if hist else None
    rcol = recent[0] if recent else hcol

    print(keywords, "| historical:", hcol, "| recent:", rcol)

    h = (
        pd.to_numeric(frame[hcol], errors="coerce")
        if hcol else pd.Series(np.nan, index=frame.index)
    )

    rr = (
        pd.to_numeric(frame[rcol], errors="coerce")
        if rcol else pd.Series(np.nan, index=frame.index)
    )

    return pd.Series(
        np.where(frame["dataset"].eq("Recent"), rr, h),
        index=frame.index
    )

v3["title_factor"] = factor_by_keywords(v3, ["title", "fight"])

# Experience: use UFC fights or UFC experience naming.
experience_options = [
    ["ufc", "experience"],
    ["ufc", "fights"],
    ["experience"],
    ["total", "fights"]
]

v3["experience_factor"] = np.nan

for keywords in experience_options:
    candidate = factor_by_keywords(v3, keywords)
    if candidate.notna().any():
        v3["experience_factor"] = candidate
        break

v3["win_pct_factor"] = factor_by_keywords(v3, ["win", "pct"])
v3["streak_factor"] = factor_by_keywords(v3, ["streak"])
v3["finish_factor"] = factor_by_keywords(v3, ["finish", "rate"])

# ---------- FINAL SCORING ----------

required = [
    "ranking_depth_score",
    "ranked_matchup_score",
    "title_factor",
    "experience_factor",
    "win_pct_factor",
    "streak_factor",
    "finish_factor"
]

v3["complete_v3"] = (
    v3[required].notna().all(axis=1)
    & v3["ranking_data_complete"].eq(True)
)

v3["card_quality_v3"] = (
    0.35 * v3["ranking_depth_score"]
    + 0.20 * v3["ranked_matchup_score"]
    + 0.15 * v3["title_factor"]
    + 0.10 * v3["experience_factor"]
    + 0.10 * v3["win_pct_factor"]
    + 0.05 * v3["streak_factor"]
    + 0.05 * v3["finish_factor"]
).where(v3["complete_v3"])

card_quality_v3_experiment = v3.copy()

print("\nCOMPLETE V3 EVENTS:", int(v3["complete_v3"].sum()))

print("\nUFC 330:")
display(
    v3.loc[
        v3["event_name"].str.contains(
            r"UFC\s+330\b", case=False, na=False
        ),
        [
            "event_name",
            "v3_total_fights",
            "v3_ranked_fighters",
            "v3_ranked_matchups",
            "ranking_depth_score",
            "ranked_matchup_score",
            "new_card_quality_test",
            "card_quality_v3",
            "complete_v3"
        ]
    ].round(2)
)

print("\n2026 V3 LEADERBOARD:")

year_v3 = v3[
    v3["event_name"].isin(set(cards_2026["event_name"]))
    & v3["complete_v3"]
].sort_values(
    "card_quality_v3", ascending=False
).copy()

year_v3["v3_rank"] = range(1, len(year_v3) + 1)

display(
    year_v3[[
        "v3_rank",
        "event_name",
        "v3_total_fights",
        "v3_ranked_fighters",
        "v3_ranked_matchups",
        "card_quality_v3"
    ]].round(2).head(25)
)

print("\nBIG CELL 52 COMPLETE")
```

```python

# BIG CELL 53 — CARD DEPTH EXPERIMENT
# Tests how ranking depth beyond the main event
# affects Card Quality. Original V3 unchanged.

import pandas as pd
import numpy as np
from IPython.display import display

r = ranking_points_final_candidate_v3.copy()
v3 = card_quality_v3_experiment.copy()

r["points"] = pd.to_numeric(
    r["ranking_points_v2"], errors="coerce"
).fillna(0)

wc = r["weightclass"].fillna("").astype(str).str.lower()

r["multiplier"] = np.where(
    wc.str.contains("women"), 1.0, 1.5
)

r["weighted_points"] = r["points"] * r["multiplier"]

r["main_event_flag"] = (
    r["is_main_event"].fillna(False).astype(bool)
)

# Ranking depth outside the main event
non_main = r[~r["main_event_flag"]].copy()

non_main_summary = non_main.groupby(
    "event_name", as_index=False
).agg(
    non_main_ranked_fighters=("points", lambda x: x.gt(0).sum()),
    non_main_weighted_points=("weighted_points", "sum"),
    non_main_appearances=("fighter", "count")
)

v3 = v3.merge(
    non_main_summary,
    on="event_name",
    how="left",
    validate="many_to_one"
)

# Outside-main-event ranking depth (0–100).
v3["non_main_depth_score"] = (
    100 * v3["non_main_weighted_points"]
    / (v3["non_main_appearances"] * 24)
).clip(0, 100)

# Reduce title contribution when the rest of the
# card has limited ranked talent.
v3["title_depth_adjustment"] = (
    0.5 + 0.5 * (
        v3["non_main_depth_score"] / 100
    )
)

v3["adjusted_title_factor"] = (
    v3["title_factor"]
    * v3["title_depth_adjustment"]
)

# Rebalance weights toward depth:
# 40% ranking depth outside main event
# 25% ranked matchups
# 10% title fights (depth adjusted)
# 10% experience
#  7% win %
#  4% streak
#  4% finish rate

v3["card_quality_v3_depth"] = (
    0.40 * v3["non_main_depth_score"]
    + 0.25 * v3["ranked_matchup_score"]
    + 0.10 * v3["adjusted_title_factor"]
    + 0.10 * v3["experience_factor"]
    + 0.07 * v3["win_pct_factor"]
    + 0.04 * v3["streak_factor"]
    + 0.04 * v3["finish_factor"]
).where(v3["complete_v3"])

card_quality_depth_experiment_v53 = v3.copy()

names = [
    "UFC 329", "UFC 330", "UFC 331"
]

focus = v3[
    v3["event_name"].str.contains(
        "|".join(names), case=False, na=False
    )
].copy()

print("UFC 329 / 330 / 331 COMPARISON:")
display(
    focus[[
        "event_name",
        "v3_ranked_fighters",
        "v3_ranked_matchups",
        "non_main_ranked_fighters",
        "non_main_depth_score",
        "title_factor",
        "adjusted_title_factor",
        "card_quality_v3",
        "card_quality_v3_depth"
    ]].round(2)
)

year = v3[
    v3["event_name"].isin(set(cards_2026["event_name"]))
    & v3["complete_v3"]
].sort_values(
    "card_quality_v3_depth", ascending=False
).copy()

year["depth_rank"] = range(1, len(year) + 1)

print("\n2026 DEPTH-BASED LEADERBOARD:")
display(
    year[[
        "depth_rank",
        "event_name",
        "v3_ranked_fighters",
        "non_main_ranked_fighters",
        "card_quality_v3_depth"
    ]].round(2).head(20)
)

print("\nBIG CELL 53 COMPLETE")
```

```python

# BIG CELL 57 — HISTORICAL RANKED DEPTH RECONSTRUCTION
# Preserve missing ranks as unknown; champions (rank 0) count as ranked.

import pandas as pd
import numpy as np
from IPython.display import display

h = scoring_fights[
    scoring_fights["event_name"].isin(
        set(historical_norm["event_name"].dropna())
    )
].copy()

for c in ["f1_rank_before", "f2_rank_before"]:
    h[c] = pd.to_numeric(h[c], errors="coerce")

division = h["ranking_division"].fillna("").astype(str).str.lower()

h["gender_multiplier"] = np.where(
    division.str.contains("women"), 1.0, 1.5
)

def ranking_points(rank):
    return np.where(
        rank.isna(),
        np.nan,
        np.where(
            rank.eq(0), 16,
            np.where(rank.between(1, 15), 16 - rank, 0)
        )
    )

h["f1_points"] = ranking_points(h["f1_rank_before"])
h["f2_points"] = ranking_points(h["f2_rank_before"])

h["f1_ranked"] = h["f1_rank_before"].between(0, 15)
h["f2_ranked"] = h["f2_rank_before"].between(0, 15)

h["ranked_fighters_known"] = (
    h["f1_ranked"].astype(int)
    + h["f2_ranked"].astype(int)
)

h["ranked_matchup_known"] = (
    h["f1_ranked"] & h["f2_ranked"]
).astype(int)

h["unknown_rank_slots"] = (
    h["f1_rank_before"].isna().astype(int)
    + h["f2_rank_before"].isna().astype(int)
)

h["known_weighted_points"] = (
    h[["f1_points", "f2_points"]]
    .fillna(0)
    .sum(axis=1)
    * h["gender_multiplier"]
)

historical_depth_v57 = h.groupby(
    "event_name", as_index=False
).agg(
    total_fights=("event_name", "size"),
    ranked_fighters_known=("ranked_fighters_known", "sum"),
    ranked_matchups_known=("ranked_matchup_known", "sum"),
    unknown_rank_slots=("unknown_rank_slots", "sum"),
    known_weighted_points=("known_weighted_points", "sum")
)

historical_depth_v57["known_depth_lower_bound"] = (
    100 * historical_depth_v57["known_weighted_points"]
    / (historical_depth_v57["total_fights"] * 48)
)

historical_depth_v57["ranking_slot_coverage_pct"] = (
    100 * (
        2 * historical_depth_v57["total_fights"]
        - historical_depth_v57["unknown_rank_slots"]
    )
    / (2 * historical_depth_v57["total_fights"])
)

print("HISTORICAL EVENTS:", len(historical_depth_v57))

print("\nUFC 205 AND UFC 300:")
display(
    historical_depth_v57[
        historical_depth_v57["event_name"].str.contains(
            r"UFC\s+(?:205|300)\b",
            case=False,
            na=False
        )
    ].round(2)
)

print("\nHIGHEST KNOWN RANKED-FIGHTER COUNTS:")
display(
    historical_depth_v57.sort_values(
        "ranked_fighters_known", ascending=False
    ).head(15).round(2)
)

print("\nRANKING COVERAGE DISTRIBUTION:")
print(
    historical_depth_v57["ranking_slot_coverage_pct"]
    .describe()
    .round(2)
)

print("\nBIG CELL 57 COMPLETE")
```

```python

# BIG CELL 63 — VERIFY ORIGINAL RANKING POINTS
# Read-only. No scores changed.

import pandas as pd
import numpy as np
from IPython.display import display

audit63 = h.copy()

for c in ["f1_rank_before", "f2_rank_before"]:
    audit63[c] = pd.to_numeric(audit63[c], errors="coerce")

def original_rank_points(rank):
    return np.where(
        rank.isna(), 0,
        np.where(
            rank.eq(0), 16,
            np.where(rank.between(1, 15), 16 - rank, 0)
        )
    )

audit63["original_points_rebuilt"] = (
    original_rank_points(audit63["f1_rank_before"])
    + original_rank_points(audit63["f2_rank_before"])
)

rebuilt63 = audit63.groupby(
    "event_name", as_index=False
)["original_points_rebuilt"].sum()

check63 = historical_raw[[
    "event_name", "total_ranking_points"
]].merge(
    rebuilt63,
    on="event_name",
    how="left",
    validate="one_to_one"
)

check63["difference"] = (
    check63["total_ranking_points"]
    - check63["original_points_rebuilt"]
)

matches63 = np.isclose(
    check63["difference"], 0, atol=0.01
)

print("EXACT MATCHES:", int(matches63.sum()),
      "OF", len(check63))

print("\nUFC 205, UFC 238, UFC 300:")
display(
    check63[
        check63["event_name"].str.contains(
            r"UFC\s+(?:205|238|300)\b",
            case=False, na=False
        )
    ].round(2)
)

print("\nNONMATCHING EVENTS:")
display(
    check63.loc[~matches63].sort_values(
        "difference", key=lambda s: s.abs(),
        ascending=False
    ).head(15).round(2)
)

print("\nBIG CELL 63 COMPLETE")
```

```python

# BIG CELL 64 — HISTORICAL V3 RANKING DEPTH EXPERIMENT
# Read-only. Original Event Strength Index remains unchanged.

import pandas as pd
import numpy as np
from IPython.display import display

hist64 = scoring_fights[
    scoring_fights["event_name"].isin(
        set(historical_norm["event_name"].dropna())
    )
].copy()

required = [
    "event_name", "f1_rank_before", "f2_rank_before",
    "ranking_division"
]
missing = [c for c in required if c not in hist64.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

for c in ["f1_rank_before", "f2_rank_before"]:
    hist64[c] = pd.to_numeric(hist64[c], errors="coerce")

# Use the original rank-points formula verified in Cell 63.
def rank_points64(s):
    return np.where(
        s.eq(0), 16,
        np.where(s.between(1, 15), 16 - s, 0)
    )

hist64["gender_weight64"] = np.where(
    hist64["ranking_division"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.contains("women"),
    1.0, 1.5
)

hist64["f1_points64"] = rank_points64(
    hist64["f1_rank_before"]
)
hist64["f2_points64"] = rank_points64(
    hist64["f2_rank_before"]
)

hist64["weighted_points64"] = (
    hist64["f1_points64"] + hist64["f2_points64"]
) * hist64["gender_weight64"]

hist64["ranked_fighters64"] = (
    hist64["f1_rank_before"].between(0, 15).astype(int)
    + hist64["f2_rank_before"].between(0, 15).astype(int)
)

hist64["ranked_matchup64"] = (
    hist64["f1_rank_before"].between(0, 15)
    & hist64["f2_rank_before"].between(0, 15)
).astype(int)

# Main-event detection: use existing marker if present.
main_cols = [
    c for c in ["is_main_event", "main_event"]
    if c in hist64.columns
]

if main_cols:
    marker = hist64[main_cols[0]]
    if pd.api.types.is_bool_dtype(marker):
        hist64["is_main64"] = marker.fillna(False)
    else:
        hist64["is_main64"] = (
            marker.astype(str).str.lower()
            .isin(["true", "1", "yes"])
        )
else:
    raise ValueError(
        "No main-event marker found. "
        "We must verify main-event identification first."
    )

hist64["outside_main_points64"] = np.where(
    ~hist64["is_main64"],
    hist64["weighted_points64"],
    0
)

hist64["outside_main_ranked64"] = np.where(
    ~hist64["is_main64"],
    hist64["ranked_fighters64"],
    0
)

historical_depth_v64 = hist64.groupby(
    "event_name", as_index=False
).agg(
    total_fights=("event_name", "size"),
    ranked_fighters=("ranked_fighters64", "sum"),
    ranked_matchups=("ranked_matchup64", "sum"),
    weighted_rank_points=("weighted_points64", "sum"),
    outside_main_ranked_fighters=(
        "outside_main_ranked64", "sum"
    ),
    outside_main_weighted_points=(
        "outside_main_points64", "sum"
    ),
    main_event_rows=("is_main64", "sum")
)

historical_depth_v64["outside_main_depth_score"] = (
    100
    * historical_depth_v64["outside_main_weighted_points"]
    / np.maximum(
        (historical_depth_v64["total_fights"] - 1) * 48,
        1
    )
)

print("HISTORICAL EVENTS:", len(historical_depth_v64))

print("\nMAIN-EVENT MARKER CHECK:")
print(historical_depth_v64["main_event_rows"].value_counts())

print("\nSELECTED EVENTS:")
display(
    historical_depth_v64[
        historical_depth_v64["event_name"].str.contains(
            r"UFC\s+(?:205|238|300)\b",
            case=False, na=False
        )
    ].round(2)
)

print("\nTOP 15 HISTORICAL CARDS BY OUTSIDE-MAIN DEPTH:")
display(
    historical_depth_v64.sort_values(
        "outside_main_depth_score", ascending=False
    ).head(15).round(2)
)

print("\nBIG CELL 64 COMPLETE")
```

```python

# BIG CELL 68 — BALANCED SUPPORTING-CARD DEPTH EXPERIMENT
# Does not change the existing leaderboard.

import pandas as pd
from IPython.display import display

depth68 = unified_depth_v67.copy()

# Restrict comparison to the modern ranking era.
dates68 = pd.concat([
    historical_raw[["event_name", "event_date"]],
    ranking_points_final_candidate_v3[[
        "event_name", "event_date"
    ]]
]).drop_duplicates("event_name")

depth68 = depth68.merge(
    dates68,
    on="event_name",
    how="left",
    validate="one_to_one"
)

depth68["event_date"] = pd.to_datetime(
    depth68["event_date"], errors="coerce"
)

modern68 = depth68[
    depth68["event_date"] >= "2013-01-01"
].copy()

# Average depth: rewards quality per supporting-card fight.
modern68["average_depth_percentile"] = (
    modern68["outside_main_depth_score"]
    .rank(pct=True) * 100
)

# Total depth: rewards the amount of ranked talent.
modern68["total_depth_percentile"] = (
    modern68["outside_main_weighted_points"]
    .rank(pct=True) * 100
)

modern68["balanced_depth_score"] = (
    0.50 * modern68["average_depth_percentile"]
    + 0.50 * modern68["total_depth_percentile"]
)

modern68["balanced_depth_rank"] = (
    modern68["balanced_depth_score"]
    .rank(method="min", ascending=False)
    .astype(int)
)

modern68 = modern68.sort_values(
    ["balanced_depth_rank", "event_name"]
)

print("MODERN EVENTS:", len(modern68))

print("\nTOP 20 BALANCED DEPTH:")
display(modern68[[
    "balanced_depth_rank",
    "event_name",
    "total_fights",
    "outside_main_ranked_fighters",
    "outside_main_depth_score",
    "outside_main_weighted_points",
    "balanced_depth_score"
]].head(20).round(2))

print("\nREFERENCE EVENTS:")
display(
    modern68[
        modern68["event_name"].str.contains(
            r"UFC\s+(?:200|205|300|329|330|331)\b",
            case=False, na=False
        )
    ][[
        "balanced_depth_rank",
        "event_name",
        "outside_main_ranked_fighters",
        "outside_main_depth_score",
        "balanced_depth_score",
        "ranking_status"
    ]].round(2)
)

print("\nBIG CELL 68 COMPLETE")
```

```python

# BIG CELL 69 — DEPTH WEIGHT SENSITIVITY TEST
# Experimental only. Does not modify V2 or V3.

import pandas as pd
from IPython.display import display

sensitivity69 = modern68.copy()

scenarios = {
    "quantity_focused": (0.25, 0.75),
    "balanced": (0.50, 0.50),
    "average_focused": (0.75, 0.25)
}

for name, (avg_weight, total_weight) in scenarios.items():
    sensitivity69[f"{name}_score"] = (
        avg_weight * sensitivity69["average_depth_percentile"]
        + total_weight * sensitivity69["total_depth_percentile"]
    )

    sensitivity69[f"{name}_rank"] = (
        sensitivity69[f"{name}_score"]
        .rank(method="min", ascending=False)
        .astype(int)
    )

rank_cols = [
    "quantity_focused_rank",
    "balanced_rank",
    "average_focused_rank"
]

sensitivity69["rank_range"] = (
    sensitivity69[rank_cols].max(axis=1)
    - sensitivity69[rank_cols].min(axis=1)
)

print("TOP 20 UNDER BALANCED WEIGHTING:")

display(
    sensitivity69.sort_values("balanced_rank")[[
        "event_name",
        "outside_main_ranked_fighters",
        "quantity_focused_rank",
        "balanced_rank",
        "average_focused_rank",
        "rank_range"
    ]].head(20)
)

print("\nREFERENCE EVENTS:")

display(
    sensitivity69[
        sensitivity69["event_name"].str.contains(
            r"UFC\s+(?:200|205|300|329|330|331)\b",
            case=False,
            na=False
        )
    ][[
        "event_name",
        "quantity_focused_rank",
        "balanced_rank",
        "average_focused_rank",
        "rank_range"
    ]]
)

print("\nAVERAGE RANK MOVEMENT:")
print(round(sensitivity69["rank_range"].mean(), 2))

print("\nBIG CELL 69 COMPLETE")
```

```python

# BIG CELL 71 — FULL MODERN-ERA CARD QUALITY V3 EXPERIMENT
# Does not modify original V2 scores.

import pandas as pd
import numpy as np
from IPython.display import display

base71 = modern68.copy()

hist71 = historical_norm.copy()
recent71 = card_quality_depth_experiment_v53.copy()

factor_map71 = {
    "experience": "avg_ufc_experience_score",
    "win_pct": "avg_ufc_win_pct_score",
    "streak": "avg_win_streak_score",
    "finish": "avg_finish_rate_score",
    "title": "title_fights_score"
}

# Build historical factor table.
historical_factors71 = hist71[["event_name"]].copy()

for short, col in factor_map71.items():
    historical_factors71[short] = pd.to_numeric(
        hist71[col], errors="coerce"
    )

historical_factors71["factor_source"] = "Historical"

# Build recent factor table using candidate-normalized columns.
recent_factors71 = recent71[
    recent71["event_name"].isin(
        ranking_points_final_candidate_v3["event_name"]
    )
].copy()

recent_factors71 = recent_factors71[
    ["event_name"] + [
        "candidate__" + c
        for c in factor_map71.values()
    ]
].drop_duplicates("event_name")

for short, col in factor_map71.items():
    recent_factors71[short] = pd.to_numeric(
        recent_factors71["candidate__" + col],
        errors="coerce"
    )

recent_factors71 = recent_factors71[
    ["event_name"] + list(factor_map71.keys())
].copy()

recent_factors71["factor_source"] = "Recent provisional"

factors71 = pd.concat(
    [historical_factors71, recent_factors71],
    ignore_index=True
)

# Retain the matching historical/recent factor source.
base71 = base71.merge(
    factors71,
    on="event_name",
    how="left",
    validate="one_to_one"
)

# Ranked matchups as a percentage of total fights.
base71["matchup_factor"] = (
    100 * base71["ranked_matchups"]
    / base71["total_fights"].clip(lower=1)
)

# Title fight component, with a modest supporting-card depth adjustment.
base71["title_depth_factor"] = (
    base71["title"]
    * (
        0.5
        + 0.5 * base71["balanced_depth_score"] / 100
    )
)

base71["card_quality_v3_full"] = (
    0.40 * base71["balanced_depth_score"]
    + 0.25 * base71["matchup_factor"]
    + 0.10 * base71["title_depth_factor"]
    + 0.10 * base71["experience"]
    + 0.07 * base71["win_pct"]
    + 0.04 * base71["streak"]
    + 0.04 * base71["finish"]
)

base71["card_quality_v3_rank"] = (
    base71["card_quality_v3_full"]
    .rank(method="min", ascending=False)
    .astype("Int64")
)

card_quality_v3_full_v71 = base71.sort_values(
    ["card_quality_v3_rank", "event_name"]
).reset_index(drop=True)

print("MODERN EVENTS:", len(card_quality_v3_full_v71))

print("\nFACTOR MISSING VALUES:")
print(
    card_quality_v3_full_v71[
        list(factor_map71.keys())
    ].isna().sum()
)

print("\nTOP 20 CARD QUALITY V3:")
display(
    card_quality_v3_full_v71[[
        "card_quality_v3_rank",
        "event_name",
        "card_quality_v3_full",
        "balanced_depth_score",
        "matchup_factor",
        "title_depth_factor",
        "experience",
        "win_pct",
        "streak",
        "finish"
    ]].head(20).round(2)
)

print("\nREFERENCE EVENTS:")
display(
    card_quality_v3_full_v71[
        card_quality_v3_full_v71[
            "event_name"
        ].str.contains(
            r"UFC\s+(?:200|205|300|329|330|331)\b",
            case=False,
            na=False
        )
    ][[
        "card_quality_v3_rank",
        "event_name",
        "card_quality_v3_full",
        "balanced_depth_score",
        "matchup_factor",
        "title_depth_factor",
        "ranking_status"
    ]].round(2)
)

print("\nBIG CELL 71 COMPLETE")
```

```python

# BIG CELL 77 — COMPLETE OVERALL UFC EVENT STRENGTH INDEX V3
# 50% Card Quality V3 + 25% Star Experience + 25% Main Event
# Preserves all original V2 datasets.

import pandas as pd
import numpy as np
from IPython.display import display

esi77 = overall_v3_v73.copy()

recent77 = overall_recent_leaderboard_v11[
    [
        "event_name",
        "star_experience_score",
        "main_event_strength_score"
    ]
].copy()

assert recent77["event_name"].is_unique
assert esi77["event_name"].is_unique

recent77 = recent77.rename(columns={
    "star_experience_score": "recent_star77",
    "main_event_strength_score": "recent_main77"
})

esi77 = esi77.merge(
    recent77,
    on="event_name",
    how="left",
    validate="one_to_one"
)

# Fill missing scores only; never replace existing scores.
esi77["star_v3"] = esi77["star_v3"].fillna(
    esi77["recent_star77"]
)

esi77["main_v3"] = esi77["main_v3"].fillna(
    esi77["recent_main77"]
)

# Require all three components before calculating.
ready77 = esi77[
    [
        "card_quality_v3_full",
        "star_v3",
        "main_v3"
    ]
].notna().all(axis=1)

esi77["event_strength_index_v3"] = np.where(
    ready77,
    0.50 * esi77["card_quality_v3_full"]
    + 0.25 * esi77["star_v3"]
    + 0.25 * esi77["main_v3"],
    np.nan
)

esi77["overall_v3_rank"] = (
    esi77["event_strength_index_v3"]
    .rank(method="min", ascending=False)
    .astype("Int64")
)

overall_v3_final_v77 = esi77.sort_values(
    ["overall_v3_rank", "event_name"],
    na_position="last"
).reset_index(drop=True)

print("=" * 70)
print("OVERALL ESI V3 — COMPLETENESS")
print("=" * 70)
print("Events:", len(overall_v3_final_v77))
print("Complete:", ready77.sum())
print("Incomplete:", (~ready77).sum())

print("\nTOP 30 OVERALL UFC EVENTS:")
display(
    overall_v3_final_v77[
        [
            "overall_v3_rank",
            "event_name",
            "event_strength_index_v3",
            "card_quality_v3_full",
            "star_v3",
            "main_v3",
            "ranking_status"
        ]
    ].head(30).round(2)
)

print("\nRECENT EVENTS:")
display(
    overall_v3_final_v77[
        overall_v3_final_v77["data_group"].eq("Recent")
    ][
        [
            "overall_v3_rank",
            "event_name",
            "event_strength_index_v3",
            "card_quality_v3_full",
            "star_v3",
            "main_v3",
            "ranking_status"
        ]
    ].round(2)
)

print("\nREFERENCE EVENTS:")
display(
    overall_v3_final_v77[
        overall_v3_final_v77["event_name"].str.contains(
            r"UFC\s+(?:200|205|300|329|330|331|332|333)\b",
            case=False,
            na=False
        )
    ][
        [
            "overall_v3_rank",
            "event_name",
            "event_strength_index_v3",
            "card_quality_v3_full",
            "star_v3",
            "main_v3"
        ]
    ].round(2)
)

print("\nBIG CELL 77 COMPLETE")
```

```python

# BIG CELL 79 — SUPPORTING CARD CORRECTION EXPERIMENT
# Read-only comparison. Original V2 and V3 remain unchanged.

import pandas as pd
import numpy as np
from IPython.display import display

# Historical and recent event scores
base79 = overall_v3_final_v77.copy()

# Fight-level data, including historical pre-fight rankings
fights79 = scoring_fights.copy()

# Standardize identifiers
fights79["event_name"] = fights79["event_name"].astype(str)
base79["event_name"] = base79["event_name"].astype(str)

# Only supporting-card fights
support79 = fights79[
    fights79["is_main_event"].fillna(False).eq(False)
].copy()

# Ranking point calculation:
# Champion = 16; #1 = 15; ... #15 = 1.
def rank_points79(series):
    rank = pd.to_numeric(series, errors="coerce")
    return (16 - rank).where(rank.between(0, 15), 0)

support79["f1_points"] = rank_points79(
    support79["f1_rank_before"]
)
support79["f2_points"] = rank_points79(
    support79["f2_rank_before"]
)

# Use division to distinguish women's fights.
# Unknown division is deliberately not assumed male.
division79 = support79["ranking_division"].fillna(
    ""
).astype(str).str.lower()

female79 = division79.str.contains(
    "women|womens|female|strawweight"
)

male79 = (
    division79.ne("")
    & ~female79
)

support79["gender_multiplier"] = np.select(
    [female79, male79],
    [1.0, 1.5],
    default=np.nan
)

support79["ranked_fighters_support"] = (
    (support79["f1_points"] > 0).astype(int)
    + (support79["f2_points"] > 0).astype(int)
)

support79["ranked_matchup_support"] = (
    (support79["f1_points"] > 0)
    & (support79["f2_points"] > 0)
).astype(int)

support79["weighted_points_support"] = (
    (support79["f1_points"] + support79["f2_points"])
    * support79["gender_multiplier"]
)

# Ranked matchups weighted by gender
support79["weighted_matchup_support"] = (
    support79["ranked_matchup_support"]
    * support79["gender_multiplier"]
)

agg79 = support79.groupby(
    "event_name", as_index=False
).agg(
    support_fights=("event_name", "size"),
    ranked_fighters_support=("ranked_fighters_support", "sum"),
    ranked_matchups_support=("ranked_matchup_support", "sum"),
    weighted_points_support=("weighted_points_support", "sum"),
    weighted_matchups_support=("weighted_matchup_support", "sum"),
    unknown_division_fights=("gender_multiplier",
                             lambda s: s.isna().sum())
)

# A ranked-vs-ranked men's fight receives 1.5 points.
# Maximum possible = 1.5 per supporting fight.
agg79["corrected_matchup_factor"] = (
    100 * agg79["weighted_matchups_support"]
    / (1.5 * agg79["support_fights"])
)

compare79 = base79.merge(
    agg79,
    on="event_name",
    how="left",
    validate="one_to_one"
)

# Original matchup factor from V3
compare79["old_matchup_factor"] = (
    compare79["matchup_factor"]
    if "matchup_factor" in compare79.columns
    else (
        100 * compare79["ranked_matchups"]
        / compare79["total_fights"]
    )
)

# Change only the matchup component (25% of Card Quality).
compare79["card_quality_test79"] = (
    compare79["card_quality_v3_full"]
    + 0.25 * (
        compare79["corrected_matchup_factor"]
        - compare79["old_matchup_factor"]
    )
)

# Keep the overall 50/25/25 weights unchanged.
compare79["overall_test79"] = (
    0.50 * compare79["card_quality_test79"]
    + 0.25 * compare79["star_v3"]
    + 0.25 * compare79["main_v3"]
)

compare79["test_rank79"] = (
    compare79["overall_test79"]
    .rank(ascending=False, method="min")
    .astype("Int64")
)

compare79["rank_change79"] = (
    compare79["overall_v3_rank"]
    - compare79["test_rank79"]
)

print("=" * 70)
print("IMPORTANT: DATA COVERAGE")
print("=" * 70)
print("Events in comparison:", len(compare79))
print("Events with supporting-fight data:",
      compare79["support_fights"].notna().sum())
print("Unknown-division supporting fights:",
      agg79["unknown_division_fights"].sum())
print("Events with a valid test score:",
      compare79["overall_test79"].notna().sum())

print("\n" + "=" * 70)
print("KEY EVENT COMPARISONS")
print("=" * 70)

key79 = compare79[
    compare79["event_name"].str.contains(
        r"UFC\s+(?:207|300|329|330|331)\b",
        case=False,
        na=False
    )
]

display(key79[[
    "event_name",
    "overall_v3_rank",
    "test_rank79",
    "event_strength_index_v3",
    "overall_test79",
    "old_matchup_factor",
    "corrected_matchup_factor",
    "ranked_matchups_support",
    "weighted_matchups_support",
    "unknown_division_fights"
]].round(2))

print("\n" + "=" * 70)
print("EXPERIMENTAL TOP 15")
print("=" * 70)

display(
    compare79.sort_values("test_rank79")[[
        "test_rank79",
        "event_name",
        "overall_test79",
        "overall_v3_rank",
        "rank_change79"
    ]].head(15).round(2)
)

print("\nBIG CELL 79 COMPLETE — NO ORIGINAL SCORES CHANGED")
```

```python

# BIG CELL 82 — RANKED MATCHUP QUALITY DIAGNOSTIC
# Read-only. No changes to ESI V3.

import pandas as pd
import numpy as np
from IPython.display import display

f82 = scoring_fights.copy()

# Historical supporting fights only
f82 = f82[
    f82["ranking_division"].notna()
    & ~f82["is_main_event"].fillna(False)
].copy()

div82 = f82["ranking_division"].astype(str).str.lower()

f82["multiplier82"] = np.where(
    div82.str.startswith("women"), 1.0, 1.5
)

def pts82(s):
    r = pd.to_numeric(s, errors="coerce")
    return (16-r).where(r.between(0,15),0)

f82["p1_82"] = pts82(f82["f1_rank_before"])
f82["p2_82"] = pts82(f82["f2_rank_before"])

f82["both_ranked82"] = (
    (f82["p1_82"] > 0) & (f82["p2_82"] > 0)
)

# Only ranked-vs-ranked bouts earn matchup quality.
# Their quality is based on the LOWER ranked fighter,
# rewarding matchups where BOTH fighters are highly ranked.
f82["matchup_quality82"] = np.where(
    f82["both_ranked82"],
    np.minimum(f82["p1_82"], f82["p2_82"])
    * f82["multiplier82"],
    0
)

agg82 = f82.groupby("event_name",as_index=False).agg(
    supporting_fights82=("event_name","size"),
    ranked_matchups82=("both_ranked82","sum"),
    matchup_quality_total82=("matchup_quality82","sum"),
    matchup_quality_per_fight82=("matchup_quality82","mean")
)

print("="*65)
print("UFC 300 VS UFC 207")
print("="*65)

display(
    agg82[
        agg82["event_name"].str.contains(
            r"UFC\s+(?:207|300)\b",case=False,na=False
        )
    ].round(2)
)

print("\n"+"="*65)
print("TOP 20 SUPPORTING CARDS BY MATCHUP QUALITY")
print("="*65)

display(
    agg82.sort_values(
        "matchup_quality_per_fight82",ascending=False
    ).head(20).round(2)
)

print("\nBIG CELL 82 COMPLETE")
```

```python

# BIG CELL 86 — UNIFIED EVENT STRENGTH INDEX
# 60% Card Quality / 20% Star Experience / 20% Main Event
# Improved supporting-card matchup quality
# Does NOT overwrite the original V3.

import pandas as pd
import numpy as np
from IPython.display import display

base86 = overall_v3_final_v77.copy()
f86 = scoring_fights.copy()
recent86 = ranking_points_final_candidate_v3.copy()

# Identify available fighter-name and rank columns
print("RECENT RANKING COLUMNS:")
print(recent86.columns.tolist())

print("\nFIGHT COLUMNS:")
print([
    c for c in f86.columns
    if any(x in c.lower() for x in [
        "fighter", "f1_", "f2_", "event_name",
        "rank", "division", "main"
    ])
])

# Historical matchup quality from Cell 83
historical_quality86 = agg83[
    ["event_name", "matchup_quality_score83"]
].rename(columns={
    "matchup_quality_score83": "matchup_quality_score86"
})

# Identify recent ranking data that can be joined
name_candidates86 = [
    "fighter_name", "fighter", "name",
    "fighter_full_name"
]

rank_candidates86 = [
    "rank_before", "fighter_rank",
    "rank", "rank_number"
]

name_col86 = next(
    (c for c in name_candidates86 if c in recent86.columns),
    None
)

rank_col86 = next(
    (c for c in rank_candidates86 if c in recent86.columns),
    None
)

if name_col86 is None or rank_col86 is None:
    print("\nRECENT RANKING COLUMN MATCH NOT AUTOMATIC.")
    print("Please send the column lists printed above.")
    print("No final leaderboard created yet.")

else:
    # Standardize fighter names for joining
    def normalize86(s):
        return (
            s.astype(str)
            .str.lower()
            .str.strip()
            .str.replace(r"[^a-z0-9]", "", regex=True)
        )

    recent86["fighter_key86"] = normalize86(
        recent86[name_col86]
    )

    # Work only with recent fights lacking historical divisions
    recent_fights86 = f86[
        f86["ranking_division"].isna()
    ].copy()

    # Identify fight fighter-name columns
    f1_candidates86 = [
        "f1_name", "fighter_1", "fighter1",
        "fighter_1_name"
    ]
    f2_candidates86 = [
        "f2_name", "fighter_2", "fighter2",
        "fighter_2_name"
    ]

    f1_col86 = next(
        (c for c in f1_candidates86 if c in recent_fights86.columns),
        None
    )
    f2_col86 = next(
        (c for c in f2_candidates86 if c in recent_fights86.columns),
        None
    )

    if f1_col86 is None or f2_col86 is None:
        print("\nFIGHTER NAME COLUMNS NOT IDENTIFIED.")
        print("Send the printed fight column list.")

    else:
        recent_fights86["f1_key86"] = normalize86(
            recent_fights86[f1_col86]
        )
        recent_fights86["f2_key86"] = normalize86(
            recent_fights86[f2_col86]
        )

        lookup86 = recent86[
            ["event_name", "fighter_key86", rank_col86]
        ].drop_duplicates(
            ["event_name", "fighter_key86"]
        )

        lookup86 = lookup86.rename(
            columns={rank_col86: "recent_rank86"}
        )

        recent_fights86 = recent_fights86.merge(
            lookup86.rename(columns={
                "fighter_key86": "f1_key86",
                "recent_rank86": "rank1_86"
            }),
            on=["event_name", "f1_key86"],
            how="left",
            validate="many_to_one"
        )

        recent_fights86 = recent_fights86.merge(
            lookup86.rename(columns={
                "fighter_key86": "f2_key86",
                "recent_rank86": "rank2_86"
            }),
            on=["event_name", "f2_key86"],
            how="left",
            validate="many_to_one"
        )

        # A missing ranking can mean unranked, or unknown.
        # Don't assume missing = unranked.
        def points86(s):
            r = pd.to_numeric(s, errors="coerce")
            return (16-r).where(r.between(0,15),0)

        recent_fights86["p1_86"] = points86(
            recent_fights86["rank1_86"]
        )
        recent_fights86["p2_86"] = points86(
            recent_fights86["rank2_86"]
        )

        # Weight women's fights 1.0, men's fights 1.5
        wc86 = recent_fights86["WEIGHTCLASS"].astype(str).str.lower()
        women86 = wc86.str.contains(
            "women|w strawweight|w flyweight|w bantamweight"
        )

        recent_fights86["weight86"] = np.where(
            women86,1.0,1.5
        )

        recent_fights86["quality86"] = np.where(
            (recent_fights86["p1_86"] > 0)
            & (recent_fights86["p2_86"] > 0),
            np.minimum(
                recent_fights86["p1_86"],
                recent_fights86["p2_86"]
            ) * recent_fights86["weight86"],
            0
        )

        # Exclude main events
        supporting86 = recent_fights86[
            ~recent_fights86["is_main_event"].fillna(False)
        ].copy()

        recent_quality86 = supporting86.groupby(
            "event_name",as_index=False
        ).agg(
            supporting_fights86=("event_name","size"),
            matchup_quality_avg86=("quality86","mean"),
            matched_rank1_86=("rank1_86","count"),
            matched_rank2_86=("rank2_86","count")
        )

        # Same historical reference distribution as Cell 83
        historical_avg86 = agg83[
            "matchup_quality_avg83"
        ].dropna().to_numpy()

        recent_quality86["matchup_quality_score86"] = (
            recent_quality86["matchup_quality_avg86"]
            .apply(
                lambda x: 100 * np.mean(
                    historical_avg86 <= x
                )
            )
        )

        all_quality86 = pd.concat([
            historical_quality86,
            recent_quality86[[
                "event_name",
                "matchup_quality_score86"
            ]]
        ],ignore_index=True)

        unified86 = base86.merge(
            all_quality86,
            on="event_name",
            how="left",
            validate="one_to_one"
        )

        old_matchup86 = (
            unified86["matchup_factor"]
            if "matchup_factor" in unified86.columns
            else (
                100 * unified86["ranked_matchups"]
                / unified86["total_fights"]
            )
        )

        unified86["card_quality_final86"] = (
            unified86["card_quality_v3_full"]
            + .25 * (
                unified86["matchup_quality_score86"]
                - old_matchup86
            )
        )

        unified86["event_strength_final86"] = (
            .60 * unified86["card_quality_final86"]
            + .20 * unified86["star_v3"]
            + .20 * unified86["main_v3"]
        )

        unified86["rank_final86"] = (
            unified86["event_strength_final86"]
            .rank(ascending=False,method="min")
            .astype("Int64")
        )

        print("\n" + "="*65)
        print("FINAL MODEL COVERAGE")
        print("="*65)
        print("Events:",len(unified86))
        print("Scored:",
              unified86["event_strength_final86"]
              .notna().sum())

        print("\nRECENT FIGHTER MATCHING:")
        display(recent_quality86[[
            "event_name",
            "supporting_fights86",
            "matched_rank1_86",
            "matched_rank2_86",
            "matchup_quality_score86"
        ]])

        print("\nKEY EVENT COMPARISON:")
        display(
            unified86[
                unified86["event_name"].str.contains(
                    r"UFC (?:207|300|330|331):",
                    regex=True,na=False
                )
            ][[
                "event_name",
                "rank_final86",
                "event_strength_final86",
                "card_quality_final86",
                "matchup_quality_score86"
            ]].round(2)
        )

        print("\nUNIFIED TOP 20:")
        display(
            unified86.sort_values(
                "rank_final86"
            )[[
                "rank_final86",
                "event_name",
                "event_strength_final86"
            ]].head(20).round(2)
        )

        print("\nBIG CELL 86 COMPLETE")
```

```python

# BIG CELL 87 — REPAIR UNIFIED MATCHUP QUALITY
# Does not overwrite original V3.

import pandas as pd
import numpy as np
from IPython.display import display

base87 = overall_v3_final_v77.copy()
f87 = scoring_fights.copy()
r87 = ranking_points_final_candidate_v3.copy()

def key87(s):
    return (s.astype(str).str.lower().str.strip()
            .str.replace(r"[^a-z0-9]", "", regex=True))

# Only events in the recent ranking table are treated as recent
recent_events87 = set(r87["event_name"].dropna())

f87 = f87[f87["event_name"].isin(recent_events87)].copy()

# One row per fighter appearance, with event and fight identity
r87["fighter_key87"] = key87(r87["fighter"])
r87["fight_key87"] = r87["fight_url"].astype(str).str.strip()

# Use already-calculated ranking points, not rank_before
r87["points87"] = pd.to_numeric(
    r87["final_ranking_points"], errors="coerce"
)

# Fall back to ranking_points_v2 only where needed
fallback87 = pd.to_numeric(
    r87["ranking_points_v2"], errors="coerce"
)
r87["points87"] = r87["points87"].fillna(fallback87)

# Keep only unique event + fighter appearances
duplicates87 = r87.duplicated(
    ["event_name", "fighter_key87"], keep=False
)

print("Duplicate recent fighter appearances:",
      duplicates87.sum())

if duplicates87.any():
    display(r87.loc[duplicates87, [
        "event_name", "fighter", "fight_url",
        "points87"
    ]].head(20))

lookup87 = r87[
    ["event_name", "fighter_key87", "points87"]
].drop_duplicates(
    ["event_name", "fighter_key87"],
    keep="first"
)

f87["f1_key87"] = key87(f87["f1_name"])
f87["f2_key87"] = key87(f87["f2_name"])

f87 = f87.merge(
    lookup87.rename(columns={
        "fighter_key87": "f1_key87",
        "points87": "p1_87"
    }),
    on=["event_name", "f1_key87"],
    how="left",
    validate="many_to_one"
)

f87 = f87.merge(
    lookup87.rename(columns={
        "fighter_key87": "f2_key87",
        "points87": "p2_87"
    }),
    on=["event_name", "f2_key87"],
    how="left",
    validate="many_to_one"
)

# Exclude main events
support87 = f87[
    ~f87["is_main_event"].fillna(False)
].copy()

wc87 = support87["WEIGHTCLASS"].astype(str).str.lower()

support87["weight87"] = np.where(
    wc87.str.contains("women|w strawweight|w flyweight|w bantamweight"),
    1.0, 1.5
)

# Points are already 16-rank.
# Both fighters need positive points for a ranked matchup.
support87["quality87"] = np.where(
    support87["p1_87"].notna()
    & support87["p2_87"].notna(),
    np.minimum(support87["p1_87"],support87["p2_87"])
    .clip(lower=0) * support87["weight87"],
    np.nan
)

recent_quality87 = support87.groupby(
    "event_name",as_index=False
).agg(
    supporting_fights87=("event_name","size"),
    known_p1_87=("p1_87","count"),
    known_p2_87=("p2_87","count"),
    quality_avg87=("quality87","mean")
)

recent_quality87["complete87"] = (
    (recent_quality87["known_p1_87"]
     == recent_quality87["supporting_fights87"])
    & (recent_quality87["known_p2_87"]
       == recent_quality87["supporting_fights87"])
)

# Do not score partially known recent cards
recent_quality87.loc[
    ~recent_quality87["complete87"],"quality_avg87"
] = np.nan

# Normalize using historical supporting matchup quality
hist_distribution87 = agg83[
    "matchup_quality_avg83"
].dropna().to_numpy()

recent_quality87["matchup_quality_score87"] = (
    recent_quality87["quality_avg87"].apply(
        lambda x: (
            100*np.mean(hist_distribution87 <= x)
            if pd.notna(x) else np.nan
        )
    )
)

# Historical quality only for events NOT in recent table
hist_quality87 = agg83[
    ~agg83["event_name"].isin(recent_events87)
][["event_name","matchup_quality_score83"]].rename(
    columns={"matchup_quality_score83":"matchup_quality_score87"}
)

all_quality87 = pd.concat([
    hist_quality87,
    recent_quality87[[
        "event_name","matchup_quality_score87"
    ]]
],ignore_index=True)

# Prevent duplicate event keys
assert not all_quality87["event_name"].duplicated().any(), \
    "Duplicate event names remain in matchup-quality table."

unified87 = base87.merge(
    all_quality87,
    on="event_name",
    how="left",
    validate="one_to_one"
)

old87 = (
    unified87["matchup_factor"]
    if "matchup_factor" in unified87.columns
    else 100*unified87["ranked_matchups"] /
         unified87["total_fights"]
)

unified87["card_quality_final87"] = (
    unified87["card_quality_v3_full"]
    + .25*(unified87["matchup_quality_score87"]-old87)
)

unified87["event_strength_final87"] = (
    .60*unified87["card_quality_final87"]
    + .20*unified87["star_v3"]
    + .20*unified87["main_v3"]
)

unified87["rank_final87"] = (
    unified87["event_strength_final87"]
    .rank(ascending=False,method="min")
    .astype("Int64")
)

print("\nRECENT MATCHING AND COMPLETENESS:")
display(recent_quality87[[
    "event_name","supporting_fights87",
    "known_p1_87","known_p2_87",
    "complete87","matchup_quality_score87"
]])

print("\nKEY EVENTS:")
display(unified87[
    unified87["event_name"].str.contains(
        r"UFC (?:207|300|330|331):",na=False
    )
][[
    "event_name","rank_final87",
    "event_strength_final87","card_quality_final87",
    "matchup_quality_score87"
]].round(2))

print("\nTOP 15:")
display(unified87.sort_values(
    "event_strength_final87",ascending=False
)[[
    "rank_final87","event_name","event_strength_final87"
]].head(15).round(2))

print("\nCOVERAGE:",
      unified87["event_strength_final87"].notna().sum(),
      "/",len(unified87))

print("\nBIG CELL 87 COMPLETE")
```

```python

# BIG CELL 94 — REPAIR CANCELED MAIN EVENT
# Uses existing historical scoring scales.
# Does not change the 70/20/10 ESI weights.

import numpy as np
import pandas as pd
from IPython.display import display

event94 = "UFC Fight Night: Blanchfield vs. Barber"

components94 = [
    "main_event_combined_ufc_fights",
    "main_event_avg_win_pct",
    "main_event_combined_win_streak",
    "main_event_combined_title_fights",
    "main_event_combined_previous_main_events",
    "main_event_combined_bonuses",
    "main_event_ranking_points"
]

# Reconstruct the actual main-event raw values.
fight94 = scoring_fights.loc[
    (scoring_fights["event_name"] == event94)
    & (scoring_fights["is_main_event"] == True)
]

assert len(fight94) == 1, "Expected exactly one actual main event."

f94 = fight94.iloc[0]

raw94 = {
    "main_event_combined_ufc_fights":
        f94["f1_fights_before"] + f94["f2_fights_before"],
    "main_event_avg_win_pct":
        np.nanmean([
            f94["f1_ufc_win_pct_before"],
            f94["f2_ufc_win_pct_before"]
        ]),
    "main_event_combined_win_streak":
        f94["f1_win_streak_before"] + f94["f2_win_streak_before"],
    "main_event_combined_title_fights":
        f94["f1_previous_title_fights"] + f94["f2_previous_title_fights"],
    "main_event_combined_previous_main_events":
        f94["f1_previous_main_events"] + f94["f2_previous_main_events"],
    "main_event_combined_bonuses":
        f94["f1_previous_total_bonuses"] + f94["f2_previous_total_bonuses"]
}

# Historical ranking-points convention:
# 16 minus ranking for a ranked fighter.
raw94["main_event_ranking_points"] = sum(
    16 - float(f94[c])
    for c in ["f1_rank_before", "f2_rank_before"]
    if pd.notna(f94[c]) and 1 <= float(f94[c]) <= 15
)

print("ACTUAL MAIN EVENT RAW VALUES:")
display(pd.DataFrame([raw94]).T.rename(columns={0: "Value"}))

# Recover each component's normalization using the historical
# raw-value to normalized-score relationship.
normalized94 = {}

for component94 in components94:
    score_col94 = component94 + "_score"

    paired94 = historical_raw[
        ["event_name", component94]
    ].merge(
        historical_norm[["event_name", score_col94]],
        on="event_name",
        how="inner"
    ).dropna()

    # Exact raw values should have consistent normalized scores.
    same94 = paired94.loc[
        np.isclose(
            paired94[component94].astype(float),
            float(raw94[component94]),
            atol=1e-9
        ),
        score_col94
    ]

    if len(same94) > 0:
        normalized94[score_col94] = float(same94.median())
    else:
        # Interpolate within observed historical scoring scale.
        points94 = (
            paired94.groupby(component94, as_index=False)[score_col94]
            .median()
            .sort_values(component94)
        )

        x94 = points94[component94].to_numpy(dtype=float)
        y94 = points94[score_col94].to_numpy(dtype=float)
        target94 = float(raw94[component94])

        if len(x94) < 2 or not (x94.min() <= target94 <= x94.max()):
            raise ValueError(
                f"Cannot safely normalize {component94}: "
                "outside observed historical range."
            )

        normalized94[score_col94] = float(
            np.interp(target94, x94, y94)
        )

print("\nRECONSTRUCTED NORMALIZED COMPONENTS:")
display(pd.DataFrame([normalized94]).T.rename(columns={0: "Score"}))

# Recover the linear combination from complete historical rows.
score_columns94 = [c + "_score" for c in components94]

training94 = historical_norm.dropna(
    subset=score_columns94 + ["main_event_score"]
).copy()

X94 = training94[score_columns94].to_numpy(dtype=float)
y94 = training94["main_event_score"].to_numpy(dtype=float)

# No intercept: score is expected to be a weighted combination.
weights94, *_ = np.linalg.lstsq(X94, y94, rcond=None)

pred94 = X94 @ weights94
max_error94 = float(np.max(np.abs(pred94 - y94)))

print("\nRECOVERED COMPONENT WEIGHTS:")
display(pd.DataFrame({
    "Component": components94,
    "Weight": weights94
}))

print("Maximum historical reconstruction error:", round(max_error94, 8))

# Refuse to apply an unvalidated approximation.
if max_error94 > 0.01:
    print("\nSTOP: Original formula was not reproduced accurately.")
    print("Leaderboard unchanged.")
else:
    replacement_main94 = float(
        np.array([
            normalized94[c] for c in score_columns94
        ]) @ weights94
    )

    repaired94 = clean90.copy()
    mask94 = repaired94["event_name"].eq(event94)

    repaired94.loc[mask94, "main_v3"] = replacement_main94

    repaired94["event_strength_v94"] = (
        0.70 * repaired94["card_quality_final87"]
        + 0.20 * repaired94["main_v3"]
        + 0.10 * repaired94["star_v3"]
    )

    repaired94["rank_v94"] = (
        repaired94["event_strength_v94"]
        .rank(ascending=False, method="min")
        .astype("Int64")
    )

    print("\nREPAIRED EVENT:")
    display(repaired94.loc[
        mask94,
        ["event_name", "main_v3", "event_strength_v94", "rank_v94"]
    ].round(2))

    print("\nEVENTS:", len(repaired94))
    print("SCORED:", repaired94["event_strength_v94"].notna().sum())
    print("MISSING:", repaired94["event_strength_v94"].isna().sum())

print("\nBIG CELL 94 COMPLETE")
```

```python

# BIG CELL 106 — UFC 333 CARD QUALITY INPUTS
# Prepares data; does not modify the 569-event leaderboard.

import pandas as pd
import numpy as np

card106 = ufc333_verified_card_v13.copy()
fighters106 = ufc333_recovery101.copy()

def points106(row):
    if row["rank_type"] == "Champion":
        return 16.0
    if pd.notna(row["rank_event_page"]):
        return float(16 - row["rank_event_page"])
    return 0.0

fighters106["ranking_points_candidate"] = fighters106.apply(
    points106, axis=1
)

# Match each fighter to the official card.
left106 = fighters106.add_prefix("f1_")
right106 = fighters106.add_prefix("f2_")

combined106 = (
    card106
    .merge(left106, left_on="f1_name", right_on="f1_fighter",
           how="left", validate="one_to_one")
    .merge(right106, left_on="f2_name", right_on="f2_fighter",
           how="left", validate="one_to_one")
)

assert len(combined106) == 14, "Fight count mismatch"

for side in ["f1", "f2"]:
    assert combined106[f"{side}_fighter"].notna().all(), (
        f"Missing fighter match for {side}"
    )

combined106["ranked_fighters_in_fight"] = (
    (combined106["f1_rank_type"].isin(["Ranked", "Champion"])).astype(int)
    + (combined106["f2_rank_type"].isin(["Ranked", "Champion"])).astype(int)
)

combined106["ranked_matchup"] = (
    combined106["ranked_fighters_in_fight"] == 2
)

combined106["ranking_points_in_fight"] = (
    combined106["f1_ranking_points_candidate"]
    + combined106["f2_ranking_points_candidate"]
)

summary106 = pd.DataFrame([{
    "event_name": card106["event_name"].iloc[0],
    "total_fights": len(combined106),
    "fighter_appearances": len(fighters106),
    "title_fights": int(card106["is_title_fight"].sum()),
    "ranked_fighters": int(
        fighters106["rank_type"].isin(["Ranked", "Champion"]).sum()
    ),
    "ranked_matchups": int(combined106["ranked_matchup"].sum()),
    "outside_main_ranked_fighters": int(
        combined106.loc[
            ~combined106["is_main_event"],
            "ranked_fighters_in_fight"
        ].sum()
    ),
    "outside_main_weighted_points": float(
        combined106.loc[
            ~combined106["is_main_event"],
            "ranking_points_in_fight"
        ].sum()
    ),
    "avg_ufc_experience": fighters106["ufc_fights_final"].mean(),
    "avg_ufc_win_pct": fighters106["ufc_win_pct_final"].mean(),
    "avg_win_streak": fighters106["win_streak"].mean(),
    "avg_previous_title_fights":
        fighters106["previous_title_fights"].mean(),
    "avg_previous_main_events":
        fighters106["previous_main_events"].mean(),
    "avg_previous_bonuses":
        fighters106["bonus_estimate"].mean(),
    "total_ranking_points":
        fighters106["ranking_points_candidate"].sum()
}])

print("UFC 333 CARD QUALITY INPUT SUMMARY")
display(summary106.T)

print("\nDATA QUALITY CHECK")
print("Missing UFC fight totals:",
      fighters106["ufc_fights_final"].isna().sum())
print("Missing UFC win percentages:",
      fighters106["ufc_win_pct_final"].isna().sum())
print("Missing winning streaks:",
      fighters106["win_streak"].isna().sum())
print("Estimated bonuses:",
      fighters106["bonus_imputed"].sum())
print("Pre-event rankings verified:",
      fighters106["ranking_pre_event_verified"].sum(), "of 28")

ufc333_card_inputs106 = summary106.copy()
ufc333_fight_inputs106 = combined106.copy()

print("\nBIG CELL 106 COMPLETE")
```

```python

# BIG CELL 109 — RECOVER FINAL CARD QUALITY FORMULA
# Read-only; original leaderboard remains unchanged.

import numpy as np
import pandas as pd

df109 = card_quality_v3_full_v71.copy()

factors109 = [
    "balanced_depth_score",
    "experience",
    "win_pct",
    "streak",
    "finish",
    "title",
    "matchup_factor",
    "title_depth_factor"
]

X109 = df109[factors109].apply(
    pd.to_numeric, errors="coerce"
)

y109 = pd.to_numeric(
    df109["card_quality_v3_full"],
    errors="coerce"
)

valid109 = X109.notna().all(axis=1) & y109.notna()

X109 = X109.loc[valid109]
y109 = y109.loc[valid109]

# Recover weights without changing historical scores.
weights109, _, _, _ = np.linalg.lstsq(
    X109.to_numpy(),
    y109.to_numpy(),
    rcond=None
)

pred109 = X109.to_numpy() @ weights109
error109 = np.abs(pred109 - y109.to_numpy())

result109 = pd.DataFrame({
    "component": factors109,
    "recovered_weight": weights109,
    "percentage": weights109 * 100
})

print("RECOVERED CARD QUALITY WEIGHTS")
display(result109)

print("\nMODEL RECONSTRUCTION CHECK")
print("Events checked:", len(X109))
print("Weight sum:", round(weights109.sum(), 8))
print("Mean absolute error:", round(error109.mean(), 8))
print("Maximum absolute error:", round(error109.max(), 8))

print("\nCOMPONENT RANGES")
display(pd.DataFrame({
    "component": factors109,
    "minimum": X109.min().values,
    "maximum": X109.max().values,
    "median": X109.median().values
}))

ufc333_card_weights109 = dict(zip(factors109, weights109))

print("\nBIG CELL 109 COMPLETE")
```

```python

# BIG CELL 111 — UFC 333 FACTOR NORMALIZATION
# Read-only; does not change the leaderboard.

import numpy as np
import pandas as pd

history111 = card_quality_v3_experiment.copy()
fighters111 = ufc333_recovery101.copy()

# UFC debut: exclude undefined winning percentage from average.
raw111 = {
    "avg_ufc_experience": fighters111["ufc_fights_final"].mean(),
    "avg_ufc_win_pct": fighters111["ufc_win_pct_final"].mean(),
    "avg_win_streak": fighters111["win_streak"].mean()
}

# Finish rate from the previously prepared UFC 333 finish audit.
print("FINISH AUDIT COLUMNS:")
print(list(ufc333_finish_audit_v23.columns))

pairs111 = [
    ("avg_ufc_experience", "experience_factor"),
    ("avg_ufc_win_pct", "win_pct_factor"),
    ("avg_win_streak", "streak_factor")
]

results111 = []

for raw_col, score_col in pairs111:
    subset = history111[[raw_col, score_col]].copy()
    subset = subset.apply(pd.to_numeric, errors="coerce").dropna()
    subset = subset.sort_values(raw_col)

    # Percentile-style interpolation based on historical model values.
    grouped = subset.groupby(raw_col)[score_col].mean()

    predicted = np.interp(
        raw111[raw_col],
        grouped.index.to_numpy(),
        grouped.to_numpy()
    )

    results111.append({
        "factor": score_col,
        "UFC_333_raw": raw111[raw_col],
        "estimated_normalized_score": predicted,
        "historical_min": subset[raw_col].min(),
        "historical_max": subset[raw_col].max()
    })

normalized111 = pd.DataFrame(results111)

print("\nUFC 333 NORMALIZED FACTORS")
display(normalized111)

print("\nBIG CELL 111 COMPLETE")
```

```python

# BIG CELL 113 — NORMALIZE UFC 333 AGAINST HISTORICAL REFERENCE
# No leaderboard modifications.

import numpy as np
import pandas as pd

fighters113 = ufc333_recovery101.copy()
finish113 = ufc333_finish_audit_v23.copy()
reference113 = card_quality_v3_experiment.copy()

# Use UFC debut correction; don't overwrite undefined win percentage.
finish113.loc[
    finish113["fighter"].eq("Asaf Chopurov"),
    "ufc_fights_final"
] = 0

raw113 = {
    "experience": fighters113["ufc_fights_final"].mean(),
    "win_pct": fighters113["ufc_win_pct_final"].mean(),
    "streak": fighters113["win_streak"].mean(),
    "finish": pd.to_numeric(
        finish113["finish_rate_baseline"],
        errors="coerce"
    ).mean()
}

mapping113 = {
    "experience": ("avg_ufc_experience", "experience_factor"),
    "win_pct": ("avg_ufc_win_pct", "win_pct_factor"),
    "streak": ("avg_win_streak", "streak_factor"),
    "finish": ("avg_finish_rate", "finish_factor")
}

scores113 = {}
rows113 = []

for factor, (raw_col, score_col) in mapping113.items():
    historical = reference113[[raw_col, score_col]].copy()
    historical = historical.apply(
        pd.to_numeric, errors="coerce"
    ).dropna()

    historical = (
        historical.groupby(raw_col, as_index=False)[score_col]
        .mean()
        .sort_values(raw_col)
    )

    score = float(np.interp(
        raw113[factor],
        historical[raw_col].to_numpy(),
        historical[score_col].to_numpy()
    ))

    scores113[factor] = score

    rows113.append({
        "factor": factor,
        "raw_value": raw113[factor],
        "normalized_score": score,
        "reference_events": len(historical),
        "outside_reference_range": bool(
            raw113[factor] < historical[raw_col].min()
            or raw113[factor] > historical[raw_col].max()
        )
    })

ufc333_factor_scores113 = scores113

print("UFC 333 CARD QUALITY FIGHTER FACTORS")
display(pd.DataFrame(rows113))

print("\nBIG CELL 113 COMPLETE")
```

```python

# BIG CELL 116 — RECOVER CARD DEPTH FORMULA
# Read-only; original leaderboard unchanged.

import numpy as np
import pandas as pd

d = card_quality_v3_full_v71.copy()

p = d["outside_main_weighted_points"].astype(float)
r = d["outside_main_ranked_fighters"].astype(float)
f = d["total_fights"].astype(float)
target = d["outside_main_depth_score"].astype(float)

# Test plausible original depth denominators.
tests = {
    "points / (2 * outside fights)": p / (2 * (f - 1)),
    "points * ranked / (2 * outside fights)": p * r / (2 * (f - 1)),
    "points * ranked / outside fights": p * r / (f - 1),
    "points * ranked / (outside fights squared)": p * r / ((f - 1) ** 2),
    "points * ranked / (4 * outside fights)": p * r / (4 * (f - 1)),
    "points / total fights": p / f,
    "points * ranked / total fights squared": p * r / (f ** 2)
}

print("DEPTH FORMULA ERRORS")
for name, prediction in tests.items():
    diff = (prediction - target).abs()
    print(f"{name}: max={diff.max():.8f}, mean={diff.mean():.8f}")

print("\nHISTORICAL DEPTH PERCENTILE VALIDATION")
for raw, pct in [
    ("outside_main_depth_score", "average_depth_percentile"),
    ("outside_main_weighted_points", "total_depth_percentile")
]:
    expected = d[raw].rank(pct=True, method="average") * 100
    diff = (expected - d[pct]).abs()
    print(f"{pct}: max error={diff.max():.8f}, mean error={diff.mean():.8f}")

print("\nBIG CELL 116 COMPLETE")
```

```python

# BIG CELL 126 — RECOVER TITLE DEPTH FORMULA
import numpy as np
import pandas as pd

d126 = repaired94.copy()

title = d126["title"].astype(float)
depth = d126["balanced_depth_score"].astype(float)
actual = d126["title_depth_factor"].astype(float)

# Calculate the actual multiplier used historically.
multiplier = actual / title.replace(0, np.nan)

results = pd.DataFrame({
    "event_name": d126["event_name"],
    "title": title,
    "depth": depth,
    "actual_multiplier": multiplier
})

print("1. ACTUAL MULTIPLIER DISTRIBUTION")
display(multiplier.describe())

print("\n2. RECOVER LINEAR MULTIPLIER RELATIONSHIP")

valid = (
    np.isfinite(multiplier)
    & np.isfinite(depth)
)

x = depth[valid].to_numpy()
y = multiplier[valid].to_numpy()

slope, intercept = np.polyfit(x, y, 1)

print("Slope:", slope)
print("Intercept:", intercept)

prediction = title * (intercept + slope * depth)
error = (prediction - actual).abs()

print("Mean error:", error.mean())
print("Maximum error:", error.max())

print("\n3. TEST PIECEWISE DEPTH MULTIPLIER")

for base in [0.4, 0.5, 0.6, 0.7]:
    pred = title * (
        base + (1 - base) * depth / 100
    )
    err = (pred - actual).abs()

    print(
        f"Base {base:.1f}: "
        f"mean={err.mean():.6f}, "
        f"max={err.max():.6f}, "
        f"exact={(err < 1e-8).sum()}"
    )

print("\n4. SHOW LOW-DEPTH EVENTS")
display(results.sort_values("depth").head(20))

print("\nBIG CELL 126 COMPLETE")
```

```python

# BIG CELL 127 — MATCHUP QUALITY PERCENTILE RECOVERY

import numpy as np
import pandas as pd

d127 = repaired94.copy()

target127 = d127["matchup_quality_score87"]

candidates127 = [
    "ranked_matchups",
    "ranked_fighters",
    "outside_main_ranked_fighters",
    "outside_main_weighted_points",
    "matchup_factor",
    "balanced_depth_score"
]

results127 = []

for col in candidates127:
    values = d127[col].astype(float)

    for method in ["average", "min", "max", "dense"]:
        percentile = values.rank(
            method=method,
            pct=True
        ) * 100

        error = (percentile - target127).abs()

        results127.append({
            "variable": col,
            "method": method,
            "mean_error": error.mean(),
            "max_error": error.max()
        })

print("1. BEST SINGLE-VARIABLE PERCENTILE MATCHES")

display(
    pd.DataFrame(results127)
    .sort_values("mean_error")
    .head(15)
)

print("\n2. CHECK MATCHUP QUALITY SCORE DISTRIBUTION")

display(target127.describe())

print("\n3. CHECK COMMON MATCHUP QUALITY VALUES")

display(
    target127.round(6)
    .value_counts()
    .head(15)
)

print("\n4. CHECK WHETHER MATCHUP QUALITY TRACKS RANKED MATCHUPS")

display(
    d127.groupby("ranked_matchups")[
        "matchup_quality_score87"
    ].agg(["count", "min", "mean", "max"])
)

print("\nBIG CELL 127 COMPLETE")
```

```python

# BIG CELL 135 — TOP 10 UFC EVENTS OF 2026

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

df135 = final134.copy()

# Convert dates and identify 2026 events
df135["event_date"] = pd.to_datetime(
    df135["event_date"], errors="coerce"
)

events2026 = df135[
    df135["event_date"].dt.year == 2026
].copy()

# Find top 10 using the final ESI score
top10_2026 = events2026.nlargest(
    10, "event_strength_v94"
).sort_values("event_strength_v94")

# Build a professional chart
fig, ax = plt.subplots(figsize=(13, 7))

fig.patch.set_facecolor("#101820")
ax.set_facecolor("#101820")

bars = ax.barh(
    top10_2026["event_name"],
    top10_2026["event_strength_v94"],
    color="#D71920"
)

ax.set_xlim(0, 105)
ax.set_title(
    "TOP 10 UFC EVENTS OF 2026",
    fontsize=20,
    fontweight="bold",
    color="white",
    pad=20
)

ax.set_xlabel("Event Strength Index (0–100)", color="white")
ax.tick_params(colors="white")
ax.spines[:].set_visible(False)

for bar, score in zip(
    bars,
    top10_2026["event_strength_v94"]
):
    ax.text(
        score + 1,
        bar.get_y() + bar.get_height()/2,
        f"{score:.2f}",
        va="center",
        color="white",
        fontsize=10
    )

fig.text(
    0.5, 0.015,
    "UFC Event Strength Index | UFC 333 provisional",
    ha="center",
    color="#BBBBBB",
    fontsize=10
)

plt.tight_layout(rect=[0, 0.04, 1, 1])

plt.savefig(
    "top_10_ufc_events_2026.png",
    dpi=220,
    bbox_inches="tight",
    facecolor=fig.get_facecolor()
)

plt.show()

print("2026 events found:", len(events2026))
print("Chart saved successfully.")

files.download("top_10_ufc_events_2026.png")
```

```python

# BIG CELL 137 — FOUR UFC ANALYTICS CHARTS

import pandas as pd
import matplotlib.pyplot as plt
import os
import zipfile
from google.colab import files

df137 = final134.copy()

# Chart title, score column, and filename
charts137 = [
    (
        "TOP 10 UFC EVENTS OF ALL TIME",
        "event_strength_v94",
        "top_10_ufc_events_all_time.png"
    ),
    (
        "TOP 10 UFC EVENTS BY CARD QUALITY",
        "card_quality_final87",
        "top_10_ufc_card_quality.png"
    ),
    (
        "TOP 10 UFC EVENTS BY STAR EXPERIENCE",
        "star_v3",
        "top_10_ufc_star_experience.png"
    ),
    (
        "TOP 10 UFC EVENTS BY MAIN EVENT STRENGTH",
        "main_v3",
        "top_10_ufc_main_event_strength.png"
    )
]

os.makedirs("ufc_charts", exist_ok=True)

for title, column, filename in charts137:

    # Select strongest 10 events for this metric
    top10 = (
        df137.nlargest(10, column)
        .sort_values(column)
    )

    fig, ax = plt.subplots(figsize=(15, 8))

    fig.patch.set_facecolor("#101820")
    ax.set_facecolor("#101820")

    bars = ax.barh(
        top10["event_name"],
        top10[column],
        color="#D71920",
        height=0.72
    )

    ax.set_xlim(0, 105)

    ax.set_title(
        title,
        fontsize=21,
        fontweight="bold",
        color="white",
        pad=22
    )

    ax.set_xlabel(
        "Score (0–100)",
        fontsize=12,
        color="white"
    )

    ax.tick_params(
        axis="both",
        colors="white",
        labelsize=11
    )

    ax.spines[:].set_visible(False)

    for bar, score in zip(bars, top10[column]):
        ax.text(
            score + 0.8,
            bar.get_y() + bar.get_height()/2,
            f"{score:.2f}",
            va="center",
            color="white",
            fontsize=11
        )

    fig.text(
        0.5, 0.015,
        "UFC Event Strength Index | 570-event dataset | UFC 333 provisional",
        ha="center",
        color="#BBBBBB",
        fontsize=10
    )

    plt.tight_layout(rect=[0, 0.04, 1, 1])

    path = os.path.join("ufc_charts", filename)

    plt.savefig(
        path,
        dpi=220,
        bbox_inches="tight",
        facecolor=fig.get_facecolor()
    )

    plt.show()
    plt.close(fig)

    print("\n", title)
    display(
        top10[["event_name", column]]
        .sort_values(column, ascending=False)
        .reset_index(drop=True)
    )

# Package all four images into one ZIP
zip_name = "ufc_analytics_charts.zip"

with zipfile.ZipFile(zip_name, "w") as z:
    for _, _, filename in charts137:
        z.write(
            os.path.join("ufc_charts", filename),
            arcname=filename
        )

files.download(zip_name)

print("\nBIG CELL 137 COMPLETE")
```
